```python
import math

def calculate_wet_bulb(t, rh):
    """
    Calculates wet bulb temperature from dry bulb temperature and relative humidity.
    
    Parameters:
    t  (float): Air (dry bulb) temperature in Celsius (e.g., 25.0)
    rh (float): Relative humidity in percent (e.g., 50.0 for 50%)
    
    Returns:
    float: Wet bulb temperature in Celsius
    """
    # Roland Stull's empirical formula
    tw = (
        t * math.atan(0.151977 * (rh + 8.313659) ** 0.5)
        + math.atan(t + rh)
        - math.atan(rh - 1.676331)
        + 0.00391838 * (rh ** 1.5) * math.atan(0.023101 * rh)
        - 4.686035
    )
    return tw

# --- Example Usage ---
temperature = 24.5  # 25°C
humidity = 46.0     # 50% RH

wet_bulb = calculate_wet_bulb(temperature, humidity)
print(f"Dry Bulb Temp: {temperature}°C")
print(f"Relative Humidity: {humidity}%")
print(f"Calculated Wet Bulb Temp: {wet_bulb:.2f}°C")
```
