# <font color='blue'> Parameter Estimation </font>

## <font color='orange'> Estimators and Bias in Statistics </font>

In statistical inference, an **estimator** is a rule or a function applied to a set of data to provide an estimate for an unknown population parameter (such as the mean $\mu$ or variance $\sigma^2$).

---

## <font color='orange'> 1. Estimator </font>
An estimator is a random variable because its value depends on the specific sample of data drawn from the population. 

* **Definition**: If $\theta$ is the true population parameter we wish to know, an estimator $\hat{\theta}$ is a function of the observed data $\mathbf{x}$.
* **Role**: It allows researchers to move from limited, observed samples to a formal inference about the underlying physical process or population.
* **Goal**: We seek estimators that are not only close to the true value but also possess desirable properties such as low variance and consistency.

---

## <font color='orange'> 2. Bias </font>
Bias is a measure of the systematic error of an estimator. It quantifies the difference between the expected value of the estimator and the true value of the parameter it is intended to estimate.

* **Mathematical Definition**: The bias $b$ of an estimator $\hat{\theta}$ is defined as:
    $$b = E[\hat{\theta}] - \theta$$
    Where $E[\hat{\theta}]$ is the expected value of the estimator over many repeated experiments.
* **Unbiased Estimator**: If the bias $b = 0$, the estimator is called "unbiased," meaning that on average, the estimator hits the true value of the parameter.
* **Biased Estimator**: If $b \neq 0$, the estimator is "biased," meaning it consistently overestimates or underestimates the true parameter.


## <font color='orange'> Mean Squared Error (MSE) </font>

The Mean Squared Error (MSE) of an estimator $\hat{\theta}$ for a true parameter $\theta$ is defined as the expected value of the squared difference between the estimate and the parameter. The derivation showing its relation to variance ($V$) and bias ($b$) is as follows:

The MSE is defined as:
$$\text{MSE}[\hat{\theta}] = E[(\hat{\theta} - \theta)^2]$$


To relate this to bias and variance, we add and subtract the expected value of the estimator, $E[\hat{\theta}]$, inside the square:
$$\text{MSE}[\hat{\theta}] = E[(\hat{\theta} - E[\hat{\theta}] + E[\hat{\theta}] - \theta)^2]$$

Expanding the expression $(a + b)^2 = a^2 + 2ab + b^2$, where $a = (\hat{\theta} - E[\hat{\theta}])$ and $b = (E[\hat{\theta}] - \theta)$:
$$\text{MSE}[\hat{\theta}] = E[(\hat{\theta} - E[\hat{\theta}])^2] + 2E[(\hat{\theta} - E[\hat{\theta}])(E[\hat{\theta}] - \theta)] + E[(E[\hat{\theta}] - \theta)^2]$$

* **First Term:** By definition, $E[(\hat{\theta} - E[\hat{\theta}])^2]$ is the variance of the estimator, $V[\hat{\theta}]$.
* **Second Term:** Since $(E[\hat{\theta}] - \theta)$ is a constant, it can be pulled out of the expectation: $2(E[\hat{\theta}] - \theta) \cdot E[(\hat{\theta} - E[\hat{\theta}])]$. Because $E[\hat{\theta} - E[\hat{\theta}]] = E[\hat{\theta}] - E[\hat{\theta}] = 0$, this entire term becomes zero.
* **Third Term:** Since $(E[\hat{\theta}] - \theta)$ is the bias $b$, the term $E[b^2]$ is simply $b^2$.

Combining the simplified terms, we arrive at the final relation:
$$\text{MSE}[\hat{\theta}] = V[\hat{\theta}] + b^2$$

---

### <font color='violet'> Glossary of Terms </font>

| Term | Definition |
| :--- | :--- |
| **Parameter ($\theta$)** | The true, fixed value describing the entire population (e.g., true signal cross-section). |
| **Estimator ($\hat{\theta}$)** | A function of the data used to estimate the unknown parameter. |
| **Expected Value ($E[\hat{\theta}]$)** | The long-term average value of the estimator if the experiment were repeated infinite times. |
| **Systematic Error** | A consistent, repeatable error that results in bias; often associated with detector calibration or model assumptions. |
| **Consistency** | The property that as the sample size $n \rightarrow \infty$, the estimator converges to the true value of the parameter. |

# <font color='blue'> Estimator for the Population Mean </font>

In statistics, to estimate the unknown population mean ($\mu$) of a distribution based on a sample, we use the **Sample Mean** ($\bar{X}$), also known as the empirical mean.

Given a random sample of $n$ independent and identically distributed (i.i.d.) random variables $X_1, X_2, \dots, X_n$, each with mean $E[X_i] = \mu$ and variance $Var(X_i) = \sigma^2$, the estimator is defined as:

$$\bar{X} = \frac{1}{n} \sum_{i=1}^{n} X_i$$

---

## <font color='orange'> Properties and Derivations </font>

To evaluate the quality of this estimator, we analyze its bias and variance.

### <font color='orange'> 1. Unbiasedness </font>
An estimator is unbiased if its expected value equals the true population parameter.

**Derivation:**
$$E[\bar{X}] = E\left[ \frac{1}{n} \sum_{i=1}^{n} X_i \right]$$

By the linearity of expectation:
$$E[\bar{X}] = \frac{1}{n} \sum_{i=1}^{n} E[X_i]$$

Since $E[X_i] = \mu$ for all $i$:
$$E[\bar{X}] = \frac{1}{n} \sum_{i=1}^{n} \mu = \frac{1}{n} (n\mu) = \mu$$

**Conclusion:** The sample mean is an **unbiased estimator** of the population mean.

---

### <font color='orange'> 2. Variance </font>
The variance of the estimator describes its precision (how much the estimate fluctuates across different samples).

**Derivation:**
$$Var(\bar{X}) = Var\left( \frac{1}{n} \sum_{i=1}^{n} X_i \right)$$

Using the property $Var(aX) = a^2 Var(X)$ and noting that for independent variables, the variance of the sum is the sum of the variances:
$$Var(\bar{X}) = \frac{1}{n^2} \sum_{i=1}^{n} Var(X_i)$$

Since $Var(X_i) = \sigma^2$:
$$Var(\bar{X}) = \frac{1}{n^2} \sum_{i=1}^{n} \sigma^2 = \frac{1}{n^2} (n\sigma^2) = \frac{\sigma^2}{n}$$

**Conclusion:** The variance of the sample mean decreases as the sample size $n$ increases, confirming the consistency of the estimator.

---

### <font color='orange'> 3. Efficiency (Minimum Variance) </font>
According to the **Gauss-Markov Theorem** (for linear estimators) and the **Cramér-Rao Bound**, the sample mean is the Best Linear Unbiased Estimator (BLUE) for the mean of a normal distribution. It achieves the minimum possible variance among unbiased estimators, making it a highly efficient choice.

# <font color='blue'> Method of Maximum Likelihood (MLE) </font>

Maximum Likelihood Estimation (MLE) is a statistical method used to estimate the parameters of a probability distribution by maximizing a **likelihood function**. The goal is to find the parameter values that make the observed data most probable.

---

## <font color='orange'> The Concept </font>

Suppose we have observed data $X = \{x_1, x_2, \dots, x_n\}$ generated from a distribution with an unknown parameter $\theta$. 

1.  **Likelihood Function $L(\theta)$**: This represents the probability (or probability density) of observing the data given a specific value of $\theta$.
    $$L(\theta) = f(x_1, x_2, \dots, x_n | \theta) = \prod_{i=1}^{n} f(x_i | \theta)$$
2.  **Objective**: Find the parameter $\hat{\theta}_{MLE}$ that maximizes $L(\theta)$:
    $$\hat{\theta}_{MLE} = \arg \max_{\theta} L(\theta)$$



---

## <font color='orange'> The Log-Likelihood Method </font>

Maximizing the product of many probabilities is computationally difficult and prone to numerical underflow. We typically maximize the **Log-Likelihood** ($\ell(\theta)$) instead, as the natural logarithm is a monotonically increasing function.

$$\ell(\theta) = \ln(L(\theta)) = \sum_{i=1}^{n} \ln(f(x_i | \theta))$$

### <font color='orange'> Derivation Steps </font>

To find the maximum, we take the derivative with respect to $\theta$ (the score function) and set it to zero:

1.  **Calculate the log-likelihood:** Take the log of the joint probability density function.
2.  **Take the derivative:** Find $\frac{d}{d\theta} \ell(\theta)$.
3.  **Solve for $\theta$:** Set the derivative to zero and solve for $\hat{\theta}$:
    $$\frac{d}{d\theta} \ell(\theta) = 0$$
4.  **Verify:** Ensure it is a maximum by checking that the second derivative is negative:
    $$\frac{d^2}{d\theta^2} \ell(\theta) < 0$$

---

## <font color='orange'> Example: Normal Distribution ($\mu$ estimation) </font>

For a sample from $N(\mu, \sigma^2)$, assuming $\sigma^2$ is known:

1.  **Likelihood:**
    $$L(\mu) = \prod_{i=1}^{n} \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x_i - \mu)^2}{2\sigma^2} \right)$$
2.  **Log-Likelihood:**
    $$\ell(\mu) = -\frac{n}{2}\ln(2\pi\sigma^2) - \sum_{i=1}^{n} \frac{(x_i - \mu)^2}{2\sigma^2}$$
3.  **Derivative:**
    $$\frac{d}{d\mu} \ell(\mu) = \frac{1}{\sigma^2} \sum_{i=1}^{n} (x_i - \mu) = 0$$
4.  **Solution:**
    $$\sum x_i - n\mu = 0 \implies \hat{\mu} = \frac{1}{n} \sum x_i = \bar{x}$$

The Maximum Likelihood Estimator for the mean of a normal distribution is simply the sample mean.

---

# <font color='blue'> Variance of Maximum Likelihood Estimators (MLE) </font>

The variance of an MLE $\hat{\theta}$ as $n \to \infty$ is determined by the **Cramér-Rao Lower Bound (CRLB)** and the **Fisher Information**. Using Taylor's theorem, we can show that the MLE is asymptotically efficient, meaning its variance approaches this lower bound.

---

## <font color='orange'> The Cramér-Rao Lower Bound (CRLB) </font>

For any unbiased estimator $\hat{\theta}$, the variance is bounded by the reciprocal of the Fisher Information $I(\theta)$:
$$Var(\hat{\theta}) \ge \frac{1}{I(\theta)}$$

Where $I(\theta)$ is the Fisher Information of the sample:
$$I(\theta) = -E\left[ \frac{d^2}{d\theta^2} \ell(\theta) \right]$$

---

# <font color='blue'> Graphical Method for Estimating Variance of MLEs </font>

The images provided describe a practical, graphical technique to approximate the variance (and standard deviation) of a Maximum Likelihood Estimator (MLE), $\hat{\theta}$. This method relies on a Taylor series expansion of the log-likelihood function around the estimate $\hat{\theta}$.

---
 
## <font color='orange'> 1. Taylor Series Expansion </font>
To understand the shape of the log-likelihood function $\log L(\theta)$ near its peak, we expand it around the MLE $\hat{\theta}$:

$$\log L(\theta) \approx \log L(\hat{\theta}) + \left[ \frac{\partial \log L}{\partial \theta} \right]_{\theta=\hat{\theta}} (\theta - \hat{\theta}) + \frac{1}{2!} \left[ \frac{\partial^2 \log L}{\partial \theta^2} \right]_{\theta=\hat{\theta}} (\theta - \hat{\theta})^2$$

### <font color='orange'> Key Simplifications: </font>
* **Stationarity:** Since $\hat{\theta}$ is the maximum, the first derivative is zero: $\left[ \frac{\partial \log L}{\partial \theta} \right]_{\theta=\hat{\theta}} = 0$. This eliminates the middle term.
* **Curvature:** The term $\log L(\hat{\theta})$ is the maximum value, denoted as $\log L_{max}$.
* **Fisher Information:** The second derivative is related to the variance: $\left[ \frac{\partial^2 \log L}{\partial \theta^2} \right]_{\theta=\hat{\theta}} \approx -\frac{1}{\sigma_{\hat{\theta}}^2}$.

---

## <font color='orange'> 2. The Quadratic Approximation </font>
Substituting these simplifications into the expansion yields a quadratic approximation of the log-likelihood function:

$$\log L(\theta) \approx \log L_{max} - \frac{(\theta - \hat{\theta})^2}{2\sigma_{\hat{\theta}}^2}$$

This formula shows that near the peak, the log-likelihood function behaves like an inverted parabola.

---

## <font color='orange'> 3. The Graphical Interpretation </font>
Equation (6.25) provides the "rule of thumb" for finding the standard deviation $\sigma_{\hat{\theta}}$ directly from a plot of the log-likelihood:

$$\log L(\hat{\theta} \pm \sigma_{\hat{\theta}}) = \log L_{max} - \frac{1}{2}$$

### <font color='orange'> How to use this method: </font>
1.  **Plot** the log-likelihood function $\log L(\theta)$ versus $\theta$.
2.  **Identify** the maximum point, which occurs at $\hat{\theta}$ with a value of $\log L_{max}$.
3.  **Find** the two points on the curve where the log-likelihood has dropped by exactly **0.5** units from its maximum value (i.e., at height $\log L_{max} - 0.5$).
4.  **Measure** the distance from these points to the central estimate $\hat{\theta}$. This horizontal distance provides an estimate of the standard deviation $\sigma_{\hat{\theta}}$ of your MLE.


This graphical approach is widely used in statistical software (such as generating likelihood profile intervals) to quickly visualize the uncertainty (precision) of parameter estimates.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. Generate a synthetic dataset
# We'll generate data from a Normal Distribution: X ~ N(mu=5, sigma=2)
np.random.seed(42)
n_samples = 1000
true_mu = 5.0
true_sigma = 2.0
data = np.random.normal(true_mu, true_sigma, n_samples)

# 2. MLE Estimation
# For a Normal Distribution, the MLE for mean is the sample mean
# The MLE for variance is the biased sample variance (1/n)
mle_mu = np.mean(data)
mle_var = np.sum((data - mle_mu)**2) / n_samples
mle_std = np.sqrt(mle_var)

# 3. Variance of the Estimator
# The variance of the sample mean (MLE) is sigma^2 / n
var_of_mle_mu = mle_var / n_samples
std_error_of_mu = np.sqrt(var_of_mle_mu)

print(f"MLE Estimate of Mean: {mle_mu:.4f}")
print(f"MLE Estimate of Variance: {mle_var:.4f}")
print(f"Estimated Standard Error of the Mean: {std_error_of_mu:.4f}")

# 4. Visualization: Log-Likelihood Profile
# This helps practice the graphical method discussed previously
mu_range = np.linspace(4.5, 5.5, 100)
log_likelihoods = []

for mu in mu_range:
    # Log-likelihood for Normal distribution: -n/2*ln(2*pi*sigma^2) - sum((x-mu)^2)/(2*sigma^2)
    ll = - (n_samples / 2) * np.log(2 * np.pi * mle_var) - np.sum((data - mu)**2) / (2 * mle_var)
    log_likelihoods.append(ll)

# 
plt.figure(figsize=(8, 5))
plt.plot(mu_range, log_likelihoods)
plt.axhline(y=max(log_likelihoods) - 0.5, color='r', linestyle='--', label='LogL_max - 0.5')
plt.title(r"Log-Likelihood Profile for $\mu$")
plt.xlabel(r"$\mu$")
plt.ylabel("Log-Likelihood")
plt.legend()
plt.show()
```

# <Method of Least Squares ($\chi^2$)

Since we have been discussing Maximum Likelihood Estimation (MLE) and variance, the Chi-Squared distribution is essential. It frequently appears when evaluating the "goodness-of-fit" of an MLE model or when constructing confidence intervals for variance.

---

## 1. Definition and Relation to Normal Distribution
If $Z_1, Z_2, \dots, Z_k$ are independent standard normal random variables $N(0, 1)$, then the sum of their squares follows a Chi-Squared distribution with $k$ degrees of freedom:
$$\chi^2 = \sum_{i=1}^{k} Z_i^2$$



---

## 2. Relation to MLE and Variance
In MLE, the Chi-Squared distribution is critical for two main reasons:

### Likelihood Ratio Tests (Wilks' Theorem)
For a large sample size $n$, the test statistic $-2 \ln \Lambda$ (where $\Lambda$ is the likelihood ratio) converges to a $\chi^2$ distribution:
$$-2 \ln \left( \frac{L(\theta_0)}{L(\hat{\theta})} \right) \xrightarrow{d} \chi^2_p$$
Where $p$ is the number of parameters being tested. This allows us to perform hypothesis testing on our MLE estimates.

### Estimation of Variance ($\sigma^2$)
If we have a sample from a normal distribution with unknown variance $\sigma^2$, the sample variance $S^2$ is related to the $\chi^2$ distribution:
$$\frac{(n-1)S^2}{\sigma^2} \sim \chi^2_{n-1}$$
This relationship is what we use to construct **confidence intervals for the population variance**.

---

## 3. Visualizing the Chi-Squared "Drop"
Recall the graphical method for variance we discussed earlier. The $\log L_{max} - 0.5$ rule is actually a specific case of the $\chi^2$ test. 

Specifically, the region where:
$$2 \left( \log L(\hat{\theta}) - \log L(\theta) \right) \le \chi^2_{1, \alpha}$$
defines the $(1-\alpha)$ confidence interval for $\theta$. For a 68% confidence interval ($\alpha=0.32$), the value of $\chi^2_{1, 0.32}$ is approximately **1.0**. 

Dividing this by 2 gives the **0.5** drop we looked at previously in the log-likelihood plot:
$$\log L(\hat{\theta}) - \log L(\theta) = \frac{1}{2} \chi^2_{1, \alpha} \approx 0.5$$
