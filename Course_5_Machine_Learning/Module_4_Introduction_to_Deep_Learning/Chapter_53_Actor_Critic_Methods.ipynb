# <font color='blue'> Chapter 53: Actor–Critic Methods </font>

Value-based methods estimate value functions and derive policies from them.

Policy Gradient methods directly optimize the policy.

Each approach has advantages and disadvantages.

Actor–Critic methods combine the strengths of both.

Instead of relying on a single model,

Actor–Critic algorithms use two cooperating components:

- an **Actor**, which selects actions,
- a **Critic**, which evaluates those actions.

The Critic provides feedback that helps the Actor improve its policy,

leading to faster and more stable learning than pure Policy Gradient methods.

Modern Reinforcement Learning algorithms,

including PPO, A2C, A3C, DDPG, TD3, and Soft Actor-Critic,

are all based on this fundamental architecture.

---

# <font color='orange'> 1. Motivation </font>

Suppose a student is learning to play the piano.

```
Student

↓

Play Music

↓

Teacher

↓

Feedback

↓

Improve
```

The student decides

what notes to play.

The teacher evaluates the performance.

Together,

the student gradually improves.

Actor–Critic methods follow exactly the same principle.

---

# <font color='orange'> 2. The Two Components </font>

The Actor chooses actions.

The Critic evaluates those actions.

```
State

↓

Actor

↓

Action

↓

Environment

↓

Reward

↓

Critic

↓

Feedback

↓

Actor Update
```

The Actor learns

**what to do**.

The Critic learns

**how good the current behaviour is**.

---

# <font color='orange'> 3. The Actor </font>

The Actor is a parameterized policy,

usually represented by a neural network.

It outputs

$$
\boxed{
\pi_\theta(a|s),
}
$$

the probability of selecting action

$$
a
$$

when the current state is

$$
s.
$$

The Actor is responsible for exploration and decision making.

---

# <font color='orange'> 4. The Critic </font>

The Critic estimates

how good the current state

or state-action pair is.

Depending on the algorithm,

it estimates either

$$
\boxed{
V(s)
}
$$

or

$$
\boxed{
Q(s,a).
}
$$

Unlike the Actor,

the Critic does not choose actions.

Its purpose is to provide learning signals that improve the policy.

---

# <font color='orange'> 5. Why Does the Critic Help? </font>

REINFORCE updates the policy using

the complete return

$$
G_t.
$$

Unfortunately,

returns can vary considerably,

leading to noisy gradient estimates.

The Critic provides a learned estimate,

which greatly reduces this variance.

Training therefore becomes

more stable

and

more sample efficient.

---

# <font color='orange'> 6. Advantage Function </font>

Rather than asking

"How good is this action?"

we ask

"Was this action better than expected?"

This quantity is called the

**Advantage Function**.

It is defined as

$$
\boxed{
A(s,a)
=
Q(s,a)
-
V(s).
}
$$

Interpretation

- Positive advantage

↓

better than expected.

- Negative advantage

↓

worse than expected.

The policy should increase the probability of actions with positive advantage,

and decrease the probability of actions with negative advantage.

---

# <font color='orange'> 7. Policy Gradient with Advantage </font>

The Policy Gradient Theorem becomes

$$
\boxed{
\nabla_\theta J(\theta)
=
\mathbb E
\left[
\nabla_\theta
\log
\pi_\theta(a|s)
A(s,a)
\right].
}
$$

Replacing

$$
Q(s,a)
$$

with

the advantage

reduces variance while preserving the correct expected gradient.

---

# <font color='orange'> 8. Temporal Difference Critic </font>

The Critic is often trained using

Temporal Difference Learning.

The TD error is

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
V(s_{t+1})
-
V(s_t).
}
$$

This error serves two purposes.

It updates

the Critic's value estimates,

and also provides a learning signal for the Actor.

---

# <font color='orange'> 9. Actor Update </font>

The Actor performs

gradient ascent

using the Critic's feedback.

Conceptually,

the update becomes

$$
\boxed{
\theta
\leftarrow
\theta
+
\alpha
\nabla_\theta
\log
\pi_\theta(a|s)
\,
\delta_t.
}
$$

If

the TD error is positive,

the probability of selecting that action increases.

If

the TD error is negative,

the probability decreases.

---

# <font color='orange'> 10. Critic Update </font>

The Critic minimizes the TD error,

typically by updating its value estimate as

$$
\boxed{
V(s_t)
\leftarrow
V(s_t)
+
\beta
\delta_t,
}
$$

where

$$
\beta
$$

is the Critic's learning rate.

Thus,

the Actor maximizes reward,

while the Critic improves its predictions.

---

# <font color='orange'> 11. Advantage Actor–Critic (A2C) </font>

One popular Actor–Critic algorithm is

**Advantage Actor–Critic (A2C)**.

Instead of using

$$
Q(s,a),
$$

it directly estimates

the advantage,

leading to

lower variance

and

faster convergence.

A2C is widely used because of its simplicity and efficiency.

---

# <font color='orange'> 12. Asynchronous Advantage Actor–Critic (A3C) </font>

A3C extends A2C

by running

multiple agents

simultaneously.

```
Worker 1

↓

Worker 2

↓

Worker 3

↓

Global Network
```

Parallel exploration produces more diverse experience,

leading to faster and more robust learning.

---

# <font color='orange'> 13. Advantages </font>

Actor–Critic methods

- reduce gradient variance,
- support continuous action spaces,
- learn online,
- are computationally efficient,
- scale well with deep neural networks.

These properties make them the basis of many state-of-the-art RL algorithms.

---

# <font color='orange'> 14. Limitations </font>

Actor–Critic methods also present challenges.

They require

- simultaneous training of two networks,
- careful tuning of learning rates,
- stable interaction between Actor and Critic.

Poor Critic estimates can lead the Actor to learn suboptimal policies.

---

# <font color='orange'> 15. Applications </font>

Actor–Critic methods are widely used in

- robotics,
- autonomous driving,
- drone control,
- industrial automation,
- finance,
- recommendation systems,
- game playing,
- language model alignment.

Many modern reinforcement learning systems rely on this architecture.

# <font color='orange'> 17. Relationship to Modern Algorithms </font>

Many state-of-the-art algorithms are extensions of the Actor–Critic framework.

| Algorithm | Main Idea |
|:---|:---|
| A2C | Synchronous Advantage Actor–Critic |
| A3C | Parallel asynchronous Actor–Critic |
| PPO | Constrains policy updates for greater stability |
| DDPG | Actor–Critic for continuous deterministic policies |
| TD3 | Improves DDPG with reduced value overestimation |
| SAC | Maximum-entropy Actor–Critic for robust exploration |

Although these algorithms differ in implementation,

they all rely on the cooperation between an Actor and a Critic.

---

# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> The Critic chooses the actions.

**False.**

The Actor selects actions. The Critic evaluates the consequences of those actions.

---

### Misconception 2

> The Critic must always estimate the action-value function \(Q(s,a)\).

**False.**

Some Actor–Critic algorithms estimate the state-value function \(V(s)\), while others estimate \(Q(s,a)\). The choice depends on the specific algorithm.

---

### Misconception 3

> Actor–Critic methods eliminate all variance in policy gradients.

**False.**

The Critic significantly reduces variance, but stochastic environments and sampling still introduce variability into the gradient estimates.

---

# <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Actor | Learns the policy \(\pi_\theta(a|s)\) |
| Critic | Estimates value functions to evaluate the Actor's decisions |
| Actor–Critic | Combines policy optimization with value estimation |
| Advantage Function | Measures whether an action is better or worse than expected |
| TD Error | Learning signal used to update both Actor and Critic |
| A2C | Synchronous Advantage Actor–Critic |
| A3C | Parallel Actor–Critic with asynchronous workers |
| Variance Reduction | Main advantage over pure Policy Gradient methods |

> **Key Insight:** Actor–Critic methods combine the strengths of policy-based and value-based reinforcement learning. The Actor learns how to act by optimizing a parameterized policy, while the Critic evaluates the quality of those actions using value estimates. By replacing noisy Monte Carlo returns with lower-variance value estimates and temporal difference errors, Actor–Critic methods achieve faster, more stable learning and form the foundation of many modern reinforcement learning algorithms, including PPO, DDPG, TD3, and Soft Actor–Critic.
