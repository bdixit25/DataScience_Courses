# <font color='blue'> Chapter 35: BERT – Bidirectional Encoder Representations from Transformers </font>

The original Transformer architecture consists of both an **encoder** and a **decoder**. While this design is ideal for sequence-to-sequence tasks such as machine translation, many Natural Language Processing (NLP) problems require only an understanding of text rather than text generation.

To address this need, researchers at Google introduced **BERT (Bidirectional Encoder Representations from Transformers)** in 2018.

Unlike earlier language models, which processed text from left to right (or right to left), BERT processes text **bidirectionally**, allowing every word to attend simultaneously to both its preceding and following context.

This ability to build rich contextual representations transformed NLP and established new state-of-the-art performance across numerous benchmark tasks.

---



# <font color='orange'> 1. Motivation </font>

Consider the sentence

> **The bank approved the loan.**

Now consider

> **The fisherman sat on the bank.**

The meaning of the word

```
bank
```

depends entirely on its surrounding words.

A language model that only reads from left to right cannot fully exploit information appearing later in the sentence.

BERT overcomes this limitation by considering both the left and right contexts simultaneously.

---



# <font color='orange'> 2. Why Was BERT Needed? </font>

Earlier language models,

including many RNN- and LSTM-based systems,

processed text sequentially.

```
Word₁

↓

Word₂

↓

Word₃

↓

Word₄
```

Each prediction depended primarily on previous words.

Consequently,

future context was unavailable.

BERT instead processes

the entire sentence at once.

Every word can attend to every other word using self-attention.

---



# <font color='orange'> 3. Encoder-Only Architecture </font>

BERT uses only the **Transformer Encoder**.

```
Input Tokens

↓

Embeddings

↓

Positional Encoding

↓

Encoder

↓

Encoder

↓

Encoder

↓

Contextual Word Representations
```

Unlike the original Transformer,

there is

- no decoder,
- no masked autoregressive generation.

The output is a contextual representation for every input token.

---



# <font color='orange'> 4. Bidirectional Context </font>

Suppose we process

```
The cat sat on the mat.
```

When computing the representation of

```
sat
```

BERT simultaneously considers

```
The

cat

sat

on

the

mat
```

rather than only the preceding words.

This bidirectional processing produces richer contextual representations.

---



# <font color='orange'> 5. Input Representation </font>

The input to BERT is the sum of three embeddings.

$$
\boxed{
\text{Input Embedding}
=
\text{Token Embedding}
+
\text{Position Embedding}
+
\text{Segment Embedding}.
}
$$

These components encode

- the identity of the token,
- its position in the sequence,
- the sentence to which it belongs.

# <font color='orange'> 7. Masked Language Modelling (MLM) </font>

The primary pretraining task for BERT is

**Masked Language Modelling**.

Randomly,

approximately

15%

of the input tokens are selected.

Some are replaced by

```
[MASK]
```

The model predicts the original token.

Example

Input

```
The cat

[MASK]

on the mat.
```

Target

```
sat
```

Unlike traditional language modelling,

the prediction can use both left and right context.

---



# <font color='orange'> 8. Next Sentence Prediction (NSP) </font>

The original BERT paper also introduced

**Next Sentence Prediction**.

The model receives two sentences.

Example

```
Sentence A

↓

Sentence B
```

The task is to determine

whether Sentence B actually follows Sentence A

in the original text.

This encourages the model to learn relationships between sentences.

Later variants,

such as RoBERTa,

removed NSP after finding that it was not essential for strong performance.

---



# <font color='orange'> 9. Fine-Tuning BERT </font>

After pretraining,

BERT can be adapted to many tasks by adding a small output layer.

Examples include

- sentiment analysis,
- spam detection,
- question answering,
- named entity recognition,
- text classification.

Only a relatively small number of task-specific parameters need to be learned.

---



# <font color='orange'> 10. BERT Workflow </font>

```
Large Text Corpus

↓

Pretraining

↓

Masked Language Modelling

↓

Pretrained BERT

↓

Fine-Tuning

↓

Specific NLP Task
```

This two-stage workflow became the standard paradigm for many NLP models.

---



# <font color='orange'> 11. Variants of BERT </font>

Several important variants have been developed.

| Model | Main Idea |
|:---|:---|
| BERT Base | Original encoder-only Transformer |
| BERT Large | Larger model with more layers and parameters |
| RoBERTa | Improved training procedure without NSP |
| DistilBERT | Smaller and faster distilled model |
| ALBERT | Parameter sharing for greater efficiency |
| ELECTRA | Replaces MLM with a discriminative pretraining task |

These models retain the encoder-only philosophy while introducing various architectural or training improvements.

---



# <font color='orange'> 12. Applications </font>

BERT has been successfully applied to

- sentiment analysis,
- document classification,
- spam filtering,
- question answering,
- information retrieval,
- search engines,
- biomedical NLP,
- legal document analysis.

Its contextual embeddings have become a standard feature representation in many NLP systems.

# <font color='red'> 13. Mathematical Foundations </font>

Suppose

the input embedding matrix is

$$
X.
$$

Each encoder layer computes

$$
H^{(l)}
=
\mathrm{Encoder}
\left(
H^{(l-1)}
\right),
$$

where

$$
H^{(0)}=X.
$$

After

$$
L
$$

encoder layers,

the final contextual representations are

$$
\boxed{
H^{(L)}
=
\mathrm{Encoder}^{L}(X).
}
$$

During Masked Language Modelling,

the probability of the masked token is

$$
\boxed{
P(w_i \mid \text{context})
=
\mathrm{Softmax}(Wh_i+b),
}
$$

where

- \(h_i\) is the final hidden representation of the masked position,
- \(W\) and \(b\) are the output layer parameters.

The training objective is to maximize the probability of the original masked tokens.

# <font color='orange'> 14. BERT vs GPT </font>

| BERT | GPT |
|:---|:---|
| Encoder-only | Decoder-only |
| Bidirectional | Left-to-right autoregressive |
| Understanding text | Generating text |
| Uses Masked Language Modelling | Uses Next Token Prediction |
| Excellent for classification | Excellent for generation |

These models share the Transformer foundation but are optimized for different objectives.

---



# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> BERT generates text like ChatGPT.

**False.**

BERT is designed primarily for understanding text rather than generating coherent long passages.

---

### Misconception 2

> Every word is masked during training.

**False.**

Only a small fraction of tokens (typically about 15%) are selected for masking during pretraining.

---

### Misconception 3

> BERT always uses Next Sentence Prediction.

**False.**

The original BERT used NSP, but several later variants, such as RoBERTa, removed it while achieving improved performance.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| BERT | Encoder-only Transformer model |
| Bidirectional Context | Every token attends to both left and right context |
| [CLS] Token | Sequence-level representation for classification |
| [SEP] Token | Separates multiple input sequences |
| [MASK] Token | Used during Masked Language Modelling |
| MLM | Predicts masked words using surrounding context |
| Fine-Tuning | Adapts pretrained BERT to downstream tasks |
| Contextual Embedding | Word representation that depends on surrounding words |

> **Key Insight:** BERT applies the Transformer encoder to learn bidirectional contextual representations of language. By predicting masked words using information from both preceding and following tokens, BERT develops a deep understanding of linguistic context. After large-scale pretraining, these learned representations can be efficiently fine-tuned for a wide range of natural language processing tasks, making BERT one of the most influential models in modern NLP.
