# <font color='blue'> Chapter 31: Query, Key, and Value (Q, K, V) — The Language of Attention </font>

In the previous chapter, we introduced the attention mechanism as a way of allowing a model to focus on the most relevant parts of an input sequence. We saw that attention assigns different importance weights to different elements of the input when making a prediction.

However, one important question remains:

> **How does the model decide which parts of the sequence are relevant?**

Modern attention mechanisms answer this question using three learned vectors for every input element:

- **Query (Q)**
- **Key (K)**
- **Value (V)**

These three vectors form the foundation of the Transformer architecture.

Although the equations are relatively simple, understanding the intuition behind Queries, Keys, and Values is essential for understanding self-attention and modern Large Language Models.

---

# <font color='orange'> 1. Motivation </font>

Suppose you are in a library looking for books about

```
Quantum Mechanics.
```

You do not examine every book in detail.

Instead,

you

- search using a keyword,
- compare it with book labels,
- retrieve the relevant books.

Attention works in exactly the same way.

Every token asks

> **Which other tokens are most relevant to me?**

---

# <font color='orange'> 2. The Library Analogy </font>

Imagine a library.

```
Your Question

↓

Query

↓

Compare With

↓

Book Labels

↓

Keys

↓

Retrieve

↓

Book Contents

↓

Values
```

The analogy is

| Library | Transformer |
|:---|:---|
| Search request | Query |
| Book label | Key |
| Book contents | Value |

The Query searches,

the Keys identify relevant information,

and the Values contain the information that is actually retrieved.

---

# <font color='orange'> 3. Every Token Creates Q, K and V </font>

Suppose we have the sentence

```
The

black

cat

sat
```

Each word is converted into an embedding vector.

Then,

every embedding generates

- one Query,
- one Key,
- one Value.

```
Embedding

↓

Query

↓

Key

↓

Value
```

Thus,

every token simultaneously

- searches,
- advertises itself,
- stores information.

---

# <font color='orange'> 4. Why Three Different Vectors? </font>

One might ask

> Why not simply use the embedding itself?

The reason is that

the model performs three different tasks.

The Query asks

> What information am I looking for?

The Key answers

> What information do I contain?

The Value contains

> What information should be passed to other tokens?

Learning separate projections allows the model to specialize these three roles independently.

---

# <font color='orange'> 5. Computing Queries, Keys and Values </font>

Suppose

the input embedding matrix is

$$
X.
$$

Three different weight matrices are learned.

Queries

$$
\boxed{
Q=XW_Q.
}
$$

Keys

$$
\boxed{
K=XW_K.
}
$$

Values

$$
\boxed{
V=XW_V.
}
$$

where

- $W_Q$,
- $W_K$,
- $W_V$

are trainable parameters.

Notice that

all three vectors originate from the same embedding,

but each undergoes a different learned transformation.

---

# <font color='orange'> 6. Comparing Queries and Keys </font>

The next step is determining

how well every Query matches every Key.

Suppose

```
Query

↓

Key
```

If they are similar,

the corresponding token is likely to be important.

Mathematically,

similarity is measured using a **dot product**.

Large dot product

→ highly relevant.

Small dot product

→ less relevant.

---

# <font color='orange'> 7. Similarity Matrix </font>

Instead of comparing one Query with one Key,

Transformers compare

every Query

with

every Key.

The result is

$$
\boxed{
QK^T.
}
$$

Suppose the sentence contains

four words.

The similarity matrix becomes

```
        K₁ K₂ K₃ K₄

Q₁

Q₂

Q₃

Q₄
```

Each entry measures

how strongly one word attends to another.

---

# <font color='orange'> 8. Why the Dot Product? </font>

Suppose

two vectors point in similar directions.

Their dot product is large.

If they point in different directions,

their dot product is small.

Thus,

the dot product naturally measures

how similar two learned representations are.

This makes it an effective similarity measure for attention.

---

# <font color='orange'> 9. Retrieving the Values </font>

After computing attention scores,

the model retrieves the Values.

High attention weight

↓

Large contribution.

Low attention weight

↓

Small contribution.

Notice that

attention is computed using

Queries

and

Keys,

but the information actually passed forward comes from the

Values.

---

# <font color='orange'> 10. Intuitive Example </font>

Consider

```
The animal

didn't cross

the street

because

it

was tired.
```

When processing

```
it
```

the Query generated by

```
it
```

compares itself with every Key.

The Key corresponding to

```
animal
```

is likely to produce the highest similarity.

Consequently,

the Value associated with

```
animal
```

contributes strongly to the new representation of

```
it.
```

---

# <font color='orange'> 11. Learning Q, K and V </font>

Initially,

the matrices

$$
W_Q,
W_K,
W_V
$$

are initialized randomly.

During training,

backpropagation adjusts them

so that

Queries learn

how to search,

Keys learn

how to describe information,

and Values learn

what information should be transmitted.

Thus,

the entire attention mechanism is learned automatically from data.

---

# <font color='red'> 12. Mathematical Foundations </font>

Suppose

the embedding matrix is

$$
X
\in
\mathbb{R}^{n\times d},
$$

where

- $n$ is the sequence length,
- $d$ is the embedding dimension.

Three learned linear projections produce

$$
\boxed{
Q=XW_Q,
}
$$

$$
\boxed{
K=XW_K,
}
$$

$$
\boxed{
V=XW_V.
}
$$

The similarity matrix is

$$
\boxed{
S=QK^T.
}
$$

The element

$$
S_{ij}
$$

represents

how strongly

token

$$
i
$$

attends to

token

$$
j.
$$

These similarity scores will later be normalized using Softmax to produce the attention weights.

---

# <font color='orange'> 13. Common Misconceptions </font>

### Misconception 1

> Queries, Keys, and Values are different input vectors.

**False.**

All three are computed from the same input embeddings using different learned linear transformations.

---

### Misconception 2

> The Query retrieves the information directly.

**False.**

The Query is used only to compute similarity with the Keys. The information that is passed forward comes from the Values.

---

### Misconception 3

> The Query, Key, and Value matrices are fixed.

**False.**

The projection matrices \(W_Q\), \(W_K\), and \(W_V\) are trainable parameters learned during training through backpropagation.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Query (Q) | Represents what a token is searching for |
| Key (K) | Represents how a token describes itself for matching |
| Value (V) | Represents the information carried by the token |
| Linear Projection | Transformation from embeddings to Q, K, and V using learned weight matrices |
| Similarity Matrix | Computed as \(QK^T\), measuring how strongly tokens attend to one another |
| Dot Product | Measures similarity between Queries and Keys |

> **Key Insight:** Every token in a sequence generates three learned representations: a Query, a Key, and a Value. The Query determines what information the token is seeking, the Key determines how other tokens can be matched, and the Value contains the information that will be shared. By comparing Queries with Keys and retrieving the corresponding Values, the attention mechanism enables each token to dynamically gather information from the most relevant parts of the sequence.
