# <font color='blue'> Chapter 4: The Artificial Neuron in Action </font>

In the previous chapters, we introduced the mathematical model of an artificial neuron and the sigmoid activation function. Although the equations appear simple, it is important to understand **how a neuron actually processes information**.

A neuron can be viewed as a tiny decision-making unit. It receives several pieces of information, evaluates their importance, combines them into a single numerical score, and finally decides how strongly it should respond.

This chapter follows the journey of data through a neuron, from the raw input values to the final output.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to determine whether a student is likely to pass an examination.

The available information includes

- Hours of Study
- Attendance
- Previous Test Score

Clearly, not every feature contributes equally.

For example,

- studying may be the most important,
- attendance moderately important,
- previous performance may also contribute.

Rather than assigning these importance values manually,

a neural network learns them automatically from data.

Each neuron performs this evaluation independently.

---

# <font color='orange'> 2. The Flow of Information </font>

Every neuron performs exactly the same sequence of operations.

```
Inputs

↓

Multiply by Weights

↓

Add Bias

↓

Weighted Sum

↓

Activation Function

↓

Output
```

This process is called the **forward pass** of the neuron.

---

# <font color='orange'> 3. Step 1 — Receiving the Inputs </font>

Suppose a neuron receives three inputs

$$
x_1,\;x_2,\;x_3.
$$

For example,

| Feature | Value |
|:---|---:|
| Hours Studied | 8 |
| Attendance (%) | 90 |
| Previous Score | 70 |

The neuron itself does not know what these quantities represent.

It simply receives numerical values.

---

# <font color='orange'> 4. Step 2 — Applying Weights </font>

Each input is multiplied by its corresponding weight.

Suppose

$$
w=
\begin{bmatrix}
0.8\\
0.4\\
0.6
\end{bmatrix}.
$$

The neuron computes

$$
0.8x_1,
\qquad
0.4x_2,
\qquad
0.6x_3.
$$

The weights determine how much attention the neuron pays to each feature.

Large weights imply

greater influence.

Small weights imply

less influence.

Negative weights suppress the contribution of the corresponding feature.

---

# <font color='orange'> 5. Step 3 — Computing the Weighted Sum </font>

The neuron combines all weighted inputs.

Mathematically,

$$
\boxed{
z
=
w_1x_1
+
w_2x_2
+
w_3x_3
+
b.
}
$$

For our example,

suppose

$$
b=-5.
$$

Then

$$
z
=
0.8(8)
+
0.4(90)
+
0.6(70)
-
5.
$$

Evaluating,

$$
z
=
6.4
+
36
+
42
-
5
=
79.4.
$$

The weighted sum represents the neuron's internal evidence before making a decision.

---

# <font color='orange'> 6. Step 4 — Applying the Activation Function </font>

The weighted sum is passed through an activation function.

Suppose we use the sigmoid activation.

The neuron outputs

$$
\boxed{
a
=
\frac{1}{1+e^{-79.4}}.
}
$$

Since

$$
79.4
$$

is very large,

the output is approximately

$$
a\approx1.
$$

The neuron therefore responds very strongly.

---

# <font color='orange'> 7. Another Example </font>

Suppose another student has

- Hours Studied = 1
- Attendance = 40
- Previous Score = 25

Then

$$
z
=
0.8(1)
+
0.4(40)
+
0.6(25)
-
5
=
26.8.
$$

Again,

the neuron produces

a high activation.

If we instead chose smaller weights,

or a much larger negative bias,

the activation would become significantly smaller.

This illustrates that **both the weights and the bias jointly determine the neuron's behaviour**.

---

# <font color='orange'> 8. Interpreting the Weights </font>

Suppose

$$
w=
\begin{bmatrix}
2\\
0.1\\
-3
\end{bmatrix}.
$$

This tells us

- Feature 1 is extremely important.
- Feature 2 contributes only weakly.
- Feature 3 decreases the activation.

Thus,

weights encode the relative importance of each feature.

Learning the weights is the primary objective of neural network training.

---

# <font color='orange'> 9. Interpreting the Bias </font>

The bias shifts the neuron's decision threshold.

Suppose

$$
z
=
w^Tx+b.
$$

If

$$
b
=
0,
$$

the neuron may activate too easily.

A negative bias requires stronger evidence before activation.

A positive bias makes the neuron activate more readily.

The bias therefore acts like an adjustable threshold.

---

# <font color='orange'> 10. Decision Making </font>

A neuron can be viewed as making a simple decision.

Suppose

$$
a=\sigma(z).
$$

If

$$
a>0.5,
$$

predict

Positive.

Otherwise,

predict

Negative.

Since

$$
\sigma(0)=0.5,
$$

this is equivalent to checking

$$
\boxed{
w^Tx+b>0.
}
$$

Thus,

every neuron divides the input space into two regions.

---

# <font color='orange'> 11. Geometric Interpretation </font>

The equation

$$
w^Tx+b=0
$$

defines a

- line in two dimensions,
- plane in three dimensions,
- hyperplane in higher dimensions.

Everything on one side of the hyperplane produces one decision.

Everything on the opposite side produces the other.

Thus,

a neuron behaves like a linear classifier.

---

# <font color='orange'> 12. What Does the Neuron Learn? </font>

During training,

the neuron repeatedly updates

- the weights,
- the bias,

to improve its predictions.

Initially,

the weights are chosen almost randomly.

After seeing many training examples,

they gradually become meaningful.

For example,

a neuron trained to recognise handwritten digits may eventually respond strongly to

- vertical lines,
- horizontal edges,
- curves,
- corners.

Thus,

neurons become specialised feature detectors.

---

# <font color='orange'> 13. From One Neuron to Many </font>

One neuron has limited capability.

It can only learn a single linear decision boundary.

Real-world problems require many neurons working together.

```
Inputs

↓

Neuron

↓

Neuron

↓

Neuron

↓

Prediction
```

Each neuron extracts a different aspect of the data.

Together,

they produce highly sophisticated models capable of learning extremely complex relationships.

---

# <font color='red'> 14. Mathematical Foundations </font>

The complete computation performed by a neuron can be summarised as

### Step 1

Compute the weighted sum

$$
z
=
\sum_{i=1}^{n}
w_ix_i+b.
$$

---

### Step 2

Apply the activation

$$
a=f(z).
$$

---

Combining both,

$$
\boxed{
a
=
f
\left(
\sum_{i=1}^{n}
w_ix_i+b
\right).
}
$$

This compact equation describes every artificial neuron used in modern deep learning.

---

# <font color='orange'> 15. Computational Complexity </font>

Suppose a neuron receives

$n$

inputs.

The neuron performs

- $n$ multiplications,
- $n-1$ additions,
- one bias addition,
- one activation evaluation.

Therefore,

the computational complexity is

$$
\boxed{
\mathcal{O}(n).
}
$$

Although a single neuron is computationally inexpensive,

modern neural networks may contain billions of neurons operating simultaneously.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Input | Numerical information supplied to the neuron |
| Weight | Determines the importance of an input |
| Bias | Controls the activation threshold |
| Weighted Sum | Internal evidence computed by the neuron |
| Activation Function | Produces the final neuron output |
| Decision Boundary | Hyperplane separating different predictions |
| Training | Learns the weights and bias from data |

> **Key Insight:** An artificial neuron is best viewed as a simple decision-making unit. It combines its inputs according to their learned importance, shifts the result using a bias, and transforms the weighted sum through a nonlinear activation function. Individually, a neuron can only represent simple linear decisions, but networks of many interacting neurons become capable of learning remarkably complex patterns from data.
