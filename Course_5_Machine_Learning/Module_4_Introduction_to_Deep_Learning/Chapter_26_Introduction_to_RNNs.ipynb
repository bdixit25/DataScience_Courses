# <font color='blue'> Chapter 26: Introduction to Recurrent Neural Networks (RNNs) </font>

In the previous chapters, we studied feedforward neural networks and Convolutional Neural Networks (CNNs). These architectures assume that every input sample is independent of every other sample.

However, many real-world problems involve **sequential data**, where the order of observations carries important information.

For example,

- the meaning of a sentence depends on the order of its words,
- the next value of a stock price depends on previous prices,
- spoken language unfolds over time,
- weather observations form time series.

Feedforward neural networks process each input independently and therefore cannot naturally model these temporal dependencies.

Recurrent Neural Networks (RNNs) overcome this limitation by introducing **memory**, allowing information from previous time steps to influence future predictions.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict the next word in the sentence

> "The cat sat on the ____"

The correct prediction depends on the previous words.

If we ignored their order,

```
The

Cat

Sat

On
```

would simply become

```
Cat

On

The

Sat
```

which loses the grammatical structure.

A successful model must therefore remember previous information.

---

# <font color='orange'> 2. Why Feedforward Networks Fail for Sequences </font>

Consider a standard neural network.

```
Input

↓

Hidden Layers

↓

Output
```

Each prediction depends only on the current input.

Previous inputs are forgotten immediately.

For example,

```
Temperature Today

↓

Predict Tomorrow
```

The model has no memory of

- yesterday,
- last week,
- seasonal patterns.

Many sequence problems require this historical context.

---

# <font color='orange'> 3. The Key Idea Behind RNNs </font>

Recurrent Neural Networks introduce a **hidden state**, which acts as the network's memory.

At each time step,

the network receives

- the current input,
- the previous hidden state.

```
Previous Hidden State

↓

Current Input

↓

RNN Cell

↓

New Hidden State

↓

Output
```

The hidden state carries information forward through time.

---

# <font color='orange'> 4. Sequential Processing </font>

Unlike CNNs,

RNNs process one element of the sequence at a time.

Example

```
Word 1

↓

Word 2

↓

Word 3

↓

Word 4
```

Each prediction depends on everything processed previously.

---

# <font color='orange'> 5. Hidden State </font>

The hidden state is the memory of the network.

Suppose

$$
\mathbf{x}_t
$$

is the input at time

$$
t.
$$

The hidden state is

$$
\mathbf{h}_t.
$$

The RNN updates its memory according to

$$
\boxed{
\mathbf{h}_t
=
f
(
W_{xh}\mathbf{x}_t
+
W_{hh}\mathbf{h}_{t-1}
+
\mathbf{b}
).
}
$$

Notice that

the new hidden state depends on

- the current input,
- the previous hidden state.

This is the defining feature of an RNN.

---

# <font color='orange'> 6. The Recurrent Connection </font>

The hidden state feeds back into the network.

```
Input

↓

Hidden State

↺

↓

Output
```

The feedback loop gives the network memory.

Unlike feedforward networks,

information can persist across multiple time steps.

---

# <font color='orange'> 7. Unrolling an RNN </font>

Although an RNN contains a loop,

during training it is easier to imagine it as a repeated sequence of identical cells.

```
x₁ → RNN → h₁

↓

x₂ → RNN → h₂

↓

x₃ → RNN → h₃

↓

x₄ → RNN → h₄
```

Every cell shares exactly the same weights.

This representation is called the **unrolled RNN**.

---

# <font color='orange'> 8. Weight Sharing Through Time </font>

Unlike a feedforward network,

an RNN does **not** learn new weights at every time step.

Instead,

the same parameters are reused.

```
Time 1

↓

W

↓

Time 2

↓

W

↓

Time 3

↓

W
```

This dramatically reduces the number of trainable parameters.

---

# <font color='orange'> 9. Many Types of Sequence Problems </font>

RNNs can solve several different tasks.

### One-to-One

```
Image

↓

Class
```

(Standard neural networks)

---

### One-to-Many

```
Image

↓

Sentence
```

(Image caption generation)

---

### Many-to-One

```
Sentence

↓

Sentiment
```

(Sentiment analysis)

---

### Many-to-Many

```
Sentence

↓

Translated Sentence
```

(Machine translation)

---

### Many-to-Many (Aligned)

```
Speech

↓

Text
```

where each time step produces an output.

---

# <font color='orange'> 10. Applications of RNNs </font>

RNNs have been applied to

- language modelling,
- speech recognition,
- handwriting recognition,
- machine translation,
- weather prediction,
- financial forecasting,
- anomaly detection,
- biological sequence analysis.

Although many NLP applications now use Transformers,

RNNs remain important for understanding sequence modelling.

---

# <font color='orange'> 11. Advantages of RNNs </font>

RNNs

- process variable-length sequences,
- share parameters across time,
- model temporal dependencies,
- naturally handle sequential data.

These properties distinguish them from feedforward architectures.

---

# <font color='orange'> 12. Limitations of Basic RNNs </font>

Standard RNNs struggle to learn long-range dependencies.

As sequences become longer,

important information may gradually disappear during training.

This phenomenon is known as the **vanishing gradient problem**.

Conversely,

gradients may also grow uncontrollably,

leading to the **exploding gradient problem**.

These issues motivated the development of

- Long Short-Term Memory (LSTM) networks,
- Gated Recurrent Units (GRUs).

---

# <font color='red'> 13. Mathematical Foundations </font>

The defining equation of a simple RNN is

$$
\boxed{
\mathbf{h}_t
=
f
(
W_{xh}\mathbf{x}_t
+
W_{hh}\mathbf{h}_{t-1}
+
\mathbf{b}
).
}
$$

The output is then

$$
\boxed{
\mathbf{y}_t
=
g
(
W_{hy}\mathbf{h}_t
+
\mathbf{c}
).
}
$$

where

- $\mathbf{x}_t$ is the input,
- $\mathbf{h}_t$ is the hidden state,
- $\mathbf{y}_t$ is the output,
- $W_{xh}$ maps the input to the hidden state,
- $W_{hh}$ maps the previous hidden state to the new hidden state,
- $W_{hy}$ maps the hidden state to the output.

Unlike feedforward networks,

the hidden state introduces a recursive dependence across time,

allowing information to persist through the sequence.

---

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> An RNN stores the entire history of the sequence perfectly.

**False.**

The hidden state is a compressed representation of previous information. It cannot perfectly retain arbitrarily long sequences, which is one reason LSTMs and GRUs were developed.

---

### Misconception 2

> RNNs use different weights at every time step.

**False.**

The same weight matrices are shared across all time steps. This parameter sharing enables RNNs to process sequences of varying lengths.

---

### Misconception 3

> RNNs have been completely replaced by Transformers.

**False.**

Transformers dominate many natural language processing tasks, but RNNs remain useful for certain sequence problems, embedded systems, streaming data, and as an essential conceptual foundation for understanding modern sequence models.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Recurrent Neural Network | Neural network designed for sequential data |
| Hidden State | Memory carried from one time step to the next |
| Recurrent Connection | Feedback loop enabling temporal memory |
| Unrolled RNN | Representation of the network across time |
| Weight Sharing | Same parameters reused at every time step |
| Many-to-One | Sequence classification |
| Many-to-Many | Sequence transformation |
| Vanishing Gradient | Gradients shrink during long sequences |
| Exploding Gradient | Gradients grow excessively during long sequences |

> **Key Insight:** Recurrent Neural Networks extend feedforward neural networks by introducing a hidden state that carries information across time. This enables them to model sequential data such as text, speech, and time series. Their defining characteristic is the recursive update of the hidden state using shared parameters, allowing the network to learn temporal dependencies. However, standard RNNs struggle with long-range dependencies, motivating more advanced architectures such as LSTMs and GRUs.
