# <font color='blue'> Chapter 67: Offline Reinforcement Learning </font>

Most Reinforcement Learning algorithms learn by interacting continuously with the environment.

After each action,

the environment provides

- the next state,
- a reward,

allowing the policy to improve over time.

This process is called **online Reinforcement Learning**.

However,

many practical problems do not permit continuous exploration.

Instead,

only a fixed dataset of past experiences is available.

Offline Reinforcement Learning aims to learn an effective policy **without collecting any additional data**.

The agent must extract as much information as possible from previously recorded interactions.

Offline RL has become an important research area because it enables Reinforcement Learning in safety-critical domains such as healthcare, autonomous driving, finance, and robotics.

---



# <font color='orange'> 1. Motivation </font>

Suppose a hospital has collected

```
10 million

patient treatments
```

Each record contains

- patient condition,
- treatment,
- outcome.

Clearly,

we cannot allow a Reinforcement Learning agent

to try random treatments

on future patients.

Instead,

the agent must learn

entirely from

the historical data.

This is an Offline RL problem.

---



# <font color='orange'> 2. Online vs Offline Reinforcement Learning </font>

### Online RL

```
Environment

↓

Action

↓

Reward

↓

Learn

↓

Repeat
```

The dataset grows continuously.

---

### Offline RL

```
Fixed Dataset

↓

Learn Policy

↓

Finished
```

No additional interactions are allowed.

---



# <font color='orange'> 3. Offline Dataset </font>

The dataset consists of transitions

$$
\boxed{
\mathcal D
=
\{
(s,a,r,s')
\}.
}
$$

These transitions were collected

before training began.

The dataset remains fixed

throughout learning.

---



# <font color='orange'> 4. Behaviour Policy </font>

The dataset was generated

by another policy,

called the

**Behaviour Policy**,

denoted

$$
\boxed{
\pi_\beta.
}
$$

This policy may

- be a human expert,
- be another RL algorithm,
- contain mistakes,
- be suboptimal.

The Offline RL algorithm must learn

a new policy

from these demonstrations.

---



# <font color='orange'> 5. Target Policy </font>

The goal is

to learn

a better policy,

written

$$
\boxed{
\pi.
}
$$

Notice

the learned policy

does not necessarily equal

the behaviour policy.

Instead,

it attempts to improve upon it

without collecting new data.

---



# <font color='orange'> 6. The Distribution Shift Problem </font>

One of the greatest challenges

in Offline RL

is

**distribution shift**.

Suppose

the dataset contains

only

```
Action A

↓

Action B
```

If the learned policy suddenly selects

```
Action C
```

the algorithm has

no experience

showing

what happens next.

The value function must therefore

guess,

which may produce large errors.

---



# <font color='orange'> 7. Extrapolation Error </font>

When evaluating actions

that never appeared

in the dataset,

the value function

must extrapolate.

These predictions are often unreliable.

Small estimation errors

can accumulate,

causing the learned policy

to become unrealistically optimistic.

This is known as

**extrapolation error**.

---



# <font color='orange'> 8. Conservative Reinforcement Learning </font>

A natural solution is

to avoid trusting

unfamiliar actions.

Many Offline RL algorithms

therefore become

**conservative**.

Rather than assigning

large Q-values

to unseen actions,

they intentionally underestimate them.

This encourages

the learned policy

to remain close

to actions supported by the dataset.

---



# <font color='orange'> 9. Conservative Q-Learning (CQL) </font>

One influential Offline RL algorithm is

**Conservative Q-Learning (CQL).**

Instead of only minimizing

Bellman error,

CQL also penalizes

large Q-values

for actions that are poorly represented

in the dataset.

Conceptually,

CQL prefers

```
Known Actions

↓

Higher Confidence

Unknown Actions

↓

Lower Confidence
```

This improves policy reliability.

---



# <font color='orange'> 10. Batch-Constrained Q-Learning (BCQ) </font>

Another important algorithm is

**Batch-Constrained Q-Learning (BCQ).**

Instead of considering

every possible action,

BCQ generates actions

similar to those

already present

in the dataset.

The learned policy therefore remains

close to

the observed behaviour policy.

---



# <font color='orange'> 11. Why Offline RL is Difficult </font>

Unlike Online RL,

the agent cannot

collect additional data

to correct mistakes.

Every error

must be resolved

using

only

the available dataset.

Consequently,

generalization becomes

one of the central challenges.

---



# <font color='orange'> 12. Advantages </font>

Offline Reinforcement Learning

- requires no further environment interaction,
- is safer for high-risk applications,
- allows learning from historical data,
- reduces deployment cost,
- enables reproducible experiments.

---



# <font color='orange'> 13. Applications </font>

Offline RL has been applied to

- healthcare,
- autonomous driving,
- robotics,
- finance,
- recommendation systems,
- industrial process optimization.

It is particularly valuable whenever online exploration is unsafe or prohibitively expensive.

---



# <font color='orange'> 14. Limitations </font>

Offline RL also has important limitations.

Performance depends strongly on

- dataset quality,
- dataset coverage,
- diversity of observed behaviours.

If important states or actions are missing,

the learned policy may perform poorly.

Unlike Online RL,

the agent cannot gather new experience to fill these gaps.

---

# <font color='red'> 15. Mathematical Foundations </font>

Offline dataset

$$
\boxed{
\mathcal D
=
\{
(s_i,a_i,r_i,s_i')
\}_{i=1}^{N}.
}
$$

Behaviour policy

$$
\boxed{
\pi_\beta(a|s).
}
$$

Target policy

$$
\boxed{
\pi(a|s).
}
$$

Standard Bellman target

$$
\boxed{
y
=
r
+
\gamma
\max_a
Q(s',a).
}
$$

In Offline RL,

this target may overestimate

actions

never observed

in

$$
\mathcal D.
$$

Conservative methods therefore modify the objective

to penalize

high Q-values

for unsupported actions.

Conceptually,

the objective becomes

$$
\boxed{
L
=
L_{\mathrm{Bellman}}
+
\lambda
L_{\mathrm{Conservative}},
}
$$

where

- \(L_{\mathrm{Bellman}}\) fits the value function to the dataset,
- \(L_{\mathrm{Conservative}}\) discourages overestimation of unseen actions,
- \(\lambda\) controls the strength of the conservative penalty.

Although different Offline RL algorithms define

$$
L_{\mathrm{Conservative}}
$$

in different ways,

they all aim to reduce extrapolation error.

---



# <font color='orange'> 16. Online RL vs Offline RL </font>

| Online RL | Offline RL |
|:---|:---|
| Learns while interacting with the environment | Learns from a fixed dataset |
| Exploration is allowed | No exploration during training |
| Dataset grows continuously | Dataset is fixed |
| Can correct mistakes through new experience | Cannot collect additional experience |
| Examples: DQN, PPO, SAC | Examples: CQL, BCQ, IQL |

---



# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> Offline RL is simply supervised learning.

**False.**

Although both use fixed datasets, Offline RL still optimizes long-term cumulative reward and must account for future consequences of actions. Supervised learning predicts labels, whereas Offline RL learns sequential decision-making policies.

---

### Misconception 2

> A larger dataset always guarantees better performance.

**False.**

The quality and coverage of the dataset are often more important than its size. A very large dataset that contains only a narrow range of behaviours may not support learning a strong policy.

---

### Misconception 3

> Offline RL completely eliminates uncertainty.

**False.**

The agent may still encounter states or actions that are poorly represented in the dataset. Conservative algorithms reduce the resulting errors but cannot remove uncertainty entirely.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Offline Reinforcement Learning | Learns entirely from a fixed dataset |
| Behaviour Policy | Policy that generated the dataset |
| Target Policy | Policy learned from the dataset |
| Distribution Shift | Learned policy selects actions unlike those in the dataset |
| Extrapolation Error | Incorrect value estimates for unseen actions |
| Conservative Q-Learning (CQL) | Penalizes optimistic Q-values for unsupported actions |
| Batch-Constrained Q-Learning (BCQ) | Restricts actions to those similar to the dataset |
| Sample Efficiency | Achieves learning without further interaction |

> **Key Insight:** Offline Reinforcement Learning enables agents to learn effective decision-making policies from previously collected experience without interacting with the environment during training. Because the learned policy may consider actions not represented in the dataset, Offline RL faces unique challenges such as distribution shift and extrapolation error. Modern algorithms such as Conservative Q-Learning and Batch-Constrained Q-Learning address these issues by encouraging the learned policy to remain consistent with the available data, making Offline RL particularly valuable for safety-critical applications where online exploration is impractical.
