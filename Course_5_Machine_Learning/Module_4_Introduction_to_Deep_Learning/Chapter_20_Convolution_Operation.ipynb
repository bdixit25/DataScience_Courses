# <font color='blue'> Chapter 20: The Convolution Operation </font>

The defining operation of a Convolutional Neural Network (CNN) is the **convolution operation**. Every convolutional layer learns a set of small filters that move across the input image, extracting increasingly meaningful features.

Although the term *convolution* originates from mathematics and signal processing, the operation used in modern deep learning is slightly different. Most deep learning libraries actually perform **cross-correlation**, a closely related operation whose filter is not reversed.

This chapter develops the convolution operation from first principles, explains the relationship between convolution and cross-correlation, derives the output size equations, and demonstrates the calculations through detailed numerical examples.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to detect vertical edges in an image.

Rather than examining the entire image simultaneously,

we move a small filter across the image,

computing a weighted sum at every position.

```
Input Image

□□□□□□□□

□□□□□□□

□□□□□□□□

↓

3 × 3 Filter

■■■

■■■

■■■

↓

Feature Map
```

The resulting feature map indicates where the desired pattern appears.

---

# <font color='orange'> 2. What is a Filter (Kernel)? </font>

A **filter** (also called a **kernel**) is a small matrix of learnable weights.

For example,

$$
K=
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}
$$

is a filter that responds strongly to vertical edges.

Typical filter sizes include

- $3\times3$
- $5\times5$
- $7\times7$

Modern CNNs usually prefer

$$
3\times3
$$

filters because they are computationally efficient while capturing local spatial information.

---

# <font color='orange'> 3. The Sliding Window Idea </font>

Instead of processing the whole image,

the filter examines one small region at a time.

```
Image

■■■□□□□

■■■□□□□

■■■□□□□

□□□□□□□□

□□□□□□□□
```

The filter then moves one step,

repeating the same calculation.

```
□■■■□□□

□■■■□□□

□■■■□□□

□□□□□□□□

□□□□□□□□
```

This continues until the filter has visited every valid position.

---

# <font color='orange'> 4. Cross-Correlation </font>

Suppose

the input image is

$$
I
$$

and the filter is

$$
K.
$$

Modern CNNs compute

$$
\boxed{
S(i,j)
=
\sum_m
\sum_n
I(i+m,j+n)
K(m,n).
}
$$

Notice that

the filter is **not reversed**.

This operation is called **cross-correlation**.

---

# <font color='orange'> 5. Mathematical Convolution </font>

The mathematical definition of convolution is

$$
\boxed{
S(i,j)
=
\sum_m
\sum_n
I(i-m,j-n)
K(m,n).
}
$$

or equivalently,

the filter is rotated by

$$
180^\circ
$$

before multiplication.

```
Original Filter

1 2

3 4

↓

Rotated

4 3

2 1
```

Deep learning libraries omit this rotation.

Since the filter weights are learned,

the network simply learns the flipped version if needed.

Therefore,

cross-correlation and convolution are effectively equivalent in CNNs.

---

# <font color='orange'> 6. Numerical Example </font>

Suppose

$$
I=
\begin{bmatrix}
1&2&0\\
3&1&1\\
0&2&2
\end{bmatrix},
$$

and

$$
K=
\begin{bmatrix}
1&0\\
-1&1
\end{bmatrix}.
$$

Compute the first output element.

Multiply corresponding entries,

$$
1(1)
+
2(0)
+
3(-1)
+
1(1).
$$

Therefore,

$$
1+0-3+1=-1.
$$

Thus,

the first feature-map value is

$$
\boxed{-1.}
$$

The same procedure is repeated at every valid filter position.

---

# <font color='orange'> 7. Feature Maps </font>

Each filter produces one feature map.

```
Input Image

↓

Filter 1

↓

Feature Map 1

↓

Filter 2

↓

Feature Map 2

↓

Filter 3

↓

Feature Map 3
```

If a convolutional layer contains

64 filters,

it produces

64 feature maps.

Each filter specializes in detecting a different visual pattern.

---

# <font color='orange'> 8. Stride </font>

The **stride** specifies how far the filter moves after each calculation.

Stride = 1

```
■■■

 □■■■

  □■■■
```

Stride = 2

```
■■■

  □□■■■
```

Larger strides

- reduce computational cost,
- decrease the output size,
- may discard fine spatial detail.

---

# <font color='orange'> 9. Padding </font>

Without padding,

the output feature map becomes progressively smaller.

To preserve information near the image borders,

extra pixels are added around the image.

Usually,

the padding consists of zeros.

```
Original

□□□□

□□□□

↓

Padding

000000

0□□□□0

0□□□□0

000000
```

Common choices are

- Valid Padding (no padding),
- Same Padding (output size approximately equals input size).

---

# <font color='orange'> 10. Output Size Formula </font>

Suppose

- input size

$$
N,
$$

- filter size

$$
F,
$$

- padding

$$
P,
$$

- stride

$$
S.
$$

Then the output size is

$$
\boxed{
\frac{N-F+2P}{S}+1.
}
$$

This quantity must be an integer.

---

## Example

Input

$$
32\times32,
$$

filter

$$
3\times3,
$$

padding

$$
1,
$$

stride

$$
1.
$$

Then

$$
\frac{32-3+2}{1}+1
=
32.
$$

Hence,

the output remains

$$
32\times32.
$$

---

# <font color='orange'> 11. Multiple Input Channels </font>

Colour images contain

three channels,

namely

- Red,
- Green,
- Blue.

Thus,

a filter must span all input channels.

For an RGB image,

a

$$
3\times3
$$

filter actually has dimensions

$$
3\times3\times3.
$$

The outputs from each channel are summed to produce one feature map.

---

# <font color='orange'> 12. Number of Parameters </font>

Suppose

- filter size

$$
3\times3,
$$

- input channels

$$
3,
$$

- filters

$$
64.
$$

Each filter contains

$$
3\times3\times3=27
$$

weights,

plus

one bias.

Thus,

parameters per filter

$$
28.
$$

Total parameters

$$
64\times28
=
1792.
$$

Compare this with a fully connected layer,

which might require millions of parameters.

---

# <font color='orange'> 13. Why Small Filters Work </font>

One might ask

> Why not simply use a very large filter?

The answer is that stacking multiple small filters

- requires fewer parameters,
- introduces more nonlinearities,
- increases representational power,
- reduces computational cost.

For example,

two consecutive

$$
3\times3
$$

convolutions have an effective receptive field similar to a

$$
5\times5
$$

filter,

while using fewer parameters.

---

# <font color='red'> 14. Mathematical Foundations </font>

The convolution operation is fundamentally a weighted sum.

For a filter of size

$$
F\times F,
$$

the output at location

$$
(i,j)
$$

is

$$
\boxed{
S(i,j)
=
\sum_{m=0}^{F-1}
\sum_{n=0}^{F-1}
I(i+m,j+n)\,
K(m,n).
}
$$

This expression is simply the dot product between

- the local image patch,
- the filter weights.

Thus,

a convolutional layer can be viewed as repeatedly applying the same linear transformation to different regions of the image.

The nonlinearity is introduced afterward by the activation function (typically ReLU).

---

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> A filter detects only one object.

**False.**

A filter detects a specific **feature**, such as an edge, corner, or texture. Higher-level objects emerge from combinations of many learned features across multiple layers.

---

### Misconception 2

> CNNs use mathematical convolution exactly as defined in signal processing.

**False.**

Most deep learning libraries implement **cross-correlation**, which omits the 180° filter rotation. Because the filter weights are learned, this distinction does not affect the expressive power of the network.

---

### Misconception 3

> Larger filters are always better.

**False.**

Modern CNN architectures often favour repeated **3 × 3** convolutions because they achieve a large effective receptive field with fewer parameters and more nonlinear transformations.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Filter (Kernel) | Small matrix of learnable weights |
| Cross-Correlation | Sliding weighted-sum operation used in CNNs |
| Mathematical Convolution | Similar operation with a flipped filter |
| Feature Map | Output produced by applying one filter |
| Stride | Number of pixels the filter moves between computations |
| Padding | Additional border pixels used to control output size |
| Output Size | $\frac{N-F+2P}{S}+1$ |
| Parameter Sharing | Same filter applied across the entire image |

> **Key Insight:** The convolution operation enables CNNs to detect local patterns efficiently by applying the same small filter across an entire image. This parameter-sharing strategy dramatically reduces the number of trainable parameters compared with fully connected networks while preserving the spatial relationships between neighbouring pixels. Combined with nonlinear activation functions, convolutions allow CNNs to learn increasingly sophisticated hierarchical representations of visual data.
