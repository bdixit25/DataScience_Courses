# <font color='blue'> Chapter 5: From a Single Neuron to a Neural Network </font>

In the previous chapters, we studied the operation of a single artificial neuron. While a single neuron can perform simple linear classification, it is far too limited to solve most real-world problems.

Modern Deep Learning achieves its remarkable performance by connecting many neurons together into large networks. Each neuron performs a simple computation, but collectively these neurons learn highly sophisticated representations of data.

This chapter explains how individual neurons are organized into layers and how information flows through an entire neural network.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to recognise handwritten digits.

A single neuron can only learn one linear decision boundary.

Unfortunately,

the boundary separating digits such as

0,

3,

5,

and

8

is far more complicated than a straight line.

Instead of relying on one neuron,

we connect many neurons together.

Some neurons learn

- edges,
- corners,
- curves,

while later neurons combine these simple features into increasingly complex patterns.

This hierarchical learning is the key idea behind neural networks.

---



# <font color='orange'> 2. Why One Neuron is Not Enough </font>

Recall that one neuron computes

$$
a=f(\mathbf{w}^T\mathbf{x}+b).
$$

Its decision boundary is

$$
\mathbf{w}^T\mathbf{x}+b=0.
$$

This is always

- a line in two dimensions,
- a plane in three dimensions,
- a hyperplane in higher dimensions.

Many real-world datasets cannot be separated by a single hyperplane.

Therefore,

we require multiple neurons.

---



# <font color='orange'> 3. Building a Layer of Neurons </font>

Suppose instead of one neuron,

we have four neurons.

```
x₁ ─────► ○

x₂ ─────► ○

x₃ ─────► ○

x₄ ─────► ○
```

Each neuron

- receives exactly the same inputs,
- has different weights,
- learns different features.

For neuron

$j$

the output is

$$
a_j=f(\mathbf{w}_j^T\mathbf{x}+b_j).
$$

Thus,

each neuron learns a different representation of the same input.

---



# <font color='orange'> 4. Hidden Layers </font>

The outputs from one layer become the inputs to the next.

```
Input Layer

     │

     ▼

Hidden Layer

     │

     ▼

Hidden Layer

     │

     ▼

Output Layer
```

Each hidden layer transforms the data into a more useful representation.

Early layers often learn simple features,

while deeper layers learn increasingly abstract concepts.

---



# <font color='orange'> 5. Types of Layers </font>

## Input Layer

The input layer simply receives the data.

No learning occurs here.

---

## Hidden Layers

These perform almost all of the computation.

Their purpose is to learn useful intermediate features.

---

## Output Layer

Produces the final prediction.

Its activation depends upon the task.

| Problem | Output Activation |
|:---|:---|
| Regression | Linear |
| Binary Classification | Sigmoid |
| Multi-class Classification | Softmax |

---



# <font color='orange'> 6. Fully Connected Networks </font>

The simplest neural network is called a

**Fully Connected Network**

or

**Dense Neural Network**.

Every neuron is connected to every neuron in the next layer.

```
Input

○   ○   ○

|\ /|\ /|

| X | X |

|/ \|/ \|

○   ○   ○

     |

     ▼

Output
```

These are also called

- Dense Networks
- Feedforward Networks
- Multi-Layer Perceptrons (MLPs)

depending on the context.

---



# <font color='orange'> 7. Information Flow </font>

Information always flows

from left to right.

```
Inputs

↓

Layer 1

↓

Layer 2

↓

Layer 3

↓

Prediction
```

This process is called

**Forward Propagation**.

No learning occurs during forward propagation.

The network simply computes predictions.

Learning happens later,

during

**Backpropagation**.

---



# <font color='orange'> 8. Matrix Representation </font>

Suppose the input vector is

$$
\mathbf{x}.
$$

A hidden layer computes

$$
\boxed{
\mathbf{z}
=
W\mathbf{x}
+
\mathbf{b}.
}
$$

Applying an activation function,

$$
\boxed{
\mathbf{a}
=
f(\mathbf{z}).
}
$$

Combining both,

$$
\boxed{
\mathbf{a}
=
f(W\mathbf{x}+\mathbf{b}).
}
$$

Notice that

the equation for an entire layer is almost identical to that of a single neuron.

The only difference is that

weights become a matrix,

and

biases become a vector.

---



# <font color='orange'> 9. Multiple Layers </font>

Suppose our network has three hidden layers.

The computations become

$$
\mathbf{a}^{(1)}
=
f
\left(
W^{(1)}
\mathbf{x}
+
\mathbf{b}^{(1)}
\right),
$$

$$
\mathbf{a}^{(2)}
=
f
\left(
W^{(2)}
\mathbf{a}^{(1)}
+
\mathbf{b}^{(2)}
\right),
$$

$$
\mathbf{a}^{(3)}
=
f
\left(
W^{(3)}
\mathbf{a}^{(2)}
+
\mathbf{b}^{(3)}
\right).
$$

Finally,

the output layer computes

$$
\hat y
=
f
\left(
W^{(4)}
\mathbf{a}^{(3)}
+
\mathbf{b}^{(4)}
\right).
$$

Every modern neural network follows this same principle.

---



# <font color='orange'> 10. Why Multiple Layers Work </font>

Each layer learns increasingly sophisticated features.

For image recognition,

a simplified hierarchy might be

```
Pixels

↓

Edges

↓

Corners

↓

Textures

↓

Shapes

↓

Objects

↓

Prediction
```

This hierarchical representation is one of the defining characteristics of deep learning.

---



# <font color='orange'> 11. Number of Parameters </font>

Suppose

- 5 input features,
- 10 hidden neurons,
- 3 output neurons.

Input → Hidden

Weights

$$
5\times10=50
$$

Biases

$$
10
$$

Hidden → Output

Weights

$$
10\times3=30
$$

Biases

$$
3
$$

Total parameters

$$
\boxed{
50+10+30+3=93.
}
$$

These parameters are learned automatically during training.

---



# <font color='orange'> 12. Why Deep Networks are Powerful </font>

A shallow network can represent many functions,

but it may require an enormous number of neurons.

Deep networks often achieve the same representation

using far fewer neurons

by learning hierarchical features.

This is one reason why deep architectures have become dominant in modern AI.

---

# <font color='red'> 13. Mathematical Foundations </font>

Suppose

layer

$l$

contains

$n_l$

neurons.

The layer computation is

$$
\boxed{
\mathbf{a}^{(l)}
=
f
\left(
W^{(l)}
\mathbf{a}^{(l-1)}
+
\mathbf{b}^{(l)}
\right).
}
$$

This single equation describes

every hidden layer

in virtually every feedforward neural network.

It will become one of the most frequently used equations throughout this course.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Layer | Collection of neurons performing the same computation |
| Hidden Layer | Learns intermediate representations |
| Dense Layer | Every neuron connected to every neuron in the next layer |
| Forward Propagation | Information flows through the network |
| Weight Matrix | Stores connections between layers |
| Bias Vector | Stores one bias for each neuron |
| Deep Network | Neural network containing multiple hidden layers |

> **Key Insight:** A neural network is simply a collection of interconnected artificial neurons arranged into layers. Each neuron performs a weighted sum followed by a nonlinear activation, while each layer transforms the data into progressively more informative representations. By stacking many such layers, neural networks can learn complex hierarchical features that are impossible for a single neuron or linear model to represent.
