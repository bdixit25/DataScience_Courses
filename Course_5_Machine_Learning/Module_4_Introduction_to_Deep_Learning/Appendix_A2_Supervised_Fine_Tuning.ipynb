# <font color='blue'> Appendix A2: Supervised Fine-Tuning (SFT) </font>

After pretraining,

a language model possesses extensive knowledge about language,

facts,

mathematics,

programming,

and reasoning.

However,

it has never been explicitly taught

how to interact with users.

Its only objective during pretraining was

to predict the next token.

Consequently,

a pretrained model may generate text that is grammatically correct,

yet fail to answer questions helpfully,

follow instructions,

or maintain natural conversations.

Supervised Fine-Tuning (SFT) addresses this problem.

Instead of learning from raw internet text,

the model is trained using carefully prepared examples of

- instructions,
- questions,
- ideal responses.

This stage transforms a general language model into an instruction-following assistant.

---



# <font color='orange'> 1. Motivation </font>

Suppose we prompt a pretrained language model with

```
Translate into French:

Good morning
```

Because the model has never been instructed to follow commands,

it might produce

```
Translate into French:

Good morning

Many people learn French because...
```

Instead,

we would like

```
Bonjour
```

The model must therefore learn

that its job is to respond to instructions,

not merely continue text.

---



# <font color='orange'> 2. What is Supervised Fine-Tuning? </font>

Supervised Fine-Tuning (SFT)

is a second stage of training

performed after pretraining.

Instead of using

raw internet documents,

SFT uses

instruction–response pairs.

Example

| Instruction | Desired Response |
|:---|:---|
| Explain Newton's First Law. | A clear explanation suitable for the user. |
| Translate "Good Morning" into French. | Bonjour |
| Write Python code to compute factorials. | Correct Python implementation. |

Each example teaches

the model

how humans expect an assistant to respond.

---



# <font color='orange'> 3. Instruction Datasets </font>

An SFT dataset consists of many examples of the form

$$
\boxed{
(\text{Instruction},\ \text{Response}).
}
$$

For example,

```
Instruction

↓

Summarize this article.

↓

Ideal Summary
```

or

```
Instruction

↓

Explain Quantum Mechanics.

↓

Helpful Explanation
```

Thousands or millions of such examples are collected.

---



# <font color='orange'> 4. Teacher Demonstrations </font>

The responses in the dataset are usually written by

- human annotators,
- subject experts,
- professional trainers,
- or high-quality AI systems reviewed by humans.

These responses are often called

**demonstrations**,

because they demonstrate

the desired behaviour.

The model learns by imitating these examples.

---



# <font color='orange'> 5. Why It Is Called "Supervised" </font>

The training process is called

**supervised learning**

because

the correct output

is already known.

For every instruction,

the dataset contains

the desired response.

The model therefore learns

to reproduce

the demonstrated behaviour.

---



# <font color='orange'> 6. Training Objective </font>

Suppose the instruction is

```
Explain photosynthesis.
```

The desired response is

```
Photosynthesis is the process by which plants convert sunlight into chemical energy...
```

The model predicts

one token at a time,

exactly as during pretraining.

The only difference is

that the target text

is now

a high-quality assistant response,

rather than arbitrary internet text.

---



# <font color='orange'> 7. Cross-Entropy Loss </font>

Interestingly,

the loss function

does **not change**.

The model still minimizes

$$
\boxed{
L
=
-
\sum_t
\log
P(x_t|x_{<t}).
}
$$

The difference lies

entirely in

the training data.

Instead of predicting internet text,

the model predicts

assistant demonstrations.

---



# <font color='orange'> 8. What the Model Learns </font>

After SFT,

the model begins to learn

- instruction following,
- question answering,
- summarization,
- translation,
- coding assistance,
- conversational behaviour,
- structured responses.

It also learns

the expected style

of an AI assistant.

---



# <font color='orange'> 9. The SFT Training Pipeline </font>

The overall workflow is

```
Pretrained Language Model

↓

Instruction Dataset

↓

Cross-Entropy Loss

↓

Backpropagation

↓

Gradient Descent

↓

Instruction-Following Model
```

Notice

that no Reinforcement Learning

has yet been used.

---



# <font color='orange'> 10. Why SFT is Not Enough </font>

Although SFT greatly improves the model,

it still has limitations.

Suppose

two responses are both reasonable.

Response A

```
Clear

Helpful

Concise
```

Response B

```
Correct

Very Long

Hard to Read
```

The SFT dataset contains

only one demonstration.

The model therefore cannot learn

which response

humans actually prefer.

This motivates

the next stage

of training.

---



# <font color='orange'> 11. Preference Learning </font>

Humans often disagree

about

which response is best.

Instead of providing

one correct answer,

they can compare

two responses

and indicate

which they prefer.

These preference comparisons become the foundation of

Reinforcement Learning from Human Feedback (RLHF).

---



# <font color='orange'> 12. Applications </font>

Supervised Fine-Tuning is used to improve

- instruction following,
- conversational AI,
- coding assistants,
- translation systems,
- summarization models,
- domain-specific assistants.

Almost every modern LLM undergoes some form of supervised fine-tuning before alignment.

---



# <font color='orange'> 13. Limitations </font>

SFT depends heavily on

- dataset quality,
- demonstration quality,
- diversity of instructions.

It also learns only

to imitate demonstrations.

It cannot directly optimize

human preferences,

helpfulness,

or harmlessness.

These goals require

additional alignment methods,

such as RLHF or DPO.

---

# <font color='red'> 14. Mathematical Foundations </font>

Suppose

an instruction is

$$
x,
$$

and

the desired response is

$$
y=(y_1,y_2,\ldots,y_T).
$$

The objective is

to maximize

$$
\boxed{
P(y|x).
}
$$

Using the chain rule,

$$
\boxed{
P(y|x)
=
\prod_{t=1}^{T}
P(y_t|x,y_{<t}).
}
$$

Training minimizes

the negative log-likelihood,

or equivalently,

the cross-entropy loss,

$$
\boxed{
L
=
-
\sum_{t=1}^{T}
\log
P(y_t|x,y_{<t}).
}
$$

Notice

that this is mathematically identical

to the objective used during pretraining.

Only the training data

have changed.

---



# <font color='orange'> 15. Pretraining vs Supervised Fine-Tuning </font>

| Pretraining | Supervised Fine-Tuning |
|:---|:---|
| Internet text | Instruction–response pairs |
| Learns language | Learns instruction following |
| Self-supervised | Supervised |
| Next-token prediction | Next-token prediction |
| General knowledge | Assistant behaviour |

Both stages optimize

cross-entropy,

but they learn different behaviours because they use different datasets.

---



# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> Supervised Fine-Tuning changes the architecture of the Transformer.

**False.**

The Transformer architecture remains exactly the same. SFT updates only the model parameters using new training data.

---

### Misconception 2

> SFT uses Reinforcement Learning.

**False.**

SFT is purely supervised learning. The model learns by imitating demonstration responses and does not receive rewards or interact with an environment.

---

### Misconception 3

> After SFT, the model is perfectly aligned with human preferences.

**False.**

SFT teaches the model to imitate demonstrations, but it does not explicitly optimize for qualities such as helpfulness, harmlessness, or user preference. Those goals motivate later stages such as RLHF and DPO.

---

# <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Supervised Fine-Tuning (SFT) | Trains a pretrained model to follow instructions |
| Instruction Dataset | Collection of instruction–response examples |
| Demonstration | High-quality response written by humans or expert systems |
| Supervised Learning | Learns from labelled input–output pairs |
| Cross-Entropy Loss | Same objective as pretraining |
| Instruction Following | Main capability learned during SFT |
| Limitation | Learns demonstrations, not human preferences |

> **Key Insight:** Supervised Fine-Tuning transforms a pretrained language model into an instruction-following assistant by training it on carefully curated instruction–response pairs. Although the learning objective remains the same as during pretraining—minimizing cross-entropy loss—the change in the training data fundamentally changes the model's behaviour. Instead of predicting arbitrary internet text, the model learns to generate helpful, structured responses to user instructions. However, SFT alone cannot capture nuanced human preferences, motivating the next stage: Reinforcement Learning from Human Feedback (RLHF).
