# <font color='blue'> Chapter 47: Monte Carlo Methods </font>

Dynamic Programming solves Reinforcement Learning problems when the complete environment is known.

However,

in most real-world applications,

the transition probabilities and reward function are unknown.

Instead of solving the Bellman equations directly,

the agent must learn from experience.

Monte Carlo (MC) Methods estimate value functions by averaging the returns obtained from many sampled episodes.

Rather than requiring a model of the environment,

they learn directly from interaction,

making them one of the first **model-free Reinforcement Learning** algorithms.

---



# <font color='orange'> 1. Motivation </font>

Suppose you wish to estimate

the average time required to travel from home to work.

Rather than building a complete mathematical traffic model,

you simply record your travel time every day.

```
Monday

↓

42 min

Tuesday

↓

39 min

Wednesday

↓

45 min

...
```

After many observations,

the average provides a good estimate.

Monte Carlo methods apply exactly the same principle.

Instead of computing expectations analytically,

they estimate them using repeated samples.

---



# <font color='orange'> 2. Why Monte Carlo Methods? </font>

Dynamic Programming requires

$$
P(s'|s,a)
$$

and

$$
R(s,a).
$$

Monte Carlo methods require only

```
Experience

↓

Episodes

↓

Returns

↓

Average
```

No knowledge of the environment is needed.

---



# <font color='orange'> 3. Learning from Complete Episodes </font>

Monte Carlo methods wait until an episode finishes.

Example

```
Start

↓

Move

↓

Move

↓

Goal

↓

Episode Ends
```

Only after reaching the terminal state

does the algorithm compute the return.

This differs from Temporal Difference learning,

which updates values before the episode ends.

---



# <font color='orange'> 4. Return </font>

For an episode,

the return from time

$$
t
$$

is

$$
\boxed{
G_t
=
r_{t+1}
+
\gamma r_{t+2}
+
\gamma^2r_{t+3}
+\cdots.
}
$$

This return serves as the target value for learning.

---



# <font color='orange'> 5. Monte Carlo Prediction </font>

Suppose

state

$$
s
$$

appears in many episodes.

Each occurrence produces

a return

$$
G.
$$

The value estimate becomes

the average

of all observed returns.

$$
\boxed{
V(s)
=
\frac{1}{N}
\sum_{i=1}^{N}
G_i.
}
$$

As the number of episodes increases,

this estimate approaches the true expected value.

---



# <font color='orange'> 6. First-Visit Monte Carlo </font>

Suppose a state appears multiple times

within one episode.

Example

```
A

↓

B

↓

A

↓

Goal
```

First-Visit Monte Carlo

uses only

the first occurrence.

Repeated visits during the same episode are ignored.

This often simplifies the analysis.

---



# <font color='orange'> 7. Every-Visit Monte Carlo </font>

Every-Visit Monte Carlo

uses

every occurrence

of a state.

```
A

↓

B

↓

A

↓

Goal
```

Both visits to

```
A
```

contribute to the value estimate.

Both First-Visit and Every-Visit methods converge to the correct value under suitable assumptions, though they may differ in learning speed and variance.

---



# <font color='orange'> 8. Incremental Mean Update </font>

Instead of storing

every return,

we can update the estimate incrementally.

Suppose

after

$$
N
$$

observations,

the current estimate is

$$
V_N.
$$

After observing a new return

$$
G,
$$

the updated estimate is

$$
\boxed{
V_{N+1}
=
V_N
+
\frac{1}{N+1}
(G-V_N).
}
$$

This requires storing only

the current estimate

and

the number of visits.

---



# <font color='orange'> 9. Exploring Starts </font>

Suppose an agent always starts

from exactly the same state

and always chooses the same action.

Some states may never be visited.

To ensure sufficient exploration,

early Monte Carlo algorithms assumed

**Exploring Starts**,

meaning

every state-action pair has a non-zero probability of being selected as an initial condition.

Although useful theoretically,

this assumption is often impractical.

Modern algorithms typically use other exploration strategies, such as \(\varepsilon\)-greedy policies.

---



# <font color='orange'> 10. On-Policy Learning </font>

An **on-policy** algorithm

learns about

the same policy

that it follows.

```
Current Policy

↓

Generate Episode

↓

Update Current Policy
```

The policy used for exploration

is also the policy being improved.

---



# <font color='orange'> 11. Off-Policy Learning </font>

An **off-policy** algorithm

learns about

one policy

while following

another.

```
Behaviour Policy

↓

Generate Episode

↓

Learn Target Policy
```

This idea becomes important later

in Q-Learning.

---



# <font color='orange'> 12. Advantages </font>

Monte Carlo methods

- require no environment model,
- are conceptually simple,
- converge to correct value estimates with sufficient sampling,
- naturally handle stochastic environments.

They are particularly useful for episodic problems.

---



# <font color='orange'> 13. Limitations </font>

Monte Carlo methods also have disadvantages.

They

- require complete episodes,
- cannot update during an episode,
- may exhibit high variance,
- learn slowly when episodes are long.

These limitations motivated the development of

Temporal Difference Learning.

---

# <font color='red'> 14. Mathematical Foundations </font>

The discounted return is

$$
\boxed{
G_t
=
\sum_{k=0}^{T-t-1}
\gamma^k
r_{t+k+1},
}
$$

where

$$
T
$$

is the terminal time step.

The Monte Carlo estimate of the value function is

$$
\boxed{
V(s)
=
\frac{1}{N(s)}
\sum_{i=1}^{N(s)}
G_i,
}
$$

where

$$
N(s)
$$

is the number of observed visits to state

$$
s.
$$

The incremental update rule is

$$
\boxed{
V(s)
\leftarrow
V(s)
+
\alpha
\left(
G
-
V(s)
\right),
}
$$

where

- \(G\) is the observed return,
- \(\alpha\) is the learning rate.

When

$$
\alpha=\frac{1}{N(s)},
$$

this update is exactly equivalent to computing the running average of all observed returns.

---



# <font color='orange'> 15. First-Visit vs Every-Visit Monte Carlo </font>

| First-Visit MC | Every-Visit MC |
|:---|:---|
| Uses only the first occurrence of a state in each episode | Uses every occurrence of a state |
| Simpler theoretical analysis | Often uses more available data |
| One update per state per episode | Potentially multiple updates per state per episode |
| Converges under standard assumptions | Also converges under standard assumptions |

Both methods estimate the same underlying value function.

---



# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> Monte Carlo methods require knowledge of the transition probabilities.

**False.**

Monte Carlo methods are model-free. They learn directly from sampled episodes without requiring \(P(s'|s,a)\) or \(R(s,a)\).

---

### Misconception 2

> Monte Carlo methods update after every action.

**False.**

Classical Monte Carlo methods wait until the entire episode has finished before updating value estimates.

---

### Misconception 3

> One episode is sufficient to estimate the value function.

**False.**

A single episode provides only one sample of the return. Accurate value estimates require averaging over many episodes.

---

# <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Monte Carlo Method | Model-free learning from complete episodes |
| Return (\(G_t\)) | Discounted cumulative reward after a state |
| Monte Carlo Prediction | Estimates state values by averaging sampled returns |
| First-Visit MC | Uses only the first visit to a state in each episode |
| Every-Visit MC | Uses every visit to a state in each episode |
| Incremental Update | Updates the value estimate without storing all returns |
| On-Policy Learning | Learns the same policy used to generate experience |
| Off-Policy Learning | Learns one policy while following another |

> **Key Insight:** Monte Carlo methods estimate value functions by averaging the returns observed over complete episodes, eliminating the need for a model of the environment. By replacing exact expectations with sampled experience, they introduce the central idea of model-free reinforcement learning. Although they converge to correct value estimates under suitable conditions, their requirement to wait until the end of each episode motivates more efficient methods such as Temporal Difference Learning.
