# <font color='blue'> Chapter 18: Weight Initialization and Regularization </font>

A neural network begins its training with randomly initialized parameters. During training, these parameters are updated using backpropagation and an optimization algorithm to minimize the loss function.

However, two important questions arise:

1. **How should the weights be initialized?**
2. **How can we prevent the network from memorizing the training data?**

The first question concerns **weight initialization**, while the second concerns **regularization**.

Both play a crucial role in determining how quickly a network learns and how well it generalizes to unseen data.

---

# <font color='orange'> 1. Why Does Weight Initialization Matter? </font>

Recall that each neuron computes

$$
z = W\mathbf{x}+b.
$$

Before training begins,

the network has no knowledge of the data.

Therefore,

the weights must be assigned some initial values.

Although these values are only the starting point,

they strongly influence

- convergence speed,
- training stability,
- gradient flow,
- final model performance.

---

# <font color='orange'> 2. Why Can't We Initialize All Weights to Zero? </font>

Suppose every weight is initialized as

$$
W = 0.
$$

Then every neuron in the same layer produces exactly the same output.

For example,

$$
z_1=z_2=z_3=\cdots=0.
$$

Since all neurons receive identical gradients during backpropagation,

they remain identical throughout training.

This is known as the **symmetry problem**.

Instead of learning different features,

every neuron learns the same feature.

Consequently,

the network behaves as though it contains only one neuron per layer.

---

# <font color='orange'> 3. Random Initialization </font>

To break this symmetry,

weights are initialized randomly.

For example,

$$
W \sim \mathcal{N}(0,\sigma^2)
$$

or

$$
W \sim U(-a,a),
$$

where

- $\mathcal{N}$ denotes a normal distribution,
- $U$ denotes a uniform distribution.

Random initialization ensures that different neurons begin with different parameters.

---

# <font color='orange'> 4. Xavier (Glorot) Initialization </font>

If the initial weights are too small,

signals shrink as they propagate through the network.

If they are too large,

signals may grow uncontrollably.

Xavier initialization chooses the variance to keep activations approximately constant across layers.

For a layer with

- $n_{\text{in}}$ input units,
- $n_{\text{out}}$ output units,

the weights are sampled from

$$
\boxed{
W
\sim
\mathcal{N}
\left(
0,
\frac{2}
{n_{\text{in}}+n_{\text{out}}}
\right).
}
$$

Xavier initialization is particularly suitable for

- Sigmoid,
- tanh.

---

# <font color='orange'> 5. He Initialization </font>

ReLU-based networks require a slightly different initialization.

He initialization uses

$$
\boxed{
W
\sim
\mathcal{N}
\left(
0,
\frac{2}
{n_{\text{in}}}
\right).
}
$$

Compared with Xavier initialization,

He initialization preserves variance more effectively when ReLU activations are used.

Consequently,

it is the default initialization for many modern neural networks.

---

# <font color='orange'> 6. What is Regularization? </font>

Regularization refers to techniques that reduce **overfitting**.

Instead of memorizing the training data,

the model learns patterns that generalize to new data.

The objective is to find a balance between

- fitting the training data,
- maintaining predictive performance on unseen data.

---

# <font color='orange'> 7. L2 Regularization (Weight Decay) </font>

The most common regularization method is **L2 Regularization**.

Instead of minimizing only the loss,

we minimize

$$
\boxed{
J_{\rm new}
=
J
+
\lambda
\sum_i w_i^2.
}
$$

The additional term penalizes large weights.

Here,

- $J$ is the original loss,
- $\lambda$ controls the strength of regularization.

Large values of

$$
\lambda
$$

encourage smaller weights.

---

## Why Does L2 Help?

Large weights often make the model overly sensitive to small changes in the input.

By discouraging excessively large weights,

L2 regularization produces smoother models that typically generalize better.

---

# <font color='orange'> 8. L1 Regularization </font>

L1 regularization replaces the squared weights with their absolute values.

$$
\boxed{
J_{\rm new}
=
J
+
\lambda
\sum_i
|w_i|.
}
$$

Unlike L2,

L1 tends to drive many weights exactly to zero.

This leads to **sparse models** and can perform implicit feature selection.

---

# <font color='orange'> 9. Comparing L1 and L2 </font>

| Property | L1 | L2 |
|:---|:---:|:---:|
| Penalty | $|w|$ | $w^2$ |
| Produces Sparse Weights | ✓ | ✗ |
| Smooth Optimization | Less | More |
| Feature Selection | ✓ | ✗ |
| Most Common in Deep Learning | Rare | ✓ |

L2 regularization is generally preferred for deep neural networks.

---

# <font color='orange'> 10. Dropout </font>

Dropout randomly deactivates neurons during training.

Example

```
Training

○ ○ ○ ○ ○

↓

○ × ○ × ○

↓

Next Layer
```

Each neuron is independently removed with probability

$$
p.
$$

Typical values are

- 0.2
- 0.3
- 0.5

During inference,

all neurons are active,

but their outputs are appropriately scaled.

---

# <font color='orange'> 11. Why Does Dropout Work? </font>

Without Dropout,

neurons may become highly dependent on one another.

Dropout forces the network to learn redundant and robust feature representations.

Conceptually,

each training step uses a slightly different neural network.

The final model behaves like an average over many such subnetworks.

---

# <font color='orange'> 12. Early Stopping as Regularization </font>

Suppose validation loss begins to increase after several epochs.

Continuing training may simply memorize the training data.

EarlyStopping halts training before significant overfitting occurs.

Thus,

EarlyStopping acts as a form of **implicit regularization**.

---

# <font color='orange'> 13. Data Augmentation </font>

Rather than collecting more data,

we can generate modified versions of existing data.

For images,

typical augmentations include

- rotation,
- flipping,
- cropping,
- zooming,
- brightness adjustment.

This increases the diversity of the training set and reduces overfitting.

---

# <font color='orange'> 14. Batch Normalization and Regularization </font>

Batch Normalization was originally introduced to stabilize training.

However,

its slight randomness during training often provides a regularizing effect.

Although it should not be viewed as a replacement for Dropout,

it can improve generalization in many architectures.

---

# <font color='orange'> 15. Choosing Regularization Techniques </font>

| Situation | Recommended Technique |
|:---|:---|
| Small Dataset | L2 + Dropout |
| Large CNN | Data Augmentation + BatchNorm |
| NLP Transformer | Weight Decay (AdamW) |
| Sparse Model | L1 Regularization |
| Unknown | Begin with L2 |

---

# <font color='red'> 16. Mathematical Foundations </font>

With L2 regularization,

the optimization objective becomes

$$
J_{\rm new}
=
J
+
\lambda
\sum_i
w_i^2.
$$

Differentiating with respect to a weight,

$$
\frac{\partial J_{\rm new}}
{\partial w}
=
\frac{\partial J}
{\partial w}
+
2\lambda w.
$$

The gradient descent update is therefore

$$
\boxed{
w
\leftarrow
w
-
\eta
\left(
\frac{\partial J}{\partial w}
+
2\lambda w
\right).
}
$$

Notice that the additional term continuously pulls the weights toward zero.

This is why L2 regularization is also known as **weight decay**.

---

# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> Initializing all weights to zero simplifies training.

**False.**

Zero initialization causes all neurons within a layer to learn identical features, preventing the network from exploiting its full representational capacity.

---

### Misconception 2

> Dropout permanently removes neurons.

**False.**

Dropout randomly deactivates neurons **only during training**. During inference, all neurons participate in the computation.

---

### Misconception 3

> Regularization always improves accuracy.

**False.**

Excessive regularization may lead to underfitting, reducing both training and test performance.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Weight Initialization | Determines the starting values of trainable parameters |
| Xavier Initialization | Suitable for Sigmoid and tanh activations |
| He Initialization | Designed for ReLU-based networks |
| L1 Regularization | Encourages sparse models by penalizing absolute weights |
| L2 Regularization | Penalizes large weights and improves generalization |
| Dropout | Randomly deactivates neurons during training |
| EarlyStopping | Stops training before significant overfitting occurs |
| Data Augmentation | Creates additional training examples through transformations |
| Batch Normalization | Stabilizes training and often provides mild regularization |

> **Key Insight:** Successful neural network training depends not only on optimization algorithms but also on sensible initialization and effective regularization. Proper weight initialization ensures stable gradient flow from the start of training, while regularization techniques such as L2 regularization, Dropout, and EarlyStopping reduce overfitting and improve the model's ability to generalize to unseen data.
