# <font color='blue'> Chapter 29: Gated Recurrent Units (GRUs) </font>

Long Short-Term Memory (LSTM) networks significantly improved the ability of recurrent neural networks to learn long-term dependencies. However, this improvement came at the cost of increased computational complexity.

Each LSTM cell contains

- three gates,
- one memory cell,
- one hidden state.

Although highly effective,

this architecture requires a relatively large number of parameters.

To simplify the LSTM while maintaining similar predictive performance, researchers introduced the **Gated Recurrent Unit (GRU)**.

A GRU combines the hidden state and cell state into a single state vector and replaces the three LSTM gates with only two gates.

Consequently,

GRUs are computationally faster while still learning long-term dependencies effectively.

---



# <font color='orange'> 1. Motivation </font>

Recall that an LSTM computes

- Forget Gate
- Input Gate
- Candidate Memory
- Cell State Update
- Output Gate

Although powerful,

this involves many matrix multiplications.

Researchers asked

> Can we achieve similar performance using fewer gates?

The answer is

**Yes.**

This led to the development of the GRU.

---



# <font color='orange'> 2. Main Idea Behind a GRU </font>

Unlike an LSTM,

a GRU

- has **no separate cell state**,
- keeps only a hidden state,
- uses only two gates.

The architecture is therefore much simpler.

```
Previous Hidden State

↓

Reset Gate

↓

Update Gate

↓

Candidate Hidden State

↓

New Hidden State
```

---



# <font color='orange'> 3. The Hidden State </font>

Instead of maintaining

$$
\mathbf{C}_t
$$

and

$$
\mathbf{h}_t,
$$

the GRU stores only

$$
\boxed{
\mathbf{h}_t.
}
$$

This hidden state contains both

- short-term information,
- long-term information.

---



# <font color='orange'> 4. The Update Gate </font>

The Update Gate determines

how much of the previous hidden state should be retained.

It is computed as

$$
\boxed{
\mathbf{z}_t
=
\sigma
\left(
W_z
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_z
\right).
}
$$

Interpretation

If

$$
z_t \approx 1,
$$

the network largely keeps the previous hidden state.

If

$$
z_t \approx 0,
$$

the network relies more heavily on new information.

---



# <font color='orange'> 5. The Reset Gate </font>

The Reset Gate determines

how much of the previous hidden state should be ignored when computing new information.

It is

$$
\boxed{
\mathbf{r}_t
=
\sigma
\left(
W_r
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_r
\right).
}
$$

Interpretation

Large values

→ remember the previous state.

Small values

→ forget much of the previous state.

---



# <font color='orange'> 6. Candidate Hidden State </font>

The network now computes a candidate hidden state.

$$
\boxed{
\tilde{\mathbf{h}}_t
=
\tanh
\left(
W_h
\begin{bmatrix}
\mathbf{r}_t
\odot
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_h
\right).
}
$$

Notice that

the Reset Gate controls

how much previous information contributes to the candidate state.

---



# <font color='orange'> 7. Updating the Hidden State </font>

The final hidden state is

$$
\boxed{
\mathbf{h}_t
=
(1-\mathbf{z}_t)
\odot
\mathbf{h}_{t-1}
+
\mathbf{z}_t
\odot
\tilde{\mathbf{h}}_t.
}
$$

The Update Gate therefore blends

- old information,
- newly computed information.

---



# <font color='orange'> 8. Complete GRU Forward Pass </font>

For each time step,

the GRU performs

```
Reset Gate

↓

Update Gate

↓

Candidate Hidden State

↓

Hidden State Update
```

Compared with an LSTM,

there are fewer computations.

---



# <font color='orange'> 9. Why GRUs Work </font>

The Update Gate provides

a direct path through time,

allowing gradients to propagate much more effectively than in a standard RNN.

Like the LSTM,

the GRU avoids repeatedly overwriting useful information.

Consequently,

GRUs are much less susceptible to the vanishing gradient problem.

---



# <font color='orange'> 10. Comparing RNN, LSTM and GRU </font>

| Property | RNN | LSTM | GRU |
|:---|:---:|:---:|:---:|
| Hidden State | ✓ | ✓ | ✓ |
| Cell State | ✗ | ✓ | ✗ |
| Gates | None | 3 | 2 |
| Long-Term Memory | Poor | Excellent | Excellent |
| Parameters | Few | Many | Moderate |
| Training Speed | Fast | Slowest | Faster than LSTM |

---



# <font color='orange'> 11. Parameter Comparison </font>

Suppose

the hidden layer contains

$$
H
$$

units.

Approximate parameter counts

Standard RNN

$$
\mathcal O(H^2)
$$

GRU

approximately

$$
3
\mathcal O(H^2)
$$

LSTM

approximately

$$
4
\mathcal O(H^2)
$$

Thus,

GRUs usually require fewer parameters than LSTMs.

# <font color='orange'> 12. Applications </font>

GRUs are widely used in

- speech recognition,
- sensor analysis,
- weather forecasting,
- financial prediction,
- industrial monitoring,
- robotics,
- embedded AI,
- healthcare.

Whenever computational efficiency is important,

GRUs are an attractive alternative to LSTMs.

---



# <font color='orange'> 13. Choosing Between RNN, LSTM and GRU </font>

General guidelines

| Situation | Recommended Model |
|:---|:---|
| Short sequences | RNN |
| Medium sequences | GRU |
| Very long dependencies | LSTM |
| Limited hardware | GRU |
| Maximum modelling flexibility | LSTM |

In practice,

the best choice often depends on the dataset and available computational resources.

# <font color='red'> 14. Mathematical Foundations </font>

The complete GRU equations are

Update Gate

$$
\boxed{
\mathbf{z}_t
=
\sigma
(W_z[\mathbf{h}_{t-1},\mathbf{x}_t]+b_z),
}
$$

Reset Gate

$$
\boxed{
\mathbf{r}_t
=
\sigma
(W_r[\mathbf{h}_{t-1},\mathbf{x}_t]+b_r),
}
$$

Candidate Hidden State

$$
\boxed{
\tilde{\mathbf{h}}_t
=
\tanh
(W_h[
\mathbf{r}_t
\odot
\mathbf{h}_{t-1},
\mathbf{x}_t]
+b_h),
}
$$

Hidden State Update

$$
\boxed{
\mathbf{h}_t
=
(1-\mathbf{z}_t)
\odot
\mathbf{h}_{t-1}
+
\mathbf{z}_t
\odot
\tilde{\mathbf{h}}_t.
}
$$

Unlike the LSTM,

there is no separate cell state.

All information is contained within

$$
\mathbf{h}_t.
$$

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> GRUs are simply smaller LSTMs.

**Not entirely.**

GRUs simplify the LSTM architecture by removing the cell state and combining certain gating operations, but they introduce a distinct update mechanism rather than merely deleting components.

---

### Misconception 2

> GRUs always outperform LSTMs.

**False.**

GRUs often achieve comparable performance with fewer parameters and faster training, but LSTMs may perform better on some tasks involving very long-range dependencies.

---

### Misconception 3

> Standard RNNs are obsolete.

**False.**

Although LSTMs and GRUs are generally preferred for long sequences, standard RNNs remain useful for educational purposes and for some simple or computationally constrained applications.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| GRU | Simplified gated recurrent neural network |
| Hidden State | Combined short-term and long-term memory |
| Update Gate | Determines how much previous information is retained |
| Reset Gate | Determines how much previous information is ignored |
| Candidate Hidden State | Proposed new memory representation |
| Parameter Efficiency | Fewer parameters than an LSTM |
| Long-Term Dependencies | Better than a standard RNN |

> **Key Insight:** The Gated Recurrent Unit simplifies the LSTM architecture by removing the separate cell state and reducing the number of gates from three to two. Despite this simplification, GRUs retain the ability to model long-term dependencies while requiring fewer parameters and less computation. They therefore provide an excellent balance between efficiency and predictive performance, making them a popular choice for many sequence modelling tasks.
