# <font color='blue'> Appendix A1: Language Models Before Reinforcement Learning </font>

Modern conversational AI systems, such as ChatGPT, Claude, Gemini, and many open-source Large Language Models (LLMs), are built through several stages of training.

The first stage is called **pretraining**.

During pretraining, the model is not taught to answer questions, follow instructions, or have conversations.

Instead, it learns one remarkably simple task:

> **Predict the next word (or more precisely, the next token) in a sequence of text.**

Although this objective appears simple, it enables the model to acquire knowledge of grammar, facts, reasoning patterns, programming languages, mathematics, and many other aspects of human language.

This chapter explains how language models are pretrained and why pretraining alone is insufficient for building useful AI assistants.

---



# <font color='orange'> 1. Motivation </font>

Imagine reading a sentence where the last word has been removed.

```
The Earth revolves around the ______
```

Most people would predict

```
Sun
```

without much difficulty.

Similarly,

consider

```
Paris is the capital of ______
```

Most people would answer

```
France
```

Language models learn by repeatedly solving problems like these,

billions of times,

using enormous collections of text.

---



# <font color='orange'> 2. What is a Language Model? </font>

A **Language Model (LM)** is a machine learning model that estimates the probability of observing a sequence of words or tokens.

Given a sequence

$$
x_1,x_2,\ldots,x_n,
$$

the model estimates

$$
\boxed{
P(x_1,x_2,\ldots,x_n).
}
$$

Rather than estimating this probability directly,

it is decomposed using the chain rule of probability.

---



# <font color='orange'> 3. The Chain Rule of Probability </font>

The probability of an entire sentence can be written as

$$
\boxed{
P(x_1,\ldots,x_n)
=
\prod_{t=1}^{n}
P(x_t|x_1,\ldots,x_{t-1}).
}
$$

This equation states that

the probability of the complete sentence

is simply

the product of the probability of each token,

conditioned on all previous tokens.

For example,

```
The

↓

cat

↓

sat

↓

on

↓

the

↓

mat
```

The model predicts each token

using all previously generated tokens.

---



# <font color='orange'> 4. What is a Token? </font>

Although we often speak about predicting words,

modern language models actually predict **tokens**.

A token is the basic unit processed by the model.

Depending on the tokenizer,

a token may represent

- an entire word,
- part of a word,
- punctuation,
- numbers,
- mathematical symbols.

Example

```
unbelievable

↓

un

↓

believ

↓

able
```

Breaking text into tokens allows the model to efficiently represent an enormous vocabulary.

---



# <font color='orange'> 5. Next-Token Prediction </font>

Suppose the input is

```
The largest planet in the Solar System is
```

The model predicts a probability distribution over all possible next tokens.

Example

| Token | Probability |
|:---|---:|
| Jupiter | 0.94 |
| Saturn | 0.03 |
| Earth | 0.01 |
| Mars | 0.01 |
| Moon | 0.01 |

The token with the highest probability is often selected,

although different decoding strategies may also be used.

---



# <font color='orange'> 6. Neural Network Output </font>

Internally,

the Transformer produces

a vector of numbers called **logits**.

```
Input Tokens

↓

Transformer

↓

Logits

↓

Softmax

↓

Probabilities
```

The logits themselves are not probabilities.

They become probabilities only after applying the Softmax function.

---



# <font color='orange'> 7. Softmax Function </font>

Suppose the Transformer outputs logits

$$
z_1,z_2,\ldots,z_K.
$$

These are converted into probabilities using

$$
\boxed{
P_i
=
\frac{e^{z_i}}
{\sum_{j=1}^{K}e^{z_j}}.
}
$$

The Softmax function ensures

- every probability is positive,
- all probabilities sum to one.

---



# <font color='orange'> 8. Cross-Entropy Loss </font>

The model must learn

to assign high probability

to the correct next token.

Suppose

the correct token is

$$
y.
$$

The training loss is

$$
\boxed{
L
=
-
\log
P(y).
}
$$

This is called the **Cross-Entropy Loss**.

If

the correct token receives

high probability,

the loss is small.

If

the model assigns low probability,

the loss becomes large.

---



# <font color='orange'> 9. Learning Through Gradient Descent </font>

Training proceeds exactly as in other neural networks.

```
Input Tokens

↓

Transformer

↓

Predicted Probabilities

↓

Cross-Entropy Loss

↓

Backpropagation

↓

Gradient Descent

↓

Update Parameters
```

This process is repeated

over trillions of tokens.

Eventually,

the model becomes increasingly accurate at predicting the next token.

---



# <font color='orange'> 10. What Does the Model Learn? </font>

Although the objective is simply next-token prediction,

the model gradually learns

- grammar,
- spelling,
- syntax,
- factual knowledge,
- reasoning patterns,
- programming languages,
- mathematical notation,
- scientific writing,
- multilingual representations.

None of these capabilities are programmed explicitly.

They emerge naturally from the prediction task.

---



# <font color='orange'> 11. Why Pretraining Works </font>

Predicting the next token requires understanding

- sentence structure,
- context,
- world knowledge,
- relationships between concepts.

For example,

consider

```
Einstein developed the theory of
```

The correct prediction requires

scientific knowledge,

not merely grammar.

Thus,

the model gradually acquires a broad internal representation of language and knowledge during pretraining.

---



# <font color='orange'> 12. Limitations of Pretraining </font>

Although pretraining produces an extremely capable language model,

it does **not** teach the model to

- follow instructions,
- answer questions politely,
- refuse harmful requests,
- maintain conversations,
- align with human preferences.

For example,

consider the prompt

```
Explain Newton's Laws.
```

A pretrained model might

- complete the prompt,
- continue an article,
- imitate internet text,

rather than providing a helpful explanation.

Pretraining optimizes prediction,

not assistance.

---



# <font color='orange'> 13. Why More Training is Needed </font>

To become a useful assistant,

the model must learn

- how humans ask questions,
- how humans prefer answers,
- conversational behaviour,
- instruction following,
- helpfulness,
- harmlessness.

These abilities are introduced

after pretraining

through additional training stages.

These stages include

- Supervised Fine-Tuning (SFT),
- Reinforcement Learning from Human Feedback (RLHF),
- Direct Preference Optimization (DPO).

---

# <font color='red'> 14. Mathematical Foundations </font>

### Language Model

The probability of a sequence is

$$
\boxed{
P(x_1,\ldots,x_n)
=
\prod_{t=1}^{n}
P(x_t|x_{<t}),
}
$$

where

$$
x_{<t}
=
(x_1,\ldots,x_{t-1}).
$$

---

### Softmax

Given logits

$$
z_i,
$$

the predicted probability of token

$$
i
$$

is

$$
\boxed{
P_i
=
\frac{e^{z_i}}
{\sum_j e^{z_j}}.
}
$$

---

### Cross-Entropy Loss

If

the correct next token is

$$
y,
$$

the loss is

$$
\boxed{
L
=
-
\log
P(y).
}
$$

For an entire sequence of length \(T\),

the total training loss is

$$
\boxed{
L
=
-
\sum_{t=1}^{T}
\log
P(x_t|x_{<t}).
}
$$

Minimizing this loss encourages the model to assign high probability to the correct next token at every position.

---



# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> Language models memorize every sentence they read.

**False.**

Although memorization can occur in specific cases, the primary goal of pretraining is to learn statistical patterns and relationships in language rather than storing exact copies of the training data.

---

### Misconception 2

> The model understands language in the same way humans do.

**Not necessarily.**

The model learns statistical relationships between tokens. These learned representations enable impressive language use, but they are not the same as human understanding or consciousness.

---

### Misconception 3

> Pretraining teaches the model to be a helpful assistant.

**False.**

Pretraining teaches the model to predict the next token. Instruction following, conversational behaviour, and alignment are introduced in later training stages such as Supervised Fine-Tuning and RLHF.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Language Model | Predicts the probability of text sequences |
| Token | Basic unit processed by the model |
| Next-Token Prediction | Core objective during pretraining |
| Chain Rule | Decomposes sequence probability into conditional probabilities |
| Logits | Raw scores produced by the Transformer |
| Softmax | Converts logits into probabilities |
| Cross-Entropy Loss | Measures prediction error for the next token |
| Pretraining | Learns language patterns from massive text corpora |
| Limitation | Does not teach instruction following or alignment |

> **Key Insight:** Modern Large Language Models are first trained through self-supervised pretraining, where the objective is simply to predict the next token in a sequence. Despite this seemingly simple task, the model acquires extensive knowledge of language, facts, and reasoning by learning statistical patterns across vast amounts of text. However, pretraining alone does not produce a conversational assistant. Additional training stages are required to teach the model how to follow instructions, interact helpfully with users, and align its behaviour with human preferences.
