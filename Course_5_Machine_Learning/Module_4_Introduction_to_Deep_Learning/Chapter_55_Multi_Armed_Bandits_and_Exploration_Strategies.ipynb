# <font color='blue'> Chapter 55: Multi-Armed Bandits and Exploration Strategies </font>

One of the fundamental challenges in Reinforcement Learning is deciding whether to

- exploit what is already known,
- or explore actions that may produce even better rewards.

This dilemma is known as the **exploration–exploitation trade-off**.

To study this problem in isolation,

we consider one of the simplest decision-making problems in Machine Learning:

the **Multi-Armed Bandit**.

Unlike a full Reinforcement Learning problem,

there are

- no states,
- no state transitions,
- no long-term planning.

The agent simply chooses one action,

receives an immediate reward,

and repeats the process.

Although simple,

Multi-Armed Bandits provide the mathematical foundation for many exploration strategies used in modern Reinforcement Learning.

---

# <font color='orange'> 1. Motivation </font>

Imagine entering a casino.

In front of you are several slot machines.

```
🎰 1

🎰 2

🎰 3

🎰 4

🎰 5
```

Each machine has an unknown probability of paying out.

Your goal is to maximize the total money won.

Should you

- keep playing the machine that has paid well so far,

or

- try another machine that might pay even better?

This is the Multi-Armed Bandit problem.

---

# <font color='orange'> 2. Why "Multi-Armed Bandit"? </font>

A slot machine is sometimes called a

**one-armed bandit**

because it has

- one lever ("one arm"),
- and it takes your money like a bandit.

If several slot machines are available,

we obtain a

**Multi-Armed Bandit**.

Each machine represents one possible action.

---

# <font color='orange'> 3. The Bandit Setting </font>

At every time step,

the agent

```
Choose Action

↓

Receive Reward

↓

Update Estimates

↓

Repeat
```

Notice that

there is

- no state,
- no next state,
- no trajectory.

Only immediate rewards matter.

---

# <font color='orange'> 4. Action Values </font>

Each action

$$
a
$$

has an unknown expected reward

$$
\boxed{
q_*(a).
}
$$

The agent does not know these values initially.

Instead,

it must estimate them from experience.

The estimated value is written

$$
\boxed{
Q_t(a),
}
$$

where

$$
t
$$

denotes the current time step.

As more rewards are observed,

the estimate improves.

---

# <font color='orange'> 5. Estimating the Action Value </font>

Suppose

action

$$
a
$$

has been selected

$$
N(a)
$$

times.

If the observed rewards are

$$
R_1,R_2,\ldots,R_{N(a)},
$$

the estimated value is simply the sample mean,

$$
\boxed{
Q(a)
=
\frac{1}{N(a)}
\sum_{i=1}^{N(a)}
R_i.
}
$$

This estimate converges to the true expected reward as the number of samples increases.

---

# <font color='orange'> 6. Incremental Update </font>

Instead of storing every reward,

the estimate can be updated incrementally.

After receiving a new reward

$$
R,
$$

the estimate becomes

$$
\boxed{
Q_{n+1}
=
Q_n
+
\frac{1}{n}
\left(
R-Q_n
\right).
}
$$

More generally,

using a constant learning rate,

$$
\boxed{
Q
\leftarrow
Q
+
\alpha
(R-Q).
}
$$

This update resembles the Temporal Difference updates encountered earlier.

---

# <font color='orange'> 7. Exploration vs Exploitation </font>

Suppose

Machine A

has produced

£500.

Machine B

has never been tried.

Should you

```
Keep Playing A

or

Try B?
```

Exploitation chooses

the action currently believed to be best.

Exploration tries actions that may eventually prove superior.

Balancing these competing objectives is the central challenge of the bandit problem.

---

# <font color='orange'> 8. Greedy Strategy </font>

The simplest strategy always selects

the action with the highest estimated value.

Mathematically,

$$
\boxed{
a_t
=
\arg\max_a
Q_t(a).
}
$$

Although simple,

pure greedy algorithms often become trapped in suboptimal actions because they stop exploring too early.

---

# <font color='orange'> 9. ε-Greedy Strategy </font>

To encourage exploration,

the ε-greedy algorithm behaves as follows.

With probability

$$
1-\varepsilon,
$$

choose

the greedy action.

With probability

$$
\varepsilon,
$$

choose

a random action.

Typical values include

$$
\varepsilon
=
0.1
\quad
\text{or}
\quad
0.01.
$$

This simple strategy is widely used in Q-Learning and Deep Q Networks.

---

# <font color='orange'> 10. Optimistic Initial Values </font>

Another exploration strategy is to initialize all action values

with unrealistically large values.

Example

```
Q(A)=100

Q(B)=100

Q(C)=100
```

Initially,

every action appears promising.

The agent naturally explores all actions before settling on the most rewarding ones.

No random exploration is required.

---

# <font color='orange'> 11. Upper Confidence Bound (UCB) </font>

Rather than exploring randomly,

Upper Confidence Bound (UCB) chooses actions according to both

- their estimated reward,
- the uncertainty in that estimate.

The action selected is

$$
\boxed{
a_t
=
\arg\max_a
\left[
Q_t(a)
+
c
\sqrt{
\frac{\ln t}
{N_t(a)}
}
\right].
}
$$

where

- \(Q_t(a)\) is the estimated reward,
- \(N_t(a)\) is the number of times action \(a\) has been selected,
- \(t\) is the total number of action selections,
- \(c\) controls the degree of exploration.

Actions that have been tried only a few times receive a larger exploration bonus.

As more data are collected,

this bonus decreases.

---

# <font color='orange'> 12. Thompson Sampling </font>

Another powerful strategy is

**Thompson Sampling**.

Instead of choosing the action with the highest estimated reward,

the algorithm

samples

a plausible reward distribution

for each action,

then selects

the action whose sampled reward is highest.

Intuitively,

actions with greater uncertainty are sampled more aggressively,

naturally balancing exploration and exploitation.

Thompson Sampling is widely used in recommendation systems,

online advertising,

and A/B testing.

---

# <font color='orange'> 13. Regret </font>

The objective is not merely to maximize reward,

but to minimize

**regret**.

Regret measures

how much reward has been lost

by not always selecting the optimal action.

If

$$
q_*
$$

is the true reward of the optimal action,

the cumulative regret after

$$
T
$$

steps is

$$
\boxed{
R_T
=
\sum_{t=1}^{T}
\left(
q_*
-
q(a_t)
\right).
}
$$

A good exploration strategy minimizes long-term regret.

---

# <font color='orange'> 14. Bandits vs Reinforcement Learning </font>

| Multi-Armed Bandit | Reinforcement Learning |
|:---|:---|
| No states | Multiple states |
| No state transitions | Environment changes after actions |
| Immediate rewards | Immediate and future rewards |
| No Bellman equations | Bellman equations required |
| No planning | Sequential decision making |

Bandit problems can therefore be viewed as the simplest form of Reinforcement Learning.

---

# <font color='orange'> 15. Applications </font>

Multi-Armed Bandits are widely used in

- recommendation systems,
- online advertising,
- A/B testing,
- clinical trials,
- adaptive pricing,
- website optimization,
- search engines.

Many practical decision-making systems rely on bandit algorithms rather than full Reinforcement Learning.

---

# <font color='red'> 16. Mathematical Foundations </font>

True action value

$$
\boxed{
q_*(a)
=
\mathbb E[R|A=a].
}
$$

Sample-average estimate

$$
\boxed{
Q(a)
=
\frac{1}{N(a)}
\sum_{i=1}^{N(a)}
R_i.
}
$$

Incremental update

$$
\boxed{
Q
\leftarrow
Q
+
\alpha
(R-Q).
}
$$

Greedy action selection

$$
\boxed{
a_t
=
\arg\max_a
Q_t(a).
}
$$

Upper Confidence Bound

$$
\boxed{
a_t
=
\arg\max_a
\left(
Q_t(a)
+
c
\sqrt{
\frac{\ln t}
{N_t(a)}
}
\right).
}
$$

$$
\boxed{

$$

These equations describe the core mathematical framework of the Multi-Armed Bandit problem.

# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> Multi-Armed Bandits are the same as Reinforcement Learning.

**False.**

Bandit problems have no states or state transitions. Every decision is independent of previous ones, except through the agent's updated estimates.

---

### Misconception 2

> Exploration is only useful at the beginning of learning.

**False.**

In many environments, especially those that change over time, continued exploration remains important because the best action may change.

---

### Misconception 3

> The greedy strategy always performs best.

**False.**

A purely greedy strategy can become stuck exploiting a suboptimal action if it stops exploring too early.

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Multi-Armed Bandit | Simplified decision-making problem with no states |
| Action Value \(q_*(a)\) | True expected reward of an action |
| Estimated Value \(Q(a)\) | Agent's estimate of an action's expected reward |
| Exploration | Trying uncertain actions to gain information |
| Exploitation | Selecting the action currently believed to be best |
| ε-Greedy | Random exploration with probability \(\varepsilon\) |
| Optimistic Initial Values | Encourages exploration by starting with high estimates |
| Upper Confidence Bound (UCB) | Balances reward estimates with uncertainty |
| Thompson Sampling | Bayesian exploration strategy based on sampling |
| Regret | Reward lost by not always selecting the optimal action |

> **Key Insight:** The Multi-Armed Bandit problem isolates the exploration–exploitation trade-off by removing the complexity of state transitions and long-term planning. An agent repeatedly chooses among several actions with unknown reward distributions, gradually improving its estimates through experience. Strategies such as ε-greedy, Upper Confidence Bound, and Thompson Sampling provide principled ways to balance exploration and exploitation, and these ideas underpin many exploration mechanisms used in modern reinforcement learning algorithms.
