# <font color='blue'> Chapter 46: Dynamic Programming for Reinforcement Learning </font>

Dynamic Programming (DP) is a collection of algorithms for solving Markov Decision Processes (MDPs) when the environment is fully known.

Unlike many Reinforcement Learning algorithms, Dynamic Programming assumes that we know

- all states,
- all actions,
- the transition probabilities,
- the reward function.

Using this information, DP repeatedly applies the Bellman equations until the value functions and policies converge.

Although these assumptions are unrealistic for many practical problems, Dynamic Programming provides the mathematical foundation for modern Reinforcement Learning algorithms such as Q-Learning, SARSA, and Deep Q Networks.

---



# <font color='orange'> 1. Motivation </font>

Suppose a robot must navigate through a maze.

```
Start

↓

Move

↓

Move

↓

Goal
```

If we know exactly

- where every wall is,
- the probability of every movement,
- the reward at every location,

then we should be able to compute the optimal path without trial and error.

Dynamic Programming provides algorithms for doing exactly this.

---



# <font color='orange'> 2. Why Dynamic Programming? </font>

The Bellman equations describe the value function,

but they do not tell us how to compute it.

Dynamic Programming provides iterative procedures that repeatedly apply the Bellman equations until the values stop changing.

This process gradually improves both

- value estimates,
- decision policies.

---



# <font color='orange'> 3. Two Main DP Algorithms </font>

Dynamic Programming consists of two classical algorithms.

```
Dynamic Programming

↓

Policy Iteration

↓

Value Iteration
```

Both algorithms compute the optimal policy,

but they update value functions differently.

---



# <font color='orange'> 4. Policy Evaluation </font>

Suppose we already have a policy

$$
\pi.
$$

The first question is

> **How good is this policy?**

Policy Evaluation computes

$$
V^\pi(s)
$$

for every state.

Using the Bellman Expectation Equation,

$$
\boxed{
V_{k+1}(s)
=
\sum_a
\pi(a|s)
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V_k(s')
\right].
}
$$

Starting from an initial guess,

the values are updated repeatedly until convergence.

---



# <font color='orange'> 5. Policy Improvement </font>

Once we know

$$
V^\pi(s),
$$

we can ask

> **Can we improve the policy?**

The answer is yes.

For every state,

choose the action with the largest expected return.

Mathematically,

$$
\boxed{
\pi'(s)
=
\arg\max_a
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V^\pi(s')
\right].
}
$$

The new policy

$$
\pi'
$$

is guaranteed to be at least as good as the previous one.

This result is known as the **Policy Improvement Theorem**.

---



# <font color='orange'> 6. Policy Iteration </font>

Policy Iteration alternates between

- Policy Evaluation,
- Policy Improvement.

The algorithm is

```
Random Policy

↓

Evaluate Policy

↓

Improve Policy

↓

Evaluate Again

↓

Improve Again

↓

Repeat Until Stable
```

When the policy no longer changes,

the optimal policy has been found.

---



# <font color='orange'> 7. Policy Iteration Algorithm </font>

Algorithm

1. Initialize a policy randomly.
2. Evaluate the policy using the Bellman Expectation Equation.
3. Improve the policy by selecting better actions.
4. Repeat until the policy remains unchanged.

Because each improvement step never decreases performance,

Policy Iteration converges to an optimal policy for finite MDPs.

---



# <font color='orange'> 8. Value Iteration </font>

Policy Evaluation can be computationally expensive.

Instead,

Value Iteration combines

evaluation

and

improvement

into one update.

The Bellman Optimality Equation becomes

$$
\boxed{
V_{k+1}(s)
=
\max_a
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V_k(s')
\right].
}
$$

Rather than evaluating a complete policy,

each update immediately assumes the best action.

---



# <font color='orange'> 9. Policy Extraction </font>

After Value Iteration converges,

the optimal policy is obtained by selecting the action with the highest expected value.

$$
\boxed{
\pi^*(s)
=
\arg\max_a
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V^*(s')
\right].
}
$$

Thus,

Value Iteration computes

the optimal value function first,

and extracts the policy afterwards.

---



# <font color='orange'> 10. Grid World Example </font>

Consider the following grid.

```
+----+----+----+

| S  |    | G  |

+----+----+----+

|    | X  |    |

+----+----+----+

|    |    |    |

+----+----+----+
```

Suppose

- reaching the goal gives

+100,

- every move costs

-1,

- obstacles cannot be entered.

Initially,

every state may have value

0.

Repeated Bellman updates gradually propagate the high reward from the goal back through the grid,

revealing the optimal path.

---



# <font color='orange'> 11. Convergence </font>

Both Policy Iteration

and

Value Iteration

converge under standard assumptions for finite discounted MDPs.

Intuitively,

each iteration improves or maintains the value estimates,

and the discount factor

$$
0 \le \gamma < 1
$$

ensures that future rewards contribute progressively less,

preventing infinite oscillations in typical continuing tasks.

---



# <font color='orange'> 12. Computational Complexity </font>

Suppose

- number of states

$$
|\mathcal S|,
$$

- number of actions

$$
|\mathcal A|.
$$

Each Bellman update requires considering

every action

and

every possible next state.

Consequently,

Dynamic Programming becomes computationally expensive for very large state spaces.

This challenge is known as the

**curse of dimensionality**.

It motivates model-free methods such as Q-Learning and Deep Reinforcement Learning.

---



# <font color='orange'> 13. Policy Iteration vs Value Iteration </font>

| Policy Iteration | Value Iteration |
|:---|:---|
| Separate evaluation and improvement steps | Combines evaluation and improvement |
| Usually fewer iterations | Often simpler updates |
| Each iteration is more computationally expensive | Each iteration is computationally cheaper |
| Computes the policy during optimization | Extracts the policy after convergence |

Both algorithms converge to the optimal policy for finite discounted MDPs.

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> Dynamic Programming can solve any Reinforcement Learning problem.

**False.**

Dynamic Programming requires complete knowledge of the environment, including transition probabilities and reward functions. In many real-world problems, these quantities are unknown.

---

### Misconception 2

> Policy Iteration and Value Iteration produce different optimal policies.

**False.**

Both algorithms converge to an optimal policy for finite discounted MDPs, although they follow different computational procedures.

---

### Misconception 3

> Value Iteration computes the policy directly.

**False.**

Value Iteration computes the optimal value function. The optimal policy is then extracted by selecting the action that maximizes the Bellman Optimality Equation.

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Dynamic Programming | Methods for solving known MDPs using Bellman equations |
| Policy Evaluation | Computes the value function for a fixed policy |
| Policy Improvement | Updates a policy using the current value estimates |
| Policy Iteration | Alternates between evaluation and improvement |
| Value Iteration | Directly applies the Bellman Optimality Equation |
| Policy Extraction | Recovers the optimal policy from the optimal value function |
| Convergence | Repeated Bellman updates converge under standard assumptions |
| Curse of Dimensionality | Computational challenge arising from large state spaces |

> **Key Insight:** Dynamic Programming transforms the Bellman equations into practical algorithms for solving Markov Decision Processes with known dynamics. Policy Iteration repeatedly evaluates and improves a policy until it becomes optimal, while Value Iteration directly updates the value function using the Bellman Optimality Equation. Although Dynamic Programming requires a complete model of the environment, its recursive update principles form the foundation of nearly all modern Reinforcement Learning algorithms.
