# <font color='blue'> Chapter 45: Value Functions and the Bellman Equation </font>

In Reinforcement Learning, the ultimate goal is to learn a policy that maximizes the total expected reward.

However,

rather than directly searching for the best policy,

it is often easier to estimate **how good** different states and actions are.

This idea leads to **Value Functions**.

Value functions estimate the expected future reward that an agent can obtain from a particular state or state-action pair.

The Bellman Equation provides a recursive relationship between these values, making it one of the most fundamental equations in Reinforcement Learning.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot is navigating a maze.

```
Start

↓

Move

↓

Move

↓

Goal (+100)
```

When the robot is at the start,

it has not yet received any reward.

Nevertheless,

the starting position is valuable because

it eventually leads to the goal.

Thus,

the value of a state is not determined solely by the immediate reward,

but by **all future rewards** that can be expected.

---

# <font color='orange'> 2. What is a Value Function? </font>

A value function measures

how good it is

to be in a particular state.

Intuitively,

it answers the question

> **If I start from this state and follow a particular policy, how much total reward should I expect to receive?**

States closer to the goal generally have higher values,

while states leading to poor outcomes have lower values.

---

# <font color='orange'> 3. State-Value Function </font>

The **state-value function** is denoted

$$
\boxed{
V^\pi(s).
}
$$

It is defined as

the expected return obtained by starting in state

$$
s
$$

and then following policy

$$
\pi.
$$

Mathematically,

$$
\boxed{
V^\pi(s)
=
\mathbb E_\pi[G_t\mid S_t=s].
}
$$

where

- \(G_t\) is the discounted return,
- the expectation is taken over all possible future trajectories generated by policy \(\pi\).

---

# <font color='orange'> 4. Action-Value Function </font>

Sometimes,

we wish to evaluate not only a state,

but a specific action taken from that state.

This leads to the **action-value function**

$$
\boxed{
Q^\pi(s,a).
}
$$

It measures

the expected return obtained by

- starting in state \(s\),
- taking action \(a\),
- then following policy \(\pi\).

Mathematically,

$$
\boxed{
Q^\pi(s,a)
=
\mathbb E_\pi
[G_t
\mid
S_t=s,
A_t=a].
}
$$

This function plays a central role in algorithms such as Q-Learning.

---

# <font color='orange'> 5. Why Are Value Functions Useful? </font>

Suppose two possible actions are available.

```
Action A

↓

Expected Return = 50
```

```
Action B

↓

Expected Return = 120
```

The rational choice is

Action B,

because it leads to a larger expected future reward.

Rather than memorizing every possible trajectory,

the agent only needs to estimate the value functions.

---

# <font color='orange'> 6. The Discounted Return </font>

Recall that

the discounted return is

$$
\boxed{
G_t
=
r_{t+1}
+
\gamma r_{t+2}
+
\gamma^2r_{t+3}
+\cdots.
}
$$

Notice that

we can separate the first reward,

$$
r_{t+1},
$$

from the remaining terms,

$$
\begin{aligned}
G_t
&=
r_{t+1}
+
\gamma
\left(
r_{t+2}
+
\gamma r_{t+3}
+\cdots
\right).
\end{aligned}
$$

The expression inside the parentheses is exactly

$$
G_{t+1}.
$$

Therefore,

$$
\boxed{
G_t
=
r_{t+1}
+
\gamma G_{t+1}.
}
$$

This recursive relationship is the key insight behind the Bellman equations.

---

# <font color='orange'> 7. Deriving the Bellman Expectation Equation </font>

Starting from

$$
V^\pi(s)
=
\mathbb E_\pi[G_t\mid S_t=s],
$$

substitute

$$
G_t
=
r_{t+1}
+
\gamma G_{t+1}.
$$

Then

$$
V^\pi(s)
=
\mathbb E_\pi
[
r_{t+1}
+
\gamma G_{t+1}
\mid
S_t=s
].
$$

Using the linearity of expectation,

$$
\boxed{
V^\pi(s)
=
\mathbb E_\pi
[
r_{t+1}
+
\gamma V^\pi(S_{t+1})
\mid
S_t=s
].
}
$$

This is the **Bellman Expectation Equation**.

It states that

> **The value of the current state equals the immediate reward plus the discounted value of the next state.**

---

# <font color='orange'> 8. Bellman Expectation Equation (Expanded Form) </font>

If the transition probabilities are known,

the Bellman equation can be written as

$$
\boxed{
V^\pi(s)
=
\sum_a
\pi(a|s)
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V^\pi(s')
\right].
}
$$

This equation averages over

- all possible actions,
- all possible next states,
- all possible rewards.

---

# <font color='orange'> 9. Bellman Equation for the Action-Value Function </font>

Similarly,

the action-value function satisfies

$$
\boxed{
Q^\pi(s,a)
=
\mathbb E_\pi
[
r_{t+1}
+
\gamma
Q^\pi(S_{t+1},A_{t+1})
].
}
$$

This recursive equation forms the basis of many value-based reinforcement learning algorithms.

---

# <font color='orange'> 10. Bellman Optimality Equation </font>

Suppose the agent always chooses

the best possible action.

Then

the optimal value function is

$$
\boxed{
V^*(s).
}
$$

The Bellman Optimality Equation becomes

$$
\boxed{
V^*(s)
=
\max_a
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma V^*(s')
\right].
}
$$

Notice that

the expectation over actions has been replaced by a

maximum,

because the optimal policy always selects the action producing the greatest expected return.

---

# <font color='orange'> 11. Optimal Action-Value Function </font>

Similarly,

the optimal action-value function satisfies

$$
\boxed{
Q^*(s,a)
=
\sum_{s',r}
P(s',r|s,a)
\left[
r
+
\gamma
\max_{a'}
Q^*(s',a')
\right].
}
$$

This equation is the foundation of

Q-Learning

and

Deep Q Networks.

---

# <font color='orange'> 12. Why is the Bellman Equation Important? </font>

The Bellman equation transforms

a difficult long-term optimization problem

into

a sequence of simpler recursive computations.

Instead of evaluating every possible future trajectory,

the agent repeatedly updates value estimates using information from the next state.

This recursive structure makes Reinforcement Learning computationally feasible.

---

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> The value of a state is equal to its immediate reward.

**False.**

The value of a state includes both the immediate reward and the expected discounted future rewards.

---

### Misconception 2

> The Bellman equation computes the optimal policy directly.

**False.**

The Bellman equations define relationships between value functions. Algorithms such as Value Iteration, Policy Iteration, and Q-Learning use these relationships to compute optimal policies.

---

### Misconception 3

> The Bellman equation only applies to deterministic environments.

**False.**

The Bellman equations explicitly account for stochastic transitions by averaging over all possible next states and rewards.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| State-Value Function \(V^\pi(s)\) | Expected return from a state while following policy \(\pi\) |
| Action-Value Function \(Q^\pi(s,a)\) | Expected return after taking action \(a\) in state \(s\) |
| Discounted Return \(G_t\) | Sum of discounted future rewards |
| Bellman Expectation Equation | Recursive equation for evaluating a policy |
| Bellman Optimality Equation | Recursive equation for the optimal value function |
| Optimal Value Function \(V^*(s)\) | Maximum expected return achievable from a state |
| Optimal Action-Value Function \(Q^*(s,a)\) | Maximum expected return achievable from a state-action pair |

> **Key Insight:** Value functions estimate the long-term usefulness of states and actions by considering both immediate and future rewards. The Bellman equations exploit the recursive nature of the discounted return, expressing the value of a state in terms of the reward received after one step and the value of the next state. This recursive decomposition is the mathematical foundation of nearly every classical and modern reinforcement learning algorithm.
