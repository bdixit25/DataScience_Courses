# <font color='blue'> Chapter 43: Introduction to Reinforcement Learning </font>

Reinforcement Learning (RL) is a branch of Machine Learning in which an **agent** learns how to make decisions by interacting with an **environment**.

Unlike supervised learning, where correct answers are provided, the agent learns through **trial and error**.

After taking an action, the environment provides feedback in the form of a **reward** or **penalty**.

The agent's objective is not simply to maximize the immediate reward, but to maximize the **total reward accumulated over time**.

This makes Reinforcement Learning particularly suitable for sequential decision-making problems such as robotics, autonomous driving, game playing, finance, and resource optimization.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to train a robot to navigate through a maze.

Initially,

the robot knows nothing about the environment.

```
Start

↓

Move

↓

Wall

↓

Penalty

↓

Try Again
```

After many attempts,

the robot gradually discovers which actions lead to the goal.

Unlike supervised learning,

no one tells the robot the correct path.

Instead,

it learns from the consequences of its actions.

---

# <font color='orange'> 2. What Makes Reinforcement Learning Different? </font>

Machine learning can be broadly divided into three paradigms.

| Learning Type | Learning Signal | Objective |
|:---|:---|:---|
| Supervised Learning | Labelled examples | Predict correct outputs |
| Unsupervised Learning | No labels | Discover hidden patterns |
| Reinforcement Learning | Rewards and penalties | Learn an optimal decision-making strategy |

In RL,

the model is not given the correct action.

Instead,

it must discover good actions through experience.

---

# <font color='orange'> 3. Agent and Environment </font>

Every Reinforcement Learning problem consists of two components.

```
+-----------+      Action      +--------------+
|           | ---------------> |              |
|   Agent   |                  | Environment  |
|           | <--------------- |              |
+-----------+   State, Reward   +--------------+
```

The **agent** makes decisions.

The **environment** responds to those decisions.

Learning emerges through repeated interaction.

---

# <font color='orange'> 4. State </font>

A **state** represents everything the agent currently knows about the environment.

Examples include

- the position of a robot,
- the arrangement of pieces on a chessboard,
- the speed and location of a self-driving car,
- the current frame in a video game.

States are usually denoted by

$$
\boxed{
s.
}
$$

At time step

$$
t,
$$

the state is written as

$$
\boxed{
s_t.
}
$$

---

# <font color='orange'> 5. Action </font>

An **action** is a decision made by the agent.

Examples

- move left,
- move right,
- accelerate,
- brake,
- pick up an object,
- play a chess move.

Actions are denoted by

$$
\boxed{
a.
}
$$

At time step

$$
t,
$$

the chosen action is

$$
\boxed{
a_t.
}
$$

---

# <font color='orange'> 6. Reward </font>

After taking an action,

the environment provides a **reward**.

Examples

```
Reach Goal

↓

+100
```

```
Hit Wall

↓

-10
```

```
Take One Step

↓

-1
```

Rewards encourage desirable behaviour and discourage poor decisions.

The reward at time step

$$
t
$$

is written as

$$
\boxed{
r_t.
}
$$

---

# <font color='orange'> 7. The Interaction Loop </font>

Reinforcement Learning proceeds through repeated interactions.

```
Current State

↓

Choose Action

↓

Environment

↓

Next State

↓

Reward

↓

Repeat
```

The agent continuously updates its behaviour based on the rewards it receives.

---

# <font color='orange'> 8. Episodes </font>

An **episode** is one complete interaction with the environment.

Example

```
Start

↓

Move

↓

Move

↓

Move

↓

Goal

↓

Episode Ends
```

Some tasks naturally terminate,

while others may continue indefinitely.

---

# <font color='orange'> 9. Policies </font>

A **policy** tells the agent which action to choose in each state.

It is usually denoted by

$$
\boxed{
\pi.
}
$$

Mathematically,

a policy maps

states

to

actions.

Deterministic policy

$$
\boxed{
a=\pi(s).
}
$$

Stochastic policy

$$
\boxed{
\pi(a|s).
}
$$

The stochastic policy gives the probability of choosing action

$$
a
$$

when the current state is

$$
s.
$$

---

# <font color='orange'> 10. Return </font>

In RL,

the objective is not to maximize the immediate reward,

but the **total future reward**.

This total is called the **return**.

For an episode of length

$$
T,
$$

the return from time step

$$
t
$$

is

$$
\boxed{
G_t
=
r_{t+1}
+
r_{t+2}
+
\cdots
+
r_T.
}
$$

The agent aims to maximize this quantity.

---

# <font color='orange'> 11. Discount Factor </font>

Future rewards are often considered less valuable than immediate rewards.

This is captured using the **discount factor**

$$
\gamma,
$$

where

$$
0\le\gamma\le1.
$$

The discounted return is

$$
\boxed{
G_t
=
r_{t+1}
+
\gamma r_{t+2}
+
\gamma^2 r_{t+3}
+
\cdots.
}
$$

Interpretation

- \( \gamma=0 \): only immediate rewards matter.
- \( \gamma\approx1 \): long-term rewards become important.

The discount factor balances short-term and long-term decision making.

---

# <font color='orange'> 12. Exploration vs Exploitation </font>

One of the defining challenges of Reinforcement Learning is the **exploration–exploitation trade-off**.

Suppose a robot has found a path that usually works.

Should it

- continue using that path (**exploitation**),

or

- explore alternative paths that might be even better (**exploration**)?

Too much exploitation may prevent discovering better strategies.

Too much exploration wastes time on poor actions.

A successful RL algorithm balances both.

# <font color='red'> 14. Mathematical Foundations </font>

At each time step,

the interaction follows

$$
\boxed{
(s_t,a_t,r_{t+1},s_{t+1}).
}
$$

The agent observes the current state

$$
s_t,
$$

chooses an action

$$
a_t,
$$

receives reward

$$
r_{t+1},
$$

and transitions to the next state

$$
s_{t+1}.
$$

The discounted return is

$$
\boxed{
G_t
=
\sum_{k=0}^{\infty}
\gamma^k
r_{t+k+1}.
}
$$

The overall objective of Reinforcement Learning is

$$
\boxed{
\text{Find a policy }
\pi
\text{ that maximizes }
\mathbb{E}[G_t].
}
$$

This optimization problem is the foundation of nearly every Reinforcement Learning algorithm.

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> Reinforcement Learning requires labelled data.

**False.**

The agent receives rewards rather than correct labels. It must discover effective actions through interaction with the environment.

---

### Misconception 2

> The agent should always maximize the immediate reward.

**False.**

A small immediate reward may lead to much larger future rewards. RL seeks to maximize the **expected cumulative reward**, not just the next reward.

---

### Misconception 3

> Exploration is always beneficial.

**False.**

Exploration helps discover better strategies, but excessive exploration can reduce performance by repeatedly trying poor actions. Effective RL balances exploration with exploitation.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Reinforcement Learning | Learning through interaction with an environment |
| Agent | Decision-maker that selects actions |
| Environment | System with which the agent interacts |
| State (\(s\)) | Current situation of the environment |
| Action (\(a\)) | Decision taken by the agent |
| Reward (\(r\)) | Feedback received after an action |
| Policy (\(\pi\)) | Strategy used by the agent |
| Episode | One complete interaction sequence |
| Return (\(G_t\)) | Total cumulative future reward |
| Discount Factor (\(\gamma\)) | Determines the importance of future rewards |
| Exploration | Trying new actions to gain information |
| Exploitation | Choosing actions known to produce good rewards |

> **Key Insight:** Reinforcement Learning differs from supervised learning by replacing labelled examples with rewards obtained through interaction. An agent repeatedly observes the current state, selects an action, receives a reward, and moves to a new state. Over time, it learns a policy that maximizes the expected cumulative reward, balancing immediate gains against long-term success. This framework forms the basis of modern algorithms used in robotics, game playing, autonomous systems, and language model alignment.
