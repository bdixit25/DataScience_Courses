# <font color='blue'> Chapter 52: Policy Gradient Methods </font>

Previous Reinforcement Learning algorithms learned value functions and then derived a policy from those values.

Policy Gradient methods take a different approach.

Instead of learning the value of states or actions,

they **learn the policy directly**.

The policy is represented by a parameterized function,

typically a neural network,

whose parameters are adjusted to maximize the expected cumulative reward.

Policy Gradient methods are particularly effective for problems with large or continuous action spaces,

where value-based methods become difficult to apply.

They form the foundation of many modern Reinforcement Learning algorithms,

including REINFORCE, Actor–Critic methods, PPO, and Soft Actor-Critic.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot arm must rotate by

```
37.4°
```

A Q-learning agent would have to discretize

every possible angle,

creating an enormous action space.

Instead,

a policy network can output

```
37.4°
```

directly.

Rather than searching over actions,

the policy learns to generate good actions.

---

# <font color='orange'> 2. Value-Based vs Policy-Based Methods </font>

Value-based methods

```
State

↓

Estimate Q-values

↓

Choose Best Action
```

Policy-based methods

```
State

↓

Policy Network

↓

Action
```

The policy itself becomes the object being optimized.

---

# <font color='orange'> 3. Parameterized Policies </font>

A policy is represented as

$$
\boxed{
\pi_\theta(a|s),
}
$$

where

- \(s\) is the current state,
- \(a\) is the chosen action,
- \(\theta\) denotes the parameters of the policy network.

The network outputs either

- probabilities over discrete actions, or
- parameters of a continuous probability distribution.

---

# <font color='orange'> 4. The Objective Function </font>

The goal is to maximize

the expected return obtained by following the policy.

This objective is written as

$$
\boxed{
J(\theta)
=
\mathbb{E}_{\pi_\theta}
\left[
G_0
\right].
}
$$

The learning problem becomes

```
Find θ

↓

Maximize Expected Return
```

Unlike supervised learning,

there are no target labels.

The objective is defined by the rewards received from the environment.

---

# <font color='orange'> 5. Why Gradients? </font>

Suppose the policy depends on parameters

$$
\theta.
$$

Small changes in

$$
\theta
$$

produce small changes in the policy,

which in turn change the expected reward.

Therefore,

we compute

$$
\nabla_\theta J(\theta),
$$

the gradient of the expected return with respect to the policy parameters.

The parameters are updated using **gradient ascent**,

because we wish to maximize,

rather than minimize,

the objective.

---

# <font color='orange'> 6. The Policy Gradient Theorem </font>

One of the most important results in Reinforcement Learning is the **Policy Gradient Theorem**.

It states that

$$
\boxed{
\nabla_\theta J(\theta)
=
\mathbb E
\left[
\nabla_\theta
\log
\pi_\theta(a|s)
\;
Q^{\pi}(s,a)
\right].
}
$$

This remarkable result allows the policy to be optimized without differentiating through the environment's dynamics.

The expectation is taken over trajectories generated by the current policy.

---

# <font color='orange'> 7. The REINFORCE Algorithm </font>

The simplest Policy Gradient algorithm is

**REINFORCE**.

Instead of using the action-value function,

it replaces

$$
Q^\pi(s,a)
$$

with the observed return

$$
G_t.
$$

The gradient estimate becomes

$$
\boxed{
\nabla_\theta J(\theta)
=
\mathbb E
\left[
\nabla_\theta
\log
\pi_\theta(a|s)
\;
G_t
\right].
}
$$

After each episode,

the policy parameters are adjusted in the direction that increases the probability of actions leading to high returns.

---

# <font color='orange'> 8. Gradient Ascent Update </font>

The policy parameters are updated using

$$
\boxed{
\theta
\leftarrow
\theta
+
\alpha
\nabla_\theta J(\theta),
}
$$

where

- \(\alpha\) is the learning rate.

Unlike supervised learning,

we move **up** the objective function because we seek to maximize reward.

---

# <font color='orange'> 9. Stochastic Policies </font>

Policy Gradient methods often use **stochastic policies**.

Instead of always selecting the same action,

the policy outputs probabilities.

Example

```
Left

0.15

Right

0.80

Up

0.03

Down

0.02
```

Sampling from these probabilities naturally provides exploration,

without requiring an external ε-greedy mechanism.

---

# <font color='orange'> 10. Advantages </font>

Policy Gradient methods

- naturally support continuous actions,
- learn stochastic policies,
- directly optimize the objective of interest,
- avoid searching over all possible actions.

These properties make them attractive for robotics and continuous control.

---

# <font color='orange'> 11. Limitations </font>

Pure Policy Gradient methods also have disadvantages.

They

- often exhibit high variance,
- may require many training episodes,
- can converge slowly,
- are sensitive to learning-rate choices.

These limitations motivate Actor–Critic methods,

which combine value estimation with policy optimization.

---

# <font color='orange'> 12. Applications </font>

Policy Gradient methods are widely used in

- robotics,
- autonomous vehicles,
- drone control,
- industrial process optimization,
- game playing,
- language model alignment,
- continuous control tasks.

They are particularly valuable when the action space is continuous rather than discrete.

---

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> Policy Gradient methods always perform better than Q-Learning.

**False.**

Policy Gradient methods excel in many continuous control problems, but Q-Learning and its variants often perform very well in discrete action spaces. The choice depends on the problem.

---

### Misconception 2

> Policy Gradient methods do not use neural networks.

**False.**

Modern Policy Gradient methods typically represent the policy using deep neural networks, enabling them to handle complex, high-dimensional inputs.

---

### Misconception 3

> REINFORCE is the algorithm used directly in most modern applications.

**False.**

REINFORCE is conceptually important but suffers from high-variance gradient estimates. Most practical systems use more advanced algorithms, such as Actor–Critic methods or PPO, which reduce variance and improve training stability.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Policy Gradient | Direct optimization of a parameterized policy |
| Policy Network | Neural network representing \(\pi_\theta(a|s)\) |
| Objective Function | Expected cumulative reward \(J(\theta)\) |
| Policy Gradient Theorem | Expresses the gradient of the expected return |
| REINFORCE | Monte Carlo Policy Gradient algorithm |
| Gradient Ascent | Optimization method for maximizing expected reward |
| Stochastic Policy | Outputs probabilities over actions rather than deterministic choices |
| Continuous Action Space | Action space with infinitely many possible actions |

> **Key Insight:** Policy Gradient methods differ fundamentally from value-based methods by optimizing the policy directly rather than first estimating value functions. By representing the policy with a parameterized function, typically a neural network, these methods can naturally handle continuous action spaces and stochastic decision-making. The Policy Gradient Theorem provides the mathematical basis for updating the policy in the direction of higher expected reward, forming the foundation of modern algorithms such as REINFORCE, Actor–Critic methods, and Proximal Policy Optimization.
