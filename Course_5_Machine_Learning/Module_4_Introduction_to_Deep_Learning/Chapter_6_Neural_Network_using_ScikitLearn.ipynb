So far, we have developed the mathematical foundations of artificial neural networks. We have learned how individual neurons perform computations and how these neurons are connected together into multilayer networks.

# <font color='orange'> 1. Why Start with Scikit-Learn? </font>

Before learning advanced deep learning frameworks, it is useful to begin with a simpler library.

Scikit-Learn automatically handles

- weight initialization,
- forward propagation,
- backpropagation,
- gradient descent,
- parameter updates.

This allows us to concentrate on understanding how neural networks behave rather than how they are implemented internally.

---

# <font color='orange'> 2. What is a Multi-Layer Perceptron (MLP)? </font>

The neural network implemented in Scikit-Learn is called a

**Multi-Layer Perceptron (MLP)**.

It consists of

- one input layer,
- one or more hidden layers,
- one output layer.

For example,

```
Input Layer

○ ○ ○ ○

      │

      ▼

Hidden Layer

○ ○ ○ ○ ○

      │

      ▼

Output Layer

○
```

Each neuron in one layer is connected to every neuron in the next layer.

Therefore,

the MLP is a **fully connected feedforward neural network**.

# <font color='orange'> 5. Creating an MLPClassifier </font>

A simple neural network can be created as follows.

```python
from sklearn.neural_network import MLPClassifier

model = MLPClassifier(
    hidden_layer_sizes=(100,),
    activation="relu",
    solver="adam",
    random_state=42
)
```

This creates

- one hidden layer,
- 100 neurons,
- ReLU activation,
- Adam optimizer.

---

# <font color='orange'> 6. Understanding the Parameters </font>

### hidden_layer_sizes

Determines

- number of hidden layers,
- number of neurons in each layer.

Examples

```python
hidden_layer_sizes=(50,)
```

One hidden layer

50 neurons.

---

```python
hidden_layer_sizes=(100,50)
```

Two hidden layers.

First layer

100 neurons.

Second layer

50 neurons.

---

```python
hidden_layer_sizes=(128,64,32)
```

Three hidden layers.

---

### activation

Specifies the activation function.

Options include

| Activation | Description |
|:---|:---|
| identity | Linear activation |
| logistic | Sigmoid activation |
| tanh | Hyperbolic tangent |
| relu | Rectified Linear Unit |

Modern deep learning almost always uses

```
relu
```

for hidden layers.

---

### solver

Determines the optimization algorithm.

Options include

| Solver | Description |
|:---|:---|
| lbfgs | Quasi-Newton optimizer |
| sgd | Stochastic Gradient Descent |
| adam | Adaptive Moment Estimation |

For most problems,

Adam provides the best balance between speed and performance.

---

### learning_rate_init

Controls the step size during optimization.

Typical value

```python
learning_rate_init=0.001
```

A learning rate that is too large

may cause unstable training.

A learning rate that is too small

may require many iterations.

---

### max_iter

Maximum number of training iterations.

Example

```python
max_iter=500
```

Training stops when

- convergence occurs,
- or the maximum number of iterations is reached.

# <font color='orange'> 11. Why Feature Scaling is Essential </font>

Neural networks perform best when features have similar numerical scales.

For example,

suppose

Age

ranges from

18

to

80,

while

Annual Income

ranges from

£20,000

to

£500,000.

Without scaling,

the larger feature dominates the weighted sums.

Therefore,

feature scaling is almost always required.

Typical preprocessing uses

```python
from sklearn.preprocessing import StandardScaler
```

or

```python
MinMaxScaler()
```

TensorFlow

Although Scikit-Learn hides the implementation,

$$
\boxed{
a=f(Wx+b).
}

The network repeatedly performs

1. Forward propagation,

2. Loss computation,

3. Backpropagation,
