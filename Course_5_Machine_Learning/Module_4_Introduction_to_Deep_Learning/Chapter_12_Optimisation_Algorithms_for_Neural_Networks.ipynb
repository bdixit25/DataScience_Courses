# <font color='blue'> Chapter 12: Optimization Algorithms for Neural Networks </font>

In the previous chapter, we learned that Gradient Descent updates the parameters of a neural network by moving in the direction of the negative gradient.

Although Gradient Descent is conceptually simple, it has several practical limitations.

For example,

- it may converge slowly,
- it may oscillate during training,
- it may become trapped on flat regions,
- it is sensitive to the learning rate.

To overcome these problems, researchers developed a variety of optimization algorithms.

These algorithms improve the speed, stability, and efficiency of training.

Today,

almost every modern neural network is trained using one of these optimizers.

---

# <font color='orange'> 1. Why Do We Need Better Optimizers? </font>

Suppose we are descending a mountain.

Standard Gradient Descent behaves like someone who

- looks only at the current slope,
- takes one step,
- stops,
- looks again,
- takes another step.

Although this eventually reaches the valley,

the journey can be very slow.

Modern optimization algorithms attempt to

- remember previous directions,
- adjust the step size automatically,
- avoid unnecessary oscillations.

---

# <font color='orange'> 2. Batch Gradient Descent </font>

Batch Gradient Descent computes the gradient using the **entire training dataset**.

Suppose the training set contains

$$
n
$$

examples.

The gradient is

$$
\boxed{
\nabla J
=
\frac1n
\sum_{i=1}^{n}
\nabla L_i.
}
$$

The update rule becomes

$$
\boxed{
\theta
=
\theta
-
\eta
\nabla J.
}
$$

### Advantages

- Stable convergence.
- Accurate gradient estimate.

### Limitations

- Very slow for large datasets.
- Requires loading the entire dataset into memory.

---

# <font color='orange'> 3. Stochastic Gradient Descent (SGD) </font>

Instead of computing the gradient using every observation,

SGD updates the parameters after **one training example**.

For one observation,

$$
\boxed{
\theta
=
\theta
-
\eta
\nabla L_i.
}
$$

Training therefore becomes much faster.

The updates are noisier,

but often escape poor local minima.

---

### Comparison

Batch GD

```
Dataset

↓

One Update
```

SGD

```
Example 1 → Update

Example 2 → Update

Example 3 → Update

...
```

---

# <font color='orange'> 4. Mini-Batch Gradient Descent </font>

Modern Deep Learning rarely uses

either

Batch GD

or

pure SGD.

Instead,

the dataset is divided into

small batches.

Typical batch sizes are

- 32
- 64
- 128
- 256

The update becomes

$$
\boxed{
\theta
=
\theta
-
\eta
\,
\nabla J_{\text{mini-batch}}.
}
$$

Mini-batch Gradient Descent combines

- the efficiency of SGD,
- the stability of Batch GD.

It is the default choice in almost every deep learning framework.

---

# <font color='orange'> 5. Momentum </font>

Gradient Descent forgets everything about previous updates.

Momentum introduces the idea of **velocity**.

Imagine pushing a heavy ball down a hill.

Once the ball starts rolling,

it continues moving even if the slope briefly becomes smaller.

The momentum update is

$$
\boxed{
v_t
=
\beta
v_{t-1}
+
(1-\beta)
\nabla J.
}
$$

The parameters become

$$
\boxed{
\theta
=
\theta
-
\eta
v_t.
}
$$

Typical value

$$
\beta=0.9.
$$

---

### Why Momentum Helps

Momentum

- accelerates learning,
- reduces oscillation,
- moves smoothly through shallow valleys.

---

# <font color='orange'> 6. Nesterov Momentum </font>

Momentum first moves,

then computes the gradient.

Nesterov Momentum

looks slightly ahead before computing the gradient.

Conceptually,

```
Current Position

↓

Look Ahead

↓

Compute Gradient

↓

Update
```

This often improves convergence.

---

# <font color='orange'> 7. AdaGrad </font>

Different parameters may require different learning rates.

AdaGrad automatically adapts the learning rate for every parameter.

The accumulated squared gradients are

$$
G_t
=
G_{t-1}
+
g_t^2.
$$

The update becomes

$$
\boxed{
\theta
=
\theta
-
\frac{\eta}
{\sqrt{G_t+\epsilon}}
g_t.
}
$$

Frequently updated parameters receive

smaller learning rates.

Rare parameters receive

larger learning rates.

AdaGrad is especially useful for sparse data.

---

# <font color='orange'> 8. RMSProp </font>

AdaGrad continually decreases the learning rate,

sometimes making it too small.

RMSProp solves this by keeping an exponentially weighted moving average.

$$
\boxed{
E[g^2]_t
=
\beta
E[g^2]_{t-1}
+
(1-\beta)
g_t^2.
}
$$

The update becomes

$$
\boxed{
\theta
=
\theta
-
\frac{\eta}
{\sqrt{E[g^2]_t+\epsilon}}
g_t.
}
$$

RMSProp remains one of the most popular optimizers,

especially for recurrent neural networks.

---

# <font color='orange'> 9. Adam Optimizer </font>

Adam

(**Adaptive Moment Estimation**)

combines

- Momentum,
- RMSProp.

It keeps track of

the first moment

(mean gradient)

and

the second moment

(squared gradient).

First moment

$$
m_t
=
\beta_1m_{t-1}
+
(1-\beta_1)g_t.
$$

Second moment

$$
v_t
=
\beta_2v_{t-1}
+
(1-\beta_2)g_t^2.
$$

Bias correction

$$
\hat m_t
=
\frac{m_t}
{1-\beta_1^t},
$$

$$
\hat v_t
=
\frac{v_t}
{1-\beta_2^t}.
$$

Final update

$$
\boxed{
\theta
=
\theta
-
\eta
\frac{\hat m_t}
{\sqrt{\hat v_t}+\epsilon}.
}
$$

Adam is currently the default optimizer for many deep learning applications.

---

# <font color='orange'> 10. AdamW </font>

Adam sometimes struggles with weight regularization.

AdamW separates

weight decay

from

gradient updates.

This usually improves generalization.

Modern Transformer models almost always use AdamW.

---

General recommendations are

| Problem | Recommended Optimizer |
|:---|:---|
| Beginner Projects | Adam |
| CNNs | Adam |
| RNNs | RMSProp or Adam |
| Transformers | AdamW |

# <font color='red'> 13. Mathematical Foundations </font>

Observe that every optimizer is still based on the same fundamental idea:

$$
\boxed{
\theta
\leftarrow
\theta
-
\eta
(\text{Gradient Information}).
}
$$

The difference lies in **how the gradient information is modified** before updating the parameters.

- Gradient Descent uses the current gradient.
- Momentum adds information from previous gradients.
- AdaGrad scales the gradient according to its history.
- RMSProp uses an exponentially weighted average of squared gradients.
- Adam combines momentum and adaptive learning rates.

Thus, all modern optimizers are refinements of the basic Gradient Descent algorithm rather than entirely new methods.

---

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> Adam is always the best optimizer.

**False.**

While Adam performs well in many applications, SGD with Momentum often achieves better generalization in some computer vision tasks.

---

### Misconception 2

> A larger batch size always leads to better training.

**False.**

Very large batches can reduce the stochasticity that helps optimization escape poor regions of the loss landscape and may require larger computational resources.

---

### Misconception 3

> Adaptive optimizers eliminate the need to tune the learning rate.

**False.**

Although optimizers such as Adam adapt the effective learning rate, the initial learning rate remains an important hyperparameter.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Batch Gradient Descent | Uses the entire dataset for each update |
| Stochastic Gradient Descent | Updates after each training example |
| Mini-Batch Gradient Descent | Uses small subsets of the data |
| Momentum | Adds velocity to parameter updates |
| AdaGrad | Adapts the learning rate using accumulated gradients |
| RMSProp | Prevents learning rates from shrinking excessively |
| Adam | Combines Momentum and RMSProp |
| AdamW | Adam with decoupled weight decay |

> **Key Insight:** All optimization algorithms seek to minimise the loss function by updating the network parameters using gradient information. They differ primarily in how they estimate, smooth, or scale the gradients. Modern optimizers such as Adam and AdamW build directly upon the principles of Gradient Descent, providing faster convergence and more stable training for deep neural networks.
