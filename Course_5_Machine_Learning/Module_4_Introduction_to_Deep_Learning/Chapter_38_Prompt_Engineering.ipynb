# <font color='blue'> Chapter 38: Prompt Engineering </font>

Large Language Models (LLMs) are powerful general-purpose systems capable of answering questions, writing code, summarizing documents, translating languages, and assisting with many other tasks.

However, the quality of an LLM's response depends heavily on **how the task is described**.

The process of designing clear, effective instructions for a language model is known as **Prompt Engineering**.

A well-designed prompt helps the model understand

- what task to perform,
- what information is available,
- what output format is expected,
- what constraints should be followed.

Prompt engineering therefore plays a crucial role in obtaining reliable and useful responses from modern LLMs.

---

# <font color='orange'> 1. What is a Prompt? </font>

A **prompt** is the input provided to a language model.

It may consist of

- a question,
- an instruction,
- contextual information,
- examples,
- formatting requirements.

Example

```
Summarize the following article in three bullet points.
```

The model uses the prompt as the context for predicting the next tokens.

---

# <font color='orange'> 2. Why Prompt Engineering Matters </font>

Consider two prompts.

### Prompt A

```
Explain machine learning.
```

### Prompt B

```
Explain machine learning to a first-year university student.

Use simple language.

Include one real-world example.

Limit the answer to 300 words.
```

Both prompts ask about the same topic,

but the second provides much clearer guidance.

Generally,

clearer instructions lead to responses that better match the user's intent.

---

# <font color='orange'> 3. Components of a Good Prompt </font>

A well-designed prompt usually contains

- the task,
- relevant context,
- constraints,
- the desired output format.

Example

```
Task:
Explain Principal Component Analysis.

Audience:
First-year physics students.

Constraints:
Use simple mathematics.

Output:
Markdown with equations.
```

Providing this information reduces ambiguity.

---

# <font color='orange'> 4. Zero-Shot Prompting </font>

In **Zero-Shot Prompting**,

the model receives only the task.

Example

```
Translate this sentence into French.

The weather is beautiful today.
```

No examples are provided.

The model relies entirely on its pretrained knowledge.

---

# <font color='orange'> 5. One-Shot Prompting </font>

One example is included before the actual task.

Example

```
English:
Hello

French:
Bonjour

English:
Thank you

French:
?
```

The example helps clarify the expected pattern.

---

# <font color='orange'> 6. Few-Shot Prompting </font>

Several examples are provided.

```
2 + 2 = 4

3 + 5 = 8

7 + 4 = 11

9 + 6 = ?
```

The model infers the relationship from the examples before producing the answer.

Few-shot prompting is particularly useful when the desired output format is unusual or highly structured.

---

# <font color='orange'> 7. Role Prompting </font>

The prompt may specify a role.

Examples

```
You are a university lecturer.
```

or

```
You are an experienced software engineer.
```

or

```
You are a scientific editor.
```

This guides the style and level of explanation,

without changing the underlying model.

---

# <font color='orange'> 8. Providing Context </font>

LLMs perform better when relevant context is supplied.

Instead of asking

```
Summarize this.
```

provide

```
Summarize the following research paper.

Focus on the methodology and results.

Ignore the literature review.
```

Context helps the model identify what information is most important.

---

# <font color='orange'> 9. Specifying the Output Format </font>

Explicitly requesting an output format often improves consistency.

Examples include

```
Return the answer as

- a Markdown table,

or

- JSON,

or

- bullet points.
```

Structured outputs are especially useful in programming and data processing workflows.

---

# <font color='orange'> 10. Prompt Templates </font>

Many tasks benefit from reusable templates.

Example

```
Task:

Context:

Constraints:

Output Format:
```

Templates make prompts easier to reproduce and modify.

---

# <font color='orange'> 11. Iterative Prompting </font>

Prompt engineering is often an iterative process.

Example

```
Initial Prompt

↓

Model Response

↓

Refined Prompt

↓

Improved Response
```

Rather than expecting the perfect answer immediately,

users often improve prompts based on previous responses.

---

# <font color='orange'> 12. Common Prompting Mistakes </font>

Examples include

- vague instructions,
- contradictory requirements,
- insufficient context,
- ambiguous terminology,
- unrealistic expectations.

For example,

```
Explain everything about AI.
```

is too broad.

A more focused prompt generally produces a more useful response.

---

# <font color='orange'> 13. Why Prompting Works </font>

Prompt engineering does **not** modify the model's parameters.

Instead,

the prompt changes the context that the model uses when predicting subsequent tokens.

Different prompts lead to different probability distributions over possible next tokens,

which explains why wording can significantly influence the generated response.

---

# <font color='orange'> 14. Applications </font>

Prompt engineering is widely used for

- programming assistance,
- scientific writing,
- mathematics,
- document summarization,
- tutoring,
- legal analysis,
- medical documentation,
- report generation,
- data analysis.

It has become an essential skill for working effectively with LLMs.

---

# <font color='red'> 15. Mathematical Foundations </font>

Suppose the prompt consists of tokens

$$
p_1,p_2,\ldots,p_m.
$$

The model predicts the next token according to

$$
\boxed{
P(x_{m+1}\mid p_1,p_2,\ldots,p_m).
}
$$

Once the next token is generated,

the context becomes

$$
(p_1,p_2,\ldots,p_m,x_{m+1}),
$$

and the process repeats.

Consequently,

different prompts create different contexts,

which lead to different probability distributions over future tokens.

This is the mathematical reason why prompt wording affects the model's responses.

---

# <font color='orange'> 16. Best Practices </font>

Effective prompts are generally

- clear,
- specific,
- unambiguous,
- appropriately detailed,
- realistic in scope,
- explicit about the desired output format.

Providing examples and relevant context often improves response quality.

---

# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> Prompt engineering is about discovering secret keywords.

**False.**

Effective prompting relies on clear communication, appropriate context, and well-defined objectives rather than hidden phrases.

---

### Misconception 2

> Longer prompts are always better.

**False.**

Additional detail is helpful only when it provides relevant information. Unnecessary or contradictory instructions can reduce response quality.

---

### Misconception 3

> Prompt engineering changes the model itself.

**False.**

Prompt engineering changes only the input context. The model's learned parameters remain unchanged.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Prompt | Input provided to an LLM |
| Prompt Engineering | Designing effective prompts to guide model behaviour |
| Zero-Shot Prompting | Task without examples |
| One-Shot Prompting | Task with one example |
| Few-Shot Prompting | Task with several examples |
| Role Prompting | Assigning a role to influence style and perspective |
| Context | Additional information supplied with the task |
| Output Format | Explicit specification of the desired response structure |
| Prompt Template | Reusable prompt structure for similar tasks |

> **Key Insight:** Prompt engineering is the practice of communicating effectively with a Large Language Model by providing clear instructions, relevant context, and well-defined output requirements. Rather than changing the model itself, a prompt shapes the context used during next-token prediction, influencing the probability distribution over possible responses. Well-designed prompts consistently produce more accurate, structured, and useful outputs.
