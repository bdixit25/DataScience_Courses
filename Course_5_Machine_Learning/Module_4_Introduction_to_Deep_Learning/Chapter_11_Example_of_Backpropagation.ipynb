# <font color='blue'> Chapter 11: A Complete Worked Example of Backpropagation </font>

In the previous chapter, we derived the Backpropagation algorithm using the Chain Rule. Although the mathematics is elegant, it can appear abstract until we apply it to a real neural network.

In this chapter, we shall work through one complete training iteration of a small neural network by hand. We shall

- perform forward propagation,
- compute the loss,
- calculate every gradient,
- update every weight,
- verify that the loss decreases.

This example demonstrates how all of the concepts introduced in previous chapters fit together.

---

# <font color='orange'> 1. The Neural Network </font>

Consider the following neural network.

```
            Hidden Layer

          h₁         h₂

         ○──────────○

        / \        / \

       /   \      /   \

      /     \    /     \

 x₁ ○         \/         ○ Output

      \       /\

       \     /  \

        \   /    \

         \ /      \

          ○────────○

          h₂
```

The network consists of

- 2 input neurons,
- 2 hidden neurons,
- 1 output neuron.

All neurons use the **Sigmoid activation function**.

---

# <font color='orange'> 2. Initial Parameters </font>

Inputs

$$
x_1=0.05,
\qquad
x_2=0.10.
$$

Target

$$
y=0.01.
$$

Hidden-layer weights

$$
W^{(1)}
=
\begin{bmatrix}
0.15 & 0.20\\
0.25 & 0.30
\end{bmatrix}.
$$

Hidden-layer biases

$$
b^{(1)}
=
\begin{bmatrix}
0.35\\
0.35
\end{bmatrix}.
$$

Output-layer weights

$$
W^{(2)}
=
\begin{bmatrix}
0.40 & 0.45
\end{bmatrix}.
$$

Output-layer bias

$$
b^{(2)}=0.60.
$$

These values are the famous example introduced by Matt Mazur and are widely used to illustrate backpropagation.

---

# <font color='orange'> 3. Forward Propagation </font>

## Hidden Neuron 1

Weighted sum

$$
z_1
=
0.05(0.15)
+
0.10(0.20)
+
0.35
=
0.3775.
$$

Activation

$$
a_1
=
\sigma(0.3775)
=
0.59327.
$$

---

## Hidden Neuron 2

$$
z_2
=
0.05(0.25)
+
0.10(0.30)
+
0.35
=
0.3925.
$$

Activation

$$
a_2
=
\sigma(0.3925)
=
0.59688.
$$

---

## Output Neuron

Weighted sum

$$
z_3
=
0.40(0.59327)
+
0.45(0.59688)
+
0.60
=
1.1059.
$$

Output

$$
\hat y
=
\sigma(1.1059)
=
0.75136.
$$

The network predicts

$$
0.75136,
$$

whereas the correct answer is

$$
0.01.
$$

Clearly,

the prediction is poor.

---

# <font color='orange'> 4. Compute the Loss </font>

Using

Mean Squared Error,

$$
L
=
\frac12
(y-\hat y)^2.
$$

Substitute the values,

$$
L
=
\frac12
(0.01-0.75136)^2
=
0.2748.
$$

This is the quantity we wish to minimise.

---

# <font color='orange'> 5. Compute the Output Gradient </font>

We first compute

$$
\frac{\partial L}{\partial \hat y}.
$$

Differentiate,

$$
\frac{\partial L}{\partial \hat y}
=
\hat y-y.
$$

Substitute,

$$
=
0.75136-0.01
=
0.74136.
$$

---

Next,

differentiate the sigmoid,

$$
\frac{\partial\hat y}{\partial z}
=
\hat y
(1-\hat y).
$$

Substitute,

$$
=
0.75136
(1-0.75136)
=
0.18682.
$$

Therefore,

the output error becomes

$$
\boxed{
\delta_{\rm output}
=
0.74136
\times
0.18682
=
0.13850.
}
$$

---

# <font color='orange'> 6. Compute the Output Weight Gradients </font>

Recall

$$
\frac{\partial z}{\partial w}=a.
$$

Therefore,

$$
\frac{\partial L}
{\partial w_5}
=
0.13850
\times
0.59327
=
0.08217.
$$

Similarly,

$$
\frac{\partial L}
{\partial w_6}
=
0.13850
\times
0.59688
=
0.08267.
$$

---

# <font color='orange'> 7. Update the Output Weights </font>

Suppose

$$
\eta=0.5.
$$

Gradient Descent gives

$$
w_{\rm new}
=
w-\eta
\frac{\partial L}{\partial w}.
$$

For

$$
w_5,
$$

$$
0.40
-
0.5(0.08217)
=
0.35892.
$$

Similarly,

$$
w_6
=
0.40866.
$$

Notice

both weights become smaller,

reducing the overly large prediction.

---

# <font color='orange'> 8. Hidden Layer Gradients </font>

Unlike the output layer,

the hidden neurons influence the loss only indirectly.

We therefore apply the Chain Rule.

For hidden neuron 1,

$$
\delta_1
=
\delta_{\rm output}
w_5
a_1
(1-a_1).
$$

Substituting,

$$
=
0.13850
\times
0.40
\times
0.59327
\times
0.40673
=
0.01337.
$$

Similarly,

hidden neuron 2 gives

$$
\delta_2
=
0.01487.
$$

These values represent the error signals propagated backwards.

---

# <font color='orange'> 9. Update the Hidden Weights </font>

For

$$
w_1,
$$

$$
\frac{\partial L}{\partial w_1}
=
\delta_1x_1.
$$

Substituting,

$$
=
0.01337
\times
0.05
=
0.000668.
$$

Therefore,

$$
w_1
=
0.15
-
0.5
(0.000668)
=
0.14967.
$$

The remaining hidden weights are updated in exactly the same way.

---

# <font color='orange'> 10. Verify the Improvement </font>

Using the updated weights,

perform another forward propagation.

The prediction moves closer to the target,

and the loss decreases.

For this example,

the loss changes from

$$
0.2748
$$

to approximately

$$
0.271.
$$

Although the improvement after one iteration is small,

repeating this process thousands of times allows the network to learn.

---

# <font color='orange'> 11. What Happened? </font>

The complete learning process was

```
Inputs

↓

Forward Propagation

↓

Prediction

↓

Loss

↓

Backpropagation

↓

Gradients

↓

Gradient Descent

↓

Updated Weights

↓

Repeat
```

Every training iteration follows exactly this sequence.

---

# <font color='orange'> 12. Why This Example Matters </font>

Although the network contains only

- two inputs,
- two hidden neurons,
- one output neuron,

the same algorithm scales naturally to

- thousands of neurons,
- hundreds of layers,
- billions of parameters.

TensorFlow and PyTorch perform exactly the same calculations,

only much faster using vectorised operations and GPUs.

---

# <font color='red'> 13. Mathematical Foundations </font>

Notice that every gradient was obtained using the same principle.

For example,

$$
\frac{\partial L}{\partial w_5}
=
\frac{\partial L}{\partial\hat y}
\cdot
\frac{\partial\hat y}{\partial z_3}
\cdot
\frac{\partial z_3}{\partial w_5}.
$$

Similarly,

the hidden weights required one additional application of the Chain Rule,

because the hidden neurons affect the loss indirectly through the output neuron.

This repeated application of the Chain Rule is precisely what Backpropagation does automatically.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Step | Operation |
|:---|:---|
| 1 | Initialize weights and biases |
| 2 | Perform forward propagation |
| 3 | Compute the prediction |
| 4 | Calculate the loss |
| 5 | Compute the output-layer gradients |
| 6 | Propagate the error backwards |
| 7 | Compute the hidden-layer gradients |
| 8 | Update all parameters using Gradient Descent |
| 9 | Repeat until the loss converges |

> **Key Insight:** Backpropagation is not a mysterious algorithm. It is simply the systematic application of the Chain Rule to every parameter in the network. By repeatedly performing forward propagation, computing the loss, propagating the error backwards, and updating the parameters, the network gradually learns to make increasingly accurate predictions.
