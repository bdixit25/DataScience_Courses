# <font color='blue'> Chapter 51: Deep Q Networks (DQN) </font>

Classical Q-Learning stores one Q-value for every state-action pair in a lookup table.

Although effective for small environments,

this approach becomes impractical when the state space is very large or continuous.

For example,

an autonomous vehicle observing camera images may encounter billions of distinct states.

Deep Q Networks (DQN) solve this problem by replacing the Q-table with a **Deep Neural Network** that approximates the action-value function.

Instead of storing every Q-value explicitly,

the network learns a function that predicts

$$
Q(s,a)
$$

for any state-action pair.

This combination of Deep Learning and Reinforcement Learning marked a major breakthrough in Artificial Intelligence.

---



# <font color='orange'> 1. Motivation </font>

Suppose an Atari game has

```
210 × 160
```

pixel images.

Even a single frame contains tens of thousands of pixel values.

A Q-table would require an entry for every possible image,

which is impossible.

Instead,

we use a neural network

```
Image

↓

Neural Network

↓

Q-values
```

The network learns to generalize across similar states.

---



# <font color='orange'> 2. From Q-Tables to Neural Networks </font>

Classical Q-Learning stores

```
(s,a)

↓

Q-value
```

DQN instead learns a function

$$
\boxed{
Q(s,a;\theta),
}
$$

where

- \(s\) is the state,
- \(a\) is the action,
- \(\theta\) represents the neural network parameters (weights and biases).

The network acts as a function approximator.

---



# <font color='orange'> 3. Q-Network Architecture </font>

The input to the network is the current state.

```
State

↓

Neural Network

↓

Q(Left)

Q(Right)

Q(Up)

Q(Down)
```

Each output corresponds to the estimated Q-value for one possible action.

The agent selects the action with the largest predicted Q-value.

---



# <font color='orange'> 4. Why Doesn't Ordinary Q-Learning Work? </font>

If we simply replace the Q-table with a neural network,

training often becomes unstable.

This happens because

- consecutive experiences are highly correlated,
- the learning targets change as the network updates,
- neural networks assume approximately independent training samples.

DQN introduces two key innovations to address these issues:

- **Experience Replay**
- **Target Networks**

---



# <font color='orange'> 5. Experience Replay </font>

Instead of learning immediately from each experience,

the agent stores experiences in a replay buffer.

Each experience consists of

$$
\boxed{
(s,a,r,s').
}
$$

```
Environment

↓

Experience

↓

Replay Buffer

↓

Random Mini-batch

↓

Neural Network
```

Random sampling breaks the correlation between consecutive experiences,

leading to more stable training.

---



# <font color='orange'> 6. Replay Buffer </font>

The replay buffer stores thousands or millions of past experiences.

Example

```
Experience 1

Experience 2

Experience 3

...

Experience N
```

During training,

mini-batches are sampled uniformly at random.

Older experiences are eventually removed as new ones are added.

---



# <font color='orange'> 7. Target Network </font>

Another problem is that the target values change continuously as the network learns.

To stabilize learning,

DQN introduces a second neural network,

called the **target network**.

```
Online Network

↓

Learning

↓

Copy Parameters

↓

Target Network
```

The target network remains fixed for several training steps before being updated.

This provides more stable learning targets.

---



# <font color='orange'> 8. DQN Training Pipeline </font>

A single training iteration proceeds as follows.

```
Current State

↓

Choose Action

↓

Environment

↓

Reward + Next State

↓

Replay Buffer

↓

Sample Mini-batch

↓

Compute Targets

↓

Gradient Descent

↓

Update Online Network
```

Periodically,

the online network parameters are copied to the target network.

---



# <font color='orange'> 9. Target Value </font>

For each sampled transition,

the target is

$$
\boxed{
y
=
r
+
\gamma
\max_{a'}
Q_{\theta^-}(s',a'),
}
$$

where

- \(\theta^-\) denotes the parameters of the target network.

Notice that

the target network,

not the online network,

is used to compute the target value.

---



# <font color='orange'> 10. Loss Function </font>

The online network predicts

$$
Q_\theta(s,a).
$$

The prediction is compared with the target.

The loss is

$$
\boxed{
L(\theta)
=
\left(
y
-
Q_\theta(s,a)
\right)^2.
}
$$

Gradient descent adjusts the network parameters to minimize this loss.

---



# <font color='orange'> 11. ε-Greedy Exploration </font>

During training,

the agent continues to explore.

With probability

$$
1-\varepsilon,
$$

choose

the action with the highest predicted Q-value.

With probability

$$
\varepsilon,
$$

choose

a random action.

As training progresses,

$$
\varepsilon
$$

is often reduced,

allowing the agent to exploit its learned policy more frequently.

---



# <font color='orange'> 12. The Atari Breakthrough </font>

One of the most famous demonstrations of DQN was learning to play Atari games directly from raw pixel inputs.

The agent received only

- the game screen,
- the game score.

No handcrafted features or game-specific rules were provided.

The same algorithm learned many different games,

demonstrating that Deep Reinforcement Learning could learn directly from high-dimensional sensory data.

---



# <font color='orange'> 13. Limitations of DQN </font>

Although highly successful,

DQN has several limitations.

It

- tends to overestimate Q-values,
- works best with discrete action spaces,
- can require large amounts of training data,
- may be sensitive to hyperparameter choices.

These limitations motivated later improvements such as

- Double DQN,
- Dueling DQN,
- Prioritized Experience Replay.

---



# <font color='orange'> 14. Applications </font>

Deep Q Networks have been applied to

- video game playing,
- robotics,
- warehouse automation,
- traffic signal optimization,
- resource scheduling,
- autonomous navigation.

They demonstrate how deep neural networks can scale Reinforcement Learning to high-dimensional problems.

---

# <font color='red'> 15. Mathematical Foundations </font>

The Q-network approximates

$$
\boxed{
Q(s,a;\theta).
}
$$

The target value is

$$
\boxed{
y
=
r
+
\gamma
\max_{a'}
Q(s',a';\theta^-),
}
$$

where

$$
\theta^-
$$

are the target-network parameters.

The loss function is

$$
\boxed{
L(\theta)
=
\left(
y
-
Q(s,a;\theta)
\right)^2.
}
$$

Gradient descent updates the parameters

$$
\theta
$$

to minimize this loss.

Periodically,

the target network parameters are updated

$$
\boxed{
\theta^-
\leftarrow
\theta.
}
$$

These mechanisms stabilize learning when neural networks are used to approximate Q-values.

---



# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> DQN stores a large Q-table internally.

**False.**

DQN replaces the Q-table with a neural network that approximates the action-value function.

---

### Misconception 2

> Experience Replay improves performance by increasing the amount of data.

**False.**

Its primary purpose is to break correlations between consecutive experiences and improve sample efficiency by reusing past interactions.

---

### Misconception 3

> The online and target networks are identical at every training step.

**False.**

The online network is updated continuously, whereas the target network is updated only periodically to provide stable learning targets.

---

# <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Deep Q Network (DQN) | Uses a neural network to approximate the Q-function |
| Q-Network | Predicts Q-values for actions given a state |
| Function Approximation | Replaces the tabular Q-function with a neural network |
| Experience Replay | Stores and randomly samples past experiences |
| Replay Buffer | Memory containing previous transitions |
| Target Network | Provides stable targets during training |
| DQN Loss | Mean squared error between target and predicted Q-values |
| ε-Greedy Exploration | Balances exploration with exploitation |

> **Key Insight:** Deep Q Networks extend classical Q-Learning by replacing the Q-table with a deep neural network capable of approximating action values in large state spaces. Experience Replay reduces correlations between training samples, while a Target Network stabilizes learning by providing slowly changing targets. Together, these innovations enabled Reinforcement Learning agents to solve complex, high-dimensional tasks such as playing Atari games directly from raw pixel inputs, establishing Deep Reinforcement Learning as a major field of modern AI.
