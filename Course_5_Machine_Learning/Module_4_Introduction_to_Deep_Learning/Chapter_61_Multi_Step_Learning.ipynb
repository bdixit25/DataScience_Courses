# <font color='blue'> Chapter 61: Multi-Step Learning (n-Step Returns) </font>

Temporal Difference Learning updates value estimates using only the immediate reward and the estimated value of the next state.

Although efficient,

this approach propagates reward information only one step at a time.

In environments where rewards are delayed,

learning can therefore be very slow.

Multi-Step Learning addresses this problem by allowing the agent to accumulate rewards over several future steps before updating its value estimates.

This combines the strengths of

- Temporal Difference Learning,
- Monte Carlo methods.

The result is faster reward propagation,

improved sample efficiency,

and better learning in many complex environments.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot must reach a goal located

20 steps away.

```
Start

↓

↓

↓

↓

↓

Goal (+100)
```

With one-step TD learning,

the reward moves backward

only one state

per update.

Learning therefore requires many episodes.

Instead,

suppose we use

the next

5

rewards simultaneously.

Reward information now travels much faster through the state space.

---

# <font color='orange'> 2. One-Step Temporal Difference Learning </font>

Recall the TD target

$$
\boxed{
G_t^{(1)}
=
R_{t+1}
+
\gamma
V(S_{t+1}).
}
$$

Only

one reward

is observed before bootstrapping.

This produces low variance,

but slower reward propagation.

---

# <font color='orange'> 3. Monte Carlo Learning </font>

Monte Carlo methods wait

until the episode finishes.

The target becomes

$$
\boxed{
G_t
=
R_{t+1}
+
\gamma
R_{t+2}
+
\gamma^2
R_{t+3}
+\cdots.
}
$$

This uses

the complete return,

eliminating bootstrapping.

Although unbiased,

the variance can become large,

especially in long episodes.

---

# <font color='orange'> 4. The Main Idea of n-Step Learning </font>

Instead of using

one reward

or

the complete episode,

we use

exactly

$$
n
$$

future rewards.

After

$$
n
$$

steps,

bootstrapping begins.

Thus,

n-step learning lies between

Temporal Difference Learning

and

Monte Carlo methods.

---

# <font color='orange'> 5. n-Step Return </font>

The n-step return is

$$
\boxed{
G_t^{(n)}
=
R_{t+1}
+
\gamma
R_{t+2}
+
\cdots
+
\gamma^{n-1}
R_{t+n}
+
\gamma^n
V(S_{t+n}).
}
$$

Notice

the first

$$
n
$$

rewards

are observed directly.

Only after that

does bootstrapping occur.

---

# <font color='orange'> 6. Special Cases </font>

If

$$
n=1,
$$

we recover

ordinary TD learning.

$$
\boxed{
G_t^{(1)}
=
R_{t+1}
+
\gamma
V(S_{t+1}).
}
$$

If

$$
n
$$

equals

the episode length,

bootstrapping disappears,

giving

Monte Carlo learning.

Thus,

n-step learning forms

a continuous bridge

between TD learning

and

Monte Carlo methods.

---

# <font color='orange'> 7. Bias–Variance Trade-off </font>

Choosing

$$
n
$$

changes the balance between

bias

and

variance.

Small

$$
n
$$

- more bootstrapping,
- higher bias,
- lower variance.

Large

$$
n
$$

- less bootstrapping,
- lower bias,
- higher variance.

Selecting

an appropriate value of

$$
n
$$

is therefore an important design choice.

---

# <font color='orange'> 8. Faster Reward Propagation </font>

Suppose

a reward appears

10

steps after the current state.

One-step TD

```
Reward

↓

↓

↓

↓

Slow
```

Five-step learning

```
Reward

↓↓↓↓↓

Much Faster
```

Multi-step returns allow distant rewards to influence earlier decisions more quickly.

---

# <font color='orange'> 9. Multi-Step Q-Learning </font>

The same idea applies

to action-value functions.

The n-step target becomes

$$
\boxed{
G_t^{(n)}
=
\sum_{k=0}^{n-1}
\gamma^k
R_{t+k+1}
+
\gamma^n
\max_a
Q(S_{t+n},a).
}
$$

The Q-function is then updated

using this target

instead of

the one-step TD target.

---

# <font color='orange'> 10. Applications </font>

Multi-step learning is widely used in

- DQN,
- Rainbow DQN,
- Actor–Critic methods,
- PPO,
- modern robotics,
- game playing.

Many state-of-the-art Reinforcement Learning algorithms employ n-step returns because they improve learning speed.

---

# <font color='orange'> 11. Advantages </font>

Multi-step learning

- accelerates reward propagation,
- improves sample efficiency,
- often converges faster,
- provides a flexible balance between TD learning and Monte Carlo methods.

---

# <font color='orange'> 12. Limitations </font>

Choosing

$$
n
$$

requires experimentation.

If

$$
n
$$

is too small,

learning may remain slow.

If

$$
n
$$

is too large,

variance increases,

reducing learning stability.

The optimal value depends on the environment and task.

---

# <font color='orange'> 13. Relationship to Rainbow DQN </font>

Rainbow DQN combines

- Double DQN,
- Dueling Networks,
- Prioritized Experience Replay,
- Distributional RL,
- Noisy Networks,
- Multi-Step Learning.

Multi-Step Learning contributes

by improving

how rewards are propagated

through the value function.

---

# <font color='red'> 14. Mathematical Foundations </font>

### One-Step TD Target

$$
\boxed{
G_t^{(1)}
=
R_{t+1}
+
\gamma
V(S_{t+1}).
}
$$

---

### Monte Carlo Return

$$
\boxed{
G_t
=
\sum_{k=0}^{T-t-1}
\gamma^k
R_{t+k+1}.
}
$$

where

$$
T
$$

is the terminal time step.

---

### General n-Step Return

$$
\boxed{
G_t^{(n)}
=
\sum_{k=0}^{n-1}
\gamma^k
R_{t+k+1}
+
\gamma^n
V(S_{t+n}).
}
$$

For action-value methods,

the target becomes

$$
\boxed{
G_t^{(n)}
=
\sum_{k=0}^{n-1}
\gamma^k
R_{t+k+1}
+
\gamma^n
\max_a
Q(S_{t+n},a).
}
$$

The update rule remains

$$
\boxed{
Q(S_t,A_t)
\leftarrow
Q(S_t,A_t)
+
\alpha
\left(
G_t^{(n)}
-
Q(S_t,A_t)
\right).
}
$$

---

# <font color='orange'> 15. One-Step TD vs n-Step vs Monte Carlo </font>

| Method | Uses Bootstrapping | Rewards Used | Bias | Variance |
|:---|:---:|:---:|:---:|:---:|
| One-Step TD | ✓ | 1 | Higher | Lower |
| n-Step Learning | ✓ | n | Medium | Medium |
| Monte Carlo | ✗ | Entire Episode | Lower | Higher |

n-step learning provides a practical compromise between the two extremes.

---

# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> Multi-step learning completely replaces bootstrapping.

**False.**

Bootstrapping still occurs after \(n\) observed rewards. Only Monte Carlo methods eliminate bootstrapping entirely.

---

### Misconception 2

> Increasing \(n\) always improves performance.

**False.**

Larger values of \(n\) reduce bias but increase variance. There is no universally optimal choice of \(n\).

---

### Misconception 3

> Multi-step learning only applies to DQN.

**False.**

The idea of n-step returns can be incorporated into many reinforcement learning algorithms, including value-based methods, Actor–Critic methods, and policy optimization algorithms.

---

# <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| One-Step TD | Uses one observed reward before bootstrapping |
| Monte Carlo | Uses the complete return without bootstrapping |
| n-Step Return | Uses \(n\) observed rewards before bootstrapping |
| Reward Propagation | Process by which future rewards influence earlier decisions |
| Bias–Variance Trade-off | Larger \(n\) reduces bias but increases variance |
| Multi-Step Learning | General framework connecting TD learning and Monte Carlo methods |
| Rainbow DQN | Uses n-step returns to improve learning efficiency |

> **Key Insight:** Multi-Step Learning generalizes both Temporal Difference Learning and Monte Carlo methods by combining observed rewards over several future steps before bootstrapping. By allowing reward information to propagate more quickly through the environment, n-step returns improve sample efficiency and accelerate learning, particularly in environments with delayed or sparse rewards. The choice of \(n\) controls the balance between bias and variance, making multi-step learning a flexible and powerful component of modern deep reinforcement learning algorithms such as Rainbow DQN.
