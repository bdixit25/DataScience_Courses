# <font color='blue'> Chapter 13: Activation Functions </font>

In the previous chapters, we learned that every neuron computes two operations:

1. A **linear transformation**
2. A **nonlinear activation function**

The activation function is one of the most important components of a neural network. It determines how information flows through the network and allows the network to learn complex nonlinear relationships.

Without activation functions, a deep neural network would behave exactly like a single linear model, regardless of the number of layers.

This chapter introduces the most widely used activation functions in modern Deep Learning, explains their mathematical properties, derives their derivatives, and discusses their practical applications.

---



# <font color='orange'> 1. Why Do We Need Activation Functions? </font>

Recall that a neuron first computes

$$
z = \mathbf{w}^T\mathbf{x}+b.
$$

The activation function then produces

$$
a=f(z).
$$

The complete neuron is therefore

$$
\boxed{
a=f(\mathbf{w}^T\mathbf{x}+b).
}
$$

The activation function transforms the weighted sum into the neuron's output.

---



# <font color='orange'> 2. What Happens Without an Activation Function? </font>

Suppose every neuron simply outputs

$$
a=z.
$$

Consider two consecutive layers.

First layer

$$
a_1=W_1x+b_1.
$$

Second layer

$$
a_2=W_2a_1+b_2.
$$

Substituting,

$$
a_2
=
W_2(W_1x+b_1)+b_2.
$$

Expanding,

$$
a_2
=
(W_2W_1)x
+
(W_2b_1+b_2).
$$

Define

$$
W=W_2W_1,
$$

and

$$
b=W_2b_1+b_2.
$$

Then

$$
\boxed{
a_2
=
Wx+b.
}
$$

This is simply another linear transformation.

Therefore,

stacking linear layers never produces a nonlinear model.

---



# <font color='orange'> 3. Characteristics of a Good Activation Function </font>

An effective activation function should ideally

- introduce nonlinearity,
- be differentiable,
- be computationally efficient,
- allow stable gradient flow,
- avoid vanishing gradients whenever possible.

Unfortunately,

no single activation function satisfies every requirement.

---



# <font color='orange'> 4. Sigmoid Function </font>

The sigmoid function is

$$
\boxed{
\sigma(z)
=
\frac1{1+e^{-z}}.
}
$$

Derivative

$$
\boxed{
\sigma'(z)
=
\sigma(z)
\left(
1-\sigma(z)
\right).
}
$$

### Properties

Output range

$$
(0,1).
$$

Advantages

- smooth,
- differentiable,
- probability interpretation.

Limitations

- vanishing gradients,
- not zero-centered,
- slow learning.

Typical use

Binary classification output layer.

---



# <font color='orange'> 5. Hyperbolic Tangent (tanh) </font>

Definition

$$
\boxed{
\tanh(z)
=
\frac{e^z-e^{-z}}
{e^z+e^{-z}}.
}
$$

Derivative

$$
\boxed{
\frac{d}{dz}\tanh(z)
=
1-\tanh^2(z).
}
$$

Range

$$
(-1,1).
$$

Advantages

- zero-centered,
- stronger gradients near zero than sigmoid.

Limitations

Still suffers from vanishing gradients.

---



# <font color='orange'> 6. ReLU (Rectified Linear Unit) </font>

Definition

$$
\boxed{
ReLU(z)
=
\max(0,z).
}
$$

Derivative

$$
\boxed{
ReLU'(z)
=
\begin{cases}
0,&z<0\\
1,&z>0
\end{cases}
}
$$

Advantages

- extremely simple,
- computationally efficient,
- avoids vanishing gradients for positive inputs,
- sparse activations.

Disadvantages

Neurons may become permanently inactive.

This is called the

**Dead ReLU Problem**.

---



# <font color='orange'> 7. Leaky ReLU </font>

Definition

$$
\boxed{
f(z)
=
\begin{cases}
z,&z>0\\
\alpha z,&z<0
\end{cases}
}
$$

Typical value

$$
\alpha=0.01.
$$

Advantages

- solves Dead ReLU,
- simple,
- computationally inexpensive.

---



# <font color='orange'> 8. ELU (Exponential Linear Unit) </font>

Definition

$$
\boxed{
ELU(z)
=
\begin{cases}
z,&z>0\\
\alpha(e^z-1),&z<0
\end{cases}
}
$$

Advantages

- smoother than ReLU,
- negative outputs improve optimization.

Disadvantages

Slightly slower due to exponential computation.

---



# <font color='orange'> 9. Softplus </font>

Definition

$$
\boxed{
Softplus(z)
=
\ln(1+e^z).
}
$$

Derivative

$$
\boxed{
Softplus'(z)
=
\sigma(z).
}
$$

Softplus is essentially

a smooth approximation of ReLU.

---



# <font color='orange'> 10. Swish </font>

Proposed by Google,

Swish is

$$
\boxed{
Swish(z)
=
z\sigma(z).
}
$$

Advantages

- smooth,
- non-monotonic,
- often outperforms ReLU in deep networks.

---



# <font color='orange'> 11. GELU (Gaussian Error Linear Unit) </font>

The GELU activation is defined as

$$
\boxed{
GELU(z)
=
z\Phi(z),
}
$$

where

$$
\Phi(z)
$$

is the cumulative distribution function (CDF) of the standard normal distribution.

A common approximation is

$$
GELU(z)
\approx
0.5z
\left(
1+
\tanh
\left(
\sqrt{\frac2\pi}
(z+0.044715z^3)
\right)
\right).
$$

Advantages

- smooth,
- excellent gradient flow,
- state-of-the-art performance.

Applications

- BERT
- GPT
- Vision Transformers

---



# <font color='orange'> 12. Softmax </font>

Unlike previous activations,

Softmax operates on an entire vector.

Suppose

$$
z=
(z_1,z_2,\ldots,z_K).
$$

Then

$$
\boxed{
Softmax(z_i)
=
\frac{e^{z_i}}
{\sum_{j=1}^{K}e^{z_j}}.
}
$$

Properties

- outputs probabilities,
- probabilities sum to one,
- used for multi-class classification.

---

# <font color='orange'> 14. Which Activation Function Should I Use? </font>

| Problem | Recommended Activation |
|:---|:---|
| Regression Output | Linear |
| Binary Classification | Sigmoid |
| Multi-Class Classification | Softmax |
| Hidden Layers (General) | ReLU |
| Very Deep Networks | GELU or Swish |
| Transformers | GELU |
| Older Networks | tanh |

---



# <font color='orange'> 15. Computational Complexity </font>

Most activation functions operate independently on each neuron.

For

$$
n
$$

neurons,

the computational complexity is

$$
\boxed{
\mathcal{O}(n).
}
$$

The dominant computational cost during training remains the matrix multiplication,

not the activation function.

---

# <font color='red'> 16. Mathematical Foundations </font>

The activation function must be differentiable (or almost everywhere differentiable) because Backpropagation requires computing

$$
\frac{\partial L}{\partial z}.
$$

Using the Chain Rule,

$$
\boxed{
\frac{\partial L}{\partial z}
=
\frac{\partial L}{\partial a}
\cdot
f'(z).
}
$$

Thus,

the derivative of the activation function directly affects how the gradients propagate through the network.

If

$$
f'(z)
$$

is close to zero,

the gradients may vanish.

This is one of the primary reasons why ReLU and GELU have largely replaced sigmoid and tanh in deep networks.

---



# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> More complicated activation functions are always better.

**False.**

ReLU remains the default choice for many applications because of its simplicity and effectiveness.

---

### Misconception 2

> Sigmoid should never be used.

**False.**

Sigmoid is still the standard activation for the output layer in binary classification.

---

### Misconception 3

> Softmax can replace sigmoid in binary classification.

**Not usually.**

Although a two-class Softmax is mathematically possible, a single sigmoid output is simpler, more efficient, and is the standard approach for binary classification.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Activation Function | Introduces nonlinearity into the network |
| Sigmoid | Binary probability output |
| tanh | Zero-centered nonlinear activation |
| ReLU | Most common hidden-layer activation |
| Leaky ReLU | Addresses the Dead ReLU problem |
| ELU | Smooth negative activations |
| Swish | Google-developed smooth activation |
| GELU | Standard activation in Transformers |
| Softmax | Converts logits into class probabilities |

> **Key Insight:** Activation functions transform simple linear neurons into powerful nonlinear models capable of learning highly complex relationships. The choice of activation function influences gradient flow, training stability, computational efficiency, and predictive performance. While ReLU remains the standard activation for most hidden layers, modern architectures increasingly employ GELU and Swish, whereas Sigmoid and Softmax remain the preferred choices for binary and multi-class output layers, respectively.
