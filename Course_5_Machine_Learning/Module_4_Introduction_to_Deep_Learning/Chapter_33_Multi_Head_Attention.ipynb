# <font color='blue'> Chapter 33: Multi-Head Attention </font>

In the previous chapter, we derived the **Scaled Dot-Product Attention** equation

$$
\boxed{
\text{Attention}(Q,K,V)
=
\text{Softmax}
\left(
\frac{QK^T}
{\sqrt{d_k}}
\right)V.
}
$$

This mechanism allows every token to attend to every other token in a sequence.

However,

a single attention computation can only learn **one type of relationship** at a time.

Modern Transformer models therefore use **Multi-Head Attention**, where several independent attention mechanisms operate simultaneously.

Each attention head learns a different representation of the same input sequence.

This greatly increases the expressive power of the Transformer.

---



# <font color='orange'> 1. Motivation </font>

Consider the sentence

> **The boy who was wearing a blue jacket kicked the ball.**

Different relationships exist simultaneously.

One attention mechanism may focus on

- grammatical structure,

another on

- subject–verb agreement,

another on

- descriptive adjectives,

and another on

- semantic meaning.

A single attention head cannot easily capture all of these relationships simultaneously.

Multiple attention heads allow the model to learn several complementary representations.

---



# <font color='orange'> 2. Why One Attention Head is Not Enough </font>

Suppose we have only one attention head.

```
Sentence

↓

Attention

↓

Output
```

The model must learn

- syntax,
- semantics,
- long-range dependencies,
- positional relationships,

using one set of parameters.

This is unnecessarily restrictive.

Instead,

we create several independent attention heads.

```
Sentence

↓

Head 1

↓

Head 2

↓

Head 3

↓

Head 4

↓

Combine
```

Each head specializes in different patterns.

---



# <font color='orange'> 3. Independent Q, K and V Matrices </font>

Every attention head learns its own projections.

For head

$$
i,
$$

we compute

$$
\boxed{
Q_i=XW_Q^{(i)},
}
$$

$$
\boxed{
K_i=XW_K^{(i)},
}
$$

$$
\boxed{
V_i=XW_V^{(i)}.
}
$$

Thus,

every head has its own

- Query matrix,
- Key matrix,
- Value matrix.

Each head therefore learns a different representation of the same input sequence.

---



# <font color='orange'> 4. Attention Within Each Head </font>

Each head independently computes

$$
\boxed{
\text{head}_i
=
\text{Softmax}
\left(
\frac{Q_iK_i^T}
{\sqrt{d_k}}
\right)
V_i.
}
$$

Notice that

this is exactly the same attention equation derived previously.

The only difference is that

every head has different learned projection matrices.

---



# <font color='orange'> 5. Parallel Computation </font>

All attention heads operate simultaneously.

```
Input

↓

Head 1

↓

Head 2

↓

Head 3

↓

Head 4

↓

Concatenate
```

Since modern hardware performs matrix operations efficiently,

these computations occur in parallel rather than sequentially.

---



# <font color='orange'> 6. Concatenating the Heads </font>

Suppose we have

$$
h
$$

attention heads.

Their outputs are concatenated.

$$
\boxed{
H
=
\text{Concat}
(
\text{head}_1,
\dots,
\text{head}_h
).
}
$$

The resulting representation combines information learned by every head.

---



# <font color='orange'> 7. Final Linear Projection </font>

After concatenation,

a final linear transformation is applied.

$$
\boxed{
Y
=
HW_O,
}
$$

where

$$
W_O
$$

is another trainable weight matrix.

This projection mixes information from all attention heads into a unified representation.

---



# <font color='orange'> 8. Complete Multi-Head Attention Equation </font>

Combining all the previous steps,

we obtain

$$
\boxed{
\text{MultiHead}(Q,K,V)
=
\text{Concat}
(
\text{head}_1,
\dots,
\text{head}_h
)
W_O.
}
$$

This is the standard Multi-Head Attention equation used in Transformer models.

---



# <font color='orange'> 9. Why Smaller Heads? </font>

Suppose

the embedding dimension is

$$
d=512,
$$

and we use

$$
8
$$

attention heads.

Each head typically uses

$$
d_k=\frac{512}{8}=64.
$$

Instead of one large attention computation,

we perform

eight smaller attention computations.

Together,

they retain the same overall dimensionality while allowing each head to specialize.

---



# <font color='orange'> 10. What Different Heads Learn </font>

Although no head is manually assigned a role,

research has shown that different heads often learn different patterns.

Examples include

- subject–verb agreement,
- pronoun resolution,
- punctuation,
- long-range dependencies,
- positional relationships,
- semantic similarity.

Some heads even become highly specialized for specific linguistic structures.

---



# <font color='orange'> 11. Matrix Dimensions </font>

Suppose

- sequence length

$$
n,
$$

- embedding dimension

$$
d,
$$

- number of heads

$$
h.
$$

Then each head usually has

$$
d_k=\frac{d}{h}.
$$

For each head

| Matrix | Shape |
|:---|:---:|
| \(Q_i\) | \(n\times d_k\) |
| \(K_i\) | \(n\times d_k\) |
| \(V_i\) | \(n\times d_k\) |
| Head Output | \(n\times d_k\) |

After concatenation,

the output becomes

$$
n\times(hd_k)=n\times d.
$$

Thus,

the final output has the same embedding dimension as the input.

---



# <font color='orange'> 12. Numerical Illustration </font>

Suppose

the embedding dimension is

$$
d=8,
$$

and we use

$$
2
$$

attention heads.

Each head processes

$$
d_k=4
$$

features.

```
Embedding (8)

↓

Head 1 (4)

↓

Attention

↓

Output (4)

+

Head 2 (4)

↓

Attention

↓

Output (4)

↓

Concatenate

↓

Output (8)
```

Thus,

the final representation retains the original dimensionality while incorporating two independent attention computations.

---



# <font color='orange'> 13. Advantages of Multi-Head Attention </font>

Compared with a single attention head,

Multi-Head Attention

- learns multiple representations simultaneously,
- captures different linguistic and semantic relationships,
- improves expressive power,
- enables richer contextual representations,
- increases model performance.

These advantages make Multi-Head Attention a central component of the Transformer architecture.

---

# <font color='red'> 14. Mathematical Foundations </font>

For each attention head,

$$
\boxed{
Q_i=XW_Q^{(i)},
}
$$

$$
\boxed{
K_i=XW_K^{(i)},
}
$$

$$
\boxed{
V_i=XW_V^{(i)}.
}
$$

Each head computes

$$
\boxed{
\text{head}_i
=
\text{Softmax}
\left(
\frac{Q_iK_i^T}
{\sqrt{d_k}}
\right)
V_i.
}
$$

The outputs are concatenated,

$$
\boxed{
H
=
\text{Concat}
(
\text{head}_1,
\dots,
\text{head}_h
),
}
$$

and projected back into the embedding space,

$$
\boxed{
Y
=
HW_O.
}
$$

This complete computation forms the Multi-Head Attention block used in every Transformer layer.

---



# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> Every attention head learns the same information.

**False.**

Each head has its own trainable projection matrices, allowing different heads to specialize in different relationships within the data.

---

### Misconception 2

> Multi-Head Attention simply repeats the same computation multiple times.

**False.**

Although the mathematical operation is the same, each head uses different learned parameters, so the resulting attention patterns are generally different.

---

### Misconception 3

> Increasing the number of attention heads always improves performance.

**False.**

Too many heads may increase computational cost and parameter count without providing additional useful representations. The optimal number of heads depends on the model architecture and task.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Attention Head | Independent attention mechanism with its own Q, K, and V projections |
| Multi-Head Attention | Parallel combination of several attention heads |
| Concatenation | Combines outputs from all heads |
| Output Projection | Linear transformation that mixes information from different heads |
| Head Dimension | Usually \(d_k=d/h\) |
| Parallel Processing | All heads compute attention simultaneously |

> **Key Insight:** Multi-Head Attention extends scaled dot-product attention by running several independent attention mechanisms in parallel. Each head learns its own representation of the input sequence, allowing the Transformer to capture multiple types of relationships simultaneously. After concatenating the outputs and applying a final linear projection, the model obtains a rich contextual representation that is significantly more expressive than that produced by a single attention head.
