# <font color='blue'> Chapter 7: Forward Propagation </font>

Thus far, we have studied the individual components of a neural network, including neurons, activation functions, and network architectures. We now examine how information flows through the network during prediction.

This process is known as **Forward Propagation**.

Forward propagation is the process of passing the input data through each layer of the neural network, computing intermediate activations until the final prediction is produced.

It is called **forward** because information always moves from the input layer towards the output layer.

Every prediction made by a neural network, whether for image classification, speech recognition, language translation, or regression, begins with forward propagation.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to classify an image of a handwritten digit.

The raw image consists of thousands of pixel values.

The neural network cannot immediately recognise the digit.

Instead, it gradually transforms the image through multiple layers.

```
Image

↓

Edges

↓

Corners

↓

Shapes

↓

Digit

↓

Prediction
```

Each layer extracts increasingly useful information.

This transformation of information from one layer to the next is called **forward propagation**.

---



# <font color='orange'> 2. What is Forward Propagation? </font>

Forward propagation is simply the repeated application of

1. Linear transformation
2. Activation function

Every layer performs exactly these two operations.

For a single neuron,

```
Inputs

↓

Weighted Sum

↓

Activation

↓

Output
```

For an entire neural network,

this process repeats across every layer.

---



# <font color='orange'> 3. Forward Propagation in a Single Neuron </font>

Recall that one neuron computes

$$
\boxed{
z
=
\mathbf{w}^T\mathbf{x}
+
b.
}
$$

The activation is

$$
\boxed{
a
=
f(z).
}
$$

Combining both,

$$
\boxed{
a
=
f(\mathbf{w}^T\mathbf{x}+b).
}
$$

This equation represents one forward propagation step.

---



# <font color='orange'> 4. From One Neuron to One Layer </font>

Suppose a layer contains

three neurons.

Each neuron has different weights,

but all receive the same input vector.

```
          x₁

          x₂

          x₃

        / | \

       /  |  \

      ○   ○   ○

      |   |   |

      ▼   ▼   ▼
```

Neuron 1 computes

$$
z_1
=
\mathbf{w}_1^T\mathbf{x}+b_1.
$$

Neuron 2 computes

$$
z_2
=
\mathbf{w}_2^T\mathbf{x}+b_2.
$$

Neuron 3 computes

$$
z_3
=
\mathbf{w}_3^T\mathbf{x}+b_3.
$$

Collecting these together,

we obtain

$$
\boxed{
\mathbf{z}
=
W\mathbf{x}
+
\mathbf{b}.
}
$$

---



# <font color='orange'> 5. Applying the Activation Function </font>

Each neuron now applies an activation function independently.

Thus,

$$
\boxed{
\mathbf{a}
=
f(\mathbf{z}).
}
$$

Substituting

$$
\mathbf{z},
$$

we obtain

$$
\boxed{
\mathbf{a}
=
f(W\mathbf{x}
+
\mathbf{b}).
}
$$

This equation describes an entire neural network layer.

---



# <font color='orange'> 6. Forward Propagation Through Multiple Layers </font>

Suppose the network contains

three hidden layers.

The computation becomes

First hidden layer

$$
\mathbf{a}^{(1)}
=
f
\left(
W^{(1)}
\mathbf{x}
+
\mathbf{b}^{(1)}
\right).
$$

Second hidden layer

$$
\mathbf{a}^{(2)}
=
f
\left(
W^{(2)}
\mathbf{a}^{(1)}
+
\mathbf{b}^{(2)}
\right).
$$

Third hidden layer

$$
\mathbf{a}^{(3)}
=
f
\left(
W^{(3)}
\mathbf{a}^{(2)}
+
\mathbf{b}^{(3)}
\right).
$$

Output layer

$$
\boxed{
\hat y
=
f
\left(
W^{(4)}
\mathbf{a}^{(3)}
+
\mathbf{b}^{(4)}
\right).
}
$$

Notice that

every layer performs exactly the same mathematical operation.

Only

- weights,
- biases,

and occasionally

- activation functions

change.

---



# <font color='orange'> 7. General Mathematical Formulation </font>

Suppose

layer

$l$

contains

$n_l$

neurons.

The previous layer contains

$n_{l-1}$

neurons.

Forward propagation is

$$
\boxed{
\mathbf{z}^{(l)}
=
W^{(l)}
\mathbf{a}^{(l-1)}
+
\mathbf{b}^{(l)}.
}
$$

Then

$$
\boxed{
\mathbf{a}^{(l)}
=
f
\left(
\mathbf{z}^{(l)}
\right).
}
$$

Combining both,

$$
\boxed{
\mathbf{a}^{(l)}
=
f
\left(
W^{(l)}
\mathbf{a}^{(l-1)}
+
\mathbf{b}^{(l)}
\right).
}
$$

This is arguably the most important equation in feedforward neural networks.

# <font color='orange'> 8. Numerical Example </font>

Suppose

$$
\mathbf{x}
=
\begin{bmatrix}
2\\
1
\end{bmatrix},
$$

$$
W=
\begin{bmatrix}
1 & -1\\
0.5 & 2
\end{bmatrix},
$$

$$
\mathbf{b}
=
\begin{bmatrix}
1\\
-2
\end{bmatrix}.
$$

Compute

$$
\mathbf{z}
=
W\mathbf{x}
+
\mathbf{b}.
$$

First,

$$
W\mathbf{x}
=
\begin{bmatrix}
1 & -1\\
0.5 & 2
\end{bmatrix}
\begin{bmatrix}
2\\
1
\end{bmatrix}
=
\begin{bmatrix}
1\\
3
\end{bmatrix}.
$$

Adding the bias,

$$
\mathbf{z}
=
\begin{bmatrix}
2\\
1
\end{bmatrix}.
$$

Applying ReLU,

$$
\boxed{
\mathbf{a}
=
\begin{bmatrix}
2\\
1
\end{bmatrix}.
}
$$

If we instead applied the sigmoid function,

the activations would become

$$
\mathbf{a}
=
\begin{bmatrix}
0.881\\
0.731
\end{bmatrix}.
$$

This example demonstrates that the linear transformation and the activation function are distinct stages of the computation.

# <font color='orange'> 9. Why Matrix Multiplication? </font>

Instead of computing each neuron individually,

modern neural networks perform one large matrix multiplication.

Advantages include

- simpler notation,
- efficient implementation,
- vectorised computation,
- GPU acceleration,
- parallel processing.

This is one reason why deep learning frameworks such as TensorFlow and PyTorch rely heavily on matrix operations.

---



# <font color='orange'> 10. Vectorised Computation </font>

Suppose a layer contains

1000 neurons.

Instead of evaluating

1000 equations separately,

the network computes

$$
W\mathbf{x}
$$

using a single optimized matrix multiplication routine.

This approach is known as **vectorisation**.

It is significantly faster than computing each neuron individually.

---



# <font color='orange'> 11. Computational Complexity </font>

Suppose

- previous layer contains

$$
n
$$

neurons,

- current layer contains

$$
m
$$

neurons.

Computing

$$
W\mathbf{x}
$$

requires approximately

$$
mn
$$

multiplications.

Therefore,

the computational complexity is

$$
\boxed{
\mathcal{O}(mn).
}
$$

For very deep networks,

this matrix multiplication dominates the computational cost.

# <font color='red'> 12. Mathematical Foundations </font>

## Why Does Every Layer Use the Same Equation?

Consider one neuron

$$
a=f(w^Tx+b).
$$

Now suppose

instead of one neuron,

we stack

$m$

neurons together.

Arrange their weight vectors into a matrix

$$
W=
\begin{bmatrix}
w_1^T\\
w_2^T\\
\vdots\\
w_m^T
\end{bmatrix}.
$$

Similarly,

collect the biases into a vector

$$
\mathbf{b}
=
\begin{bmatrix}
b_1\\
b_2\\
\vdots\\
b_m
\end{bmatrix}.
$$

The output of all neurons can now be written simultaneously as

$$
\boxed{
\mathbf{z}
=
W\mathbf{x}
+
\mathbf{b}.
}
$$

Applying the activation function element-wise,

$$
\boxed{
\mathbf{a}
=
f(W\mathbf{x}+\mathbf{b}).
}
$$

Thus,

the matrix equation is simply a compact way of writing the individual equations for every neuron in the layer.

# <font color='orange'> 13. Common Misconceptions </font>

### Misconception 1

> Forward propagation changes the weights.

**False.**

During forward propagation, the weights remain fixed. The network only computes predictions.

---

### Misconception 2

> The activation function is applied to the whole matrix.

**False.**

The activation function is applied **element-wise** to every component of the vector

$$
\mathbf{z}.
$$

---

### Misconception 3

> Every layer must use the same activation function.

**False.**

Different layers can use different activation functions. For example,

- Hidden layers: ReLU
- Output layer (binary classification): Sigmoid
- Output layer (multi-class classification): Softmax

The choice depends on the learning task.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Forward Propagation | Process of computing predictions by passing data through the network |
| Linear Transformation | Computes the weighted sum, $W\mathbf{x}+\mathbf{b}$ |
| Activation Function | Introduces nonlinearity by transforming the weighted sum |
| Layer Equation | $\mathbf{a}=f(W\mathbf{x}+\mathbf{b})$ |
| Vectorisation | Simultaneous computation of all neurons in a layer |
| Matrix Multiplication | Efficient implementation of the linear transformation |

> **Key Insight:** Forward propagation is the mechanism by which a neural network transforms raw input data into predictions. Each layer performs the same two operations—a linear transformation followed by a nonlinear activation. By repeatedly applying this simple process, the network gradually extracts increasingly informative features. Understanding forward propagation is essential because it forms the foundation upon which loss functions, gradient descent, and backpropagation are built.
