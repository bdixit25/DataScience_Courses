# <font color='blue'> Chapter 2: The Artificial Neuron </font>

The **artificial neuron** is the fundamental computational unit of every neural network. Regardless of whether we are building a simple feedforward network, a Convolutional Neural Network (CNN), a Recurrent Neural Network (RNN), or a Transformer, each model ultimately consists of many interconnected neurons performing the same basic computation.

The purpose of a neuron is remarkably simple:

- receive information,
- combine the information,
- determine its importance,
- and produce an output.

Although the computation performed by a single neuron is mathematically straightforward, millions or even billions of neurons working together are capable of solving extremely complex problems.

Throughout this chapter, we shall study the mathematical formulation, geometric interpretation, and intuition behind artificial neurons.

---

# <font color='orange'> 1. Motivation </font>

Suppose we want to predict whether a patient has a certain disease.

The available information might include

- Age
- Blood Pressure
- Cholesterol Level
- Body Mass Index
- Heart Rate

Clearly,

not every variable contributes equally to the diagnosis.

For example,

blood pressure may be more informative than age,

while cholesterol may contribute negatively or positively depending on the disease.

The neuron learns **how important** each input should be.

Rather than assigning these importance values manually,

the neural network learns them automatically from data.

---

# <font color='orange'> 2. The Biological Inspiration </font>

Artificial neurons were originally inspired by biological neurons.

A biological neuron contains

```
Dendrites
     │
     ▼
Cell Body
     │
     ▼
 Axon
     │
     ▼
Synapse
```

Each neuron receives electrical signals from many neighbouring neurons.

The signals are combined inside the cell body.

If the total signal exceeds a certain threshold,

the neuron fires and sends an electrical impulse through the axon.

Although artificial neurons are vastly simpler than biological neurons,

the overall idea remains similar.

---

# <font color='orange'> 3. Mathematical Model of a Neuron </font>

Suppose a neuron receives

$n$

inputs

$$
x_1,x_2,\ldots,x_n.
$$

Each input has an associated weight

$$
w_1,w_2,\ldots,w_n.
$$

The neuron computes a weighted sum

$$
\boxed{
z
=
w_1x_1
+
w_2x_2
+
\cdots
+
w_nx_n
+
b,
}
$$

where

- $w_i$ are the weights,
- $x_i$ are the inputs,
- $b$ is the bias.

The weighted sum is often called the

**pre-activation value**.

---

The neuron then applies an activation function

$$
\boxed{
a
=
f(z).
}
$$

This value

$a$

becomes the output of the neuron.

Therefore,

the complete neuron is described by

$$
\boxed{
a
=
f
\left(
\sum_{i=1}^{n}
w_ix_i+b
\right).
}
$$

This equation is one of the most important equations in Deep Learning.

---

# <font color='orange'> 4. Understanding Each Component </font>

## Inputs

Inputs represent the information entering the neuron.

Examples include

- house size,
- image pixels,
- audio amplitudes,
- temperatures,
- stock prices,
- word embeddings.

The neuron itself does not know what the inputs represent.

It simply processes numbers.

---

## Weights

Weights determine the importance of each input.

If

$$
w_i
=
0,
$$

then the corresponding input has no influence.

If

$$
|w_i|
$$

is large,

the corresponding feature strongly influences the output.

Positive weights increase the activation,

while negative weights suppress it.

---

## Bias

The bias allows the neuron to shift its decision boundary.

Without the bias,

every decision boundary would pass through the origin.

Mathematically,

the bias is simply another parameter learned during training.

---

## Activation Function

After computing

$$
z,
$$

the neuron applies

$$
a=f(z).
$$

Without this nonlinear function,

deep neural networks would reduce to ordinary linear regression,

regardless of the number of layers.

Activation functions therefore give neural networks their expressive power.

---

# <font color='orange'> 5. Why Weighted Sums? </font>

Consider two students applying for a scholarship.

Suppose the admission committee considers

- Exam Score
- Interview Score

Clearly,

both are important,

but perhaps the exam should count twice as much.

A simple model might compute

$$
\boxed{
Score
=
2\times
(\text{Exam})
+
1\times
(\text{Interview}).
}
$$

The coefficients

2

and

1

represent the relative importance.

A neuron performs exactly the same calculation,

except that the coefficients are learned automatically.

---

# <font color='orange'> 6. Vector Representation </font>

Instead of writing

$$
w_1x_1+w_2x_2+\cdots+w_nx_n,
$$

we define

$$
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2\\
\vdots\\
x_n
\end{bmatrix},
$$

and

$$
\mathbf{w}
=
\begin{bmatrix}
w_1\\
w_2\\
\vdots\\
w_n
\end{bmatrix}.
$$

The weighted sum becomes

$$
\boxed{
z
=
\mathbf{w}^T
\mathbf{x}
+
b.
}
$$

This compact notation is used throughout deep learning.

---

# <font color='orange'> 7. Geometric Interpretation </font>

The quantity

$$
\mathbf{w}^T\mathbf{x}
$$

is the dot product between two vectors.

Recall from linear algebra

$$
\boxed{
\mathbf{w}^T\mathbf{x}
=
||\mathbf{w}||
||
\mathbf{x}
||
\cos\theta.
}
$$

This tells us

- the neuron measures similarity,
- similar vectors produce large positive values,
- opposite vectors produce negative values.

Thus,

a neuron can be viewed as a similarity detector.

This geometric interpretation is extremely important,

especially later when studying embeddings and attention mechanisms.

---

# <font color='orange'> 8. Decision Boundaries </font>

Suppose a neuron outputs

Positive

whenever

$$
z>0,
$$

and

Negative

otherwise.

The decision boundary is

$$
\boxed{
w^Tx+b=0.
}
$$

In two dimensions,

this equation describes a straight line.

In three dimensions,

it describes a plane.

In higher dimensions,

it becomes a **hyperplane**.

Thus,

a single neuron divides the feature space into two regions.

---

# <font color='orange'> 9. Numerical Example </font>

Suppose

$$
x=
\begin{bmatrix}
2\\
3
\end{bmatrix},
$$

weights are

$$
w=
\begin{bmatrix}
1.5\\
-0.5
\end{bmatrix},
$$

and

$$
b=2.
$$

The weighted sum becomes

$$
z
=
1.5(2)
-
0.5(3)
+
2
=
3
-
1.5
+
2
=
3.5.
$$

If the activation function is ReLU,

$$
a
=
\max(0,3.5)
=
3.5.
$$

If the activation is Sigmoid,

$$
a
=
\frac{1}{1+e^{-3.5}}
\approx
0.9707.
$$

Notice that

the same neuron

produces different outputs

depending only upon the activation function.

---

# <font color='orange'> 10. A Neuron is a Linear Model </font>

Ignoring the activation,

the neuron computes

$$
z
=
w^Tx+b.
$$

This is exactly the same mathematical form as

- Linear Regression
- Logistic Regression
- Perceptron

Neural networks become powerful

only after introducing nonlinear activation functions and stacking many neurons together.

Thus,

the neuron can be viewed as the natural generalization of the linear models studied previously.

---

# <font color='orange'> 11. Limitations of a Single Neuron </font>

A single neuron can only learn

a linear decision boundary.

Consequently,

it cannot solve problems such as

- XOR classification,
- image recognition,
- speech recognition,
- language translation.

These tasks require multiple neurons arranged in several layers.

This motivates the study of **multilayer neural networks**.

---

# <font color='red'> 12. Mathematical Foundations </font>

## Derivation of the Neuron Equation

Suppose the input vector is

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)^T.
$$

Each feature contributes proportionally to its importance.

Hence,

the total contribution is

$$
w_1x_1
+
w_2x_2
+
\cdots
+
w_nx_n.
$$

This is precisely the definition of the dot product,

which gives

$$
\boxed{
\mathbf{w}^T\mathbf{x}.
}
$$

Adding a bias,

we obtain

$$
\boxed{
z=\mathbf{w}^T\mathbf{x}+b.
}
$$

Finally,

applying an activation function,

we obtain

$$
\boxed{
a=f(\mathbf{w}^T\mathbf{x}+b).
}
$$

This equation defines every artificial neuron used throughout modern deep learning.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Input | Information entering the neuron |
| Weight | Learns the importance of an input |
| Bias | Shifts the decision boundary |
| Weighted Sum | Linear combination of inputs |
| Activation | Introduces nonlinearity |
| Dot Product | Measures similarity between vectors |
| Hyperplane | Decision boundary produced by a neuron |

> **Key Insight:** An artificial neuron performs a simple computation: it forms a weighted combination of its inputs, adds a bias, and applies a nonlinear activation function. Individually, a neuron is only capable of representing linear relationships, but when many neurons are connected together, they can approximate highly complex nonlinear functions. The neuron is therefore the fundamental building block from which all modern deep learning architectures are constructed.
