# <font color='blue'> Chapter 48: Temporal Difference (TD) Learning </font>

Monte Carlo methods estimate value functions by averaging the returns obtained from complete episodes.

Although this approach is mathematically simple,

it requires waiting until an episode has terminated before any learning can occur.

Temporal Difference (TD) Learning removes this limitation.

Instead of waiting until the end of an episode,

TD Learning updates value estimates immediately after each interaction with the environment.

This allows an agent to learn continuously while interacting with the environment,

making TD Learning one of the most important techniques in Reinforcement Learning.

Many modern algorithms,

including SARSA,

Q-Learning,

Deep Q Networks (DQN),

and Actor-Critic methods,

are built upon the principles introduced in this chapter.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot is navigating a maze.

With Monte Carlo learning,

the robot waits until

```
Start

↓

Move

↓

Move

↓

Move

↓

Goal

↓

Update Values
```

Only after reaching the goal does learning begin.

Temporal Difference Learning instead performs

```
Start

↓

Move

↓

Update

↓

Move

↓

Update

↓

Move

↓

Update
```

Learning occurs after every step,

greatly improving efficiency.

---

# <font color='orange'> 2. The Central Idea </font>

Suppose we currently estimate

the value of a state as

$$
V(s).
$$

After taking one action,

we observe

- the immediate reward,
- the next state.

Rather than waiting for the complete return,

we use our **current estimate** of the next state's value.

This idea is called

**bootstrapping**.

---

# <font color='orange'> 3. What is Bootstrapping? </font>

Bootstrapping means

using an existing estimate

to improve itself.

Instead of computing

$$
G_t
$$

exactly,

we estimate it as

$$
\boxed{
r_{t+1}
+
\gamma V(s_{t+1}).
}
$$

Notice that

the right-hand side already contains

our current estimate

of

$$
V(s_{t+1}).
$$

The algorithm therefore learns

from its own predictions.

---

# <font color='orange'> 4. Monte Carlo vs Temporal Difference </font>

Monte Carlo target

$$
\boxed{
G_t.
}
$$

Requires

the complete episode.

---

Temporal Difference target

$$
\boxed{
r_{t+1}
+
\gamma
V(s_{t+1}).
}
$$

Requires only

one transition.

This is the key difference.

---

# <font color='orange'> 5. The TD Error </font>

Suppose

our current estimate is

$$
V(s_t).
$$

After observing

the reward

and

the next state,

our new estimate becomes

$$
r_{t+1}
+
\gamma
V(s_{t+1}).
$$

The difference between these quantities is called the

**Temporal Difference Error**

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
V(s_{t+1})
-
V(s_t).
}
$$

Interpretation

- Positive error

↓

state is better than expected.

- Negative error

↓

state is worse than expected.

---

# <font color='orange'> 6. TD Update Rule </font>

The value estimate is updated according to

$$
\boxed{
V(s_t)
\leftarrow
V(s_t)
+
\alpha
\delta_t,
}
$$

where

- \(\alpha\) is the learning rate.

Substituting the TD error,

$$
\boxed{
V(s_t)
\leftarrow
V(s_t)
+
\alpha
\left[
r_{t+1}
+
\gamma
V(s_{t+1})
-
V(s_t)
\right].
}
$$

This is the

TD(0)

update equation.

---

# <font color='orange'> 7. Why is it Called "Temporal Difference"? </font>

The update compares

two estimates made

at consecutive time steps.

```
Current Estimate

↓

Next Estimate

↓

Difference

↓

Update
```

The algorithm therefore learns from the

difference across time,

hence the name

**Temporal Difference Learning**.

---

# <font color='orange'> 8. TD(0) Algorithm </font>

Algorithm

1. Observe current state.
2. Select an action.
3. Receive reward.
4. Observe next state.
5. Compute TD error.
6. Update value estimate.
7. Repeat.

Unlike Monte Carlo,

the episode does not need to terminate before learning occurs.

---

# <font color='orange'> 9. Online Learning </font>

Because updates occur immediately,

TD Learning naturally supports

**online learning**.

```
Interaction

↓

Update

↓

Interaction

↓

Update
```

This makes it suitable for

robotics,

autonomous vehicles,

financial trading,

and other continuously operating systems.

---

# <font color='orange'> 10. Bias and Variance </font>

Monte Carlo methods use the true return,

making them

unbiased,

but often high variance.

Dynamic Programming uses exact expectations,

making it low variance,

but requiring a complete model.

Temporal Difference Learning lies between these extremes.

It introduces some bias through bootstrapping,

but usually has much lower variance than Monte Carlo methods,

often leading to faster and more stable learning.

---

# <font color='orange'> 11. Dynamic Programming vs Monte Carlo vs TD </font>

| Dynamic Programming | Monte Carlo | TD Learning |
|:---|:---|:---|
| Requires environment model | Model-free | Model-free |
| Bootstrapping | No bootstrapping | Bootstrapping |
| Updates every step | End of episode | Every step |
| Uses expectations | Uses sampled returns | Uses sampled transitions |

TD Learning combines the advantages of both Dynamic Programming and Monte Carlo methods.

---

# <font color='orange'> 12. Applications </font>

Temporal Difference Learning forms the basis of

- SARSA,
- Q-Learning,
- Deep Q Networks,
- Actor-Critic methods,
- modern game-playing agents,
- robotics,
- autonomous navigation.

Its online nature makes it one of the most widely used learning paradigms in Reinforcement Learning.

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> TD Learning requires complete episodes.

**False.**

TD Learning updates after every transition and does not require waiting until an episode has finished.

---

### Misconception 2

> TD Learning knows the true future reward.

**False.**

TD Learning estimates future rewards using its current value function. This process is called bootstrapping.

---

### Misconception 3

> Bootstrapping always produces exact value estimates.

**False.**

Bootstrapping relies on current estimates, which may initially be inaccurate. However, repeated updates typically improve these estimates over time.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Temporal Difference Learning | Learns from sampled transitions rather than complete episodes |
| Bootstrapping | Updating estimates using other current estimates |
| TD Target | Immediate reward plus discounted value of the next state |
| TD Error (\(\delta_t\)) | Difference between the TD target and the current estimate |
| TD(0) | Simplest Temporal Difference algorithm |
| Online Learning | Updates values after every interaction |
| Bias–Variance Trade-off | TD introduces some bias but often reduces variance compared with Monte Carlo methods |

> **Key Insight:** Temporal Difference Learning combines the model-free nature of Monte Carlo methods with the bootstrapping idea from Dynamic Programming. Instead of waiting until an episode has finished, TD Learning updates value estimates after every interaction using the immediate reward and the estimated value of the next state. This makes learning faster, more memory efficient, and suitable for continuing tasks, forming the foundation of many modern Reinforcement Learning algorithms.
