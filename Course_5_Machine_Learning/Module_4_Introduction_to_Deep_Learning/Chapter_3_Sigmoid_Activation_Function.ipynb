# <font color='blue'> Chapter 3: The Sigmoid Activation Function </font>

In the previous chapter, we learned that an artificial neuron computes a weighted sum of its inputs,

$$
z = \mathbf{w}^T\mathbf{x}+b.
$$

However, if the neuron simply outputs this weighted sum, then no matter how many neurons we stack together, the entire network remains a **linear model**.

The key idea that transformed simple linear models into modern Deep Learning was the introduction of **activation functions**.

An activation function introduces **nonlinearity**, allowing neural networks to learn highly complex relationships that cannot be represented by linear equations alone.

The first activation function to become widely popular was the **Sigmoid Function**, which we study in this chapter.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to classify emails into

- Spam
- Not Spam

A linear model may compute

$$
z = \mathbf{w}^T\mathbf{x}+b.
$$

The value of

$$
z
$$

could be

- -15
- 2.8
- 103
- -47

These numbers are difficult to interpret directly.

Ideally,

we would like the model to output something like

$$
0.95
$$

meaning

> "There is a 95% probability that this email is spam."

To convert any real number into a value between

0

and

1,

we require a nonlinear function.

The Sigmoid Function does exactly this.

---



# <font color='orange'> 2. Definition of the Sigmoid Function </font>

The Sigmoid Function is defined as

$$
\boxed{
\sigma(z)
=
\frac{1}{1+e^{-z}}.
}
$$

where

- $z$ is the weighted sum,

$$
z=\mathbf{w}^T\mathbf{x}+b.
$$

The neuron output therefore becomes

$$
\boxed{
a=\sigma(z)
=
\frac{1}{1+e^{-z}}.
}
$$

---



# <font color='orange'> 3. Why is it Called "Sigmoid"? </font>

The word

**sigmoid**

comes from the Greek word

*"sigma"*

meaning

**S-shaped**.

The graph has the characteristic shape

```
1.0 |                         *****
    |                     ****
0.8 |                 ****
    |             ****
0.6 |          ***
    |       ***
0.5 |------*--------------------
    |    ***
0.4 |  ***
    | **
0.2 |*
    |
0.0 +------------------------------->
          z
```

Notice

- very negative inputs approach

$$
0,
$$

- very positive inputs approach

$$
1,
$$

- the midpoint occurs at

$$
z=0.
$$

---



# <font color='orange'> 4. Properties of the Sigmoid Function </font>

The sigmoid function has several useful mathematical properties.

### Range

The output always satisfies

$$
\boxed{
0<\sigma(z)<1.
}
$$

Therefore,

it is ideal for representing probabilities.

---

### Smoothness

The sigmoid function is continuous and differentiable everywhere.

This property makes it suitable for optimization using gradient descent.

---

### Monotonicity

As

$$
z
$$

increases,

the output also increases.

Thus,

larger weighted sums correspond to higher predicted probabilities.

---

### Symmetry

Although the sigmoid function itself is not symmetric,

it satisfies

$$
\boxed{
\sigma(-z)
=
1-\sigma(z).
}
$$

This elegant property is often useful in mathematical derivations.

---



# <font color='orange'> 5. Behaviour of the Sigmoid Function </font>

Let us evaluate the sigmoid for several values of

$$
z.
$$

| $z$ | $\sigma(z)$ |
|:---:|:---:|
| -5 | 0.0067 |
| -3 | 0.0474 |
| -2 | 0.1192 |
| -1 | 0.2689 |
| 0 | 0.5000 |
| 1 | 0.7311 |
| 2 | 0.8808 |
| 3 | 0.9526 |
| 5 | 0.9933 |

Notice

- large negative values approach zero,
- large positive values approach one,
- the transition occurs smoothly around zero.

---



# <font color='orange'> 6. Decision Boundary </font>

Suppose we classify an observation as

Positive

whenever

$$
\sigma(z)\ge0.5.
$$

Since

$$
\sigma(0)=0.5,
$$

this condition becomes

$$
\boxed{
z\ge0.
}
$$

Since

$$
z=\mathbf{w}^T\mathbf{x}+b,
$$

the decision boundary is

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0.
}
$$

Interestingly,

the sigmoid changes the output,

but **does not change the location of the decision boundary**.

---



# <font color='orange'> 7. Interpretation as Probability </font>

One reason the sigmoid function became extremely popular is that

its output naturally resembles a probability.

For example,

$$
\sigma(z)=0.92
$$

can be interpreted as

> "The model estimates a 92% probability that the observation belongs to the positive class."

This interpretation forms the basis of

**Logistic Regression**,

which is essentially a single neuron with a sigmoid activation function.

---



# <font color='orange'> 8. Why is Sigmoid Nonlinear? </font>

Suppose the activation were simply

$$
a=z.
$$

Then

two consecutive layers become

$$
a_2
=
W_2(W_1x+b_1)+b_2,
$$

which simplifies to

$$
a_2
=
Wx+b.
$$

No additional expressive power is gained.

However,

if

$$
a=\sigma(z),
$$

then

$$
a_2
=
\sigma
\left(
W_2
\sigma
(W_1x+b_1)
+b_2
\right),
$$

which can no longer be simplified into a single linear equation.

This is why activation functions are essential.

---



# <font color='orange'> 9. Derivative of the Sigmoid Function </font>

The derivative of the sigmoid function plays a central role in **Backpropagation**.

Starting with

$$
\sigma(z)
=
\frac{1}{1+e^{-z}},
$$

differentiate with respect to

$$
z.
$$

Using the chain rule,

$$
\frac{d\sigma}{dz}
=
-(1+e^{-z})^{-2}
(-e^{-z}).
$$

Simplifying,

$$
=
\frac{e^{-z}}
{(1+e^{-z})^2}.
$$

Now multiply the numerator and denominator by

$$
1.
$$

Observe that

$$
\sigma(z)
=
\frac{1}{1+e^{-z}}
$$

and

$$
1-\sigma(z)
=
\frac{e^{-z}}
{1+e^{-z}}.
$$

Therefore,

$$
\boxed{
\frac{d\sigma}{dz}
=
\sigma(z)
\left(
1-\sigma(z)
\right).
}
$$

This remarkably elegant result is one reason sigmoid became historically important.

---



# <font color='orange'> 10. Vanishing Gradient Problem </font>

Notice that

$$
\sigma(z)
$$

approaches

0

or

1

for large values of

$$
|z|.
$$

Its derivative

$$
\sigma(z)(1-\sigma(z))
$$

then becomes extremely small.

For example,

$$
\sigma(8)\approx0.9997,
$$

so

$$
\sigma'(8)\approx0.0003.
$$

During backpropagation,

many such small derivatives are multiplied together,

causing gradients to become almost zero.

This phenomenon is called the

**Vanishing Gradient Problem**.

It is one of the main reasons modern deep learning rarely uses sigmoid in hidden layers.

---



# <font color='orange'> 11. Advantages of the Sigmoid Function </font>

The sigmoid function

- is smooth,
- is differentiable everywhere,
- produces outputs between 0 and 1,
- naturally represents probabilities,
- is mathematically elegant,
- works well for binary classification outputs.

---



# <font color='orange'> 12. Limitations of the Sigmoid Function </font>

The sigmoid function

- suffers from vanishing gradients,
- saturates for large inputs,
- is computationally more expensive than ReLU,
- is not zero-centered,
- often slows training in deep networks.

Consequently,

modern neural networks typically use

- ReLU,
- Leaky ReLU,
- GELU,

for hidden layers,

while sigmoid is primarily reserved for **binary classification output layers**.

---

# <font color='red'> 13. Mathematical Foundations </font>

## Why Does the Output Always Lie Between 0 and 1?

Since

$$
e^{-z}>0
$$

for every real number,

we have

$$
1+e^{-z}>1.
$$

Therefore,

$$
0<
\frac{1}{1+e^{-z}}
<1.
$$

Hence,

$$
\boxed{
0<\sigma(z)<1.
}
$$

This proves that the sigmoid function always produces values strictly between

0

and

1.

---

## Maximum Derivative

Since

$$
\sigma'(z)
=
\sigma(z)(1-\sigma(z)),
$$

let

$$
a=\sigma(z).
$$

Then

$$
f(a)=a(1-a).
$$

Differentiate,

$$
\frac{df}{da}
=
1-2a.
$$

Setting this equal to zero,

$$
a=\frac12.
$$

Thus,

the maximum derivative occurs when

$$
\sigma(z)=0.5,
$$

which corresponds to

$$
z=0.
$$

The maximum value is

$$
\boxed{
\sigma'(0)=\frac14.
}
$$

Thus,

the sigmoid function can never have a slope greater than

$$
0.25.
$$

This fact contributes to the vanishing gradient problem in deep networks.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Sigmoid Function | Maps any real number to the interval $(0,1)$ |
| Formula | $\sigma(z)=\frac{1}{1+e^{-z}}$ |
| Output Range | $(0,1)$ |
| Midpoint | $\sigma(0)=0.5$ |
| Derivative | $\sigma'(z)=\sigma(z)(1-\sigma(z))$ |
| Main Use Today | Output layer for binary classification |
| Major Limitation | Vanishing gradients |

> **Key Insight:** The sigmoid activation function was one of the first nonlinear activation functions used in neural networks. By mapping arbitrary real-valued inputs into the interval $(0,1)$, it enables probabilistic interpretations and transforms simple linear models into nonlinear ones. Although its elegant mathematical properties make it historically significant, its tendency to produce vanishing gradients means that modern deep learning typically reserves sigmoid for binary output layers while using activations such as ReLU or GELU in hidden layers.
