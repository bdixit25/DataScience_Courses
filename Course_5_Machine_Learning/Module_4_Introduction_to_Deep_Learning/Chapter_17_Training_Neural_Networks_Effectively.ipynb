# <font color='blue'> Chapter 17: Training Neural Networks Effectively </font>

Building a neural network is only the first step in solving a machine learning problem. Equally important is ensuring that the network is trained efficiently and that it generalises well to unseen data.

Training a neural network involves much more than repeatedly calling `model.fit()`. Modern deep learning workflows make extensive use of **callbacks**, **learning curves**, **model checkpoints**, **early stopping**, and **learning-rate scheduling** to improve both performance and reliability.

This chapter introduces these practical techniques and explains why they are essential for training modern neural networks.

---



# <font color='orange'> 1. The Complete Training Pipeline </font>

A typical deep learning workflow is

```
Prepare Data

↓

Create Model

↓

Compile Model

↓

Train Model

↓

Monitor Performance

↓

Save Best Model

↓

Evaluate

↓

Deploy
```

Training is therefore only one stage of the complete machine learning pipeline.

---



# <font color='orange'> 2. Epochs, Batches, and Iterations </font>

These three terms are often confused.

### Epoch

One complete pass through the entire training dataset.

```
Entire Dataset

↓

One Epoch
```

---

### Batch

A small subset of the training data.

Example

```
60,000 Images

↓

Batch Size = 64

↓

938 batches
```

---

### Iteration

One parameter update.

If

- dataset size = 60,000
- batch size = 100

then

```
Iterations per Epoch

=

600
```

# <font color='orange'> 5. Underfitting and Overfitting </font>

## Underfitting

Characteristics

- high training loss,
- high validation loss.

```
Training

High Error

Validation

High Error
```

The model is too simple.

Possible solutions

- larger network,
- more epochs,
- better features.

---

## Overfitting

Characteristics

- very low training loss,
- increasing validation loss.

```
Training

↓

Improves

Validation

↓

Gets Worse
```

The model memorises the training data rather than learning general patterns.

Possible solutions

- Dropout,
- Early Stopping,
- Data Augmentation,
- L2 Regularization.

---



# <font color='orange'> 6. What are Callbacks? </font>

Callbacks are functions executed automatically during training.

Examples include

- stopping training,
- saving models,
- reducing learning rates,
- recording statistics.

Callbacks allow the training process to adapt dynamically.

---



# <font color='orange'> 7. EarlyStopping </font>

One of the most useful callbacks is

```python
keras.callbacks.EarlyStopping()
```

Example

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=5,
    restore_best_weights=True
)
```

Meaning

- monitor validation loss,
- wait five epochs,
- stop if no improvement,
- restore the best model.

Early stopping prevents unnecessary training and reduces overfitting.

---



# <font color='orange'> 8. ModelCheckpoint </font>

Training may take

hours

or even

days.

ModelCheckpoint automatically saves the best model.

Example

```python
checkpoint = keras.callbacks.ModelCheckpoint(
    "best_model.keras",
    save_best_only=True
)
```

If training stops unexpectedly,

the best model is preserved.

---



# <font color='orange'> 9. ReduceLROnPlateau </font>

Sometimes the loss stops improving.

Rather than stopping immediately,

we reduce the learning rate.

Example

```python
reduce_lr = keras.callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    factor=0.5,
    patience=3
)
```

This halves the learning rate whenever progress stalls.

A smaller learning rate often allows the optimizer to converge to a better minimum.

---



# <font color='orange'> 10. TensorBoard </font>

TensorBoard is TensorFlow's visualisation toolkit.

It displays

- loss curves,
- accuracy,
- computational graphs,
- histograms,
- learning rates.

Example

```python
tensorboard = keras.callbacks.TensorBoard(
    log_dir="logs"
)
```

TensorBoard is invaluable for monitoring long training runs.

# <font color='orange'> 12. Using Multiple Callbacks </font>

```python
history = model.fit(

    X_train,

    y_train,

    callbacks=[
        early_stop,
        checkpoint,
        reduce_lr,
        tensorboard
    ]

)
```

# <font color='orange'> 14. Hyperparameters </font>

Hyperparameters are values chosen **before** training.

Examples

- learning rate,
- batch size,
- number of hidden layers,
- neurons per layer,
- optimizer,
- activation function,
- dropout rate.

Unlike weights,

hyperparameters are **not learned** during training.

| Batch Size | Characteristics |

# <font color='red'> 19. Mathematical Foundations </font>

Training seeks to minimise

$$
\boxed{
J(W,b).
}
$$

Callbacks do **not** alter the mathematics of optimisation.

Instead,

they control **how** the optimisation proceeds.

For example,

EarlyStopping halts optimisation when further iterations are unlikely to reduce

$$
J(W,b).
$$

ReduceLROnPlateau modifies

the learning rate

$$
\eta
$$

during optimisation,

changing the update rule

$$
\boxed{
W
\leftarrow
W
-
\eta
\nabla J.
}
$$

The mathematical objective remains unchanged.

# <font color='orange'> 20. Common Misconceptions </font>

### Misconception 1

> More epochs always improve performance.

**False.**

After a certain point,

additional epochs may increase overfitting and reduce performance on unseen data.

---

### Misconception 2

> EarlyStopping means the model is underfitting.

**False.**

EarlyStopping aims to stop training near the point where validation performance is best, not necessarily when the training loss is lowest.

---

### Misconception 3

> Saving only the weights is equivalent to saving the whole model.

**Not always.**

Saving only the weights requires the model architecture to be recreated before loading. Saving the entire model preserves both the architecture and the trained parameters.
