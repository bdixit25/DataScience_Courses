# <font color='blue'> Chapter 9: Gradient Descent </font>

Once a neural network has made a prediction and the loss has been calculated, the next objective is to improve the model.

The parameters of the neural network, namely the weights and biases, determine the predictions produced during forward propagation. If the predictions are poor, then these parameters must be adjusted.

The challenge is determining **how** the parameters should change in order to reduce the loss.

Gradient Descent provides the answer.

Gradient Descent is one of the most important optimization algorithms in Machine Learning and Deep Learning. It uses information from calculus to determine the direction in which the loss decreases most rapidly, allowing the neural network to gradually improve its predictions.

Almost every modern deep learning algorithm is based upon Gradient Descent or one of its extensions.

---



# <font color='orange'> 1. Motivation </font>

Imagine standing on a mountain in thick fog.

Your objective is to reach the bottom of the valley.

Unfortunately,

you cannot see the entire landscape.

Instead,

you can only observe

the slope immediately beneath your feet.

The sensible strategy is

- determine which direction slopes downward,
- take a small step,
- repeat the process.

Eventually,

you arrive at the lowest point.

Gradient Descent follows exactly the same principle.

Instead of descending a mountain,

it descends the **loss function**.

---



# <font color='orange'> 2. The Optimization Problem </font>

Suppose the loss function depends upon a parameter

$$
\theta.
$$

Our objective is

$$
\boxed{
\min_{\theta} J(\theta),
}
$$

where

- $\theta$ represents the model parameters,
- $J(\theta)$ is the cost function.

Training a neural network is therefore an optimization problem.

---



# <font color='orange'> 3. The Idea Behind Gradient Descent </font>

Suppose we are currently at

$$
\theta.
$$

Should we move

- left?
- right?
- by how much?

Calculus provides the answer.

The derivative

$$
\frac{dJ}{d\theta}
$$

tells us

- the direction in which the function increases,
- how rapidly it increases.

Therefore,

to decrease the loss,

we move in the **opposite direction**.

Hence,

$$
\boxed{
\text{Move opposite to the gradient.}
}
$$

---



# <font color='orange'> 4. Understanding the Gradient </font>

For a function

$$
J(\theta),
$$

the derivative

$$
\frac{dJ}{d\theta}
$$

represents the slope.

If

$$
\frac{dJ}{d\theta}>0,
$$

the function is increasing.

Move left.

---

If

$$
\frac{dJ}{d\theta}<0,
$$

the function is decreasing.

Move right.

---

If

$$
\frac{dJ}{d\theta}=0,
$$

we have reached

- a minimum,
- a maximum,
- or a saddle point.

Gradient Descent seeks the minimum.

---



# <font color='orange'> 5. Deriving the Gradient Descent Update Rule </font>

Suppose we wish to move from

$$
\theta
$$

to

$$
\theta+\Delta\theta.
$$

Using the first-order Taylor expansion,

$$
J(\theta+\Delta\theta)
\approx
J(\theta)
+
\frac{dJ}{d\theta}
\Delta\theta.
$$

We want

$$
J(\theta+\Delta\theta)
<
J(\theta).
$$

Therefore,

the second term should be negative.

A natural choice is

$$
\boxed{
\Delta\theta
=
-
\eta
\frac{dJ}{d\theta},
}
$$

where

$$
\eta>0
$$

is called the **learning rate**.

Substituting,

$$
J(\theta+\Delta\theta)
\approx
J(\theta)
-
\eta
\left(
\frac{dJ}{d\theta}
\right)^2.
$$

Since

$$
\left(
\frac{dJ}{d\theta}
\right)^2\ge0,
$$

the loss decreases whenever

$$
\eta>0.
$$

Thus,

the parameter update becomes

$$
\boxed{
\theta_{\text{new}}
=
\theta
-
\eta
\frac{dJ}{d\theta}.
}
$$

This is the Gradient Descent algorithm.

---



# <font color='orange'> 6. Learning Rate </font>

The quantity

$$
\eta
$$

is called the **learning rate**.

It determines the size of each step.

---



# <font color='orange'> 7. Gradient Descent in Multiple Dimensions </font>

Neural networks contain thousands or even millions of parameters.

Instead of one parameter,

we now have

$$
\boldsymbol{\theta}
=
(\theta_1,\theta_2,\ldots,\theta_n).
$$

The derivative becomes the **gradient vector**

$$
\boxed{
\nabla J
=
\begin{bmatrix}
\frac{\partial J}{\partial\theta_1}\\
\frac{\partial J}{\partial\theta_2}\\
\vdots\\
\frac{\partial J}{\partial\theta_n}
\end{bmatrix}.
}
$$

The update rule becomes

$$
\boxed{
\boldsymbol{\theta}
=
\boldsymbol{\theta}
-
\eta
\nabla J.
}
$$

This equation is used throughout Deep Learning.

---



# <font color='orange'> 8. Geometric Interpretation </font>

The gradient points in the direction of **steepest ascent**.

Therefore,

its negative,

$$
-\nabla J,
$$

points in the direction of **steepest descent**.

This is why Gradient Descent always updates the parameters using

$$
-\nabla J.
$$

---



# <font color='orange'> 9. Example </font>

Consider

$$
J(\theta)
=
\theta^2.
$$

Its derivative is

$$
\frac{dJ}{d\theta}
=
2\theta.
$$

Suppose

$$
\theta=5,
$$

and

$$
\eta=0.1.
$$

The update becomes

$$
\theta_{\text{new}}
=
5
-
0.1(10)
=
4.
$$

Repeating,

$$
4
\rightarrow
3.2
\rightarrow
2.56
\rightarrow
\cdots
\rightarrow
0.
$$

The algorithm gradually approaches the minimum.

---



# <font color='orange'> 10. Why Does Gradient Descent Work? </font>

The gradient tells us

how sensitive the loss is to changes in the parameters.

If the slope is steep,

large improvements are possible.

If the slope is small,

only minor adjustments are needed.

Gradient Descent continuously follows the local slope until the loss can no longer be reduced.

---



# <font color='orange'> 11. Limitations </font>

Gradient Descent is powerful,

but not perfect.

Potential difficulties include

- local minima,
- saddle points,
- plateaus,
- slow convergence,
- poor learning-rate selection.

Modern optimizers address many of these challenges.

---

# <font color='red'> 12. Mathematical Foundations </font>

## Why Does the Negative Gradient Give the Steepest Descent?

Consider a small step

$$
\Delta\boldsymbol{\theta}.
$$

Using the multivariable Taylor expansion,

$$
J(\boldsymbol{\theta}+\Delta\boldsymbol{\theta})
\approx
J(\boldsymbol{\theta})
+
\nabla J^T
\Delta\boldsymbol{\theta}.
$$

The change in the cost function is

$$
\Delta J
=
\nabla J^T
\Delta\boldsymbol{\theta}.
$$

Suppose the step has fixed length

$$
||\Delta\boldsymbol{\theta}||=c.
$$

Using the dot product,

$$
\Delta J
=
||\nabla J||
\,c\,
\cos\phi,
$$

where

$$
\phi
$$

is the angle between

$$
\nabla J
$$

and

$$
\Delta\boldsymbol{\theta}.
$$

The greatest decrease occurs when

$$
\cos\phi=-1,
$$

that is,

$$
\phi=180^\circ.
$$

Therefore,

the optimal direction is

$$
\boxed{
\Delta\boldsymbol{\theta}
\propto
-\nabla J.
}
$$

This proves that the negative gradient gives the direction of steepest descent.

---



# <font color='orange'> 13. Gradient Descent Algorithm </font>

```
Initialize weights randomly

↓

Forward Propagation

↓

Compute Loss

↓

Compute Gradient

↓

Update Parameters

↓

Repeat until convergence
```

This loop forms the basis of neural network training.

---



# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> Gradient Descent always finds the global minimum.

**False.**

For non-convex loss functions, Gradient Descent may converge to a local minimum or saddle point.

---

### Misconception 2

> A larger learning rate always speeds up training.

**False.**

An excessively large learning rate can cause the optimization to overshoot the minimum or diverge entirely.

---

### Misconception 3

> Gradient Descent updates only the weights.

**False.**

Both the **weights** and the **biases** are trainable parameters and are updated during optimization.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Gradient | Direction of steepest increase of the loss |
| Negative Gradient | Direction of steepest decrease |
| Learning Rate ($\eta$) | Controls the size of each parameter update |
| Gradient Descent | Iterative optimization algorithm |
| Update Rule | $\theta \leftarrow \theta - \eta \nabla J$ |
| Objective | Minimise the loss function |

> **Key Insight:** Gradient Descent is the optimization engine that enables neural networks to learn. By repeatedly computing the gradient of the loss function and updating the parameters in the direction of the negative gradient, the network progressively reduces its prediction error. This simple iterative procedure underlies nearly every modern deep learning algorithm and provides the mathematical foundation for more advanced optimizers such as Stochastic Gradient Descent, Momentum, RMSProp, and Adam.
