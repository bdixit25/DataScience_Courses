# <font color='blue'> Chapter 8: Loss Functions (Cost Functions) </font>

A neural network learns by making predictions and comparing them with the true values. If the predictions are inaccurate, the network must determine **how wrong** they are before it can improve.

This measure of prediction error is called the **loss function**.

The loss function quantifies the difference between the predicted output and the true output. During training, the objective of the neural network is to minimize this loss by adjusting its weights and biases.

Every learning algorithm in Deep Learning ultimately attempts to solve the same optimization problem:

> **Find the parameters that minimize the loss function.**

For this reason, the loss function is often called the **objective function**, **error function**, or **cost function**.

Although these terms are sometimes used interchangeably, they have slightly different meanings, which we shall discuss in this chapter.

---

# <font color='orange'> 1. Motivation </font>

Suppose we build a neural network to predict house prices.

The true selling price is

$$
£350,000.
$$

The network predicts

$$
£420,000.
$$

Clearly,

the prediction is incorrect.

But how incorrect?

We need a numerical quantity that measures the error.

This quantity is called the **loss**.

A small loss indicates good predictions.

A large loss indicates poor predictions.

The goal of training is to make the loss as small as possible.

---

# <font color='orange'> 2. What is a Loss Function? </font>

A **loss function** measures the error for **a single training example**.

Suppose

- the true value is

$$
y,
$$

- the prediction is

$$
\hat y.
$$

Then

$$
\boxed{
L(y,\hat y)
}
$$

represents the loss.

The exact mathematical expression depends upon the learning problem.

---

# <font color='orange'> 3. Loss vs Cost Function </font>

Although many books use these terms interchangeably,

there is a subtle distinction.

### Loss Function

Measures the error for **one observation**.

$$
L(y,\hat y)
$$

---

### Cost Function

Measures the average loss over the **entire training dataset**.

Suppose

there are

$n$

training examples.

Then

$$
\boxed{
J
=
\frac1n
\sum_{i=1}^{n}
L(y_i,\hat y_i).
}
$$

Here,

- $L$ measures individual prediction errors,
- $J$ measures overall model performance.

During training,

gradient descent minimizes

$$
J.
$$

---

# <font color='orange'> 4. Why Do We Need a Loss Function? </font>

Imagine asking a student to improve their exam performance.

If you simply say

> "Do better."

they have no objective.

Instead,

suppose you tell them

"You scored

52%.

Try to reach

90%."

Now the student has a measurable goal.

The loss function provides exactly this objective for the neural network.

It tells the network

how far it is from the desired answer.

---

# <font color='orange'> 5. Mean Squared Error (MSE) </font>

For regression problems,

the most common loss function is the **Mean Squared Error (MSE)**.

For a single observation,

the squared error is

$$
\boxed{
L=(y-\hat y)^2.
}
$$

For the entire dataset,

$$
\boxed{
MSE
=
\frac1n
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

---

## Why Square the Errors?

Squaring has several advantages.

- Positive and negative errors cannot cancel.
- Large errors receive greater penalty.
- The function is smooth and differentiable.
- It is mathematically convenient for optimization.

---

# <font color='orange'> 6. Mean Absolute Error (MAE) </font>

Another common regression loss is

$$
\boxed{
MAE
=
\frac1n
\sum_{i=1}^{n}
|y_i-\hat y_i|.
}
$$

Unlike MSE,

the error increases linearly.

Advantages

- robust to outliers,
- easier to interpret.

Disadvantages

- not differentiable at

$$
0.
$$

---

# <font color='orange'> 7. Binary Cross-Entropy Loss </font>

For binary classification,

Mean Squared Error is usually not the best choice.

Instead,

we use

**Binary Cross-Entropy**.

The loss for one observation is

$$
\boxed{
L
=
-
\left[
y\log(\hat y)
+
(1-y)\log(1-\hat y)
\right].
}
$$

where

- $y\in\{0,1\}$,
- $\hat y$ is the predicted probability.

This loss becomes

very small

when the prediction is correct,

and

very large

when the prediction is confidently wrong.

---

# <font color='orange'> 8. Categorical Cross-Entropy </font>

For multi-class classification,

the loss becomes

$$
\boxed{
L
=
-
\sum_{k=1}^{K}
y_k
\log(\hat y_k).
}
$$

where

- $K$ is the number of classes,
- $y_k$ is the true label (usually one-hot encoded),
- $\hat y_k$ is the predicted probability.

This loss is almost universally used with the Softmax activation function.

---

# <font color='orange'> 9. Visualising the Effect of Different Losses </font>

Suppose the true value is

$$
y=10.
$$

| Prediction | Squared Error | Absolute Error |
|:---:|:---:|:---:|
| 10 | 0 | 0 |
| 9 | 1 | 1 |
| 8 | 4 | 2 |
| 7 | 9 | 3 |
| 5 | 25 | 5 |

Notice

- MSE penalises large errors much more heavily.
- MAE grows linearly.

This difference influences how the model learns.

---

# <font color='orange'> 10. The Optimization Objective </font>

Training a neural network means finding the parameters

$$
W,\;b
$$

that minimise the cost function.

Mathematically,

$$
\boxed{
\min_{W,b}
J(W,b).
}
$$

Notice that

the loss depends upon the network parameters,

since

the predictions

$$
\hat y
$$

are produced through forward propagation.

Thus,

training becomes an optimization problem.

---

# <font color='orange'> 11. Relationship to Forward Propagation </font>

Forward propagation computes

$$
\hat y.
$$

The loss function compares

$$
\hat y
$$

with

$$
y.
$$

```
Input

↓

Forward Propagation

↓

Prediction

↓

Loss Function

↓

Error
```

The next chapter will explain

how this error is used to update the weights.

---

# <font color='orange'> 12. Choosing the Correct Loss Function </font>

| Problem | Loss Function |
|:---|:---|
| Linear Regression | Mean Squared Error |
| Robust Regression | Mean Absolute Error |
| Binary Classification | Binary Cross-Entropy |
| Multi-Class Classification | Categorical Cross-Entropy |

Selecting an appropriate loss function is essential because it determines **what the network learns to optimize**.

---

# <font color='red'> 13. Mathematical Foundations </font>

## Why Does MSE Use the Mean?

Suppose we have

$n$

training examples.

If we simply summed the squared errors,

the total error would grow with the dataset size.

Dividing by

$n$

produces

an average error

that is independent of the number of training samples.

Hence,

$$
\boxed{
MSE
=
\frac1n
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

This makes it easier to compare models trained on datasets of different sizes.

---

## Derivative of the Mean Squared Error

Consider the loss

$$
L
=
(y-\hat y)^2.
$$

Differentiate with respect to

$$
\hat y.
$$

Using the chain rule,

$$
\boxed{
\frac{\partial L}
{\partial\hat y}
=
-2(y-\hat y)
=
2(\hat y-y).
}
$$

This derivative tells us

how the loss changes

when the prediction changes.

It will appear repeatedly during **Backpropagation**.

---

# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> Accuracy can always replace the loss function.

**False.**

Accuracy is not differentiable,

so it cannot be optimized directly using gradient descent.

---

### Misconception 2

> Lower loss always means higher accuracy.

Not necessarily.

A model may have the same accuracy but very different predicted probabilities, resulting in different cross-entropy losses.

---

### Misconception 3

> There is one universal loss function.

**False.**

The appropriate loss depends on the learning task.

Regression and classification require different objective functions.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Loss Function | Measures prediction error for one observation |
| Cost Function | Average loss over the entire dataset |
| Mean Squared Error | Standard loss for regression |
| Mean Absolute Error | Robust regression loss |
| Binary Cross-Entropy | Loss for binary classification |
| Categorical Cross-Entropy | Loss for multi-class classification |
| Optimization Objective | Minimise the cost function |

> **Key Insight:** A neural network cannot learn unless it has an objective. The loss function provides this objective by measuring the discrepancy between predictions and true values. Training is therefore an optimization problem in which the network repeatedly adjusts its parameters to minimise the cost function. Once the loss has been computed, the next challenge is determining **how to change the weights** to reduce it—a problem solved by gradient descent and backpropagation.
