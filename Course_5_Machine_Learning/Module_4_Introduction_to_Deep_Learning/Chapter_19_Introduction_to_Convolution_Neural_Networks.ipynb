# <font color='blue'> Chapter 19: Introduction to Convolutional Neural Networks (CNNs) </font>

In the previous chapters, we studied feedforward neural networks (also called **Multilayer Perceptrons (MLPs)**), where every neuron in one layer is connected to every neuron in the next layer.

Although these networks perform well on tabular data, they become inefficient when applied to images.

Convolutional Neural Networks (CNNs) were developed to overcome these limitations. Instead of treating every pixel independently, CNNs exploit the spatial structure of images by learning local features such as edges, textures, corners, and shapes.

Today, CNNs are among the most successful architectures in Deep Learning and have transformed the fields of computer vision, medical imaging, robotics, autonomous vehicles, and remote sensing.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to classify handwritten digits.

Each image has size

$$
28\times28.
$$

A fully connected neural network first converts the image into a vector of length

$$
28\times28 = 784.
$$

```
28 × 28 Image

↓

Flatten

↓

784 Numbers

↓

Dense Layer
```

Flattening destroys the spatial relationships between neighbouring pixels.

For example,

two adjacent pixels become just two numbers in a long vector.

The network no longer knows that they were neighbours in the original image.

However,

spatial relationships are precisely what define visual patterns.

---



# <font color='orange'> 2. Why Fully Connected Networks Are Inefficient </font>

Consider an image of size

$$
256\times256\times3,
$$

where

- 256 = image height,
- 256 = image width,
- 3 = RGB colour channels.

The total number of input values is

$$
256\times256\times3
=
196\,608.
$$

Suppose the first hidden layer contains

1000 neurons.

The number of weights is

$$
196\,608
\times
1000
=
196\,608\,000.
$$

Nearly

200 million

parameters are required before the network has even processed the first hidden layer.

Such networks are

- memory intensive,
- computationally expensive,
- prone to overfitting.

Clearly,

a different approach is needed.

---



# <font color='orange'> 3. The Key Idea Behind CNNs </font>

Instead of connecting every neuron to every pixel,

CNNs use small filters that move across the image.

```
Image

□□□□□□□□

□□■□□□

□□□□□□□□

↓

Small Filter

■■■

■■■

■■■

↓

Moves Across Image
```

Each filter examines only a small neighbourhood of pixels at a time.

This allows the network to learn local features while using far fewer parameters.

---



# <font color='orange'> 4. Local Receptive Fields </font>

Each neuron in a convolutional layer sees only a small region of the input image.

This region is called the **local receptive field**.

For example,

a

$$
3\times3
$$

filter examines only

nine neighbouring pixels.

```
□□□□□□□□

□■■■□□□□

□■■■□□□□

□■■■□□□□

□□□□□□□□
```

Instead of analysing the entire image,

the neuron focuses on a small local region.

---



# <font color='orange'> 5. Feature Extraction </font>

Different filters learn different visual patterns.

For example,

one filter may detect

- horizontal edges,

another may detect

- vertical edges,

while others detect

- corners,
- curves,
- textures.

As information moves through deeper layers,

simple features combine into more complex features.

```
Pixels

↓

Edges

↓

Corners

↓

Shapes

↓

Objects

↓

Prediction
```

This hierarchical feature learning is one of the defining characteristics of CNNs.

---



# <font color='orange'> 6. Parameter Sharing </font>

Unlike Dense layers,

the same filter is applied across the entire image.

Suppose we have a single

$$
3\times3
$$

filter.

Instead of learning a different set of weights for every image location,

the filter uses the same

nine weights

everywhere.

This principle is known as **parameter sharing**.

Parameter sharing greatly reduces the number of trainable parameters.

---



# <font color='orange'> 7. Translation Equivariance </font>

Suppose an image contains a cat.

If the cat moves slightly to the left or right,

we would still like the network to recognise it.

Because the same filter is applied across the image,

CNNs naturally respond to the same feature regardless of its location.

This property is known as **translation equivariance**.

It is one of the major advantages of convolutional layers over fully connected layers.

---



# <font color='orange'> 8. Typical CNN Architecture </font>

A convolutional neural network usually consists of

```
Input Image

↓

Convolution

↓

Activation

↓

Pooling

↓

Convolution

↓

Activation

↓

Pooling

↓

Flatten

↓

Dense

↓

Output
```

The early layers learn simple features,

while deeper layers learn increasingly abstract representations.

---



# <font color='orange'> 9. Advantages of CNNs </font>

Compared with fully connected networks,

CNNs

- require fewer parameters,
- exploit spatial information,
- learn hierarchical features,
- are more computationally efficient,
- generalize better for image data.

These advantages have made CNNs the standard architecture for image analysis.

---



# <font color='orange'> 10. Applications of CNNs </font>

CNNs are widely used in

- image classification,
- facial recognition,
- medical image analysis,
- object detection,
- image segmentation,
- handwriting recognition,
- satellite imagery,
- autonomous driving.

Although originally developed for images,

CNNs have also been applied to audio, time series, and certain natural language processing tasks.

---

# <font color='red'> 11. Mathematical Foundations </font>

Suppose the input image is represented by a matrix

$$
I.
$$

A convolutional filter is represented by a smaller matrix

$$
K.
$$

The output of the convolution is another matrix,

called the **feature map**,

which is obtained by sliding the filter across the image and computing a weighted sum at each location.

Symbolically,

$$
\boxed{
\text{Feature Map}
=
I * K,
}
$$

where

$$
*
$$

denotes the convolution operation.

In practice, many deep learning libraries implement **cross-correlation** rather than the strict mathematical definition of convolution. Cross-correlation applies the filter directly without flipping it first. Since the filter weights are learned during training, this distinction has no practical effect on the model's ability to learn.

The precise computation of the convolution operation will be derived in the next chapter.

# <font color='orange'> 12. Common Misconceptions </font>

### Misconception 1

> CNNs only work for images.

**False.**

Although CNNs were developed for computer vision, they have also been successfully applied to audio signals, one-dimensional sensor data, genomics, and some natural language processing tasks.

---

### Misconception 2

> CNNs automatically understand images.

**False.**

CNNs begin with randomly initialized filters. During training, backpropagation adjusts these filters so that they respond to informative visual patterns.

---

### Misconception 3

> Flattening an image preserves all visual information.

**False.**

Flattening preserves the pixel values but destroys the explicit spatial arrangement of neighbouring pixels. CNNs are designed to exploit this spatial structure before any flattening occurs.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Convolutional Neural Network | Neural network designed for spatially structured data |
| Local Receptive Field | Small region of the image processed by one neuron |
| Filter (Kernel) | Small matrix that detects local features |
| Parameter Sharing | Same filter weights applied across the image |
| Feature Map | Output produced by applying a filter |
| Translation Equivariance | Ability to detect a feature regardless of its position |
| Hierarchical Features | Simple features combine into increasingly complex representations |

> **Key Insight:** Convolutional Neural Networks overcome the limitations of fully connected networks by exploiting the spatial structure of images. Instead of learning separate parameters for every pixel connection, CNNs use small shared filters that detect local patterns throughout the image. This dramatically reduces the number of parameters while enabling the network to learn increasingly complex visual features, making CNNs the foundation of modern computer vision.
