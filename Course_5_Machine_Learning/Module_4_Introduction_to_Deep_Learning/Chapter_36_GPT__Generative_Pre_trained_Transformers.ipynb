# <font color='blue'> Chapter 36: GPT – Generative Pre-trained Transformers </font>

Generative Pre-trained Transformers (GPT) are a family of language models based on the **Transformer Decoder** architecture.

Unlike BERT, which learns to understand language by predicting masked words, GPT learns to **generate language** by predicting the next token in a sequence.

This simple training objective allows GPT models to learn grammar, syntax, semantics, reasoning patterns, and factual knowledge directly from large collections of text.

Today, GPT-style models form the foundation of many modern Large Language Models (LLMs), including conversational AI systems, code assistants, and content generation tools.

---



# <font color='orange'> 1. Motivation </font>

Suppose we begin typing

```
The Earth revolves around the
```

A human naturally predicts

```
Sun.
```

GPT is trained to perform exactly this task.

Given

all previous tokens,

predict

the next token.

After generating one token,

that token becomes part of the input,

and the process repeats.

---



# <font color='orange'> 2. Decoder-Only Architecture </font>

Unlike the original Transformer,

GPT removes the encoder completely.

Its architecture becomes

```
Input Tokens

↓

Token Embeddings

↓

Positional Encoding

↓

Masked Multi-Head Self-Attention

↓

Feed-Forward Network

↓

Linear Layer

↓

Softmax

↓

Next Token Prediction
```

Only the Transformer Decoder stack remains.

---



# <font color='orange'> 3. Autoregressive Language Modelling </font>

GPT generates text

one token at a time.

Suppose the sentence is

```
The

cat

sat

on

the

mat.
```

Training proceeds as

Input

```
The
```

Target

```
cat
```

---

```
The cat
```

Target

```
sat
```

---

Input

```
The cat sat
```

Target

```
on
```

# <font color='orange'> 4. Why Masked Self-Attention? </font>

During prediction,

the model must **not** see future words.

Suppose the sentence is

```
The cat sat on the mat.
```

When predicting

```
sat,
```

the model may use

```
The

cat
```

but **not**

```
on

the

mat.
```

Therefore,

future positions are masked.

This is called

**Causal Masking**

or

**Masked Self-Attention**.

---



# <font color='orange'> 5. Causal Attention Mask </font>

For a sequence of length four,

the attention mask is

```
        1 2 3 4

1       ✓ ✗ ✗ ✗

2       ✓ ✓ ✗ ✗

3       ✓ ✓ ✓ ✗

4       ✓ ✓ ✓ ✓
```

Each token attends only to

itself

and

previous tokens.

Future tokens remain invisible.

---



# <font color='orange'> 6. Next Token Prediction </font>

Suppose the vocabulary contains

$$
V
$$

possible tokens.

For every position,

the model outputs

a probability distribution

over all vocabulary items.

Example

```
Input

The cat sat on the

↓

Probabilities

mat      0.81

chair    0.07

floor    0.04

roof     0.02

...
```

The token with the highest probability is often selected,

although alternative sampling strategies may also be used.

---



# <font color='orange'> 7. Mathematical Objective </font>

Suppose

a sequence is

$$
x_1,x_2,\ldots,x_n.
$$

GPT models the joint probability as

$$
\boxed{
P(x_1,x_2,\ldots,x_n)
=
\prod_{t=1}^{n}
P(x_t \mid x_1,\ldots,x_{t-1}).
}
$$

This equation is known as

the **chain rule of probability**.

Rather than predicting an entire sentence at once,

GPT predicts one token at a time,

conditioning on all previous tokens.

---



# <font color='orange'> 8. Training Loss </font>

The predicted probability for the correct token is maximized using

**Cross-Entropy Loss**.

For one token,

the loss is

$$
\boxed{
L
=
-\log P(x_t|x_{<t}).
}
$$

For an entire sequence,

the loss becomes

$$
\boxed{
L
=
-
\sum_{t=1}^{n}
\log
P(x_t|x_{<t}).
}
$$

Minimizing this loss encourages the model to assign high probability to the correct next token.

---



# <font color='orange'> 9. Text Generation </font>

After training,

GPT generates text iteratively.

```
Prompt

↓

Predict Next Token

↓

Append Token

↓

Predict Again

↓

Append Again

↓

Continue...
```

Each generated token becomes part of the input for the next prediction.

This autoregressive process continues until a stopping condition is reached.

---



# <font color='orange'> 10. Sampling Strategies </font>

The model produces probabilities,

not fixed answers.

Different decoding strategies produce different outputs.

Common approaches include

- Greedy Decoding
- Beam Search
- Top-k Sampling
- Top-p (Nucleus) Sampling
- Temperature Sampling

These methods control the balance between diversity and determinism during text generation.

---



# <font color='orange'> 11. GPT Evolution </font>

The GPT family has evolved through increasingly larger models and datasets.

| Model | Main Contribution |
|:---|:---|
| GPT-1 | Demonstrated the effectiveness of generative pretraining |
| GPT-2 | Showed impressive long-form text generation |
| GPT-3 | Introduced large-scale in-context learning with many more parameters |
| Later GPT-style models | Improved reasoning, multimodal capabilities, instruction following, and efficiency through architectural and training advances |

While details vary across generations, they all share the decoder-only autoregressive Transformer foundation.

---



# <font color='orange'> 12. GPT vs BERT </font>

| BERT | GPT |
|:---|:---|
| Encoder-only | Decoder-only |
| Bidirectional | Left-to-right autoregressive |
| Predicts masked words | Predicts next token |
| Designed for understanding | Designed for generation |
| Uses full context during encoding | Uses only previous context during generation |

Both models are based on Transformers,

but they optimize different learning objectives.

# <font color='red'> 14. Mathematical Foundations </font>

Let

$$
X=(x_1,x_2,\ldots,x_n)
$$

represent a sequence of tokens.

The decoder computes contextual representations

$$
H
=
\mathrm{Decoder}(X).
$$

A linear layer maps each hidden representation to vocabulary logits

$$
\boxed{
z_t
=
Wh_t+b.
}
$$

Applying the Softmax function gives

$$
\boxed{
P(x_t|x_{<t})
=
\frac{\exp(z_t)}
{\sum_j\exp(z_j)}.
}
$$

The training objective minimizes

$$
\boxed{
L
=
-
\sum_{t=1}^{n}
\log
P(x_t|x_{<t}).
}
$$

This objective encourages the model to predict the correct next token at every position in the sequence.

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> GPT memorizes every sentence it has seen.

**False.**

GPT learns statistical patterns and representations from data. While memorization can occur in some cases, especially for frequently repeated content, the model primarily learns to generalize rather than simply store complete sentences.

---

### Misconception 2

> GPT predicts an entire paragraph at once.

**False.**

GPT predicts one token at a time. Each newly generated token is added to the context before predicting the next one.

---

### Misconception 3

> GPT understands language exactly as humans do.

**False.**

GPT builds powerful statistical representations of language through next-token prediction. Although these representations enable impressive reasoning and generation, they are not equivalent to human understanding.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| GPT | Decoder-only Transformer model |
| Autoregressive Model | Predicts one token conditioned on previous tokens |
| Causal Mask | Prevents access to future tokens during training and generation |
| Next Token Prediction | Primary learning objective |
| Cross-Entropy Loss | Measures prediction error for the correct next token |
| Chain Rule | Decomposes sequence probability into conditional probabilities |
| Sampling Strategy | Method for selecting the next generated token |

> **Key Insight:** GPT uses a decoder-only Transformer trained with next-token prediction to model language autoregressively. By repeatedly predicting the next token conditioned on all previous tokens, the model learns rich linguistic and semantic representations that enable coherent text generation across a wide range of tasks. This simple yet powerful training objective underlies many modern generative AI systems.
