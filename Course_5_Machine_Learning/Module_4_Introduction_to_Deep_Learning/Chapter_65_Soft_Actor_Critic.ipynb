# <font color='blue'> Chapter 65: Soft Actor–Critic (SAC) </font>

Deep Deterministic Policy Gradient (DDPG) and Twin Delayed DDPG (TD3) learn deterministic policies for continuous action spaces.

Although highly successful,

deterministic policies often struggle with exploration.

Soft Actor–Critic (SAC) addresses this problem by learning a **stochastic policy**.

Rather than maximizing only the expected cumulative reward,

SAC maximizes both

- expected reward,
- policy entropy.

This encourages the agent to continue exploring while simultaneously learning high-reward behaviours.

As training progresses,

the amount of exploration naturally decreases,

leading to robust and sample-efficient learning.

Today,

SAC is one of the most widely used algorithms for robotics,

continuous control,

and autonomous systems.

---



# <font color='orange'> 1. Motivation </font>

Suppose a robot must grasp an object.

A deterministic policy always chooses

```
Grip Force

↓

25 N
```

A stochastic policy instead learns

```
24.8 N

↓

25.0 N

↓

25.2 N
```

Initially,

trying slightly different actions helps discover better strategies.

Eventually,

the policy becomes increasingly confident.

---



# <font color='orange'> 2. Why Deterministic Policies Can Be Limiting </font>

Suppose

the Actor always outputs

$$
25^\circ.
$$

If

$$
26^\circ
$$

is actually better,

the agent may never discover it.

A stochastic policy naturally explores nearby actions,

improving robustness.

---



# <font color='orange'> 3. Stochastic Policies </font>

Instead of outputting

one action,

the Actor outputs

a probability distribution.

$$
\boxed{
\pi_\theta(a|s).
}
$$

An action is sampled

from this distribution.

Thus,

the same state

may produce different actions on different occasions.

---



# <font color='orange'> 4. Entropy </font>

Entropy measures

the uncertainty

of a probability distribution.

Mathematically,

$$
\boxed{
\mathcal H(\pi)
=
-
\mathbb E
\left[
\log
\pi(a|s)
\right].
}
$$

Interpretation

- High entropy

↓

many actions are plausible.

- Low entropy

↓

one action dominates.

Entropy therefore measures

how exploratory

the policy is.

---



# <font color='orange'> 5. Maximum Entropy Reinforcement Learning </font>

Classical Reinforcement Learning maximizes

expected reward.

SAC instead maximizes

reward

plus

entropy.

The objective becomes

$$
\boxed{
J(\pi)
=
\sum_t
\mathbb E
\left[
R_t
+
\alpha
\mathcal H
(\pi(\cdot|S_t))
\right].
}
$$

The parameter

$$
\alpha
$$

controls

the importance of entropy.

---



# <font color='orange'> 6. Temperature Parameter </font>

The coefficient

$$
\alpha
$$

is called

the **temperature**.

Large

$$
\alpha
$$

↓

more exploration.

Small

$$
\alpha
$$

↓

more exploitation.

Modern SAC implementations

often learn

$$
\alpha
$$

automatically.

---



# <font color='orange'> 7. Actor–Critic Architecture </font>

SAC uses

an Actor

and

two Critic networks.

```
State

↓

Actor

↓

Probability Distribution

↓

Sample Action

↓

Twin Critics

↓

Q₁,Q₂
```

The twin Critics reduce overestimation,

just as in TD3.

---



# <font color='orange'> 8. Soft Bellman Equation </font>

Because entropy contributes to the objective,

the Bellman equation changes.

The soft target becomes

$$
\boxed{
Q(s,a)
=
r
+
\gamma
\left(
V(s')
\right),
}
$$

where

the state value now includes

an entropy bonus.

Conceptually,

future rewards are replaced by

future rewards

plus

future exploration.

---



# <font color='orange'> 9. Soft Value Function </font>

The value function becomes

$$
\boxed{
V(s)
=
\mathbb E
\left[
Q(s,a)
-
\alpha
\log
\pi(a|s)
\right].
}
$$

Notice

the entropy term

appears explicitly.

Actions with

both

high reward

and

reasonable uncertainty

receive larger values.

---



# <font color='orange'> 10. Critic Update </font>

The Critic minimizes

the Bellman error,

using targets that include

the entropy bonus.

Twin Critics are employed,

and the smaller estimate is typically used,

following the same philosophy as TD3.

---



# <font color='orange'> 11. Actor Update </font>

The Actor seeks

actions that

- maximize the Critic's estimate,
- maintain sufficient entropy.

Instead of converging immediately

to one deterministic action,

the policy gradually becomes more confident

as learning progresses.

---



# <font color='orange'> 12. Automatic Entropy Tuning </font>

One of SAC's major innovations

is that

the algorithm can learn

the best value of

$$
\alpha.
$$

If

the policy becomes

too deterministic,

entropy increases.

If

the policy becomes

too random,

entropy decreases.

Exploration therefore adjusts automatically.

---



# <font color='orange'> 13. Advantages </font>

Soft Actor–Critic

- learns stochastic policies,
- provides excellent exploration,
- is highly sample efficient,
- remains stable during training,
- performs well across many continuous-control tasks.

These advantages have made SAC one of the most popular reinforcement learning algorithms.

---



# <font color='orange'> 14. Applications </font>

SAC has been applied successfully to

- robotic manipulation,
- autonomous vehicles,
- drone control,
- industrial automation,
- continuous optimization,
- reinforcement learning research.

It is now considered a standard benchmark algorithm for continuous control.

---



# <font color='orange'> 15. Limitations </font>

Although powerful,

SAC

- requires multiple neural networks,
- is computationally more demanding than DDPG,
- introduces additional hyperparameters,
- can be more difficult to implement.

Nevertheless,

its robustness often outweighs these disadvantages.

---

# <font color='orange'> 17. TD3 vs SAC </font>

| Feature | TD3 | SAC |
|:---|:---:|:---:|
| Policy | Deterministic | Stochastic |
| Exploration | Action-space noise | Entropy maximization |
| Twin Critics | ✓ | ✓ |
| Automatic Exploration | ✗ | ✓ |
| Sample Efficiency | High | Very High |
| Stability | High | Very High |

SAC extends many of TD3's ideas while introducing entropy-regularized policy optimization.

---



# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> SAC tries to maximize randomness instead of reward.

**False.**

SAC maximizes **both** expected reward and entropy. Entropy encourages exploration, but actions that consistently lead to poor rewards are still discouraged.

---

### Misconception 2

> A stochastic policy means the agent behaves randomly forever.

**False.**

Although SAC maintains stochasticity, the learned probability distribution gradually concentrates on high-reward actions. The policy becomes more confident while still retaining enough uncertainty for effective exploration.

---

### Misconception 3

> The temperature parameter \(\alpha\) is always fixed.

**False.**

Modern implementations usually adjust \(\alpha\) automatically so that the policy maintains an appropriate level of entropy throughout training.

---

# <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Soft Actor–Critic | Entropy-regularized Actor–Critic algorithm |
| Stochastic Policy | Outputs a probability distribution over actions |
| Entropy | Measures the uncertainty of the policy |
| Maximum Entropy RL | Optimizes both reward and exploration |
| Temperature \(\alpha\) | Controls the balance between reward and entropy |
| Twin Critics | Reduce overestimation bias |
| Automatic Entropy Tuning | Learns the exploration level during training |

> **Key Insight:** Soft Actor–Critic extends the Actor–Critic framework by replacing deterministic policies with stochastic ones and by maximizing both expected reward and policy entropy. This maximum-entropy objective encourages efficient exploration while maintaining stable learning. Combined with twin Critics and automatic entropy tuning, SAC achieves outstanding performance and sample efficiency, making it one of the most successful algorithms for continuous control in modern reinforcement learning.
