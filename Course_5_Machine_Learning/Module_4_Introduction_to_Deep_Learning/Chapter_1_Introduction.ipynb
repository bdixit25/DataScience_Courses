# <font color='blue'> Chapter 1: Introduction to Neural Networks </font>

Neural Networks are among the most powerful machine learning models ever developed. They are capable of learning extremely complex relationships directly from data and have become the foundation of modern Artificial Intelligence (AI).

Today, neural networks power technologies such as

- image recognition,
- speech recognition,
- natural language processing,
- medical diagnosis,
- autonomous vehicles,
- recommendation systems,
- scientific simulations,
- reinforcement learning,
- and generative AI systems such as ChatGPT.

Unlike traditional machine learning algorithms, neural networks do not require humans to manually design complex features. Instead, they automatically learn useful representations from data through multiple layers of computation.

The field of **Deep Learning** refers to neural networks containing many hidden layers, allowing them to model highly nonlinear and hierarchical relationships.

Throughout this chapter we shall gradually develop neural networks from the simplest mathematical models, building the intuition required for the remainder of this course.

---

# <font color='orange'> 1. Motivation </font>

Imagine trying to predict the selling price of a house.

For many datasets, Linear Regression performs surprisingly well because it assumes that the relationship between the input variables and the target is approximately linear.

For example,

$$
\boxed{
Price
=
\beta_0
+
\beta_1
(\text{Area})
}
$$

works reasonably well if house prices increase approximately linearly with floor area.

However, many real-world relationships are far more complicated.

Examples include

- recognizing handwritten digits,
- detecting cancer in medical images,
- identifying objects in photographs,
- translating languages,
- speech recognition,
- predicting protein structures.

None of these problems can be described by a simple straight line.

Neural Networks were developed to learn these complex nonlinear relationships automatically.

---

# <font color='orange'> 2. From Linear Models to Neural Networks </font>

Recall that Linear Regression predicts

$$
\boxed{
\hat y
=
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+\cdots+
\beta_px_p.
}
$$

This equation represents a weighted combination of the input variables.

Notice something interesting.

The prediction consists of

- multiplying each input by a weight,
- adding the weighted values,
- producing an output.

This simple computation forms the basis of every neural network.

The only difference is that neural networks introduce **nonlinear activation functions** between successive weighted sums.

Instead of

$$
\hat y
=
Wx+b,
$$

a neural network computes

$$
\boxed{
\hat y
=
f(Wx+b),
}
$$

where

$f(\cdot)$

is called the **activation function**.

Although this appears to be a very small modification, it dramatically increases the expressive power of the model.

---

# <font color='orange'> 3. Why Do We Need Neural Networks? </font>

Suppose we wish to classify images of cats and dogs.

```
Image

↓

Extract features

↓

Predict
```

Traditional machine learning requires engineers to manually design useful features.

Examples include

- edge detectors,
- colour histograms,
- texture descriptors,
- object shapes.

This process is known as **feature engineering**.

Neural Networks remove much of this manual effort.

Instead,

they automatically learn increasingly sophisticated features during training.

A simplified view is

```
Image

↓

Edges

↓

Corners

↓

Textures

↓

Eyes

↓

Face

↓

Entire Animal

↓

Prediction
```

Each layer learns features that become progressively more abstract.

# <font color='orange'> 4. Inspiration from Biology </font>

Artificial Neural Networks were originally inspired by biological neurons found in the human brain.

A biological neuron consists of

- dendrites,
- cell body,
- axon,
- synapses.

```
Inputs

↓

[Dendrites]

↓

(Cell Body)

↓

Axon

↓

Output
```

The biological neuron receives electrical signals from many neighbouring neurons.

If the combined signal exceeds a certain threshold,

the neuron "fires"

and transmits a signal to other neurons.

Although modern neural networks are only loosely inspired by biology,

this simple idea motivated the development of artificial neurons.

---

# <font color='orange'> 5. Artificial Neuron </font>

An artificial neuron receives several numerical inputs,

multiplies each input by a weight,

adds them together,

adds a bias,

and finally applies an activation function.

Mathematically,

$$
\boxed{
z
=
w_1x_1
+
w_2x_2
+
\cdots
+
w_nx_n
+
b.
}
$$

The neuron output is

$$
\boxed{
a=f(z).
}
$$

where

- $x_i$ are the inputs,
- $w_i$ are the weights,
- $b$ is the bias,
- $z$ is the weighted sum,
- $f$ is the activation function,
- $a$ is the neuron output.

This equation is arguably the single most important equation in all of Deep Learning.

Everything else in this course builds upon it.

---

# <font color='orange'> 6. Understanding the Components </font>

### Inputs

The inputs represent information supplied to the model.

Examples include

- house area,
- age,
- salary,
- image pixels,
- sound amplitudes,
- word embeddings.

---

### Weights

Weights determine the importance of each input.

Large positive weights

increase the output.

Negative weights

decrease the output.

Weights are the parameters learned during training.

---

### Bias

The bias shifts the activation function left or right.

Without the bias,

the neuron would always pass through the origin,

greatly limiting its flexibility.

---

### Activation Function

The activation function introduces nonlinearity.

Without it,

multiple layers would collapse into one equivalent linear transformation.

This is why activation functions are essential.

We shall study them in detail later.

---

# <font color='orange'> 7. Why Nonlinearity is Necessary </font>

Suppose every layer performed only

$$
Wx+b.
$$

Then two consecutive layers become

$$
y
=
W_2(W_1x+b_1)+b_2.
$$

Expanding,

$$
y
=
W_2W_1x
+
W_2b_1
+
b_2.
$$

Define

$$
W=W_2W_1,
$$

and

$$
b=W_2b_1+b_2.
$$

Therefore,

$$
\boxed{
y
=
Wx+b.
}
$$

This is simply another linear model.

No matter how many purely linear layers are stacked together,

the result remains linear.

Therefore,

deep networks require nonlinear activation functions.

---

# <font color='orange'> 8. Layers of a Neural Network </font>

A neural network consists of several layers.

```
Input Layer

↓

Hidden Layer

↓

Hidden Layer

↓

Output Layer
```

### Input Layer

Receives the data.

No computation is performed here.

---

### Hidden Layers

Transform the data into increasingly useful representations.

Most learning occurs inside these layers.

---

### Output Layer

Produces the final prediction.

The choice of activation depends upon the problem.

Examples include

- Linear
- Sigmoid
- Softmax

---

# <font color='orange'> 9. Deep Learning </font>

A neural network containing many hidden layers is called a

**Deep Neural Network**.

The word

**deep**

simply refers to the number of hidden layers.

```
Input

↓

Layer 1

↓

Layer 2

↓

Layer 3

↓

Layer 4

↓

Prediction
```

Deep networks are capable of learning hierarchical representations that shallow models cannot easily capture.

---

# <font color='orange'> 10. Universal Approximation Theorem </font>

One of the most remarkable theoretical results in neural network theory is the **Universal Approximation Theorem**.

It states that

> A feedforward neural network with at least one hidden layer and a suitable nonlinear activation function can approximate any continuous function on a compact domain to arbitrary accuracy, provided it contains sufficiently many neurons.

This theorem explains why neural networks are extraordinarily flexible.

However,

it does **not** guarantee that

- training will be easy,
- the required network will be small,
- or that the optimal parameters can be found efficiently.

In practice, network architecture and optimization remain crucial.

# <font color='orange'> 11. Advantages of Neural Networks </font>

Neural Networks

- learn complex nonlinear relationships,
- automatically learn features,
- scale well to large datasets,
- work for structured and unstructured data,
- achieve state-of-the-art performance across many applications,
- can approximate extremely complicated functions.

---

# <font color='orange'> 12. Limitations </font>

Neural Networks

- require large datasets,
- are computationally expensive,
- require careful hyperparameter tuning,
- are often less interpretable than linear models,
- may overfit without regularization,
- usually require GPUs for large-scale training.

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Artificial Neuron | Fundamental computational unit of a neural network |
| Weight | Learns the importance of an input |
| Bias | Allows flexible shifting of the activation |
| Activation Function | Introduces nonlinearity |
| Hidden Layer | Learns intermediate representations |
| Deep Learning | Neural networks containing multiple hidden layers |
| Universal Approximation Theorem | Neural networks can approximate arbitrary continuous functions under suitable conditions |

> **Key Insight:** A neural network can be viewed as a generalized linear model in which each layer performs a linear transformation followed by a nonlinear activation. The linear transformations learn how to combine information, while the nonlinear activations allow the network to model highly complex relationships. This simple combination of linear algebra and nonlinear functions forms the foundation of modern deep learning.
