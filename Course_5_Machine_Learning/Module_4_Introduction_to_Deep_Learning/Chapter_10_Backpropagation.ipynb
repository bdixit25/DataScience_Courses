# <font color='blue'> Chapter 10: Backpropagation </font>

Backpropagation is the learning algorithm that enables neural networks to update their parameters efficiently.

After a forward pass through the network, the prediction is compared with the true target using a loss function. The network must then determine how each weight and bias contributed to the final error so that these parameters can be adjusted.

Computing these gradients independently would be extremely inefficient for modern neural networks containing millions of parameters.

Backpropagation solves this problem by repeatedly applying the **Chain Rule** of calculus, allowing the gradients to be computed efficiently layer by layer.

Together,

- Forward Propagation,
- Loss Functions,
- Gradient Descent,
- Backpropagation,

form the complete learning algorithm of a neural network.

---

# <font color='orange'> 1. Motivation </font>

Suppose a student takes an examination and scores

40%.

The student now asks

> "Which mistakes caused me to lose marks?"

Without identifying the source of the mistakes,

the student cannot improve.

A neural network faces exactly the same problem.

After computing the loss,

it must determine

- which weights contributed most to the error,
- by how much each weight should change,
- in which direction they should change.

Backpropagation answers these questions.

---

# <font color='orange'> 2. The Complete Learning Cycle </font>

Training a neural network consists of four repeated steps.

```
Training Data

↓

Forward Propagation

↓

Prediction

↓

Loss Function

↓

Backpropagation

↓

Gradient Descent

↓

Updated Parameters

↓

Repeat
```

Notice that

Forward Propagation computes predictions,

while

Backpropagation computes gradients.

---

# <font color='orange'> 3. Why Can't We Differentiate Everything Directly? </font>

Consider a network

```
Input

↓

Hidden Layer

↓

Output Layer

↓

Loss
```

The loss depends upon

- output,

which depends upon

- hidden activations,

which depend upon

- hidden weights,

which depend upon

- the input.

The dependencies are nested.

Therefore,

ordinary differentiation is insufficient.

Instead,

we must use the **Chain Rule**.

---

# <font color='orange'> 4. Review of the Chain Rule </font>

Suppose

$$
y=f(u),
$$

and

$$
u=g(x).
$$

Then

$$
y=f(g(x)).
$$

The Chain Rule states

$$
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}.
}
$$

This simple rule is the mathematical foundation of Backpropagation.

---

# <font color='orange'> 5. A Single Neuron </font>

Consider one neuron.

The computations are

$$
z=wx+b,
$$

$$
a=f(z),
$$

$$
L=L(a).
$$

Notice the dependency

```
w

↓

z

↓

a

↓

L
```

The loss depends on the weight only indirectly.

Applying the Chain Rule,

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\cdot
\frac{\partial a}{\partial z}
\cdot
\frac{\partial z}{\partial w}.
}
$$

This is the central idea behind Backpropagation.

---

# <font color='orange'> 6. Computing Each Term </font>

Suppose

$$
z=wx+b.
$$

Then

$$
\boxed{
\frac{\partial z}{\partial w}=x.
}
$$

---

Suppose

$$
a=\sigma(z).
$$

We already proved

$$
\boxed{
\frac{\partial a}{\partial z}
=
a(1-a).
}
$$

---

Suppose

Mean Squared Error

$$
L=\frac12(y-a)^2.
$$

Then

$$
\boxed{
\frac{\partial L}{\partial a}
=
a-y.
}
$$

Combining all three,

$$
\boxed{
\frac{\partial L}{\partial w}
=
(a-y)
\,
a(1-a)
\,
x.
}
$$

This tells us exactly how changing the weight affects the loss.

---

# <font color='orange'> 7. Updating the Weight </font>

Once the gradient has been computed,

Gradient Descent performs

$$
\boxed{
w_{\text{new}}
=
w
-
\eta
\frac{\partial L}{\partial w}.
}
$$

Thus,

Backpropagation computes the gradient,

while

Gradient Descent uses that gradient.

---

# <font color='orange'> 8. Backpropagation Through Multiple Layers </font>

Suppose the network contains

Input

↓

Hidden Layer

↓

Output Layer

Now the loss depends upon

both layers.

For the output layer,

the gradients are computed directly.

For hidden layers,

the error must first pass through later layers.

This is why the algorithm is called

**Backpropagation**.

The error propagates backwards through the network.

---

# <font color='orange'> 9. Error Propagation </font>

Suppose

Layer 3

produces an error.

Part of that error originated from

Layer 2.

Similarly,

Layer 2

received information from

Layer 1.

Therefore,

the error is propagated backwards

one layer at a time.

```
Prediction

↑

Layer 3

↑

Layer 2

↑

Layer 1

↑

Input
```

Notice that

the data moves forwards,

while

the error moves backwards.

---

# <font color='orange'> 10. Matrix Formulation </font>

Forward propagation computes

$$
\mathbf{z}^{(l)}
=
W^{(l)}
\mathbf{a}^{(l-1)}
+
\mathbf{b}^{(l)}.
$$

The activations are

$$
\mathbf{a}^{(l)}
=
f(\mathbf{z}^{(l)}).
$$

Define the error vector

$$
\boxed{
\delta^{(l)}
=
\frac{\partial L}
{\partial \mathbf{z}^{(l)}}.
}
$$

This quantity measures

how sensitive the loss is

to the weighted inputs

of layer

$l$.

---

# <font color='orange'> 11. Output Layer Error </font>

For the final layer,

the error is

$$
\boxed{
\delta^{(L)}
=
\frac{\partial L}
{\partial \mathbf{a}^{(L)}}
\odot
f'
\left(
\mathbf{z}^{(L)}
\right),
}
$$

where

$$
\odot
$$

denotes element-wise multiplication.

---

# <font color='orange'> 12. Hidden Layer Error </font>

For earlier layers,

the error becomes

$$
\boxed{
\delta^{(l)}
=
\left(
W^{(l+1)}
\right)^T
\delta^{(l+1)}
\odot
f'
\left(
\mathbf{z}^{(l)}
\right).
}
$$

This equation is one of the most important results in Deep Learning.

It explains why the error travels backwards through the network.

---

# <font color='orange'> 13. Computing the Gradients </font>

Once

$$
\delta^{(l)}
$$

is known,

the gradients become

Weights

$$
\boxed{
\frac{\partial L}
{\partial W^{(l)}}
=
\delta^{(l)}
\left(
\mathbf{a}^{(l-1)}
\right)^T.
}
$$

Biases

$$
\boxed{
\frac{\partial L}
{\partial \mathbf{b}^{(l)}}
=
\delta^{(l)}.
}
$$

These are precisely the quantities required by Gradient Descent.

---

# <font color='orange'> 14. Complete Backpropagation Algorithm </font>

```
Forward Propagation

↓

Prediction

↓

Compute Loss

↓

Compute Output Error

↓

Propagate Error Backwards

↓

Compute Gradients

↓

Update Weights

↓

Repeat
```

Every neural network library performs these steps automatically.

---

# <font color='orange'> 15. Why Backpropagation is Efficient </font>

A naive approach would compute

one derivative

for every weight independently.

Modern networks contain

millions

or even

billions

of parameters.

Backpropagation avoids repeated computations by

reusing intermediate derivatives.

This reduces the computational cost dramatically,

making deep learning practical.

---

# <font color='orange'> 16. Computational Complexity </font>

Remarkably,

Backpropagation has approximately the same computational complexity as Forward Propagation.

Thus,

training requires roughly

```
Forward Pass

+

Backward Pass

≈

2 × Forward Pass
```

This efficiency is one of the reasons neural networks can be trained on very large datasets.

---

# <font color='red'> 17. Mathematical Foundations </font>

The Chain Rule explains every equation in Backpropagation.

Suppose

$$
L
=
L(a),
$$

$$
a
=
f(z),
$$

$$
z
=
wx+b.
$$

Then

$$
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial a}
\cdot
\frac{\partial a}{\partial z}
\cdot
\frac{\partial z}{\partial w}.
$$

Each derivative answers a different question.

| Derivative | Interpretation |
|:---|:---|
| $\frac{\partial L}{\partial a}$ | How does the loss change when the activation changes? |
| $\frac{\partial a}{\partial z}$ | How does the activation change when the weighted sum changes? |
| $\frac{\partial z}{\partial w}$ | How does the weighted sum change when the weight changes? |

Multiplying these quantities together yields the total effect of the weight on the loss.

---

# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> Backpropagation updates the weights.

**False.**

Backpropagation computes gradients.

Gradient Descent updates the weights.

---

### Misconception 2

> The error travels forwards.

**False.**

The predictions travel forwards.

The error travels backwards.

---

### Misconception 3

> Backpropagation only works with sigmoid activation.

**False.**

Backpropagation applies to any differentiable activation function, including ReLU, tanh, GELU, and others.

---

# <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Forward Propagation | Computes predictions |
| Loss Function | Measures prediction error |
| Chain Rule | Computes gradients through composed functions |
| Backpropagation | Efficiently computes gradients for all trainable parameters |
| Gradient Descent | Updates parameters using the computed gradients |
| Error Vector ($\delta$) | Sensitivity of the loss to the weighted inputs |

> **Key Insight:** Backpropagation is the algorithm that enables neural networks to learn efficiently. By applying the Chain Rule repeatedly, it determines how much each weight and bias contributed to the final prediction error. These gradients are then used by Gradient Descent to update the network parameters. Together, Forward Propagation, Loss Functions, Backpropagation, and Gradient Descent form the complete learning cycle of every modern neural network.
