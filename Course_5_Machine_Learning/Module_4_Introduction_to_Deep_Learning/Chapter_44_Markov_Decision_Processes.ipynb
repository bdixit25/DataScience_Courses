# <font color='blue'> Chapter 44: Markov Decision Processes (MDPs) </font>

Reinforcement Learning problems involve an agent interacting with an environment over time.

To analyze these interactions mathematically, we require a formal framework that describes

- the possible situations,
- the available actions,
- how the environment changes,
- the rewards received.

This framework is called a **Markov Decision Process (MDP).**

Nearly every classical Reinforcement Learning algorithm assumes that the environment can be modeled as an MDP.

Understanding MDPs is therefore essential before studying Bellman Equations, Dynamic Programming, Q-Learning, and Deep Reinforcement Learning.

---



# <font color='orange'> 1. Motivation </font>

Imagine a robot navigating a maze.

At every intersection,

the robot must decide

- move left,
- move right,
- move forward,
- move backward.

Each decision changes the robot's position.

Some decisions bring it closer to the goal,

others lead to dead ends.

The robot must therefore learn

which action should be chosen in every situation.

An MDP provides a mathematical description of this decision-making problem.

---



# <font color='orange'> 2. The Markov Property </font>

The defining characteristic of an MDP is the **Markov Property**.

It states

> **The future depends only on the present state, not on the past history.**

Suppose

```
Past

↓

Current State

↓

Future
```

Once the current state is known,

earlier states provide no additional information for predicting the future.

Mathematically,

$$
\boxed{
P(s_{t+1}\mid s_t,s_{t-1},\ldots,s_0)
=
P(s_{t+1}\mid s_t).
}
$$

This is called the **memoryless property**.

The system is not literally forgetting the past. Instead, the current state is assumed to contain all the information needed to predict future behaviour.

---



# <font color='orange'> 3. Why is the Markov Property Useful? </font>

Suppose a chess player must choose a move.

The current board position already contains

- the locations of all pieces,
- whose turn it is,
- castling rights,
- en passant opportunities.

There is no need to replay the entire game.

Similarly,

in many Reinforcement Learning problems,

the current state summarizes everything necessary for making an optimal decision.

This simplification makes mathematical analysis tractable.

---



# <font color='orange'> 4. Components of an MDP </font>

A Markov Decision Process is defined by five components.

$$
\boxed{
(\mathcal S,\mathcal A,P,R,\gamma)
}
$$

where

| Symbol | Meaning |
|:---:|:---|
| \(\mathcal S\) | Set of all possible states |
| \(\mathcal A\) | Set of all possible actions |
| \(P\) | State transition probabilities |
| \(R\) | Reward function |
| \(\gamma\) | Discount factor |

Together,

these completely specify the Reinforcement Learning problem.

---



# <font color='orange'> 5. States </font>

The state space

$$
\boxed{
\mathcal S
}
$$

contains every possible situation the agent may encounter.

Examples

Robot navigation

```
Grid Position
```

Chess

```
Every legal board configuration
```

Autonomous driving

```
Vehicle position

Speed

Nearby vehicles

Traffic signals
```

The current state at time

$$
t
$$

is written

$$
\boxed{
s_t.
}
$$

---



# <font color='orange'> 6. Actions </font>

For every state,

the agent chooses an action.

The action space is

$$
\boxed{
\mathcal A.
}
$$

Examples

Robot

```
Left

Right

Forward

Backward
```

Chess

```
All legal moves
```

The selected action is denoted

$$
\boxed{
a_t.
}
$$

---



# <font color='orange'> 7. State Transition Probabilities </font>

After taking an action,

the environment changes state.

This is described by the transition probability

$$
\boxed{
P(s'|s,a).
}
$$

Interpretation

> Starting in state

$$
s,
$$

taking action

$$
a,
$$

what is the probability of reaching state

$$
s'?
$$

For deterministic environments,

this probability is either

0

or

1.

For stochastic environments,

multiple next states may be possible.

---



# <font color='orange'> 8. Reward Function </font>

The reward function specifies the immediate feedback after an action.

It is commonly written as

$$
\boxed{
R(s,a)
}
$$

or, in a more general form,

$$
\boxed{
R(s,a,s').
}
$$

Examples

```
Reach Goal

↓

+100
```

```
Fall Into Trap

↓

-50
```

```
Take One Step

↓

-1
```

Rewards define what the agent is trying to achieve.

---



# <font color='orange'> 9. Discount Factor </font>

Future rewards are discounted by

$$
\boxed{
\gamma,
}
$$

where

$$
0\le\gamma\le1.
$$

Interpretation

| Value | Meaning |
|:---:|:---|
| \(0\) | Only immediate rewards matter |
| \(0.5\) | Moderate emphasis on future rewards |
| \(0.99\) | Long-term planning is important |
| \(1\) | All future rewards weighted equally (for suitable tasks) |

The discount factor ensures that the total return remains well-defined in many continuing tasks and controls how far into the future the agent plans.

---



# <font color='orange'> 10. Episodic and Continuing Tasks </font>

Some tasks naturally terminate.

Example

```
Maze

↓

Goal Reached

↓

Episode Ends
```

These are called

**episodic tasks**.

Other tasks continue indefinitely.

Example

```
Traffic Signal Control
```

These are called

**continuing tasks**.

The mathematical treatment differs slightly,

but both can be modeled using MDPs.

---



# <font color='orange'> 11. Grid World Example </font>

Consider the following environment.

```
+---+---+---+

| S |   | G |

+---+---+---+

|   | X |   |

+---+---+---+

|   |   |   |

+---+---+---+
```

where

- S = Start
- G = Goal
- X = Obstacle

Possible actions

- Up
- Down
- Left
- Right

Each movement changes the state,

and the reward depends on the resulting position.

This simple environment is widely used for illustrating Reinforcement Learning algorithms.

---



# <font color='orange'> 12. Deterministic vs Stochastic Environments </font>

In a deterministic environment,

every action has exactly one outcome.

Example

```
Move Right

↓

Always Move Right
```

In a stochastic environment,

the same action may lead to different outcomes.

Example

```
Move Right

↓

80% Right

10% Up

10% Down
```

Many real-world problems,

including robotics and finance,

are stochastic.

---

# <font color='red'> 13. Mathematical Foundations </font>

An MDP is formally defined as

$$
\boxed{
(\mathcal S,\mathcal A,P,R,\gamma).
}
$$

The transition probability satisfies

$$
\boxed{
P(s'|s,a)
=
\Pr(S_{t+1}=s'
\mid
S_t=s,
A_t=a).
}
$$

The Markov Property states

$$
\boxed{
P(s_{t+1}|s_t,s_{t-1},\ldots)
=
P(s_{t+1}|s_t).
}
$$

The expected discounted return is

$$
\boxed{
G_t
=
\sum_{k=0}^{\infty}
\gamma^k
r_{t+k+1}.
}
$$

The objective of Reinforcement Learning is to find a policy

$$
\pi
$$

that maximizes

$$
\boxed{
\mathbb E_\pi[G_t].
}
$$

This optimization problem is defined entirely within the MDP framework.

---



# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> The Markov Property means the system has no memory.

**False.**

The Markov Property means that the **state representation** contains all the information needed to predict the future. The underlying process may depend on past events, but those effects are assumed to be fully summarized by the current state.

---

### Misconception 2

> Every Reinforcement Learning problem is deterministic.

**False.**

Many environments are stochastic, meaning that the same action can lead to different outcomes with different probabilities.

---

### Misconception 3

> The reward function tells the agent which action to take.

**False.**

The reward function only provides feedback after an action. The agent must learn an effective policy through experience.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Markov Decision Process (MDP) | Mathematical framework for sequential decision making |
| Markov Property | Future depends only on the current state |
| State Space (\(\mathcal S\)) | All possible states of the environment |
| Action Space (\(\mathcal A\)) | All actions available to the agent |
| Transition Probability (\(P(s'|s,a)\)) | Probability of moving to the next state |
| Reward Function (\(R\)) | Immediate feedback received after an action |
| Discount Factor (\(\gamma\)) | Determines the importance of future rewards |
| Episodic Task | Task with a terminal state |
| Continuing Task | Task without a natural termination |

> **Key Insight:** A Markov Decision Process provides the mathematical framework for Reinforcement Learning by describing the environment in terms of states, actions, transition probabilities, rewards, and a discount factor. The defining Markov Property assumes that the current state contains all the information needed to predict future behaviour, allowing complex sequential decision-making problems to be analyzed and solved using rigorous mathematical methods.
