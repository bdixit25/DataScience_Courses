# <font color='blue'> Appendix A7: Modern Alignment Methods </font>

Large Language Models continue to evolve rapidly.

Although Reinforcement Learning from Human Feedback (RLHF) and Direct Preference Optimization (DPO) remain highly influential,

researchers have proposed numerous alternative methods that simplify training,

improve stability,

reduce computational cost,

or decrease dependence on human annotations.

Rather than representing completely different paradigms,

most modern alignment methods build upon the same fundamental idea:

> **Use preference information to make language models more helpful, truthful, and aligned with human intentions.**

This chapter surveys several important alignment methods used in contemporary Large Language Model research.

---



# <font color='orange'> 1. Motivation </font>

Suppose two language models produce responses of similar quality.

One requires

```
Reward Model

+

Value Network

+

PPO
```

while another requires only

```
Preference Dataset

↓

Gradient Descent
```

Naturally,

researchers seek the simpler approach.

Modern alignment methods largely focus on simplifying optimization while preserving performance.

---



# <font color='orange'> 2. Evolution of Alignment </font>

The development of alignment methods can be viewed as a progression.

```
Pretraining

↓

Supervised Fine-Tuning

↓

RLHF

↓

DPO

↓

Modern Preference Optimization
```

Each stage reduces complexity or improves performance.

---



# <font color='orange'> 3. Reinforcement Learning from Human Feedback (RLHF) </font>

RLHF introduced the idea of

optimizing language models

using

human preferences.

Pipeline

```
Preference Data

↓

Reward Model

↓

PPO

↓

Aligned Model
```

Advantages

- strong empirical performance,
- flexible reward optimization.

Limitations

- computationally expensive,
- multiple neural networks,
- PPO tuning.

---



# <font color='orange'> 4. Direct Preference Optimization (DPO) </font>

DPO removes

the Reward Model

and PPO.

Pipeline

```
Preference Data

↓

DPO Loss

↓

Aligned Model
```

Advantages

- simpler,
- stable,
- computationally efficient.

Today,

DPO is widely used in research and many open-source models.

---



# <font color='orange'> 5. Implicit Preference Optimization (IPO) </font>

Implicit Preference Optimization (IPO)

modifies the DPO objective

to improve optimization behaviour.

Rather than aggressively separating

preferred

and rejected responses,

IPO encourages

more balanced updates.

This can improve training stability,

particularly for very large models.

Conceptually,

IPO seeks to preserve the strengths of DPO while reducing overconfidence during optimization.

---



# <font color='orange'> 6. Odds Ratio Preference Optimization (ORPO) </font>

ORPO simplifies the training pipeline even further.

Instead of performing

Supervised Fine-Tuning

followed by

DPO,

ORPO combines

instruction tuning

and preference optimization

into a single optimization stage.

This reduces computational cost

and simplifies training.

---



# <font color='orange'> 7. Simple Preference Optimization (SimPO) </font>

SimPO proposes

an even simpler preference optimization objective.

Instead of explicitly comparing

the current model

to

a reference model,

SimPO directly optimizes

relative preference scores.

Removing the reference model reduces memory usage

and simplifies implementation.

---



# <font color='orange'> 8. Group Relative Policy Optimization (GRPO) </font>

Group Relative Policy Optimization (GRPO)

compares

multiple candidate responses

rather than only two.

```
Prompt

↓

Response 1

↓

Response 2

↓

Response 3

↓

Rank Responses

↓

Optimize
```

Learning from groups

can provide richer information

than isolated pairwise comparisons.

GRPO has attracted attention because it enables efficient optimization from ranked candidate responses and has been used in reasoning-focused training.

---



# <font color='orange'> 9. Reinforcement Learning from AI Feedback (RLAIF) </font>

Collecting

human preference data

is expensive.

Instead,

another language model

can generate preference comparisons.

```
Prompt

↓

Two Responses

↓

AI Judge

↓

Preference
```

This approach is called

**Reinforcement Learning from AI Feedback (RLAIF).**

Human oversight remains important,

especially for evaluating the quality of the AI-generated feedback.

---



# <font color='orange'> 10. Constitutional AI </font>

Rather than relying solely on human comparisons,

Constitutional AI provides

the model

with

a written set of principles,

or

a **constitution**.

Example principles include

- be truthful,
- avoid harmful advice,
- explain uncertainty,
- respect user intent.

During training,

the model critiques

its own responses

using these principles,

and then revises them.

This reduces the need for extensive human annotation while encouraging consistent behaviour.

---



# <font color='orange'> 11. Why So Many Methods? </font>

Different applications

require different trade-offs.

Some prioritize

- computational efficiency,

while others prioritize

- reasoning,
- safety,
- scalability,
- data efficiency.

Consequently,

there is no universally best alignment algorithm.

Research continues rapidly.

---



# <font color='orange'> 12. Applications </font>

Modern alignment methods are used in

- conversational assistants,
- coding assistants,
- reasoning models,
- educational systems,
- scientific assistants,
- enterprise AI.

The choice of method depends on the model,

available data,

and deployment requirements.

---



# <font color='orange'> 13. Limitations </font>

All alignment methods share common challenges.

These include

- dependence on high-quality preference data or evaluators,
- potential biases in preferences,
- imperfect evaluation of truthfulness,
- balancing helpfulness with safety,
- ensuring robustness across diverse tasks.

Alignment remains an active area of research rather than a solved problem.

---

# <font color='red'> 14. Conceptual Comparison </font>

| Method | Reward Model | PPO | Reference Model | Human Preferences |
|:---|:---:|:---:|:---:|:---:|
| RLHF | ✓ | ✓ | ✓ | ✓ |
| DPO | ✗ | ✗ | ✓ | ✓ |
| IPO | ✗ | ✗ | ✓ | ✓ |
| ORPO | ✗ | ✗ | Implicit | ✓ |
| SimPO | ✗ | ✗ | ✗ | ✓ |
| GRPO | Optional | Relative Optimization | Varies | Rankings |
| RLAIF | AI-generated | Optional | Varies | AI feedback |
| Constitutional AI | Optional | Optional | Varies | Constitution + critique |

Although these methods differ,

they all aim to improve alignment with desired behaviour.

---



# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> DPO has completely replaced RLHF.

**False.**

DPO has become very influential, but RLHF remains valuable in settings where explicit reward optimization is beneficial. Different organizations and applications may choose different alignment strategies.

---

### Misconception 2

> AI feedback completely eliminates the need for humans.

**False.**

AI-generated feedback can reduce annotation costs, but human oversight is still important to verify quality, identify systematic errors, and establish the principles the AI should follow.

---

### Misconception 3

> Alignment is a solved problem.

**False.**

Modern alignment methods have significantly improved language models, but challenges such as factual accuracy, robustness, bias, and handling ambiguous instructions remain active research topics.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Method | Main Idea |
|:---|:---|
| RLHF | Optimize a reward model using PPO |
| DPO | Optimize directly from preferences |
| IPO | More conservative preference optimization |
| ORPO | Combine instruction tuning and preference optimization |
| SimPO | Simplify preference optimization without a reference model |
| GRPO | Learn from groups of ranked responses |
| RLAIF | Use AI-generated preference feedback |
| Constitutional AI | Align behaviour using explicit guiding principles |

> **Key Insight:** Modern alignment methods all share the same objective: improving language model behaviour so that it better reflects human preferences and intended use. The field has progressed from the relatively complex RLHF pipeline toward increasingly direct and computationally efficient preference optimization methods such as DPO, while also exploring alternatives like AI-generated feedback and constitutional guidance. Rather than a single best solution, alignment has become a rich family of techniques, each offering different trade-offs in efficiency, scalability, and capability.
