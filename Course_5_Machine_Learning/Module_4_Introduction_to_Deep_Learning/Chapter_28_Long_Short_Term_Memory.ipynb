# <font color='blue'> Chapter 28: Long Short-Term Memory (LSTM) Networks </font>

In the previous chapter, we learned that Recurrent Neural Networks (RNNs) are trained using **Backpropagation Through Time (BPTT)**. We also saw that repeated multiplication of gradients causes the **vanishing gradient problem**, making it difficult for standard RNNs to learn long-range dependencies.

Long Short-Term Memory (LSTM) networks were developed to overcome this limitation.

The key idea behind an LSTM is to introduce a **memory cell** that can preserve information over long periods while allowing the network to selectively forget irrelevant information.

Instead of relying on a single hidden state, an LSTM maintains two pieces of information:

- a **hidden state**, which represents the current output,
- a **cell state**, which stores long-term memory.

The flow of information is controlled by three gates:

- Forget Gate
- Input Gate
- Output Gate

These gates enable the network to decide what information should be remembered, updated, or discarded.

---



# <font color='orange'> 1. Motivation </font>

Consider the sentence

> *The conference began on Monday. It lasted three days. The final presentation was held on ______.*

To predict the missing word,

the network must remember that the conference started on **Monday**.

This information appeared several time steps earlier.

A standard RNN may gradually forget this information,

whereas an LSTM is specifically designed to preserve it.

---



# <font color='orange'> 2. The Main Components of an LSTM </font>

Each LSTM cell contains

```
Previous Hidden State

↓

Previous Cell State

↓

Forget Gate

↓

Input Gate

↓

Cell Update

↓

Output Gate

↓

New Hidden State

↓

New Cell State
```

Unlike a simple RNN,

the memory stored in the cell state can persist over many time steps.

---



# <font color='orange'> 3. Hidden State vs Cell State </font>

The hidden state

$$
\mathbf{h}_t
$$

is passed to the next layer (or output).

The cell state

$$
\mathbf{C}_t
$$

acts as the network's long-term memory.

Think of the hidden state as the **working memory**,

while the cell state is the **long-term memory**.

---



# <font color='orange'> 4. The Forget Gate </font>

The first step is deciding what information should be forgotten.

The forget gate computes

$$
\boxed{
\mathbf{f}_t
=
\sigma
\left(
W_f
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_f
\right).
}
$$

where

- $\sigma$ is the sigmoid function,
- values lie between

$$
0
\quad\text{and}\quad
1.
$$

Interpretation

- 0 → completely forget,
- 1 → completely remember.

---



# <font color='orange'> 5. The Input Gate </font>

The network now decides

what new information should be stored.

The input gate is

$$
\boxed{
\mathbf{i}_t
=
\sigma
\left(
W_i
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_i
\right).
}
$$

Again,

values lie between

0

and

1.

---



# <font color='orange'> 6. Candidate Cell State </font>

Next,

the network generates candidate memory values.

$$
\boxed{
\tilde{\mathbf{C}}_t
=
\tanh
\left(
W_C
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_C
\right).
}
$$

Unlike the sigmoid gates,

the candidate values lie between

$$
-1
\quad\text{and}\quad
1.
$$

---



# <font color='orange'> 7. Updating the Cell State </font>

The new cell state combines

- remembered information,
- newly learned information.

The update equation is

$$
\boxed{
\mathbf{C}_t
=
\mathbf{f}_t
\odot
\mathbf{C}_{t-1}
+
\mathbf{i}_t
\odot
\tilde{\mathbf{C}}_t.
}
$$

where

$$
\odot
$$

denotes element-wise multiplication.

This equation is the heart of the LSTM.

It allows information to flow through many time steps with minimal attenuation.

---



# <font color='orange'> 8. The Output Gate </font>

The final gate determines

which information should be exposed.

It computes

$$
\boxed{
\mathbf{o}_t
=
\sigma
\left(
W_o
\begin{bmatrix}
\mathbf{h}_{t-1}\\
\mathbf{x}_t
\end{bmatrix}
+
\mathbf{b}_o
\right).
}
$$

---



# <font color='orange'> 9. Hidden State Update </font>

The hidden state becomes

$$
\boxed{
\mathbf{h}_t
=
\mathbf{o}_t
\odot
\tanh(\mathbf{C}_t).
}
$$

Thus,

the hidden state is a filtered version of the cell state.

---



# <font color='orange'> 10. Complete LSTM Forward Pass </font>

For each time step,

the LSTM performs

```
Forget Gate

↓

Input Gate

↓

Candidate Memory

↓

Update Cell State

↓

Output Gate

↓

Update Hidden State
```

These six operations are repeated at every time step.

---



# <font color='orange'> 11. Why LSTMs Solve the Vanishing Gradient Problem </font>

Recall that in a simple RNN,

the hidden state is repeatedly transformed,

causing gradients to shrink or grow exponentially.

In an LSTM,

the cell state update

$$
\mathbf{C}_t
=
\mathbf{f}_t
\odot
\mathbf{C}_{t-1}
+
...
$$

contains an **additive** path.

Because information can pass through this additive connection,

gradients are much less likely to vanish.

This is sometimes called the **Constant Error Carousel (CEC)**.

---



# <font color='orange'> 12. RNN vs LSTM </font>

| Standard RNN | LSTM |
|:---|:---|
| Single hidden state | Hidden state + cell state |
| Simple recurrence | Memory controlled by gates |
| Suffers from vanishing gradients | Much better at long-term dependencies |
| Fewer parameters | More parameters |
| Faster | Slower but more powerful |

---



# <font color='orange'> 13. Applications of LSTMs </font>

LSTMs have been widely used in

- language modelling,
- speech recognition,
- machine translation,
- handwriting recognition,
- financial forecasting,
- weather prediction,
- anomaly detection,
- healthcare time series.

Although Transformers now dominate many NLP tasks,

LSTMs remain important for many sequence modelling applications and for understanding gated recurrent networks.

# <font color='red'> 14. Mathematical Foundations </font>

The complete forward pass of an LSTM is

$$
\boxed{
\mathbf{f}_t
=
\sigma
(W_f[\mathbf{h}_{t-1},\mathbf{x}_t]+b_f),
}
$$

$$
\boxed{
\mathbf{i}_t
=
\sigma
(W_i[\mathbf{h}_{t-1},\mathbf{x}_t]+b_i),
}
$$

$$
\boxed{
\tilde{\mathbf{C}}_t
=
\tanh
(W_C[\mathbf{h}_{t-1},\mathbf{x}_t]+b_C),
}
$$

$$
\boxed{
\mathbf{C}_t
=
\mathbf{f}_t
\odot
\mathbf{C}_{t-1}
+
\mathbf{i}_t
\odot
\tilde{\mathbf{C}}_t,
}
$$

$$
\boxed{
\mathbf{o}_t
=
\sigma
(W_o[\mathbf{h}_{t-1},\mathbf{x}_t]+b_o),
}
$$

$$
\boxed{
\mathbf{h}_t
=
\mathbf{o}_t
\odot
\tanh(\mathbf{C}_t).
}
$$

Together,

these equations describe the complete computation performed by an LSTM cell at every time step.

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> LSTMs completely eliminate the vanishing gradient problem.

**False.**

LSTMs significantly reduce the problem by providing a more direct path for gradients through the cell state, but they do not guarantee perfect learning over arbitrarily long sequences.

---

### Misconception 2

> The cell state and hidden state are identical.

**False.**

The cell state stores long-term information, while the hidden state represents the information exposed to the next layer or output at the current time step.

---

### Misconception 3

> The gates are manually programmed.

**False.**

The gate parameters are learned automatically during training through backpropagation, just like all other trainable weights in the network.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Cell State | Long-term memory carried through the sequence |
| Hidden State | Current output representation |
| Forget Gate | Decides what information to discard |
| Input Gate | Decides what new information to store |
| Candidate Cell State | Proposed memory update |
| Output Gate | Controls what information is exposed |
| Constant Error Carousel | Additive memory path that helps preserve gradients |
| Long-Term Dependency | Information that must be retained across many time steps |

> **Key Insight:** Long Short-Term Memory networks extend recurrent neural networks by introducing a dedicated memory cell and learnable gating mechanisms. The forget, input, and output gates regulate the flow of information, allowing the network to preserve relevant long-term information while discarding irrelevant details. This architecture greatly alleviates the vanishing gradient problem and enables effective learning from long sequential data.
