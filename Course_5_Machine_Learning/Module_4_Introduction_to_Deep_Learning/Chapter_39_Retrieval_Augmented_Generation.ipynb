# <font color='blue'> Chapter 39: Retrieval-Augmented Generation (RAG) </font>

Large Language Models (LLMs) possess impressive language understanding and generation capabilities. However, they have an important limitation:

Their knowledge is primarily derived from the data used during pretraining.

Consequently,

they may

- lack recent information,
- be unaware of private documents,
- generate incorrect or fabricated answers,
- struggle with organization-specific knowledge.

Retrieval-Augmented Generation (RAG) addresses these limitations by combining **information retrieval** with **language generation**.

Instead of relying solely on its internal parameters,

the language model first retrieves relevant documents from an external knowledge base,

then uses those documents to generate an informed response.

---



# <font color='orange'> 1. Motivation </font>

Suppose we ask

```
What is our company's latest leave policy?
```

A pretrained LLM cannot answer reliably unless that information appeared in its training data.

Instead,

a RAG system

- searches the company's document database,
- retrieves the relevant policy,
- provides the retrieved information to the LLM,
- generates an answer based on the retrieved documents.

Thus,

the model can answer questions about information it was never explicitly trained on.

---



# <font color='orange'> 2. Why RAG is Needed </font>

Pretraining has several limitations.

Knowledge becomes outdated.

Private company documents are unavailable.

Scientific literature grows daily.

Medical and legal regulations change.

Retraining an LLM every time new information appears would be extremely expensive.

RAG avoids this problem by retrieving information at inference time.

---



# <font color='orange'> 3. The Big Idea </font>

Instead of asking

```
Question

↓

LLM

↓

Answer
```

RAG performs

```
Question

↓

Retriever

↓

Relevant Documents

↓

LLM

↓

Answer
```

The retrieved documents become additional context for the language model.

---



# <font color='orange'> 4. The Complete RAG Pipeline </font>

A typical RAG system follows these steps.

```
Documents

↓

Chunking

↓

Embeddings

↓

Vector Database

↓

User Question

↓

Question Embedding

↓

Similarity Search

↓

Relevant Chunks

↓

LLM

↓

Generated Answer
```

Each stage plays an important role in the overall system.

---



# <font color='orange'> 5. Document Chunking </font>

Large documents are usually divided into smaller pieces called

**chunks**.

For example,

```
100-page report

↓

Paragraphs

↓

Chunks
```

Chunking improves retrieval because the system returns only the most relevant sections rather than entire documents.

Choosing an appropriate chunk size is an important design decision.

---



# <font color='orange'> 6. Embeddings </font>

Every chunk is converted into an embedding vector.

```
Chunk

↓

Embedding Model

↓

Dense Vector
```

Semantically similar chunks produce nearby vectors in the embedding space.

These embeddings enable efficient semantic search.

---



# <font color='orange'> 7. Vector Databases </font>

The embedding vectors are stored in a **vector database**.

Unlike a traditional SQL database,

a vector database searches using similarity rather than exact matches.

Popular vector databases include

- FAISS,
- Chroma,
- Pinecone,
- Milvus,
- Weaviate.

Their primary purpose is efficient nearest-neighbour search in high-dimensional spaces.

---



# <font color='orange'> 8. Embedding the User Question </font>

When a user asks a question,

the same embedding model converts the question into a vector.

```
Question

↓

Embedding

↓

Vector
```

The question vector is then compared with all stored document vectors.

---



# <font color='orange'> 9. Similarity Search </font>

The retriever searches for document chunks whose embeddings are closest to the question embedding.

Common similarity measures include

- Cosine Similarity,
- Dot Product,
- Euclidean Distance.

The most similar chunks are returned to the LLM.

---



# <font color='orange'> 10. Cosine Similarity </font>

The most widely used similarity measure is **Cosine Similarity**.

For two vectors

$$
\mathbf{u}
\quad\text{and}\quad
\mathbf{v},
$$

the similarity is

$$
\boxed{
\mathrm{Cosine}
=
\frac{\mathbf{u}\cdot\mathbf{v}}
{\|\mathbf{u}\|\,\|\mathbf{v}\|}.
}
$$

Values close to

$$
1
$$

indicate highly similar meanings,

while values near

$$
0
$$

indicate little semantic similarity.

---



# <font color='orange'> 11. Retrieval </font>

Suppose

the top three chunks are retrieved.

```
Chunk 17

↓

Chunk 84

↓

Chunk 212
```

These chunks are inserted into the prompt.

The LLM now answers

using both

- the user's question,
- the retrieved context.

---



# <font color='orange'> 12. Generation </font>

The prompt becomes

```
Retrieved Context

+

User Question

↓

LLM

↓

Answer
```

The model generates an answer grounded in the retrieved documents rather than relying solely on its pretrained knowledge.

---



# <font color='orange'> 13. Advantages of RAG </font>

RAG

- provides access to recent information,
- reduces hallucinations,
- incorporates private knowledge,
- avoids expensive retraining,
- improves factual accuracy,
- enables explainable answers by citing retrieved sources.

These properties make RAG the preferred architecture for many enterprise AI systems.

---



# <font color='orange'> 14. Limitations </font>

RAG is not perfect.

Its performance depends on

- document quality,
- chunking strategy,
- embedding quality,
- retrieval accuracy,
- context window limitations.

If irrelevant documents are retrieved,

the generated answer may also be poor.

---



# <font color='orange'> 15. Applications </font>

RAG is widely used for

- enterprise knowledge assistants,
- legal document search,
- medical information systems,
- customer support,
- scientific literature review,
- technical documentation,
- financial analysis,
- educational assistants.

Whenever answers depend on external or frequently updated knowledge,

RAG is often preferred over relying solely on the pretrained model.

---

# <font color='red'> 16. Mathematical Foundations </font>

Suppose

the document embedding is

$$
\mathbf{d}_i,
$$

and the question embedding is

$$
\mathbf{q}.
$$

The retrieval score is often computed using cosine similarity,

$$
\boxed{
\mathrm{score}_i
=
\frac{\mathbf{q}\cdot\mathbf{d}_i}
{\|\mathbf{q}\|\,\|\mathbf{d}_i\|}.
}
$$

The retriever selects the

$$
k
$$

documents with the highest scores.

The LLM then models

$$
\boxed{
P(\text{Answer}\mid
\text{Question},
\text{Retrieved Documents}).
}
$$

Compared with a standard LLM,

the model now conditions its response on both the prompt and the retrieved external knowledge.

---



# <font color='orange'> 17. RAG vs Fine-Tuning </font>

| Retrieval-Augmented Generation | Fine-Tuning |
|:---|:---|
| Adds external knowledge at inference time | Updates the model's parameters |
| Easy to update by changing the document collection | Requires additional training |
| Suitable for frequently changing information | Suitable for changing model behaviour or specialization |
| Preserves the original pretrained model | Produces a new adapted model |

In practice,

RAG and fine-tuning are often complementary rather than competing approaches.

---



# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> RAG changes the model's knowledge permanently.

**False.**

RAG retrieves external information only at inference time. The underlying model parameters remain unchanged.

---

### Misconception 2

> RAG completely eliminates hallucinations.

**False.**

RAG can reduce hallucinations by grounding responses in retrieved documents, but poor retrieval, ambiguous questions, or incorrect source material can still lead to inaccurate answers.

---

### Misconception 3

> A vector database replaces the language model.

**False.**

The vector database retrieves relevant information, while the LLM interprets that information and generates the final response.

---

# <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Retrieval-Augmented Generation (RAG) | Combines document retrieval with LLM generation |
| Chunking | Splitting large documents into smaller sections |
| Embedding | Dense vector representation of text |
| Vector Database | Stores embeddings for similarity search |
| Retriever | Finds documents relevant to a query |
| Cosine Similarity | Measures semantic similarity between embeddings |
| Retrieved Context | External information provided to the LLM |
| Grounded Generation | Responses based on retrieved documents rather than only pretrained knowledge |

> **Key Insight:** Retrieval-Augmented Generation enhances Large Language Models by allowing them to access external knowledge at inference time. Documents are converted into embeddings and stored in a vector database, where similarity search retrieves the most relevant information for a user's query. The retrieved context is then provided to the LLM, enabling more accurate, up-to-date, and context-aware responses without retraining the underlying model.
