# <font color='blue'> Chapter 21: Pooling Layers </font>

Convolutional layers extract features from an image by applying learnable filters. However, after several convolutional operations, the resulting feature maps can become very large, leading to increased computational cost and memory usage.

Pooling layers address this problem by reducing the spatial dimensions of the feature maps while preserving their most important information.

Pooling is therefore a form of **downsampling**, allowing Convolutional Neural Networks (CNNs) to become faster, more memory efficient, and more robust to small variations in the input.

Unlike convolutional layers, pooling layers contain **no trainable parameters**.

---

# <font color='orange'> 1. Motivation </font>

Suppose a convolutional layer produces a feature map of size

$$
64\times64.
$$

If we continue stacking convolutional layers without reducing its size,

every subsequent layer must process

$$
64\times64=4096
$$

values per feature map.

This rapidly becomes computationally expensive.

Pooling reduces the feature map size.

For example,

```
64 × 64

↓

Pooling

↓

32 × 32
```

The number of activations decreases by a factor of four.

---

# <font color='orange'> 2. What is Pooling? </font>

Pooling replaces a small neighbourhood of values with a single representative value.

For example,

using a

$$
2\times2
$$

window,

```
8  3

5  7
```

may become

```
8
```

when Max Pooling is used.

The pooling window then slides across the feature map,

just like a convolutional filter.

---

# <font color='orange'> 3. Max Pooling </font>

The most widely used pooling method is **Max Pooling**.

For every pooling window,

the largest value is selected.

Example

Original feature map

$$
\begin{bmatrix}
1 & 5\\
4 & 2
\end{bmatrix}
$$

Applying

$$
2\times2
$$

Max Pooling gives

$$
\boxed{
5.
}
$$

The strongest activation is preserved,

while the remaining values are discarded.

---

# <font color='orange'> 4. Numerical Example </font>

Consider the feature map

$$
\begin{bmatrix}
1 & 3 & 2 & 4\\
5 & 6 & 1 & 2\\
7 & 8 & 3 & 1\\
4 & 2 & 5 & 9
\end{bmatrix}
$$

Apply

$$
2\times2
$$

Max Pooling

with stride

$$
2.
$$

Each window becomes

Window 1

$$
\begin{bmatrix}
1 & 3\\
5 & 6
\end{bmatrix}
\rightarrow
6
$$

Window 2

$$
\begin{bmatrix}
2 & 4\\
1 & 2
\end{bmatrix}
\rightarrow
4
$$

Window 3

$$
\begin{bmatrix}
7 & 8\\
4 & 2
\end{bmatrix}
\rightarrow
8
$$

Window 4

$$
\begin{bmatrix}
3 & 1\\
5 & 9
\end{bmatrix}
\rightarrow
9
$$

Final output

$$
\boxed{
\begin{bmatrix}
6 & 4\\
8 & 9
\end{bmatrix}.
}
$$

---

# <font color='orange'> 5. Average Pooling </font>

Instead of selecting the maximum value,

Average Pooling computes the mean.

Using

$$
\begin{bmatrix}
1 & 5\\
3 & 7
\end{bmatrix}
$$

Average Pooling produces

$$
\boxed{
\frac{1+5+3+7}{4}=4.
}
$$

Average Pooling retains more overall information but may weaken strong feature responses.

---

# <font color='orange'> 6. Global Average Pooling </font>

Modern CNNs often replace the Flatten layer with **Global Average Pooling (GAP)**.

Suppose a feature map is

$$
\begin{bmatrix}
2 & 4\\
6 & 8
\end{bmatrix}.
$$

Its global average is

$$
\boxed{
\frac{2+4+6+8}{4}=5.
}
$$

Each feature map is reduced to a single value.

Advantages include

- fewer parameters,
- reduced overfitting,
- simpler architectures.

Many modern architectures such as **ResNet** and **EfficientNet** employ Global Average Pooling before the final classification layer.

---

# <font color='orange'> 7. Stride in Pooling </font>

Pooling also has a stride.

Most commonly,

- window size =

$$
2\times2,
$$

- stride =

$$
2.
$$

This halves both the height and width of the feature map.

Example

```
32 × 32

↓

2 × 2 Pooling

↓

16 × 16
```

---

# <font color='orange'> 8. Output Size Formula </font>

Pooling follows the same output size equation as convolution.

Suppose

- input size

$$
N,
$$

- pooling window

$$
F,
$$

- padding

$$
P,
$$

- stride

$$
S.
$$

Then

$$
\boxed{
\text{Output Size}
=
\frac{N-F+2P}{S}+1.
}
$$

For example,

Input

$$
32\times32,
$$

Pooling

$$
2\times2,
$$

Stride

$$
2,
$$

Padding

$$
0.
$$

The output becomes

$$
\frac{32-2}{2}+1=16.
$$

Thus,

the output size is

$$
16\times16.
$$

---

# <font color='orange'> 9. Why Does Pooling Help? </font>

Pooling provides several advantages.

- Reduces computation.
- Reduces memory usage.
- Increases the effective receptive field of deeper layers.
- Makes the network less sensitive to small translations or distortions.
- Helps reduce overfitting.

Pooling therefore allows deeper CNNs to remain computationally feasible.

---

# <font color='orange'> 10. Pooling vs Convolution </font>

| Convolution | Pooling |
|:---|:---|
| Learns filters | No trainable parameters |
| Extracts features | Summarizes features |
| Changes feature values | Reduces spatial dimensions |
| Requires backpropagation for weights | No weights to update |

Convolution and pooling complement one another rather than competing.

---

# <font color='orange'> 11. Is Pooling Always Necessary? </font>

Early CNN architectures,

such as **LeNet** and **AlexNet**,

used pooling extensively.

However,

many modern architectures reduce spatial dimensions using **strided convolutions** instead of separate pooling layers.

Examples include

- ResNet,
- DenseNet,
- EfficientNet.

Pooling remains an important concept and is still widely used, particularly Max Pooling and Global Average Pooling.

---

# <font color='red'> 12. Mathematical Foundations </font>

Suppose the feature map is

$$
F(i,j).
$$

For Max Pooling,

the pooled value at position

$$
(p,q)
$$

is

$$
\boxed{
P(p,q)
=
\max_{(i,j)\in\Omega_{pq}}
F(i,j),
}
$$

where

$$
\Omega_{pq}
$$

represents the pooling window.

For Average Pooling,

the pooled value is

$$
\boxed{
P(p,q)
=
\frac{1}{|\Omega_{pq}|}
\sum_{(i,j)\in\Omega_{pq}}
F(i,j).
}
$$

Unlike convolution,

these operations contain **no learnable parameters**.

---

# <font color='orange'> 13. Common Misconceptions </font>

### Misconception 1

> Pooling learns new features.

**False.**

Pooling summarizes existing features. Feature extraction is performed by convolutional layers.

---

### Misconception 2

> Max Pooling always outperforms Average Pooling.

**False.**

Max Pooling is often preferred for object recognition because it preserves strong activations, but Average Pooling can be advantageous in tasks where overall feature intensity is more informative.

---

### Misconception 3

> Pooling always improves performance.

**False.**

Pooling discards spatial information. While this can improve robustness and efficiency, excessive pooling may remove details that are important for tasks such as image segmentation or keypoint detection.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Pooling | Downsamples feature maps |
| Max Pooling | Selects the largest value within each window |
| Average Pooling | Computes the average value within each window |
| Global Average Pooling | Reduces each feature map to a single value |
| Stride | Determines how far the pooling window moves |
| Output Size | $\frac{N-F+2P}{S}+1$ |
| Trainable Parameters | None |

> **Key Insight:** Pooling layers reduce the spatial dimensions of feature maps while preserving their most important information. By decreasing computational cost and improving robustness to small translations, pooling enables CNNs to learn hierarchical feature representations efficiently. Although modern architectures sometimes replace traditional pooling with strided convolutions, Max Pooling and Global Average Pooling remain fundamental components of many successful convolutional neural networks.
