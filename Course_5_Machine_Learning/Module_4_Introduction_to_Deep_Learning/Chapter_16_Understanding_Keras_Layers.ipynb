# <font color='blue'> Chapter 16: Understanding Keras Layers </font>

In the previous chapter, we built our first neural network using the Keras Sequential API. Although the model required only a few lines of code, each layer performed an important mathematical operation.

A neural network is simply a sequence of layers, where each layer transforms its input into a more useful representation.

Understanding the purpose of each layer is essential for designing effective neural network architectures.

In this chapter, we study the most commonly used Keras layers and explain both their implementation and their mathematical interpretation.

---

# <font color='orange'> 1. What is a Layer? </font>

A layer is a computational unit that receives one or more tensors as input, performs a mathematical transformation, and produces another tensor as output.

```
Input Tensor

↓

Layer

↓

Output Tensor
```

Each layer performs a specific task.

Examples include

- linear transformations,
- nonlinear activations,
- reshaping data,
- normalization,
- regularization,
- feature extraction.

---

# <font color='orange'> 2. The Input Layer </font>

Every neural network begins with an input layer.

The input layer defines

- the shape of the data,
- the number of input features.

Example

```python
keras.Input(shape=(10,))
```

This specifies

- 10 input features,
- batch size determined automatically.

For images

```python
keras.Input(shape=(28,28,1))
```

means

- height = 28
- width = 28
- one colour channel

Many datasets are multidimensional.

Example

```
28 × 28 Image
```

Most fully connected layers expect

```
784-dimensional vector
```

Flatten converts

```
28 × 28

↓

784
```

Code

```python
layers.Flatten()
```

It simply reshapes the data.

# <font color='orange'> 4. Dense Layer </font>

The Dense layer is the most important layer in feedforward neural networks.

Example

```python
layers.Dense(128)
```

creates

128 neurons.

Each neuron computes

$$
\boxed{
a=f(Wx+b).
}
$$

Every input is connected to every neuron.

Hence,

Dense layers are also called

**Fully Connected Layers**.

---

## Number of Parameters

Suppose

- input dimension = 50
- neurons = 128

Weights

$$
50\times128=6400
$$

Biases

$$
128
$$

Total

$$
\boxed{
6528
}
$$

In general,

$$
\boxed{
\text{Parameters}
=
(n_{\rm input}+1)
\times
n_{\rm neurons}.
}
$$

# <font color='orange'> 5. Activation Layer </font>

Instead of specifying

```python
layers.Dense(128,
             activation="relu")
```

we may separate the activation

```python
layers.Dense(128)

layers.Activation("relu")
```

Both approaches are mathematically equivalent.

---

# <font color='orange'> 6. Dropout Layer </font>

Dropout is a regularization technique.

During training,

some neurons are randomly ignored.

```
Training

○ ○ ○ ○ ○

↓

○ X ○ X ○

↓

Next Layer
```

Example

```python
layers.Dropout(0.5)
```

Here,

50% of neurons are randomly deactivated during each training step.

This reduces overfitting.

---

# <font color='orange'> 7. Batch Normalization </font>

Batch Normalization normalizes the activations within a mini-batch.

```python
layers.BatchNormalization()
```

Benefits

- faster convergence,
- more stable gradients,
- higher learning rates,
- improved generalization.

Instead of every layer learning from wildly different activation ranges,

Batch Normalization keeps activations approximately centred and scaled.

# <font color='red'> 17. Mathematical Foundations </font>

Every trainable layer ultimately performs

$$
\boxed{
a
=
f(Wx+b).
}
$$

Some layers,

such as

Flatten,

Reshape,

and

Masking,

do not perform learning.

They merely transform the data before it reaches trainable layers.

The Dense layer remains the fundamental computational unit of feedforward neural networks.

---

# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> Every layer contains trainable weights.

**False.**

Layers such as `Flatten`, `Reshape`, and `Dropout` have no trainable parameters.

---

### Misconception 2

> More Dense layers always improve performance.

**False.**

Increasing depth can improve representational power, but it also increases the risk of overfitting and optimization difficulties if not paired with appropriate regularization and architecture choices.

---

### Misconception 3

> Dropout permanently removes neurons.

**False.**

Dropout temporarily deactivates randomly selected neurons **only during training**. During inference, all neurons are used.

|:---|:---|

| Flatten | Converts multidimensional data into a vector |
| Dense | Fully connected neural layer |

> **Key Insight:** A neural network is built by composing layers, each of which performs a specific transformation on the input data. Some layers, such as `Dense`, learn trainable parameters through backpropagation, while others, such as `Flatten` and `Dropout`, modify the flow of information without learning parameters themselves. Choosing the right combination of layers is one of the most important aspects of neural network design.
