# <font color='blue'> Chapter 59: Distributional Reinforcement Learning </font>

Classical Reinforcement Learning algorithms estimate only the **expected return** obtained after taking an action.

For many problems,

this is sufficient.

However,

the expected return ignores important information,

such as

- uncertainty,
- variability,
- risk,
- multiple possible future outcomes.

Distributional Reinforcement Learning (Distributional RL) addresses this limitation by learning the **entire probability distribution of future returns** instead of only their mean.

Rather than predicting a single number,

the agent predicts all possible returns together with their probabilities.

This richer representation often produces better policies,

improved stability,

and faster learning.

---



# <font color='orange'> 1. Motivation </font>

Suppose two actions both have an average return of

```
£100
```

Action A

```
Always £100
```

Action B

```
£0

or

£200
```

Although both actions have the same expected reward,

their uncertainty is very different.

Traditional Q-Learning treats them as identical.

Distributional RL does not.

---



# <font color='orange'> 2. Expected Return </font>

Classical Reinforcement Learning learns

$$
\boxed{
Q(s,a)
=
\mathbb E[G],
}
$$

where

$$
G
$$

is the discounted return.

Only the average value is retained.

All information about the distribution is discarded.

---



# <font color='orange'> 3. Random Return </font>

Instead of treating

the return

as a single number,

Distributional RL treats it as a random variable.

It is written

$$
\boxed{
Z(s,a).
}
$$

Different episodes may produce different returns,

even when the same action is chosen.

The goal becomes

learning

the probability distribution of

$$
Z(s,a).
$$

---



# <font color='orange'> 4. Relationship Between Z and Q </font>

The expected return is simply

the mean

of the return distribution.

$$
\boxed{
Q(s,a)
=
\mathbb E
\left[
Z(s,a)
\right].
}
$$

Therefore,

traditional Q-Learning can be viewed

as learning only

one statistic

of a much richer distribution.

---



# <font color='orange'> 5. Why Learn the Whole Distribution? </font>

Knowing the entire distribution provides additional information,

including

- uncertainty,
- variance,
- skewness,
- multimodal behaviour.

Example

```
Return

↓

20%

£0

↓

60%

£100

↓

20%

£200
```

A single expected value cannot represent this structure.

---



# <font color='orange'> 6. Distributional Bellman Equation </font>

The Bellman equation can also be written

for distributions.

Instead of

updating expected values,

we update

the distribution itself.

Conceptually,

$$
\boxed{
Z(s,a)
=
r
+
\gamma
Z(s',a').
}
$$

Unlike the classical Bellman equation,

both sides now represent random variables rather than deterministic expectations.

---



# <font color='orange'> 7. Categorical Distribution (C51) </font>

The first successful Distributional RL algorithm was

**Categorical DQN (C51).**

Instead of representing

all possible returns,

C51 approximates the return distribution using

a fixed number of discrete support points,

called

**atoms**.

Example

```
Return Axis

↓

-10

↓

-5

↓

0

↓

5

↓

10
```

The network predicts

the probability associated with each atom.

---



# <font color='orange'> 8. Atoms </font>

Suppose

there are

$$
N
$$

atoms,

equally spaced between

$$
V_{\min}
$$

and

$$
V_{\max}.
$$

Each atom is

$$
\boxed{
z_i
=
V_{\min}
+
i\Delta z,
}
$$

where

$$
\Delta z
=
\frac{V_{\max}-V_{\min}}{N-1}.
$$

The neural network predicts

the probability of each atom.

---



# <font color='orange'> 9. Probability Distribution </font>

Instead of one output,

the network predicts

```
Atom

↓

Probability
```

Example

| Return | Probability |
|:---|---:|
| -10 | 0.02 |
| -5 | 0.08 |
| 0 | 0.15 |
| 5 | 0.35 |
| 10 | 0.40 |

The probabilities sum to one.

---



# <font color='orange'> 10. Expected Return from the Distribution </font>

Although the network predicts

a distribution,

the expected return can still be recovered.

$$
\boxed{
Q(s,a)
=
\sum_i
z_i
p_i,
}
$$

where

- \(z_i\) is the atom,
- \(p_i\) is its probability.

Thus,

standard action selection remains possible.

---



# <font color='orange'> 11. Quantile Regression DQN (QR-DQN) </font>

C51 uses

fixed atoms.

Another approach,

called

**Quantile Regression DQN (QR-DQN),**

learns the locations of the atoms themselves.

Instead of predicting probabilities for fixed values,

the network predicts

selected quantiles of the return distribution.

This often provides a more flexible and accurate approximation.

---



# <font color='orange'> 12. Why Distributional RL Works Better </font>

Learning the entire distribution allows the network to

- distinguish risky and safe actions,
- capture uncertainty,
- preserve more information during Bellman updates,
- learn richer state representations.

Although the final decision may still use the expected return,

training benefits from the additional distributional information.

---



# <font color='orange'> 13. Applications </font>

Distributional RL has been applied to

- Atari games,
- robotics,
- finance,
- autonomous systems,
- resource allocation,
- game playing.

It is an important component of Rainbow DQN and many later Deep Reinforcement Learning algorithms.

---

Return

$$
\boxed{
G_t
=
\sum_{k=0}^{\infty}
\gamma^k
R_{t+k+1}.
}
$$

Distributional Return

$$
\boxed{
Z(s,a).
}
$$

Expected Return

$$
\boxed{
Q(s,a)
=
\mathbb E
\left[
Z(s,a)
\right].
}
$$

Distributional Bellman Equation

$$
\boxed{
Z(s,a)
\overset{D}{=}
R
+
\gamma
Z(s',a'),
}
$$

where

$$
\overset{D}{=}
$$

denotes **equality in distribution**, meaning that the random variables on both sides have the same probability distribution, even though they are not necessarily equal in every individual outcome.

For C51,

the support atoms are

$$
\boxed{
z_i
=
V_{\min}
+
i
\frac{V_{\max}-V_{\min}}
{N-1}.
}
$$

The expected Q-value is recovered using

$$
\boxed{
Q(s,a)
=
\sum_i
z_i
p_i.
}
$$

# <font color='orange'> 15. C51 vs QR-DQN </font>

| C51 | QR-DQN |
|:---|:---|
| Fixed support atoms | Learnable quantiles |
| Predicts probabilities | Predicts quantile locations |
| Simpler implementation | More flexible approximation |
| Original Distributional DQN | Improved distributional representation |

Both algorithms model the full return distribution rather than only its expectation.

---



# <font color='orange'> 16. Relationship to Rainbow DQN </font>

Rainbow DQN combines

- Double DQN,
- Dueling Networks,
- Prioritized Experience Replay,
- Multi-Step Learning,
- Distributional RL,
- Noisy Networks.

Distributional RL contributes by changing

**what the network predicts**,

namely,

an entire return distribution instead of a single expected value.

---



# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> Distributional RL predicts multiple future trajectories.

**False.**

Distributional RL predicts the **probability distribution of returns**, not individual future state sequences.

---

### Misconception 2

> Distributional RL replaces the Q-function.

**False.**

The Q-function is still available. It is simply obtained as the expected value of the learned return distribution.

---

### Misconception 3

> A larger expected return always implies a better action.

**Not necessarily.**

Two actions may have identical expected returns but very different distributions. One may be highly reliable, while the other is much more uncertain. Distributional RL captures this distinction, even if a downstream decision rule ultimately uses the expectation.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Distributional RL | Learns the full probability distribution of future returns |
| Random Return \(Z(s,a)\) | Distribution of discounted returns |
| Expected Return | Mean of the return distribution |
| Distributional Bellman Equation | Bellman equation applied to return distributions |
| C51 | Uses fixed support atoms with learned probabilities |
| QR-DQN | Learns quantiles of the return distribution |
| Rainbow DQN | Combines Distributional RL with several other DQN improvements |

> **Key Insight:** Distributional Reinforcement Learning extends classical value-based reinforcement learning by modeling the entire probability distribution of future returns rather than only their expected value. This richer representation captures uncertainty and variability in future outcomes, leading to more informative value estimates and improved learning. Algorithms such as C51 and QR-DQN demonstrate that predicting return distributions, rather than single expected values, can significantly improve the performance and stability of deep reinforcement learning systems.
