# <font color='blue'> Appendix A3: Reinforcement Learning from Human Feedback (RLHF) </font>

After Supervised Fine-Tuning,

a language model can follow instructions and answer questions.

However,

many different responses may all be correct,

yet humans clearly prefer some responses over others.

For example,

humans generally prefer responses that are

- helpful,
- truthful,
- concise,
- polite,
- safe.

These qualities are difficult to specify using ordinary supervised learning.

Reinforcement Learning from Human Feedback (RLHF) addresses this problem.

Instead of learning directly from demonstration responses,

the model learns from **human preferences**.

Rather than asking

"What is the correct answer?",

RLHF asks

"Which answer do humans prefer?"

This transforms language model alignment into a Reinforcement Learning problem.

---

# <font color='orange'> 1. Motivation </font>

Suppose we ask

```
Explain gravity.
```

The language model generates two responses.

### Response A

```
Technically correct

Long

Difficult to read
```

### Response B

```
Simple

Clear

Easy to understand
```

A human reviewer chooses

```
Response B
```

This comparison provides valuable information,

even though both responses are correct.

RLHF learns from these preference comparisons.

---

# <font color='orange'> 2. Why Supervised Fine-Tuning is Not Enough </font>

Supervised Fine-Tuning teaches

one demonstration

for each instruction.

However,

language is inherently flexible.

Many different answers may be acceptable.

SFT cannot express

which acceptable answer

is **better**.

Preference comparisons solve this problem naturally.

---

# <font color='orange'> 3. Human Preference Data </font>

Annotators are presented with

an instruction

and two candidate responses.

```
Prompt

↓

Response A

↓

Response B

↓

Choose Preferred Response
```

The chosen response becomes

a preference example.

Large collections of these comparisons are used to train a reward model.

---

# <font color='orange'> 4. The RLHF Pipeline </font>

The complete RLHF process consists of five stages.

```
Internet Text

↓

Pretraining

↓

Supervised Fine-Tuning

↓

Collect Human Preferences

↓

Train Reward Model

↓

Reinforcement Learning (PPO)

↓

Aligned Assistant
```

Each stage builds upon the previous one.

---

# <font color='orange'> 5. Reward Model </font>

Instead of asking humans

to score every future response,

we train another neural network

called the **Reward Model**.

Its purpose is

to predict

how much humans would like

a response.

The Reward Model outputs

a single scalar value

$$
\boxed{
r(x,y),
}
$$

where

- \(x\) is the prompt,
- \(y\) is the generated response.

Higher rewards indicate

greater human preference.

---

# <font color='orange'> 6. Reinforcement Learning Formulation </font>

After the Reward Model has been trained,

the language model becomes

the RL agent.

The interaction is

```
Prompt

↓

Policy (Language Model)

↓

Generated Response

↓

Reward Model

↓

Reward

↓

Policy Update
```

Unlike classical Reinforcement Learning,

the "environment"

is the Reward Model rather than a physical world.

---

# <font color='orange'> 7. Policy </font>

The policy is

the language model itself.

Instead of choosing

robot actions,

it chooses

the next token.

Mathematically,

the policy is

$$
\boxed{
\pi_\theta(y|x),
}
$$

which represents

the probability

of generating response

$$
y
$$

given prompt

$$
x.
$$

---

# <font color='orange'> 8. Reward </font>

After generating

a complete response,

the Reward Model computes

$$
\boxed{
r(x,y).
}
$$

Unlike pretraining,

the model is no longer optimizing

next-token prediction.

Instead,

it attempts to generate responses

that receive larger rewards.

---

# <font color='orange'> 9. Why PPO is Used </font>

The model already knows

how to write.

We do not want

Reinforcement Learning

to completely overwrite

its language ability.

Instead,

PPO makes

small,

stable updates

to the policy.

This allows the model

to improve gradually

while preserving fluency.

---

# <font color='orange'> 10. KL-Divergence Penalty </font>

Suppose RL optimization

changes the model too aggressively.

The resulting language model

may become unnatural

or repetitive.

To prevent this,

RLHF introduces

a **KL-divergence penalty**.

The optimization objective becomes

```
High Reward

+

Stay Close to Original Model
```

The policy is therefore encouraged

to improve,

but not change too dramatically.

---

# <font color='orange'> 11. RLHF Objective </font>

Conceptually,

the optimization objective becomes

$$
\boxed{
\text{Reward}
-
\beta
\times
\text{KL Penalty}.
}
$$

The parameter

$$
\beta
$$

controls

how strongly

the updated model

is encouraged to remain close

to the original Supervised Fine-Tuned model.

---

# <font color='orange'> 12. Why RLHF Works </font>

Instead of memorizing

fixed demonstrations,

the model learns

general principles such as

- helpfulness,
- politeness,
- clarity,
- harmlessness.

These properties are encoded

through human preferences,

rather than explicit programming.

---

# <font color='orange'> 13. Applications </font>

RLHF has been used to improve

- conversational assistants,
- coding assistants,
- summarization systems,
- question-answering systems,
- dialogue agents.

It was a key technique used in early generations of conversational language models.

---

# <font color='orange'> 14. Limitations </font>

RLHF also has limitations.

It requires

- large numbers of human preference comparisons,
- expensive Reward Model training,
- reinforcement learning optimization,
- careful tuning of PPO.

These requirements make RLHF computationally expensive.

This motivates newer methods,

such as

Direct Preference Optimization (DPO),

which eliminate several of these stages.

---

# <font color='red'> 15. Mathematical Foundations </font>

### Policy

The language model defines

$$
\boxed{
\pi_\theta(y|x),
}
$$

the probability of generating response

$$
y
$$

for prompt

$$
x.
$$

---

### Reward Model

The Reward Model predicts

$$
\boxed{
r(x,y).
}
$$

Higher values indicate

that humans are expected to prefer

the generated response.

---

### KL Divergence

The updated policy

should remain close

to the reference policy

obtained after Supervised Fine-Tuning.

The KL divergence is

$$
\boxed{
D_{\mathrm{KL}}
(\pi_\theta\|\pi_{\mathrm{ref}}).
}
$$

A larger KL divergence indicates

greater deviation

from the reference model.

---

### RLHF Objective

The objective optimized during RLHF can be written conceptually as

$$
\boxed{
J(\theta)
=
\mathbb E
\left[
r(x,y)
\right]
-
\beta
D_{\mathrm{KL}}
\left(
\pi_\theta
\|
\pi_{\mathrm{ref}}
\right).
}
$$

The first term encourages responses that humans prefer,

while the second term prevents the policy from drifting too far from the supervised fine-tuned model.

In practice,

PPO is commonly used to approximately optimize this objective.

---

# <font color='orange'> 16. Supervised Fine-Tuning vs RLHF </font>

| Supervised Fine-Tuning | RLHF |
|:---|:---|
| Learns from demonstrations | Learns from preferences |
| Cross-Entropy Loss | Reinforcement Learning objective |
| One target response | Relative ranking of responses |
| Imitates examples | Optimizes human preference |
| No reward model | Uses a learned reward model |

The two stages complement one another.

SFT provides a strong initial policy,

while RLHF aligns it with human preferences.

---

# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> RLHF teaches the model new factual knowledge.

**False.**

Most factual knowledge is acquired during pretraining. RLHF primarily changes *how* the model presents information—making responses more helpful, safer, and better aligned with human preferences.

---

### Misconception 2

> Humans assign rewards during every conversation.

**False.**

Humans provide preference data only during the training process. After training, the Reward Model estimates human preferences automatically.

---

### Misconception 3

> RLHF completely replaces Supervised Fine-Tuning.

**False.**

RLHF depends on a strong Supervised Fine-Tuned model. Without SFT, reinforcement learning would begin from a much weaker policy and would be considerably less effective.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| RLHF | Aligns language models using human preferences |
| Preference Comparison | Humans rank candidate responses |
| Reward Model | Predicts how much humans would prefer a response |
| Policy | The language model being optimized |
| PPO | Reinforcement learning algorithm used to improve the policy |
| KL Penalty | Prevents the policy from changing too drastically |
| Alignment | Adjusting model behaviour to better match human preferences |

> **Key Insight:** Reinforcement Learning from Human Feedback transforms language model alignment into a reinforcement learning problem. Rather than optimizing only next-token prediction, the model learns to generate responses that humans are likely to prefer. A Reward Model estimates these preferences from human comparison data, and PPO updates the language model while a KL-divergence penalty keeps it close to the supervised fine-tuned policy. RLHF played a central role in the development of early conversational AI systems by enabling models to produce responses that are more helpful, coherent, and aligned with human expectations.
