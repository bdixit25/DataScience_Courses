# <font color='blue'> Chapter 63: Deep Deterministic Policy Gradient (DDPG) </font>

Deep Q Networks are highly successful for problems with **discrete action spaces**.

However,

many real-world control problems require actions that vary continuously,

such as

- steering angles,
- motor torques,
- robot joint positions,
- throttle settings.

Searching over infinitely many actions is impossible.

Deep Deterministic Policy Gradient (DDPG) solves this problem by learning

- a policy that directly outputs continuous actions,
- and a value function that evaluates those actions.

DDPG combines ideas from

- Deep Q Networks,
- Actor–Critic methods,
- Deterministic Policy Gradients.

It is one of the first successful Deep Reinforcement Learning algorithms for continuous control.

---



# <font color='orange'> 1. Motivation </font>

Suppose a robot arm must rotate by

```
37.26°
```

A DQN would need to evaluate

```
37.25°

37.26°

37.27°

37.28°

...
```

There are infinitely many possibilities.

Instead,

DDPG learns

a neural network that outputs

```
37.26°
```

directly.

No search is required.

---



# <font color='orange'> 2. Discrete vs Continuous Actions </font>

Discrete actions

```
Left

Right

Jump

Stop
```

Continuous actions

```
Steering Angle

↓

-30°

↓

...

↓

12.57°

↓

...

↓

+30°
```

Continuous action spaces contain infinitely many possible actions.

---



# <font color='orange'> 3. Deterministic Policies </font>

Most policy-gradient methods learn

a stochastic policy

$$
\pi(a|s),
$$

which outputs

a probability distribution over actions.

DDPG instead learns

a deterministic policy,

written

$$
\boxed{
a
=
\mu_\theta(s),
}
$$

where

- \(s\) is the current state,
- \(\mu_\theta\) is the Actor network.

For every state,

the Actor outputs exactly one action.

---



# <font color='orange'> 4. The Actor–Critic Architecture </font>

DDPG uses two neural networks.

```
State

↓

Actor

↓

Action

↓

Critic

↓

Q-value
```

The Actor decides

which action to take.

The Critic estimates

how good

that action is.

---



# <font color='orange'> 5. The Actor </font>

The Actor represents

the deterministic policy

$$
\boxed{
a
=
\mu_\theta(s).
}
$$

Its objective is

to choose actions

that maximize

the Critic's estimated Q-value.

Unlike DQN,

the Actor never searches over actions.

---



# <font color='orange'> 6. The Critic </font>

The Critic estimates

$$
\boxed{
Q(s,a).
}
$$

Its input consists of

both

- the state,
- the action.

```
State

+

Action

↓

Critic

↓

Q-value
```

The Critic therefore evaluates

the action proposed by the Actor.

---



# <font color='orange'> 7. Bellman Target </font>

The Critic learns

using a Bellman target similar to DQN.

The target is

$$
\boxed{
y
=
r
+
\gamma
Q'
\left(
s',
\mu'(s')
\right),
}
$$

where

- \(Q'\) is the target Critic,
- \(\mu'\) is the target Actor.

Instead of maximizing over actions,

the target Actor directly provides the next action.

---



# <font color='orange'> 8. Actor Objective </font>

The Actor aims

to maximize

the Critic's prediction.

Conceptually,

it solves

$$
\boxed{
\max_\theta
Q
\left(
s,
\mu_\theta(s)
\right).
}
$$

The Actor therefore learns

actions that the Critic believes will produce high returns.

---



# <font color='orange'> 9. Deterministic Policy Gradient </font>

Unlike stochastic policy gradients,

DDPG uses

the deterministic policy gradient theorem.

The gradient is

$$
\boxed{
\nabla_\theta J
=
\mathbb E
\left[
\nabla_a
Q(s,a)
\,
\nabla_\theta
\mu_\theta(s)
\right].
}
$$

The Critic tells the Actor

how changing the action

changes the expected return.

The Actor then adjusts its parameters

to produce better actions.

---



# <font color='orange'> 10. Target Networks </font>

Like DQN,

DDPG maintains

two target networks.

```
Actor

↓

Target Actor

Critic

↓

Target Critic
```

These target networks stabilize learning.

---



# <font color='orange'> 11. Soft Target Updates </font>

Unlike DQN,

DDPG does not copy parameters abruptly.

Instead,

target parameters are updated gradually.

$$
\boxed{
\theta'
\leftarrow
\tau\theta
+
(1-\tau)\theta',
}
$$

where

$$
\tau
\ll
1,
$$

typically

$$
\tau
=
0.001.
$$

This process is called

**soft updating**.

---



# <font color='orange'> 12. Experience Replay </font>

DDPG also uses

Experience Replay.

Transitions

$$
(s,a,r,s')
$$

are stored

inside a replay buffer

and sampled randomly.

This reduces correlations

between consecutive experiences.

---



# <font color='orange'> 13. Exploration </font>

Since

the policy is deterministic,

exploration must be added explicitly.

The selected action becomes

$$
\boxed{
a
=
\mu(s)
+
\mathcal N,
}
$$

where

$$
\mathcal N
$$

is exploration noise.

Common choices include

- Gaussian noise,
- Ornstein–Uhlenbeck (OU) noise.

This allows the agent to explore nearby actions during training.

---



# <font color='orange'> 14. Applications </font>

DDPG has been successfully applied to

- robotic manipulation,
- autonomous driving,
- drone control,
- industrial process control,
- continuous optimization.

It was one of the first algorithms capable of solving complex continuous-control tasks using deep neural networks.

---



# <font color='orange'> 15. Limitations </font>

Although powerful,

DDPG has several weaknesses.

It

- can overestimate Q-values,
- is sensitive to hyperparameters,
- may become unstable,
- often requires careful tuning.

These limitations motivated

Twin Delayed DDPG (TD3),

which addresses many of these problems.

---

# <font color='orange'> 17. DQN vs DDPG </font>

| DQN | DDPG |
|:---|:---|
| Discrete actions | Continuous actions |
| Searches over actions | Actor outputs actions directly |
| One Q-network | Actor + Critic |
| ε-greedy exploration | Action-space noise |
| Hard target updates | Soft target updates |

DDPG extends many of the ideas from DQN while making them suitable for continuous control.

---



# <font color='orange'> 18. Common Misconceptions </font>

### Misconception 1

> DDPG is simply DQN with continuous actions.

**False.**

DDPG uses an Actor–Critic architecture and deterministic policy gradients rather than searching over actions with a Q-network.

---

### Misconception 2

> DDPG does not require exploration.

**False.**

Because the policy is deterministic, exploration must be added explicitly by injecting noise into the Actor's output during training.

---

### Misconception 3

> Soft target updates and hard target updates are identical.

**False.**

Hard updates copy the online network parameters to the target network at fixed intervals. Soft updates gradually move the target parameters toward the online parameters, leading to smoother and more stable learning.

---

# <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| DDPG | Actor–Critic algorithm for continuous action spaces |
| Deterministic Policy | Outputs a single action for each state |
| Actor | Learns the policy \(\mu_\theta(s)\) |
| Critic | Estimates \(Q(s,a)\) |
| Deterministic Policy Gradient | Updates the Actor using gradients from the Critic |
| Experience Replay | Stores transitions for stable training |
| Soft Target Updates | Gradually update target networks |
| Exploration Noise | Enables exploration in deterministic policies |

> **Key Insight:** Deep Deterministic Policy Gradient extends deep reinforcement learning to continuous action spaces by combining an Actor–Critic architecture with deterministic policy gradients. Instead of searching over infinitely many possible actions, the Actor directly outputs a continuous action, while the Critic evaluates its quality. Experience Replay and soft target updates stabilize learning, and exploration is achieved by adding noise to the Actor's actions. DDPG laid the foundation for later continuous-control algorithms such as TD3 and Soft Actor–Critic.
