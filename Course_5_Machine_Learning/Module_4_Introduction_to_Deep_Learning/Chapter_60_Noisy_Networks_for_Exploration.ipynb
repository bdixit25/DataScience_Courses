# <font color='blue'> Chapter 60: Noisy Networks for Exploration (NoisyNet) </font>

Efficient exploration is one of the central challenges in Reinforcement Learning.

Traditional algorithms often use ε-greedy exploration,

where random actions are occasionally selected to discover new behaviours.

Although simple,

ε-greedy exploration ignores what the agent has already learned.

Every exploratory action is completely random,

regardless of whether the agent is highly uncertain or already confident.

Noisy Networks (NoisyNet) introduce a fundamentally different approach.

Instead of adding randomness to the selected actions,

they introduce **learnable stochasticity into the neural network itself**.

The network automatically learns

- when exploration is useful,
- where exploration is useful,
- and how much exploration is required.

This produces more intelligent and adaptive exploration strategies.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot has almost learned to walk.

With ε-greedy exploration,

```
Walk

↓

Walk

↓

Random Jump

↓

Fall
```

The random exploratory action ignores everything the robot has learned.

Instead,

we would like exploration to be

small,

structured,

and dependent upon the current uncertainty.

---

# <font color='orange'> 2. Why ε-Greedy Has Limitations </font>

Recall the ε-greedy strategy.

With probability

$$
1-\varepsilon,
$$

the greedy action is selected.

With probability

$$
\varepsilon,
$$

a completely random action is selected.

Problems include

- exploration is unstructured,
- ε must be manually tuned,
- exploration does not depend on the current state,
- exploration continues even when unnecessary.

---

# <font color='orange'> 3. A Different Idea </font>

Rather than making

the actions random,

suppose

the neural network itself becomes slightly random.

```
State

↓

Noisy Neural Network

↓

Q-values

↓

Action
```

Because the network changes slightly,

its predictions also change,

naturally producing exploration.

---

# <font color='orange'> 4. Noisy Linear Layers </font>

A standard linear layer computes

$$
\boxed{
y
=
Wx+b,
}
$$

where

- \(W\) is the weight matrix,
- \(b\) is the bias vector.

NoisyNet modifies this layer by adding random perturbations.

---

# <font color='orange'> 5. Adding Learnable Noise </font>

Instead of fixed weights,

the network uses

$$
\boxed{
W
=
\mu_W
+
\sigma_W
\odot
\varepsilon_W,
}
$$

where

- \(\mu_W\) is the learned mean weight,
- \(\sigma_W\) controls the magnitude of the noise,
- \(\varepsilon_W\) is random noise,
- \(\odot\) denotes element-wise multiplication.

The bias is modified similarly.

---

# <font color='orange'> 6. Complete Noisy Layer </font>

The noisy linear transformation becomes

$$
\boxed{
y
=
(W+\sigma_W\odot\varepsilon_W)x
+
(b+\sigma_b\odot\varepsilon_b).
}
$$

Unlike ε-greedy,

the amount of exploration is now determined by the network itself.

---

# <font color='orange'> 7. Learning the Noise </font>

The important observation is that

$$
\sigma_W
$$

is **trainable**.

If exploration is useful,

gradient descent increases

$$
\sigma_W.
$$

If exploration is no longer beneficial,

training naturally reduces

$$
\sigma_W
\rightarrow 0.
$$

The network therefore learns

how much randomness it needs.

---

# <font color='orange'> 8. Structured Exploration </font>

Because

every forward pass

uses slightly different parameters,

the policy changes

smoothly.

Instead of taking

isolated random actions,

the agent explores

consistent sequences of actions.

This often produces much more meaningful exploration.

---

# <font color='orange'> 9. Training </font>

Training proceeds almost exactly like DQN.

```
State

↓

Noisy Network

↓

Q-values

↓

Action

↓

Replay Buffer

↓

Gradient Descent
```

The only architectural change

is that some fully connected layers

are replaced

by noisy linear layers.

---

# <font color='orange'> 10. Advantages </font>

Noisy Networks

- eliminate manual ε scheduling,
- produce adaptive exploration,
- learn state-dependent exploration,
- often improve sample efficiency,
- integrate naturally with DQN.

---

# <font color='orange'> 11. Limitations </font>

Although effective,

Noisy Networks

- increase the number of trainable parameters,
- introduce additional computational overhead,
- do not eliminate the need for good network architectures,
- may not outperform ε-greedy in every environment.

---

# <font color='orange'> 12. Applications </font>

Noisy Networks have been successfully applied to

- Atari games,
- robotics,
- autonomous navigation,
- game playing,
- industrial optimization.

They are commonly used as one component of Rainbow DQN.

---

# <font color='orange'> 13. Relationship to Rainbow DQN </font>

Rainbow combines

- Double DQN,
- Dueling Networks,
- Prioritized Experience Replay,
- Multi-Step Learning,
- Distributional RL,
- Noisy Networks.

Noisy Networks replace

ε-greedy exploration

with

learnable parameter-space exploration.

---

# <font color='red'> 14. Mathematical Foundations </font>

Standard linear layer

$$
\boxed{
y
=
Wx+b.
}
$$

Noisy weights

$$
\boxed{
W
=
\mu_W
+
\sigma_W
\odot
\varepsilon_W.
}
$$

Noisy bias

$$
\boxed{
b
=
\mu_b
+
\sigma_b
\odot
\varepsilon_b.
}
$$

Complete noisy layer

$$
\boxed{
y
=
(W+\sigma_W\odot\varepsilon_W)x
+
(b+\sigma_b\odot\varepsilon_b).
}
$$

where

- \(\mu\) are learned parameters,
- \(\sigma\) are learned noise scales,
- \(\varepsilon\) are random variables, typically sampled from a standard Gaussian distribution.

As training progresses,

gradient descent optimizes both

the network parameters

and

the amount of exploration.

---

# <font color='orange'> 15. ε-Greedy vs Noisy Networks </font>

| ε-Greedy | Noisy Networks |
|:---|:---|
| Random actions | Random network parameters |
| Fixed exploration schedule | Learnable exploration |
| Requires tuning ε | No ε parameter |
| Exploration independent of state | Exploration adapts through the network |
| Simple implementation | Slightly more complex architecture |

---

# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> Noisy Networks randomly change the rewards.

**False.**

The rewards remain unchanged. Noise is added only to the neural network parameters, affecting the predicted Q-values and therefore the chosen actions.

---

### Misconception 2

> The amount of noise is fixed throughout training.

**False.**

The noise scale parameters \(\sigma\) are learned during training. The network can increase or decrease exploration depending on what improves performance.

---

### Misconception 3

> Noisy Networks always outperform ε-greedy exploration.

**False.**

Noisy Networks often improve exploration, particularly in complex environments, but their effectiveness depends on the task and the overall learning algorithm.

---

# <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Noisy Networks | Neural networks with learnable stochastic parameters |
| Parameter-Space Exploration | Exploration generated by perturbing network weights rather than actions |
| Noisy Linear Layer | Linear layer with trainable mean and noise parameters |
| Learnable Noise Scale | Controls the amount of exploration during training |
| Adaptive Exploration | Exploration decreases or increases automatically as learning progresses |
| Rainbow DQN | Incorporates Noisy Networks as one of its core improvements |

> **Key Insight:** Noisy Networks replace the external exploration strategy used in traditional reinforcement learning with learnable stochasticity inside the neural network itself. By perturbing the network parameters rather than the selected actions, exploration becomes structured, adaptive, and state-dependent. Because the magnitude of the noise is learned through gradient descent, the agent automatically discovers how much exploration is needed at different stages of training, making Noisy Networks an effective alternative to ε-greedy exploration in modern deep reinforcement learning.
