# <font color='blue'> Chapter 23: Understanding CNN Feature Maps and Filters </font>

After training a Convolutional Neural Network (CNN), an obvious question arises:

> **What has the network actually learned?**

Unlike traditional machine learning models, CNNs learn their own features directly from the training data. These learned features are stored in the weights of the convolutional filters.

Fortunately, these filters and their outputs can be visualized, allowing us to understand how CNNs transform raw pixels into increasingly sophisticated visual representations.

This chapter explores feature maps, learned filters, activation visualizations, saliency maps, and Grad-CAM, providing an introduction to model interpretability in deep learning.

---



# <font color='orange'> 1. What Does a CNN Learn? </font>

Recall that every convolutional layer contains several learnable filters.

Initially,

these filters contain random weights.

```
Random Filter

↓

Random Feature Map
```

After training,

the filters become specialized.

Examples include

- horizontal edges,
- vertical edges,
- corners,
- textures,
- curves,
- object parts.

The network learns these features automatically through backpropagation.

---



# <font color='orange'> 2. Filters vs Feature Maps </font>

These terms are often confused.

### Filter (Kernel)

A filter is

- a small matrix of trainable weights,
- learned during training.

Example

$$
K=
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}
$$

---

### Feature Map

A feature map is

the output obtained after applying a filter to the input image.

```
Image

↓

Filter

↓

Feature Map
```

One filter produces one feature map.

If a layer contains

64 filters,

it produces

64 feature maps.

---



# <font color='orange'> 3. Early Layers Learn Simple Features </font>

The first convolutional layers usually detect

- edges,
- gradients,
- corners,
- simple textures.

```
Input

↓

Vertical Edge

↓

Horizontal Edge

↓

Diagonal Edge
```

These low-level features appear in almost every natural image.

---



# <font color='orange'> 4. Deeper Layers Learn Complex Features </font>

As information passes through the network,

simple features combine into increasingly complex representations.

```
Edges

↓

Corners

↓

Shapes

↓

Object Parts

↓

Complete Objects
```

For example,

a network trained on animals may gradually learn

- whiskers,
- ears,
- eyes,
- noses,

before finally recognizing

a cat.

---

# <font color='orange'> 7. What Do Feature Maps Show? </font>

Suppose

one feature map responds strongly to vertical edges.

Bright regions indicate

large activations.

Dark regions indicate

weak activations.

Thus,

feature maps reveal

which image regions activate a particular filter.

---



# <font color='orange'> 8. Activation Maximization </font>

Instead of asking

"What does this image activate?"

we ask

"What image would maximally activate this neuron?"

Starting from random noise,

gradient ascent modifies the image until one neuron becomes highly active.

The resulting image provides insight into

what that neuron has learned to detect.

---



# <font color='orange'> 9. Saliency Maps </font>

A saliency map answers

> Which pixels influenced the prediction most?

Mathematically,

we compute

$$
\boxed{
\frac{\partial y}
{\partial I},
}
$$

where

- $I$ is the input image,
- $y$ is the predicted score.

Pixels with large gradients have the greatest influence on the prediction.

---



# <font color='orange'> 10. Grad-CAM </font>

Grad-CAM

(**Gradient-weighted Class Activation Mapping**)

is one of the most widely used CNN interpretation techniques.

Instead of highlighting individual pixels,

Grad-CAM highlights

important image regions.

```
Original Image

↓

CNN

↓

Prediction

↓

Grad-CAM

↓

Heat Map
```

Bright colours indicate regions that most influenced the prediction.

---



# <font color='orange'> 11. Why Grad-CAM Works </font>

Grad-CAM computes

the gradients of the predicted class

with respect to

the final convolutional feature maps.

These gradients determine

which feature maps were most important.

The weighted feature maps are then combined into a heat map.

---



# <font color='orange'> 12. Occlusion Sensitivity </font>

Another interpretability technique is

Occlusion Sensitivity.

Instead of computing gradients,

small image patches are hidden one at a time.

If hiding a region greatly reduces confidence,

that region was important for the prediction.

```
Image

↓

Hide Patch

↓

Prediction Drops

↓

Important Region
```

---



# <font color='orange'> 13. Why Explainability Matters </font>

Model interpretation is especially important in

- medicine,
- autonomous driving,
- finance,
- scientific research.

For example,

a medical CNN should classify an X-ray

based on the tumour,

not based on

image labels

or

hospital markings.

Explainability helps identify such issues.

---



# <font color='orange'> 14. Limitations of Visualization </font>

Although visualization techniques provide valuable insight,

they do not fully explain

how neural networks reason.

Different interpretation methods may highlight different regions,

and visual explanations should be treated as aids rather than definitive proofs of the model's decision process.

---

# <font color='red'> 15. Mathematical Foundations </font>

Suppose

the CNN predicts

$$
y_c
$$

for class

$$
c.
$$

The importance of an input pixel is measured by

$$
\boxed{
\frac{\partial y_c}
{\partial I_{ij}}.
}
$$

Large values indicate

that small changes in pixel

$$
(i,j)
$$

strongly influence the prediction.

Grad-CAM instead computes

the gradients with respect to the final convolutional feature maps,

combining them into a coarse localization heat map.

Thus,

Backpropagation,

which was originally developed for learning,

also enables model interpretation.

# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> CNN filters are manually designed.

**False.**

CNN filters are initialized randomly and learned automatically during training through backpropagation.

---

### Misconception 2

> One filter detects one object.

**False.**

A filter usually detects a visual feature, such as an edge, texture, or pattern. Object recognition emerges from the combined responses of many filters across multiple layers.

---

### Misconception 3

> Grad-CAM proves why a model made a decision.

**False.**

Grad-CAM provides a useful visualization of influential regions, but it is an interpretability tool rather than a complete explanation of the model's reasoning.

| Saliency Map | Highlights influential pixels |
| Grad-CAM | Highlights image regions most responsible for a prediction |

> **Key Insight:** Although CNNs are often viewed as black-box models, their internal representations can be explored through feature visualization techniques. Learned filters detect increasingly complex visual patterns, while methods such as saliency maps and Grad-CAM reveal which regions of an image influenced the network's prediction. These techniques are fundamental to Explainable AI and are particularly important in high-stakes applications where understanding model decisions is essential.
