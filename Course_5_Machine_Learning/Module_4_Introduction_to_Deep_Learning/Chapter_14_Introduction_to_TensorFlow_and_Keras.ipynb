# <font color='blue'> Chapter 14: Introduction to TensorFlow and Keras </font>

In the previous chapters, we studied the mathematical foundations of Deep Learning. We learned how neural networks make predictions using forward propagation, measure errors using loss functions, and improve themselves using backpropagation and gradient descent.

Although it is possible to implement these algorithms from scratch using only NumPy, doing so becomes impractical for large neural networks.

Modern Deep Learning therefore relies on specialised software frameworks that automatically perform tensor operations, compute gradients, optimise parameters, and efficiently utilise modern hardware such as GPUs and TPUs.

The most widely used of these frameworks is **TensorFlow**, which includes the high-level neural network API known as **Keras**.

This chapter introduces TensorFlow and Keras, explains their relationship, and prepares us for building our first deep learning models.

---



# <font color='orange'> 1. Why Do We Need Deep Learning Frameworks? </font>

Suppose we wish to train a neural network containing

- 50 input features,
- 10 hidden layers,
- 1000 neurons per layer.

Such a network contains millions of parameters.

Implementing

- forward propagation,
- backpropagation,
- gradient descent,

manually would require thousands of lines of code.

Deep learning frameworks automate these tasks.

They provide

- automatic differentiation,
- optimized linear algebra,
- GPU acceleration,
- efficient memory management,
- high-level neural network APIs.

---



# <font color='orange'> 2. What is TensorFlow? </font>

**TensorFlow** is an open-source machine learning framework developed by Google.

It provides tools for

- numerical computation,
- automatic differentiation,
- deep learning,
- distributed computing,
- GPU and TPU acceleration.

TensorFlow is designed to build and train machine learning models efficiently on devices ranging from laptops to large computing clusters.

---



# <font color='orange'> 3. What is Keras? </font>

**Keras** is a high-level API for building neural networks.

Originally,

Keras was developed as an independent library.

Today,

Keras is tightly integrated into TensorFlow and is accessed through

```python
tensorflow.keras
```

Keras hides much of the mathematical complexity of neural network implementation, allowing users to define sophisticated models using only a few lines of code.

---



# <font color='orange'> 4. TensorFlow vs Keras </font>

A common misconception is that TensorFlow and Keras are competing libraries.

In reality,

they work together.

```
Application

↓

Keras API

↓

TensorFlow Backend

↓

CPU / GPU / TPU
```

TensorFlow performs the computations,

while Keras provides a user-friendly interface.

---



# <font color='orange'> 5. What is a Tensor? </font>

The word **TensorFlow** comes from two ideas:

- Tensor
- Flow

A **tensor** is a mathematical object that generalises

- scalars,
- vectors,
- matrices,

to higher dimensions.

Examples

| Object | Rank | Example |
|:---|:---:|:---|
| Scalar | 0 | 5 |
| Vector | 1 | $[1,2,3]$ |
| Matrix | 2 | $\begin{bmatrix}1&2\\3&4\end{bmatrix}$ |
| Tensor | 3+ | Images, videos, batches of data |

Most neural network computations involve tensors.

---



# <font color='orange'> 6. TensorFlow Architecture </font>

A simplified view of TensorFlow is

```
Python Code

↓

Keras

↓

TensorFlow

↓

Linear Algebra Library

↓

CPU / GPU / TPU
```

Every layer that we create in Keras is eventually converted into tensor operations executed by TensorFlow.

---



# <font color='orange'> 7. Automatic Differentiation </font>

One of TensorFlow's most powerful features is **Automatic Differentiation**.

Recall that Backpropagation requires computing

$$
\frac{\partial L}{\partial W}
$$

for every parameter.

Instead of deriving these gradients manually,

TensorFlow computes them automatically.

This capability is known as **Automatic Differentiation** (AutoDiff).

Internally,

TensorFlow constructs a computational graph and applies the Chain Rule to compute gradients efficiently.

---



# <font color='orange'> 8. GPUs and TPUs </font>

Training modern neural networks requires billions of floating-point operations.

Matrix multiplication dominates the computational cost.

GPUs are designed to perform thousands of matrix operations simultaneously.

Consequently,

deep learning frameworks make extensive use of GPU acceleration.

For even larger models,

Google developed specialised hardware known as **Tensor Processing Units (TPUs)**.

# <font color='orange'> 10. Three Ways to Build Models in Keras </font>

Keras provides three modelling approaches.

### Sequential API

Suitable for simple feedforward networks.

```
Input

↓

Layer

↓

Layer

↓

Output
```

---

### Functional API

Supports

- multiple inputs,
- multiple outputs,
- branching architectures,
- residual connections.

---

### Model Subclassing

Provides maximum flexibility.

Useful for

- research,
- custom architectures,
- novel neural networks.

# <font color='red'> 15. Mathematical Foundations </font>

Every Dense layer in Keras performs the same computation we derived earlier.

For layer

$l$,

$$
\boxed{
\mathbf{a}^{(l)}
=
f\left(
W^{(l)}
\mathbf{a}^{(l-1)}
+
\mathbf{b}^{(l)}
\right).
}
$$

During training,

TensorFlow automatically computes

$$
\frac{\partial L}{\partial W^{(l)}}
\quad\text{and}\quad
\frac{\partial L}{\partial \mathbf{b}^{(l)}}
$$

using Backpropagation.

Thus,

Keras is not replacing the mathematics—it is implementing it efficiently.

# <font color='orange'> 16. Common Misconceptions </font>

### Misconception 1

> Keras is a different deep learning framework from TensorFlow.

**False.**

Modern Keras is integrated into TensorFlow and serves as its high-level API.

---

### Misconception 2

> Using Keras means you no longer need to understand neural networks.

**False.**

Keras simplifies implementation but does not replace the need for understanding the underlying mathematics.

---

### Misconception 3

> TensorFlow automatically makes every model fast.

**False.**

Model architecture, data preprocessing, hardware, and hyperparameter choices all influence training speed and performance.
