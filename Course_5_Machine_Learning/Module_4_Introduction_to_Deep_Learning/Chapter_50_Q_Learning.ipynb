# <font color='blue'> Chapter 50: Q-Learning </font>

SARSA introduced the idea of learning action-value functions using Temporal Difference Learning.

However,

SARSA learns the value of the policy that the agent is currently following.

In many situations,

we are more interested in learning the value of the **optimal policy**, even while the agent is still exploring.

Q-Learning achieves this objective.

It is an **off-policy** Reinforcement Learning algorithm that learns the optimal action-value function directly,

regardless of the behaviour policy used to explore the environment.

Q-Learning is one of the most influential algorithms in Reinforcement Learning and forms the basis of many modern Deep Reinforcement Learning methods.

---

# <font color='orange'> 1. Motivation </font>

Suppose a robot is exploring a maze.

Occasionally,

it makes random exploratory moves.

Should these exploratory actions determine

what the robot believes is optimal?

SARSA answers

**Yes**,

because it learns from the actions actually taken.

Q-Learning answers

**No**.

Instead,

it always assumes that

from the next state onward,

the best possible action will be chosen.

This allows the algorithm to learn the optimal policy even while exploring.

---

# <font color='orange'> 2. From SARSA to Q-Learning </font>

Recall the SARSA update rule

$$
Q(s,a)
\leftarrow
Q(s,a)
+
\alpha
\left[
r
+
\gamma
Q(s',a')
-
Q(s,a)
\right].
$$

Notice that

the update depends on

the next action actually selected,

$$
a'.
$$

Q-Learning replaces

this action

with

the best possible action.

---

# <font color='orange'> 3. The Bellman Optimality Equation </font>

Recall that

the optimal action-value function satisfies

$$
\boxed{
Q^*(s,a)
=
\mathbb E
\left[
r
+
\gamma
\max_{a'}
Q^*(s',a')
\right].
}
$$

Instead of evaluating

the current policy,

the Bellman Optimality Equation assumes

optimal decisions are always made in future states.

This equation provides the theoretical foundation of Q-Learning.

---

# <font color='orange'> 4. The Q-Learning Update Rule </font>

Replacing

the next action

by

the maximum future Q-value,

we obtain

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\left[
r_{t+1}
+
\gamma
\max_{a}
Q(s_{t+1},a)
-
Q(s_t,a_t)
\right].
}
$$

This is the Q-Learning algorithm.

Unlike SARSA,

the update no longer depends on

the action actually chosen during exploration.

---

# <font color='orange'> 5. The TD Error </font>

The Temporal Difference error becomes

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
\max_a
Q(s_{t+1},a)
-
Q(s_t,a_t).
}
$$

Interpretation

- Positive TD error

↓

better than expected.

- Negative TD error

↓

worse than expected.

The Q-value is adjusted according to this error.

---

# <font color='orange'> 6. Why the Maximum Operator? </font>

Suppose

the next state's Q-values are

| Action | Q-value |
|:---|---:|
| Left | 25 |
| Right | 80 |
| Up | 40 |
| Down | 10 |

Even if the exploration policy chooses

```
Left
```

Q-Learning updates using

```
Right
```

because

80

is the maximum.

The algorithm therefore learns

the value of the **optimal future policy**.

---

# <font color='orange'> 7. Off-Policy Learning </font>

Q-Learning is an

**off-policy**

algorithm.

This means

the policy used for exploration

does not need to be the same

as the policy being learned.

```
Behaviour Policy

↓

Generate Experience

↓

Learn Optimal Policy
```

Typically,

the behaviour policy is

ε-greedy,

while the learned target policy is greedy.

---

# <font color='orange'> 8. ε-Greedy Exploration </font>

To ensure sufficient exploration,

Q-Learning commonly uses an

ε-greedy behaviour policy.

With probability

$$
1-\varepsilon,
$$

the greedy action is selected.

With probability

$$
\varepsilon,
$$

a random action is selected.

Although random actions are executed,

the update still assumes

optimal future behaviour.

---

# <font color='orange'> 9. Complete Q-Learning Algorithm </font>

```
Initialize Q(s,a)

↓

Observe Current State

↓

Choose Action (ε-greedy)

↓

Execute Action

↓

Observe Reward

↓

Observe Next State

↓

Update Q-value

↓

Repeat
```

Unlike SARSA,

there is

no need

to choose the next action before updating.

---

# <font color='orange'> 10. SARSA vs Q-Learning </font>

Suppose a cliff exists beside the shortest path.

```
Goal

██████████████

Safe Path

██████████████

Cliff
```

With ε-greedy exploration,

the agent occasionally makes random moves.

SARSA learns

that exploratory moves near the cliff are dangerous,

so it often prefers a safer route.

Q-Learning assumes

future behaviour will always be optimal,

so it learns the shortest path,

even if exploration occasionally causes the agent to fall during training.

This example illustrates the difference between **on-policy** and **off-policy** learning.

---

# <font color='orange'> 11. Convergence </font>

Under suitable assumptions,

including

- every state-action pair being visited infinitely often,
- an appropriate learning-rate schedule,
- a stationary environment,

Q-Learning converges to the optimal action-value function

$$
Q^*(s,a).
$$

This is one of the major theoretical strengths of the algorithm.

---

# <font color='orange'> 12. Applications </font>

Q-Learning has been applied to

- robot navigation,
- warehouse automation,
- traffic signal control,
- resource allocation,
- scheduling,
- game playing,
- autonomous systems.

Its ideas also form the basis of Deep Q Networks, where neural networks approximate the Q-function.

# <font color='red'> 13. Mathematical Foundations </font>

The Bellman Optimality Equation is

$$
\boxed{
Q^*(s,a)
=
\mathbb E
\left[
r
+
\gamma
\max_{a'}
Q^*(s',a')
\right].
}
$$

The TD target becomes

$$
\boxed{
r_{t+1}
+
\gamma
\max_a
Q(s_{t+1},a).
}
$$

The TD error is

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
\max_a
Q(s_{t+1},a)
-
Q(s_t,a_t).
}
$$

The Q-Learning update rule is

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\delta_t.
}
$$

Substituting the TD error,

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\left[
r_{t+1}
+
\gamma
\max_a
Q(s_{t+1},a)
-
Q(s_t,a_t)
\right].
}
$$

This recursive update allows the agent to approximate the optimal action-value function without requiring a model of the environment.

# <font color='orange'> 14. SARSA vs Q-Learning Comparison </font>

| Feature | SARSA | Q-Learning |
|:---|:---|:---|
| Learning Type | On-policy | Off-policy |
| Update Uses | Action actually taken | Best possible next action |
| Update Rule | \(Q(s',a')\) | \(\max_a Q(s',a)\) |
| Behaviour During Exploration | Learns consequences of exploration | Learns optimal policy despite exploration |
| Typical Behaviour | More conservative | More aggressive |

---

# <font color='orange'> 15. Common Misconceptions </font>

### Misconception 1

> Q-Learning always follows the optimal policy while training.

**False.**

The behaviour policy often includes exploration (e.g., ε-greedy). Q-Learning learns the optimal policy even though it may not follow it during training.

---

### Misconception 2

> Q-Learning requires a model of the environment.

**False.**

Q-Learning is a model-free algorithm. It learns directly from sampled interactions with the environment.

---

### Misconception 3

> Q-Learning and SARSA differ only in notation.

**False.**

The algorithms optimize different objectives. SARSA evaluates the current behaviour policy, while Q-Learning approximates the optimal policy by using the maximum future action value.

---

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Q-Learning | Off-policy TD control algorithm |
| Action-Value Function \(Q(s,a)\) | Expected return for taking action \(a\) in state \(s\) |
| Bellman Optimality Equation | Recursive equation defining the optimal Q-function |
| TD Error | Difference between the TD target and current Q-value |
| Max Operator | Assumes optimal future actions |
| Off-Policy Learning | Learns a target policy different from the behaviour policy |
| ε-Greedy Exploration | Balances exploration with exploitation during training |

> **Key Insight:** Q-Learning extends Temporal Difference Learning by directly approximating the optimal action-value function. Unlike SARSA, which learns the value of the policy actually being followed, Q-Learning assumes that the best possible action will always be selected in future states. This off-policy approach enables the algorithm to learn the optimal policy while still exploring the environment, making Q-Learning one of the foundational algorithms in modern Reinforcement Learning.
