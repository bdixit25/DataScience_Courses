# <font color='blue'> Chapter 42: Reinforcement Learning from Human Feedback (RLHF) </font>

Large Language Models acquire language ability through large-scale pretraining and can be adapted to specific tasks using supervised fine-tuning.

However,

neither of these stages directly teaches a model

- what humans prefer,
- how to follow instructions consistently,
- how to provide helpful answers,
- how to avoid undesirable responses.

To address these challenges,

modern language models often undergo an additional alignment stage known as

**Reinforcement Learning from Human Feedback (RLHF).**

Rather than learning only from text,

the model learns from **human preferences**,

allowing it to produce responses that better match human expectations.

RLHF combines ideas from supervised learning,

reinforcement learning,

and human evaluation.

---

# <font color='orange'> 1. Why is RLHF Needed? </font>

Suppose a pretrained model receives the prompt

```
Explain black holes.
```

Possible responses include

Response A

```
A highly technical explanation using tensor calculus.
```

Response B

```
A simple explanation suitable for high-school students.
```

Both responses may be factually correct.

However,

most users would prefer

Response B.

Traditional next-token prediction has no mechanism for expressing this preference.

RLHF teaches the model

which responses humans actually prefer.

---

# <font color='orange'> 2. The Three-Stage Training Pipeline </font>

Modern language models are often trained in three stages.

```
Large Text Corpus

↓

Pretraining

↓

Supervised Fine-Tuning (SFT)

↓

Reinforcement Learning from Human Feedback (RLHF)

↓

Aligned Language Model
```

Each stage has a different purpose.

---

# <font color='orange'> 3. Stage 1 — Pretraining </font>

During pretraining,

the model performs

Next Token Prediction.

```
Input

↓

Predict Next Token

↓

Update Parameters
```

The model learns

- grammar,
- vocabulary,
- reasoning patterns,
- world knowledge.

However,

it has not yet learned human preferences.

---

# <font color='orange'> 4. Stage 2 — Supervised Fine-Tuning (SFT) </font>

Human-written demonstrations are collected.

Example

Prompt

```
Explain gravity.
```

Desired Response

```
Gravity is the attractive force between masses...
```

The model learns

to imitate high-quality human responses.

This improves instruction following,

but it still does not compare alternative answers.

---

# <font color='orange'> 5. Stage 3 — Human Preference Collection </font>

Human annotators compare multiple responses.

Example

Prompt

```
What is machine learning?
```

Response A

↓

Preferred

Response B

↓

Less Preferred

The model does **not** receive a numerical score.

Instead,

it receives information about

which response humans prefer.

---

# <font color='orange'> 6. Reward Model </font>

The collected human preferences are used to train

a separate neural network,

called the

**Reward Model**.

```
Prompt

+

Response

↓

Reward Model

↓

Reward Score
```

Higher scores correspond to responses that humans generally prefer.

The reward model approximates human judgement.

---

# <font color='orange'> 7. Reinforcement Learning </font>

The language model is now viewed as an

**agent**.

```
Prompt

↓

Agent

↓

Response

↓

Reward Model

↓

Reward

↓

Update Policy
```

The objective becomes

producing responses that maximize the reward predicted by the reward model.

---

# <font color='orange'> 8. Policy </font>

In reinforcement learning,

the model's behaviour is called its

**policy**.

The policy determines

which token the model generates next,

given the current context.

RLHF improves this policy

so that preferred responses become more likely.

---

# <font color='orange'> 9. Policy Optimization </font>

A common optimization algorithm used in early RLHF systems is

**Proximal Policy Optimization (PPO).**

Rather than allowing the model to change arbitrarily,

PPO updates the policy gradually,

preventing excessively large changes that could destabilize training.

This helps preserve useful language abilities while improving alignment.

---

# <font color='orange'> 10. Why Not Optimize Only the Reward? </font>

Suppose we maximize only the reward model.

The language model might discover undesirable strategies,

such as exploiting weaknesses in the reward model,

rather than genuinely producing better responses.

To reduce this risk,

training often includes a penalty for moving too far away from the original pretrained model.

This encourages the model to remain fluent while improving alignment.

---

# <font color='orange'> 11. RLHF Workflow </font>

The complete RLHF pipeline is

```
Large Text Corpus

↓

Pretraining

↓

Supervised Fine-Tuning

↓

Generate Candidate Responses

↓

Human Preference Ranking

↓

Train Reward Model

↓

Reinforcement Learning

↓

Aligned Language Model
```

Each stage builds upon the previous one.

---

# <font color='orange'> 12. Advantages of RLHF </font>

RLHF improves

- instruction following,
- helpfulness,
- conversational quality,
- politeness,
- consistency,
- user satisfaction.

It allows language models to better match human expectations without manually programming every behaviour.

---

# <font color='orange'> 13. Limitations </font>

RLHF also has limitations.

Human preferences

- may differ,
- may contain biases,
- may change over time.

Furthermore,

the reward model is only an approximation of human judgement.

Poorly designed reward functions may encourage undesirable behaviours.

Alignment therefore remains an active area of research.

---

# <font color='orange'> 14. Newer Alignment Methods </font>

Although RLHF has been highly successful,

newer methods have emerged.

One important example is

**Direct Preference Optimization (DPO).**

Instead of using reinforcement learning,

DPO directly optimizes the model using preference comparisons,

simplifying training while achieving competitive performance.

Other alignment methods continue to be developed,

but the underlying objective remains the same:

to align language model behaviour with human preferences.

---

# <font color='red'> 15. Mathematical Foundations </font>

Suppose

the language model has policy

$$
\pi_\theta.
$$

For a prompt

$$
x,
$$

the model generates a response

$$
y.
$$

The reward model assigns a score

$$
\boxed{
R(x,y).
}
$$

The objective is to maximize the expected reward

$$
\boxed{
J(\theta)
=
\mathbb{E}_{y\sim\pi_\theta}
\left[
R(x,y)
\right].
}
$$

To prevent the updated model from drifting too far from the original policy,

RLHF commonly includes a regularization term that penalizes large deviations from a reference model.

Conceptually,

the optimization objective becomes

$$
\boxed{
\text{Maximize Reward}
-
\text{Penalty for Excessive Policy Change}.
}
$$

This balances alignment with language quality.

---

# <font color='orange'> 16. RLHF vs Supervised Fine-Tuning </font>

| Supervised Fine-Tuning | RLHF |
|:---|:---|
| Learns from correct demonstrations | Learns from human preferences between responses |
| Uses labelled prompt–response pairs | Uses ranked or preferred responses |
| Imitates examples | Optimizes for preferred behaviour |
| No reward model | Uses a learned reward model |
| Supervised learning | Reinforcement learning |

Both stages are complementary and are commonly used together.

---

# <font color='orange'> 17. Common Misconceptions </font>

### Misconception 1

> RLHF teaches factual knowledge.

**False.**

Most factual knowledge is acquired during pretraining. RLHF primarily teaches the model how to present information in ways that better align with human preferences.

---

### Misconception 2

> The reward model is a human.

**False.**

Humans provide preference data, but the reward model is a neural network trained to approximate those preferences.

---

### Misconception 3

> RLHF guarantees perfect behaviour.

**False.**

RLHF improves alignment but does not eliminate mistakes, biases, or incorrect responses. Alignment remains an active area of research.

---

# <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| RLHF | Aligns language models using human preference data |
| Pretraining | Learns language through next-token prediction |
| Supervised Fine-Tuning | Learns from high-quality human demonstrations |
| Human Preferences | Rankings of alternative model responses |
| Reward Model | Predicts how humans are likely to evaluate a response |
| Policy | The model's strategy for generating responses |
| PPO | Reinforcement learning algorithm used in many early RLHF systems |
| DPO | Directly optimizes preference data without a separate reinforcement learning stage |

> **Key Insight:** Reinforcement Learning from Human Feedback extends pretrained language models beyond simple next-token prediction by incorporating human preferences into the training process. Human evaluators compare candidate responses, a reward model learns these preferences, and the language model is optimized to generate responses that maximize the predicted reward while preserving its general language capabilities. RLHF has played a central role in making modern conversational AI systems more helpful, reliable, and aligned with user expectations.
