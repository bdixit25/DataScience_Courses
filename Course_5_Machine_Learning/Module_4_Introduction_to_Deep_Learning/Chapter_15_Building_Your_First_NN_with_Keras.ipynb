# <font color='blue'> Chapter 15: Building Your First Neural Network with Keras </font>

In the previous chapter, we introduced TensorFlow and Keras and explained how these frameworks automate the mathematical operations involved in training neural networks.

In this chapter, we shall build our first neural network using the **Keras Sequential API**.

Rather than treating the code as a black box, we shall relate every line of code to the mathematical concepts developed throughout this course.

By the end of this chapter, you will understand how to

- create a neural network,
- define its architecture,
- compile the model,
- train it,
- evaluate its performance,
- and make predictions on unseen data.

---

# <font color='orange'> 1. The Keras Workflow </font>

Training a neural network in Keras follows six main steps.

```
Load Data

↓

Preprocess Data

↓

Define Model

↓

Compile Model

↓

Train Model

↓

Evaluate Model

↓

Predict
```

Although Keras performs many computations internally,

these are exactly the same operations that we studied mathematically.

---

# <font color='orange'> 2. Importing the Required Libraries </font>

The first step is importing TensorFlow and Keras.

```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
```

Here,

- **TensorFlow** performs the numerical computations,
- **Keras** provides the high-level neural network interface,
- **layers** contains the building blocks used to construct models.

---

# <font color='orange'> 3. Loading a Dataset </font>

For demonstration purposes,

Keras provides several built-in datasets.

For example,

the Fashion-MNIST dataset can be loaded using

```python
fashion_mnist = keras.datasets.fashion_mnist

(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

This dataset contains grayscale images of clothing items belonging to ten classes.

---

# <font color='orange'> 4. Preprocessing the Data </font>

Neural networks generally perform better when the input values are scaled.

Since image pixels lie between

$$
0
\quad\text{and}\quad
255,
$$

we divide every pixel by

$$
255.
$$

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

This rescales the data into

$$
[0,1].
$$

Feature scaling improves numerical stability and usually accelerates training.

---

# <font color='orange'> 5. Building the Neural Network </font>

The simplest way to construct a neural network is using the **Sequential API**.

```python
model = keras.Sequential([
    layers.Flatten(input_shape=(28,28)),
    layers.Dense(128, activation="relu"),
    layers.Dense(64, activation="relu"),
    layers.Dense(10, activation="softmax")
])
```

This network consists of

- an input layer,
- two hidden layers,
- one output layer.

---

# <font color='orange'> 6. Understanding Each Layer </font>

## Flatten Layer

```python
layers.Flatten(input_shape=(28,28))
```

The Fashion-MNIST images are

$$
28\times28
$$

matrices.

Flatten converts each image into a vector of length

$$
784.
$$

Mathematically,

this operation **does not change the data**.

It merely reshapes it.

---

## Dense Layer

```python
layers.Dense(128, activation="relu")
```

A Dense layer is a fully connected layer.

Each neuron computes

$$
\boxed{
a=f(Wx+b).
}
$$

Here,

- 128 neurons are created,
- every neuron is connected to every input.

---

## Output Layer

```python
layers.Dense(10, activation="softmax")
```

Since Fashion-MNIST contains

10 classes,

the output layer contains

10 neurons.

The Softmax activation converts the outputs into probabilities.

---

# <font color='orange'> 7. Visualising the Architecture </font>

```
Input Image

28 × 28

↓

Flatten

784 Features

↓

Dense (128)

↓

Dense (64)

↓

Dense (10)

↓

Predicted Class
```

---

# <font color='orange'> 8. Compiling the Model </font>

Before training,

the model must be compiled.

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

Compilation specifies three important components.

---

## Optimizer

```python
optimizer="adam"
```

This tells Keras to update the parameters using the Adam optimization algorithm.

Recall that Adam combines

- Momentum,
- RMSProp.

---

## Loss Function

```python
loss="sparse_categorical_crossentropy"
```

Since this is a multi-class classification problem,

we use cross-entropy loss.

The adjective **sparse** indicates that the labels are stored as integers

rather than one-hot vectors.

---

## Metrics

```python
metrics=["accuracy"]
```

During training,

Keras reports the classification accuracy after every epoch.

---

# <font color='orange'> 9. Training the Model </font>

Training is performed using

```python
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    validation_split=0.2
)
```

Each epoch means

one complete pass through the training dataset.

During every epoch,

Keras performs

```
Forward Propagation

↓

Compute Loss

↓

Backpropagation

↓

Gradient Descent

↓

Repeat
```

exactly as derived in previous chapters.

---

# <font color='orange'> 10. Evaluating the Model </font>

After training,

evaluate the model using

```python
test_loss, test_accuracy = model.evaluate(
    X_test,
    y_test
)
```

Typical output

```text
Test Accuracy: 0.89
```

The model is now evaluated on data it has never seen before.

---

# <font color='orange'> 11. Making Predictions </font>

Predictions are obtained using

```python
predictions = model.predict(X_test)
```

Each prediction consists of

10 probabilities.

For example,

```text
[0.01,
 0.00,
 0.02,
 ...
 0.92,
 ...
]
```

The predicted class is

```python
import numpy as np

predicted_class = np.argmax(predictions[0])
```

The class with the highest probability is selected.

# <font color='orange'> 12. Viewing the Model Summary </font>

Keras provides

```python
model.summary()
```

Typical output

```
Layer

↓

Output Shape

↓

Number of Parameters
```

This command is extremely useful because it shows

- every layer,
- every output shape,
- every trainable parameter.

# <font color='orange'> 13. Relationship to the Mathematics </font>

Every line of Keras code corresponds directly to an equation studied earlier.

| Keras Code | Mathematical Interpretation |
|:---|:---|
| Dense Layer | $a=f(Wx+b)$ |
| activation="relu" | Applies the ReLU function |
| loss | Computes $L(y,\hat y)$ |
| optimizer | Performs Gradient Descent (Adam) |
| fit() | Repeats Forward Propagation → Backpropagation |
| predict() | Performs Forward Propagation only |

Notice that Keras automates the implementation but does not change the underlying mathematics.

---

# <font color='orange'> 14. Common Beginner Mistakes </font>

### Forgetting to Normalize the Data

Large input values may slow training or produce unstable optimization.

---

### Choosing the Wrong Loss Function

Use

- Mean Squared Error for regression,
- Binary Cross-Entropy for binary classification,
- Sparse Categorical Cross-Entropy for integer-labelled multi-class classification.

---

### Too Few Epochs

The model may stop before learning meaningful patterns.

---

### Too Many Epochs

The model may overfit the training data.

Later chapters will introduce **Early Stopping** to address this issue.

# <font color='red'> 15. Mathematical Foundations </font>

Consider one Dense layer containing

$m$

neurons.

Given the input vector

$$
\mathbf{x},
$$

the layer computes

$$
\boxed{
\mathbf{z}
=
W\mathbf{x}
+
\mathbf{b}.
}
$$

Applying the activation function,

$$
\boxed{
\mathbf{a}
=
f(\mathbf{z}).
}
$$

During training,

Keras repeatedly computes

$$
\boxed{
\mathbf{a}
=
f(W\mathbf{x}+\mathbf{b}),
}
$$

followed by

- loss evaluation,
- backpropagation,
- parameter updates.

Thus,

every training iteration follows exactly the mathematical framework developed in earlier chapters.

# <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Sequential API | Simplest way to build feedforward neural networks |
| Flatten Layer | Converts multidimensional input into a vector |
| Dense Layer | Fully connected layer implementing $a=f(Wx+b)$ |
| Compile | Specifies optimizer, loss function, and metrics |
| Fit | Trains the model using forward and backward propagation |
| Evaluate | Measures performance on unseen data |
| Predict | Produces predictions using forward propagation only |

> **Key Insight:** Building a neural network in Keras is a direct implementation of the mathematics developed throughout this course. Each Dense layer performs a linear transformation followed by a nonlinear activation, the loss function measures prediction error, the optimizer updates the parameters, and the training loop repeatedly performs forward propagation, backpropagation, and gradient descent until the model converges.
