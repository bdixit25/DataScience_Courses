# <font color='blue'> Chapter 34: The Transformer Architecture </font>

The Transformer is one of the most influential neural network architectures ever developed.

Introduced in the landmark 2017 paper **Attention Is All You Need**, the Transformer replaced recurrent computations with attention mechanisms, allowing models to process entire sequences simultaneously.

Unlike Recurrent Neural Networks (RNNs), which process one token at a time, Transformers process all tokens in parallel while allowing every token to attend to every other token in the sequence.

This parallel processing dramatically improves training efficiency and enables models to capture long-range dependencies much more effectively than traditional recurrent architectures.

Today, nearly every state-of-the-art Large Language Model (LLM) is based on the Transformer architecture.

---



# <font color='orange'> 1. Motivation </font>

Recall the limitations of RNNs.

```
Word 1

↓

Word 2

↓

Word 3

↓

Word 4
```

Each token must wait until the previous token has been processed.

Consequences

- slow training,
- difficult parallelization,
- vanishing gradients,
- difficulty modelling long sequences.

The Transformer removes recurrence completely.

Instead,

every token is processed simultaneously.

---



# <font color='orange'> 2. The Big Idea </font>

Instead of remembering previous words,

each token directly attends to every other token.

```
Sentence

↓

Self-Attention

↓

Updated Representations

↓

Feed-Forward Network
```

Information no longer needs to travel through many recurrent steps.

Instead,

it can be accessed directly.

---



# <font color='orange'> 3. Overall Transformer Architecture </font>

The original Transformer consists of

```
Input Tokens

↓

Input Embeddings

↓

Positional Encoding

↓

Encoder Stack

↓

Encoder Output

↓

Decoder Stack

↓

Linear Layer

↓

Softmax

↓

Output Tokens
```

The encoder processes the input sequence,

while the decoder generates the output sequence.

---



# <font color='orange'> 4. Input Embeddings </font>

Words cannot be processed directly.

Each token is converted into a dense vector,

called an **embedding**.

Suppose

```
Cat
```

becomes

$$
\mathbf{x}
=
[0.13,\,
-0.71,\,
0.42,\,
\dots].
$$

These embeddings are learned during training and capture semantic relationships between words.

---



# <font color='orange'> 5. Why Positional Encoding is Necessary </font>

Unlike RNNs,

Transformers process all tokens simultaneously.

Consequently,

they have no inherent notion of word order.

For example,

the sentences

```
Dog bites man
```

and

```
Man bites dog
```

contain the same words,

but have different meanings.

Therefore,

positional information must be added to the embeddings.

---



# <font color='orange'> 6. Positional Encoding </font>

The original Transformer uses sinusoidal positional encodings.

For position

$$
p
$$

and embedding dimension

$$
i,
$$

the encoding is

$$
\boxed{
PE(p,2i)
=
\sin
\left(
\frac{p}
{10000^{2i/d}}
\right),
}
$$

$$
\boxed{
PE(p,2i+1)
=
\cos
\left(
\frac{p}
{10000^{2i/d}}
\right).
}
$$

The positional encoding is added to the word embedding,

allowing the model to incorporate sequence order.

---



# <font color='orange'> 7. Encoder Block </font>

Each encoder layer consists of

```
Input

↓

Multi-Head Self-Attention

↓

Add & Layer Normalization

↓

Feed-Forward Network

↓

Add & Layer Normalization

↓

Output
```

Multiple encoder layers are stacked to build a deeper model.

---



# <font color='orange'> 8. Multi-Head Self-Attention </font>

The first component is

Multi-Head Attention,

which we derived previously.

Every token attends to every other token,

producing contextualized representations.

The attention computation is

$$
\boxed{
\mathrm{Attention}(Q,K,V)
=
\mathrm{Softmax}
\left(
\frac{QK^T}
{\sqrt{d_k}}
\right)V.
}
$$

---



# <font color='orange'> 9. Residual Connections </font>

Instead of replacing the input,

the Transformer adds it back to the output.

```
Input

↓

Attention

↓

+

↓

Output
```

Mathematically,

$$
\boxed{
\mathbf{y}
=
F(\mathbf{x})
+
\mathbf{x}.
}
$$

Residual connections improve gradient flow and make very deep Transformers easier to train.

---



# <font color='orange'> 10. Layer Normalization </font>

After each residual connection,

Layer Normalization is applied.

Unlike Batch Normalization,

Layer Normalization normalizes across the features of a single token rather than across a mini-batch.

This stabilizes training and improves convergence.

---



# <font color='orange'> 11. Feed-Forward Network </font>

Every encoder layer also contains a small fully connected neural network.

Typically,

$$
\boxed{
FFN(x)
=
W_2
\,
ReLU
(
W_1x+b_1
)
+b_2.
}
$$

This network operates independently on every token.

It introduces additional nonlinear transformations after attention.

---



# <font color='orange'> 12. Decoder Block </font>

The decoder resembles the encoder,

but contains one additional attention layer.

```
Masked Self-Attention

↓

Encoder–Decoder Attention

↓

Feed-Forward Network
```

The second attention layer allows the decoder to attend to the encoder outputs.

---



# <font color='orange'> 13. Masked Self-Attention </font>

When generating text,

future words must remain unknown.

Therefore,

the decoder masks future positions.

```
Word 1

↓

Can See

Word 1

Word 2

↓

Can See

Word 1

Word 2

Word 3

↓

Can See

Word 1

Word 2

Word 3
```

Future words are hidden during training.

This ensures autoregressive generation.

---



# <font color='orange'> 14. Encoder–Decoder Attention </font>

The decoder attends

not only to previous output tokens,

but also to the encoder outputs.

This allows the generated sequence to remain consistent with the input sequence.

This mechanism is essential for tasks such as

- machine translation,
- summarization,
- question answering.

---



# <font color='orange'> 15. Stacking Transformer Layers </font>

A practical Transformer contains many encoder and decoder blocks.

For example,

```
Encoder

↓

Encoder

↓

Encoder

↓

...

↓

Decoder

↓

Decoder

↓

Decoder
```

Each successive layer builds more abstract contextual representations.

---



# <font color='orange'> 16. Why Transformers Outperform RNNs </font>

Compared with RNNs,

Transformers

- process tokens in parallel,
- capture long-range dependencies directly,
- avoid recurrent computations,
- train more efficiently on GPUs,
- scale effectively to billions of parameters.

These advantages have made Transformers the dominant architecture in modern AI.

---

# <font color='red'> 17. Mathematical Foundations </font>

For one encoder block,

the computation is

Input embeddings

$$
X.
$$

Add positional encoding

$$
X+PE.
$$

Multi-Head Attention

$$
\boxed{
A
=
\mathrm{MultiHead}(X,X,X).
}
$$

Residual connection

$$
\boxed{
Z
=
A
+
X.
}
$$

Layer Normalization

$$
\boxed{
H
=
\mathrm{LayerNorm}(Z).
}
$$

Feed-Forward Network

$$
\boxed{
F
=
FFN(H).
}
$$

Second residual connection

$$
\boxed{
Y
=
\mathrm{LayerNorm}
(
F+H
).
}
$$

This output becomes the input to the next encoder layer.

---



# <font color='orange'> 18. Computational Complexity </font>

For a sequence of length

$$
n,
$$

self-attention requires

$$
\boxed{
\mathcal O(n^2)
}
$$

operations,

because every token attends to every other token.

This quadratic complexity is one of the primary limitations of standard Transformers and has motivated the development of more efficient attention mechanisms.

---



# <font color='orange'> 19. Common Misconceptions </font>

### Misconception 1

> Transformers have no memory.

**False.**

Transformers do not use recurrent memory like RNNs, but they build contextual representations through self-attention, allowing every token to access information from the entire sequence.

---

### Misconception 2

> Positional encoding is optional.

**False.**

Without positional information, the Transformer cannot distinguish between different token orders, since self-attention alone is permutation invariant.

---

### Misconception 3

> The encoder and decoder are always used together.

**False.**

Many modern models use only part of the original Transformer architecture. For example, BERT uses only the encoder stack, while GPT uses only the decoder stack.

---

# <font color='purple'> 20. Conceptual Summary </font>

| Component | Purpose |
|:---|:---|
| Input Embedding | Converts tokens into dense vectors |
| Positional Encoding | Encodes the order of tokens |
| Multi-Head Self-Attention | Learns contextual relationships between tokens |
| Residual Connection | Improves gradient flow and enables deep networks |
| Layer Normalization | Stabilizes training |
| Feed-Forward Network | Applies nonlinear transformations to each token |
| Encoder | Builds contextual representations of the input |
| Decoder | Generates output tokens autoregressively |
| Masked Attention | Prevents access to future tokens during generation |

> **Key Insight:** The Transformer replaces recurrence with self-attention, allowing every token in a sequence to interact directly with every other token. By combining positional encodings, multi-head attention, residual connections, layer normalization, and feed-forward networks, the Transformer efficiently learns rich contextual representations while enabling highly parallel computation. This architecture forms the foundation of nearly all modern large language models.
