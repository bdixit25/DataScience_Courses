# <font color='blue'> Chapter 49: SARSA (State–Action–Reward–State–Action) </font>

Temporal Difference Learning introduced the idea of learning state values from experience.

However,

many Reinforcement Learning problems require the agent to decide **which action** should be taken in each state.

Rather than estimating the value of a state alone,

we estimate the value of taking a particular action in that state.

This quantity is called the **Action-Value Function**.

SARSA is one of the earliest and most influential algorithms for learning action-value functions directly from interaction with the environment.

It is an **on-policy** algorithm,

meaning it learns the value of the same policy that it follows during exploration.

---



# <font color='orange'> 1. Motivation </font>

Suppose a robot is standing at an intersection.

```
Current State

↓

Go Left

↓

?

Go Right

↓

?
```

Knowing only the value of the intersection

is not enough.

Instead,

we want to know

```
How good is each possible action?
```

This allows the agent to compare actions directly and choose the one expected to produce the greatest long-term reward.

---



# <font color='orange'> 2. Action-Value Function </font>

The action-value function is written as

$$
\boxed{
Q^\pi(s,a).
}
$$

It represents

the expected discounted return obtained by

- starting in state \(s\),
- taking action \(a\),
- then following policy \(\pi\).

Mathematically,

$$
\boxed{
Q^\pi(s,a)
=
\mathbb{E}_\pi
\left[
G_t
\mid
S_t=s,
A_t=a
\right].
}
$$

Unlike the state-value function,

the action-value function evaluates individual decisions.

---



# <font color='orange'> 3. Why Learn Q(s,a)? </font>

Suppose

```
State

↓

Action A → Return = 25

Action B → Return = 90
```

The state itself has only one value,

but the actions differ significantly.

Learning

$$
Q(s,a)
$$

allows the agent to choose

the best action directly.

---



# <font color='orange'> 4. The SARSA Sequence </font>

The name

SARSA

comes from the sequence

$$
\boxed{
S
\rightarrow
A
\rightarrow
R
\rightarrow
S'
\rightarrow
A'.
}
$$

The algorithm performs the following steps.

```
Current State

↓

Choose Action

↓

Receive Reward

↓

Observe Next State

↓

Choose Next Action

↓

Update Q-value
```

Every update depends on this complete sequence.

---



# <font color='orange'> 5. The SARSA Update Rule </font>

The TD target becomes

$$
\boxed{
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1}).
}
$$

Therefore,

the update equation is

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\left[
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1})
-
Q(s_t,a_t)
\right].
}
$$

Notice that

the next action

$$
a_{t+1}
$$

is the action actually chosen by the current policy.

---



# <font color='orange'> 6. The SARSA TD Error </font>

The Temporal Difference error becomes

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1})
-
Q(s_t,a_t).
}
$$

Interpretation

- Positive TD error

↓

better than expected.

- Negative TD error

↓

worse than expected.

The TD error determines how much the Q-value is adjusted.

---



# <font color='orange'> 7. On-Policy Learning </font>

SARSA is an

**on-policy**

algorithm.

This means

the policy used for learning

is exactly the same policy

used for interacting with the environment.

```
Current Policy

↓

Generate Experience

↓

Update Current Policy
```

The agent therefore learns

the value of its own behaviour.

---



# <font color='orange'> 8. ε-Greedy Exploration </font>

To avoid always choosing the same action,

SARSA often uses an

**ε-greedy policy**.

With probability

$$
1-\varepsilon,
$$

the agent selects

the action with the highest Q-value.

With probability

$$
\varepsilon,
$$

the agent chooses

a random action.

This balances

- exploitation,
- exploration.

Typical values include

$$
\varepsilon
=
0.1
\quad\text{or}\quad
0.01.
$$

---



# <font color='orange'> 9. Complete SARSA Algorithm </font>

```
Initialize Q(s,a)

↓

Observe Initial State

↓

Choose Action (ε-greedy)

↓

Take Action

↓

Observe Reward

↓

Observe Next State

↓

Choose Next Action

↓

Update Q

↓

Repeat
```

Learning continues until the policy converges or the desired performance is achieved.

---



# <font color='orange'> 10. Why Does SARSA Learn Safely? </font>

Because SARSA updates using

the action actually taken,

its value estimates naturally reflect

the consequences of exploration.

For example,

suppose

a risky shortcut exists near a cliff.

Even if the optimal path uses the shortcut,

an ε-greedy policy may occasionally explore,

causing the agent to fall.

SARSA learns from these exploratory mistakes,

often preferring a safer path.

This makes SARSA relatively conservative.

---



# <font color='orange'> 11. Applications </font>

SARSA has been applied to

- robot navigation,
- game playing,
- autonomous control,
- resource management,
- inventory optimization,
- educational tutoring systems.

Although newer algorithms are often preferred,

SARSA remains an important conceptual foundation.

---

# <font color='red'> 12. Mathematical Foundations </font>

The action-value function is

$$
\boxed{
Q^\pi(s,a)
=
\mathbb{E}_\pi
[G_t
\mid
S_t=s,
A_t=a].
}
$$

The SARSA TD target is

$$
\boxed{
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1}).
}
$$

The TD error becomes

$$
\boxed{
\delta_t
=
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1})
-
Q(s_t,a_t).
}
$$

The SARSA update rule is

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\delta_t.
}
$$

Substituting the TD error,

$$
\boxed{
Q(s_t,a_t)
\leftarrow
Q(s_t,a_t)
+
\alpha
\left[
r_{t+1}
+
\gamma
Q(s_{t+1},a_{t+1})
-
Q(s_t,a_t)
\right].
}
$$

Unlike state-value methods,

SARSA directly learns the quality of state-action pairs,

allowing the agent to derive a policy by selecting the action with the highest learned value.

---



# <font color='orange'> 13. Convergence </font>

Under standard assumptions,

including

- sufficient exploration of all state-action pairs,
- appropriately decreasing learning rates,
- a stationary environment,

SARSA converges to the optimal action-value function for the policy being followed. If the exploration strategy becomes increasingly greedy while still ensuring adequate exploration, SARSA can converge to an optimal policy in finite MDPs.

---



# <font color='orange'> 14. Common Misconceptions </font>

### Misconception 1

> SARSA always follows the optimal policy.

**False.**

During learning,

SARSA usually follows an exploratory policy,

such as ε-greedy.

It learns the value of this behaviour,

not an idealized greedy policy.

---

### Misconception 2

> SARSA ignores exploratory actions.

**False.**

Exploratory actions directly influence the update,

because the algorithm uses the action actually selected at the next state.

---

### Misconception 3

> SARSA and Q-Learning use identical update rules.

**False.**

SARSA uses the value of the **next action actually taken**,

whereas Q-Learning uses the **maximum possible future Q-value**,

regardless of which action is taken.

This distinction makes SARSA on-policy and Q-Learning off-policy.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Action-Value Function \(Q(s,a)\) | Expected return for taking action \(a\) in state \(s\) |
| SARSA | On-policy TD control algorithm |
| SARSA Sequence | State → Action → Reward → Next State → Next Action |
| TD Error | Difference between the TD target and the current Q-value |
| ε-Greedy Policy | Balances exploration and exploitation |
| On-Policy Learning | Learns the value of the same policy used for interaction |
| Safe Learning | Learns values that reflect the consequences of exploration |

> **Key Insight:** SARSA extends Temporal Difference Learning from state values to action values, allowing an agent to learn which actions are most beneficial in each state. By updating its estimates using the action actually selected by its current policy, SARSA learns the consequences of its own exploratory behaviour. This on-policy approach often leads to more conservative and stable learning, making SARSA a foundational algorithm in Reinforcement Learning.
