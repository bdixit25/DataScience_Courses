# <font color='blue'> Introduction to Supervised Machine Learning </font>

Machine Learning (ML) is one of the most rapidly growing fields of Artificial Intelligence (AI). Rather than being explicitly programmed to solve every possible problem, machine learning algorithms learn patterns and relationships directly from data. Once these patterns have been learned, the algorithms can make predictions or decisions on new, previously unseen data.

Supervised Machine Learning is the most widely used branch of machine learning. In supervised learning, an algorithm is trained using a dataset that contains both input variables (features) and the corresponding correct outputs (targets). The objective is to learn the relationship between the inputs and outputs so that the model can accurately predict the target variable for new observations.

This chapter introduces the fundamental concepts of supervised machine learning, explains the different categories of machine learning, discusses the distinction between regression and classification, and provides the foundation for the regression models studied throughout this notebook.

# <font color='orange'> 2. What is Machine Learning? </font>

Machine Learning is a branch of Artificial Intelligence in which computers learn patterns directly from data instead of following explicitly programmed rules.

Arthur Samuel (1959), one of the pioneers of machine learning, defined it as

> "The field of study that gives computers the ability to learn without being explicitly programmed."

Later, Tom Mitchell (1997) provided a more formal definition:

> *A computer program is said to learn from experience **E** with respect to some task **T** and performance measure **P** if its performance at task **T**, as measured by **P**, improves with experience **E**.*

This definition highlights three essential components of every machine learning problem:

* **Task (T):** What the algorithm is trying to accomplish.
* **Experience (E):** The data from which the algorithm learns.
* **Performance Measure (P):** A metric used to evaluate how well the algorithm performs.

For example,

| Component | Example |
| :--- | :--- |
| Task | Predict house prices |
| Experience | Historical housing data |
| Performance Measure | Root Mean Squared Error (RMSE) |

# <font color='orange'> 3. Why Do We Need Machine Learning? </font>

Traditional programming works well when a problem can be described using a fixed set of rules.

```
Input Data

↓

Program (Rules)

↓

Output
```

However, many real-world problems are too complex to describe using explicit rules.

Examples include

* Identifying fraudulent transactions.
* Recognizing handwritten digits.
* Predicting stock prices.
* Estimating the mass of a particle.
* Translating languages.

For such problems,

it is often easier to allow the computer to discover the rules directly from data.

Machine learning reverses the traditional programming paradigm.

```
Input Data + Correct Outputs

↓

Machine Learning Algorithm

↓

Learned Model
```

The learned model can then be used to make predictions on new data.

# <font color='orange'> 4. Types of Machine Learning </font>

Machine learning algorithms are generally divided into four major categories.

---

## 4.1 Supervised Learning

The training dataset contains both

* input variables (features),
* correct output values (labels).

The algorithm learns the relationship between them.

Typical applications include

* House price prediction
* Disease diagnosis
* Credit scoring
* Fraud detection
* Energy demand forecasting

Supervised learning is the focus of this notebook.

---

## 4.2 Unsupervised Learning

Only input variables are available.

There are no target labels.

The objective is to discover hidden structure within the data.

Common tasks include

* Clustering
* Dimensionality reduction
* Anomaly detection
* Association rule mining

Examples

* Customer segmentation
* Topic modelling
* Market basket analysis

---

## 4.3 Semi-Supervised Learning

Only a small portion of the data is labelled.

The remaining observations are unlabelled.

The algorithm combines both sources of information during training.

Applications include

* Medical imaging
* Speech recognition
* Document classification

where obtaining labels is expensive.

---

## 4.4 Reinforcement Learning

An intelligent agent interacts with an environment.

Rather than receiving labelled examples,

the agent receives rewards or penalties.

The objective is to learn a policy that maximizes the cumulative reward.

Applications include

* Robotics
* Autonomous driving
* Game playing
* Resource allocation

# <font color='orange'> 5. What is Supervised Learning? </font>

Supervised learning learns a mapping

from

input variables

to

output variables.

Mathematically,

the objective is to learn an unknown function

$$
f : X \rightarrow Y,
$$

such that

$$
Y = f(X).
$$

During training,

the algorithm receives examples

$$
(\mathbf{x}_1,y_1),
(\mathbf{x}_2,y_2),
\ldots,
(\mathbf{x}_n,y_n),
$$

where

* $\mathbf{x}_i$ represents the feature vector,
* $y_i$ is the corresponding target value.

The model learns from these examples and then predicts the target for new observations.

# <font color='blue'> Introduction to Regression </font>

Regression is one of the oldest, most widely studied, and most important techniques in statistics, data science, and machine learning. The primary objective of regression is to model the relationship between one or more input variables (features) and a continuous numerical target variable. Once this relationship has been learned from historical data, the model can be used to predict the target variable for new observations.

Unlike classification, where the objective is to assign observations to discrete categories, regression predicts numerical quantities that can take any value within a continuous range. Examples include predicting house prices, stock prices, energy consumption, rainfall, patient survival time, or the energy of particles produced in high-energy physics experiments.

Throughout this notebook, regression will serve as the foundation for understanding many of the mathematical principles that underpin modern machine learning algorithms.

---



# <font color='orange'> 1. What is Regression? </font>

Regression is a supervised learning technique used to predict a **continuous numerical value** from one or more input variables.

Given a set of input features

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_p),
$$

our goal is to learn a function

$$
f(\mathbf{x})
$$

such that

$$
\boxed{
y \approx f(\mathbf{x}),
}
$$

where

* $y$ is the target variable,
* $\mathbf{x}$ represents the predictor variables,
* $f(\mathbf{x})$ is the unknown relationship between them.

Once this relationship has been learned,

the model can predict

the target value

for new observations.

---



# <font color='orange'> 2. Why is it Called "Regression"? </font>

The term **regression** dates back to the work of the British statistician **Sir Francis Galton** in the late nineteenth century.

Galton studied the relationship between the heights of parents and their children. He observed that extremely tall parents tended to have children who were still tall but, on average, slightly closer to the population mean. Likewise, very short parents tended to have children who were slightly taller than themselves.

He referred to this phenomenon as

> **"Regression toward the mean."**

Although the modern use of regression extends far beyond this original application, the name has remained.

Today, regression refers more generally to modelling relationships between variables and making predictions.

# <font color='orange'> 7. Parametric and Non-Parametric Regression </font>

Regression algorithms may be broadly divided into two categories.

---

## Parametric Models

Parametric models assume a specific mathematical form for the relationship between the features and the target.

Examples include

* Linear Regression
* Ridge Regression
* LASSO
* Elastic Net

Advantages

* Fast training
* Easy interpretation
* Simple mathematical formulation

Limitations

* Less flexible
* May underfit complex relationships

---

## Non-Parametric Models

Non-parametric models do not assume a fixed mathematical form.

Examples include

* Decision Trees
* Random Forests
* Gradient Boosting
* Support Vector Regression
* KNN Regression

Advantages

* Flexible
* Capture nonlinear relationships

Limitations

* More computationally expensive
* Often less interpretable
