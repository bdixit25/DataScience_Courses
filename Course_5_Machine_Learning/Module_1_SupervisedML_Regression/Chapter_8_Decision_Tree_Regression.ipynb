# <font color='blue'> Decision Tree Regression </font>

Linear regression models assume that the relationship between the predictor variables and the target can be represented by a mathematical equation. Although these models are often effective, many real-world relationships are considerably more complex and cannot be adequately described by a single global function.

Decision Tree Regression provides a fundamentally different approach. Rather than fitting one equation to the entire dataset, a decision tree recursively partitions the feature space into smaller and more homogeneous regions. Within each region, the model predicts a constant value, typically the average target value of the training observations that fall into that region.

Because decision trees make very few assumptions about the underlying relationship between the predictors and the response, they can naturally model nonlinear relationships, variable interactions, and heterogeneous datasets. They are therefore among the most flexible and intuitive machine learning algorithms.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict house prices using

* Floor Area
* Number of Bedrooms
* Distance from the City Centre

Linear Regression attempts to fit

one equation

to the entire dataset

$$
y
=
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+
\beta_3x_3.
$$

However,

real housing markets rarely behave so simply.

For example,

houses may follow completely different pricing patterns depending on whether they are

* urban,
* suburban,
* rural.

A single equation may struggle to capture these distinct behaviours.

Decision Trees address this by dividing the data into smaller groups,

allowing different regions of the feature space to behave differently.

---

# <font color='orange'> 2. What is a Decision Tree? </font>

A Decision Tree is a hierarchical predictive model consisting of

* decision nodes,
* branches,
* leaf nodes.

Each internal node asks a question about one feature.

Example

```
Is Area < 120 m² ?

            Yes
           /
Root -----
           \
            No
```

The answer determines which branch is followed.

Eventually,

the observation reaches

a leaf node,

which contains the prediction.

---

# <font color='orange'> 3. Decision Tree Regression vs Classification </font>

Decision Trees can be used for both

* classification,
* regression.

The difference lies only in the prediction produced at each leaf.

---

### Classification Tree

Leaf node predicts

```
Spam

Healthy

Signal
```

---

### Regression Tree

Leaf node predicts

```
£315,000

17.8°C

42.6 GeV
```

Thus,

the tree-building procedure is very similar,

but the prediction target differs.

---

# <font color='orange'> 4. How Does a Regression Tree Work? </font>

Suppose our first split is

```
Area < 120 m² ?
```

The tree becomes

```
                 Area < 120 ?

               /             \

           Yes               No

      Predict 240k      Predict 420k
```

Instead of fitting

one equation,

the algorithm has created

two different prediction regions.

Additional splits continue dividing the data into increasingly homogeneous groups.

---

# <font color='orange'> 5. Recursive Partitioning </font>

The process of repeatedly dividing the feature space is known as

**recursive binary partitioning**.

```
Entire Dataset

↓

Split 1

↓

Two Regions

↓

Split 2

↓

Four Regions

↓

Split 3

↓

Eight Regions

↓

...
```

Each split attempts to reduce the variation of the target variable within the resulting regions.

---

# <font color='orange'> 6. Choosing the Best Split </font>

At every node,

the algorithm considers

many possible

* features,
* split values.

Example

```
Area < 100?

Area < 120?

Area < 150?

Bedrooms < 3?

Bedrooms < 4?

Distance < 5 km?

...
```

For each candidate split,

the algorithm measures

how much it improves the homogeneity of the target values.

The split that produces the greatest improvement is selected.

---

# <font color='orange'> 7. Measuring Node Impurity </font>

Unlike classification trees,

which use measures such as Gini Impurity or Entropy,

regression trees aim to reduce the variability of the target values.

The most common criteria are

* Mean Squared Error (MSE)
* Variance Reduction
* Mean Absolute Error (less common)

For a node containing observations

$$
y_1,y_2,\ldots,y_n,
$$

the node impurity using MSE is

$$
\boxed{
MSE
=
\frac1n
\sum_{i=1}^{n}
(y_i-\bar y)^2,
}
$$

where

$$
\bar y
$$

is the mean target value within the node.

The algorithm chooses the split that produces the largest reduction in impurity.

---

# <font color='orange'> 8. Leaf Node Predictions </font>

After the tree has been constructed,

each leaf contains

a subset of the training observations.

The prediction is usually

the average target value

within that leaf.

Example

```
Leaf

House Prices

240

235

245

250
```

Prediction

$$
=
242.5
$$

Every future observation reaching this leaf receives the same prediction.

---

# <font color='orange'> 9. Stopping Criteria </font>

If unrestricted,

a Decision Tree can continue splitting until every training observation occupies its own leaf.

This almost always leads to overfitting.

To prevent this,

the tree-building process is stopped according to predefined criteria,

such as

* maximum depth,
* minimum number of samples required to split,
* minimum number of samples per leaf,
* minimum impurity decrease.

These hyperparameters control the complexity of the tree.

---

# <font color='orange'> 10. Advantages </font>

Decision Tree Regression offers several advantages.

* Naturally models nonlinear relationships.
* Captures interactions between variables automatically.
* Requires little data preprocessing.
* Does not require feature scaling.
* Handles numerical and categorical variables.
* Easy to visualize.
* Easy to explain to non-technical audiences.

---

# <font color='orange'> 11. Limitations </font>

Decision Trees also have important limitations.

* Highly prone to overfitting.
* High variance.
* Small changes in the data may produce very different trees.
* Predictions are piecewise constant rather than smooth.
* Often less accurate than ensemble methods.

These limitations motivate Random Forests and Gradient Boosting.

---

# <font color='orange'> 12. Hyperparameters </font>

Common Decision Tree hyperparameters include

| Hyperparameter | Purpose |
| :--- | :--- |
| `max_depth` | Maximum tree depth |
| `min_samples_split` | Minimum samples required to split a node |
| `min_samples_leaf` | Minimum samples in each leaf |
| `max_leaf_nodes` | Maximum number of terminal nodes |
| `max_features` | Number of features considered at each split |

These hyperparameters control the complexity of the tree and help reduce overfitting.

# <font color='blue'> The CART Algorithm: How Decision Trees Are Built </font>

The previous chapter introduced the fundamental idea behind Decision Tree Regression: recursively partitioning the feature space into increasingly homogeneous regions. An important question naturally follows:

> **How does the algorithm decide where to split the data?**

The answer is provided by the **Classification and Regression Trees (CART)** algorithm. Developed by Leo Breiman, Jerome Friedman, Richard Olshen, and Charles Stone in 1984, CART is one of the most influential algorithms in modern statistics and machine learning.

CART constructs decision trees by repeatedly selecting the feature and split point that produce the greatest improvement according to a predefined criterion. For regression, this criterion is typically based on reducing the variability of the target values within each region.

---

# <font color='orange'> 1. The Goal of CART </font>

Suppose we have collected housing data.

```
Area

Bedrooms

Age

↓

House Price
```

Rather than fitting

one mathematical equation,

CART repeatedly asks

questions such as

```
Area < 120 m²?
```

or

```
Bedrooms < 3?
```

Each question divides the data into two smaller groups.

The objective is to create groups in which

the target values

are as similar as possible.

---

# <font color='orange'> 2. Binary Recursive Partitioning </font>

CART always performs

**binary splits**.

Each node is divided into exactly

two

child nodes.

Example

```
Entire Dataset

↓

Area < 120?

↓

Yes         No

↓

Two Child Nodes
```

Each child node may then be split again.

```
Root

↓

Split

↓

Two Nodes

↓

Split Again

↓

Four Nodes

↓

...
```

This repeated process is called

**recursive binary partitioning**.

---

# <font color='orange'> 3. Candidate Splits </font>

Suppose

Area

contains

```
80

100

120

150

180
```

Possible split points include

```
90

110

135

165
```

Similarly,

every predictor contributes its own possible split values.

The algorithm evaluates

all candidate splits

and selects

the one that produces the greatest improvement.

---

# <font color='orange'> 4. Node Impurity </font>

To decide whether a split is useful,

the algorithm measures

how similar

the target values are within a node.

For regression,

the most common impurity measure is

the **Mean Squared Error (MSE)**.

Suppose a node contains

$$
y_1,y_2,\ldots,y_n.
$$

Its impurity is

$$
\boxed{
MSE
=
\frac1n
\sum_{i=1}^{n}
(y_i-\bar y)^2,
}
$$

where

$$
\bar y
$$

is the mean target value.

A node with

very similar target values

has

low impurity.

---

# <font color='orange'> 5. Variance Reduction </font>

Suppose a node is split into

Left

and

Right

child nodes.

The quality of the split is measured by

how much

the impurity decreases.

Mathematically,

$$
\boxed{
\Delta I
=
I_{\text{parent}}
-
\left(
\frac{n_L}{n}I_L
+
\frac{n_R}{n}I_R
\right),
}
$$

where

* \(I_{\text{parent}}\) is the impurity of the parent node,
* \(I_L\) and \(I_R\) are the impurities of the left and right child nodes,
* \(n_L\) and \(n_R\) are the numbers of observations in each child.

The split with the largest

$$
\Delta I
$$

is selected.

---

# <font color='orange'> 6. Greedy Optimization </font>

An important property of CART is that it is

**greedy**.

At each node,

the algorithm chooses

the best immediate split,

without considering whether another sequence of splits might produce a better overall tree.

```
Current Node

↓

Best Immediate Split

↓

Repeat
```

The algorithm never revisits earlier decisions.

Although this strategy does not guarantee the globally optimal tree,

it is computationally efficient and performs remarkably well in practice.

---

# <font color='orange'> 7. Growing the Tree </font>

The algorithm continues splitting until one of the stopping criteria is met.

```
Root

↓

Split

↓

Split Again

↓

Split Again

↓

Leaf Nodes
```

Without restrictions,

the tree would continue growing until

each leaf contains

only one observation.

This almost always results in severe overfitting.

---

# <font color='orange'> 8. Stopping Criteria (Pre-Pruning) </font>

To control model complexity,

tree growth can be stopped early.

Common stopping criteria include

| Hyperparameter | Purpose |
| :--- | :--- |
| `max_depth` | Maximum tree depth |
| `min_samples_split` | Minimum observations required to split a node |
| `min_samples_leaf` | Minimum observations required in a leaf |
| `max_leaf_nodes` | Maximum number of leaves |
| `min_impurity_decrease` | Minimum reduction in impurity required to split |

These techniques are known collectively as

**pre-pruning**,

because they prevent unnecessary growth during tree construction.

---

# <font color='orange'> 9. Post-Pruning </font>

An alternative strategy is

to first grow

a very large tree,

then

remove unnecessary branches afterwards.

This process is called

**post-pruning**.

```
Large Tree

↓

Remove Weak Branches

↓

Simpler Tree
```

One common method is

**Cost-Complexity Pruning**,

which balances

prediction accuracy

against

tree size.

---

# <font color='orange'> 10. Prediction </font>

Once the tree has been built,

prediction is straightforward.

Suppose a new observation has

```
Area = 145

Bedrooms = 3
```

The observation starts

at the root

and follows the decision rules.

```
Area < 120?

↓

No

↓

Bedrooms < 4?

↓

Yes

↓

Leaf
```

The prediction is simply

the average target value

stored in that leaf.

No further calculations are required.

---

# <font color='orange'> 11. Computational Complexity </font>

Constructing a decision tree involves evaluating many candidate splits.

For a dataset with

* \(n\) observations,
* \(p\) predictors,

training is approximately

$$
O(p\,n\log n)
$$

for balanced trees using efficient implementations.

Prediction is much faster,

requiring only one path from the root to a leaf.

For a balanced tree,

prediction typically scales with the tree depth rather than the total number of observations.

---

# <font color='orange'> 12. Advantages of CART </font>

CART offers several advantages.

* Handles nonlinear relationships naturally.
* Captures interactions automatically.
* No feature scaling required.
* Works with numerical and categorical predictors.
* Easy to visualize.
* Fast prediction.
* Forms the basis of many ensemble methods.

---

# <font color='orange'> 13. Limitations of CART </font>

The CART algorithm also has several limitations.

* Greedy optimization may not find the globally optimal tree.
* High variance.
* Sensitive to small changes in the data.
* Can overfit if allowed to grow without restriction.
* Piecewise-constant predictions may appear discontinuous.

These limitations motivate ensemble methods such as Random Forests and Gradient Boosting.

---

# <font color='orange'> 14. CART in Modern Machine Learning </font>

Many modern algorithms are built directly upon CART.

```
CART

↓

Decision Trees

↓

Random Forest

↓

Extra Trees

↓

Gradient Boosting

↓

XGBoost

↓

LightGBM

↓

CatBoost
```

Although these algorithms differ in how multiple trees are combined,

each individual tree is typically constructed using a CART-like procedure.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| CART | Classification and Regression Trees algorithm |
| Binary Splitting | Every split creates exactly two child nodes |
| Recursive Partitioning | Repeatedly divides the feature space into more homogeneous regions |
| Split Criterion | Maximizes the reduction in node impurity |
| Greedy Optimization | Chooses the best immediate split at each node |
| Pre-Pruning | Stops tree growth early using hyperparameters |

> **Key Insight:** The CART algorithm constructs decision trees by recursively applying binary splits that maximize the reduction in node impurity. Through greedy optimization, the feature space is partitioned into increasingly homogeneous regions, producing a hierarchy of decision rules that can model complex nonlinear relationships. Although a single CART tree is prone to overfitting and high variance, its simplicity, flexibility, and computational efficiency make it the fundamental building block of many of the most powerful ensemble learning algorithms in modern machine learning.

# <font color='blue'> Overfitting, Pruning, and Hyperparameter Tuning in Decision Trees </font>

Decision Trees are among the most flexible machine learning algorithms. By recursively partitioning the feature space, they can model highly complex nonlinear relationships with relatively few assumptions about the underlying data. However, this flexibility comes at a cost. If allowed to grow without restriction, a decision tree can become excessively complex, memorizing the training data rather than learning the underlying relationship. This phenomenon is known as **overfitting**.

To obtain models that generalize well to unseen data, the complexity of the tree must be carefully controlled. This chapter examines why overfitting occurs in decision trees, introduces the concepts of pruning and regularization, and discusses the most important hyperparameters used to control tree complexity.

---

# <font color='orange'> 1. Why Do Decision Trees Overfit? </font>

Recall that the CART algorithm grows a tree by repeatedly selecting the split that maximally reduces the impurity.

If no restrictions are imposed,

the algorithm continues splitting until

* every leaf contains only one observation, or
* no further improvement is possible.

For example,

```
Training Data

↓

Split

↓

Split Again

↓

Split Again

↓

...

↓

One Sample Per Leaf
```

Such a tree fits the training data almost perfectly.

However,

it has learned

not only the underlying relationship,

but also

the random noise

present in the training dataset.

---

# <font color='orange'> 2. Underfitting vs Overfitting </font>

The complexity of a decision tree is largely determined by its depth.

### Very Shallow Tree

```
Root

↓

Two Leaves
```

Characteristics

* High bias
* Low variance
* Underfitting

---

### Very Deep Tree

```
Root

↓

Many Splits

↓

Many Leaves
```

Characteristics

* Low bias
* High variance
* Overfitting

---

The objective is to find an appropriate balance between these two extremes.

---

# <font color='orange'> 3. Decision Trees and the Bias–Variance Trade-off </font>

Decision Trees provide an excellent illustration of the Bias–Variance Trade-off.

```
Increase Tree Depth

↓

Bias

↓

Decreases
```

```
Increase Tree Depth

↓

Variance

↓

Increases
```

Initially,

adding more splits improves predictive performance.

Eventually,

additional splits begin fitting noise,

causing the test error to increase.

---

# <font color='orange'> 4. Controlling Tree Complexity </font>

The complexity of a decision tree can be controlled in two general ways.

### Pre-Pruning

Restrict tree growth while the tree is being constructed.

---

### Post-Pruning

Allow the tree to grow fully,

then remove unnecessary branches afterwards.

Both approaches seek to improve generalization by reducing model complexity.

---

# <font color='orange'> 5. Pre-Pruning Hyperparameters </font>

Scikit-Learn provides several hyperparameters for limiting tree growth.

---

## Maximum Depth (`max_depth`)

Limits the number of levels in the tree.

```
Depth = 2

↓

Simple Tree
```

```
Depth = 20

↓

Complex Tree
```

Smaller values reduce overfitting but may increase underfitting.

---

## Minimum Samples Required to Split (`min_samples_split`)

Specifies the minimum number of observations required before a node can be divided.

Example

```
min_samples_split = 10
```

A node containing fewer than ten observations cannot be split further.

Larger values generally produce simpler trees.

---

## Minimum Samples per Leaf (`min_samples_leaf`)

Each terminal node must contain at least the specified number of observations.

Example

```
min_samples_leaf = 5
```

Every leaf must contain at least five training examples.

This often improves prediction stability.

---

## Maximum Number of Leaf Nodes (`max_leaf_nodes`)

Limits the total number of terminal nodes.

Fewer leaves generally reduce model complexity.

---

## Minimum Impurity Decrease (`min_impurity_decrease`)

A split is accepted only if it reduces the impurity by at least a specified amount.

Very small improvements are ignored.

---

# <font color='orange'> 6. Post-Pruning (Cost-Complexity Pruning) </font>

Instead of restricting growth,

another strategy is

```
Grow Large Tree

↓

Evaluate Branches

↓

Remove Weak Branches

↓

Simpler Tree
```

Scikit-Learn implements

**Minimal Cost-Complexity Pruning**.

The optimization objective becomes

$$
\boxed{
R_\alpha(T)
=
R(T)
+
\alpha |T|,
}
$$

where

* \(R(T)\) is the training error,
* \(|T|\) is the number of leaf nodes,
* \(\alpha\) controls the penalty for tree complexity.

As

\(\alpha\)

increases,

the tree becomes progressively smaller.

Notice the similarity to Ridge and LASSO:

a penalty term is added to discourage unnecessary complexity.

---

# <font color='orange'> 7. Visualizing Tree Complexity </font>

Conceptually,

```
Small α

↓

Large Tree

↓

Low Bias

↓

High Variance
```

```
Large α

↓

Small Tree

↓

Higher Bias

↓

Lower Variance
```

Choosing an appropriate value of

\(\alpha\)

is therefore another example of the Bias–Variance Trade-off.

---

# <font color='orange'> 8. Hyperparameter Tuning </font>

The tree hyperparameters should rarely be chosen manually.

Instead,

they should be optimized using

* K-Fold Cross-Validation
* Grid Search
* Random Search

Typical hyperparameters include

| Hyperparameter | Typical Values |
| :--- | :--- |
| `max_depth` | 3–30 |
| `min_samples_split` | 2–20 |
| `min_samples_leaf` | 1–10 |
| `max_leaf_nodes` | Problem dependent |
| `ccp_alpha` | Selected by cross-validation |

The optimal values depend on the dataset and the prediction task.

---

# <font color='orange'> 9. Common Signs of Overfitting </font>

A decision tree may be overfitting if

* training error is extremely small,
* validation error is much larger,
* the tree is very deep,
* many leaves contain only one or two observations,
* predictions change dramatically after small changes to the training data.

Cross-validation is an effective way to detect these issues.

---

# <font color='orange'> 10. Practical Implementation </font>

A constrained decision tree in Scikit-Learn might be created as follows:

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    max_depth=6,
    min_samples_split=10,
    min_samples_leaf=5,
    ccp_alpha=0.001,
    random_state=42
)

model.fit(X_train, y_train)
```

These parameters help produce a tree that is less likely to overfit than a fully grown tree.

---

# <font color='orange'> 11. Best Practices </font>

When building decision trees,

consider the following recommendations.

* Begin with a reasonably simple tree.
* Use cross-validation to tune hyperparameters.
* Monitor the difference between training and validation error.
* Avoid unnecessarily deep trees.
* Use pruning to improve generalization.
* Consider ensemble methods if a single tree remains unstable.

These practices typically lead to models with better predictive performance.

---

# <font color='orange'> 12. Why This Leads to Random Forests </font>

Even after careful tuning,

a single decision tree remains a **high-variance estimator**.

Two trees trained on slightly different datasets may have very different structures.

This observation motivates the next major algorithm.

Instead of relying on

one carefully tuned tree,

Random Forest Regression trains

**many different decision trees**

and averages their predictions.

```
Tree 1

↓

Prediction
```

```
Tree 2

↓

Prediction
```

```
Tree 3

↓

Prediction
```

```
...

↓

Average Prediction
```

Averaging many trees substantially reduces variance while preserving the flexibility of individual decision trees.

Controlling tree complexity through pre-pruning, post-pruning, and careful hyperparameter tuning reduces variance and improves generalization. Nevertheless, even well-pruned trees remain relatively unstable, motivating ensemble methods such as Random Forests, which reduce variance by averaging the predictions of many independently constructed trees.
