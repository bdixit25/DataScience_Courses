# <font color='blue'> Introduction to Regularization </font>

As machine learning models become increasingly flexible, they gain the ability to capture complex relationships within the data. However, increased flexibility also increases the risk of fitting random fluctuations and noise rather than the underlying relationship. This phenomenon is known as **overfitting**.

Regularization is a general strategy for reducing overfitting by discouraging unnecessarily complex models. Instead of optimizing only the prediction error, regularization modifies the objective function by introducing a penalty for model complexity. This encourages simpler models that often generalize better to previously unseen data.

Modern regression techniques such as Ridge Regression, LASSO, Elastic Net, Support Vector Regression, and even many deep learning algorithms rely upon regularization to improve predictive performance.

---



# <font color='orange'> 1. Why Do We Need Regularization? </font>

Consider fitting a Polynomial Regression model.

### Degree 1

```
Simple

↓

High Bias

↓

Underfitting
```

---

### Degree 20

```
Very Flexible

↓

Fits Noise

↓

Overfitting
```

Although the high-degree polynomial may achieve an extremely small training error,

its performance on new data is often poor.

The objective of machine learning is **generalization**, not memorization.

Regularization helps achieve this objective.

---



# <font color='orange'> 2. The Idea Behind Regularization </font>

Without regularization,

Ordinary Least Squares minimizes

$$
\boxed{
RSS
=
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

Regularization modifies this objective by adding a penalty term.

The general form becomes

$$
\boxed{
\text{Loss}
=
\text{Prediction Error}
+
\text{Complexity Penalty}.
}
$$

Instead of minimizing prediction error alone,

the algorithm now seeks a balance between

* fitting the data accurately,
* keeping the model simple.

---



# <font color='orange'> 3. Model Complexity </font>

A model becomes more complex when

* coefficients become very large,
* many predictors contribute to the model,
* the model becomes highly sensitive to small changes in the data.

Large coefficients often indicate that the model is fitting noise rather than meaningful signal.

Regularization discourages excessively large coefficients.

---



# <font color='orange'> 4. Shrinkage </font>

The central idea of regularization is **shrinkage**.

Rather than allowing regression coefficients to grow without restriction,

regularization pulls them towards zero.

```
Without Regularization

β₁ = 8.4

β₂ = -6.9

β₃ = 5.7
```

↓

```
With Regularization

β₁ = 5.2

β₂ = -3.8

β₃ = 3.1
```

Notice that the coefficients remain useful,

but become less extreme.

This usually improves generalization.

---



# <font color='orange'> 5. The Bias–Variance Perspective </font>

Regularization deliberately introduces

a small amount of bias

in order to reduce variance.

```
Regularization

↓

Slightly Higher Bias

↓

Much Lower Variance

↓

Better Test Performance
```

This is a practical application of the Bias–Variance Trade-off.

Different regression algorithms penalize model complexity in different ways.

### Elastic Net

Combines both penalties.

These three methods differ primarily in how they measure model complexity.

# <font color='blue'> Ridge Regression (L2 Regularization) </font>

Linear Regression estimates the regression coefficients by minimizing the Residual Sum of Squares (RSS). Although this approach often produces accurate models, it can suffer from overfitting when the predictors are numerous, highly correlated, or when the training data contain substantial noise. In such situations, the estimated coefficients may become excessively large, leading to unstable predictions and poor generalization.

Ridge Regression addresses this problem by extending Ordinary Least Squares with an additional penalty on the magnitude of the regression coefficients. Rather than minimizing the prediction error alone, Ridge Regression simultaneously seeks accurate predictions and small coefficient values. This simple modification often produces models that are more stable, less sensitive to noise, and better able to generalize to previously unseen data.

---



# <font color='orange'> 1. Motivation </font>

Suppose we have fitted a Linear Regression model.

The estimated coefficients are

| Feature | Coefficient |
| :--- | ---: |
| Area | 12.8 |
| Bedrooms | -9.7 |
| Age | 15.6 |
| Distance | -13.2 |

Notice that several coefficients are quite large.

Large coefficients indicate that

small changes in the data

can produce

large changes in the predictions.

This is one symptom of a high-variance model.

Ridge Regression reduces this instability by shrinking the coefficients.

---



# <font color='orange'> 2. Ordinary Least Squares Review </font>

Linear Regression minimizes

$$
\boxed{
RSS
=
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

Only the prediction error is minimized.

The complexity of the model is ignored.

Consequently,

the coefficients may become unnecessarily large.

---



# <font color='orange'> 3. Ridge Regression Objective Function </font>

Ridge Regression modifies the objective function by adding an **L2 penalty**.

The objective becomes

$$
\boxed{
\min_{\boldsymbol{\beta}}
\left[
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
+
\lambda
\sum_{j=1}^{p}
\beta_j^2
\right].
}
$$

The first term measures

prediction error.

The second term penalizes

large coefficients.

The parameter

$$
\lambda
$$

controls

the strength of the regularization.

---



# <font color='orange'> 4. Understanding the L2 Penalty </font>

The Ridge penalty is

$$
\boxed{
\sum_{j=1}^{p}\beta_j^2.
}
$$

This quantity is simply the squared Euclidean length (L2 norm squared) of the coefficient vector.

Large coefficients

produce

large penalties.

Small coefficients

produce

small penalties.

The optimization therefore favours

smaller coefficients.

---



# <font color='orange'> 5. The Role of λ </font>

The regularization parameter

$$
\lambda
$$

determines the balance between

prediction accuracy

and

model simplicity.

### λ = 0

```
Ridge Regression

↓

Ordinary Linear Regression
```

No regularization is applied.

---

### Small λ

```
Weak Penalty

↓

Small Amount of Shrinkage
```

---

### Large λ

```
Strong Penalty

↓

Very Small Coefficients
```

If λ becomes excessively large,

the model may underfit.

Selecting λ is therefore a model-selection problem, typically addressed using cross-validation.

---

Every coefficient has moved

Notice,

# <font color='orange'> 7. Geometric Interpretation </font>

Ridge Regression can be interpreted geometrically.

OLS searches for the point that minimizes the RSS.

Ridge introduces the constraint

$$
\boxed{
\sum_{j=1}^{p}\beta_j^2
\le t,
}
$$

where

$t$

is determined by

$\lambda$.

The feasible region is therefore

a circle (two parameters)

or,

more generally,

a hypersphere.

```
      RSS Contours

         ○

      ○     ○

    ○    ●

  -----------

  Ridge Constraint
```

The solution occurs where

the RSS contours

first touch

the constraint boundary.

This geometric viewpoint explains why Ridge shrinks coefficients smoothly without setting them exactly to zero.

# <font color='blue'> LASSO Regression (L1 Regularization) </font>

Ridge Regression improves the generalization performance of Linear Regression by shrinking regression coefficients towards zero. However, Ridge Regression rarely eliminates predictors entirely. Every feature remains in the final model, even if its contribution is very small.

LASSO (Least Absolute Shrinkage and Selection Operator) extends Linear Regression using a different regularization penalty. Instead of penalizing the squared magnitude of the coefficients, LASSO penalizes their absolute values. This seemingly small mathematical change has profound consequences: some coefficients become exactly zero, allowing LASSO to perform **automatic feature selection** while simultaneously controlling overfitting.

LASSO is therefore both a regression method and a feature selection technique, making it particularly valuable for high-dimensional datasets where many predictors may be irrelevant.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict house prices using

* Area
* Bedrooms
* Age
* Garage Size
* Garden Area
* Distance to School
* Number of Fireplaces
* Swimming Pool
* Roof Colour

Do all of these variables genuinely contribute to predicting house prices?

Probably not.

Ideally,

the model should

* retain important predictors,
* discard irrelevant ones automatically.

LASSO is designed to achieve precisely this.

---



# <font color='orange'> 2. Review: Ordinary Least Squares </font>

Linear Regression minimizes

$$
\boxed{
RSS
=
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

Every predictor contributes to the model,

regardless of its usefulness.

This can lead to

* overfitting,
* unnecessarily complex models,
* difficult interpretation.

---



# <font color='orange'> 3. LASSO Objective Function </font>

LASSO modifies the objective function by adding an **L1 penalty**.

The optimization problem becomes

$$
\boxed{
\min_{\boldsymbol{\beta}}
\left[
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
+
\lambda
\sum_{j=1}^{p}
|\beta_j|
\right].
}
$$

The first term measures prediction error.

The second term penalizes large coefficients using their **absolute values**.

---



# <font color='orange'> 4. Understanding the L1 Penalty </font>

The LASSO penalty is

$$
\boxed{
\sum_{j=1}^{p}
|\beta_j|.
}
$$

Unlike Ridge,

which squares the coefficients,

LASSO penalizes them linearly.

Large coefficients receive larger penalties,

but the shape of the penalty encourages some coefficients to become **exactly zero**.

This property is what makes LASSO unique.

---



# <font color='orange'> 5. The Role of λ </font>

The parameter

$$
\lambda
$$

controls the strength of the regularization.

---

### λ = 0

```
LASSO

↓

Ordinary Linear Regression
```

No regularization is applied.

---

### Small λ

```
Weak Penalty

↓

Slight Coefficient Shrinkage
```

---

### Large λ

```
Strong Penalty

↓

Many Coefficients Become Zero
```

If λ becomes too large,

important predictors may also be removed,

leading to underfitting.

Selecting λ therefore requires careful tuning,

typically using cross-validation.

---

| Feature | OLS |

| Garage | 0.8 |

| :--- | ---: |
| Area | 6.9 |
| Bedrooms | 3.5 |
| Age | -1.8 |
| Garage | 0.4 |
| Roof Colour | **0** |
| Swimming Pool | **0** |

The last two variables are removed from the model.

This automatic feature selection often improves both interpretability and predictive performance.

---



# <font color='orange'> 7. Why Does LASSO Produce Zero Coefficients? </font>

The answer lies in the geometry of the optimization problem.

LASSO constrains the coefficients using

$$
\boxed{
\sum_{j=1}^{p}
|\beta_j|
\le t,
}
$$

where

$t$

depends on

$\lambda$.

For two coefficients,

this constraint forms a **diamond-shaped** region.

```
        β₂

         ▲

        ◇

      ◇   ◇

    ◇       ◇

--------●--------► β₁

    ◇       ◇

      ◇   ◇

        ◇
```

Notice that the corners of the diamond lie on the coordinate axes.

The optimization frequently reaches one of these corners,

forcing one or more coefficients to be exactly zero.

This geometric property explains why LASSO performs feature selection.

# <font color='blue'> Elastic Net Regression </font>

Ridge Regression and LASSO Regression are two of the most widely used regularization techniques in machine learning. Ridge Regression effectively reduces overfitting and handles multicollinearity by shrinking regression coefficients towards zero, whereas LASSO performs automatic feature selection by forcing some coefficients to become exactly zero.

Although both methods are powerful, neither is universally optimal. Ridge Regression retains all predictors even when some are irrelevant, while LASSO can behave unpredictably when several predictors are highly correlated, often selecting one variable and discarding the others.

Elastic Net Regression combines the strengths of both Ridge and LASSO by incorporating both L1 and L2 penalties into a single objective function. The resulting model simultaneously performs coefficient shrinkage and feature selection while maintaining greater stability when correlated predictors are present.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict house prices using

* Living Area
* Floor Area
* Building Size
* Number of Rooms
* Garage Size
* Garden Area

Notice that

```
Living Area

↓

Floor Area

↓

Building Size
```

are all highly correlated.

---

### Ridge Regression

Keeps

all three variables.

---

### LASSO

May keep

only one variable

and discard the others.

---

Neither behaviour is always desirable.

Elastic Net provides a compromise.

---



# <font color='orange'> 2. Review of Ridge and LASSO </font>

### Ridge Regression

Minimizes

$$
RSS
+
\lambda
\sum \beta_j^2.
$$

Characteristics

* Shrinks coefficients.
* Keeps every predictor.

---

### LASSO Regression

Minimizes

$$
RSS
+
\lambda
\sum |\beta_j|.
$$

Characteristics

* Shrinks coefficients.
* Performs feature selection.

---

Elastic Net combines both ideas.

---



# <font color='orange'> 3. Elastic Net Objective Function </font>

Elastic Net minimizes

$$
\boxed{
\min_{\boldsymbol{\beta}}
\left[
RSS
+
\lambda_1
\sum |\beta_j|
+
\lambda_2
\sum \beta_j^2
\right].
}
$$

The objective therefore contains

* prediction error,
* L1 penalty,
* L2 penalty.

Many software libraries express the same objective using a single overall regularization parameter and a mixing parameter, but the underlying idea remains the same: both penalties are applied simultaneously.

---



# <font color='orange'> 4. Understanding the Two Penalties </font>

The two penalties serve different purposes.

---

## L1 Penalty

Encourages

```
Sparse Models

↓

Feature Selection
```

Some coefficients become

exactly zero.

---

## L2 Penalty

Encourages

```
Small Coefficients

↓

Stable Estimates
```

Highly correlated predictors remain more stable than with LASSO alone.

---

Elastic Net combines

both behaviours.

---



# <font color='orange'> 5. The Mixing Parameter </font>

In practice,

Elastic Net is usually controlled by two hyperparameters.

---

### Overall Regularization Strength

Determines

how strongly

the coefficients are penalized.

---

### Mixing Parameter

Determines

the balance between

LASSO

and

Ridge.

```
0

↓

Pure Ridge
```

```
0.5

↓

Balanced Elastic Net
```

```
1

↓

Pure LASSO
```

Thus,

Elastic Net spans the entire spectrum between Ridge and LASSO.

# <font color='blue'> Comparing Linear Regression, Ridge Regression, LASSO Regression, and Elastic Net </font>

Linear Regression, Ridge Regression, LASSO Regression, and Elastic Net belong to the same family of supervised learning algorithms. Each models the target variable as a linear combination of the predictor variables, but they differ in how they estimate the regression coefficients and how they control model complexity.

Ordinary Least Squares focuses solely on minimizing the prediction error, whereas the regularized methods augment the objective function with penalty terms that discourage unnecessarily complex models. These penalties improve generalization, reduce overfitting, and, in some cases, perform automatic feature selection.

This chapter summarizes the similarities, differences, strengths, limitations, and practical applications of these four regression algorithms.

---



# <font color='orange'> 1. The Evolution of Linear Models </font>

Each method builds naturally upon the previous one.

```
Linear Regression

↓

Minimize Prediction Error
```

↓

```
Ridge Regression

↓

Prediction Error

+

L2 Penalty
```

↓

```
LASSO Regression

↓

Prediction Error

+

L1 Penalty
```

↓

```
Elastic Net

↓

Prediction Error

+

L1 Penalty

+

L2 Penalty
```

Rather than viewing these as unrelated algorithms,

it is more useful to regard them as successive refinements of the same underlying idea.

---



# <font color='orange'> 2. Objective Functions </font>

The four methods differ primarily in the objective function they optimize.

| Algorithm | Objective Function |
| :--- | :--- |
| Linear Regression | \(RSS\) |
| Ridge Regression | \(RSS+\lambda\sum\beta_j^2\) |
| LASSO Regression | \(RSS+\lambda\sum|\beta_j|\) |
| Elastic Net | \(RSS+\lambda_1\sum|\beta_j|+\lambda_2\sum\beta_j^2\) |

Each additional penalty modifies the balance between prediction accuracy and model complexity.

---

| Algorithm | Coefficients |
| :--- | :--- |
| Linear Regression | Estimated freely |
| Ridge Regression | Shrunk towards zero |
| LASSO Regression | Some coefficients become exactly zero |
| Elastic Net | Combination of shrinkage and feature selection |

Only LASSO and Elastic Net remove predictors completely.

# <font color='orange'> 5. Handling Multicollinearity </font>

Highly correlated predictors can destabilize coefficient estimates.

Suppose the dataset contains

* Living Area
* Floor Area
* Building Area

These variables convey similar information.

| Algorithm | Behaviour |
| :--- | :--- |
| Linear Regression | Unstable coefficients |
| Ridge Regression | Shares information across correlated predictors |
| LASSO Regression | Often keeps one predictor and removes the others |
| Elastic Net | Often retains groups of correlated predictors |

Elastic Net therefore combines the stability of Ridge with the sparsity of LASSO.

| Algorithm | Bias | Variance |
| :--- | :---: | :---: |
| Linear Regression | Low | High |
| Ridge Regression | Moderate | Lower |
| LASSO Regression | Moderate | Lower |
| Elastic Net | Moderate | Lower |

Regularized methods deliberately introduce a small amount of bias in exchange for a larger reduction in variance, often improving generalization.
