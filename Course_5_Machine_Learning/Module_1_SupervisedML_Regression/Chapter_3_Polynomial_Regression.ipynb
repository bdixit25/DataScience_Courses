# <font color='blue'> Polynomial Regression </font>

Linear Regression assumes that the expected value of the target variable is a linear function of the predictor variables. While this assumption is appropriate for many applications, numerous real-world relationships exhibit nonlinear behaviour. Examples include population growth, projectile motion, chemical reaction rates, biological growth processes, and many engineering systems.

Polynomial Regression extends Linear Regression by introducing polynomial terms of the predictor variables. Although the resulting relationship between the predictor and the target is nonlinear, the model remains **linear in its parameters**, allowing the same estimation techniques used in Linear Regression to be applied.

Polynomial Regression therefore provides a simple yet powerful method for modelling curved relationships without abandoning the familiar framework of linear models.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict the stopping distance of a car as a function of its speed.

A scatter plot of the observations might look like

```
Stopping Distance

^

|                          •

|                    •

|               •

|          •

|     •

|  •

+---------------------------------->

            Speed
```

Clearly,

the relationship is **curved** rather than a straight line.

If we fit a simple Linear Regression model,

the fitted line may systematically underestimate some observations and overestimate others.

This indicates that the linear assumption is inadequate.

---



# <font color='orange'> 2. Why Linear Regression May Fail </font>

Simple Linear Regression assumes

$$
\boxed{
y=\beta_0+\beta_1x+\varepsilon.
}
$$

This implies that

every one-unit increase in

$x$

produces the same expected change in

$y$.

Many natural processes do not behave this way.

Examples include

* Population growth
* Drug concentration
* Projectile motion
* Learning curves
* Equipment degradation

These systems often exhibit nonlinear relationships.

---



# <font color='orange'> 3. The Idea Behind Polynomial Regression </font>

Rather than fitting

only

$x$,

we also include higher-order powers of the predictor.

For example,

a quadratic model becomes

$$
\boxed{
y
=
\beta_0
+
\beta_1x
+
\beta_2x^2
+
\varepsilon.
}
$$

A cubic model becomes

$$
\boxed{
y
=
\beta_0
+
\beta_1x
+
\beta_2x^2
+
\beta_3x^3
+
\varepsilon.
}
$$

More generally,

a polynomial of degree

$d$

is

$$
\boxed{
y
=
\beta_0
+
\sum_{k=1}^{d}
\beta_kx^k
+
\varepsilon.
}
$$

---



# <font color='orange'> 4. Why is Polynomial Regression Still "Linear"? </font>

This is one of the most common points of confusion.

Although the predictor variable appears with higher powers,

the coefficients

$$
\beta_0,\beta_1,\ldots,\beta_d
$$

still appear **linearly**.

For example,

$$
y
=
\beta_0
+
\beta_1x
+
\beta_2x^2
$$

is linear in the unknown parameters

$$
\beta_0,\beta_1,\beta_2.
$$

Consequently,

Ordinary Least Squares can still estimate the coefficients.

Polynomial Regression is therefore a special case of Linear Regression with transformed features.

---



# <font color='orange'> 5. Feature Transformation </font>

Instead of fitting

```
x
```

we create additional features

```
x

↓

x²

↓

x³

↓

...

↓

xᵈ
```

The transformed dataset becomes

| Original Feature | New Features |
| :---: | :--- |
| \(x\) | \(x,\;x^2,\;x^3,\ldots,x^d\) |

Linear Regression is then applied to the transformed features.

---



# <font color='orange'> 6. Example </font>

Suppose the original feature is

```
House Age
```

After polynomial expansion of degree three,

the feature matrix becomes

| Original | Expanded Features |
| :---: | :--- |
| 10 | \(10,\;100,\;1000\) |
| 20 | \(20,\;400,\;8000\) |
| 30 | \(30,\;900,\;27000\) |

The regression algorithm now estimates one coefficient for each transformed feature.

# <font color='orange'> 8. Underfitting and Overfitting </font>

Choosing the polynomial degree is an important modelling decision.

---

### Degree Too Low

```
Actual Data

••••••••

↓

Straight Line
```

The model is too simple.

This leads to

**underfitting**.

---

### Degree Too High

```
Data

••••••••

↓

Highly Oscillating Curve
```

The model begins to fit

random noise

rather than

the underlying relationship.

This is

**overfitting**.

---

An appropriate degree balances model flexibility and generalization.

---



# <font color='orange'> 9. Choosing the Polynomial Degree </font>

The degree should not be chosen by visual inspection alone.

Instead,

evaluate several candidate degrees using

* Validation Set
* Cross Validation

Select the degree that minimizes the validation error rather than the training error.

---



# <font color='orange'> 10. Advantages </font>

Polynomial Regression offers several advantages.

* Captures curved relationships.
* Simple extension of Linear Regression.
* Uses Ordinary Least Squares.
* Easy to implement.
* Interpretable coefficients.

For moderately nonlinear problems,

it often provides excellent performance.

---



# <font color='orange'> 11. Limitations </font>

Polynomial Regression also has limitations.

### High Variance

High-degree polynomials

can overfit the data.

---

### Poor Extrapolation

Predictions outside the observed data range

may become unrealistic.

---

### Numerical Instability

Very large powers

may produce poorly conditioned design matrices.

Feature scaling often improves numerical stability.

---

### Global Model

A single polynomial attempts to fit the entire dataset.

Highly complex relationships may be better modelled using tree-based methods or splines.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Polynomial Regression | Linear Regression using polynomial features |
| Purpose | Model nonlinear relationships |
| Feature Expansion | Introduces powers of the predictor variables |
| Linear in Parameters | Yes |
| Estimation Method | Ordinary Least Squares |
| Main Hyperparameter | Polynomial degree |
| Main Risk | Overfitting for high degrees |

> **Key Insight:** Polynomial Regression extends Linear Regression by transforming the input features into polynomial terms while preserving linearity in the model parameters. This enables the model to capture nonlinear relationships using the familiar Ordinary Least Squares framework. The polynomial degree controls the model's flexibility: low degrees may underfit the data, whereas excessively high degrees may overfit by modelling random noise. Selecting an appropriate degree using validation or cross-validation is therefore essential for achieving good generalization performance.
