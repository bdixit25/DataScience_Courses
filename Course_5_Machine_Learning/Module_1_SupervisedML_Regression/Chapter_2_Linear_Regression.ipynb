# <font color='blue'> Linear Regression </font>

Linear Regression is the simplest and one of the most important supervised learning algorithms. It models the relationship between one or more input variables and a continuous target variable by assuming that the expected value of the target can be expressed as a linear combination of the input features.

Despite its simplicity, Linear Regression remains one of the most widely used predictive models in statistics, economics, engineering, finance, medicine, and the natural sciences. It provides not only accurate predictions for many problems but also interpretable relationships between variables, making it an essential tool for both prediction and scientific understanding.

Furthermore, many modern machine learning algorithms—including Ridge Regression, LASSO, Elastic Net, Generalized Linear Models, and even Neural Networks—can be viewed as extensions or generalizations of the ideas introduced in this chapter.

---

# <font color='orange'> 1. Motivation </font>

Suppose you want to predict the selling price of a house.

You collect data containing

| House Size (m²) | Price (£) |
| :---: | ---: |
| 80 | 220,000 |
| 100 | 270,000 |
| 120 | 315,000 |
| 150 | 390,000 |
| 180 | 460,000 |

A scatter plot of the data might reveal an approximately straight-line trend.

```
Price (£)

^

|

|                      •

|                 •

|            •

|        •

|   •

+-------------------------------->

         House Size
```

Rather than memorizing every observation individually,

we would like to learn a mathematical relationship that summarizes the overall trend.

Once that relationship has been learned,

we can predict the price of houses that have never been observed before.

---

# <font color='orange'> 2. What is a Linear Relationship? </font>

A relationship is said to be **linear** if the expected change in the target variable is proportional to changes in the input variables.

For a single predictor,

this relationship is represented by the equation of a straight line

$$
\boxed{
y = mx + c,
}
$$

where

* \(m\) is the slope,
* \(c\) is the intercept.

Machine learning adopts exactly the same idea,

except that the parameters are estimated from data rather than chosen manually.

---

# <font color='orange'> 3. Simple Linear Regression </font>

Simple Linear Regression involves

* one predictor,
* one continuous target.

The model is

$$
\boxed{
y=\beta_0+\beta_1x+\varepsilon,
}
$$

where

* \(y\) is the observed target,
* \(x\) is the predictor,
* \(\beta_0\) is the intercept,
* \(\beta_1\) is the slope,
* \(\varepsilon\) is the random error term.

The predicted value is

$$
\boxed{
\hat y=\beta_0+\beta_1x.
}
$$

Notice that the prediction does **not** include the error term,

because the error is unknown for future observations.

---

# <font color='orange'> 4. Interpretation of the Parameters </font>

The regression coefficients have intuitive interpretations.

### Intercept (\(\beta_0\))

The predicted value of the target when

$$
x=0.
$$

Depending on the application,

this may or may not have a meaningful physical interpretation.

---

### Slope (\(\beta_1\))

The expected change in the target variable associated with a one-unit increase in the predictor,

assuming all other variables remain unchanged.

For example,

if

$$
\beta_1=2500,
$$

then every additional square metre of house area increases the predicted price by

£2,500,

on average.

---

# <font color='orange'> 5. Multiple Linear Regression </font>

Real-world problems rarely depend on only one predictor.

Suppose we wish to predict house prices using

* Floor Area,
* Number of Bedrooms,
* Property Age.

The model becomes

$$
\boxed{
y
=
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+
\cdots
+
\beta_px_p
+
\varepsilon.
}
$$

The corresponding prediction equation is

$$
\boxed{
\hat y
=
\beta_0
+
\sum_{j=1}^{p}\beta_jx_j.
}
$$

Although this model contains many variables,

it remains **linear** because it is linear in the unknown coefficients \(\beta_j\).

---

# <font color='orange'> 6. Why is it Called "Linear"? </font>

A common misconception is that "linear" refers to the shape of the data.

In fact,

Linear Regression is called **linear** because it is **linear in the parameters**, not necessarily in the predictor variables.

For example,

the model

$$
y
=
\beta_0
+
\beta_1x
+
\beta_2x^2
$$

is still a linear model because the coefficients

\(\beta_0,\beta_1,\beta_2\)

appear linearly,

even though the relationship with \(x\) is curved.

This observation provides the foundation for **Polynomial Regression**, which we will study later.

---

# <font color='orange'> 7. Predictions and Residuals </font>

For every observation,

the model produces a prediction

$$
\hat y_i.
$$

The difference between the observed value and the prediction is called the **residual**

$$
\boxed{
e_i
=
y_i-\hat y_i.
}
$$

Residuals measure how well the model fits the observed data.

A good regression model produces residuals that are

* small,
* randomly distributed,
* centered around zero.

Residual analysis is an essential part of regression diagnostics.

---

# <font color='orange'> 8. Ordinary Least Squares (OLS) </font>

How should we choose the coefficients

\(\beta_0,\beta_1,\ldots,\beta_p\)?

The most common approach is **Ordinary Least Squares (OLS)**.

OLS selects the coefficients that minimize the **Residual Sum of Squares (RSS)**

$$
\boxed{
RSS
=
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

Squaring the residuals ensures that

* positive and negative errors do not cancel,
* larger errors receive greater emphasis.

OLS therefore identifies the line (or hyperplane) that best fits the training data in the least-squares sense.

---

# <font color='orange'> 9. Matrix Formulation </font>

Multiple Linear Regression can be expressed compactly using matrix notation.

Let

$$
\mathbf{X}
=
\begin{bmatrix}
1 & x_{11} & \cdots & x_{1p}\\
1 & x_{21} & \cdots & x_{2p}\\
\vdots & \vdots & \ddots & \vdots\\
1 & x_{n1} & \cdots & x_{np}
\end{bmatrix},
$$

and

$$
\boldsymbol{\beta}
=
\begin{bmatrix}
\beta_0\\
\beta_1\\
\vdots\\
\beta_p
\end{bmatrix}.
$$

The regression model becomes

$$
\boxed{
\mathbf{y}
=
\mathbf{X}\boldsymbol{\beta}
+
\boldsymbol{\varepsilon}.
}
$$

This compact representation simplifies both theoretical analysis and software implementation.

---

# <font color='orange'> 10. Advantages of Linear Regression </font>

Linear Regression remains popular because it is

* simple,
* computationally efficient,
* highly interpretable,
* easy to implement,
* supported by strong statistical theory.

It often performs surprisingly well when relationships are approximately linear.

---

# <font color='orange'> 11. Limitations </font>

Linear Regression also has important limitations.

It assumes

* approximately linear relationships,
* independent observations,
* constant error variance,
* absence of severe multicollinearity.

It may perform poorly when

* relationships are highly nonlinear,
* strong interactions exist,
* outliers dominate the data.

These limitations motivate more flexible regression algorithms such as Decision Tree Regression, Random Forest Regression, and Gradient Boosting.

---

# <font color='orange'> 12. Typical Applications </font>

Linear Regression is widely used for

* House price prediction
* Sales forecasting
* Economic modelling
* Medical research
* Engineering calibration
* Environmental modelling
* Scientific data analysis

Even when more sophisticated models are ultimately adopted,

Linear Regression often serves as the baseline against which all other models are compared.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Linear Regression | Models a continuous target as a linear combination of predictors |
| Simple Linear Regression | One predictor |
| Multiple Linear Regression | Two or more predictors |
| Intercept (\(\beta_0\)) | Predicted target when all predictors are zero |
| Slope (\(\beta_j\)) | Expected change in the target for a one-unit increase in a predictor |
| Residual | Difference between the observed and predicted values |
| Ordinary Least Squares | Estimates coefficients by minimizing the Residual Sum of Squares (RSS) |

> **Key Insight:** Linear Regression models the relationship between one or more predictor variables and a continuous response by assuming that the expected value of the response can be expressed as a linear combination of the predictors. Its coefficients have clear interpretations, its mathematical formulation is elegant, and its parameters can be estimated efficiently using Ordinary Least Squares. Although more flexible regression models exist, Linear Regression remains the cornerstone of regression analysis and provides the conceptual and mathematical foundation for many advanced machine learning algorithms.

# <font color='blue'> Assumptions of Linear Regression </font>

Linear Regression is one of the simplest and most widely used predictive models in statistics and machine learning. However, the validity of its parameter estimates, confidence intervals, hypothesis tests, and interpretation depends upon several underlying assumptions about the data and the relationship between the predictors and the target variable.

These assumptions do not guarantee that a regression model will produce accurate predictions, nor do small violations necessarily invalidate the model. Instead, they provide a mathematical framework under which the Ordinary Least Squares (OLS) estimator possesses desirable statistical properties such as unbiasedness, efficiency, and consistency.

Understanding these assumptions is therefore essential for determining when Linear Regression is appropriate and for diagnosing potential modelling problems.

---

# <font color='orange'> 1. Why Do We Need Assumptions? </font>

Linear Regression estimates the coefficients

$$
\beta_0,\beta_1,\ldots,\beta_p
$$

using Ordinary Least Squares.

If certain assumptions hold,

these estimates possess several desirable properties.

They are

* unbiased,
* consistent,
* efficient,
* statistically interpretable.

If important assumptions are violated,

the model may

* produce biased coefficients,
* underestimate uncertainty,
* yield misleading hypothesis tests,
* make less reliable predictions.

Consequently,

every regression analysis should include an assessment of the underlying assumptions.

---

# <font color='orange'> 2. The Seven Classical Assumptions </font>

The assumptions of Linear Regression can be summarized as follows.

| Assumption | Purpose |
| :--- | :--- |
| Linearity | Relationship is approximately linear |
| Independence | Observations are independent |
| Exogeneity | Errors have zero conditional mean |
| Homoscedasticity | Constant error variance |
| Normality | Residuals are approximately normally distributed |
| No Multicollinearity | Predictors are not highly correlated |
| No Influential Outliers | Individual observations do not dominate the fit |

Each assumption is discussed in detail below.

---

# <font color='orange'> 3. Assumption 1: Linearity </font>

The expected value of the target should be a linear function of the predictors.

Mathematically,

$$
E[Y|X]
=
\beta_0
+
\beta_1X_1
+
\cdots
+
\beta_pX_p.
$$

Notice that the assumption concerns the **expected value** of the response, not every individual observation.

---

### Interpretation

Each predictor contributes additively and linearly to the expected response.

This does **not** imply that the observed data points lie exactly on a straight line.

Random noise is expected.

---

### Detecting Violations

Common diagnostic tools include

* Scatter plots
* Residual vs fitted plots
* Partial residual plots

Curved patterns in the residuals often indicate that the linear model is inadequate.

---

### Possible Solutions

* Polynomial Regression
* Feature transformations
* Interaction terms
* Tree-based regression models

---

# <font color='orange'> 4. Assumption 2: Independence of Observations </font>

The observations in the dataset should be statistically independent.

In other words,

the error associated with one observation should not depend upon the error of another.

Mathematically,

$$
\operatorname{Cov}(\varepsilon_i,\varepsilon_j)=0
\qquad
(i\neq j).
$$

---

### When This Assumption Fails

Examples include

* Time-series data
* Repeated measurements
* Longitudinal studies
* Clustered observations

Ignoring dependence often leads to underestimated standard errors and overconfident statistical conclusions.

---

### Diagnostic Methods

* Residual plots versus time
* Durbin–Watson statistic
* Autocorrelation function (ACF)

---

### Possible Solutions

* Time-series models
* Mixed-effects models
* Generalized Least Squares (GLS)

---

# <font color='orange'> 5. Assumption 3: Exogeneity (Zero Conditional Mean) </font>

Perhaps the most important assumption is

$$
\boxed{
E[\varepsilon|X]=0.
}
$$

This means that,

given the predictor variables,

the average error is zero.

---

### Interpretation

All systematic information relevant to predicting the response has been captured by the predictors.

The remaining errors are purely random.

---

### Violations

This assumption fails when

* important predictors are omitted,
* measurement error exists in the predictors,
* reverse causality is present,
* endogeneity occurs.

Violating this assumption leads to biased coefficient estimates.

---

# <font color='orange'> 6. Assumption 4: Homoscedasticity </font>

The variance of the errors should remain constant across all values of the predictors.

Mathematically,

$$
\boxed{
\operatorname{Var}(\varepsilon|X)=\sigma^2.
}
$$

---

### Homoscedastic Errors

```
Residual

^

|  • • • • •

| • • • • •

|  • • • • •

+------------------>

      Fitted
```

---

### Heteroscedastic Errors

```
Residual

^

|        •

|      •   •

|    •      •

|  •          •

+------------------>

      Fitted
```

The spread increases with the fitted values.

---

### Diagnostic Methods

* Residual vs fitted plots
* Breusch–Pagan test
* White test

---

### Possible Solutions

* Log transformation
* Weighted Least Squares
* Robust standard errors

---

# <font color='orange'> 7. Assumption 5: Normality of Residuals </font>

The residuals should be approximately normally distributed.

Mathematically,

$$
\varepsilon
\sim
N(0,\sigma^2).
$$

---

### Why Does It Matter?

Normality is primarily required for

* confidence intervals,
* hypothesis tests,
* prediction intervals.

OLS coefficient estimates remain unbiased even when this assumption is not perfectly satisfied.

---

### Diagnostic Methods

* Histogram
* Q-Q plot
* Shapiro–Wilk test

---

### Possible Solutions

* Transform the response
* Collect more data
* Robust regression methods

---

# <font color='orange'> 8. Assumption 6: No Multicollinearity </font>

The predictor variables should not exhibit strong linear dependence.

For example,

```
Height (cm)

Height (m)
```

contain essentially the same information.

---

### Why Is It Problematic?

High multicollinearity

* inflates coefficient variances,
* makes coefficients unstable,
* complicates interpretation.

---

### Diagnostic Methods

* Correlation matrix
* Variance Inflation Factor (VIF)

As a practical guideline,

* VIF < 5 is generally acceptable.
* VIF > 10 often indicates serious multicollinearity.

---

### Possible Solutions

* Remove redundant predictors
* Principal Component Analysis (PCA)
* Ridge Regression
* LASSO

---

# <font color='orange'> 9. Assumption 7: No Influential Outliers </font>

Some observations may exert a disproportionately large influence on the fitted regression line.

These are called **influential observations**.

---

### Why Are They Important?

A single extreme observation can

* substantially alter the estimated coefficients,
* distort predictions,
* affect statistical inference.

---

### Diagnostic Methods

* Leverage
* Cook's Distance
* DFFITS
* DFBetas

---

### Possible Solutions

* Investigate data quality
* Correct data entry errors
* Robust regression
* Remove observations only with appropriate justification

---

# <font color='orange'> 10. Summary of Diagnostic Tools </font>

| Assumption | Common Diagnostic |
| :--- | :--- |
| Linearity | Residual vs fitted plot |
| Independence | Durbin–Watson statistic |
| Exogeneity | Subject-matter knowledge, study design |
| Homoscedasticity | Breusch–Pagan test, residual plot |
| Normality | Q-Q plot, Shapiro–Wilk test |
| Multicollinearity | Variance Inflation Factor (VIF) |
| Influential Outliers | Cook's Distance, leverage plots |

---

# <font color='orange'> 11. What Happens If Assumptions Are Violated? </font>

Not all violations have the same consequences.

| Assumption Violated | Possible Consequence |
| :--- | :--- |
| Linearity | Poor predictive performance |
| Independence | Incorrect standard errors |
| Exogeneity | Biased coefficient estimates |
| Homoscedasticity | Inefficient estimates, unreliable inference |
| Normality | Invalid confidence intervals for small samples |
| Multicollinearity | Unstable coefficients |
| Influential Outliers | Distorted regression line |

Modern machine learning often prioritizes predictive performance over statistical inference. Consequently, some assumption violations may have little impact on prediction accuracy but significantly affect interpretability and inference.

---

# <font color='purple'> 12. Conceptual Summary </font>

| Homoscedasticity | Error variance is constant |

| No Multicollinearity | Predictors are not highly correlated |

> **Key Insight:** Linear Regression relies on several assumptions that ensure the Ordinary Least Squares estimator possesses desirable statistical properties. These assumptions govern the validity of coefficient estimates, confidence intervals, and hypothesis tests, while also providing guidance for diagnosing modelling problems. Although minor violations do not necessarily prevent accurate prediction, understanding and assessing these assumptions is essential for building reliable, interpretable, and scientifically meaningful regression models.

# <font color='blue'> Training, Validation, and Test Splits </font>

One of the primary objectives of supervised machine learning is to develop models that generalize well to previously unseen data. A regression model that performs exceptionally well on the data used during training may nevertheless perform poorly when applied to new observations. This phenomenon, known as **overfitting**, highlights the importance of evaluating models using data that were not involved in the training process.

To obtain an unbiased estimate of a model's predictive performance, the available dataset is typically divided into separate subsets for training, validation, and testing. Each subset serves a distinct purpose during model development and helps ensure that the reported performance reflects the model's ability to generalize beyond the training data.

---

# <font color='orange'> 1. Why Do We Split the Dataset? </font>

Suppose we have collected

```
10,000

house sales
```

A natural question is

> **Should we use all of the data to train the model?**

Although this might seem reasonable,

doing so creates an important problem.

If the model is evaluated using the same data on which it was trained,

it may simply memorize the training observations.

As a result,

the reported prediction error would be unrealistically optimistic.

The true objective of machine learning is not to memorize the training data,

but to make accurate predictions for **new, unseen observations**.

Dataset splitting provides an unbiased estimate of this ability.

---

# <font color='orange'> 2. Generalization </font>

A machine learning model should learn

the underlying relationship

between the predictors

and

the target,

rather than memorizing individual observations.

```
Training Data

↓

Learn Relationship

↓

Generalize

↓

New Data
```

Generalization is therefore the central objective of supervised learning.

---

# <font color='orange'> 3. The Three Datasets </font>

The complete dataset is usually divided into three parts.

```
Entire Dataset

↓

Training Set

↓

Validation Set

↓

Test Set
```

Each serves a different purpose.

---

# <font color='orange'> 4. Training Set </font>

The training set is the subset used to estimate the model parameters.

For Linear Regression,

this means estimating

$$
\beta_0,\beta_1,\ldots,\beta_p.
$$

The model has full access to these observations during learning.

Typical proportion

```
60–80%
```

of the available data.

---

## Purpose

* Learn model parameters
* Fit regression coefficients
* Minimize the chosen loss function

The model should **never** be evaluated solely on the training set.

---

# <font color='orange'> 5. Validation Set </font>

The validation set is used during model development,

but **not** during parameter estimation.

Its primary purpose is to compare different models and tune hyperparameters.

Examples include selecting

* polynomial degree,
* regularization strength (\(\lambda\)),
* tree depth,
* number of estimators.

Typical proportion

```
10–20%
```

of the data.

---

## Purpose

* Hyperparameter tuning
* Model comparison
* Early stopping
* Feature selection

The validation set acts as a "practice exam" before the final evaluation.

---

# <font color='orange'> 6. Test Set </font>

The test set is reserved until the very end of the modelling process.

The model should never use the test data

during

* training,
* hyperparameter tuning,
* feature engineering,
* model selection.

Its only purpose is to provide an unbiased estimate of the final model's predictive performance.

Typical proportion

```
10–20%
```

of the data.

---

## Purpose

* Final performance evaluation
* Estimate real-world generalization

Think of the test set as the final examination.

Once it has been used,

its results should not influence further model development.

---

# <font color='orange'> 7. Typical Splits </font>

Several splitting strategies are commonly used.

| Training | Validation | Test |
| :---: | :---: | :---: |
| 60% | 20% | 20% |
| 70% | 15% | 15% |
| 80% | 10% | 10% |

The optimal proportions depend on the size of the dataset.

Larger datasets can allocate relatively smaller proportions to validation and testing.

---

# <font color='orange'> 8. Hold-Out Validation </font>

The simplest approach is the **hold-out method**.

```
Dataset

↓

Random Split

↓

Training

↓

Testing
```

Advantages

* Simple
* Fast
* Easy to implement

Limitations

* Performance estimate depends on a single random split.
* May exhibit high variance for small datasets.

These limitations motivate **Cross-Validation**, which will be discussed in the next chapter.

---

# <font color='orange'> 9. Random Splitting </font>

For most regression problems,

observations are randomly assigned to the different subsets.

Randomization helps ensure that

each subset is representative of the overall data distribution.

In Scikit-Learn,

this is commonly implemented using

```python
train_test_split()
```

---

# <font color='orange'> 10. Data Leakage </font>

One of the most serious mistakes in machine learning is **data leakage**.

Data leakage occurs when information from the validation or test set influences model training.

Examples include

* Scaling the entire dataset before splitting.
* Performing feature selection using all observations.
* Using future information in time-series prediction.
* Selecting hyperparameters based on test performance.

Data leakage leads to overly optimistic performance estimates that do not reflect real-world behaviour.

---

# <font color='orange'> 11. Correct Workflow </font>

The correct sequence is

```
Raw Dataset

↓

Split Data

↓

Fit Preprocessing

(Training Set Only)

↓

Transform Validation/Test

↓

Train Model

↓

Tune Hyperparameters

↓

Final Test Evaluation
```

Any preprocessing step that learns from the data

(such as scaling, imputation, or feature selection)

must be fitted using the **training data only** and then applied to the validation and test sets.

---

# <font color='orange'> 12. Regression vs Classification Splits </font>

The overall principles are identical,

but one difference deserves attention.

### Classification

Class proportions should usually be preserved using **stratified splitting**.

---

### Regression

Because the target is continuous,

standard random splitting is generally appropriate.

For highly skewed continuous targets,

some practitioners perform approximate stratification by binning the target variable before splitting. While this can help maintain similar target distributions across subsets, it should be applied thoughtfully, as the bins are an approximation rather than true classes.

---

# <font color='orange'> 13. Time-Series Regression </font>

Random splitting is **not appropriate** for time-dependent data.

Example

Predicting

```
Tomorrow's Stock Price
```

We cannot train using future observations to predict the past.

Instead,

the split should respect chronological order.

```
Past

↓

Training

↓

Validation

↓

Test

↓

Future
```

Time-series regression therefore requires specialized validation strategies such as rolling-window or expanding-window evaluation.

---

# <font color='orange'> 14. Best Practices </font>

When splitting regression datasets,

consider the following recommendations.

* Shuffle observations unless temporal order is important.
* Reserve an independent test set.
* Use a validation set or cross-validation for model selection.
* Avoid data leakage.
* Fit preprocessing only on the training data.
* Never tune hyperparameters using the test set.

Following these principles helps ensure that performance estimates remain reliable and reproducible.

---

# <font color='orange'> 15. Practical Example </font>

Suppose we have

```
5,000

housing observations
```

A common split is

```
Training

↓

3,500

(70%)
```

```
Validation

↓

750

(15%)
```

```
Test

↓

750

(15%)
```

The model is trained using the first subset,

tuned using the second,

and evaluated only once using the third.

| Dataset | Purpose |
| :--- | :--- |
| Training Set | Learn model parameters |
| Validation Set | Tune hyperparameters and compare models |
| Test Set | Estimate final generalization performance |
| Hold-Out Validation | Single random split |
| Data Leakage | Information from validation/test influences training |
| Generalization | Performance on previously unseen data |

> **Key Insight:** Splitting a dataset into training, validation, and test subsets is essential for developing regression models that generalize well beyond the observed data. The training set is used to estimate model parameters, the validation set guides model selection and hyperparameter tuning, and the test set provides an unbiased assessment of the final model's predictive performance. By carefully separating these stages and avoiding data leakage, practitioners obtain realistic estimates of model performance and reduce the risk of overfitting.
