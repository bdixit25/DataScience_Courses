# <font color='blue'> Ensemble Learning: Combining Multiple Models </font>

One of the central ideas in modern machine learning is that combining the predictions of multiple models often produces better performance than relying on a single model. This principle forms the basis of **ensemble learning**, a family of techniques that construct multiple predictive models and combine their outputs to obtain more accurate, more robust, and more stable predictions.

The motivation behind ensemble learning is simple. Every individual model has strengths and weaknesses. A single decision tree, for example, may fit one particular training dataset extremely well but perform poorly on new observations because of its high variance. By combining many diverse models, individual errors tend to cancel one another, resulting in improved generalization.

Many of the most successful machine learning algorithms—including Random Forests, Gradient Boosting, XGBoost, LightGBM, and CatBoost—are ensemble methods.

---

# <font color='orange'> 1. Why Do We Need Ensemble Learning? </font>

Suppose we ask

one doctor

to diagnose a patient.

The diagnosis may be accurate,

but it is also influenced by that doctor's experience, judgement, and possible errors.

Now suppose

five independent specialists

examine the same patient.

Their collective opinion is usually more reliable than that of any single doctor.

Machine learning follows exactly the same principle.

Instead of relying on

one predictive model,

we combine

many models.

---

# <font color='orange'> 2. What is an Ensemble? </font>

An **ensemble** is a collection of multiple learning algorithms whose predictions are combined to produce a single final prediction.

Conceptually,

```
Model 1

↓

Prediction
```

```
Model 2

↓

Prediction
```

```
Model 3

↓

Prediction
```

```
...

↓

Combine Predictions

↓

Final Prediction
```

The individual models are often called **base learners** or **weak learners**.

---

# <font color='orange'> 3. Why Does It Work? </font>

Individual models make different errors.

If these errors are

* relatively independent,
* not systematically biased in the same direction,

then averaging or combining the predictions reduces the overall prediction error.

For regression,

the final prediction is often

the average

of the individual predictions.

For classification,

the final prediction is commonly

majority voting

or

averaged class probabilities.

---

# <font color='orange'> 4. Diversity is the Key </font>

An ensemble is most effective when its individual models are **diverse**.

Suppose we build

100 identical decision trees

using exactly the same training data.

```
Tree 1

↓

240
```

```
Tree 2

↓

240
```

```
Tree 3

↓

240
```

Averaging identical predictions provides no benefit.

Instead,

we want

different trees

that make

slightly different mistakes.

The diversity of the models allows many individual errors to cancel when combined.

---

# <font color='orange'> 5. Two Major Families of Ensemble Methods </font>

Most ensemble algorithms belong to one of two categories.

---

## Bagging (Bootstrap Aggregating)

Models are trained

**independently**

and

their predictions are combined.

```
Training Data

↓

Bootstrap Samples

↓

Tree 1

Tree 2

Tree 3

↓

Average
```

Examples

* Random Forest
* Extra Trees

---

## Boosting

Models are trained

**sequentially**.

Each new model attempts to correct the mistakes made by the previous models.

```
Model 1

↓

Errors

↓

Model 2

↓

Errors

↓

Model 3

↓

Final Model
```

Examples

* AdaBoost
* Gradient Boosting
* XGBoost
* LightGBM
* CatBoost

---

# <font color='orange'> 6. Bagging </font>

Bagging stands for

**Bootstrap Aggregating**.

The procedure is

1. Generate many bootstrap samples from the original dataset.
2. Train one model on each bootstrap sample.
3. Combine the predictions.

For regression,

the predictions are averaged.

Bagging primarily reduces

**variance**.

Decision Trees benefit greatly from bagging because they are naturally high-variance models.

---

# <font color='orange'> 7. Boosting </font>

Boosting follows a different philosophy.

Instead of training models independently,

each model learns from the errors made by its predecessors.

```
Weak Model

↓

Identify Errors

↓

Improve Model

↓

Repeat
```

Each successive model focuses increasingly on observations that were difficult to predict.

Boosting primarily reduces

**bias**,

although it can also reduce variance depending on the algorithm.

---

| Examples | Random Forest | Gradient Boosting, XGBoost |

Both methods improve predictive performance,

but they achieve this in fundamentally different ways.

# <font color='blue'> Random Forest Regression </font>

Decision Tree Regression is a powerful and flexible algorithm capable of modelling highly nonlinear relationships. However, individual decision trees suffer from an important limitation: they are highly sensitive to the particular training dataset. Small changes in the data can produce substantially different trees, leading to high variance and poor generalization.

Random Forest Regression addresses this problem by combining a large collection of decision trees into a single ensemble model. Rather than relying on the prediction of one tree, Random Forest trains many diverse trees using different subsets of the data and predictor variables, then averages their predictions. This simple yet effective strategy substantially reduces variance while retaining the flexibility of individual decision trees.

Random Forest is one of the most widely used machine learning algorithms because it typically achieves excellent predictive performance with relatively little hyperparameter tuning.

---

# <font color='orange'> 1. Motivation </font>

Suppose we train a single Decision Tree on a housing dataset.

```
Training Data

↓

Decision Tree

↓

Prediction
```

The prediction depends strongly on the exact training observations.

If we slightly modify the dataset,

the tree may change completely.

This instability leads to high variance.

Instead,

imagine training

```
500

different trees.
```

Each tree makes a slightly different prediction.

The final prediction is

the average

of all trees.

Individual errors tend to cancel,

producing much more stable predictions.

---

# <font color='orange'> 2. What is a Random Forest? </font>

A Random Forest is an ensemble of decision trees.

Each tree is trained

* on a different bootstrap sample,
* using a random subset of the available features at each split.

For regression,

the final prediction is

$$
\boxed{
\hat y
=
\frac1B
\sum_{b=1}^{B}
\hat y_b,
}
$$

where

* \(B\) is the number of trees,
* \(\hat y_b\) is the prediction of the \(b\)-th tree.

Averaging the predictions reduces variance and improves generalization.

---

# <font color='orange'> 3. Bootstrap Sampling </font>

The first source of randomness comes from

**bootstrap sampling**.

Instead of training every tree on the entire dataset,

each tree is trained on a bootstrap sample obtained by sampling the training data **with replacement**.

Suppose the original dataset contains

```
1000

observations.
```

Each tree receives

```
1000

samples
```

drawn randomly **with replacement**.

Consequently,

* some observations appear multiple times,
* some observations are omitted.

Each tree therefore sees a slightly different version of the data.

---

# <font color='orange'> 4. Random Feature Selection </font>

The second source of randomness occurs during tree construction.

Instead of considering

all predictors

at every split,

the algorithm considers only

a random subset.

Example

Suppose

```
20

predictors
```

are available.

At each split,

the algorithm may consider only

```
5

randomly selected predictors.
```

This encourages diversity among the trees.

Without this step,

many trees would become nearly identical.

---

# <font color='orange'> 5. Building the Forest </font>

The training procedure can be summarized as follows.

```
Training Data

↓

Bootstrap Sample 1

↓

Tree 1
```

```
Training Data

↓

Bootstrap Sample 2

↓

Tree 2
```

```
Training Data

↓

Bootstrap Sample 3

↓

Tree 3
```

```
...

↓

Average Predictions
```

Each tree is grown independently,

making Random Forest highly parallelizable.

---

# <font color='orange'> 6. Why Averaging Works </font>

Suppose three trees predict

```
240

245

255
```

The Random Forest prediction is

$$
\frac{240+245+255}{3}
=
246.7.
$$

The averaging process reduces the effect of unusually high or low predictions made by individual trees.

If the trees make different errors,

their average is often much more accurate than any single prediction.

---

# <font color='orange'> 7. Out-of-Bag (OOB) Samples </font>

Because bootstrap sampling is performed with replacement,

each tree does **not** use every training observation.

The observations omitted from a tree's bootstrap sample are called **Out-of-Bag (OOB) samples**.

These samples can be used as an internal validation set.

The resulting **Out-of-Bag error** provides an estimate of the model's generalization performance without requiring a separate validation dataset.

Although OOB estimates are often very useful, many practitioners still use cross-validation when comparing different models or tuning hyperparameters.

---

# <font color='orange'> 8. Feature Importance </font>

Random Forest can estimate the relative importance of each predictor.

Two commonly used approaches are

### Mean Decrease in Impurity (MDI)

Measures

how much

each feature reduces impurity

across all trees.

---

### Permutation Importance

Randomly shuffles one feature

and measures

how much

the prediction accuracy decreases.

Features whose randomization substantially worsens performance are considered more important.

Permutation importance is generally more reliable because it is less affected by biases that can arise in impurity-based importance measures.

---

Among these,

`n_estimators`

# <font color='blue'> Extra Trees Regression (Extremely Randomized Trees) </font>

Random Forest Regression improves the stability and predictive performance of individual decision trees by combining many independently trained trees using bootstrap sampling and random feature selection. Although this approach substantially reduces variance, each decision tree still searches for the best split among the randomly selected features.

Extra Trees Regression (Extremely Randomized Trees) introduces an additional level of randomness. Instead of searching for the optimal split threshold, Extra Trees chooses split thresholds at random and then selects the best among these randomly generated candidates. This additional randomization increases the diversity of the individual trees, often leading to further reductions in variance and, in many cases, improved generalization.

Extra Trees therefore represents another important ensemble method based on bagging, differing from Random Forest primarily in how the individual trees are constructed.

---

# <font color='orange'> 1. Motivation </font>

Recall how Random Forest builds its trees.

For each split,

it

* selects a random subset of features,
* searches for the **best split** among those features.

Although the trees differ because of bootstrap sampling and random feature selection,

each split is still chosen optimally.

Extra Trees asks an interesting question:

> **What if we also randomize the split locations?**

Surprisingly,

adding more randomness often improves predictive performance.

---

# <font color='orange'> 2. From Decision Trees to Extra Trees </font>

The progression is

```
Decision Tree

↓

One Tree
```

↓

```
Random Forest

↓

Many Trees

↓

Bootstrap Sampling

↓

Random Features
```

↓

```
Extra Trees

↓

Many Trees

↓

Random Features

↓

Random Split Thresholds
```

Every stage increases model diversity.

---

# <font color='orange'> 3. How Does Extra Trees Work? </font>

For each split,

the algorithm performs the following steps.

1. Select a random subset of predictors.
2. Generate random split thresholds for each selected predictor.
3. Evaluate the candidate splits.
4. Choose the best of these randomly generated splits.
5. Continue recursively.

Unlike Random Forest,

the algorithm does **not** search exhaustively for the globally best split threshold.

---

### Extra Trees

Generates a few random split points

```
95

145
```

and selects the better of those.

The split search is therefore much more random.

Some implementations also allow bootstrap sampling, but the standard Extra Trees algorithm does not require it.

If the individual trees are less correlated,

their errors cancel more effectively when averaged.

This often produces lower variance than Random Forest.

# <font color='blue'> Boosting: Learning from Mistakes </font>

Ensemble learning improves predictive performance by combining multiple models into a single, more powerful predictor. The previous chapters introduced **bagging**, where multiple models are trained independently and their predictions are averaged to reduce variance. Random Forest and Extra Trees are two prominent examples of this approach.

Boosting represents a fundamentally different philosophy. Instead of training models independently, boosting trains them **sequentially**, with each new model focusing on the mistakes made by the previous models. Rather than averaging many independent predictors, boosting constructs a strong learner by gradually improving a sequence of weak learners.

This chapter introduces the fundamental concepts of boosting and provides the conceptual foundation for Gradient Boosting, XGBoost, LightGBM, and CatBoost.

---

# <font color='orange'> 1. Why Do We Need Boosting? </font>

Recall the purpose of bagging.

```
Decision Tree

↓

High Variance

↓

Bagging

↓

Average Many Trees

↓

Lower Variance
```

Bagging is excellent for reducing variance.

However,

what if the individual trees themselves are systematically inaccurate?

In other words,

what if the model has **high bias**?

Averaging many weak models that all make similar systematic errors may not fully solve the problem.

Boosting addresses this issue by improving the models sequentially.

---

# <font color='orange'> 2. The Central Idea </font>

Rather than building

many independent trees,

boosting builds

many dependent trees.

Each new tree attempts to correct the mistakes made by the previous trees.

Conceptually,

```
Tree 1

↓

Errors

↓

Tree 2

↓

Remaining Errors

↓

Tree 3

↓

Remaining Errors

↓

...

↓

Final Model
```

The final prediction combines the contributions of all trees.

---

# <font color='orange'> 3. Weak Learners </font>

Boosting is based on the concept of a **weak learner**.

A weak learner is a model that performs only slightly better than random guessing (for classification) or provides only a modest improvement over a simple baseline (for regression).

Typical weak learners include

* shallow decision trees,
* decision stumps (trees with only one split).

Individually,

these models are not very accurate.

Collectively,

they can become extremely powerful.

---

# <font color='orange'> 4. From Weak Learners to Strong Learners </font>

Boosting constructs

a strong predictive model

by combining

many weak learners.

```
Weak Tree

↓

Weak Tree

↓

Weak Tree

↓

...

↓

Strong Ensemble
```

Each learner contributes only a small improvement,

but the cumulative effect can be substantial.

---

# <font color='orange'> 5. Learning from Mistakes </font>

Suppose

the first tree predicts house prices.

Some houses are predicted accurately,

others are not.

Instead of ignoring these errors,

boosting asks

> **Where did the model perform poorly?**

The next tree is trained specifically to improve those predictions.

Each subsequent tree focuses on what the previous ensemble has not yet learned.

---

# <font color='orange'> 6. Sequential Learning </font>

Unlike bagging,

where every tree is trained independently,

boosting follows a sequential procedure.

```
Training Data

↓

Tree 1

↓

Errors

↓

Tree 2

↓

Errors

↓

Tree 3

↓

Errors

↓

...

↓

Final Prediction
```

Because each tree depends on the previous ones,

boosting cannot usually be parallelized as easily as bagging.

---

# <font color='orange'> 7. Residual Correction in Regression </font>

For regression,

boosting often focuses on the **residuals**.

Recall that the residual for observation \(i\) is

$$
\boxed{
e_i = y_i - \hat{y}_i.
}
$$

After the first tree is fitted,

the residuals represent the remaining prediction errors.

The next tree is trained to predict these residuals.

The ensemble prediction is then updated.

This process is repeated,

gradually reducing the overall prediction error.

This idea forms the basis of **Gradient Boosting Regression**.

---

# <font color='orange'> 8. AdaBoost: The First Major Boosting Algorithm </font>

One of the earliest successful boosting algorithms is **AdaBoost** (Adaptive Boosting).

Its central idea is simple.

After each learner is trained,

observations that were predicted poorly receive greater emphasis,

while well-predicted observations receive less emphasis.

The next learner therefore concentrates on the difficult cases.

AdaBoost is primarily associated with classification,

but its core idea of sequential improvement inspired many later boosting algorithms.

# <font color='blue'> Gradient Boosting Regression </font>

Random Forest Regression improves predictive performance by averaging the predictions of many independently trained decision trees. Although this substantially reduces variance, each tree is constructed without knowledge of the mistakes made by the others.

Gradient Boosting Regression adopts a fundamentally different strategy. Instead of building trees independently, it constructs them sequentially, with each new tree specifically designed to improve the predictions of the existing ensemble. The model begins with a simple initial prediction and then repeatedly adds new trees that correct the remaining prediction errors. Each additional tree makes a small improvement, gradually producing an increasingly accurate model.

This sequential refinement process allows Gradient Boosting to achieve exceptional predictive performance and forms the conceptual foundation of modern boosting algorithms such as XGBoost, LightGBM, and CatBoost.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict house prices.

A single Decision Tree produces

```
Prediction

↓

Errors
```

Random Forest averages

many trees

to reduce variance.

However,

suppose the predictions are still systematically too low for expensive houses and too high for inexpensive houses.

Rather than averaging additional independent trees,

Gradient Boosting asks

> **How can we directly correct the remaining mistakes?**

The answer is simple.

Train another tree

whose purpose is

to predict the errors made by the current model.

---

# <font color='orange'> 2. The Central Idea </font>

Instead of constructing

many independent trees,

Gradient Boosting builds

an additive sequence of trees.

```
Initial Model

↓

Tree 1

↓

Improved Model

↓

Tree 2

↓

Improved Model

↓

Tree 3

↓

...

↓

Final Model
```

Each tree contributes

a small correction

to the existing prediction.

---

# <font color='orange'> 3. Initial Prediction </font>

The algorithm begins with a very simple model.

For squared-error regression,

the optimal initial prediction is

the mean of the target values.

$$
\boxed{
F_0(x)
=
\bar y.
}
$$

Example

Suppose

```
House Prices

220

240

260

280
```

The initial prediction becomes

$$
\bar y
=
250.
$$

Every observation is initially predicted as

250.

---

# <font color='orange'> 4. Computing the Residuals </font>

The prediction errors are then computed.

The residual for observation

\(i\)

is

$$
\boxed{
r_i
=
y_i
-
\hat y_i.
}
$$

Suppose

| Actual | Predicted | Residual |
| :---: | :---: | :---: |
| 220 | 250 | -30 |
| 240 | 250 | -10 |
| 260 | 250 | 10 |
| 280 | 250 | 30 |

These residuals represent

the information

that the current model has not yet learned.

---

# <font color='orange'> 5. Fitting a Tree to the Residuals </font>

Instead of predicting

house prices,

the next tree predicts

the residuals.

```
Training Data

↓

Residuals

↓

Decision Tree
```

The tree therefore learns

how the existing model should be corrected.

---

# <font color='orange'> 6. Updating the Model </font>

Once the residual tree has been trained,

its predictions are added to the existing model.

Mathematically,

$$
\boxed{
F_1(x)
=
F_0(x)
+
h_1(x),
}
$$

where

* \(F_0(x)\) is the current model,
* \(h_1(x)\) is the first residual tree.

The prediction becomes

```
Old Prediction

+

Correction

=

Improved Prediction
```

This process repeats after every tree.

---

# <font color='orange'> 7. Iterative Learning </font>

The procedure continues iteratively.

```
Initial Model

↓

Residuals

↓

Tree 1

↓

Updated Model

↓

New Residuals

↓

Tree 2

↓

Updated Model

↓

...

↓

Final Ensemble
```

Each tree focuses only on

what the previous ensemble has not yet explained.

---

# <font color='orange'> 8. Learning Rate (Shrinkage) </font>

Rather than adding each tree completely,

Gradient Boosting usually scales its contribution.

The updated model becomes

$$
\boxed{
F_m(x)
=
F_{m-1}(x)
+
\eta
h_m(x),
}
$$

where

$$
0<\eta\le1
$$

is the

**learning rate**.

---

### Small Learning Rate

```
Slow Learning

↓

More Trees

↓

Often Better Generalization
```

---

### Large Learning Rate

```
Fast Learning

↓

Fewer Trees

↓

Greater Risk of Overfitting
```

The learning rate is one of the most important hyperparameters in Gradient Boosting.

---

# <font color='orange'> 9. Why is it Called "Gradient" Boosting? </font>

The previous discussion focused on residuals.

More generally,

Gradient Boosting minimizes

a chosen loss function.

At each iteration,

the algorithm fits the new tree to

the **negative gradient** of the loss function with respect to the current predictions.

For squared-error loss,

the negative gradient is simply

the residuals.

Thus,

ordinary residual fitting

is a special case

of a more general optimization procedure.

This interpretation explains the name

**Gradient Boosting**.

# <font color='blue'> Gradient Boosting as Gradient Descent </font>

The previous chapter introduced Gradient Boosting as a procedure that sequentially fits decision trees to the residuals of the current model. While this provides an intuitive understanding of the algorithm, it does not explain why fitting residuals is mathematically justified.

The key insight is that Gradient Boosting is an optimization algorithm. Rather than directly minimizing a loss function with respect to a finite set of model parameters, Gradient Boosting performs **gradient descent in the space of functions**. At each iteration, a new decision tree is fitted to the negative gradient of the chosen loss function, thereby moving the model in the direction that most rapidly decreases the prediction error.

This elegant mathematical interpretation explains both the name **Gradient Boosting** and its ability to optimize a wide variety of loss functions.

---

# <font color='orange'> 1. The Optimization Problem </font>

Suppose we have

* training observations

$$
(x_i,y_i),
$$

* a predictive model

$$
F(x).
$$

Our objective is to minimize a loss function

$$
L(y,F(x)).
$$

The optimization problem can therefore be written as

$$
\boxed{
\min_F
\sum_{i=1}^{n}
L(y_i,F(x_i)).
}
$$

Unlike Linear Regression,

we are no longer optimizing

a finite number of coefficients.

Instead,

we are optimizing

an entire prediction function.

---

# <font color='orange'> 2. Review: Ordinary Gradient Descent </font>

Recall ordinary gradient descent.

Suppose

a function

$$
f(\theta)
$$

depends on a parameter

$$
\theta.
$$

Gradient descent updates the parameter according to

$$
\boxed{
\theta_{new}
=
\theta_{old}
-
\eta
\frac{\partial f}{\partial\theta},
}
$$

where

* \(\eta\) is the learning rate,
* the gradient points in the direction of greatest increase,
* the negative gradient points in the direction of greatest decrease.

This idea underlies much of modern machine learning.

---

# <font color='orange'> 3. What Changes in Gradient Boosting? </font>

In Gradient Boosting,

we are not adjusting

a few numerical parameters.

Instead,

we update

the prediction function itself.

Conceptually,

```
Current Model

↓

Compute Gradient

↓

Fit Decision Tree

↓

Update Model

↓

Repeat
```

The new tree acts as

a small correction

to the existing prediction function.

---

# <font color='orange'> 4. The General Update Rule </font>

Suppose

the current model is

$$
F_{m-1}(x).
$$

The updated model becomes

$$
\boxed{
F_m(x)
=
F_{m-1}(x)
+
\eta\,h_m(x),
}
$$

where

* \(h_m(x)\) is the new decision tree,
* \(\eta\) is the learning rate.

The question becomes

> **How should we choose \(h_m(x)\)?**

The answer is

the negative gradient.

---

# <font color='orange'> 5. The Negative Gradient </font>

For every observation,

the algorithm computes

$$
\boxed{
g_i
=
-
\left.
\frac{\partial
L(y_i,F(x_i))}
{\partial F(x_i)}
\right|_{F=F_{m-1}}.
}
$$

These values represent

the direction in which

the predictions should change

to reduce the loss most rapidly.

The new decision tree is trained

to predict

these negative gradients.

---

# <font color='orange'> 6. Why Do Residuals Appear? </font>

Consider the squared-error loss

$$
L
=
\frac12
(y-\hat y)^2.
$$

Differentiating with respect to the prediction

\(\hat y\)

gives

$$
\frac{\partial L}{\partial\hat y}
=
-(y-\hat y).
$$

Therefore,

the negative gradient is

$$
\boxed{
-
\frac{\partial L}{\partial\hat y}
=
y-\hat y.
}
$$

But

$$
y-\hat y
$$

is simply

the residual.

Therefore,

for squared-error regression,

```
Negative Gradient

=

Residual
```

This explains why Gradient Boosting fits residuals.

Residual fitting is not an approximation;

it is exactly gradient descent for the squared-error loss.

---

# <font color='orange'> 7. Beyond Squared Error </font>

The true power of Gradient Boosting becomes apparent when different loss functions are used.

Examples include

| Loss Function | Negative Gradient Represents |
| :--- | :--- |
| Squared Error | Residuals |
| Absolute Error | Sign of the residual |
| Huber Loss | Robust correction term |
| Logistic Loss | Classification gradient |
| Poisson Loss | Count-data gradient |

Thus,

Gradient Boosting is not limited to ordinary regression.

Any differentiable loss function can, in principle, be optimized using the same framework.

---

# <font color='orange'> 8. Function Space </font>

Traditional gradient descent updates

parameters.

Gradient Boosting updates

functions.

Conceptually,

```
Linear Regression

↓

Adjust Coefficients
```

↓

```
Gradient Boosting

↓

Add New Tree

↓

New Prediction Function
```

Each tree becomes

one additional component

of the prediction function.

---

# <font color='orange'> 9. The Complete Algorithm </font>

The Gradient Boosting procedure can now be summarized.

### Step 1

Initialize

$$
F_0(x).
$$

---

### Step 2

Compute the negative gradients.

---

### Step 3

Fit a decision tree to the gradients.

---

### Step 4

Multiply the tree by the learning rate.

---

### Step 5

Update the model.

---

### Step 6

Repeat until

the desired number of trees

has been constructed.

Each iteration moves the model closer to the minimum of the chosen loss function.

---

# <font color='orange'> 10. Why Is This Powerful? </font>

This mathematical framework makes Gradient Boosting extraordinarily flexible.

Instead of redesigning an entirely new algorithm for every prediction task,

we simply change

the loss function.

The optimization procedure remains the same.

This explains why Gradient Boosting can be applied to

* regression,
* classification,
* ranking,
* count prediction,
* survival analysis,

and many other problems.

# <font color='orange'> 11. Connection to XGBoost </font>

Classical Gradient Boosting uses

only the first derivative

(the gradient)

of the loss function.

XGBoost extends this idea by incorporating

the **second derivative**

(the Hessian).

```
Gradient Boosting

↓

Gradient
```

↓

```
XGBoost

↓

Gradient

+

Hessian
```

This second-order information allows XGBoost to optimize the objective more efficiently and accurately.
