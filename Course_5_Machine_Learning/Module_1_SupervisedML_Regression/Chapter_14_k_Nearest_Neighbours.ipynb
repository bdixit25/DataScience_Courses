# <font color='blue'> k-Nearest Neighbours (k-NN) Regression </font>

The previous chapter introduced Instance-Based Learning, a learning paradigm in which predictions are made by comparing new observations with previously observed examples rather than by fitting an explicit mathematical model. The most widely used instance-based regression algorithm is **k-Nearest Neighbours (k-NN) Regression**.

The fundamental assumption underlying k-NN Regression is that observations with similar predictor values tend to have similar target values. Consequently, the prediction for a new observation is obtained by identifying the **k** most similar training observations and combining their target values, typically through averaging or distance-weighted averaging.

Unlike linear regression or decision tree methods, k-NN is a **non-parametric** algorithm because it does not assume any specific functional relationship between the predictors and the response variable.

---

# <font color='orange'> 1. The Central Idea </font>

Suppose we wish to predict the price of a new house.

Instead of fitting a regression equation,

k-NN asks

> **Which previously sold houses are most similar to this one?**

The prices of those neighbouring houses are then used to estimate the price of the new house.

Conceptually,

```
New House

↓

Find Similar Houses

↓

Combine Their Prices

↓

Prediction
```

---

# <font color='orange'> 2. The k-NN Algorithm </font>

The algorithm proceeds through the following steps.

### Step 1

Store the training dataset.

---

### Step 2

For a new observation,

compute its distance to every training observation.

---

### Step 3

Identify the

**k nearest neighbours**.

---

### Step 4

Combine their target values.

Most commonly,

their average is computed.

---

### Step 5

Return this value as the prediction.

---

# <font color='orange'> 3. A Simple Example </font>

Suppose we wish to predict the price of a new house.

The three nearest neighbours have prices

| Neighbour | Price (£1000s) |
| :--- | ---: |
| 1 | 210 |
| 2 | 220 |
| 3 | 230 |

Using

$$
k=3,
$$

the predicted price is

$$
\boxed{
\hat y
=
\frac{210+220+230}{3}
=
220.
}
$$

The prediction is simply the average of the neighbouring target values.

---

# <font color='orange'> 4. Choosing the Value of \(k\) </font>

The choice of

$$
k
$$

is the most important hyperparameter.

---

### Small \(k\)

Example

$$
k=1.
$$

Prediction depends on only one neighbour.

Advantages

* captures local structure,
* highly flexible.

Disadvantages

* sensitive to noise,
* high variance.

---

### Large \(k\)

Example

$$
k=50.
$$

Prediction averages many neighbours.

Advantages

* smoother predictions,
* reduced variance.

Disadvantages

* may overlook local patterns,
* higher bias.

Thus,

\(k\)

controls the familiar Bias–Variance Trade-off.

---

# <font color='orange'> 5. Uniform vs Distance-Weighted Averaging </font>

### Uniform Weighting

Every neighbour contributes equally.

$$
\boxed{
\hat y
=
\frac1k
\sum_{i=1}^{k}
y_i.
}
$$

---

### Distance Weighting

Closer neighbours receive greater weight.

A common weighting scheme is

$$
\boxed{
w_i
=
\frac1{d_i},
}
$$

where

$$
d_i
$$

is the distance to neighbour

\(i\).

The prediction becomes

$$
\boxed{
\hat y
=
\frac{
\sum_i
w_i
y_i
}{
\sum_i
w_i
}.
}
$$

Distance weighting often improves prediction accuracy when nearby observations are substantially more informative than distant ones.

---

# <font color='orange'> 6. Local Regression </font>

Unlike Linear Regression,

which fits

one global model,

k-NN constructs

a local prediction

for every new observation.

```
Entire Dataset

↓

New Observation

↓

Local Neighbourhood

↓

Prediction
```

Different regions of the feature space therefore have different effective regression models.

---

# <font color='orange'> 7. Feature Scaling </font>

Because neighbour selection depends on distances,

feature scaling is essential.

Example

| Feature | Range |
| :--- | :---: |
| Age | 18–70 |
| Income (£) | 20,000–200,000 |

Without scaling,

income dominates the distance calculation,

causing age to contribute very little.

Standardization is therefore strongly recommended before applying k-NN.

---

# <font color='orange'> 8. Computational Complexity </font>

Unlike eager learners,

k-NN performs almost no computation during training.

### Training

```
Store Data
```

Very fast.

---

### Prediction

```
Compute Distances

↓

Find Neighbours

↓

Average Targets
```

Prediction becomes increasingly expensive as the training dataset grows.

This computational cost is one of the principal limitations of k-NN.

---

# <font color='orange'> 9. Hyperparameters </font>

Important k-NN hyperparameters include

| Hyperparameter | Purpose |
| :--- | :--- |
| `n_neighbors` | Number of neighbours (\(k\)) |
| `weights` | Uniform or distance weighting |
| `metric` | Distance metric (Euclidean, Manhattan, etc.) |
| `algorithm` | Search method (`auto`, `kd_tree`, `ball_tree`, `brute`) |

The distance metric will be discussed in the next chapter.

# <font color='orange'> 10. Practical Implementation </font>

Using Scikit-Learn,

# <font color='orange'> 11. Advantages </font>

k-NN Regression offers several attractive properties.

* Extremely simple algorithm.
* No assumptions about the functional form of the data.
* Naturally models nonlinear relationships.
* Minimal training time.
* Easy to interpret conceptually.

---

# <font color='orange'> 12. Limitations </font>

k-NN also has important limitations.

* Prediction becomes slow for large datasets.
* Requires feature scaling.
* Sensitive to irrelevant predictors.
* Suffers from the curse of dimensionality.
* Memory intensive because all training observations must be stored.

---

# <font color='orange'> 13. When Does k-NN Work Well? </font>

k-NN is often effective when

* datasets are relatively small to medium in size,
* observations form meaningful local clusters,
* predictor variables have been properly scaled,
* the number of features is moderate.

It is generally less suitable for very high-dimensional datasets or problems requiring rapid prediction.

# <font color='blue'> Distance Metrics for k-Nearest Neighbours </font>

The k-Nearest Neighbours algorithm makes predictions by identifying observations that are most similar to a given query point. Consequently, the definition of **similarity** plays a central role in the behaviour of the algorithm. In practice, similarity is usually quantified using a **distance metric**, which assigns a numerical measure of separation between two observations.

Different distance metrics emphasize different aspects of the data and may lead to different nearest neighbours, even when applied to the same dataset. The choice of distance metric therefore has a direct influence on the predictive performance of k-NN Regression.

This chapter introduces the most commonly used distance metrics in machine learning, discusses their mathematical definitions, and explains when each metric is most appropriate.

---

# <font color='orange'> 1. Why Do Distance Metrics Matter? </font>

Suppose we wish to predict the price of a house.

The k-NN algorithm asks

> **Which houses are closest to the new house?**

The answer depends entirely on

how "closest"

is defined.

Different distance metrics may identify different neighbours,

leading to different predictions.

Thus,

the distance metric is a fundamental component of the k-NN algorithm.

---

# <font color='orange'> 2. Euclidean Distance </font>

The most common distance metric is the **Euclidean distance**.

For two observations

$$
\mathbf{x}
=
(x_1,\ldots,x_p)
$$

and

$$
\mathbf{z}
=
(z_1,\ldots,z_p),
$$

the Euclidean distance is

$$
\boxed{
d(\mathbf{x},\mathbf{z})
=
\sqrt{
\sum_{i=1}^{p}
(x_i-z_i)^2
}.
}
$$

This is simply the ordinary straight-line distance between two points.

---

### Example

Suppose

$$
\mathbf{x}
=
(2,3),
$$

$$
\mathbf{z}
=
(5,7).
$$

Then

$$
d
=
\sqrt{(5-2)^2+(7-3)^2}
=
\sqrt{9+16}
=
5.
$$

---

# <font color='orange'> 3. Manhattan Distance </font>

The **Manhattan distance** measures the total distance travelled along the coordinate axes.

It is defined as

$$
\boxed{
d(\mathbf{x},\mathbf{z})
=
\sum_{i=1}^{p}
|x_i-z_i|.
}
$$

Unlike Euclidean distance,

no squaring occurs.

---

### Example

For

$$
(2,3)
$$

and

$$
(5,7),
$$

the Manhattan distance is

$$
3+4=7.
$$

The name originates from travelling along the rectangular street grid of Manhattan rather than moving directly "as the crow flies."

---

# <font color='orange'> 4. Minkowski Distance </font>

Euclidean and Manhattan distances are both special cases of the **Minkowski distance**.

It is defined as

$$
\boxed{
d(\mathbf{x},\mathbf{z})
=
\left(
\sum_{i=1}^{p}
|x_i-z_i|^q
\right)^{1/q},
}
$$

where

$$
q\ge1.
$$

Special cases include

| \(q\) | Distance |
| :---: | :--- |
| 1 | Manhattan |
| 2 | Euclidean |
| ∞ | Chebyshev |

Thus,

the Minkowski distance provides a unified mathematical framework for several commonly used distance metrics.

---

# <font color='orange'> 5. Chebyshev Distance </font>

The **Chebyshev distance** measures only the largest coordinate difference.

It is defined as

$$
\boxed{
d(\mathbf{x},\mathbf{z})
=
\max_i
|x_i-z_i|.
}
$$

Example

For

$$
(2,3)
$$

and

$$
(5,7),
$$

the coordinate differences are

```
3

4
```

The Chebyshev distance is therefore

$$
4.
$$

This metric is useful when the largest deviation dominates the notion of similarity.

---

# <font color='orange'> 6. Cosine Similarity and Cosine Distance </font>

Unlike the previous metrics,

Cosine Similarity measures the angle between two vectors rather than the physical distance separating them.

It is defined as

$$
\boxed{
\text{Cosine Similarity}
=
\frac{
\mathbf{x}^T\mathbf{z}
}{
\|\mathbf{x}\|
\|\mathbf{z}\|
}.
}
$$

Values range from

* +1 → identical direction,
* 0 → orthogonal,
* −1 → opposite direction.

Cosine Distance is commonly defined as

$$
\boxed{
1-\text{Cosine Similarity}.
}
$$

Cosine-based measures are widely used in

* text mining,
* document retrieval,
* recommendation systems,
* natural language processing.

---

# <font color='orange'> 7. Mahalanobis Distance </font>

The previous distance metrics treat each feature independently.

However,

features are often correlated.

The **Mahalanobis distance** accounts for both

* feature scaling,
* correlations between features.

It is defined as

$$
\boxed{
d(\mathbf{x},\mathbf{z})
=
\sqrt{
(\mathbf{x}-\mathbf{z})^T
\Sigma^{-1}
(\mathbf{x}-\mathbf{z})
},
}
$$

where

$$
\Sigma
$$

is the covariance matrix of the predictors.

Mahalanobis distance is particularly useful when

* features are strongly correlated,
* variables have different variances,
* multivariate statistical structure is important.

---

# <font color='orange'> 8. Feature Scaling Revisited </font>

Distance-based algorithms are highly sensitive to the scale of the predictors.

Suppose

| Feature | Range |
| :--- | :---: |
| Height (m) | 1.5–2.0 |
| Income (£) | 20,000–200,000 |

Without scaling,

income dominates the distance calculation.

Therefore,

standardization or normalization is generally required before applying

* k-NN,
* K-Means,
* SVM,
* PCA (when based on covariance scaling considerations).

---

# <font color='orange'> 9. Choosing a Distance Metric </font>

The appropriate metric depends on the application.

| Metric | Typical Use |
| :--- | :--- |
| Euclidean | Continuous numerical features |
| Manhattan | Robust when coordinate-wise differences are important |
| Minkowski | General-purpose family of metrics |
| Chebyshev | Largest coordinate difference dominates |
| Cosine | Text data, embeddings, high-dimensional sparse vectors |
| Mahalanobis | Correlated numerical features |

There is no universally optimal choice.

The best metric is typically selected using cross-validation.

---

# <font color='orange'> 10. Practical Implementation </font>

In Scikit-Learn,

different metrics are selected using the

```python
metric
```

parameter.

Example

```python
KNeighborsRegressor(
    n_neighbors=5,
    metric="euclidean"
)
```

Other common choices include

```python
metric="manhattan"

metric="minkowski"

metric="chebyshev"

metric="cosine"
```

For Minkowski distance,

the exponent is specified using

```python
p
```

For example,

```python
metric="minkowski",
p=1
```

produces Manhattan distance,

while

```python
p=2
```

produces Euclidean distance.

---

# <font color='orange'> 11. Advantages and Limitations </font>

Different distance metrics emphasize different notions of similarity.

Choosing an inappropriate metric may

* identify poor neighbours,
* reduce predictive accuracy,
* increase sensitivity to irrelevant features.

Consequently,

the distance metric should be regarded as an important hyperparameter rather than a fixed design choice.

---

# <font color='purple'> 12. Conceptual Summary </font>

| Metric | Formula | Typical Application |
| :--- | :--- | :--- |
| Euclidean | Straight-line distance | General numerical data |
| Manhattan | Sum of absolute differences | Grid-like movement, robust regression |
| Minkowski | Generalized distance family | Flexible framework |
| Chebyshev | Maximum coordinate difference | Worst-case deviation |
| Cosine | Angle between vectors | Text, embeddings, recommender systems |
| Mahalanobis | Covariance-adjusted distance | Correlated numerical features |

> **Key Insight:** Distance metrics define the notion of similarity upon which instance-based learning algorithms rely. Different metrics capture different geometric relationships between observations, leading to different nearest neighbours and, consequently, different predictions. The choice of distance metric, together with appropriate feature scaling, is therefore a critical component of successful applications of k-Nearest Neighbours and many other distance-based machine learning algorithms.

# <font color='blue'> Hyperparameter Tuning for k-Nearest Neighbours (k-NN) Regression </font>

The predictive performance of k-Nearest Neighbours Regression depends strongly on several design choices, including the number of neighbours, the distance metric, the weighting scheme, and the preprocessing applied to the predictor variables. Unlike many parametric models, k-NN does not estimate model coefficients during training. Instead, its behaviour is determined almost entirely by these hyperparameters.

Selecting appropriate hyperparameters is therefore essential for achieving good predictive performance and avoiding both underfitting and overfitting. This chapter discusses the most important hyperparameters of k-NN Regression and presents practical strategies for optimizing them using cross-validation, Grid Search, and Random Search.

---

# <font color='orange'> 1. Why Hyperparameter Tuning Matters </font>

Unlike Linear Regression,

k-NN does not automatically determine the complexity of the model.

Instead,

the user must specify

* the number of neighbours,
* the distance metric,
* the weighting scheme,
* the neighbour search algorithm.

Poor choices can produce

* unstable predictions,
* oversmoothed predictions,
* unnecessary computation.

Systematic tuning is therefore essential.

---

# <font color='orange'> 2. Choosing the Number of Neighbours (\(k\)) </font>

The most influential hyperparameter is

$$
k.
$$

Recall

### Small \(k\)

```
Highly Flexible

↓

Low Bias

↓

High Variance
```

The model may overfit the training data.

---

### Large \(k\)

```
Smooth Predictions

↓

Higher Bias

↓

Lower Variance
```

The model may overlook important local structure.

---

Typical candidate values include

```
1

3

5

7

9

11

15

21
```

The optimal value should be selected using cross-validation rather than intuition alone.

---

# <font color='orange'> 3. Choosing the Distance Metric </font>

Different distance metrics define different neighbourhoods.

Common choices include

| Metric | Typical Use |
| :--- | :--- |
| Euclidean | General numerical data |
| Manhattan | Robust to large coordinate differences |
| Minkowski | Flexible generalization |
| Chebyshev | Maximum coordinate difference |
| Cosine | High-dimensional sparse data |

The appropriate metric depends on the characteristics of the dataset.

It should therefore be treated as another hyperparameter.

---

# <font color='orange'> 4. Uniform vs Distance Weighting </font>

Scikit-Learn provides two common weighting schemes.

### Uniform

Every neighbour contributes equally.

```
Prediction

↓

Average
```

---

### Distance

Closer neighbours receive larger weights.

```
Closest Neighbours

↓

Larger Influence

↓

Prediction
```

Distance weighting often performs better when nearby observations are substantially more informative than distant ones.

---

# <font color='orange'> 5. Feature Scaling </font>

Because neighbour selection depends on distances,

feature scaling is essential.

The recommended workflow is

```
Original Data

↓

StandardScaler

↓

k-NN Regression
```

Using a Scikit-Learn `Pipeline` ensures that scaling is performed correctly within each cross-validation fold, preventing data leakage.

---

# <font color='orange'> 6. Cross-Validation </font>

Hyperparameters should be evaluated using unseen validation data.

The standard approach is

```
Training Data

↓

K-Fold Cross-Validation

↓

Average Validation Error

↓

Select Best Hyperparameters
```

Common choices include

* 5-fold cross-validation,
* 10-fold cross-validation.

---

# <font color='orange'> 7. Grid Search </font>

Grid Search evaluates every combination of candidate hyperparameters.

Example

| Hyperparameter | Candidate Values |
| :--- | :--- |
| `n_neighbors` | 3, 5, 7, 9, 11 |
| `weights` | uniform, distance |
| `metric` | euclidean, manhattan |

Every possible combination is evaluated.

Advantages

* Simple.
* Exhaustive.

Limitation

* Computational cost grows rapidly as the search space increases.

---

# <font color='orange'> 8. Random Search </font>

Random Search evaluates only a randomly selected subset of hyperparameter combinations.

Advantages include

* lower computational cost,
* efficient exploration of large search spaces,
* often comparable predictive performance.

Random Search is frequently preferred when many hyperparameters are being tuned.

---

# <font color='orange'> 9. Choosing the Search Algorithm </font>

Finding the nearest neighbours can itself be performed using different algorithms.

Scikit-Learn supports

| Algorithm | Description |
| :--- | :--- |
| `brute` | Computes all distances directly |
| `kd_tree` | Efficient for low-dimensional numerical data |
| `ball_tree` | Better suited to some higher-dimensional datasets |
| `auto` | Automatically selects an appropriate algorithm |

For small datasets,

the choice has little influence.

For larger datasets,

it can substantially affect prediction speed.

---

# <font color='orange'> 10. Practical Scikit-Learn Workflow </font>

A typical workflow combines

* feature scaling,
* k-NN,
* Grid Search.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsRegressor())
])

param_grid = {
    "knn__n_neighbors": [3, 5, 7, 9, 11],
    "knn__weights": ["uniform", "distance"],
    "knn__metric": ["euclidean", "manhattan"]
}

grid = GridSearchCV(
    pipeline,
    param_grid,
    cv=5,
    scoring="neg_root_mean_squared_error"
)

grid.fit(X_train, y_train)
```

The best model is obtained using

```python
grid.best_estimator_
```

while the optimal hyperparameters are available through

```python
grid.best_params_
```

---

# <font color='orange'> 11. Common Pitfalls </font>

Several issues frequently reduce k-NN performance.

* Forgetting to standardize the predictors.
* Choosing \(k=1\), resulting in overfitting.
* Choosing excessively large values of \(k\), producing oversmoothed predictions.
* Using an inappropriate distance metric.
* Including many irrelevant predictors.
* Applying k-NN to very high-dimensional datasets without prior dimensionality reduction.

These problems are often revealed during cross-validation.

---

# <font color='orange'> 12. Practical Recommendations </font>

For many regression problems,

the following workflow is effective.

1. Standardize the predictor variables.
2. Begin with Euclidean distance.
3. Compare uniform and distance weighting.
4. Tune \(k\) using cross-validation.
5. Evaluate alternative distance metrics if performance is unsatisfactory.
6. Compare the tuned k-NN model with stronger baselines such as Random Forest, Gradient Boosting, and SVR.

Empirical evaluation remains the most reliable method for selecting the best model.

By combining feature scaling, cross-validation, and systematic hyperparameter optimization within a pipeline, practitioners can develop robust and reproducible k-NN regression models that generalize effectively to unseen data.
