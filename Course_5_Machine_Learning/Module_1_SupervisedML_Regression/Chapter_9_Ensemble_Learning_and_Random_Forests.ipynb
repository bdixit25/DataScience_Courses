# <font color='blue'> Ensemble Learning: Combining Multiple Models </font>

One of the central ideas in modern machine learning is that combining the predictions of multiple models often produces better performance than relying on a single model. This principle forms the basis of **ensemble learning**, a family of techniques that construct multiple predictive models and combine their outputs to obtain more accurate, more robust, and more stable predictions.

The motivation behind ensemble learning is simple. Every individual model has strengths and weaknesses. A single decision tree, for example, may fit one particular training dataset extremely well but perform poorly on new observations because of its high variance. By combining many diverse models, individual errors tend to cancel one another, resulting in improved generalization.

Many of the most successful machine learning algorithms—including Random Forests, Gradient Boosting, XGBoost, LightGBM, and CatBoost—are ensemble methods.

---



# <font color='orange'> 1. Why Do We Need Ensemble Learning? </font>

Suppose we ask

one doctor

to diagnose a patient.

The diagnosis may be accurate,

but it is also influenced by that doctor's experience, judgement, and possible errors.

Now suppose

five independent specialists

examine the same patient.

Their collective opinion is usually more reliable than that of any single doctor.

Machine learning follows exactly the same principle.

Instead of relying on

one predictive model,

we combine

many models.

---



# <font color='orange'> 2. What is an Ensemble? </font>

An **ensemble** is a collection of multiple learning algorithms whose predictions are combined to produce a single final prediction.

Conceptually,

```
Model 1

↓

Prediction
```

```
Model 2

↓

Prediction
```

```
Model 3

↓

Prediction
```

```
...

↓

Combine Predictions

↓

Final Prediction
```

The individual models are often called **base learners** or **weak learners**.

---



# <font color='orange'> 3. Why Does It Work? </font>

Individual models make different errors.

If these errors are

* relatively independent,
* not systematically biased in the same direction,

then averaging or combining the predictions reduces the overall prediction error.

For regression,

the final prediction is often

the average

of the individual predictions.

For classification,

the final prediction is commonly

majority voting

or

averaged class probabilities.

---



# <font color='orange'> 4. Diversity is the Key </font>

An ensemble is most effective when its individual models are **diverse**.

Suppose we build

100 identical decision trees

using exactly the same training data.

```
Tree 1

↓

240
```

```
Tree 2

↓

240
```

```
Tree 3

↓

240
```

Averaging identical predictions provides no benefit.

Instead,

we want

different trees

that make

slightly different mistakes.

The diversity of the models allows many individual errors to cancel when combined.

---



# <font color='orange'> 5. Two Major Families of Ensemble Methods </font>

Most ensemble algorithms belong to one of two categories.

---

## Bagging (Bootstrap Aggregating)

Models are trained

**independently**

and

their predictions are combined.

```
Training Data

↓

Bootstrap Samples

↓

Tree 1

Tree 2

Tree 3

↓

Average
```

Examples

* Random Forest
* Extra Trees

---

## Boosting

Models are trained

**sequentially**.

Each new model attempts to correct the mistakes made by the previous models.

```
Model 1

↓

Errors

↓

Model 2

↓

Errors

↓

Model 3

↓

Final Model
```

Examples

* AdaBoost
* Gradient Boosting
* XGBoost
* LightGBM
* CatBoost

---



# <font color='orange'> 6. Bagging </font>

Bagging stands for

**Bootstrap Aggregating**.

The procedure is

1. Generate many bootstrap samples from the original dataset.
2. Train one model on each bootstrap sample.
3. Combine the predictions.

For regression,

the predictions are averaged.

Bagging primarily reduces

**variance**.

Decision Trees benefit greatly from bagging because they are naturally high-variance models.

---



# <font color='orange'> 7. Boosting </font>

Boosting follows a different philosophy.

Instead of training models independently,

each model learns from the errors made by its predecessors.

```
Weak Model

↓

Identify Errors

↓

Improve Model

↓

Repeat
```

Each successive model focuses increasingly on observations that were difficult to predict.

Boosting primarily reduces

**bias**,

although it can also reduce variance depending on the algorithm.

---



# <font color='orange'> 8. Bagging vs Boosting </font>

| Property | Bagging | Boosting |
| :--- | :---: | :---: |
| Training | Independent | Sequential |
| Main Objective | Reduce variance | Reduce bias (and often improve overall accuracy) |
| Parallel Training | ✔ | ✘ |
| Sensitive to Noise | Lower | Higher |
| Typical Base Learner | Decision Tree | Shallow Decision Tree |
| Examples | Random Forest | Gradient Boosting, XGBoost |

Both methods improve predictive performance,

but they achieve this in fundamentally different ways.

---



# <font color='orange'> 9. Ensemble Learning and the Bias–Variance Trade-off </font>

### Boosting

# <font color='blue'> Random Forest Regression </font>

Decision Tree Regression is a powerful and flexible algorithm capable of modelling highly nonlinear relationships. However, individual decision trees suffer from an important limitation: they are highly sensitive to the particular training dataset. Small changes in the data can produce substantially different trees, leading to high variance and poor generalization.

Random Forest Regression addresses this problem by combining a large collection of decision trees into a single ensemble model. Rather than relying on the prediction of one tree, Random Forest trains many diverse trees using different subsets of the data and predictor variables, then averages their predictions. This simple yet effective strategy substantially reduces variance while retaining the flexibility of individual decision trees.

Random Forest is one of the most widely used machine learning algorithms because it typically achieves excellent predictive performance with relatively little hyperparameter tuning.

---



# <font color='orange'> 1. Motivation </font>

Suppose we train a single Decision Tree on a housing dataset.

```
Training Data

↓

Decision Tree

↓

Prediction
```

The prediction depends strongly on the exact training observations.

If we slightly modify the dataset,

the tree may change completely.

This instability leads to high variance.

Instead,

imagine training

```
500

different trees.
```

Each tree makes a slightly different prediction.

The final prediction is

the average

of all trees.

Individual errors tend to cancel,

producing much more stable predictions.

---



# <font color='orange'> 2. What is a Random Forest? </font>

A Random Forest is an ensemble of decision trees.

Each tree is trained

* on a different bootstrap sample,
* using a random subset of the available features at each split.

For regression,

the final prediction is

$$
\boxed{
\hat y
=
\frac1B
\sum_{b=1}^{B}
\hat y_b,
}
$$

where

* \(B\) is the number of trees,
* \(\hat y_b\) is the prediction of the \(b\)-th tree.

Averaging the predictions reduces variance and improves generalization.

---



# <font color='orange'> 3. Bootstrap Sampling </font>

The first source of randomness comes from

**bootstrap sampling**.

Instead of training every tree on the entire dataset,

each tree is trained on a bootstrap sample obtained by sampling the training data **with replacement**.

Suppose the original dataset contains

```
1000

observations.
```

Each tree receives

```
1000

samples
```

drawn randomly **with replacement**.

Consequently,

* some observations appear multiple times,
* some observations are omitted.

Each tree therefore sees a slightly different version of the data.

---



# <font color='orange'> 4. Random Feature Selection </font>

The second source of randomness occurs during tree construction.

Instead of considering

all predictors

at every split,

the algorithm considers only

a random subset.

Example

Suppose

```
20

predictors
```

are available.

At each split,

the algorithm may consider only

```
5

randomly selected predictors.
```

This encourages diversity among the trees.

Without this step,

many trees would become nearly identical.

---



# <font color='orange'> 5. Building the Forest </font>

The training procedure can be summarized as follows.

```
Training Data

↓

Bootstrap Sample 1

↓

Tree 1
```

```
Training Data

↓

Bootstrap Sample 2

↓

Tree 2
```

```
Training Data

↓

Bootstrap Sample 3

↓

Tree 3
```

```
...

↓

Average Predictions
```

Each tree is grown independently,

making Random Forest highly parallelizable.

---



# <font color='orange'> 6. Why Averaging Works </font>

Suppose three trees predict

```
240

245

255
```

The Random Forest prediction is

$$
\frac{240+245+255}{3}
=
246.7.
$$

The averaging process reduces the effect of unusually high or low predictions made by individual trees.

If the trees make different errors,

their average is often much more accurate than any single prediction.

---



# <font color='orange'> 7. Out-of-Bag (OOB) Samples </font>

Because bootstrap sampling is performed with replacement,

each tree does **not** use every training observation.

The observations omitted from a tree's bootstrap sample are called **Out-of-Bag (OOB) samples**.

These samples can be used as an internal validation set.

The resulting **Out-of-Bag error** provides an estimate of the model's generalization performance without requiring a separate validation dataset.

Although OOB estimates are often very useful, many practitioners still use cross-validation when comparing different models or tuning hyperparameters.

---



# <font color='orange'> 8. Feature Importance </font>

Random Forest can estimate the relative importance of each predictor.

Two commonly used approaches are

### Mean Decrease in Impurity (MDI)

Measures

how much

each feature reduces impurity

across all trees.

---

### Permutation Importance

Randomly shuffles one feature

and measures

how much

the prediction accuracy decreases.

Features whose randomization substantially worsens performance are considered more important.

Permutation importance is generally more reliable because it is less affected by biases that can arise in impurity-based importance measures.

---

Several hyperparameters influence Random Forest performance.

| Hyperparameter | Purpose |
| :--- | :--- |

| `max_depth` | Maximum depth of each tree |
| `min_samples_split` | Minimum samples required to split a node |
| `min_samples_leaf` | Minimum samples in each leaf |
| `max_features` | Number of candidate features considered at each split |
| `bootstrap` | Whether bootstrap sampling is used |

Among these,

`n_estimators`

and

`max_features`

often have the greatest influence on performance.

# <font color='blue'> Extra Trees Regression (Extremely Randomized Trees) </font>

Random Forest Regression improves the stability and predictive performance of individual decision trees by combining many independently trained trees using bootstrap sampling and random feature selection. Although this approach substantially reduces variance, each decision tree still searches for the best split among the randomly selected features.

Extra Trees Regression (Extremely Randomized Trees) introduces an additional level of randomness. Instead of searching for the optimal split threshold, Extra Trees chooses split thresholds at random and then selects the best among these randomly generated candidates. This additional randomization increases the diversity of the individual trees, often leading to further reductions in variance and, in many cases, improved generalization.

Extra Trees therefore represents another important ensemble method based on bagging, differing from Random Forest primarily in how the individual trees are constructed.

---



# <font color='orange'> 1. Motivation </font>

Recall how Random Forest builds its trees.

For each split,

it

* selects a random subset of features,
* searches for the **best split** among those features.

Although the trees differ because of bootstrap sampling and random feature selection,

each split is still chosen optimally.

Extra Trees asks an interesting question:

> **What if we also randomize the split locations?**

Surprisingly,

adding more randomness often improves predictive performance.

---



# <font color='orange'> 2. From Decision Trees to Extra Trees </font>

The progression is

```
Decision Tree

↓

One Tree
```

↓

```
Random Forest

↓

Many Trees

↓

Bootstrap Sampling

↓

Random Features
```

↓

```
Extra Trees

↓

Many Trees

↓

Random Features

↓

Random Split Thresholds
```

Every stage increases model diversity.

---



# <font color='orange'> 3. How Does Extra Trees Work? </font>

For each split,

the algorithm performs the following steps.

1. Select a random subset of predictors.
2. Generate random split thresholds for each selected predictor.
3. Evaluate the candidate splits.
4. Choose the best of these randomly generated splits.
5. Continue recursively.

Unlike Random Forest,

the algorithm does **not** search exhaustively for the globally best split threshold.

---



# <font color='orange'> 4. Random Forest vs Extra Trees </font>

Suppose

```
Area

=

80

100

120

140

160
```

---

### Random Forest

Evaluates many possible split points

```
90

110

130

150
```

and selects

---

### Extra Trees

Generates a few random split points

```
95

145
```

and selects the better of those.

The split search is therefore much more random.

---



# <font color='orange'> 5. Bootstrap Sampling </font>

Another important distinction concerns the training data.

### Random Forest

Uses

bootstrap samples

Each tree sees a different dataset.

---

### Extra Trees

Typically uses

the **entire training dataset**

for every tree,

while introducing randomness through the split selection process.

Some implementations also allow bootstrap sampling, but the standard Extra Trees algorithm does not require it.

---



# <font color='orange'> 6. Why Does More Randomness Help? </font>

The additional randomness increases the diversity of the trees.

```
Tree 1

↓

Prediction
```

```
Tree 2

↓

Prediction
```

```
Tree 3

↓

Prediction
```

↓

```
Average
```

If the individual trees are less correlated,

their errors cancel more effectively when averaged.

This often produces lower variance than Random Forest.

---
