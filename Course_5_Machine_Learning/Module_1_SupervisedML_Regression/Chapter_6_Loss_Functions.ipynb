# <font color='blue'> Loss Functions for Regression </font>

Every supervised learning algorithm is designed to make predictions that are as close as possible to the true target values. To determine how well a model performs, we require a quantitative measure of prediction error. This measure is known as a **loss function**.

A loss function assigns a numerical value to the discrepancy between the observed target values and the model's predictions. During training, the objective of the learning algorithm is to find the model parameters that minimize this loss.

Different regression algorithms employ different loss functions depending on the desired balance between prediction accuracy, robustness to outliers, computational efficiency, and statistical properties. Consequently, understanding loss functions is fundamental to understanding modern regression algorithms.

---

# <font color='orange'> 1. Why Do We Need a Loss Function? </font>

Suppose a regression model predicts the prices of five houses.

| Actual (£) | Predicted (£) |
| :---: | :---: |
| 250,000 | 248,000 |
| 320,000 | 325,000 |
| 410,000 | 402,000 |
| 500,000 | 510,000 |
| 620,000 | 615,000 |

Some predictions are very close,

others are less accurate.

How should we quantify

the overall prediction error?

A loss function provides a mathematical answer to this question.

Without a loss function,

there is no objective way to determine which model is better.

---

# <font color='orange'> 2. Prediction Errors and Residuals </font>

For each observation,

the prediction error (or **residual** in the training data) is

$$
\boxed{
e_i = y_i-\hat y_i,
}
$$

where

* $y_i$ is the observed value,
* $\hat y_i$ is the predicted value.

If

$$
e_i=0,
$$

the prediction is perfect.

The loss function combines these individual errors into a single numerical quantity.

---

# <font color='orange'> 3. What is a Loss Function? </font>

A regression loss function is a mathematical function

$$
L(y,\hat y)
$$

that measures the disagreement between

the observed target

and

the predicted target.

The learning algorithm searches for the model parameters that minimize the total loss.

```
Training Data

↓

Predictions

↓

Compute Loss

↓

Adjust Parameters

↓

Repeat
```

This iterative optimization process lies at the heart of nearly every machine learning algorithm.

---

# <font color='orange'> 4. Mean Squared Error (MSE) </font>

The most common regression loss is the **Mean Squared Error**.

It is defined as

$$
\boxed{
MSE
=
\frac1n
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

---

### Why Square the Errors?

Squaring provides several advantages.

* Positive and negative errors cannot cancel.
* Larger errors receive greater weight.
* The function is smooth and differentiable.
* Efficient optimization is possible.

Because of these properties,

Linear Regression uses MSE through the Ordinary Least Squares (OLS) criterion.

---

### Example

Residuals

```
2

−1

3
```

Squared residuals

```
4

1

9
```

MSE

$$
=
\frac{4+1+9}{3}
=
4.67.
$$

---

# <font color='orange'> 5. Residual Sum of Squares (RSS) </font>

OLS actually minimizes the

**Residual Sum of Squares**

$$
\boxed{
RSS
=
\sum_{i=1}^{n}
(y_i-\hat y_i)^2.
}
$$

Notice that

$$
MSE
=
\frac{RSS}{n}.
$$

Since

$n$

is constant,

minimizing RSS

is equivalent to minimizing MSE.

---

# <font color='orange'> 6. Mean Absolute Error (MAE) </font>

Another common loss is the

Mean Absolute Error.

$$
\boxed{
MAE
=
\frac1n
\sum_{i=1}^{n}
|y_i-\hat y_i|.
}
$$

---

### Advantages

* Easy to interpret.
* Less sensitive to outliers.
* Measures the average prediction error directly.

---

### Limitations

* Not differentiable at zero.
* Harder to optimize analytically.

---

# <font color='orange'> 7. Comparing MSE and MAE </font>

Consider

```
Residuals

1

2

10
```

MAE

$$
=
\frac{1+2+10}{3}
=
4.33
$$

MSE

$$
=
\frac{1+4+100}{3}
=
35.
$$

Notice how

MSE penalizes

the large error

much more strongly.

This makes MSE more sensitive to outliers.

---

# <font color='orange'> 8. Huber Loss </font>

Huber Loss combines the strengths of

MSE

and

MAE.

For small errors,

it behaves like MSE.

For large errors,

it behaves like MAE.

Mathematically,

$$
L_\delta(e)=
\begin{cases}
\frac12e^2,&|e|\le\delta,\\
\delta\left(|e|-\frac12\delta\right),&|e|>\delta.
\end{cases}
$$

---

### Advantages

* Robust to outliers.
* Smooth and differentiable.
* Widely used in robust regression.

---

# <font color='orange'> 9. Quantile Loss </font>

Sometimes,

we wish to estimate

a particular percentile

rather than the mean.

For example,

predicting

the 90th percentile

of electricity demand.

Quantile Loss is designed for this purpose.

It forms the basis of

**Quantile Regression**.

---

# <font color='orange'> 10. Log-Cosh Loss </font>

Another smooth alternative is

Log-Cosh Loss

$$
L(e)
=
\log(\cosh(e)).
$$

Characteristics

* Similar to MSE for small errors.
* Similar to MAE for large errors.
* Smooth everywhere.

It is occasionally used in deep learning.

---

# <font color='orange'> 11. Which Algorithms Use Which Loss? </font>

| Algorithm | Primary Loss Function |
| :--- | :--- |
| Linear Regression | RSS / MSE |
| Ridge Regression | MSE + L2 Penalty |
| LASSO | MSE + L1 Penalty |
| Elastic Net | MSE + L1 + L2 Penalties |
| Huber Regressor | Huber Loss |
| Quantile Regression | Quantile Loss |
| Support Vector Regression | ε-Insensitive Loss |
| Gradient Boosting | User-selectable (e.g., squared error, absolute error, Huber) |
| XGBoost | Configurable objective (commonly squared error for regression) |
| LightGBM | Configurable objective |
| CatBoost | Multiple regression losses |

Different regression algorithms therefore differ not only in their model structure,

but also in the objective function they optimize.

---

# <font color='orange'> 12. Choosing a Loss Function </font>

The appropriate loss function depends on the problem.

| Situation | Recommended Loss |
| :--- | :--- |
| Standard Regression | MSE |
| Outliers Present | MAE or Huber |
| Need Robustness | Huber |
| Predict Quantiles | Quantile Loss |
| Smooth Optimization | MSE or Log-Cosh |

The loss function should reflect the practical cost of prediction errors in the application.

---

# <font color='orange'> 13. Loss Functions vs Evaluation Metrics </font>

A common source of confusion is the distinction between

**loss functions**

and

**evaluation metrics**.

| Loss Function | Evaluation Metric |
| :--- | :--- |
| Optimized during training | Reported after training |
| Guides parameter updates | Measures model performance |
| MSE | RMSE |
| Huber Loss | MAE |
| Quantile Loss | R² |

Sometimes the same quantity,

such as MSE,

can serve as both a training loss and an evaluation metric,

but their roles are conceptually different.

---

# <font color='orange'> 14. Why This Matters </font>

Nearly every regression algorithm studied later in this notebook

can be understood as minimizing a particular objective function.

```
Linear Regression

↓

MSE
```

```
Ridge

↓

MSE + L2 Penalty
```

```
LASSO

↓

MSE + L1 Penalty
```

```
Elastic Net

↓

MSE + L1 + L2 Penalties
```

```
Support Vector Regression

↓

ε-Insensitive Loss
```

```
Gradient Boosting

↓

Sequentially Minimize Loss
```

Thus,

learning the loss function is often more important than memorizing the algorithm itself.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Residual | Difference between observed and predicted values |
| Loss Function | Quantifies prediction error during training |
| MSE | Squares residuals; emphasizes large errors |
| MAE | Uses absolute residuals; robust to outliers |
| Huber Loss | Hybrid of MSE and MAE |
| Quantile Loss | Predicts conditional quantiles |
| Objective | Minimize total loss over the training data |

> **Key Insight:** Loss functions provide the mathematical objective that every regression algorithm seeks to optimize. By quantifying the discrepancy between observed and predicted values, they determine how model parameters are updated during training. While Mean Squared Error remains the standard choice for many regression problems, alternative losses such as MAE, Huber, and Quantile Loss offer greater robustness or flexibility for specific applications. Understanding loss functions is therefore essential for understanding not only Linear Regression but also Regularization, Support Vector Regression, Gradient Boosting, and many modern machine learning algorithms.
