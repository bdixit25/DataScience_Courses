# <font color='blue'> Introduction to Probabilistic Regression </font>

The regression algorithms introduced throughout the previous chapters—including Linear Regression, Decision Trees, Random Forests, Gradient Boosting, Support Vector Regression, and k-Nearest Neighbours Regression—share a common objective: given a set of predictor variables, they estimate a single value for the target variable. Such models are often referred to as **deterministic regression models** because each input corresponds to one predicted output.

In many scientific, engineering, and industrial applications, however, obtaining only a single prediction is insufficient. Decision-making often depends not only on the predicted value itself but also on the confidence that can be placed in that prediction. For example, a weather forecast, a medical diagnosis, or the prediction of a physical measurement is substantially more informative when accompanied by an estimate of its uncertainty.

**Probabilistic regression** addresses this limitation by modelling uncertainty explicitly. Rather than predicting a single value, probabilistic models estimate a probability distribution over possible target values, allowing both the expected prediction and the uncertainty associated with that prediction to be quantified.

This chapter introduces the fundamental concepts of probabilistic regression and provides the motivation for Gaussian Process Regression, which will be developed in the following chapters.

---



# <font color='orange'> 1. Deterministic vs Probabilistic Prediction </font>

Most regression algorithms answer the question

> **What is the predicted value?**

For example,

```
Input

↓

Regression Model

↓

Predicted House Price

=

£420,000
```

Only one value is returned.

---

Probabilistic regression instead asks

> **What values are plausible, and how confident are we?**

The prediction becomes

```
Predicted Mean

=

£420,000

±

Prediction Uncertainty
```

The uncertainty is just as important as the prediction itself.

---

# <font color='orange'> 3. Sources of Uncertainty </font>

Prediction uncertainty arises from several sources.

### Measurement Noise

Experimental observations are never perfectly precise.

---

### Limited Training Data

Predictions far from the observed data are naturally more uncertain.

---

### Model Uncertainty

Different models may explain the observed data equally well.

---

### Intrinsic Randomness

Some systems are genuinely stochastic.

A good regression model should ideally distinguish between these different forms of uncertainty.

# <font color='blue'> Gaussian Processes: Distributions Over Functions </font>

The previous chapter introduced probabilistic regression and motivated the need for predictive uncertainty. Unlike deterministic regression methods, probabilistic models describe predictions using probability distributions rather than single numerical values. The natural question that follows is how such probability distributions can be constructed for regression functions.

Gaussian Processes provide an elegant answer. Instead of assigning probability distributions to a finite collection of model parameters, Gaussian Processes assign probability distributions directly to functions. Before observing any training data, infinitely many regression functions are considered possible, each associated with a probability determined by a mean function and a covariance function.

Although this perspective may initially appear abstract, it follows naturally from the familiar Gaussian distribution. This chapter develops the concept gradually, beginning with ordinary Gaussian random variables, extending to multivariate Gaussian distributions, and finally introducing Gaussian Processes as probability distributions over functions.

---



# <font color='orange'> 1. Review: The Gaussian Distribution </font>

Consider a random variable

$$
X.
$$

If

$$
X
\sim
\mathcal N(\mu,\sigma^2),
$$

then

* the mean is

$$
\mu,
$$

* the variance is

$$
\sigma^2.
$$

Graphically,

```
          /\
         /  \
       /      \
_____/__________\_____

        μ
```

The Gaussian distribution describes uncertainty about **one number**.

---



# <font color='orange'> 2. From One Variable to Many Variables </font>

Suppose we now have several random variables,

$$
X_1,\;
X_2,\;
X_3.
$$

Instead of modelling each variable separately,

we describe them jointly.

This leads to the **multivariate Gaussian distribution**.

Mathematically,

$$
\boxed{
\mathbf X
\sim
\mathcal N(\boldsymbol\mu,\Sigma),
}
$$

where

* \(\boldsymbol\mu\) is the mean vector,
* \(\Sigma\) is the covariance matrix.

---



# <font color='orange'> 3. The Covariance Matrix </font>

The covariance matrix describes

how the variables vary together.

For three variables,

$$
\Sigma
=
\begin{bmatrix}
\sigma_1^2 & \sigma_{12} & \sigma_{13}\\
\sigma_{21} & \sigma_2^2 & \sigma_{23}\\
\sigma_{31} & \sigma_{32} & \sigma_3^2
\end{bmatrix}.
$$

The diagonal entries contain

the variances,

while the off-diagonal entries measure

the correlations between variables.

---



# <font color='orange'> 4. From Vectors to Functions </font>

Now imagine

that instead of three variables,

we have

one random variable

for every possible input value

$$
x.
$$

Instead of

```
X₁

X₂

X₃
```

we now have

```
f(x₁)

f(x₂)

f(x₃)

...

f(x)
```

The collection becomes

an entire function.

Rather than assigning probabilities to individual numbers,

we assign probabilities to complete functions.

---



# <font color='orange'> 5. A Distribution Over Functions </font>

Before observing any data,

many regression functions appear plausible.

Conceptually,

```
Function 1

~~~~~~

Function 2

--------

Function 3

........

Function 4

_______
```

Each curve represents one possible explanation of the unknown relationship.

A Gaussian Process assigns probabilities to all of these functions.

After observing data,

unlikely functions become less probable,

while functions consistent with the observations become more probable.

---



# <font color='orange'> 6. Definition of a Gaussian Process </font>

A Gaussian Process is formally written as

$$
\boxed{
f(x)
\sim
\mathcal{GP}
\left(
m(x),
k(x,x')
\right),
}
$$

where

* \(m(x)\) is the **mean function**,
* \(k(x,x')\) is the **covariance function** (or kernel).

This notation is analogous to

$$
X
\sim
\mathcal N(\mu,\sigma^2),
$$

except that the Gaussian Process describes

functions

rather than individual variables.

---



# <font color='orange'> 7. The Mean Function </font>

The mean function specifies

the expected value of the function at every input.

It is defined as

$$
\boxed{
m(x)
=
\mathbb E[f(x)].
}
$$

In many practical applications,

the mean function is assumed to be

$$
m(x)=0,
$$

allowing the covariance function to capture the structure of the data.

---



# <font color='orange'> 8. The Covariance Function (Kernel) </font>

The covariance function determines

how similar two function values are.

It is defined as

$$
\boxed{
k(x,x')
=
\operatorname{Cov}
\left[
f(x),
f(x')
\right].
}
$$

If

two inputs are close,

their outputs are usually expected to be similar.

Consequently,

their covariance is large.

If

the inputs are far apart,

their covariance is generally smaller.

Thus,

the kernel determines the smoothness and structure of the possible functions.

---



# <font color='orange'> 9. Sampling Functions </font>

A remarkable property of Gaussian Processes is that

we can randomly sample complete functions.

Before observing data,

the sampled functions may vary substantially.

```
Function A

~~~~~~~

Function B

--------

Function C

........

Function D

________
```

After observing data,

the sampled functions become constrained to pass near the observed training points.

The uncertainty also decreases near the observations.

---



# <font color='orange'> 10. Prior and Posterior </font>

Before observing data,

the Gaussian Process describes our

**prior belief**

about plausible functions.

```
Prior

↓

Observe Data

↓

Posterior
```

After conditioning on the observations,

we obtain the

**posterior distribution**,

which contains only functions consistent with both

* the prior assumptions,
* the observed data.

This Bayesian updating process forms the basis of Gaussian Process Regression.

---



# <font color='orange'> 11. Why Is This Useful? </font>

Instead of producing

one regression curve,

Gaussian Processes produce

an entire probability distribution over curves.

Consequently,

every prediction includes

* a mean prediction,
* an uncertainty estimate.

Far from the observed data,

the uncertainty naturally increases.

Near the training observations,

the uncertainty decreases.

This behaviour emerges automatically from the probabilistic framework.

---



# <font color='orange'> 12. Relationship to Earlier Models </font>

Notice how the ideas developed throughout this notebook connect naturally.

* Linear Regression specifies one linear function.
* Polynomial Regression specifies one polynomial.
* Decision Trees specify one tree.
* Random Forest averages many trees.
* Gaussian Processes consider infinitely many functions simultaneously and assign probabilities to them.

Thus,

Gaussian Processes represent one of the most flexible regression frameworks encountered so far.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Gaussian Distribution | Probability distribution over a single random variable |
| Multivariate Gaussian | Joint distribution over several correlated variables |
| Gaussian Process | Probability distribution over functions |
| Mean Function | Expected value of the function at each input |
| Covariance Function | Specifies how function values at different inputs are related |
| Prior | Distribution over functions before observing data |
| Posterior | Updated distribution after incorporating observations |

> **Key Insight:** A Gaussian Process extends the familiar Gaussian distribution from individual random variables to entire functions. Rather than assuming a single regression function, it assigns probabilities to infinitely many possible functions, each characterized by a mean function and a covariance function. Observing training data updates this distribution through Bayesian inference, producing a posterior distribution that not only predicts the expected function but also quantifies the uncertainty associated with every prediction. This ability to model both predictions and their uncertainty is the defining characteristic of Gaussian Process Regression.

# <font color='blue'> Gaussian Process Regression (GPR) </font>

The previous chapter introduced Gaussian Processes as probability distributions over functions. Before observing any data, a Gaussian Process describes our prior beliefs regarding the possible functions that could explain the relationship between the predictor variables and the target variable. Once training observations become available, these prior beliefs are updated through Bayesian inference to produce a posterior distribution over functions.

Gaussian Process Regression (GPR) uses this posterior distribution to make predictions. Unlike deterministic regression algorithms that return only a single predicted value, GPR produces both an expected prediction and an associated measure of uncertainty. This uncertainty naturally increases in regions where little or no training data are available and decreases near observed data points.

This ability to combine flexible nonlinear modelling with principled uncertainty quantification makes Gaussian Process Regression one of the most elegant and powerful regression techniques in modern machine learning.

---



# <font color='orange'> 1. Review: Gaussian Processes </font>

Recall that a Gaussian Process is written as

$$
\boxed{
f(x)
\sim
\mathcal{GP}
\left(
m(x),
k(x,x')
\right),
}
$$

where

* \(m(x)\) is the mean function,
* \(k(x,x')\) is the covariance (kernel) function.

The Gaussian Process specifies

a probability distribution over possible regression functions.

---



# <font color='orange'> 2. Prior Distribution </font>

Before observing any data,

our knowledge of the regression function is limited.

The Gaussian Process therefore represents a

**prior distribution**

over functions.

Conceptually,

```
Many Possible Functions

↓

Prior Distribution
```

Each function is considered plausible according to the specified mean and covariance functions.

---



# <font color='orange'> 3. Observing Training Data </font>

Suppose we observe

$$
n
$$

training observations

$$
(X,\mathbf y).
$$

These observations provide information about

which functions are consistent with reality.

The prior distribution is therefore updated using Bayes' theorem.

```
Prior

↓

Training Data

↓

Posterior
```

This posterior distribution forms the regression model.

---



# <font color='orange'> 4. Joint Gaussian Distribution </font>

Suppose

we wish to predict

the function value

at a new input

$$
x_*.
$$

The observed targets

and

the unknown prediction

are jointly Gaussian.

Mathematically,

$$
\boxed{
\begin{bmatrix}
\mathbf y\\
f_*
\end{bmatrix}
\sim
\mathcal N
\left(
0,
\begin{bmatrix}
K(X,X)+\sigma_n^2I &
K(X,x_*)
\\
K(x_*,X) &
K(x_*,x_*)
\end{bmatrix}
\right).
}
$$

Here,

* \(K(X,X)\) is the covariance matrix of the training inputs,
* \(\sigma_n^2\) represents observation noise,
* \(K(X,x_*)\) describes the covariance between the training inputs and the new input.

---



# <font color='orange'> 5. Conditioning on the Observations </font>

The key step in Gaussian Process Regression is

conditioning

the joint Gaussian distribution

on the observed targets.

This produces

the posterior predictive distribution.

Conceptually,

```
Joint Gaussian

↓

Condition on Training Data

↓

Posterior Prediction
```

This is simply Bayesian updating applied to Gaussian distributions.

---



# <font color='orange'> 6. Posterior Mean </font>

The predictive mean is

$$
\boxed{
\mu_*
=
K(x_*,X)
\left[
K(X,X)+\sigma_n^2I
\right]^{-1}
\mathbf y.
}
$$

This is the expected prediction.

Unlike Linear Regression,

the prediction depends on

the covariance between the new observation and every training observation.

---



# <font color='orange'> 7. Posterior Variance </font>

The predictive uncertainty is

$$
\boxed{
\sigma_*^2
=
K(x_*,x_*)
-
K(x_*,X)
\left[
K(X,X)+\sigma_n^2I
\right]^{-1}
K(X,x_*).
}
$$

This variance quantifies

how uncertain

the prediction is.

---

### Near Training Data

```
Low Uncertainty
```

---

### Far from Training Data

```
High Uncertainty
```

This behaviour emerges automatically from the probabilistic model.

---



# <font color='orange'> 8. Prediction with Uncertainty </font>

Unlike deterministic regression,

the prediction becomes

$$
\boxed{
f_*
\sim
\mathcal N
(
\mu_*,
\sigma_*^2
).
}
$$

Thus,

every prediction consists of

* an expected value,
* an uncertainty estimate.

Confidence intervals are obtained directly from this predictive distribution.

---



# <font color='orange'> 9. The Role of the Kernel </font>

The kernel function determines

how information is shared between observations.

If two inputs are similar,

their outputs are expected to be similar.

Different kernels produce different classes of regression functions.

Examples include

* Radial Basis Function (RBF),
* Matérn,
* Rational Quadratic,
* Periodic.

Kernel selection therefore determines the smoothness and flexibility of the regression model.

---



# <font color='orange'> 10. Observation Noise </font>

Real-world observations are noisy.

Gaussian Process Regression models this by assuming

$$
y
=
f(x)
+
\epsilon,
$$

where

$$
\epsilon
\sim
\mathcal N
(0,\sigma_n^2).
$$

The parameter

$$
\sigma_n^2
$$

represents the variance of the measurement noise.

Larger noise produces smoother posterior functions.

# <font color='blue'> Kernel Functions in Gaussian Process Regression </font>

Kernel functions play a central role in Gaussian Process Regression (GPR). While kernels were previously introduced in the context of Support Vector Machines as a computational tool for constructing nonlinear decision boundaries, their role in Gaussian Processes is fundamentally different. In GPR, the kernel function defines the covariance between function values at different input locations and therefore specifies the class of functions that the model considers plausible before observing any data.

Different kernel functions encode different assumptions regarding smoothness, periodicity, correlation length, and variability. Consequently, selecting an appropriate kernel is one of the most important modelling decisions in Gaussian Process Regression.

This chapter introduces the most widely used covariance functions, explains their mathematical forms, and discusses the types of regression problems for which each kernel is most appropriate.

---



# <font color='orange'> 1. Why Are Kernels Important? </font>

Recall that a Gaussian Process is defined as

$$
f(x)
\sim
\mathcal{GP}
\left(
m(x),
k(x,x')
\right).
$$

The mean function

$$
m(x)
$$

determines the average behaviour of the function.

The kernel

$$
k(x,x')
$$

determines

* how nearby function values are related,
* how smooth the function is,
* how rapidly it changes,
* and how uncertainty propagates across the input space.

In practice,

the kernel determines almost all of the modelling power of a Gaussian Process.

---



# <font color='orange'> 2. Covariance and Similarity </font>

Suppose two inputs are

$$
x
\quad\text{and}\quad
x'.
$$

If they are close together,

we usually expect

$$
f(x)
$$

and

$$
f(x')
$$

to have similar values.

Therefore,

their covariance should be large.

Conversely,

inputs that are far apart generally have lower covariance.

The kernel mathematically encodes this intuition.

---



# <font color='orange'> 3. Radial Basis Function (RBF) Kernel </font>

The most widely used kernel is the **Radial Basis Function (RBF)** kernel, also known as the **Squared Exponential** or **Gaussian** kernel.

It is defined as

$$
\boxed{
k(x,x')
=
\sigma_f^2
\exp
\left(
-\frac{(x-x')^2}
{2l^2}
\right),
}
$$

where

* \(l\) is the **length scale**,
* \(\sigma_f^2\) is the signal variance.

---

### Properties

* Infinitely differentiable.
* Produces very smooth functions.
* Nearby points are highly correlated.
* Covariance decreases exponentially with distance.

---

### Typical Applications

* Physical systems.
* Environmental modelling.
* Smooth scientific data.
* General-purpose regression.

The RBF kernel is usually the first kernel to try in practice.

---



# <font color='orange'> 4. The Length Scale </font>

The parameter

$$
l
$$

controls how quickly the function changes.

### Large Length Scale

```
Slowly Varying

↓

Very Smooth Function
```

---

### Small Length Scale

```
Rapidly Changing

↓

Highly Flexible Function
```

Thus,

the length scale controls the complexity of the regression function.

---



# <font color='orange'> 5. Matérn Kernel </font>

The Matérn kernel generalizes the RBF kernel by allowing different levels of smoothness.

One common form is

$$
\boxed{
k(r)
=
\sigma_f^2
\left(
1+
\frac{\sqrt3r}{l}
\right)
\exp
\left(
-\frac{\sqrt3r}{l}
\right),
}
$$

where

$$
r=|x-x'|.
$$

Unlike the RBF kernel,

the Matérn kernel can model functions that are less smooth and therefore often more realistic for physical phenomena.

---

### Typical Applications

* Robotics.
* Geostatistics.
* Engineering.
* Bayesian optimization.

The Matérn kernel is often preferred when the underlying function is expected to have abrupt but continuous changes.

---



# <font color='orange'> 6. Rational Quadratic Kernel </font>

The Rational Quadratic kernel can be viewed as a weighted combination of RBF kernels with different length scales.

It is defined as

$$
\boxed{
k(x,x')
=
\sigma_f^2
\left(
1+
\frac{(x-x')^2}
{2\alpha l^2}
\right)^{-\alpha},
}
$$

where

$$
\alpha
$$

controls the relative weighting of different length scales.

---

### Properties

* Handles functions with varying smoothness.
* Suitable when different regions of the input space exhibit different characteristic scales.

---



# <font color='orange'> 7. Periodic Kernel </font>

Many real-world phenomena repeat periodically.

Examples include

* seasons,
* tides,
* biological rhythms,
* electrical signals.

The Periodic kernel is

$$
\boxed{
k(x,x')
=
\sigma_f^2
\exp
\left[
-
\frac{
2
\sin^2
\left(
\pi|x-x'|/p
\right)
}
{l^2}
\right],
}
$$

where

* \(p\) is the period.

---

### Applications

* Weather forecasting.
* Time-series analysis.
* Signal processing.
* Astronomy.

---



# <font color='orange'> 8. White Noise Kernel </font>

Real observations contain measurement noise.

The White Noise kernel models this uncertainty.

It contributes covariance only when

the two inputs are identical.

Conceptually,

```
Noise

↓

Independent Observations

↓

Diagonal Covariance Matrix
```

It is frequently added to other kernels.

---



# <font color='orange'> 9. Constant Kernel </font>

The Constant kernel models an unknown overall scale.

It is often multiplied by another kernel.

For example,

```
Constant

×

RBF

↓

Scaled RBF Kernel
```

This allows the model to learn the overall magnitude of the variations.

---



# <font color='orange'> 10. Composite Kernels </font>

One of the greatest strengths of Gaussian Processes is that kernels can be combined.

### Addition

```
RBF

+

Periodic

↓

Smooth Trend

+

Seasonal Variation
```

---

### Multiplication

# <font color='blue'> Hyperparameter Optimization for Gaussian Process Regression </font>

The predictive performance of a Gaussian Process Regression (GPR) model depends strongly on the choice of kernel function and its associated hyperparameters. These hyperparameters determine the smoothness of the regression function, the characteristic length scale of variations, the signal variance, and the level of observation noise. Unlike many machine learning algorithms that rely primarily on cross-validation or exhaustive search techniques, Gaussian Process Regression typically estimates these quantities by maximizing the **log marginal likelihood**, also known as the model evidence.

The log marginal likelihood provides a principled Bayesian criterion for selecting hyperparameters because it naturally balances model complexity against goodness of fit. Consequently, Gaussian Process Regression is able to learn suitable kernel parameters directly from the data while avoiding unnecessary model complexity.

---



# <font color='orange'> 1. Why Hyperparameter Optimization Matters </font>

Recall that a Gaussian Process is defined by

$$
f(x)
\sim
\mathcal{GP}
(m(x),k(x,x')).
$$

The kernel function

determines the class of functions

considered plausible.

However,

the kernel itself contains parameters.

For example,

the RBF kernel contains

* length scale,

$$
l,
$$

* signal variance,

$$
\sigma_f^2.
$$

Similarly,

the observation noise introduces

$$
\sigma_n^2.
$$

Selecting appropriate values for these parameters is essential for obtaining accurate predictions.

---



# <font color='orange'> 2. Typical Kernel Hyperparameters </font>

Different kernels introduce different hyperparameters.

For the RBF kernel,

| Hyperparameter | Purpose |
| :--- | :--- |
| Length scale (\(l\)) | Controls how rapidly the function varies |
| Signal variance (\(\sigma_f^2\)) | Controls the amplitude of the function |
| Noise variance (\(\sigma_n^2\)) | Models measurement noise |

These parameters determine both the predictive mean and the predictive uncertainty.

---



# <font color='orange'> 3. The Effect of the Length Scale </font>

The length scale governs the smoothness of the regression function.

### Large Length Scale

```
Nearby points remain strongly correlated

↓

Very smooth function

↓

Higher bias
```

---

### Small Length Scale

```
Correlation decreases rapidly

↓

Rapidly varying function

↓

Higher variance
```

Thus,

the length scale controls model complexity.

---



# <font color='orange'> 4. Observation Noise </font>

The observation model is

$$
y
=
f(x)
+
\epsilon,
$$

where

$$
\epsilon
\sim
\mathcal N
(0,\sigma_n^2).
$$

---

### Small Noise Variance

The model assumes

the observations are highly reliable.

The regression function tends to pass close to every observation.

---

### Large Noise Variance

The model assumes

the observations are noisy.

The resulting regression function becomes smoother.

---



# <font color='orange'> 5. Why Not Use Grid Search? </font>

In previous chapters,

hyperparameters were selected using

* Grid Search,
* Random Search,
* Cross-Validation.

Although these methods can also be applied to GPR,

they become inefficient because

* kernel parameters are continuous,
* the search space is high-dimensional,
* evaluating every combination is computationally expensive.

Instead,

Gaussian Processes optimize the hyperparameters directly.

---



# <font color='orange'> 6. The Log Marginal Likelihood </font>

Suppose the hyperparameters are denoted by

$$
\theta.
$$

The marginal likelihood is

$$
p(\mathbf y|X,\theta).
$$

Rather than maximizing the likelihood of the regression function,

Gaussian Processes maximize

the probability of the observed data

after integrating over all possible functions.

In practice,

the logarithm is optimized,

yielding the **log marginal likelihood (LML)**.

---



# <font color='orange'> 7. The Log Marginal Likelihood Formula </font>

The log marginal likelihood is

$$
\boxed{
\log p(\mathbf y|X,\theta)
=
-\frac12
\mathbf y^T
K^{-1}
\mathbf y
-
\frac12
\log|K|
-
\frac n2
\log(2\pi),
}
$$

where

$$
K
=
K(X,X)+\sigma_n^2I.
$$

Each term has an intuitive interpretation.

---

### Data Fit

$$
-\frac12
\mathbf y^TK^{-1}\mathbf y
$$

Measures

how well

the model explains the observed data.

---

### Complexity Penalty

$$
-\frac12
\log|K|
$$

Penalizes unnecessarily complex models.

---

### Constant Term

$$
-\frac n2\log(2\pi)
$$

Normalizes the probability distribution.

---



# <font color='orange'> 8. Occam's Razor </font>

The log marginal likelihood naturally implements

**Occam's Razor**.

```
Simple Model

↓

Good Fit

↓

High Marginal Likelihood
```

---

```
Overly Complex Model

↓

Only Slightly Better Fit

↓

Lower Marginal Likelihood
```

Thus,

Bayesian model evidence automatically balances

accuracy

and

complexity.

---



# <font color='orange'> 9. Automatic Hyperparameter Optimization </font>

Scikit-Learn automatically maximizes the log marginal likelihood.

Internally,

numerical optimization methods such as

* L-BFGS,

are employed to estimate

the optimal kernel parameters.

The user usually needs only to specify

reasonable initial values.

---



# <font color='orange'> 10. Practical Implementation </font>

Example

```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import (
    ConstantKernel,
    RBF,
    WhiteKernel
)

kernel = (
    ConstantKernel(1.0)
    * RBF(length_scale=1.0)
    + WhiteKernel()
)

model = GaussianProcessRegressor(
    kernel=kernel,
    n_restarts_optimizer=10,
    random_state=42
)

model.fit(X_train, y_train)
```

The parameter

```python
n_restarts_optimizer
```

runs the optimization from several different initial values,

reducing the risk of converging to a poor local optimum.

# <font color='orange'> 12. Common Pitfalls </font>

Several issues commonly arise when fitting Gaussian Process models.

* Choosing an inappropriate kernel.
* Optimizing from poor initial parameter values.
* Applying GPR to very large datasets.
* Ignoring feature scaling when predictor variables have very different ranges.
* Overinterpreting uncertainty estimates when the chosen kernel poorly reflects the true data-generating process.

Careful kernel selection and model validation remain important.

---



# <font color='orange'> 13. Practical Recommendations </font>

For many regression problems,

the following workflow is effective.

1. Standardize the predictor variables.
2. Begin with an RBF kernel.
3. Include a White Noise kernel if measurement noise is expected.
4. Allow automatic optimization of kernel hyperparameters.
5. Use multiple optimizer restarts.
6. Compare different kernel choices using the log marginal likelihood and predictive performance on validation data.

Although the log marginal likelihood provides a powerful optimization criterion, evaluating performance on held-out data remains an important safeguard against poor modelling assumptions.
