# <font color='blue'> Cross-Validation </font>

One of the primary goals of supervised machine learning is to develop models that perform well not only on the training data but also on previously unseen observations. While a simple train-test split provides one estimate of a model's predictive performance, this estimate can vary considerably depending on how the data were partitioned.

Cross-validation is a resampling technique designed to provide a more reliable estimate of a model's ability to generalize. Rather than evaluating the model using a single train-test split, cross-validation repeatedly divides the data into different training and validation subsets, trains the model multiple times, and averages the resulting performance. This reduces the dependence on any particular random split and provides a more stable assessment of model quality.

Cross-validation is one of the most widely used techniques for model selection, hyperparameter tuning, and performance evaluation in machine learning.

---

# <font color='orange'> 1. Why is a Single Train-Test Split Not Enough? </font>

Suppose we randomly divide a dataset into

```
80%

Training

↓

20%

Testing
```

After fitting a regression model,

we obtain

```
RMSE = 12.4
```

The natural question is

> **Would we obtain the same RMSE if the data had been split differently?**

Often,

the answer is **no**.

A different random split may produce

```
RMSE = 10.8
```

or

```
RMSE = 14.1
```

Thus,

a single train-test split provides only one possible estimate of model performance.

Cross-validation reduces this dependence on a particular split.

---

# <font color='orange'> 2. The Central Idea </font>

Instead of evaluating the model once,

we evaluate it

many times

using different subsets of the data.

```
Split 1

↓

Train

↓

Validate

↓

Performance
```

```
Split 2

↓

Train

↓

Validate

↓

Performance
```

```
...

↓

Average Performance
```

The average performance provides a more reliable estimate of the model's ability to generalize.

---

# <font color='orange'> 3. K-Fold Cross-Validation </font>

The most widely used cross-validation method is **K-Fold Cross-Validation**.

The dataset is divided into

```
K

equal-sized folds.
```

For each iteration,

one fold is used for validation,

while the remaining

\(K-1\)

folds are used for training.

For example,

if

```
K = 5
```

the procedure becomes

```
Fold 1

Validation

Training

Training

Training

Training
```

```
Fold 2

Training

Validation

Training

Training

Training
```

Continue until every fold has served once as the validation set.

---

# <font color='orange'> 4. Example: 5-Fold Cross-Validation </font>

Suppose

we have

```
1,000

observations.
```

Each fold contains

```
200

samples.
```

The model is trained and evaluated five times.

| Fold | Training Samples | Validation Samples |
| :---: | :---: | :---: |
| 1 | 800 | 200 |
| 2 | 800 | 200 |
| 3 | 800 | 200 |
| 4 | 800 | 200 |
| 5 | 800 | 200 |

The final performance is

the average

of the five validation scores.

---

# <font color='orange'> 5. Choosing the Number of Folds </font>

Typical choices include

| K | Characteristics |
| :---: | :--- |
| 5 | Good balance between bias and variance |
| 10 | Common default in many applications |
| Large K | Lower bias, higher computational cost |

In practice,

**5-fold** and **10-fold**

cross-validation are the most commonly used.

---

# <font color='orange'> 6. Advantages of K-Fold Cross-Validation </font>

Compared with a single train-test split,

K-Fold Cross-Validation

* uses all observations for both training and validation,
* produces more stable performance estimates,
* reduces dependence on one random split,
* is particularly useful for small and medium-sized datasets.

---

# <font color='orange'> 7. Leave-One-Out Cross-Validation (LOOCV) </font>

LOOCV represents the extreme case of K-Fold Cross-Validation.

```
K

=

Number of Observations
```

Each iteration leaves

exactly one observation

for validation.

For a dataset containing

100 observations,

the model is trained

100 times.

---

### Advantages

* Uses almost all data for training.
* Very low bias.

---

### Limitations

* Computationally expensive.
* Higher variance than moderate K-fold methods.

LOOCV is mainly useful for very small datasets.

---

# <font color='orange'> 8. Repeated K-Fold Cross-Validation </font>

Even K-Fold Cross-Validation depends slightly on the initial random partition.

To reduce this effect,

the entire K-Fold procedure can be repeated multiple times using different random splits.

```
5-Fold

↓

Repeat 10 Times

↓

50 Validation Scores

↓

Average
```

This provides an even more stable estimate of performance.

---

# <font color='orange'> 9. Nested Cross-Validation </font>

When hyperparameter tuning is performed,

using the same cross-validation folds for both tuning and evaluation can lead to optimistic performance estimates.

Nested Cross-Validation avoids this problem by introducing two levels of cross-validation.

```
Outer CV

↓

Performance Evaluation

↓

Inner CV

↓

Hyperparameter Tuning
```

The **inner loop** selects the best hyperparameters,

while the **outer loop** estimates the model's generalization performance.

Nested Cross-Validation is considered the gold standard for unbiased model comparison.

---

# <font color='orange'> 10. Cross-Validation for Regression </font>

In regression,

each fold is typically created using **random sampling** because the target variable is continuous.

The performance metric is computed for each fold.

Common choices include

* RMSE
* MAE
* MSE
* R²

The average metric across all folds estimates the expected performance on unseen data.

---

# <font color='orange'> 11. Cross-Validation for Classification </font>

For completeness,

classification problems often use

**Stratified K-Fold Cross-Validation**.

Stratification preserves the class proportions within each fold,

which is especially important for imbalanced datasets.

Because regression targets are continuous,

standard K-Fold Cross-Validation is usually sufficient.

---

# <font color='orange'> 12. Time-Series Cross-Validation </font>

Randomly shuffling observations is inappropriate for time-series data.

Instead,

the training data should always precede the validation data chronologically.

A common strategy is

```
Train

↓

Validate

↓

Expand Training Window

↓

Validate Again
```

This approach respects the temporal ordering of the observations and prevents information from the future leaking into the past.

---

# <font color='orange'> 13. Cross-Validation and Hyperparameter Tuning </font>

Cross-validation is extensively used for selecting hyperparameters.

Example

Polynomial Regression

```
Degree = 2

↓

RMSE = 15.1
```

```
Degree = 3

↓

RMSE = 11.8
```

```
Degree = 5

↓

RMSE = 18.4
```

Cross-validation suggests

Degree = 3

because it produces the lowest average validation error.

The same idea applies to choosing

* Ridge penalty (\(\lambda\)),
* LASSO penalty,
* Tree depth,
* Number of estimators,
* Learning rate.

# <font color='orange'> 14. Practical Implementation </font>

Scikit-Learn provides built-in support.

Example

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(
    model,
    X,
    y,
    cv=5,
    scoring="neg_root_mean_squared_error"
)
```

The returned scores correspond to the performance obtained on each validation fold.

The average score estimates the model's generalization performance.

---

# <font color='orange'> 15. Advantages and Limitations </font>

### Advantages

* More reliable than a single train-test split.
* Efficient use of available data.
* Reduces variance in performance estimates.
* Supports model selection and hyperparameter tuning.

---

### Limitations

* Computationally more expensive.
* May still exhibit some variability.
* Standard K-Fold is unsuitable for time-series data.

---

# <font color='orange'> 16. Best Practices </font>

* Use 5-fold or 10-fold Cross-Validation for most regression problems.
* Use repeated K-Fold when performance estimates are unstable.
* Use nested Cross-Validation for unbiased model comparison during hyperparameter tuning.
* Use time-series cross-validation for sequential data.
* Reserve the independent test set until the final evaluation.

> **Key Insight:** Cross-validation is a fundamental technique for estimating how well a regression model will generalize to unseen data. By repeatedly training and validating the model on different subsets of the dataset, cross-validation provides a more reliable estimate of predictive performance than a single train-test split. It also forms the basis for model selection and hyperparameter tuning, helping practitioners choose models that perform well beyond the training data while reducing the risk of overfitting.
