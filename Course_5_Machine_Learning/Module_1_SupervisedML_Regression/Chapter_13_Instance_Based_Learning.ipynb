# <font color='blue'> Instance-Based Learning: A Different Approach to Regression </font>

Most supervised learning algorithms construct an explicit mathematical model from the training data. Linear Regression estimates coefficients, Decision Trees learn hierarchical decision rules, and Support Vector Regression determines an optimal regression function through constrained optimization. Once training is complete, these learned models are used to make predictions on new observations.

Instance-Based Learning adopts a fundamentally different philosophy. Rather than constructing a global predictive model during training, it stores the training observations themselves and postpones most of the computation until predictions are required. When a new observation is encountered, its prediction is obtained by comparing it with similar observations in the training dataset.

Because learning occurs primarily at prediction time rather than during training, instance-based methods are often referred to as **lazy learning algorithms**. The most widely used instance-based regression algorithm is **k-Nearest Neighbours (k-NN) Regression**, which will be introduced in the following chapter.

---



# <font color='orange'> 1. Two Different Learning Philosophies </font>

Most machine learning algorithms follow the pattern

```
Training Data

↓

Learn Model

↓

Discard Training Details

↓

Predict
```

Examples include

* Linear Regression
* Decision Trees
* Random Forest
* Gradient Boosting
* Support Vector Regression

These algorithms are called **eager learners** because they perform most of the learning during training.

---

Instance-Based Learning follows a different strategy.

```
Training Data

↓

Store Observations

↓

New Sample Arrives

↓

Find Similar Observations

↓

Predict
```

Very little learning occurs during training.

Most computation is deferred until prediction.

---



# <font color='orange'> 2. Eager Learning vs Lazy Learning </font>

| Eager Learning | Lazy Learning |
| :--- | :--- |
| Builds a predictive model during training | Stores training observations |
| Training is computationally expensive | Training is inexpensive |
| Prediction is usually fast | Prediction can be computationally expensive |
| Learns global relationships | Uses local neighbourhood information |

Examples of eager learners include

* Linear Regression
* Random Forest
* XGBoost
* SVR

The best-known lazy learner is

* k-Nearest Neighbours (k-NN).

---



# <font color='orange'> 3. Learning by Similarity </font>

Suppose we wish to estimate the price of a house.

Instead of fitting a regression equation,

we ask

> **Which previously sold houses most closely resemble this one?**

If similar houses sold for similar prices,

their prices provide useful information about the likely price of the new house.

Thus,

predictions are based on **similarity** rather than an explicitly learned mathematical model.

---



# <font color='orange'> 4. The Importance of Distance </font>

To identify similar observations,

we require a way to measure similarity.

For numerical predictors,

this is usually accomplished using a **distance metric**.

Conceptually,

```
New Observation

↓

Compute Distances

↓

Find Nearest Neighbours

↓

Combine Their Target Values

↓

Prediction
```

The choice of distance metric is therefore central to instance-based learning.

---



# <font color='orange'> 5. Local Learning </font>

Many regression algorithms attempt to describe the entire dataset using a single global model.

Instance-based learning instead focuses on

the neighbourhood surrounding the query observation.

```
Entire Dataset

↓

Locate Nearby Observations

↓

Use Local Information

↓

Predict
```

This allows k-NN to adapt naturally to complex nonlinear relationships.

---



# <font color='orange'> 6. Advantages of Local Models </font>

Because predictions depend only on nearby observations,

instance-based methods can model

* nonlinear relationships,
* irregular decision surfaces,
* locally varying behaviour.

No assumptions are made regarding the overall functional form of the relationship between the predictors and the response.

---



# <font color='orange'> 7. Computational Trade-Off </font>

Instance-based learning shifts the computational burden.

### During Training

```
Store Data
```

Training is extremely fast.

---

### During Prediction

```
Compute Distances

↓

Find Neighbours

↓

Predict
```

Prediction is comparatively more expensive because distances must be computed for each new observation.

---



# <font color='orange'> 8. Feature Scaling </font>

Because predictions depend on distances,

feature scaling becomes essential.

Suppose two predictors are

| Feature | Range |
| :--- | :---: |
| Age | 18–70 |
| Annual Income (£) | 20,000–200,000 |

Without scaling,

the income variable dominates the distance calculation,

making age almost irrelevant.

Consequently,

standardization or normalization is typically performed before applying k-NN.

---



# <font color='orange'> 9. Curse of Dimensionality </font>

Instance-based learning performs well in low- and moderate-dimensional spaces.

However,

as the number of predictors increases,

all observations tend to become similarly distant.

This phenomenon is known as the **curse of dimensionality**.

Consequences include

* reduced distinction between neighbours,
* decreased predictive accuracy,
* increased computational cost.

Dimensionality reduction techniques such as PCA are often applied before k-NN when dealing with high-dimensional data.

---



# <font color='orange'> 10. Applications </font>

Instance-based learning has many practical applications.

* House price prediction
* Recommendation systems
* Medical diagnosis
* Image recognition
* Pattern recognition
* Anomaly detection

Its simplicity and flexibility make it an attractive baseline method for many regression and classification problems.

---



# <font color='orange'> 11. Advantages </font>

Instance-based learning offers several benefits.

* Simple to understand and implement.
* Naturally models nonlinear relationships.
* No explicit training phase.
* Few hyperparameters.
* Adapts to local data structure.

---



# <font color='orange'> 12. Limitations </font>

It also has important limitations.

* Prediction can be slow for large datasets.
* Requires feature scaling.
* Sensitive to irrelevant features.
* Suffers from the curse of dimensionality.
* Memory-intensive because all training observations must be stored.

---



# <font color='orange'> 13. Relationship to k-Nearest Neighbours </font>

Instance-Based Learning is not itself a specific algorithm.

Rather,

it is a general learning philosophy.

The most important instance-based regression algorithm is

```
Instance-Based Learning

↓

Similarity

↓

Distance Metrics

↓

Nearest Neighbours

↓

k-Nearest Neighbours Regression
```

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Instance-Based Learning | Predicts using stored training observations rather than an explicit model |
| Lazy Learning | Learning occurs primarily during prediction |
| Eager Learning | Learning occurs primarily during training |
| Local Learning | Predictions are based on nearby observations |
| Distance Metric | Quantifies similarity between observations |
| Feature Scaling | Essential because distances determine predictions |
| Curse of Dimensionality | Reduced effectiveness in high-dimensional feature spaces |

> **Key Insight:** Instance-Based Learning represents a fundamentally different approach to supervised learning. Instead of constructing a global predictive model during training, it stores the training observations and makes predictions by comparing new observations with similar examples. This local, similarity-based philosophy allows the method to model complex nonlinear relationships without assuming a specific functional form, but it also makes prediction dependent on distance calculations, feature scaling, and the availability of representative neighbouring observations.
