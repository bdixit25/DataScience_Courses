# <font color='blue'> Comparing Dimensionality Reduction Techniques </font>

Dimensionality reduction plays a central role in modern machine learning by enabling high-dimensional datasets to be represented using a smaller number of informative variables. Throughout this section, we have studied several of the most widely used dimensionality reduction techniques, each based on different mathematical principles and designed for different objectives.

No single dimensionality reduction algorithm is universally optimal. Some methods prioritize preserving global variance, while others emphasize local neighbourhood relationships or nonlinear manifold structure. Likewise, some techniques are primarily intended for feature engineering, whereas others are designed almost exclusively for visualization.

This chapter compares the principal dimensionality reduction methods, discusses their assumptions and computational characteristics, and provides practical guidance for selecting the most appropriate technique for a given application.

---



# <font color='orange'> 1. Why Compare Dimensionality Reduction Methods? </font>

Different datasets possess different geometric structures.

Some datasets are

* approximately linear,
* highly nonlinear,
* noisy,
* extremely high-dimensional.

Consequently,

different dimensionality reduction techniques preserve different aspects of the data.

Choosing an appropriate method is therefore essential.

---



# <font color='orange'> 2. The Major Families </font>

The techniques studied in this notebook belong to two broad categories.

| Method | Category |
| :--- | :--- |
| PCA | Linear |
| t-SNE | Nonlinear |
| UMAP | Nonlinear |
| Isomap | Nonlinear |
| LLE | Nonlinear |

PCA is fundamentally different because it assumes linear relationships between variables.

---



# <font color='orange'> 3. What Does Each Method Preserve? </font>

The methods differ primarily in what they attempt to preserve.

| Method | Preserves |
| :--- | :--- |
| PCA | Global variance |
| t-SNE | Local neighbourhoods |
| UMAP | Local neighbourhoods with better preservation of some global structure |
| Isomap | Global manifold geometry (via geodesic distances) |
| LLE | Local linear relationships |

Understanding this distinction is often the key to selecting the appropriate algorithm.

---



# <font color='orange'> 4. Linear vs Nonlinear Methods </font>

### Linear Methods

Assume that

important structure

can be represented by

linear combinations

of the original features.

Example

```
PCA
```

---

### Nonlinear Methods

Assume the observations lie on

a curved manifold.

Examples

```
t-SNE

UMAP

Isomap

LLE
```

These algorithms are capable of revealing structures that PCA cannot capture.

# <font color='orange'> 14. Practical Recommendations </font>

The following guidelines summarize common practice.

- Use **PCA** when you need an efficient, interpretable dimensionality reduction method for preprocessing, compression, or feature engineering.
- Use **t-SNE** when your primary goal is to visualize complex local cluster structures in small- to medium-sized datasets.
- Use **UMAP** when working with large datasets or when you want a balance between local structure preservation, computational efficiency, and visualization quality.
- Use **Isomap** when preserving the global geometry of a smooth nonlinear manifold is particularly important.
- Use **LLE** when the data are believed to satisfy locally linear relationships and preserving those relationships is the primary objective.

Whenever possible, compare multiple methods on the same dataset rather than assuming one algorithm will always provide the most informative representation.

> **Key Insight:** Dimensionality reduction algorithms differ fundamentally in the information they preserve and the problems they are designed to solve. PCA provides an efficient and interpretable linear representation suitable for feature engineering, while nonlinear manifold learning techniques such as t-SNE, UMAP, Isomap, and LLE reveal complex geometric structures that cannot be captured by linear projections. Selecting an appropriate method requires balancing interpretability, computational efficiency, scalability, and the specific objectives of the analysis rather than searching for a universally superior algorithm.
