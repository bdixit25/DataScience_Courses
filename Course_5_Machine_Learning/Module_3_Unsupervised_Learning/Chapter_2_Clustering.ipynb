# <font color='blue'> Introduction to Clustering </font>

Clustering is one of the most important tasks in unsupervised machine learning. Its objective is to discover natural groupings within unlabelled data by identifying observations that are more similar to one another than to the rest of the dataset. Unlike supervised classification, where predefined class labels are available during training, clustering algorithms must infer the underlying structure directly from the observed features.

Clustering has applications across a wide range of scientific and industrial disciplines, including customer segmentation, image analysis, bioinformatics, astronomy, document organization, and social network analysis. By revealing hidden structures within complex datasets, clustering provides valuable insights that can support exploratory data analysis, feature engineering, anomaly detection, and decision-making.

This chapter introduces the fundamental concepts of clustering, explains the characteristics of meaningful clusters, and discusses the major categories of clustering algorithms that will be explored throughout this notebook.

---

# <font color='orange'> 1. What is Clustering? </font>

Clustering is the process of dividing a collection of observations into groups, called **clusters**, such that observations within the same cluster are more similar to one another than to observations in different clusters.

Conceptually,

```
Unlabelled Data

↓

Measure Similarity

↓

Group Similar Observations

↓

Clusters
```

Unlike classification,

the clusters are **not predefined**.

Instead,

they are discovered automatically from the data.

---

# <font color='orange'> 2. Why Do We Need Clustering? </font>

Many datasets contain hidden structure that is not immediately obvious.

For example,

* customers may naturally separate into purchasing groups,
* galaxies may form distinct morphological classes,
* genes may exhibit similar expression patterns,
* news articles may cover similar topics.

Clustering enables these hidden patterns to be discovered without requiring labelled examples.

---

# <font color='orange'> 3. What Defines a Good Cluster? </font>

A good clustering solution satisfies two important properties.

### High Intra-Cluster Similarity

Observations within the same cluster should be highly similar.

```
Cluster

● ● ● ● ●

Small Distances
```

---

### Low Inter-Cluster Similarity

Different clusters should be well separated.

```
Cluster A        Cluster B

● ● ●          ● ● ●

Large Distance
```

The goal is therefore

* maximize similarity within clusters,
* minimize similarity between clusters.

---

# <font color='orange'> 4. Similarity and Distance </font>

Most clustering algorithms define similarity using a **distance metric**.

Smaller distances imply

greater similarity.

Larger distances imply

less similarity.

For numerical data,

common distance measures include

* Euclidean distance,
* Manhattan distance,
* Cosine distance,
* Mahalanobis distance.

The choice of distance metric can significantly influence the clustering result.

---

# <font color='orange'> 5. Hard vs. Soft Clustering </font>

Clustering algorithms can be broadly classified into two categories.

### Hard Clustering

Each observation belongs to **exactly one** cluster.

Example

```
Point

↓

Cluster 1
```

Algorithms

* K-Means
* Hierarchical Clustering

---

### Soft Clustering

Each observation belongs to multiple clusters with different probabilities.

Example

```
Point

↓

Cluster A : 70%

Cluster B : 30%
```

Algorithms

* Gaussian Mixture Models (GMMs)

Soft clustering is particularly useful when cluster boundaries overlap.

---

# <font color='orange'> 6. Major Types of Clustering Algorithms </font>

Different algorithms make different assumptions about the structure of the data.

### Partition-Based Clustering

The dataset is divided into a fixed number of clusters.

Examples

* K-Means
* K-Medoids

---

### Hierarchical Clustering

Clusters are organized into a hierarchy.

```
One Large Cluster

↓

Split

↓

Smaller Clusters
```

or

```
Individual Points

↓

Merge

↓

Hierarchy
```

---

### Density-Based Clustering

Clusters correspond to regions of high point density.

Example

* DBSCAN
* OPTICS

These methods naturally identify noise and outliers.

---

### Model-Based Clustering

Assumes the data are generated from an underlying probabilistic model.

Example

* Gaussian Mixture Models

---

### Graph-Based Clustering

Represents observations as a graph and partitions the graph into clusters.

Example

* Spectral Clustering

---

# <font color='orange'> 7. Challenges in Clustering </font>

Unlike supervised learning,

there is usually no correct answer.

Several factors make clustering challenging.

### Unknown Number of Clusters

The appropriate number of clusters is often unknown.

---

### Irregular Cluster Shapes

Real-world clusters are not always spherical.

---

### Different Cluster Sizes

Clusters may contain very different numbers of observations.

---

### Varying Densities

Some clusters may be dense,

others sparse.

---

### Noise and Outliers

Unusual observations can distort clustering results.

---

### High Dimensionality

Distance measures become less informative in very high-dimensional spaces.

---

# <font color='orange'> 8. Feature Scaling </font>

Many clustering algorithms depend directly on distances.

Consequently,

variables measured on different scales may dominate the clustering process.

Example

```
Age

18–80

Income

£20,000–£200,000
```

Without scaling,

income contributes much more to the distance calculation.

Therefore,

standardization is usually recommended before clustering.

Tree-based algorithms are largely unaffected by feature scaling,

whereas K-Means and Hierarchical Clustering are highly sensitive to it.

# <font color='orange'> 10. Evaluating Clustering </font>

Because no labels are available,

evaluating clustering is more challenging than evaluating supervised learning models.

Common evaluation approaches include

* Silhouette Score,
* Davies–Bouldin Index,
* Calinski–Harabasz Index,
* visual inspection,
* domain expertise.

These evaluation techniques will be discussed in later chapters.

# <font color='orange'> 11. Clustering Workflow </font>

A typical clustering project follows the sequence

```
Collect Data

↓

Clean Data

↓

Feature Scaling

↓

Choose Distance Metric

↓

Choose Clustering Algorithm

↓

Fit Model

↓

Evaluate Clusters

↓

Interpret Results
```

This workflow emphasizes understanding the structure of the data rather than making predictions.

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Clustering | Grouping similar observations without labels |
| Intra-Cluster Similarity | Observations within a cluster should be similar |
| Inter-Cluster Similarity | Different clusters should be well separated |
| Hard Clustering | Each observation belongs to one cluster |
| Soft Clustering | Observations belong to multiple clusters with probabilities |
| Partition-Based Clustering | Divides data into a fixed number of clusters |
| Density-Based Clustering | Identifies clusters as dense regions |
| Model-Based Clustering | Uses probabilistic models to represent clusters |
| Graph-Based Clustering | Uses graph connectivity to identify clusters |

> **Key Insight:** Clustering seeks to uncover the hidden structure within unlabelled data by grouping similar observations into meaningful clusters. Because different datasets exhibit different characteristics—such as varying densities, irregular shapes, or overlapping groups—no single clustering algorithm is universally optimal. Understanding the assumptions underlying each clustering approach is therefore essential for selecting an appropriate algorithm and correctly interpreting the resulting clusters.
