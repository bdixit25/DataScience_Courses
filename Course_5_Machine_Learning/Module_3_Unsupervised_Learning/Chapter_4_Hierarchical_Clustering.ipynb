# <font color='blue'> Introduction to Hierarchical Clustering </font>

While K-Means clustering partitions a dataset into a fixed number of clusters, many practical problems do not naturally suggest how many clusters should exist. In such situations, it is often more useful to construct a hierarchy of clusters that can be examined at different levels of granularity. Hierarchical Clustering provides precisely this capability.

Hierarchical Clustering builds a nested sequence of clusters by either progressively merging smaller clusters into larger ones or recursively splitting larger clusters into smaller ones. The resulting hierarchy is represented by a tree-like diagram known as a **dendrogram**, which provides a visual summary of the relationships among observations and allows the user to determine the desired number of clusters after the algorithm has completed.

Unlike K-Means, Hierarchical Clustering does not rely on centroids or iterative optimization of an objective function. Instead, it is based entirely on measuring distances between observations or clusters. This makes it a versatile technique for exploratory data analysis and one of the most widely used clustering methods in biology, chemistry, social sciences, and many other disciplines.

The final result is

```
Hierarchy

↓

Tree Structure

↓

Dendrogram
```

# <font color='orange'> 3. Two Approaches </font>

Hierarchical clustering can be performed in two fundamentally different ways.

### Agglomerative Clustering

Also known as

**bottom-up clustering**.

```
Individual Observations

↓

Merge Closest Pair

↓

Larger Clusters

↓

One Large Cluster
```

Agglomerative clustering is by far the most commonly used approach and is the focus of most practical implementations.

---

### Divisive Clustering

Also known as

**top-down clustering**.

```
Entire Dataset

↓

Split into Two Clusters

↓

Split Again

↓

Individual Observations
```

Although conceptually appealing,

divisive clustering is computationally much more expensive and is less commonly used.

# <font color='orange'> 6. Distance Between Clusters </font>

Unlike K-Means,

hierarchical clustering does not compute centroids.

Instead,

it repeatedly asks

> **Which two clusters should be merged next?**

Answering this question requires defining the distance between clusters.

Different definitions produce different clustering solutions.

Common linkage methods include

* Single linkage
* Complete linkage
* Average linkage
* Ward's linkage

These methods will be studied in the next chapter.

# <font color='orange'> 7. The Dendrogram </font>

The hierarchy is visualized using a **dendrogram**.

Conceptually,

```
        ───────

      ──       ──

    ──           ──

  A   B       C    D
```

The leaves represent individual observations,

while branches indicate cluster mergers.

The height of each branch corresponds to the distance at which the clusters were merged.

# <font color='orange'> 8. Choosing the Number of Clusters </font>

Unlike K-Means,

hierarchical clustering produces

all possible clustering levels simultaneously.

To obtain a clustering solution,

the dendrogram is simply cut at a chosen height.

Conceptually,

```
Dendrogram

↓

Cut Here

↓

Three Clusters
```

Different cutting heights produce different numbers of clusters.

Hierarchical clustering offers several important advantages.

# <font color='orange'> 10. Limitations </font>

Hierarchical clustering also has several limitations.

* Computationally expensive for large datasets.
* Sensitive to noisy observations.
* Early merge decisions cannot be reversed.
* Different linkage methods may produce different hierarchies.

Consequently,

careful selection of the linkage criterion is important.

# <font color='blue'> Linkage Methods and Dendrograms </font>

The previous chapter introduced Hierarchical Clustering and explained how observations are progressively merged into larger clusters or recursively divided into smaller ones. A fundamental question naturally arises during the agglomerative clustering process:

> **Which two clusters should be merged next?**

Unlike K-Means, where observations are assigned to the nearest centroid, Hierarchical Clustering must define the distance between entire clusters rather than between individual observations. This is accomplished through a **linkage criterion**, which specifies how the distance between two clusters should be calculated.

Different linkage methods make different assumptions about cluster shape and separation, and consequently they can produce substantially different clustering hierarchies from the same dataset. The resulting hierarchy is visualized using a **dendrogram**, which provides a graphical representation of the sequence of cluster mergers.

This chapter introduces the most commonly used linkage methods, explains how dendrograms are constructed and interpreted, and discusses practical considerations when selecting a linkage criterion.

---



# <font color='orange'> 1. Why Do We Need Linkage Methods? </font>

Suppose we have already formed two clusters.

```
Cluster A

● ● ●


Cluster B

● ● ●
```

Before continuing the clustering process,

we must determine

how far apart these clusters are.

Unlike individual observations,

clusters contain multiple points.

Therefore,

the distance between clusters is **not uniquely defined**.

Different linkage methods answer this question differently.

---



# <font color='orange'> 2. Single Linkage (Nearest Neighbour) </font>

Single linkage defines the distance between two clusters as

the **smallest distance**

between any pair of observations,

one from each cluster.

Mathematically,

$$
\boxed{
d(A,B)
=
\min_{x_i\in A,\;x_j\in B}
d(x_i,x_j).
}
$$

Conceptually,

```
Cluster A

● ● ●

      ← smallest distance →

            ● ● ●

Cluster B
```

Only the closest pair of observations determines the cluster distance.

---

### Advantages

* Can discover irregularly shaped clusters.
* Effective for elongated structures.

---

### Limitations

* Sensitive to noise.
* May produce long "chains" of observations.

This phenomenon is called the

**chaining effect**.

---



# <font color='orange'> 3. Complete Linkage (Farthest Neighbour) </font>

Complete linkage defines the cluster distance as

the **largest distance**

between any pair of observations.

Mathematically,

$$
\boxed{
d(A,B)
=
\max_{x_i\in A,\;x_j\in B}
d(x_i,x_j).
}
$$

Conceptually,

```
Cluster A

● ● ●

← largest distance →

           ● ● ●

Cluster B
```

The most distant observations determine the cluster separation.

---

### Advantages

* Produces compact clusters.
* Less affected by chaining.

---

### Limitations

* Sensitive to outliers.
* May split naturally elongated clusters.

---



# <font color='orange'> 4. Average Linkage </font>

Average linkage considers

the average distance

between every pair of observations in the two clusters.

Mathematically,

$$
\boxed{
d(A,B)
=
\frac1{|A||B|}
\sum_{x_i\in A}
\sum_{x_j\in B}
d(x_i,x_j).
}
$$

Conceptually,

```
All Pairwise Distances

↓

Average

↓

Cluster Distance
```

Average linkage balances the behaviour of single and complete linkage.

---

### Advantages

* More robust than single linkage.
* Less sensitive to extreme observations.
* Often produces balanced clusters.

---



# <font color='orange'> 5. Ward's Linkage </font>

Ward's method follows a different philosophy.

Rather than directly measuring distances,

it merges the pair of clusters that produces

the **smallest increase in Within-Cluster Sum of Squares (WCSS).**

Conceptually,

```
Possible Merge

↓

Increase in WCSS

↓

Choose Smallest Increase
```

Ward's linkage therefore attempts to preserve compact clusters,

making it conceptually similar to the objective optimized by K-Means.

---

### Advantages

* Produces compact, nearly spherical clusters.
* Often performs very well on numerical data.
* Frequently the default choice in practice.

---

### Limitations

* Best suited to Euclidean distance.
* Less effective for irregularly shaped clusters.

---



# <font color='orange'> 6. Comparing the Linkage Methods </font>

| Linkage Method | Cluster Distance |
| :--- | :--- |
| Single | Minimum pairwise distance |
| Complete | Maximum pairwise distance |
| Average | Mean pairwise distance |
| Ward | Minimum increase in WCSS |

Each method represents a different definition of cluster similarity.

---



# <font color='orange'> 7. The Dendrogram </font>

The sequence of cluster mergers is represented using a

**dendrogram**.

Conceptually,

```
          ─────────────

      ─────          ─────

    ───                 ───

   A   B               C   D
```

The dendrogram summarizes

the complete hierarchical clustering process.

---

### Leaves

The bottom nodes represent

individual observations.

```
A   B   C   D
```

---

### Branches

Branches indicate

cluster mergers.

```
A

↓

Merge with B

↓

Cluster AB
```

---

### Height

The vertical height of each merge

corresponds to

the distance at which the merge occurred.

Large vertical jumps indicate

well-separated clusters.

---



# <font color='orange'> 8. Choosing the Number of Clusters </font>

One of the greatest advantages of hierarchical clustering

is that

the number of clusters is chosen

after

the dendrogram has been constructed.

Conceptually,

```
Dendrogram

↓

Choose Cut Height

↓

Final Clusters
```

A horizontal cut through the dendrogram

intersects several branches.

The number of intersections equals

the number of clusters.

---



# <font color='orange'> 9. Example Interpretation </font>

Suppose the dendrogram contains

a very large vertical gap.

```
──────

      │

      │

──────
```

Cutting within this gap

often produces

well-separated clusters.

Conversely,

cutting where many branches merge close together

may produce unstable clustering solutions.

---



# <font color='orange'> 10. Practical Guidelines </font>

The choice of linkage method depends on the characteristics of the data.

| Situation | Recommended Linkage |
| :--- | :--- |
| Irregular cluster shapes | Single |
| Compact clusters | Complete |
| General-purpose analysis | Average |
| Numerical data with compact clusters | Ward |

No linkage method is universally optimal.

The best choice depends on the geometry of the dataset.
