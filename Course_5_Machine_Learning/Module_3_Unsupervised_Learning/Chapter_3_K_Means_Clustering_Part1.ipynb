# <font color='blue'> K-Means Clustering: Foundations and Intuition </font>

Among the many clustering algorithms available in unsupervised machine learning, **K-Means** is one of the simplest, most widely used, and computationally efficient methods for partitioning data into groups. The algorithm seeks to divide a dataset into a predetermined number of clusters such that observations within each cluster are as similar as possible while observations belonging to different clusters remain as distinct as possible.

Unlike supervised learning algorithms, K-Means does not rely on labelled examples. Instead, it discovers clusters directly from the data by iteratively assigning observations to representative points known as **centroids**. These centroids summarize the location of each cluster, and the algorithm repeatedly updates them until a stable clustering solution is obtained.

This chapter develops an intuitive understanding of K-Means clustering, introduces the concept of centroids, and explains the iterative process through which clusters are formed.

---



# <font color='orange'> 1. The Goal of K-Means </font>

Suppose we are given a collection of observations.

Our objective is to divide these observations into

$$
K
$$

distinct groups.

Conceptually,

```
Dataset

↓

Divide into K Groups

↓

Cluster 1

Cluster 2

...

Cluster K
```

Each observation should belong to the cluster whose representative point is closest.

---



# <font color='orange'> 2. Why Is It Called "K-Means"? </font>

The name consists of two parts.

### K

Represents the number of clusters chosen by the user.

For example,

```
K = 2

↓

Two Clusters
```

or

```
K = 5

↓

Five Clusters
```

---

### Means

Each cluster is represented by the arithmetic mean of the observations assigned to it.

This mean is called the **centroid**.

Thus,

```
Cluster

↓

Average Position

↓

Centroid
```

The algorithm repeatedly updates these centroids during training.

---



# <font color='orange'> 3. What Is a Centroid? </font>

A centroid is the geometric centre of a cluster.

For a two-dimensional dataset,

suppose a cluster contains the points

```
(2,3)

(3,2)

(4,4)
```

The centroid is obtained by averaging each coordinate independently.

$$
x_{\text{centroid}}
=
\frac{2+3+4}{3}
=
3
$$

$$
y_{\text{centroid}}
=
\frac{3+2+4}{3}
=
3
$$

Therefore,

the centroid is

$$
(3,3).
$$

Unlike the individual observations,

the centroid is generally **not** required to coincide with an existing data point.

---



# <font color='orange'> 4. The Basic Idea </font>

The K-Means algorithm alternates between two simple steps.

### Step 1

Assign every observation to its nearest centroid.

```
Observations

↓

Nearest Centroid

↓

Cluster Assignment
```

---

### Step 2

Recompute the centroid of every cluster.

```
Cluster Members

↓

Average Position

↓

Updated Centroid
```

These two steps are repeated until the cluster assignments no longer change.

---



# <font color='orange'> 5. A Simple Example </font>

Suppose we have six observations.

```
● ● ●

            ● ● ●
```

We choose

```
K = 2
```

Initially,

two centroids are placed somewhere within the dataset.

```
×      ● ● ●

●

                 ×

          ● ● ●
```

Each observation is assigned to its nearest centroid.

After assignment,

the centroid positions are updated.

This process continues until the centroids stop moving significantly.

Eventually,

the observations naturally separate into two clusters.

---



# <font color='orange'> 6. Why Does the Algorithm Iterate? </font>

Initially,

the centroids are only rough guesses.

Consequently,

some observations may initially be assigned to the wrong cluster.

After recomputing the centroids,

the cluster centres move closer to the true centres of the data.

The assignments therefore improve.

```
Initial Guess

↓

Assign Points

↓

Update Centroids

↓

Better Guess

↓

Repeat
```

Each iteration improves the clustering solution.

---



# <font color='orange'> 7. The Importance of the Initial Centroids </font>

The starting positions of the centroids influence the final clustering.

Consider the same dataset.

### Good Initialization

```
×      ●●●

          ●●●      ×
```

Produces

```
Correct Clusters
```

---

### Poor Initialization

```
● × ●

● × ●
```

May produce

```
Poor Clusters
```

Different initializations can therefore lead to different solutions.

This issue motivates the **K-Means++ initialization algorithm**, which will be discussed later.

---



# <font color='orange'> 8. Convergence </font>

The algorithm eventually reaches a point where

* no observations change clusters, or
* the centroid locations change only negligibly.

At this point,

the algorithm has **converged**.

```
Assign

↓

Update

↓

Assign

↓

Update

↓

Stable Solution
```

Convergence usually occurs after only a relatively small number of iterations.

---



# <font color='orange'> 9. Local Optima </font>

Although K-Means always converges,

it is **not guaranteed** to find the best possible clustering.

Instead,

it may converge to a **local optimum** that depends on the initial centroid positions.

Consequently,

the algorithm is often executed multiple times using different random initializations,

and the best solution is retained.

---



# <font color='orange'> 10. Strengths of K-Means </font>

K-Means offers several important advantages.

* Simple to understand.
* Easy to implement.
* Computationally efficient.
* Scales well to large datasets.
* Performs well for compact, spherical clusters.
* Widely supported by machine learning libraries.

These properties make K-Means one of the most popular clustering algorithms.

---



# <font color='orange'> 11. Limitations </font>

Despite its popularity,

K-Means has several important limitations.

* The number of clusters must be specified in advance.
* Sensitive to initialization.
* Sensitive to feature scaling.
* Performs poorly for non-spherical clusters.
* Sensitive to outliers.
* Assumes clusters have similar sizes and densities.

These limitations motivate the development of alternative clustering algorithms such as DBSCAN and Gaussian Mixture Models.

---



# <font color='orange'> 12. Applications </font>

K-Means is widely used in practice.

Examples include

| Application | Example |
| :--- | :--- |
| Marketing | Customer segmentation |
| Image Processing | Colour quantization |
| Biology | Gene clustering |
| Astronomy | Grouping galaxies by observed properties |
| Document Analysis | Document clustering |
| Retail | Product categorization |
| Finance | Customer risk segmentation |

Its simplicity and efficiency make it an excellent first-choice clustering algorithm for many exploratory analyses.

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| K | Number of clusters specified by the user |
| Centroid | Mean position of a cluster |
| Assignment Step | Assign observations to the nearest centroid |
| Update Step | Recompute centroid locations |
| Iterative Algorithm | Repeats assignment and update until convergence |
| Local Optimum | Final solution depends on initialization |
| Best Use Case | Compact, well-separated, approximately spherical clusters |

> **Key Insight:** K-Means clustering partitions a dataset into a predetermined number of clusters by representing each cluster with its centroid, the arithmetic mean of its observations. Through an iterative process of assigning observations to the nearest centroid and updating centroid locations, the algorithm progressively improves the clustering solution until convergence. While K-Means is computationally efficient and easy to interpret, its performance depends strongly on the initial centroid positions and the assumption that clusters are compact and approximately spherical.

# <font color='blue'> The K-Means Objective Function and Lloyd's Algorithm </font>

The previous chapter introduced the intuition behind K-Means clustering by explaining how observations are repeatedly assigned to the nearest cluster centroid and how the centroid locations are subsequently updated. While this iterative procedure appears simple, it is in fact solving a well-defined optimization problem.

The objective of K-Means is to partition a dataset into a fixed number of clusters such that the observations within each cluster are as close as possible to their corresponding centroid. This objective is quantified using the **Within-Cluster Sum of Squares (WCSS)**, also known as the **cluster inertia**. Lloyd's Algorithm provides an efficient iterative procedure for approximately minimizing this objective.

This chapter develops the mathematical formulation of K-Means, explains the meaning of the objective function, and describes the optimization procedure used to obtain the final clustering solution.

---



# <font color='orange'> 1. The Optimization Problem </font>

Suppose we have

* \(n\) observations,
* \(d\) features,
* and wish to divide the data into

$$
K
$$

clusters.

Our objective is simple:

> **Make the observations within each cluster as close as possible to their centroid.**

Conceptually,

```
Cluster

↓

Small Distances

↓

Compact Cluster
```

The more compact the clusters,

the better the clustering solution.

---



# <font color='orange'> 2. Measuring Cluster Compactness </font>

To measure how compact a cluster is,

consider one observation

$$
\mathbf{x}_i
$$

and the centroid of its assigned cluster

$$
\boldsymbol{\mu}_k.
$$

The Euclidean distance between them is

$$
\|\mathbf{x}_i-\boldsymbol{\mu}_k\|.
$$

Rather than minimizing the distance itself,

K-Means minimizes the **squared distance**

$$
\boxed{
\|\mathbf{x}_i-\boldsymbol{\mu}_k\|^2.
}
$$

Squaring the distance has two advantages:

* it avoids square roots, making computation simpler,
* it penalizes large distances more heavily than small ones.

---



# <font color='orange'> 3. Within-Cluster Sum of Squares (WCSS) </font>

The total clustering error is obtained by summing the squared distances for all observations.

This quantity is called the **Within-Cluster Sum of Squares (WCSS)**.

Mathematically,

$$
\boxed{
\text{WCSS}
=
\sum_{k=1}^{K}
\sum_{\mathbf{x}_i\in C_k}
\|\mathbf{x}_i-\boldsymbol{\mu}_k\|^2,
}
$$

where

* \(C_k\) denotes the set of observations assigned to cluster \(k\),
* \(\boldsymbol{\mu}_k\) is the centroid of cluster \(k\).

The objective of K-Means is therefore

$$
\boxed{
\min \text{WCSS}.
}
$$

---



# <font color='orange'> 4. Interpreting WCSS </font>

A small value of WCSS indicates

* observations are close to their centroids,
* clusters are compact,
* clustering quality is generally good.

Conversely,

a large WCSS indicates

* observations are widely dispersed,
* clusters are less compact.

Thus,

K-Means seeks the partition that minimizes the total within-cluster variation.

---



# <font color='orange'> 5. Why Are Centroids the Mean? </font>

Suppose the cluster assignments are fixed.

What point minimizes the sum of squared distances?

The answer is

the arithmetic mean of the observations.

For a cluster containing

$$
n_k
$$

observations,

the centroid is

$$
\boxed{
\boldsymbol{\mu}_k
=
\frac1{n_k}
\sum_{\mathbf{x}_i\in C_k}
\mathbf{x}_i.
}
$$

This mathematical property explains why the algorithm is called **K-Means**.

The centroid is not chosen arbitrarily—it is the point that minimizes the sum of squared Euclidean distances within the cluster.

---



# <font color='orange'> 6. Lloyd's Algorithm </font>

Directly minimizing WCSS over all possible cluster assignments is computationally infeasible because the number of possible partitions grows exponentially with the number of observations.

Instead,

K-Means uses an iterative optimization procedure known as **Lloyd's Algorithm**.

The algorithm alternates between two steps.

---

### Assignment Step

Assign each observation to the nearest centroid.

```
Observation

↓

Nearest Centroid

↓

Cluster Assignment
```

This step decreases (or leaves unchanged) the WCSS because each observation is assigned to the cluster that minimizes its squared distance.

---

### Update Step

Recompute each centroid as the mean of the observations currently assigned to its cluster.

```
Cluster Members

↓

Arithmetic Mean

↓

Updated Centroid
```

This step also decreases (or leaves unchanged) the WCSS because the mean minimizes the sum of squared distances within the cluster.

---



# <font color='orange'> 7. Why Does Lloyd's Algorithm Converge? </font>

Each iteration consists of

1. an assignment step,
2. an update step.

Neither step can increase the WCSS.

Therefore,

the objective function either

* decreases,
* or remains unchanged.

Since the WCSS is bounded below by zero,

the algorithm must eventually terminate.

Conceptually,

```
Iteration 1

↓

Lower WCSS

↓

Iteration 2

↓

Lower WCSS

↓

...

↓

Convergence
```

---



# <font color='orange'> 8. A Numerical Example </font>

Suppose a cluster contains the following one-dimensional observations

$$
2,\;4,\;6.
$$

The centroid is

$$
\mu
=
\frac{2+4+6}{3}
=
4.
$$

The squared distances are

$$
(2-4)^2=4,
$$

$$
(4-4)^2=0,
$$

$$
(6-4)^2=4.
$$

Hence,

$$
\boxed{
\text{WCSS}
=
4+0+4
=
8.
}
$$

If we moved the centroid away from the mean,

the WCSS would increase.

This simple example illustrates why the arithmetic mean is the optimal cluster representative.

---



# <font color='orange'> 9. Computational Complexity </font>

For each iteration,

the assignment step computes the distance from every observation to every centroid.

For

* \(n\) observations,
* \(K\) clusters,
* \(d\) features,

the computational complexity is approximately

$$
\boxed{
\mathcal{O}(nKd).
}
$$

If the algorithm requires

$$
T
$$

iterations,

the total complexity becomes

$$
\boxed{
\mathcal{O}(nKdT).
}
$$

In practice,

\(T\) is usually modest, making K-Means efficient for many applications.

---



# <font color='orange'> 10. Local Optimum </font>

Although Lloyd's Algorithm always decreases WCSS,

it is **not guaranteed** to find the global minimum.

Different initial centroid positions may lead to different local minima.

This explains why modern implementations often run K-Means multiple times with different initializations and retain the solution with the smallest WCSS.

---



# <font color='orange'> 11. Relationship Between Intuition and Mathematics </font>

The intuitive description from the previous chapter now has a precise mathematical interpretation.

| Intuition | Mathematical Interpretation |
| :--- | :--- |
| Find the nearest centroid | Minimize squared distance |
| Move the centroid | Compute the cluster mean |
| Improve the clustering | Reduce WCSS |
| Stop iterating | WCSS no longer decreases |

Thus, every step of Lloyd's Algorithm directly contributes to minimizing the objective function.

---

# <font color='purple'> 12. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| WCSS | Sum of squared distances between observations and their assigned centroids |
| Objective | Minimize WCSS |
| Centroid | Arithmetic mean of the observations in a cluster |
| Assignment Step | Assign each observation to the nearest centroid |
| Update Step | Recompute centroids as cluster means |
| Lloyd's Algorithm | Iterative optimization procedure for K-Means |
| Convergence | WCSS decreases until a stable solution is reached |
| Limitation | Converges to a local, not necessarily global, optimum |

> **Key Insight:** K-Means clustering is fundamentally an optimization algorithm that seeks to minimize the Within-Cluster Sum of Squares (WCSS), a measure of cluster compactness. Lloyd's Algorithm achieves this objective through alternating assignment and update steps, each of which reduces or preserves the WCSS. Although the algorithm is guaranteed to converge, the final solution depends on the initial centroid positions, making initialization an important practical consideration in K-Means clustering.
