# <font color='blue'> Principal Component Analysis (PCA): Foundations and Intuition </font>

Principal Component Analysis (PCA) is one of the most widely used techniques for dimensionality reduction in machine learning and statistics. Its objective is to transform a high-dimensional dataset into a lower-dimensional representation while preserving as much of the important information as possible. Unlike feature selection methods, which retain a subset of the original variables, PCA constructs entirely new variables known as **principal components** that capture the dominant patterns of variation within the data.

The key insight underlying PCA is that many datasets contain substantial redundancy because several variables are highly correlated. Rather than analysing these correlated variables individually, PCA identifies new orthogonal directions along which the data exhibit the greatest variation. By projecting the observations onto these directions, the dimensionality of the dataset can often be reduced considerably with only a small loss of information.

This chapter develops an intuitive understanding of Principal Component Analysis before introducing its mathematical formulation in subsequent chapters.

---



# <font color='orange'> 1. What is PCA? </font>

Principal Component Analysis is a

**feature extraction**

technique.

Rather than selecting a subset of the original features,

PCA constructs

new variables

called

**principal components**.

Conceptually,

```
Original Features

↓

Transformation

↓

Principal Components
```

These new variables summarize the most important patterns present in the data.

---



# <font color='orange'> 2. The Central Idea </font>

Suppose a dataset contains

many correlated variables.

Instead of analysing every variable separately,

PCA asks

> **Can we describe the data using fewer variables without losing much information?**

If the answer is yes,

the data can be compressed into a lower-dimensional representation.

---



# <font color='orange'> 3. What Does "Information" Mean? </font>

In PCA,

information is measured by

**variance**.

Recall that variance measures

how much a variable changes.

```
Large Variance

↓

Many Differences

↓

More Information
```

Conversely,

```
Small Variance

↓

Little Variation

↓

Less Information
```

PCA therefore seeks directions along which the data vary the most.

---



# <font color='orange'> 4. Why Variance? </font>

Consider two variables.

```
Variable A

↓

Almost Constant
```

```
Variable B

↓

Large Spread
```

Variable B contains more information because it better distinguishes between observations.

PCA therefore prioritizes directions with the greatest variance.

---



# <font color='orange'> 5. A Geometric Interpretation </font>

Imagine plotting observations in two dimensions.

```
        ●

      ●

    ●

  ●

●
```

The observations lie approximately along

one diagonal direction.

Although the data are two-dimensional,

most of the variation occurs

along a single line.

Instead of using

two coordinates,

we could describe the data using

one new coordinate

aligned with this direction.

This is precisely the idea behind PCA.

---



# <font color='orange'> 6. Principal Components </font>

A **principal component**

is a new axis

constructed from the original variables.

The first principal component

(PC1)

points in the direction of

maximum variance.

```
Data

↓↓↓↓

PC1
```

The second principal component

(PC2)

captures

the largest remaining variance

subject to being perpendicular to PC1.

---



# <font color='orange'> 7. Why Must Principal Components Be Orthogonal? </font>

Each principal component should describe

new information.

If two components pointed in nearly the same direction,

they would capture largely the same variation.

To avoid redundancy,

PCA requires every principal component to be

**orthogonal**

(perpendicular)

to all previous components.

Conceptually,

```
PC1

→

PC2

↑
```

Each component therefore contributes unique information.

---



# <font color='orange'> 8. Projection </font>

Once the principal components have been identified,

every observation is projected onto these new axes.

Conceptually,

```
Original Data

↓

Projection

↓

Principal Component Coordinates
```

The original variables are replaced by

coordinates along the principal components.

---



# <font color='orange'> 9. Dimensionality Reduction </font>

Suppose a dataset contains

100 variables.

After applying PCA,

we may find that

the first

10 principal components

capture

95% of the total variance.

Instead of analysing

100 variables,

we analyse

only 10 principal components.

```
100 Features

↓

PCA

↓

10 Principal Components

↓

Approximately Same Information
```

This greatly simplifies subsequent analyses.

---



# <font color='orange'> 10. Feature Extraction vs Feature Selection </font>

PCA performs

**feature extraction**,

not

feature selection.

Feature Selection

```
Original Features

↓

Choose Best Features
```

Original variables remain unchanged.

---

Feature Extraction

```
Original Features

↓

Construct New Variables
```

The principal components are

linear combinations

of the original variables.

They do not correspond to any single original feature.

---

# <font color='orange'> 12. Limitations of PCA </font>

Although PCA is extremely useful,

it also has limitations.

* Captures only linear relationships.
* Sensitive to feature scaling.
* Principal components may be difficult to interpret.
* Assumes variance corresponds to useful information.
* May discard low-variance features that are nevertheless important for certain tasks.

These limitations motivate more advanced nonlinear dimensionality reduction techniques.

# <font color='blue'> The Mathematics of PCA: Covariance Matrices, Eigenvectors, and Eigenvalues </font>

The previous chapter introduced the intuition behind Principal Component Analysis (PCA), explaining that the algorithm identifies new coordinate axes along which the data exhibit the greatest variation. The remaining question is how these directions are determined mathematically.

The answer lies in the covariance matrix of the dataset. The covariance matrix summarizes how pairs of variables vary together, capturing both their individual variances and their mutual relationships. By analysing this matrix through an eigenvalue decomposition, PCA identifies a set of orthogonal directions, known as **eigenvectors**, along which the variance of the data is maximized. The corresponding **eigenvalues** quantify how much of the total variance is explained by each direction.

This chapter develops the mathematical foundations of PCA by introducing the covariance matrix, explaining the role of eigenvectors and eigenvalues, and showing how these concepts combine to identify the principal components.

---



# <font color='orange'> 1. Why Does PCA Need Mathematics? </font>

In the previous chapter,

we stated that PCA finds

the directions of maximum variance.

The natural question is

> **How can a computer determine these directions automatically?**

The answer involves

```
Dataset

↓

Covariance Matrix

↓

Eigenvalue Decomposition

↓

Principal Components
```

This sequence forms the mathematical foundation of PCA.

---



# <font color='orange'> 2. Organizing the Dataset </font>

Suppose our dataset contains

* \(n\) observations,
* \(p\) features.

The data are arranged in a matrix

$$
\mathbf{X}
=
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1p}\\
x_{21} & x_{22} & \cdots & x_{2p}\\
\vdots & \vdots & \ddots & \vdots\\
x_{n1} & x_{n2} & \cdots & x_{np}
\end{bmatrix}.
$$

Each row represents one observation,

while each column corresponds to one feature.

Before applying PCA,

the features are usually centred by subtracting their means.

---



# <font color='orange'> 3. The Covariance Matrix </font>

The first mathematical object computed by PCA is the covariance matrix.

For centred data,

the covariance matrix is

$$
\boxed{
\mathbf{\Sigma}
=
\frac{1}{n-1}
\mathbf{X}^\top
\mathbf{X}.
}
$$

The covariance matrix is a

\(p \times p\)

square matrix.

Each element

$$
\Sigma_{ij}
$$

measures how features

\(i\)

and

\(j\)

vary together.

---



# <font color='orange'> 4. Interpreting the Covariance Matrix </font>

The covariance matrix contains two types of information.

### Diagonal Elements

The diagonal entries are the variances of the individual features.

$$
\Sigma_{ii}
=
\mathrm{Var}(X_i).
$$

Large values indicate that the corresponding feature varies substantially.

---

### Off-Diagonal Elements

The off-diagonal entries are covariances.

$$
\Sigma_{ij}
=
\mathrm{Cov}(X_i,X_j).
$$

These quantify the linear relationship between different features.

Positive covariance indicates that two variables tend to increase together,

while negative covariance indicates that one tends to increase as the other decreases.

---



# <font color='orange'> 5. Why Covariance Matters </font>

Suppose two features are highly correlated.

```
Feature A

↑

↓

Feature B
```

They contain largely the same information.

PCA seeks new directions that summarize this shared variation,

thereby reducing redundancy.

The covariance matrix therefore captures precisely the information needed to identify redundant features.

---



# <font color='orange'> 6. Eigenvectors </font>

Once the covariance matrix has been constructed,

PCA computes its **eigenvectors**.

Each eigenvector represents

a direction in the feature space.

Conceptually,

```
Feature Space

↓

Direction

↓

Eigenvector
```

These directions define the new coordinate system used by PCA.

---



# <font color='orange'> 7. Eigenvalues </font>

Associated with every eigenvector is an

**eigenvalue**.

The eigenvalue measures

how much variance lies along the corresponding eigenvector.

Large eigenvalue

```
↓

Important Direction
```

Small eigenvalue

```
↓

Little Information
```

PCA therefore ranks the principal components by decreasing eigenvalue.

---



# <font color='orange'> 8. Principal Components </font>

The principal components are simply

the eigenvectors of the covariance matrix,

ordered according to their eigenvalues.

Specifically,

* **PC1** is the eigenvector with the largest eigenvalue.
* **PC2** is the eigenvector with the second-largest eigenvalue.
* And so on.

Because the covariance matrix is symmetric,

its eigenvectors are mutually orthogonal.

This ensures that each principal component captures new information not already explained by previous components.

---



# <font color='orange'> 9. Explained Variance </font>

Suppose the covariance matrix has eigenvalues

$$
\lambda_1,\lambda_2,\ldots,\lambda_p.
$$

The total variance in the dataset is

$$
\boxed{
\sum_{i=1}^{p}\lambda_i.
}
$$

The proportion of variance explained by the \(i\)-th principal component is

$$
\boxed{
\frac{\lambda_i}
{\sum_j\lambda_j}.
}
$$

This quantity is known as the **explained variance ratio**.

It indicates how much information is retained by each principal component.

---



# <font color='orange'> 10. A Simple Example </font>

Suppose the covariance matrix has eigenvalues

$$
6,\;3,\;1.
$$

The total variance is

$$
6+3+1=10.
$$

The explained variance ratios are

| Principal Component | Explained Variance |
| :--- | ---: |
| PC1 | 60% |
| PC2 | 30% |
| PC3 | 10% |

Retaining only the first two principal components preserves

90%

of the total variance.

---



# <font color='orange'> 11. Why Eigenvectors Give the Best Directions </font>

Among all possible directions in feature space,

the first eigenvector identifies the direction along which the projected data have the greatest variance.

The second eigenvector identifies the direction of the greatest remaining variance,

subject to being orthogonal to the first.

This process continues until all principal components have been identified.

Thus,

PCA finds the optimal sequence of orthogonal directions that successively maximize variance.

---



# <font color='orange'> 12. PCA Workflow </font>

The complete mathematical procedure is

```
Raw Data

↓

Center Features

↓

Compute Covariance Matrix

↓

Eigenvalue Decomposition

↓

Sort Eigenvalues

↓

Select Leading Components

↓

Project Data

↓

Lower-Dimensional Representation
```

Each step builds directly on the previous one.

---



# <font color='orange'> 13. Practical Notes </font>

In practice,

modern software often computes PCA using the **Singular Value Decomposition (SVD)** rather than explicitly forming the covariance matrix.

SVD is

* numerically more stable,
* computationally efficient,
* particularly advantageous for large or high-dimensional datasets.

Nevertheless,

the covariance matrix formulation remains the most intuitive way to understand the mathematical principles underlying PCA.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Covariance Matrix | Summarizes variances and covariances between features |
| Eigenvector | Direction in feature space defining a principal component |
| Eigenvalue | Variance explained by the corresponding eigenvector |
| Principal Component | Eigenvector ordered by decreasing eigenvalue |
| Explained Variance Ratio | Fraction of total variance explained by a principal component |
| Orthogonality | Ensures principal components capture unique information |
| Goal | Find the directions that maximize variance |

> **Key Insight:** Principal Component Analysis transforms a dataset by analysing the covariance matrix of its features. The eigenvectors of the covariance matrix define the principal components, while the corresponding eigenvalues measure the amount of variance captured by each component. By ranking the components according to their explained variance, PCA provides a mathematically optimal way to represent high-dimensional data using a smaller number of orthogonal variables that preserve as much information as possible.

# <font color='blue'> The Principal Component Analysis (PCA) Algorithm: Step-by-Step </font>

The previous chapter introduced the mathematical foundations of Principal Component Analysis (PCA), explaining how the covariance matrix, eigenvectors, and eigenvalues determine the directions of maximum variance within a dataset. These concepts now come together to form the complete PCA algorithm.

The objective of PCA is to transform the original dataset into a new coordinate system whose axes, the principal components, capture the greatest possible variance. The transformation is performed through a sequence of well-defined mathematical operations, beginning with feature standardization and ending with the projection of the original observations onto a lower-dimensional subspace.

This chapter presents the complete PCA workflow, explains the purpose of each computational step, and illustrates how the transformed dataset is obtained.

---



# <font color='orange'> 1. Overview of the PCA Algorithm </font>

The PCA algorithm consists of the following steps.

```
Raw Dataset

↓

Standardize Features

↓

Compute Covariance Matrix

↓

Eigenvalue Decomposition

↓

Sort Eigenvalues

↓

Select Principal Components

↓

Construct Projection Matrix

↓

Project the Data

↓

Lower-Dimensional Dataset
```

Each step contributes to identifying the directions that preserve the greatest amount of information.

---



# <font color='orange'> 2. Step 1 — Standardize the Features </font>

Before applying PCA,

the features should usually be standardized.

For each feature,

$$
\boxed{
z_i
=
\frac{x_i-\mu}{\sigma},
}
$$

where

* \(\mu\) is the feature mean,
* \(\sigma\) is the feature standard deviation.

Standardization ensures that

* every feature has mean zero,
* every feature has unit variance.

---

### Why is Standardization Important?

Suppose we have two variables.

| Feature | Typical Values |
| :--- | ---: |
| Height (m) | 1.5–2.0 |
| Annual Income (£) | 20,000–200,000 |

Without standardization,

the variable with the larger numerical scale dominates the covariance matrix and, consequently, the principal components.

Standardization gives every feature an equal opportunity to influence the analysis.

---



# <font color='orange'> 3. Step 2 — Compute the Covariance Matrix </font>

After centering (and usually standardizing) the data,

compute the covariance matrix.

For a dataset containing

\(n\)

observations,

$$
\boxed{
\Sigma
=
\frac{1}{n-1}
X^T X.
}
$$

The covariance matrix summarizes

* feature variances,
* pairwise covariances.

It captures the relationships between all variables.

---



# <font color='orange'> 4. Step 3 — Perform the Eigenvalue Decomposition </font>

Next,

compute the eigenvalues and eigenvectors of the covariance matrix.

Mathematically,

$$
\boxed{
\Sigma v
=
\lambda v.
}
$$

The solution produces

* eigenvectors

→ principal component directions,

* eigenvalues

→ variance explained along those directions.

---



# <font color='orange'> 5. Step 4 — Sort the Eigenvalues </font>

The eigenvalues are arranged in descending order.

$$
\lambda_1
>
\lambda_2
>
\cdots
>
\lambda_p.
$$

The corresponding eigenvectors are reordered in exactly the same way.

Consequently,

```
Largest Eigenvalue

↓

First Principal Component

↓

Most Information
```

---



# <font color='orange'> 6. Step 5 — Compute the Explained Variance </font>

The total variance equals

$$
\boxed{
\sum_i \lambda_i.
}
$$

The explained variance ratio for component

\(i\)

is

$$
\boxed{
\frac{\lambda_i}
{\sum_j\lambda_j}.
}
$$

Example

| Component | Eigenvalue | Explained Variance |
| :--- | ---: | ---: |
| PC1 | 5.6 | 56% |
| PC2 | 2.9 | 29% |
| PC3 | 1.0 | 10% |
| PC4 | 0.5 | 5% |

This information helps determine how many components should be retained.

---



# <font color='orange'> 7. Step 6 — Select the Principal Components </font>

Rather than keeping every principal component,

we retain only the leading components.

Typical criteria include

* explaining 90–95% of the total variance,
* the scree plot,
* the cumulative explained variance curve,
* domain-specific considerations.

Suppose

```
PC1

PC2

PC3
```

together explain

95%

of the variance.

The remaining components may then be discarded.

---



# <font color='orange'> 8. Step 7 — Construct the Projection Matrix </font>

Suppose we retain

\(k\)

principal components.

Arrange the corresponding eigenvectors as columns of a matrix,

$$
\boxed{
W
=
[v_1 \; v_2 \; \cdots \; v_k].
}
$$

This matrix defines the new coordinate system.

---



# <font color='orange'> 9. Step 8 — Project the Data </font>

The original dataset is transformed into the new feature space by

$$
\boxed{
Z
=
XW.
}
$$

where

* \(X\) is the standardized data matrix,
* \(W\) is the projection matrix,
* \(Z\) contains the principal component scores.

The rows of

\(Z\)

represent the observations expressed in terms of the new principal components.

---



# <font color='orange'> 10. A Simple Numerical Example </font>

Suppose PCA produces

three eigenvalues

$$
8,\;1.5,\;0.5.
$$

The total variance is

$$
10.
$$

The explained variance ratios are

| Component | Explained Variance |
| :--- | ---: |
| PC1 | 80% |
| PC2 | 15% |
| PC3 | 5% |

Keeping only the first two components preserves

95%

of the information while reducing the dimensionality from three variables to two.

---



# <font color='orange'> 11. Covariance Matrix vs Singular Value Decomposition </font>

The covariance matrix approach is conceptually simple and is commonly used to explain PCA.

In practice, however, numerical libraries often compute PCA using the **Singular Value Decomposition (SVD)** of the data matrix.

### Covariance Matrix Approach

```
Standardized Data

↓

Covariance Matrix

↓

Eigenvalue Decomposition

↓

Principal Components
```

### Singular Value Decomposition

```
Standardized Data

↓

SVD

↓

Principal Components
```

SVD is generally preferred because it is

* numerically more stable,
* computationally efficient,
* well suited to high-dimensional datasets.

Although the computations differ,

both approaches produce the same principal components (up to sign changes under standard conditions).

# <font color='orange'> 13. Complete PCA Workflow </font>

```
Raw Data

↓

Handle Missing Values

↓

Standardize Features

↓

Compute Covariance Matrix
(or perform SVD)

↓

Compute Eigenvalues and Eigenvectors

↓

Sort Components

↓

Choose Number of Components

↓

Project Data

↓

Reduced Dataset

↓

Machine Learning / Visualization
```

This workflow summarizes the complete PCA procedure from raw data to the final reduced representation.

# <font color='purple'> 14. Conceptual Summary </font>

| Step | Purpose |
| :--- | :--- |
| Standardize Features | Remove scale differences |
| Compute Covariance Matrix | Measure relationships between features |
| Eigenvalue Decomposition | Identify directions of maximum variance |
| Sort Eigenvalues | Rank principal components |
| Select Components | Retain the most informative directions |
| Construct Projection Matrix | Define the new coordinate system |
| Project Data | Obtain the lower-dimensional representation |
| SVD (Practical Alternative) | Compute PCA efficiently and robustly |

> **Key Insight:** Principal Component Analysis transforms a high-dimensional dataset into a lower-dimensional representation through a sequence of mathematically well-defined steps. After standardizing the features, PCA identifies the directions of greatest variance by analysing the covariance matrix (or, equivalently in practice, by performing a Singular Value Decomposition). The leading principal components are then selected according to their explained variance and used to project the original observations into a new coordinate system that preserves the maximum possible information with fewer dimensions.

# <font color='blue'> Interpreting Principal Component Analysis </font>

After applying Principal Component Analysis (PCA), the original features are replaced by a new set of variables called **principal components**. While the computational procedure for obtaining these components is well defined, interpreting their meaning requires careful examination of several quantities produced by the algorithm, including the principal component scores, component loadings, eigenvalues, explained variance ratios, and cumulative explained variance.

Interpreting these quantities enables us to understand how much information has been retained, determine the appropriate number of principal components to keep, identify which original variables contribute most strongly to each component, and assess whether dimensionality reduction has been successful.

This chapter introduces the principal outputs of PCA and explains how they should be interpreted in practical machine learning applications.

---



# <font color='orange'> 1. The Outputs of PCA </font>

After fitting a PCA model,

we obtain several quantities.

```
Dataset

↓

PCA

↓

Principal Components

↓

Scores

↓

Loadings

↓

Eigenvalues

↓

Explained Variance
```

Each output provides different information about the transformed dataset.

---



# <font color='orange'> 2. Principal Component Scores </font>

The **scores** are the coordinates of each observation in the new principal component space.

Suppose the original dataset contains

100 observations.

After projecting onto two principal components,

each observation becomes

```
Observation

↓

(PC1, PC2)
```

rather than

```
(x₁,x₂,...,xₚ)
```

The scores therefore describe

where each observation lies

in the reduced-dimensional space.

---

### Interpretation

Observations that have similar scores

are close together

in the transformed space.

Consequently,

score plots are widely used for

* visualization,
* clustering,
* identifying outliers.

---



# <font color='orange'> 3. Component Loadings </font>

The **loadings**

describe

how strongly each original feature contributes

to each principal component.

Mathematically,

the loadings are the coefficients of the eigenvectors.

Example

| Feature | PC1 Loading |
| :--- | ---: |
| Height | 0.63 |
| Weight | 0.61 |
| Age | -0.47 |

---

### Interpretation

Large positive loading

↓

feature contributes strongly.

Large negative loading

↓

feature contributes strongly

in the opposite direction.

Loadings close to zero

↓

little influence.

---



# <font color='orange'> 4. Scores vs Loadings </font>

These quantities are often confused.

| Scores | Loadings |
| :--- | :--- |
| Describe observations | Describe features |
| Coordinates of data points | Coefficients of principal components |
| Used for visualization | Used for interpretation |

Remember

```
Scores

↓

Observations
```

```
Loadings

↓

Variables
```

---



# <font color='orange'> 5. Explained Variance </font>

Every principal component has

an associated eigenvalue.

The eigenvalue measures

how much variance

that component explains.

Suppose

```
Eigenvalues

6

3

1
```

Total variance

```
10
```

Explained variance becomes

| Component | Variance Explained |
| :--- | ---: |
| PC1 | 60% |
| PC2 | 30% |
| PC3 | 10% |

The first component therefore captures most of the information.

---



# <font color='orange'> 6. Cumulative Explained Variance </font>

Rather than considering individual components,

we often examine

the cumulative explained variance.

Example

| Component | Cumulative Variance |
| :--- | ---: |
| PC1 | 60% |
| PC1 + PC2 | 90% |
| PC1 + PC2 + PC3 | 100% |

This tells us

how much information has been retained

after keeping several principal components.

---



# <font color='orange'> 7. Choosing the Number of Components </font>

Several strategies are commonly used.

### Explained Variance Threshold

Retain enough components to explain

```
90–95%
```

of the total variance.

---

### Kaiser Criterion

For standardized data,

retain components whose eigenvalues exceed

```
1
```

This criterion is most commonly applied when PCA is performed on the correlation matrix and should be used with caution outside that setting.

---

### Scree Plot

Plot

```
Eigenvalue

↓

Component Number
```

Conceptually,

```
Eigenvalue

│\

│ \

│  \__

│      \___

└────────────
```

The "elbow"

indicates

where additional components contribute relatively little additional variance.

---



# <font color='orange'> 8. Biplots </font>

A **biplot**

Conceptually,

```
Scores

+

Loadings

↓

Biplot
```

Biplots provide a powerful visual summary of

* observation relationships,
* feature contributions,
* dominant patterns.

---



# <font color='orange'> 9. Interpreting Principal Components </font>

# <font color='orange'> 10. Sign Ambiguity </font>

An important property of PCA is

that the sign of every eigenvector is arbitrary.

For example,

```
PC1

↓

(+0.7,+0.7)
```

and

```
(-0.7,-0.7)
```

represent exactly the same principal component.

Consequently,

only

the relative magnitudes

of the loadings are meaningful.

The overall sign should not be interpreted.

---



# <font color='orange'> 11. Common Pitfalls </font>

Several mistakes frequently occur.

### Interpreting Every Component

Later principal components

often explain very little variance.

---

### Ignoring Feature Scaling

Without standardization,

large-scale variables dominate.

---

### Assuming Components Have Physical Meaning

Principal components are mathematical constructs.

Some may have clear interpretations,

others may not.

---

### Assuming Low Variance Means Unimportant

Low-variance components can still contain valuable information,

particularly for anomaly detection or certain scientific applications.

Variance alone should not always determine importance.

> **Key Insight:** The value of Principal Component Analysis lies not only in reducing dimensionality but also in providing interpretable summaries of complex datasets. Scores reveal the positions of observations in the reduced feature space, loadings identify the variables that define each principal component, and explained variance quantifies the amount of information retained. Together, these outputs enable practitioners to balance dimensionality reduction with information preservation while gaining insight into the underlying structure of the data.

# <font color='orange'> 2. Importing the Required Libraries </font>

The required packages include

```python
import numpy as np
import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

import matplotlib.pyplot as plt
```

These libraries provide

* numerical computation,
* data manipulation,
* feature scaling,
* PCA,
* visualization.

---



# <font color='orange'> 3. Loading the Dataset </font>

Suppose the dataset has already been loaded into

```python
df
```

Separate the features from any target variable.

```python
X = df.drop(columns=["target"])
```

For unsupervised learning,

a target variable is generally **not** required.

If a labelled dataset is being used purely for visualization or later evaluation,

the target labels should **not** be included when fitting PCA.

---



# <font color='orange'> 4. Standardizing the Features </font>

Feature scaling is usually the most important preprocessing step.

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

After scaling,

every feature has

* mean zero,
* standard deviation one.

This prevents variables with large numerical ranges from dominating the principal components.

---



# <font color='orange'> 5. Fitting PCA </font>

Create the PCA model.

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)
```

By default,

Scikit-Learn computes every principal component.

The transformed observations are stored in

```python
X_pca
```

Each column corresponds to one principal component.

---



# <font color='orange'> 6. Examining the Explained Variance </font>

The explained variance ratio is obtained using

```python
pca.explained_variance_ratio_
```

Example

```python
print(
    pca.explained_variance_ratio_
)
```

Possible output

```
[0.54
 0.27
 0.12
 0.05
 0.02]
```

Interpretation

* PC1 explains 54% of the variance.
* PC2 explains 27%.
* Together they explain 81%.

---



# <font color='orange'> 7. Cumulative Explained Variance </font>

The cumulative explained variance is computed as

```python
cumulative_variance = np.cumsum(
    pca.explained_variance_ratio_
)
```

Example

```python
print(cumulative_variance)
```

Possible output

```
[0.54
 0.81
 0.93
 0.98
 1.00]
```

This information is commonly used to determine how many principal components should be retained.

---



# <font color='orange'> 8. Choosing the Number of Components </font>

Suppose we wish to retain

95%

of the variance.

Scikit-Learn allows this directly.

```python
pca = PCA(
    n_components=0.95
)

X_reduced = pca.fit_transform(
    X_scaled
)
```

The algorithm automatically selects the smallest number of principal components whose cumulative explained variance is at least 95%.

---



# <font color='orange'> 9. Examining the Component Loadings </font>

The principal component directions are stored in

```python
pca.components_
```

Example

```python
loadings = pd.DataFrame(

    pca.components_.T,

    index=X.columns,

    columns=[
        f"PC{i+1}"
        for i in range(
            pca.n_components_
        )
    ]
)

print(loadings)
```

Large positive or negative values indicate that the corresponding feature contributes strongly to the principal component.

---



# <font color='orange'> 10. Visualizing the Principal Components </font>

A two-dimensional scatter plot can be constructed using the first two principal components.

```python
plt.figure(figsize=(7,5))

plt.scatter(

    X_pca[:,0],

    X_pca[:,1]

)

plt.xlabel("PC1")

plt.ylabel("PC2")

plt.title(
    "Projection onto the First Two Principal Components"
)

plt.show()
```

This visualization often reveals

* clusters,
* trends,
* outliers,

that are difficult to identify in the original high-dimensional feature space.

---



# <font color='orange'> 11. Creating a Scree Plot </font>

The scree plot displays the explained variance of each principal component.

```python
plt.figure(figsize=(7,5))

plt.plot(

    range(

        1,

        len(
            pca.explained_variance_ratio_
        )+1

    ),

    pca.explained_variance_ratio_,

    marker="o"

)

plt.xlabel(
    "Principal Component"
)

plt.ylabel(
    "Explained Variance Ratio"
)

plt.title(
    "Scree Plot"
)

plt.show()
```

The scree plot helps identify

where additional components contribute relatively little information.

When applying PCA,

remember the following guidelines.

* Interpret cumulative explained variance before selecting the number of components.

* Use score plots to identify clusters or outliers.
* Remember that PCA captures only linear relationships.
