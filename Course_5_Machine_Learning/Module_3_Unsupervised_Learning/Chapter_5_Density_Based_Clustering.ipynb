# <font color='blue'> Density-Based Clustering: DBSCAN </font>

The clustering algorithms studied thus far have relied primarily on partitioning observations around centroids or constructing hierarchical relationships between clusters. Although these approaches are effective for many datasets, they often struggle when clusters have irregular shapes, varying sizes, or contain significant numbers of outliers. Furthermore, methods such as K-Means require the number of clusters to be specified before the analysis begins.

Density-Based Spatial Clustering of Applications with Noise (DBSCAN) addresses these limitations by defining clusters as regions of high observation density separated by regions of low density. Instead of representing clusters using centroids, DBSCAN identifies dense neighbourhoods of observations and expands clusters outward from these regions. Observations that do not belong to any sufficiently dense region are classified as **noise**, allowing the algorithm to perform clustering and outlier detection simultaneously.

This chapter introduces the intuition behind density-based clustering, explains the fundamental concepts of DBSCAN, and discusses its advantages and limitations.

The radius defining the local neighbourhood around each observation.

```
Point

↓

Circle of Radius ε

↓

Neighbourhood
```

The minimum number of observations required within an ε-neighbourhood for that region to be considered dense.

Typical values include

```
MinPts = 4

or

MinPts = 5
```

A **border point**

# <font color='orange'> 8. Building a Cluster </font>

The algorithm proceeds conceptually as follows.

```
Choose Observation

↓

Core Point?

↓

Yes

↓

Expand Cluster

↓

Add Density-Reachable Points

↓

Continue Expansion

↓

Cluster Complete
```

The process repeats until all observations have been examined.

---

# <font color='orange'> 9. Density Reachability </font>

Clusters grow by connecting neighbouring core points.

```
Core Point

↓

Neighbouring Core Point

↓

Neighbouring Core Point

↓

Same Cluster
```

Thus,

clusters expand through connected regions of sufficiently high density.

# <font color='orange'> 10. Advantages </font>

DBSCAN offers several important advantages.

* Does not require specifying the number of clusters.
* Detects arbitrarily shaped clusters.
* Naturally identifies outliers.
* Robust to isolated noisy observations.
* Performs well when clusters are separated by low-density regions.

These characteristics distinguish DBSCAN from K-Means and Hierarchical Clustering.

---

# <font color='orange'> 11. Limitations </font>

Despite its strengths,

DBSCAN also has limitations.

* Choosing ε can be difficult.
* Sensitive to the choice of MinPts.
* Performs poorly when cluster densities differ substantially.
* High-dimensional data reduce the usefulness of distance-based density estimation.

Consequently,

parameter selection is an important practical consideration.

By introducing the concepts of core points, border points, and noise points, DBSCAN can discover clusters of arbitrary shape while simultaneously identifying anomalous observations. This combination of flexibility and robustness has made DBSCAN one of the most widely used density-based clustering algorithms in modern machine learning.

# <font color='blue'> The DBSCAN Algorithm </font>

The previous chapter introduced the intuition behind Density-Based Spatial Clustering of Applications with Noise (DBSCAN), explaining how clusters are defined as regions of high observation density separated by regions of low density. While this intuition is useful, the algorithm relies on several precise mathematical concepts that determine how clusters are formed and expanded.

DBSCAN identifies clusters by examining the local neighbourhood surrounding each observation. Observations that satisfy a specified density criterion become **core points**, from which clusters are recursively expanded. Observations lying on the boundaries of these dense regions become **border points**, while isolated observations are classified as **noise**.

This chapter formalizes the concepts underlying DBSCAN, presents the complete clustering algorithm, discusses practical parameter selection, and introduces its implementation in Scikit-Learn.

---

# <font color='orange'> 1. Review of the Two Parameters </font>

DBSCAN requires two parameters.

### Epsilon

$$
\varepsilon
$$

defines the radius of the neighbourhood surrounding every observation.

```
Observation

↓

ε-Neighbourhood
```

---

### Minimum Points

```
MinPts
```

defines the minimum number of observations required inside an ε-neighbourhood for that region to be considered dense.

Together,

these two parameters determine the density threshold used to define clusters.

---

# <font color='orange'> 2. ε-Neighbourhood </font>

For an observation

$$
x,
$$

its ε-neighbourhood consists of every observation lying within distance

$$
\varepsilon.
$$

Mathematically,

$$
\boxed{
N_\varepsilon(x)
=
\{
y
:
d(x,y)
\le
\varepsilon
\}.
}
$$

Here,

$$
d(x,y)
$$

may represent Euclidean distance or another appropriate distance metric.

The ε-neighbourhood is therefore the local region examined around every observation.

---

# <font color='orange'> 3. Core Points </font>

An observation is called a **core point** if its ε-neighbourhood contains at least

```
MinPts
```

observations (including the point itself).

Mathematically,

$$
\boxed{
|N_\varepsilon(x)|
\ge
\text{MinPts}.
}
$$

Conceptually,

```
Many Nearby Observations

↓

High Density

↓

Core Point
```

Core points serve as the seeds from which clusters are expanded.

---

# <font color='orange'> 4. Border Points </font>

A border point satisfies two conditions.

* It lies inside the ε-neighbourhood of a core point.
* It does not itself contain enough neighbours to become a core point.

Conceptually,

```
Core Point

↓

Border Point

↓

Cluster Member
```

Border points belong to clusters,

but they do not initiate further cluster expansion.

---

# <font color='orange'> 5. Noise Points </font>

An observation is classified as **noise** if

* it is not a core point,
* and it does not lie within the ε-neighbourhood of any core point.

Conceptually,

```
Isolated Observation

↓

Low Density

↓

Noise
```

Unlike K-Means,

noise points remain unassigned.

---

# <font color='orange'> 6. Density Reachability </font>

Suppose

$$
x
$$

is a core point.

Another observation

$$
y
$$

is said to be **directly density-reachable** from

$$
x
$$

if

$$
y
\in
N_\varepsilon(x).
$$

Conceptually,

```
Core Point

↓

Neighbour Within ε

↓

Directly Density-Reachable
```

Only core points can directly reach neighbouring observations.

---

# <font color='orange'> 7. Density Connectivity </font>

Two observations belong to the same cluster if they are connected through a sequence of density-reachable core points.

Conceptually,

```
Core

↓

Core

↓

Core

↓

Border

↓

Same Cluster
```

Clusters therefore grow through connected dense regions rather than through centroid assignment.

---

# <font color='orange'> 8. The DBSCAN Algorithm </font>

The complete algorithm proceeds as follows.

---

### Step 1

Select an unvisited observation.

---

### Step 2

Compute its ε-neighbourhood.

---

### Step 3

If the neighbourhood contains fewer than MinPts observations,

label the observation temporarily as noise.

---

### Step 4

Otherwise,

declare it a core point

and begin a new cluster.

---

### Step 5

Add every directly density-reachable observation to the cluster.

---

### Step 6

For every newly discovered core point,

repeat the neighbourhood search.

Continue expanding the cluster until

no additional density-reachable observations remain.

---

### Step 7

Return to Step 1

until every observation has been visited.

---

# <font color='orange'> 9. Algorithm Flow </font>

```
Choose Observation

↓

Visited?

↓

No

↓

Compute ε-Neighbourhood

↓

Core Point?

↓

No

↓

Noise

```

or

```
Yes

↓

Create Cluster

↓

Expand Cluster

↓

Repeat

↓

Cluster Complete
```

---

# <font color='orange'> 10. Choosing ε </font>

Selecting

$$
\varepsilon
$$

is often the most challenging part of DBSCAN.

If

$$
\varepsilon
$$

is too small,

```
Many Points

↓

Become Noise

↓

Too Many Clusters
```

---

If

$$
\varepsilon
$$

is too large,

```
Separate Clusters

↓

Merge Together

↓

Too Few Clusters
```

Careful parameter selection is therefore essential.

---

# <font color='orange'> 11. The k-Distance Plot </font>

A common method for selecting

$$
\varepsilon
$$

is the **k-distance plot**.

Procedure

1. Compute the distance to the \(k\)-th nearest neighbour for every observation.
2. Sort these distances.
3. Plot the sorted distances.

Conceptually,

```
Distance

│

│

│       /

│     /

│   /

│__/

──────────────
```

The point where the curve begins to rise sharply (the "knee" or "elbow") is often chosen as a suitable value of

$$
\varepsilon.
$$

General practical guidelines include

| Data Dimension | Typical MinPts |
| :---: | :---: |
| 2 | 4 |
| 3 | 5 |
| Higher dimensions | At least \(2 \times\) the number of dimensions |

These are heuristics rather than strict rules.

Domain knowledge remains valuable.

# <font color='orange'> 13. Computational Complexity </font>

Without specialized indexing,

DBSCAN compares many pairs of observations,

leading to

$$
\boxed{
\mathcal O(n^2)
}
$$

time complexity.

Using spatial indexing structures such as

* KD-Trees,
* Ball Trees,

can substantially improve efficiency for low-dimensional data.

> **Key Insight:** DBSCAN constructs clusters by recursively expanding connected regions of high observation density rather than by optimizing a centroid-based objective function. The concepts of ε-neighbourhoods, core points, border points, and density connectivity provide a mathematically rigorous framework for identifying clusters of arbitrary shape while simultaneously detecting outliers. Although the algorithm depends critically on the choice of ε and MinPts, its ability to discover complex cluster structures without specifying the number of clusters beforehand makes DBSCAN one of the most powerful and widely used density-based clustering algorithms.
