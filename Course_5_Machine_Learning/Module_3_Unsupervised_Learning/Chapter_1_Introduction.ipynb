# <font color='blue'> Introduction to Unsupervised Machine Learning </font>

Machine learning algorithms are commonly categorized into supervised, unsupervised, semi-supervised, and reinforcement learning, depending on the type of information available during training. In supervised learning, algorithms learn a relationship between predictor variables and known target values, enabling them to make predictions for unseen observations. However, in many practical situations, labelled data are unavailable, expensive to obtain, or impossible to define. Instead, only the predictor variables are observed, and the objective becomes discovering meaningful patterns, structures, or relationships within the data itself.

**Unsupervised Machine Learning** addresses this class of problems. Rather than learning from labelled examples, unsupervised learning algorithms analyze unlabelled datasets to identify hidden structures, natural groupings, lower-dimensional representations, and latent features. These methods play a central role in exploratory data analysis, pattern recognition, recommendation systems, scientific discovery, and numerous other applications where the underlying structure of the data is unknown.

This chapter introduces the fundamental concepts of unsupervised learning, explains how it differs from supervised learning, and provides an overview of the major categories of unsupervised learning algorithms that will be studied throughout this notebook.

# <font color='orange'> 1. What is Machine Learning? </font>

Machine learning is a branch of artificial intelligence that develops algorithms capable of learning patterns from data without being explicitly programmed for every possible situation.

The general workflow is

```
Data

↓

Learning Algorithm

↓

Model

↓

Knowledge or Predictions
```

Depending on the information available during training, machine learning can be divided into several categories.

---



# <font color='orange'> 2. Types of Machine Learning </font>

The four major categories are

| Learning Type | Training Data | Objective |
| :--- | :--- | :--- |
| Supervised Learning | Labelled data | Predict known target values |
| Unsupervised Learning | Unlabelled data | Discover hidden patterns |
| Semi-Supervised Learning | Small amount of labelled data + large amount of unlabelled data | Improve prediction using both |
| Reinforcement Learning | Rewards and penalties | Learn optimal actions through interaction |

This notebook focuses on **unsupervised learning**.

---



# <font color='orange'> 3. What is Unsupervised Learning? </font>

In unsupervised learning,

the dataset contains only predictor variables.

There are **no target labels** indicating the correct answer.

Instead of asking

> "Can we predict the output?"

we ask questions such as

* Are there natural groups in the data?
* Can we simplify the dataset?
* Which observations are unusual?
* Are there hidden relationships?
* Can the data be represented more efficiently?

The objective is therefore

```
Unlabelled Data

↓

Discover Structure

↓

Extract Knowledge
```

---



# <font color='orange'> 4. Supervised vs Unsupervised Learning </font>

The fundamental difference lies in the availability of labels.

### Supervised Learning

```
Features

+

Known Target

↓

Learn Mapping

↓

Predict Target
```

Example

Predict the price of a house using

* size,
* location,
* age,

when the sale prices are already known.

---

### Unsupervised Learning

```
Features Only

↓

Discover Patterns

↓

Clusters

Relationships

Low-Dimensional Representations
```

Example

Group houses into similar neighbourhoods without knowing their prices.

---

# <font color='orange'> 7. Characteristics of Unsupervised Learning </font>

Compared with supervised learning,

unsupervised learning

* does not require labels,
* is primarily exploratory,
* often has no unique "correct" answer,
* relies heavily on similarity and structure,
* is frequently used before supervised learning.

Consequently,

evaluating unsupervised models is generally more difficult than evaluating supervised models.

# <font color='orange'> 10. Limitations </font>

Despite its strengths,

unsupervised learning also presents challenges.

* No ground truth labels for validation.
* Results may be difficult to interpret.
* Different algorithms may produce different structures.
* Sensitive to feature scaling and preprocessing.
* Model evaluation is often less straightforward than in supervised learning.

Careful interpretation and domain knowledge are therefore particularly important.

# <font color='blue'> Types of Unsupervised Machine Learning </font>

The previous chapter introduced the fundamental concepts of unsupervised learning and explained how it differs from supervised learning. Unlike supervised learning, which uses labelled examples to learn predictive relationships, unsupervised learning seeks to discover hidden structure within unlabelled datasets. Because there is no single target variable to predict, different unsupervised learning algorithms are designed to solve different types of problems.

Broadly speaking, unsupervised learning can be divided into several major categories, including clustering, dimensionality reduction, density estimation, association rule learning, matrix factorization, and anomaly detection. Each category addresses a distinct objective and relies on different mathematical principles. Understanding these categories provides a conceptual framework for selecting appropriate algorithms and interpreting their results.

This chapter introduces the principal types of unsupervised learning that will be explored throughout this notebook.

---



# <font color='orange'> 1. Why Different Types of Unsupervised Learning? </font>

Not all unsupervised learning problems are the same.

For example,

one dataset may require identifying groups of similar observations,

while another may require reducing thousands of variables to only a few informative features.

Consequently,

different algorithms have been developed for different objectives.

```
Unlabelled Data

↓

Different Objectives

↓

Different Algorithms
```

Understanding the objective is the first step toward selecting an appropriate method.

---



# <font color='orange'> 2. Clustering </font>

Clustering is the most widely used form of unsupervised learning.

The objective is to divide observations into groups such that

* observations within the same cluster are similar,
* observations belonging to different clusters are dissimilar.

Conceptually,

```
Unlabelled Data

↓

Similarity Measurement

↓

Clusters
```

Unlike classification,

the groups are **not predefined**.

Instead,

the algorithm discovers them automatically.

---

### Typical Applications

* Customer segmentation
* Gene expression analysis
* Image segmentation
* Document clustering
* Social network analysis

---

### Common Algorithms

* K-Means
* Hierarchical Clustering
* DBSCAN
* Gaussian Mixture Models
* Spectral Clustering

---



# <font color='orange'> 3. Dimensionality Reduction </font>

Modern datasets often contain hundreds or even thousands of variables.

Many of these variables may be

* redundant,
* highly correlated,
* noisy.

Dimensionality reduction seeks to represent the data using fewer variables while preserving as much useful information as possible.

```
High-Dimensional Data

↓

Projection

↓

Low-Dimensional Representation
```

---

### Typical Applications

* Data visualization
* Noise reduction
* Feature extraction
* Data compression
* Preprocessing before supervised learning

---

### Common Algorithms

* Principal Component Analysis (PCA)
* Kernel PCA
* t-SNE
* UMAP
* Isomap

---



# <font color='orange'> 4. Density Estimation </font>

Some algorithms attempt to estimate

the probability distribution

from which the data were generated.

Rather than assigning observations to clusters,

these methods estimate

where data are likely to occur.

Conceptually,

```
Observations

↓

Estimate Probability Density

↓

Underlying Distribution
```

---

### Applications

* Anomaly detection
* Bayesian statistics
* Simulation
* Generative modelling

---

### Common Methods

* Kernel Density Estimation (KDE)
* Gaussian Mixture Models
* Bayesian Density Estimation

---



# <font color='orange'> 5. Association Rule Learning </font>

Association rule learning discovers relationships between variables.

Rather than grouping observations,

it identifies patterns that frequently occur together.

Example

```
Bread

↓

Butter
```

or

```
Laptop

↓

Wireless Mouse
```

These relationships are called **association rules**.

---

### Applications

* Market basket analysis
* Product recommendation
* Customer behaviour analysis
* Retail analytics

---

### Common Algorithms

* Apriori
* FP-Growth
* ECLAT

---



# <font color='orange'> 6. Matrix Factorization </font>

Large datasets can often be represented more compactly by decomposing a matrix into smaller matrices.

Conceptually,

```
Large Matrix

↓

Factorization

↓

Latent Features
```

Rather than analysing the original variables,

matrix factorization discovers hidden factors that explain the observed data.

---

### Applications

* Recommendation systems
* Topic modelling
* Image compression
* Natural language processing

---

### Common Methods

* Singular Value Decomposition (SVD)
* Non-negative Matrix Factorization (NMF)

---



# <font color='orange'> 7. Anomaly Detection </font>

Some observations differ substantially from the majority of the data.

These observations are called

* anomalies,
* outliers,
* novelties.

The objective is

```
Normal Data

↓

Identify Rare Observations

↓

Anomalies
```

Unlike clustering,

anomaly detection focuses on

unusual observations

rather than natural groups.

---

### Applications

* Fraud detection
* Cybersecurity
* Equipment monitoring
* Medical diagnosis
* Manufacturing quality control

---

### Common Algorithms

* Isolation Forest
* Local Outlier Factor (LOF)
* One-Class SVM
* DBSCAN (noise points)

---
