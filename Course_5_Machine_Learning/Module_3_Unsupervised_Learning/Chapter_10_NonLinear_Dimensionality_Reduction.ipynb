# <font color='blue'> Beyond PCA: Nonlinear Dimensionality Reduction </font>

Principal Component Analysis (PCA) is one of the most widely used dimensionality reduction techniques because of its mathematical elegance, computational efficiency, and ability to preserve the dominant linear structure of a dataset. However, PCA assumes that the important relationships between observations can be represented by linear combinations of the original variables. While this assumption is often reasonable, many real-world datasets possess fundamentally nonlinear structures that cannot be adequately captured by linear projections.

Nonlinear dimensionality reduction techniques, often referred to collectively as **manifold learning algorithms**, address this limitation by attempting to preserve the intrinsic geometry of the data rather than simply maximizing variance. These methods assume that although the observations may exist in a high-dimensional space, they often lie on or near a much lower-dimensional nonlinear surface called a **manifold**.

This chapter introduces the motivation for nonlinear dimensionality reduction, explains the concept of manifolds, and provides an overview of the principal nonlinear techniques that will be studied in the following chapters.

---



# <font color='orange'> 1. The Limitation of PCA </font>

PCA searches for

straight-line directions

along which the variance is maximized.

Conceptually,

```
Data

↓

Linear Projection

↓

Principal Components
```

This works well when

the underlying structure is approximately linear.

However,

many datasets are inherently nonlinear.

---



# <font color='orange'> 2. A Nonlinear Example </font>

Imagine observations arranged along a curved surface.

```
********

      ****

           ****

                ****
```

Although the data appear high-dimensional,

they actually lie on a smooth curved structure.

No straight line can accurately summarize this geometry.

Consequently,

PCA cannot fully preserve the relationships between observations.

---



# <font color='orange'> 3. The Swiss Roll Dataset </font>

A classic example is the **Swiss Roll**.

Conceptually,

```
Flat Sheet

↓

Roll Up

↓

Spiral Surface
```

Although the Swiss Roll exists in three-dimensional space,

its intrinsic structure is only two-dimensional.

PCA projects the data onto a plane,

causing distant layers of the spiral to overlap.

The original neighbourhood relationships are therefore lost.

---



# <font color='orange'> 4. What is a Manifold? </font>

A **manifold**

is a lower-dimensional surface embedded within a higher-dimensional space.

Examples include

* a curve inside a plane,
* a sheet inside three-dimensional space,
* a curved surface embedded within many dimensions.

Conceptually,

```
High-Dimensional Space

↓

Hidden Low-Dimensional Surface

↓

Manifold
```

Many real-world datasets are believed to lie approximately on such manifolds.

---



# <font color='orange'> 5. The Goal of Manifold Learning </font>

Instead of preserving

global variance,

manifold learning attempts to preserve

local neighbourhood relationships.

Conceptually,

```
Nearby Observations

↓

Remain Nearby

↓

After Dimensionality Reduction
```

This often produces much more faithful representations of nonlinear datasets.

---



# <font color='orange'> 6. Linear vs Nonlinear Methods </font>

The fundamental difference is

what each method attempts to preserve.

| Linear Methods | Nonlinear Methods |
| :--- | :--- |
| Global variance | Local neighbourhood structure |
| Straight-line projections | Curved manifolds |
| Linear relationships | Nonlinear relationships |

Both approaches are valuable,

but they solve different problems.

---



# <font color='orange'> 7. Popular Nonlinear Techniques </font>

Several manifold learning algorithms have become standard tools in machine learning.

| Algorithm | Primary Goal |
| :--- | :--- |
| t-SNE | Data visualization |
| UMAP | Visualization and clustering |
| Isomap | Preserve global manifold geometry |
| Locally Linear Embedding (LLE) | Preserve local linear structure |

Each algorithm uses a different mathematical strategy to uncover the underlying manifold.

---



# <font color='orange'> 8. t-SNE </font>

**t-distributed Stochastic Neighbor Embedding (t-SNE)**

focuses on preserving

local neighbourhoods.

Observations that are close together in the original space remain close together in the lower-dimensional representation.

Applications include

* image visualization,
* single-cell genomics,
* natural language processing,
* exploratory data analysis.

Its primary strength is producing visually meaningful two-dimensional embeddings.

---



# <font color='orange'> 9. UMAP </font>

**Uniform Manifold Approximation and Projection (UMAP)**

also preserves local neighbourhoods,

but generally

* runs faster,
* scales better,
* preserves more global structure,

than t-SNE.

UMAP has become increasingly popular for

large datasets,

particularly in bioinformatics and machine learning.

---



# <font color='orange'> 10. Isomap </font>

Isomap extends classical multidimensional scaling by replacing straight-line distances with

**geodesic distances**

measured along the manifold.

Conceptually,

```
Straight Line

↓

Euclidean Distance
```

versus

```
Along Curved Surface

↓

Geodesic Distance
```

This allows Isomap to recover certain nonlinear structures that PCA cannot.

---



# <font color='orange'> 11. Locally Linear Embedding (LLE) </font>

LLE assumes

that every observation

can be reconstructed

from its nearby neighbours.

Dimensionality reduction then seeks

a lower-dimensional representation

that preserves these local reconstruction relationships.

LLE is particularly effective when the manifold is smooth and well sampled.

---



# <font color='orange'> 12. Comparing PCA and Manifold Learning </font>

| Property | PCA | Manifold Learning |
| :--- | :---: | :---: |
| Linear | ✔ | ✘ |
| Fast | ✔ | Moderate |
| Interpretable Components | ✔ | Generally ✘ |
| Preserves Global Variance | ✔ | Limited |
| Preserves Local Geometry | Limited | ✔ |
| Suitable for Visualization | Good | Excellent |

Each approach serves different objectives.

---



# <font color='orange'> 13. When Should You Use Nonlinear Methods? </font>

Nonlinear dimensionality reduction is particularly useful when

* PCA fails to reveal meaningful structure,
* visualization is the primary objective,
* clusters appear highly curved,
* the data are believed to lie on a nonlinear manifold.

For feature engineering in predictive modelling,

PCA often remains the preferred first choice because of its simplicity and interpretability.

---



# <font color='orange'> 14. Limitations </font>

Nonlinear methods also have important limitations.

* More computationally expensive than PCA.
* Sensitive to hyperparameter choices.
* Results may vary between runs.
* Axes usually lack direct physical interpretation.
* Often intended for visualization rather than feature extraction.

These considerations should be kept in mind when selecting an appropriate dimensionality reduction technique.

> **Key Insight:** While Principal Component Analysis is highly effective for datasets whose structure is approximately linear, many real-world datasets lie on nonlinear manifolds embedded within higher-dimensional spaces. Manifold learning techniques such as t-SNE, UMAP, Isomap, and Locally Linear Embedding address this limitation by preserving neighbourhood relationships and the intrinsic geometry of the data rather than maximizing variance. These methods provide powerful tools for visualization and exploratory analysis when linear dimensionality reduction is insufficient.

# <font color='blue'> t-distributed Stochastic Neighbor Embedding (t-SNE) </font>

Principal Component Analysis (PCA) reduces dimensionality by identifying linear combinations of the original variables that maximize variance. Although this approach is computationally efficient and mathematically elegant, it often fails to preserve the complex nonlinear relationships present in many real-world datasets. In particular, observations that are close together in the original high-dimensional space may become widely separated after a linear projection, while distant observations may appear artificially close.

t-distributed Stochastic Neighbor Embedding (t-SNE) was developed specifically to address this problem. Rather than preserving global variance, t-SNE focuses on preserving **local neighbourhood relationships**. It converts pairwise similarities between observations into probability distributions and then constructs a low-dimensional embedding whose probability distribution closely matches that of the original data.

Because of its ability to reveal complex cluster structures, t-SNE has become one of the most widely used visualization techniques for high-dimensional datasets.

---



# <font color='orange'> 1. Motivation </font>

Suppose we have a dataset with

```
500 Features
```

and

```
20,000 Observations.
```

Direct visualization is impossible.

PCA may reduce the data to two dimensions,

but important nonlinear relationships may disappear.

Ideally,

we would like nearby observations to remain nearby after dimensionality reduction.

This is precisely the objective of t-SNE.

---



# <font color='orange'> 2. Why PCA Can Fail for Visualization </font>

PCA seeks directions of maximum variance.

```
High-Dimensional Data

↓

Linear Projection

↓

Maximum Variance
```

However,

observations that are neighbours in the original space

may no longer remain neighbours after projection.

Consequently,

clusters may overlap

or become distorted.

t-SNE instead prioritizes

the preservation of local neighbourhoods.

---



# <font color='orange'> 3. The Central Idea of t-SNE </font>

Instead of preserving variance,

t-SNE preserves

**pairwise similarities**.

Conceptually,

```
Nearby Observations

↓

High Probability

↓

Remain Nearby

↓

Low-Dimensional Map
```

Similarly,

observations that are far apart

have low probabilities of being neighbours.

The resulting visualization attempts to preserve these local relationships.

---



# <font color='orange'> 4. Measuring Similarity in High Dimensions </font>

For every observation,

t-SNE computes the probability that another observation is its neighbour.

Conceptually,

```
Observation

↓

Nearby Points

↓

Large Probability
```

```
Far Away Points

↓

Small Probability
```

Thus,

the original dataset is represented as

a probability distribution over pairwise neighbourhood relationships.

---



# <font color='orange'> 5. Measuring Similarity in Low Dimensions </font>

After constructing a two-dimensional embedding,

t-SNE computes a second probability distribution

using the projected observations.

The objective is to make

the low-dimensional probabilities

match

the high-dimensional probabilities

as closely as possible.

---



# <font color='orange'> 6. Why the Student's t-Distribution? </font>

One of the defining features of t-SNE is its use of the

**Student's t-distribution**

when computing similarities in the low-dimensional space.

Unlike a Gaussian distribution,

the Student's *t*-distribution has

**heavier tails**.

Conceptually,

```
Gaussian

↓

Rapidly Decreasing Tails
```

versus

```
Student's t

↓

Heavier Tails
```

This helps prevent unrelated clusters from being artificially crowded together,

a problem known as the **crowding problem**.

---



# <font color='orange'> 7. The Crowding Problem </font>

Projecting many dimensions into only two dimensions inevitably compresses the available space.

Without correction,

many moderately distant observations would be forced too close together.

The heavy tails of the Student's *t*-distribution provide more space in the embedding,

allowing well-separated clusters to remain distinct.

---



# <font color='orange'> 8. Perplexity </font>

One of the most important hyperparameters is

**perplexity**.

Perplexity controls

the effective number of neighbouring observations considered during the construction of the probability distributions.

Conceptually,

```
Small Perplexity

↓

Very Local Structure
```

```
Large Perplexity

↓

Broader Neighbourhoods
```

Typical values range from

```
5

to

50
```

although the optimal choice depends on the dataset.

---



# <font color='orange'> 9. Optimization </font>

t-SNE seeks an embedding that makes the high-dimensional and low-dimensional probability distributions as similar as possible.

The similarity between the two distributions is measured using the

**Kullback–Leibler (KL) divergence**.

Conceptually,

```
High-Dimensional Probabilities

↓

KL Divergence

↓

Low-Dimensional Probabilities

↓

Optimization
```

The algorithm iteratively adjusts the positions of the points until the KL divergence is minimized.

---



# <font color='orange'> 10. Interpreting a t-SNE Plot </font>

A t-SNE visualization is useful for identifying

* clusters,
* local neighbourhoods,
* outliers,
* subpopulations.

However,

the distances between well-separated clusters

should be interpreted with caution.

t-SNE primarily preserves

**local relationships**,

not global geometry.

For example, if two clusters appear far apart in a t-SNE plot, this does **not** necessarily imply that they are more different than two clusters that appear closer together.

---



# <font color='orange'> 11. Advantages </font>

t-SNE offers several important advantages.

* Excellent visualization of high-dimensional data.
* Preserves local neighbourhoods.
* Reveals nonlinear cluster structures.
* Often produces visually meaningful embeddings.

These properties have made it extremely popular for exploratory data analysis.

---



# <font color='orange'> 12. Limitations </font>

Despite its strengths,

t-SNE also has important limitations.

* Computationally expensive.
* Sensitive to hyperparameters such as perplexity.
* Different random initializations may produce different embeddings.
* Does not naturally preserve global distances.
* Primarily intended for visualization rather than feature extraction.

These limitations motivate newer techniques such as UMAP.

| Concept | Description |
| :--- | :--- |

# <font color='blue'> Uniform Manifold Approximation and Projection (UMAP) </font>

t-distributed Stochastic Neighbor Embedding (t-SNE) has become one of the most widely used techniques for visualizing high-dimensional datasets because of its ability to preserve local neighbourhood relationships. However, t-SNE is computationally demanding, often struggles to preserve global structure, and can be sensitive to its hyperparameters. These limitations motivated the development of **Uniform Manifold Approximation and Projection (UMAP)**, a more recent nonlinear dimensionality reduction technique that combines strong theoretical foundations with excellent practical performance.

UMAP assumes that the observations lie approximately on a low-dimensional manifold embedded within a higher-dimensional space. Rather than modelling pairwise similarities directly through probability distributions, UMAP constructs a graph representing the local neighbourhood structure of the data and then optimizes a low-dimensional embedding that preserves this graph as faithfully as possible.

Because of its speed, scalability, and ability to preserve both local and some global structure, UMAP has become one of the most widely used dimensionality reduction techniques for visualization and exploratory data analysis.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to visualize

```
100,000 Observations

↓

500 Features
```

While t-SNE can often produce excellent visualizations,

its computational cost increases rapidly as the dataset grows.

Ideally,

we would like an algorithm that is

* faster,
* scalable,
* preserves local structure,
* retains more global information.

This is the motivation behind UMAP.

---



# <font color='orange'> 2. The Central Idea </font>

Like t-SNE,

UMAP assumes that

high-dimensional data lie on

a lower-dimensional manifold.

However,

instead of comparing probability distributions,

UMAP builds

a graph connecting neighbouring observations.

Conceptually,

```
Dataset

↓

Neighbourhood Graph

↓

Optimize Layout

↓

Low-Dimensional Embedding
```

The objective is to preserve the relationships encoded in this graph.

---



# <font color='orange'> 3. Constructing the Neighbourhood Graph </font>

For every observation,

UMAP identifies its nearest neighbours.

Conceptually,

```
Observation

↓

Nearest Neighbours

↓

Graph Connections
```

Every observation becomes

a node,

while neighbouring observations are connected by edges.

The resulting graph approximates the local geometry of the underlying manifold.

---



# <font color='orange'> 4. Optimizing the Embedding </font>

Once the graph has been constructed,

UMAP searches for

a low-dimensional arrangement

that preserves these neighbourhood relationships.

Conceptually,

```
Neighbourhood Graph

↓

Optimization

↓

2D or 3D Embedding
```

Nearby observations remain close together,

while unrelated observations are separated whenever possible.

---



# <font color='orange'> 5. The `n_neighbors` Hyperparameter </font>

One of the most important parameters is

```python
n_neighbors
```

It determines

how many neighbouring observations are considered when constructing the graph.

Small values

```
↓

Emphasize Local Structure
```

Large values

```
↓

Preserve More Global Structure
```

Typical values range between

```
10

and

50.
```

---



# <font color='orange'> 6. The `min_dist` Hyperparameter </font>

Another important parameter is

```python
min_dist
```

It controls

how closely observations may be packed together

in the final embedding.

Small values

```
↓

Compact Clusters
```

Large values

```
↓

More Evenly Spread Embedding
```

Typical values range between

```
0.0

and

0.5.
```

The optimal choice depends on the desired visualization.

---



# <font color='orange'> 7. UMAP vs t-SNE </font>

Although both methods preserve neighbourhood relationships,

their behaviour differs.

| Property | t-SNE | UMAP |
| :--- | :---: | :---: |
| Speed | Moderate | Fast |
| Scalability | Moderate | Excellent |
| Local Structure | Excellent | Excellent |
| Global Structure | Limited | Better |
| Embedding New Data | Difficult | Naturally supported through the fitted model |
| Typical Dataset Size | Small–Medium | Medium–Very Large |

UMAP is therefore often preferred for large datasets.

---



# <font color='orange'> 8. Interpreting UMAP Plots </font>

UMAP embeddings reveal

* clusters,
* local neighbourhoods,
* outliers,
* continuous transitions between groups.

However,

as with t-SNE,

the axes themselves

have **no intrinsic physical meaning**.

Only the relative arrangement of the observations should be interpreted.

Although UMAP generally preserves more global structure than t-SNE, distances between widely separated clusters should still be interpreted cautiously.

---



# <font color='orange'> 9. Advantages </font>

UMAP offers several important advantages.

* Fast execution.
* Excellent scalability.
* Preserves local neighbourhoods.
* Better preservation of global relationships than t-SNE.
* Naturally supports transforming new observations after fitting.
* Produces high-quality visualizations.

These properties have made it increasingly popular across scientific disciplines.

---



# <font color='orange'> 10. Limitations </font>

Despite its strengths,

UMAP also has limitations.

* Sensitive to hyperparameter selection.
* Embeddings may differ between runs unless a fixed random seed is used.
* Components are not directly interpretable.
* Primarily intended for visualization and exploratory analysis.

As with any manifold learning technique,

careful interpretation is essential.

---
