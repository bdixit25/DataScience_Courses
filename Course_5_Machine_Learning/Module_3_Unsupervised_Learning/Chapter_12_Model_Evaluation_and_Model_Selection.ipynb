# <font color='blue'> Introduction to Model Evaluation and Model Selection </font>

Building a machine learning model is only one part of the predictive modelling process. Equally important is determining whether the model has learned meaningful patterns from the data and whether it is likely to perform well on previously unseen observations. A model that performs exceptionally well on the training data may nevertheless fail completely when applied to new data if it has simply memorized the training examples rather than learning the underlying relationships.

Model evaluation provides a systematic framework for measuring how well a machine learning model generalizes beyond the data on which it was trained. By separating model development from model assessment, we obtain realistic estimates of predictive performance and reduce the risk of drawing overly optimistic conclusions.

Closely related to model evaluation is **model selection**, the process of comparing multiple candidate models and choosing the one that provides the best balance between predictive performance, complexity, computational cost, and interpretability. Together, model evaluation and model selection form the foundation of reliable machine learning.

This chapter introduces the motivation for model evaluation, explains why proper assessment is essential, and provides an overview of the topics that will be studied throughout this module.

# <font color='blue'> Training, Validation, and Test Sets </font>

A fundamental objective of machine learning is to develop models that perform well not only on the data used for training but also on previously unseen observations. To obtain an unbiased estimate of a model's predictive performance, the available data must therefore be divided into separate subsets that serve different purposes during model development.

The dataset is typically partitioned into **training**, **validation**, and **test** sets. The training set is used to learn the model parameters, the validation set is used to tune hyperparameters and compare competing models, and the test set is reserved for the final assessment of the selected model. Keeping these datasets separate ensures that model evaluation reflects genuine predictive performance rather than memorization of the available data.

This chapter introduces the purpose of each dataset split, discusses common splitting strategies, explains the importance of preventing data leakage, and demonstrates how these partitions are implemented in Scikit-Learn.

---

# <font color='orange'> 1. Why Split the Dataset? </font>

Suppose we have a dataset containing

```
10,000 Observations
```

If we train and evaluate the model

using exactly the same data,

the performance estimate is likely to be overly optimistic.

Instead,

we divide the dataset into separate subsets.

Conceptually,

```
Complete Dataset

↓

Split

↓

Training

Validation

Test
```

Each subset serves a different purpose.

---

# <font color='orange'> 2. The Training Set </font>

The **training set** is used to

learn the model.

During training,

the algorithm estimates

* model parameters,
* decision boundaries,
* regression coefficients,
* neural network weights,

depending on the model.

Conceptually,

```
Training Data

↓

Learn Patterns

↓

Model
```

The model is allowed to "see" only the training data during learning.

---

# <font color='orange'> 3. The Validation Set </font>

The **validation set** is used during model development,

but **not** for learning the model parameters.

Instead,

it is used to

* tune hyperparameters,
* compare different models,
* monitor overfitting.

Conceptually,

```
Candidate Models

↓

Validation Data

↓

Choose Best Model
```

The validation set provides feedback during model development.

---

# <font color='orange'> 4. The Test Set </font>

The **test set** is reserved until the very end of the project.

Its purpose is to provide

an unbiased estimate

of the model's performance on

future unseen data.

Conceptually,

```
Final Model

↓

Test Set

↓

Performance Estimate
```

The test set should **never** influence model training or model selection.

---

# <font color='orange'> 5. The Complete Workflow </font>

The overall workflow is

```
Dataset

↓

Training Set

↓

Train Model

↓

Validation Set

↓

Tune Model

↓

Test Set

↓

Final Evaluation
```

Notice that the test set appears **only once**, after the model has been finalized.

---

# <font color='orange'> 6. Common Splitting Ratios </font>

Several splitting strategies are widely used.

| Training | Validation | Test |
| :---: | :---: | :---: |
| 80% | — | 20% |
| 70% | 15% | 15% |
| 60% | 20% | 20% |
| 80% | 10% | 10% |

The most appropriate choice depends on

* dataset size,
* model complexity,
* availability of data.

For very small datasets,

cross-validation is often preferred over a fixed validation set.

---

# <font color='orange'> 7. Why the Test Set Must Be Used Only Once </font>

Suppose we repeatedly evaluate the model

on the test set,

adjusting the model after each evaluation.

Eventually,

the model becomes optimized

for the test data itself.

Conceptually,

```
Model

↓

Test Set

↓

Modify Model

↓

Repeat

↓

Optimistic Performance
```

The test set is no longer "unseen."

The reported performance therefore becomes biased.

---

# <font color='orange'> 8. Data Leakage </font>

One of the most common mistakes in machine learning is

**data leakage**.

Data leakage occurs when information from outside the training set is inadvertently used during model development.

Examples include

* scaling the entire dataset before splitting,
* performing feature selection before splitting,
* imputing missing values using statistics computed from the full dataset,
* repeatedly tuning hyperparameters using the test set.

Data leakage leads to unrealistically optimistic evaluation results.

---

# <font color='orange'> 9. Avoiding Data Leakage </font>

The correct workflow is

```
Split Dataset

↓

Fit Preprocessing

↓

Training Set Only

↓

Apply Same Transformation

↓

Validation/Test Sets
```

For example,

a `StandardScaler`

should be fitted using

only the training data,

and then applied to the validation and test sets.

This ensures that information from unseen data does not influence the training process.

---

# <font color='orange'> 10. Stratified Splitting </font>

For classification problems,

the class proportions should usually be preserved in every dataset split.

This is known as

**stratified sampling**.

Example

Original dataset

```
90% Class A

10% Class B
```

After stratification,

every split retains approximately

```
90% Class A

10% Class B
```

This helps ensure that each subset is representative of the full dataset.

For regression problems,

simple random splitting is common, although specialized strategies may be appropriate for certain datasets (e.g., grouped or time-ordered data).

---

# <font color='orange'> 11. Time-Series Data </font>

Random splitting is generally **not appropriate** for time-series data.

Instead,

the data should be divided chronologically.

Conceptually,

```
Past

↓

Training

↓

Validation

↓

Future

↓

Test
```

Using future observations to predict the past introduces data leakage and produces unrealistic performance estimates.

# <font color='blue'> Cross-Validation </font>

A single train-test split provides only one estimate of a model's predictive performance. Although this approach is simple and computationally efficient, the estimated performance can vary substantially depending on how the data happen to be partitioned. Two different random splits of the same dataset may produce noticeably different evaluation results, particularly when the dataset is small or contains substantial variability.

Cross-validation addresses this limitation by repeatedly dividing the dataset into different training and validation subsets. Rather than relying on a single evaluation, the model is trained and validated multiple times, with each observation serving as validation data exactly once (in the case of standard k-fold cross-validation). The resulting performance estimates are then averaged, providing a more robust assessment of the model's ability to generalize to unseen data.

This chapter introduces the principles of cross-validation, discusses several commonly used cross-validation strategies, explains their advantages and limitations, and demonstrates their implementation using Scikit-Learn.

---

# <font color='orange'> 1. Why Isn't a Single Train-Test Split Enough? </font>

Suppose we randomly divide a dataset into

```
80% Training

20% Testing
```

The estimated performance depends on

which observations happen to be included in the test set.

A different random split may produce

a different accuracy,

a different RMSE,

or a different F1-score.

Consequently,

a single train-test split provides only

one estimate

of model performance.

---

# <font color='orange'> 2. The Idea Behind Cross-Validation </font>

Instead of evaluating the model once,

cross-validation evaluates it

multiple times

using different partitions of the data.

Conceptually,

```
Dataset

↓

Split 1

↓

Evaluate

↓

Split 2

↓

Evaluate

↓

...

↓

Average Performance
```

The final estimate is therefore less sensitive to a particular random split.

---

# <font color='orange'> 3. k-Fold Cross-Validation </font>

The most widely used approach is

**k-fold cross-validation**.

The dataset is divided into

```
k
```

approximately equal-sized subsets called

**folds**.

The procedure is

1. Reserve one fold for validation.
2. Train on the remaining folds.
3. Evaluate the model.
4. Repeat until every fold has served as the validation set exactly once.

---

### Example: Five-Fold Cross-Validation

```
Fold 1

[V][T][T][T][T]

Fold 2

[T][V][T][T][T]

Fold 3

[T][T][V][T][T]

Fold 4

[T][T][T][V][T]

Fold 5

[T][T][T][T][V]
```

where

```
T = Training

V = Validation
```

Every observation is used

* for training,
* for validation,

exactly once as validation.

---

# <font color='orange'> 4. Final Performance Estimate </font>

Suppose five-fold cross-validation produces

| Fold | Accuracy |
| :--- | ---: |
| 1 | 92% |
| 2 | 90% |
| 3 | 91% |
| 4 | 93% |
| 5 | 89% |

The overall estimate becomes

$$
\boxed{
\frac{92+90+91+93+89}{5}
=
91\%.
}
$$

The standard deviation across folds is also useful because it indicates how stable the model's performance is.

---

# <font color='orange'> 5. Choosing the Value of \(k\) </font>

Typical choices are

| Value of \(k\) | Characteristics |
| :---: | :--- |
| 5 | Fast, widely used |
| 10 | Lower variance estimate, common default |
| Larger \(k\) | More training data per fold, greater computational cost |

Five-fold and ten-fold cross-validation are the most common choices in practice.

---

# <font color='orange'> 6. Stratified k-Fold Cross-Validation </font>

For classification problems,

it is important to preserve

the class proportions

within every fold.

This is achieved using

**Stratified k-Fold Cross-Validation**.

Example

Original dataset

```
90% Class A

10% Class B
```

Each fold retains approximately

```
90% Class A

10% Class B
```

This produces more reliable estimates,

particularly for imbalanced datasets.

---

# <font color='orange'> 7. Leave-One-Out Cross-Validation (LOOCV) </font>

An extreme form of cross-validation is

**Leave-One-Out Cross-Validation**.

Suppose the dataset contains

```
n
```

observations.

Each iteration uses

```
n−1

Training
```

and

```
1

Validation
```

This process is repeated

once for every observation.

---

### Advantages

* Uses nearly all available data for training.
* Produces an almost unbiased estimate of predictive performance.

### Limitations

* Extremely computationally expensive.
* High variance because each validation set contains only one observation.

LOOCV is therefore most appropriate for very small datasets.

---

# <font color='orange'> 8. Repeated k-Fold Cross-Validation </font>

The results of ordinary k-fold cross-validation still depend slightly on the initial random partition.

To reduce this dependence,

the entire k-fold procedure can be repeated

multiple times

using different random splits.

Conceptually,

```
5-Fold CV

↓

Repeat

↓

Repeat

↓

Average All Results
```

This provides an even more robust estimate of model performance.

---

# <font color='orange'> 9. Advantages of Cross-Validation </font>

Cross-validation offers several important benefits.

* Better estimate of generalization performance.
* More efficient use of limited data.
* Less dependent on a single random split.
* Suitable for comparing competing models.
* Widely used during hyperparameter tuning.

These advantages make it one of the standard evaluation techniques in machine learning.

---

# <font color='orange'> 10. Limitations </font>

Cross-validation also has limitations.

* Computationally more expensive than a single train-test split.
* May be impractical for extremely large datasets.
* Standard k-fold cross-validation is not appropriate for time-series data.

Specialized cross-validation methods are required for ordered or grouped data.

---

# <font color='orange'> 11. Cross-Validation and the Test Set </font>

An important distinction should be made.

Cross-validation is performed

using the **training data**.

The independent test set

remains untouched

until the final evaluation.

Conceptually,

```
Dataset

↓

Train/Test Split

↓

Training Set

↓

Cross-Validation

↓

Choose Best Model

↓

Final Evaluation

↓

Test Set
```

This ensures that the final performance estimate remains unbiased.

# <font color='blue'> The Bias–Variance Trade-Off </font>

One of the central challenges in machine learning is constructing models that generalize well to previously unseen data. A model that is too simple may fail to capture the underlying relationships within the data, whereas a model that is too complex may begin fitting random fluctuations rather than meaningful patterns. These two undesirable behaviours are known as **underfitting** and **overfitting**, respectively.

The balance between these competing effects is described by the **bias–variance trade-off**, one of the most fundamental concepts in statistical learning. Bias measures the error introduced by simplifying assumptions made by the model, while variance measures the sensitivity of the model to small fluctuations in the training data. Successful machine learning seeks an appropriate balance between these two sources of error in order to achieve good predictive performance on unseen observations.

This chapter introduces the concepts of underfitting, overfitting, bias, and variance, explains how they influence model performance, and discusses practical strategies for balancing model complexity.

---

# <font color='orange'> 1. The Goal of Machine Learning </font>

The purpose of a machine learning model is

not simply to perform well on the training data,

but to make accurate predictions on

previously unseen observations.

Conceptually,

```
Training Data

↓

Learn General Patterns

↓

Good Predictions

↓

New Data
```

Achieving this objective requires finding the appropriate level of model complexity.

---

# <font color='orange'> 2. Underfitting </font>

A model is said to

**underfit**

when it is

too simple

to capture the important relationships in the data.

Conceptually,

```
Simple Model

↓

Misses Important Patterns

↓

Large Errors
```

Characteristics include

* poor performance on the training set,
* poor performance on the test set,
* inability to capture the underlying trend.

---

### Example

Suppose the true relationship is highly curved,

but we fit

a straight line.

The model cannot adequately describe the data,

regardless of how much training data are available.

---

# <font color='orange'> 3. Overfitting </font>

A model is said to

**overfit**

when it becomes

too complex.

Instead of learning the underlying relationship,

it begins fitting

random noise

present in the training data.

Conceptually,

```
Very Complex Model

↓

Fits Noise

↓

Excellent Training Performance

↓

Poor Test Performance
```

Characteristics include

* extremely low training error,
* significantly higher test error,
* poor generalization.

---

### Example

Imagine drawing a curve that passes through every training point exactly.

Although the training error becomes nearly zero,

the curve behaves unpredictably between observations,

resulting in poor predictions for new data.

---

# <font color='orange'> 4. The Ideal Situation </font>

The goal is neither the simplest model

nor the most complex model.

Instead,

we seek

```
Appropriate Complexity

↓

Captures Genuine Patterns

↓

Good Generalization
```

This balance is the essence of the bias–variance trade-off.

---

# <font color='orange'> 5. Introducing Bias </font>

**Bias**

measures the error introduced by

overly simplifying assumptions

made by a model.

High bias implies that

the model systematically fails

to capture the true relationship.

Conceptually,

```
Simple Model

↓

High Bias

↓

Underfitting
```

Examples include

* linear regression applied to a highly nonlinear relationship,
* a shallow decision tree for a complex classification problem.

---

# <font color='orange'> 6. Introducing Variance </font>

**Variance**

measures

how sensitive a model is

to small changes in the training data.

A high-variance model may produce

very different predictions

when trained on slightly different datasets.

Conceptually,

```
Complex Model

↓

High Variance

↓

Overfitting
```

Examples include

* deep decision trees,
* high-degree polynomial regression,
* overly flexible neural networks trained on limited data.

---

# <font color='orange'> 7. Bias and Variance Together </font>

Bias and variance change in opposite directions as model complexity increases.

```
Increasing Model Complexity

↓

Bias Decreases

↓

Variance Increases
```

Conversely,

```
Decreasing Model Complexity

↓

Bias Increases

↓

Variance Decreases
```

Finding the appropriate balance between these competing effects is one of the principal objectives of machine learning.

---

# <font color='orange'> 8. The Bias–Variance Trade-Off </font>

The expected prediction error can be viewed conceptually as

```
Prediction Error

=

Bias²

+

Variance

+

Irreducible Noise
```

where

* **Bias²** represents systematic error due to simplifying assumptions,
* **Variance** represents sensitivity to the training data,
* **Irreducible Noise** represents randomness inherent in the data that no model can eliminate.

Although the exact mathematical decomposition depends on the learning setting, this expression captures the central intuition behind the trade-off.

---

# <font color='orange'> 9. Training Error vs Test Error </font>

As model complexity increases,

the training error usually decreases continuously.

The test error,

however,

typically follows a different pattern.

Conceptually,

```
Test Error

│

│        ●

│      ●

│    ●

│  ●

│ ●

│  ●

│    ●

└──────────────────────

      Model Complexity
```

Initially,

increasing complexity improves generalization.

Beyond a certain point,

additional complexity causes overfitting,

and the test error begins to increase.

The minimum of this curve corresponds to the optimal model complexity.

---

# <font color='orange'> 10. Practical Examples </font>

| Model | Typical Behaviour |
| :--- | :--- |
| Linear Regression | Higher bias, lower variance |
| Polynomial Regression (very high degree) | Lower bias, higher variance |
| Shallow Decision Tree | Higher bias, lower variance |
| Deep Decision Tree | Lower bias, higher variance |
| Random Forest | Reduced variance through averaging |
| Gradient Boosting | Can reduce bias while controlling variance with appropriate tuning |

These examples illustrate how different algorithms occupy different positions on the bias–variance spectrum.

---

# <font color='orange'> 11. Reducing Bias </font>

High bias can often be reduced by

* increasing model complexity,
* engineering more informative features,
* reducing excessive regularization,
* selecting a more expressive model.

The goal is to allow the model to capture more of the underlying structure in the data.

---

# <font color='orange'> 12. Reducing Variance </font>

High variance can often be reduced by

* collecting more training data,
* simplifying the model,
* increasing regularization,
* pruning decision trees,
* using ensemble methods such as Random Forests,
* applying cross-validation for model selection.

These techniques help improve generalization to unseen data.

---

# <font color='orange'> 13. Practical Workflow </font>

A typical approach to balancing bias and variance is

```
Train Model

↓

Evaluate Performance

↓

High Bias?

↓

Increase Complexity

↓

High Variance?

↓

Reduce Complexity
or Increase Regularization

↓

Re-evaluate
```

This iterative process is common in practical machine learning projects.

| Concept | Description |
| :--- | :--- |
| Underfitting | Model too simple to capture important patterns |
| Overfitting | Model too complex and fits noise |
| Bias | Error caused by simplifying assumptions |
| Variance | Sensitivity to changes in the training data |
| Bias–Variance Trade-Off | Balancing systematic error and model flexibility |
| Irreducible Noise | Random variation that no model can eliminate |
| Goal | Achieve the lowest possible test error by balancing bias and variance |

> **Key Insight:** Successful machine learning requires balancing two competing sources of prediction error. Models that are too simple exhibit high bias and underfit the data, while overly complex models exhibit high variance and overfit random noise. The optimal model lies between these extremes, capturing the underlying patterns without becoming excessively sensitive to the training data. Understanding the bias–variance trade-off provides the theoretical foundation for model selection, regularization, cross-validation, and many of the techniques used throughout modern machine learning.

# <font color='blue'> Hyperparameter Tuning </font>

Most machine learning algorithms contain settings that control how the model learns from the data. These settings, known as **hyperparameters**, are specified before training begins and influence the behaviour, flexibility, and complexity of the resulting model. Selecting appropriate hyperparameter values is essential because poor choices can lead to underfitting, overfitting, or unnecessarily expensive computations.

Hyperparameter tuning is the process of systematically searching for the combination of hyperparameter values that yields the best predictive performance. Unlike model parameters, which are learned automatically during training, hyperparameters must be chosen by the practitioner. Modern machine learning therefore combines optimization algorithms with robust evaluation techniques, such as cross-validation, to identify high-performing hyperparameter configurations.

This chapter introduces the concept of hyperparameters, distinguishes them from model parameters, discusses several commonly used tuning strategies, and demonstrates how hyperparameter optimization is performed using Scikit-Learn.

---

# <font color='orange'> 1. Parameters vs Hyperparameters </font>

Before discussing tuning,

it is important to distinguish between

**parameters**

and

**hyperparameters**.

### Parameters

Parameters are

**learned automatically**

during model training.

Examples include

* regression coefficients,
* decision tree split values,
* neural network weights,
* cluster centroids in K-Means.

The algorithm estimates these directly from the training data.

---

### Hyperparameters

Hyperparameters are

**chosen before training begins**.

They determine

how the model learns,

rather than

what the model learns.

Examples include

* tree depth,
* number of neighbours,
* learning rate,
* regularization strength,
* number of trees.

---

| :--- | :--- |

These choices strongly influence model performance.

# <font color='blue'> Learning Curves and Validation Curves </font>

Developing an accurate machine learning model involves more than simply selecting an algorithm and tuning its hyperparameters. It is equally important to understand *how* a model learns from data and *why* it succeeds or fails. Learning curves and validation curves provide valuable diagnostic tools for analysing model behaviour, helping practitioners identify underfitting, overfitting, and opportunities for improvement.

A **learning curve** illustrates how model performance changes as the amount of training data increases. By comparing training and validation performance, learning curves reveal whether additional data are likely to improve the model or whether the model's capacity should instead be adjusted.

A **validation curve**, in contrast, examines the effect of varying a single hyperparameter while keeping the training data fixed. Validation curves provide insight into the relationship between model complexity and predictive performance, making them particularly useful during hyperparameter tuning.

This chapter introduces both learning curves and validation curves, explains how they should be interpreted, and demonstrates their implementation using Scikit-Learn.

---

# <font color='orange'> 1. Why Diagnostic Curves Matter </font>

Suppose a model performs poorly.

There are many possible reasons.

* Too little training data.
* Model too simple.
* Model too complex.
* Poor hyperparameter choices.

Evaluation metrics alone cannot identify the cause.

Diagnostic curves help answer

**why**

the model behaves as it does.

---

# <font color='orange'> 2. What is a Learning Curve? </font>

A **learning curve**

shows how model performance changes

as the amount of training data increases.

Conceptually,

```
Training Size

↓

Train Model

↓

Evaluate

↓

Repeat
```

Both

* training performance,
* validation performance,

are plotted as functions of the training set size.

---

# <font color='orange'> 3. Components of a Learning Curve </font>

A typical learning curve contains two curves.

Performance measured on

the training data.

---

Performance measured on

Conceptually,

```
Training Score

──────────────

Validation Score

──────────────
```

Comparing these curves reveals how well the model generalizes.

Characteristics

* low training score,
* low validation score,
* small gap between the curves.

This indicates

high bias.

Characteristics

* high training score,
* much lower validation score,
* large gap between the curves.

This indicates

high variance.

# <font color='orange'> 6. Well-Generalized Model </font>

A well-performing model exhibits

```
Training Score

↓

High

↓

Validation Score

↓

Nearly Equal
```

The two curves converge

to a high performance level,

indicating good generalization.

Learning curves help answer an important practical question.

# <font color='blue'> Evaluation Metrics for Regression Models </font>

After training a regression model, the next step is to determine how accurately it predicts continuous target variables. This requires quantitative measures that compare the predicted values produced by the model with the true observed values. These measures, known as **evaluation metrics**, provide objective criteria for assessing predictive performance, comparing competing models, and guiding model improvement.

No single evaluation metric is universally optimal. Some metrics emphasize the average size of prediction errors, others penalize large errors more heavily, while others measure how much of the variability in the target variable is explained by the model. Consequently, selecting an appropriate evaluation metric depends on the objectives of the analysis, the characteristics of the data, and the practical consequences of prediction errors.

This chapter introduces the most widely used regression evaluation metrics, explains how they are computed and interpreted, discusses their advantages and limitations, and provides guidance for selecting the most appropriate metric for a given application.

---

# <font color='orange'> 1. Why Do We Need Evaluation Metrics? </font>

Suppose a regression model predicts

house prices.

For one property,

the true price is

```
£250,000
```

while the model predicts

```
£245,000.
```

The prediction is not perfect,

but

**how good is it?**

Evaluation metrics provide

a quantitative answer

to this question.

---

# <font color='orange'> 2. Prediction Errors (Residuals) </font>

Regression metrics are based on

the difference between

the true value

and

the predicted value.

This difference is called the

**residual** (or prediction error).

$$
\boxed{
e_i = y_i-\hat{y}_i
}
$$

where

* \(y_i\) is the true value,
* \(\hat{y}_i\) is the predicted value.

A perfect model produces

```
Residual = 0
```

for every observation.

---

# <font color='orange'> 3. Mean Absolute Error (MAE) </font>

The **Mean Absolute Error (MAE)**

is the average absolute prediction error.

$$
\boxed{
\mathrm{MAE}
=
\frac{1}{n}
\sum_{i=1}^{n}
|y_i-\hat{y}_i|
}
$$

---

### Interpretation

MAE answers the question

> **On average, how far are the predictions from the true values?**

Example

```
MAE = £3,500
```

means

the predictions differ from the true values

by approximately

£3,500

on average.

---

### Advantages

* Easy to interpret.
* Same units as the target variable.
* Less sensitive to outliers than MSE.

---

### Limitations

* Treats all errors equally.
* Does not strongly penalize very large errors.

---

# <font color='orange'> 4. Mean Squared Error (MSE) </font>

The **Mean Squared Error (MSE)**

squares every prediction error before averaging.

$$
\boxed{
\mathrm{MSE}
=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
$$

---

### Interpretation

Large errors contribute much more heavily

than small errors.

For example,

```
Error = 10

↓

Squared Error = 100
```

whereas

```
Error = 2

↓

Squared Error = 4
```

This makes MSE particularly sensitive to outliers.

---

### Advantages

* Strongly penalizes large prediction errors.
* Smooth mathematical properties make it convenient for optimization.

---

### Limitations

* Difficult to interpret because the units are squared.
* Highly sensitive to outliers.

---

# <font color='orange'> 5. Root Mean Squared Error (RMSE) </font>

The **Root Mean Squared Error (RMSE)**

is simply the square root of the MSE.

$$
\boxed{
\mathrm{RMSE}
=
\sqrt{
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat{y}_i)^2
}
}
$$

---

### Interpretation

RMSE restores the original units

of the target variable.

Example

```
RMSE = 4.8 kg
```

means

the typical prediction error is approximately

4.8 kg.

---

### Advantages

* Same units as the target.
* Penalizes large errors.
* Widely used in regression problems.

---

### Limitations

* More sensitive to outliers than MAE.

---

# <font color='orange'> 6. Mean Absolute Percentage Error (MAPE) </font>

The **Mean Absolute Percentage Error (MAPE)**

expresses prediction errors as percentages.

$$
\boxed{
\mathrm{MAPE}
=
\frac{100}{n}
\sum_{i=1}^{n}
\left|
\frac{y_i-\hat{y}_i}
{y_i}
\right|
}
$$

---

### Interpretation

Example

```
MAPE = 6%
```

means

the predictions differ from the true values

by approximately

6%

on average.

---

### Advantages

* Easy to communicate.
* Scale independent.

---

### Limitations

* Undefined when \(y_i = 0\).
* Can become misleading when true values are very close to zero.
* May overweight errors for small target values.

---

# <font color='orange'> 7. Coefficient of Determination (\(R^2\)) </font>

The **coefficient of determination**

measures

how much of the variation

in the target variable

is explained by the model.

$$
\boxed{
R^2
=
1
-
\frac{
\sum(y_i-\hat{y}_i)^2
}{
\sum(y_i-\bar{y})^2
}
}
$$

where

\(\bar{y}\)

is the mean of the observed values.

---

### Interpretation

| \(R^2\) | Meaning |
| :---: | :--- |
| 1.0 | Perfect predictions |
| 0.0 | No improvement over predicting the mean |
| < 0 | Worse than predicting the mean |

An \(R^2\) of

```
0.85
```

means

approximately

85%

of the variance in the target variable

is explained by the model.

---

### Limitations

* Does not measure prediction error directly.
* Can increase simply by adding more features, even if they provide little real predictive value.

---

# <font color='orange'> 8. Adjusted \(R^2\) </font>

Adjusted \(R^2\)

corrects for the number of predictors in the model.

$$
\boxed{
R^2_{\text{adj}}
=
1-
(1-R^2)
\frac{n-1}{n-p-1}
}
$$

where

* \(n\) is the number of observations,
* \(p\) is the number of predictors.

Unlike ordinary \(R^2\),

Adjusted \(R^2\)

penalizes

unnecessary predictors.

It is particularly useful when comparing regression models with different numbers of input features.

---

# <font color='orange'> 9. Comparing the Metrics </font>

| Metric | Units | Penalizes Large Errors | Robust to Outliers |
| :--- | :---: | :---: | :---: |
| MAE | ✔ | ✘ | ✔ |
| MSE | Squared | ✔✔ | ✘ |
| RMSE | ✔ | ✔ | ✘ |
| MAPE | Percentage | Moderate | Limited |
| \(R^2\) | Unitless | — | — |
| Adjusted \(R^2\) | Unitless | — | — |

Each metric emphasizes different aspects of predictive performance.

---

# <font color='orange'> 10. Choosing the Right Metric </font>

The choice depends on the problem.

Use **MAE** when

* interpretability is important,
* outliers should not dominate the evaluation.

Use **RMSE** when

* large prediction errors are especially undesirable,
* optimization using squared errors is appropriate.

Use **MAPE** when

* percentage errors are meaningful,
* the target variable is strictly positive and not close to zero.

Use **\(R^2\)** or **Adjusted \(R^2\)** when

* assessing how well the model explains the variability in the target,
* comparing regression models (Adjusted \(R^2\) is preferable when the number of predictors differs).

In many real-world applications, it is good practice to report multiple metrics rather than relying on only one.

---

# <font color='orange'> 11. Practical Implementation </font>

Scikit-Learn provides implementations of all major regression metrics.

```python
from sklearn.metrics import (

    mean_absolute_error,

    mean_squared_error,

    mean_absolute_percentage_error,

    r2_score

)

mae = mean_absolute_error(
    y_true,
    y_pred
)

mse = mean_squared_error(
    y_true,
    y_pred
)

rmse = mean_squared_error(
    y_true,
    y_pred,
    squared=False
)

mape = mean_absolute_percentage_error(
    y_true,
    y_pred
)

r2 = r2_score(
    y_true,
    y_pred
)
```

These functions are commonly used to evaluate regression models after prediction.

---

# <font color='orange'> 12. Practical Workflow </font>

A typical regression evaluation workflow is

```
Train Model

↓

Predict Test Data

↓

Compute Residuals

↓

Calculate MAE

↓

Calculate RMSE

↓

Calculate R²

↓

Interpret Results

↓

Compare Models
```

Reporting several complementary metrics provides a more complete assessment of model performance.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Metric | Primary Purpose |
| :--- | :--- |
| MAE | Average magnitude of prediction errors |
| MSE | Penalize large errors strongly |
| RMSE | Typical prediction error in the original units |
| MAPE | Average percentage error |
| \(R^2\) | Proportion of variance explained |
| Adjusted \(R^2\) | Compare models with different numbers of predictors |
| Main Goal | Quantify predictive accuracy and compare regression models |

> **Key Insight:** No single regression metric captures every aspect of model performance. MAE provides an easily interpretable measure of the average prediction error, MSE and RMSE emphasize large errors, MAPE expresses errors relative to the magnitude of the target variable, and \(R^2\) quantifies how well the model explains the observed variability. Selecting appropriate evaluation metrics—and often reporting several together—provides a more complete and reliable assessment of regression model performance.

# <font color='blue'> Evaluation Metrics for Classification Models </font>

Classification models predict discrete class labels rather than continuous numerical values. Consequently, evaluating classification models requires different performance measures from those used in regression. Simply counting the proportion of correct predictions is often insufficient because different types of prediction errors may have very different practical consequences. For example, failing to detect a serious disease may be far more costly than incorrectly diagnosing a healthy patient.

Most classification evaluation metrics are derived from the **confusion matrix**, which summarizes the relationship between the true class labels and the model's predictions. From this matrix, a variety of metrics—including accuracy, precision, recall, specificity, F1-score, ROC curves, and precision–recall curves—can be computed, each highlighting different aspects of model performance.

This chapter introduces the principal evaluation metrics used for classification, explains how they are computed and interpreted, discusses their advantages and limitations, and provides guidance for selecting appropriate metrics in different machine learning applications.

---

# <font color='orange'> 1. Why Classification Metrics Are Different </font>

Suppose we build a model to detect

```
Cancer

↓

Positive

Negative
```

Two models may achieve exactly the same

```
95% Accuracy
```

yet one may miss nearly every cancer patient.

Clearly,

accuracy alone cannot always measure model quality.

Different applications require different evaluation metrics.

---

# <font color='orange'> 2. The Confusion Matrix </font>

Almost every classification metric begins with the

**confusion matrix**.

For a binary classification problem,

the confusion matrix is

| | Predicted Positive | Predicted Negative |
| :--- | :---: | :---: |
| **Actual Positive** | True Positive (TP) | False Negative (FN) |
| **Actual Negative** | False Positive (FP) | True Negative (TN) |

This table summarizes every possible prediction outcome.

---

# <font color='orange'> 3. Understanding the Four Outcomes </font>

### True Positive (TP)

The model correctly predicts

the positive class.

Example

```
Patient has disease

↓

Model predicts disease
```

---

### True Negative (TN)

The model correctly predicts

the negative class.

---

### False Positive (FP)

The model predicts

positive,

but the true class is negative.

Also known as

a **Type I Error**.

---

### False Negative (FN)

The model predicts

negative,

but the true class is positive.

Also known as

a **Type II Error**.

---

# <font color='orange'> 4. Accuracy </font>

Accuracy measures

the proportion of correct predictions.

$$
\boxed{
\mathrm{Accuracy}
=
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

---

### Interpretation

```
Accuracy = 96%
```

means

96%

of all predictions

were correct.

---

### Advantages

* Simple.
* Easy to understand.

---

### Limitations

Accuracy can be highly misleading

for imbalanced datasets.

---

# <font color='orange'> 5. Why Accuracy Can Be Misleading </font>

Suppose

```
99 Healthy Patients

1 Cancer Patient
```

A model predicts

```
Healthy

for everyone.
```

Accuracy becomes

```
99%
```

Yet

the model fails to detect

every cancer case.

The model is therefore practically useless,

despite its high accuracy.

---

# <font color='orange'> 6. Precision </font>

Precision answers the question

> **When the model predicts the positive class, how often is it correct?**

$$
\boxed{
\mathrm{Precision}
=
\frac{TP}
{TP+FP}
}
$$

---

### Interpretation

High precision

means

few false positives.

Precision is particularly important when

false alarms are costly.

Examples include

* spam detection,
* fraud detection,
* recommendation systems.

---

# <font color='orange'> 7. Recall (Sensitivity) </font>

Recall answers the question

> **How many of the actual positive cases did the model detect?**

$$
\boxed{
\mathrm{Recall}
=
\frac{TP}
{TP+FN}
}
$$

Recall is also called

**Sensitivity**

or the

**True Positive Rate (TPR)**.

---

### Interpretation

High recall

means

few false negatives.

Recall is particularly important when

missing positive cases

is costly.

Examples include

* cancer diagnosis,
* fault detection,
* disaster warning systems.

---

# <font color='orange'> 8. Specificity </font>

Specificity measures

how well the model identifies

negative cases.

$$
\boxed{
\mathrm{Specificity}
=
\frac{TN}
{TN+FP}
}
$$

Specificity is also called

the

**True Negative Rate (TNR).**

---

### Interpretation

High specificity

means

few false positives.

Medical screening tests often seek both

high sensitivity

and

high specificity,

although improving one may reduce the other.

---

# <font color='orange'> 9. F1-Score </font>

Precision and recall often compete.

The **F1-score**

combines them

using the harmonic mean.

$$
\boxed{
F_1
=
2
\frac{
\mathrm{Precision}
\times
\mathrm{Recall}
}{
\mathrm{Precision}
+
\mathrm{Recall}
}
}
$$

---

### Interpretation

The F1-score is high

only when

both precision

and recall

are high.

It is particularly useful for

imbalanced datasets.

---

# <font color='orange'> 10. ROC Curve and AUC </font>

Many classifiers produce

prediction probabilities

rather than fixed class labels.

Different probability thresholds

produce different

precision and recall values.

The **Receiver Operating Characteristic (ROC) curve**

plots

```
True Positive Rate

vs

False Positive Rate
```

for all possible classification thresholds.

The **Area Under the Curve (AUC)**

summarizes the overall discriminatory ability of the classifier.

| AUC | Interpretation |
| :---: | :--- |
| 1.0 | Perfect classifier |
| 0.5 | Random guessing |
| <0.5 | Worse than random (often indicates a systematic reversal of predictions) |

A higher AUC indicates better ability to distinguish between the classes across different thresholds.

---

# <font color='orange'> 11. Precision–Recall Curve </font>

For highly imbalanced datasets,

the **Precision–Recall (PR) Curve**

is often more informative

than the ROC curve.

It plots

```
Precision

vs

Recall
```

across different probability thresholds.

The **Average Precision (AP)** summarizes the overall precision–recall trade-off into a single value.

When the positive class is rare,

the PR curve typically provides a clearer picture of classifier performance than the ROC curve.

---

# <font color='orange'> 12. Log Loss </font>

Some classifiers produce

predicted probabilities

rather than only class labels.

Log Loss

measures

how well calibrated

those probabilities are.

For binary classification,

$$
\boxed{
\mathrm{LogLoss}
=
-\frac{1}{n}
\sum_{i=1}^{n}
\left[
y_i\log(p_i)
+
(1-y_i)\log(1-p_i)
\right]
}
$$

where

* \(y_i\) is the true class label (0 or 1),
* \(p_i\) is the predicted probability of the positive class.

Lower values indicate better probabilistic predictions.

Log Loss heavily penalizes confident but incorrect predictions.

---

# <font color='orange'> 13. Choosing the Right Metric </font>

The most appropriate metric depends on the application.

| Situation | Recommended Metric |
| :--- | :--- |
| Balanced classes | Accuracy |
| False positives are expensive | Precision |
| False negatives are expensive | Recall |
| Need balance between precision and recall | F1-score |
| Compare classifiers across thresholds | ROC-AUC |
| Highly imbalanced classes | Precision–Recall Curve / Average Precision |
| Evaluate predicted probabilities | Log Loss |

Selecting the correct metric requires understanding the real-world consequences of different types of prediction errors.

# <font color='blue'> Comparing and Selecting Machine Learning Models </font>

Developing a machine learning model rarely involves training a single algorithm. Instead, practitioners typically evaluate multiple candidate models, compare their predictive performance, tune their hyperparameters, and select the model that best satisfies the objectives of the problem. While predictive accuracy is undoubtedly important, it is rarely the only consideration. Factors such as interpretability, computational efficiency, robustness, scalability, and deployment constraints often play equally significant roles in determining which model should ultimately be used.

Model selection is therefore a multi-criteria decision-making process rather than simply choosing the model with the highest evaluation score. Reliable model comparison requires consistent evaluation procedures, appropriate performance metrics, and an understanding of the trade-offs between model complexity and practical utility.

This chapter summarizes the principles of fair model comparison, discusses the factors that influence model selection, and presents a practical workflow for choosing machine learning models in real-world applications.

---

# <font color='orange'> 1. Why Compare Multiple Models? </font>

There is no universally best machine learning algorithm.

The performance of a model depends on

* the dataset,
* the problem,
* the available features,
* the evaluation metric.

Consequently,

several candidate models should usually be evaluated before selecting one.

---

# <font color='orange'> 2. A Typical Model Comparison Workflow </font>

Model selection typically follows the sequence

```
Prepare Data

↓

Choose Candidate Models

↓

Cross-Validation

↓

Hyperparameter Tuning

↓

Compare Performance

↓

Select Final Model

↓

Evaluate Once on Test Set

↓

Deploy
```

Notice that the independent test set is used only after the final model has been chosen.

---

# <font color='orange'> 3. Comparing Models Fairly </font>

A fair comparison requires that all models be evaluated under the same conditions.

This means

* using the same training and validation splits,
* applying identical preprocessing,
* using the same cross-validation strategy,
* evaluating with the same performance metric.

Changing any of these factors makes the comparison unreliable.

---

# <font color='orange'> 4. Choosing the Right Evaluation Metric </font>

Different problems require different metrics.

| Problem | Typical Metrics |
| :--- | :--- |
| Regression | MAE, RMSE, \(R^2\), Adjusted \(R^2\) |
| Balanced Classification | Accuracy, ROC-AUC |
| Imbalanced Classification | Precision, Recall, F1-score, Precision–Recall Curve |

The evaluation metric should reflect the real-world objective rather than simply the easiest quantity to compute.

---

# <font color='orange'> 5. Cross-Validation Results </font>

Cross-validation produces more than a single score.

Example

```
Accuracy

91%

92%

90%

93%

91%
```

Instead of reporting only

```
91.4%
```

it is often useful to report

```
Mean ± Standard Deviation

91.4% ± 1.1%
```

The standard deviation indicates how consistently the model performs across different folds.

A model with slightly lower mean performance but substantially lower variability may be preferable in applications where reliability is important.

---

# <font color='orange'> 6. Simplicity vs Complexity </font>

Suppose two models produce

```
Model A

Accuracy = 94%
```

```
Model B

Accuracy = 94.3%
```

If Model A is substantially simpler,

easier to interpret,

and faster to deploy,

it may be the better choice.

This idea is closely related to

**Occam's Razor**,

which favors simpler explanations when competing alternatives perform similarly.

---

# <font color='orange'> 7. Interpretability vs Predictive Performance </font>

Some applications require

high interpretability.

Examples include

* healthcare,
* finance,
* legal decision-making.

Other applications prioritize

maximum predictive accuracy.

Examples include

* recommendation systems,
* image recognition,
* speech recognition.

Choosing a model therefore often involves balancing

interpretability

against

predictive performance.

---

# <font color='orange'> 8. Computational Cost </font>

Machine learning models differ substantially in computational requirements.

Important considerations include

* training time,
* prediction latency,
* memory usage,
* scalability to large datasets.

For example,

a model with slightly lower accuracy but much faster inference may be preferable for real-time systems.

---

# <font color='orange'> 9. Robustness </font>

A useful model should perform reliably

when presented with

new observations

that are representative of the same problem.

Indicators of robustness include

* consistent cross-validation scores,
* stable performance across different datasets,
* low sensitivity to small perturbations in the training data.

Robustness is often more valuable than marginal improvements in average performance.

---

# <font color='orange'> 10. Practical Decision Criteria </font>

When comparing models,

consider

| Criterion | Questions to Ask |
| :--- | :--- |
| Predictive Performance | How accurate is the model? |
| Generalization | Does it perform well on unseen data? |
| Interpretability | Can its predictions be explained? |
| Computational Cost | Is training or inference expensive? |
| Robustness | Are results consistent across folds? |
| Scalability | Can it handle larger datasets? |
| Deployment Constraints | Can it satisfy real-world requirements? |

Rarely does one model dominate every criterion.

---

# <font color='orange'> 11. Practical Example </font>

Suppose we compare three classifiers.

| Model | Accuracy | Training Time | Interpretability |
| :--- | ---: | :---: | :---: |
| Logistic Regression | 92% | Fast | High |
| Random Forest | 95% | Moderate | Moderate |

Which model is best?

The answer depends on the application.

For a clinical decision support system,

interpretability may justify selecting Logistic Regression.

For an offline fraud detection system,

Gradient Boosting may be preferred because maximizing predictive performance outweighs interpretability.

# <font color='orange'> 12. Reporting Results </font>

A complete machine learning report should include

* dataset description,
* preprocessing steps,
* cross-validation strategy,
* evaluation metrics,

* final test performance,
* limitations of the model.

Transparent reporting enables others to reproduce and evaluate the results.

Following this workflow reduces the risk of selecting a model that performs well only by chance.

By comparing candidate models using consistent evaluation procedures and selecting the model that best aligns with the objectives of the application, practitioners can build machine learning systems that are both accurate and practically useful.
