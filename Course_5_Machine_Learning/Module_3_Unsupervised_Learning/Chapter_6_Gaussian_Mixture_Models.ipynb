# <font color='blue'> Gaussian Mixture Models (GMMs): Probabilistic Clustering </font>

The clustering algorithms studied thus far have assigned each observation to a single cluster. This approach, known as **hard clustering**, assumes that cluster boundaries are well defined and that every observation belongs exclusively to one group. While appropriate for many datasets, this assumption is often unrealistic in practice. Real-world data frequently exhibit overlapping clusters in which observations cannot be assigned unambiguously to a single category.

Gaussian Mixture Models (GMMs) address this limitation by adopting a probabilistic perspective. Rather than assigning observations exclusively to one cluster, GMMs model the data as originating from a mixture of Gaussian probability distributions. Each observation is assigned a probability of belonging to every cluster, resulting in a **soft clustering** solution.

This probabilistic framework provides considerably greater flexibility than K-Means, allowing clusters to differ in shape, orientation, and size while naturally representing uncertainty in cluster membership.

---

# <font color='orange'> 1. Motivation </font>

Suppose we observe the following data.

```
Cluster A        Cluster B

● ● ● ●     ● ● ● ●
     ● ● ●
```

Some observations lie

between

Should they belong to

Cluster A

or

Cluster B?

K-Means forces a decision.

Gaussian Mixture Models instead answer

> **How likely is this observation to belong to each cluster?**

# <font color='orange'> 2. Hard Clustering vs Soft Clustering </font>

### Hard Clustering

Each observation belongs to exactly one cluster.

```
Observation

↓

Cluster 2
```

Probability

```
Cluster 1 : 0%

Cluster 2 : 100%

Cluster 3 : 0%
```

---

### Soft Clustering

Each observation belongs to every cluster

with different probabilities.

Example

```
Cluster 1 : 25%

Cluster 2 : 70%

Cluster 3 : 5%
```

The probabilities always sum to

$$
1.
$$

---

# <font color='orange'> 3. The Central Idea </font>

Instead of representing clusters by centroids,

GMM assumes

each cluster is generated by

a Gaussian distribution.

Conceptually,

```
Gaussian 1

+

Gaussian 2

+

Gaussian 3

↓

Observed Dataset
```

The complete dataset is therefore viewed as

a mixture of Gaussian distributions.

---

# <font color='orange'> 4. What is a Gaussian Distribution? </font>

Recall from the supervised learning notebook,

a Gaussian distribution is characterized by

* a mean,

$$
\mu,
$$

* a variance,

$$
\sigma^2.
$$

In multiple dimensions,

the variance becomes

the covariance matrix,

$$
\Sigma.
$$

Each Gaussian therefore describes

both

* the centre,

and

* the shape

of a cluster.

---

# <font color='orange'> 5. Components of a Gaussian Mixture Model </font>

A GMM consists of

multiple Gaussian components.

Each component has

### Mean

$$
\mu_k
$$

The centre of the Gaussian.

---

### Covariance Matrix

$$
\Sigma_k
$$

Determines

* size,
* orientation,
* spread.

---

### Mixing Coefficient

$$
\pi_k
$$

Represents

the probability that

an observation originates from

component

$$
k.
$$

The mixing coefficients satisfy

$$
\sum_{k=1}^{K}\pi_k=1.
$$

---

# <font color='orange'> 6. The Mixture Model </font>

Suppose

there are

$$
K
$$

Gaussian components.

The probability density of an observation is

$$
\boxed{
p(x)
=
\sum_{k=1}^{K}
\pi_k
\,
\mathcal N
(
x
\mid
\mu_k,
\Sigma_k
).
}
$$

This equation states

that the observed data are generated

by combining several Gaussian distributions,

each weighted by its mixing coefficient.

# <font color='blue'> Expectation-Maximization (EM) Algorithm for Gaussian Mixture Models </font>

The previous chapter introduced Gaussian Mixture Models (GMMs) as probabilistic clustering models in which the observed data are assumed to be generated by a weighted combination of Gaussian distributions. Unlike K-Means, which assigns each observation to a single cluster, GMMs estimate the probability that every observation belongs to each Gaussian component.

A natural question therefore arises: **how can the parameters of these Gaussian components be estimated when the true cluster memberships are unknown?**

This problem is solved using the **Expectation-Maximization (EM) algorithm**, an iterative optimization procedure for estimating model parameters in the presence of hidden or latent variables. In the context of GMMs, the latent variables correspond to the unknown cluster memberships of the observations. EM alternates between estimating these hidden memberships and updating the Gaussian parameters until the model converges.

This chapter introduces the intuition behind the EM algorithm, derives its two principal steps, and explains how it is used to train Gaussian Mixture Models.

---

# <font color='orange'> 1. Why Do We Need EM? </font>

Suppose we observe a dataset

```
● ● ● ● ● ● ●
```

We assume that the data were generated by

three Gaussian distributions.

However,

we do **not** know

which Gaussian generated each observation.

The unknown cluster memberships are called

**latent variables**.

The EM algorithm estimates

both

* the hidden memberships,
* the Gaussian parameters.

---

# <font color='orange'> 2. The Unknown Quantities </font>

For each Gaussian component,

we wish to estimate

* the mean,

$$
\mu_k,
$$

* the covariance matrix,

$$
\Sigma_k,
$$

* the mixing coefficient,

$$
\pi_k.
$$

Unfortunately,

these quantities depend on knowing

which observations belong to each component.

At the same time,

the cluster memberships depend on

the Gaussian parameters.

This circular dependence motivates the iterative EM procedure.

---

# <font color='orange'> 3. The Basic Idea </font>

Instead of solving everything simultaneously,

EM alternates between

estimating

and

updating.

Conceptually,

```
Estimate Memberships

↓

Update Parameters

↓

Estimate Memberships

↓

Update Parameters

↓

Convergence
```

Each iteration improves the model.

---

# <font color='orange'> 4. Latent Variables </font>

Introduce a hidden variable

$$
z_i,
$$

for each observation

$$
x_i.
$$

The variable

$$
z_i
$$

indicates

which Gaussian component generated

the observation.

Unlike supervised learning,

these variables are **not observed**.

The EM algorithm estimates them probabilistically.

---

# <font color='orange'> 5. The Expectation (E) Step </font>

During the E-step,

the current model parameters are assumed to be known.

The objective is to compute

the probability that each Gaussian component generated each observation.

These probabilities are called the **responsibilities**.

Mathematically,

$$
\boxed{
\gamma(z_{ik})
=
P(z_i=k\mid x_i).
}
$$

Each responsibility satisfies

$$
0
\le
\gamma(z_{ik})
\le
1,
$$

and for every observation,

$$
\boxed{
\sum_{k=1}^{K}
\gamma(z_{ik})
=
1.
}
$$

---

### Interpretation

Example

| Cluster | Responsibility |
| :--- | ---: |
| 1 | 0.75 |
| 2 | 0.20 |
| 3 | 0.05 |

Rather than assigning the observation exclusively to one cluster,

the model represents uncertainty using probabilities.

---

# <font color='orange'> 6. The Maximization (M) Step </font>

During the M-step,

the responsibilities computed in the E-step are treated as weights.

The Gaussian parameters are then updated.

---

### Mixing Coefficients

$$
\boxed{
\pi_k
=
\frac{1}{n}
\sum_{i=1}^{n}
\gamma(z_{ik}).
}
$$

---

### Means

$$
\boxed{
\mu_k
=
\frac{
\sum_i
\gamma(z_{ik})x_i
}{
\sum_i
\gamma(z_{ik})
}.
}
$$

The updated mean is therefore a

**weighted average**

of the observations.

---

### Covariance Matrices

$$
\boxed{
\Sigma_k
=
\frac{
\sum_i
\gamma(z_{ik})
(x_i-\mu_k)
(x_i-\mu_k)^T
}{
\sum_i
\gamma(z_{ik})
}.
}
$$

The covariance matrix is likewise estimated using the responsibilities as weights.

---

# <font color='orange'> 7. The EM Algorithm </font>

The complete procedure is

---

### Step 1

Initialize

* means,
* covariance matrices,
* mixing coefficients.

---

### Step 2

Perform the E-step.

Compute the responsibilities.

---

### Step 3

Perform the M-step.

Update

* means,
* covariance matrices,
* mixing coefficients.

---

### Step 4

Repeat

Steps 2 and 3

until convergence.

---

# <font color='orange'> 8. Log-Likelihood </font>

Rather than minimizing WCSS,

GMMs maximize

the **log-likelihood**

of the observed data.

The log-likelihood is

$$
\boxed{
\log L
=
\sum_{i=1}^{n}
\log
\left(
\sum_{k=1}^{K}
\pi_k
\,
\mathcal N
(
x_i
|
\mu_k,
\Sigma_k
)
\right).
}
$$

Each EM iteration

increases (or leaves unchanged)

this log-likelihood.

---

# <font color='orange'> 9. Convergence </font>

EM continues until

the increase in log-likelihood becomes negligible.

Conceptually,

```
Iteration

↓

Higher Log-Likelihood

↓

Higher Log-Likelihood

↓

Convergence
```

The algorithm is guaranteed to converge,

although only to a **local optimum**.

---

# <font color='orange'> 10. Relationship to K-Means </font>

The similarities are striking.

| K-Means | EM for GMM |
| :--- | :--- |
| Hard cluster assignments | Probabilistic responsibilities |
| Centroids | Gaussian means |
| WCSS minimization | Log-likelihood maximization |
| One cluster per observation | Probability for every cluster |

K-Means may therefore be viewed as a special case of Gaussian Mixture Models in which

* covariance matrices are identical,
* clusters are spherical,
* assignments are hard rather than probabilistic.

---

# <font color='orange'> 11. Choosing the Number of Components </font>

Like K-Means,

Gaussian Mixture Models require

the number of components

$$
K.
$$

Instead of the Elbow Method,

probabilistic models are commonly compared using

* Akaike Information Criterion (AIC),
* Bayesian Information Criterion (BIC).

Both criteria balance

* goodness of fit,
* model complexity.

Smaller values generally indicate better models.

| Concept | Description |
| :--- | :--- |

> **Key Insight:** The Expectation-Maximization algorithm enables Gaussian Mixture Models to estimate cluster parameters even when the true cluster memberships are unknown. By alternating between estimating the probability that each observation belongs to every Gaussian component (Expectation step) and updating the model parameters using these probabilities (Maximization step), EM progressively increases the likelihood of the observed data. This iterative probabilistic framework provides a flexible alternative to hard clustering methods and forms one of the most important optimization techniques in statistical machine learning.
