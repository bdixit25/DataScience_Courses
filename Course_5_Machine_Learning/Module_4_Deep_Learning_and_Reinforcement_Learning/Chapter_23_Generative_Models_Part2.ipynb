# <font color='blue'> Variational Autoencoders (VAEs): Learning a Probabilistic Latent Space </font>

Standard autoencoders learn compact latent representations by compressing input data into deterministic latent vectors and reconstructing the original observations. Although these latent representations are often useful for dimensionality reduction and feature learning, they do not possess any particular probabilistic structure. Consequently, randomly sampling points from the latent space does not necessarily produce meaningful outputs.

A **Variational Autoencoder (VAE)** addresses this limitation by learning a **probability distribution** over the latent space instead of a single deterministic representation. Rather than mapping each input to one latent vector, the encoder predicts the parameters of a probability distribution—typically a multivariate Gaussian—from which latent vectors are sampled.

This probabilistic formulation enables VAEs to generate entirely new data by sampling from the learned latent distribution, making them one of the first practical deep generative models.

---

# <font color='orange'> 1. Motivation </font>

Consider

a standard autoencoder.

```
Image

↓

Encoder

↓

Latent Vector

↓

Decoder

↓

Reconstruction
```

Each image

maps to

one point

in latent space.

Now suppose

we randomly choose

another point.

```
Random Point

↓

Decoder

↓

?
```

There is no guarantee

that the decoder

will generate

a meaningful image.

The latent space

is irregular

and poorly organized.

---

# <font color='orange'> 2. The Central Idea </font>

Instead of learning

one latent vector,

a VAE learns

a probability distribution.

Conceptually,

```
Image

↓

Encoder

↓

Mean

Variance

↓

Sampling

↓

Latent Vector

↓

Decoder

↓

Generated Image
```

Every input

corresponds to

a region

of latent space,

not a single point.

---

# <font color='orange'> 3. Gaussian Latent Distribution </font>

The encoder predicts

two vectors.

```
μ

(mean)
```

and

```
σ²

(variance)
```

These define

a Gaussian distribution

$$
\boxed{
z
\sim
\mathcal N(\mu,\sigma^2)
}
$$

Rather than storing

one latent vector,

the network stores

the distribution

from which

latent vectors

can be sampled.

---

# <font color='orange'> 4. Why Learn a Distribution? </font>

Suppose

two images

are similar.

```
Cat 1

↓

Latent Region
```

```
Cat 2

↓

Nearby Latent Region
```

The latent space becomes

continuous

and

smooth.

Nearby latent points

produce

similar outputs.

This makes

interpolation

and sampling

possible.

---

# <font color='orange'> 5. Sampling Problem </font>

A challenge arises.

Sampling

is random.

```
μ

↓

Random Sampling

↓

z
```

Random operations

are not directly differentiable,

making gradient-based learning difficult.

How can we train

the encoder?

---

# <font color='orange'> 6. The Reparameterization Trick </font>

The solution is

the **reparameterization trick**.

Instead of sampling

directly,

we write

$$
\boxed{
z
=
\mu
+
\sigma
\odot
\epsilon
}
$$

where

$$
\epsilon
\sim
\mathcal N(0,1).
$$

The randomness

is isolated

in

\(\epsilon\),

while

\(\mu\)

and

\(\sigma\)

remain differentiable.

This allows

backpropagation

to proceed normally.

---

# <font color='orange'> 7. KL-Divergence Regularization </font>

To ensure

a well-behaved latent space,

the learned distribution

is encouraged

to resemble

a standard normal distribution

$$
\boxed{
\mathcal N(0,1).
}
$$

This is achieved

using

the

**Kullback–Leibler (KL) divergence**.

For a Gaussian latent distribution,

the KL loss is

$$
\boxed{
L_{KL}
=
-\frac12
\sum
\left(
1
+
\log\sigma^2
-
\mu^2
-
\sigma^2
\right)
}
$$

This regularization

encourages

different latent regions

to connect smoothly.

---

# <font color='orange'> 8. ELBO Loss </font>

The total VAE loss

contains

two terms.

$$
\boxed{
L
=
L_{reconstruction}
+
L_{KL}
}
$$

The reconstruction term

encourages

accurate reconstruction.

The KL term

encourages

a smooth,

continuous,

Gaussian latent space.

Together,

these define

the

**Evidence Lower Bound (ELBO)**,

which is the objective maximized during VAE training (equivalently, its negative is minimized as the loss).

---

# <font color='orange'> 9. Information Flow </font>

Training proceeds as

```
Input

↓

Encoder

↓

μ , σ

↓

Sampling

↓

Latent Vector

↓

Decoder

↓

Reconstruction

↓

Reconstruction Loss

+

KL Loss

↓

Update Parameters
```

Unlike

a standard autoencoder,

the latent space

is explicitly regularized.

---

# <font color='orange'> 10. Sampling New Data </font>

After training,

generation becomes simple.

```
Random Sample

↓

N(0,1)

↓

Decoder

↓

New Image
```

No input image

is required.

The decoder

can create

entirely new samples.

This makes

VAEs

true generative models.

# <font color='orange'> 15. Common Misconceptions </font>

### VAEs Memorize Training Images

False.

VAEs learn a probability distribution over the latent space rather than storing exact copies of training examples.

---

### Sampling Uses the Mean Only

False.

Sampling combines the predicted mean, variance, and random noise through the reparameterization trick.

---

### KL-Divergence Improves Reconstruction

Not directly.

The KL term regularizes the latent space, while the reconstruction term encourages accurate outputs. There is often a trade-off between these objectives.

---

### Every Latent Dimension Has an Obvious Meaning

False.

Although some latent dimensions may become interpretable, this is not guaranteed without additional constraints or specialized training objectives.

| :--- | :--- |

This probabilistic framework established one of the foundational approaches to deep generative modelling.

# <font color='blue'> Generative Adversarial Networks (GANs): Learning Through Competition </font>

Variational Autoencoders introduced probabilistic latent spaces that allow new data to be generated through sampling. Although effective, VAEs often produce images that appear overly smooth or blurry because they optimize a reconstruction objective that averages over many possible outputs.

In 2014, Ian Goodfellow and collaborators proposed a fundamentally different approach called the **Generative Adversarial Network (GAN)**. Rather than explicitly minimizing reconstruction error, a GAN trains two neural networks simultaneously in a competitive game.

The first network, called the **Generator**, attempts to produce synthetic data that resemble the training data. The second network, called the **Discriminator**, attempts to distinguish real data from generated data. Through this adversarial process, both networks continually improve, often producing remarkably realistic synthetic images.

GANs became one of the most influential developments in generative modelling and remain widely used for image synthesis, data augmentation, super-resolution, and scientific simulation.

---

# <font color='orange'> 1. Motivation </font>

Suppose

we wish to generate

realistic faces.

Instead of

measuring

pixel-by-pixel reconstruction,

we ask

another neural network

to judge

whether the generated image

looks real.

The generator therefore learns

to fool

the discriminator.

---

# <font color='orange'> 2. Two Neural Networks </font>

A GAN contains

two models.

### Generator

```
Random Noise

↓

Generator

↓

Fake Image
```

The generator

creates synthetic data.

---

### Discriminator

```
Image

↓

Discriminator

↓

Real

or

Fake
```

The discriminator

acts as

a binary classifier.

---

# <font color='orange'> 3. Adversarial Training </font>

Training proceeds

as a competition.

```
Random Noise

↓

Generator

↓

Fake Image

↓

Discriminator

↓

Fake?
```

Simultaneously,

real images

are also presented

to the discriminator.

The discriminator learns

to distinguish

real

from

generated samples.

The generator learns

to fool

the discriminator.

---

# <font color='orange'> 4. Generator </font>

The generator receives

a latent vector

sampled from

a simple probability distribution,

typically

$$
\boxed{
z
\sim
\mathcal N(0,1)
}
$$

or

a uniform distribution.

The generator computes

$$
\boxed{
G(z)
}
$$

which represents

a synthetic sample.

Initially,

these samples

look like random noise.

As training progresses,

they become

increasingly realistic.

---

# <font color='orange'> 5. Discriminator </font>

The discriminator

receives either

```
Real Image
```

or

```
Generated Image
```

and predicts

the probability

that the image

is real.

Mathematically,

$$
\boxed{
D(x)
\in
[0,1]
}
$$

where

* 1 indicates "real,"
* 0 indicates "fake."

---

# <font color='orange'> 6. The Minimax Game </font>

The GAN objective

is formulated

as a minimax optimization problem.

$$
\boxed{
\min_G
\;
\max_D
\;
V(D,G)
}
$$

where

$$
V(D,G)
=
\mathbb E_{x\sim p_{data}}
[
\log D(x)
]
+
\mathbb E_{z}
[
\log(1-D(G(z)))
].
$$

The discriminator

tries to maximize

this objective,

while

the generator

tries to minimize it

by producing samples

that the discriminator

classifies as real.

---

# <font color='orange'> 7. Training Procedure </font>

Training alternates

between

the two networks.

Step 1

```
Update Discriminator
```

using

* real images,
* generated images.

---

Step 2

```
Freeze Discriminator

↓

Update Generator
```

The generator

receives gradients

through the discriminator,

encouraging

more realistic outputs.

---

# <font color='orange'> 8. Nash Equilibrium </font>

Ideally,

training converges

to a point where

```
Generator

≈

Real Data Distribution
```

At this stage,

the discriminator

cannot distinguish

real

from

generated samples

better than random guessing.

This corresponds

to a Nash equilibrium

in game theory.

---

# <font color='orange'> 13. Common Misconceptions </font>

### The Generator Learns Alone

False.

The generator improves only because the discriminator provides informative gradients during training.

---

### The Discriminator Is Used During Generation

Typically, no.

After training,

False.

# <font color='blue'> Conditional Generative Adversarial Networks (cGANs): Controlled Data Generation </font>

The original Generative Adversarial Network generates synthetic data by transforming random noise into samples that resemble the training distribution. While this approach can produce realistic outputs, the generator has no direct control over what type of sample is created. Each generated example is determined solely by the random latent vector.

A **Conditional Generative Adversarial Network (cGAN)** extends the GAN framework by conditioning both the generator and discriminator on additional information. This conditioning information may include class labels, text descriptions, segmentation masks, other images, or any auxiliary data relevant to the generation task.

By incorporating conditioning variables, cGANs enable **controlled generation**, allowing users to specify desired properties of the generated output. This simple extension greatly expanded the practical usefulness of GANs and laid the foundation for image-to-image translation, super-resolution, text-guided image synthesis, and many other conditional generative models.

---

Suppose

we train

a standard GAN

```
Noise

↓

Generator

↓

Random Digit
```

The model

may generate

```
3
```

or

```
8
```

or

```
5
```

but we cannot choose

which digit

is produced.

Instead,

we would like to say

```
Generate

the digit 7.
```

This is the motivation

for conditional generation.

---

# <font color='orange'> 2. The Central Idea </font>

\boxed{

}
$$

# <font color='blue'> CycleGAN: Image-to-Image Translation Without Paired Data </font>

Many image-to-image translation problems involve learning a mapping between two visual domains. Traditional conditional generative models assume the availability of paired training examples, where each input image has a corresponding target image. While this assumption holds for some tasks, many important applications lack such paired datasets.

The **Cycle-Consistent Generative Adversarial Network (CycleGAN)** overcomes this limitation by learning mappings between two image domains using only **unpaired data**. Rather than requiring exact input-output pairs, CycleGAN introduces the principle of **cycle consistency**, ensuring that translating an image to another domain and back again approximately recovers the original image.

This innovation enabled high-quality image translation across many domains without expensive paired datasets and became one of the most influential GAN architectures.

---

# <font color='orange'> 1. Motivation </font>

Suppose

we have

two datasets.

```
Horse Images
```

and

```
Zebra Images
```

Notice

there are

no matching pairs.

We only know

```
This is a horse.
```

or

```
This is a zebra.
```

Can a model still learn

horse → zebra

and

zebra → horse?

CycleGAN answers

yes.

---

# <font color='orange'> 2. The Central Idea </font>

Instead of learning

only one mapping,

CycleGAN learns

two mappings.

```
Horse

↓

Generator G

↓

Zebra
```

and

```
Zebra

↓

Generator F

↓

Horse
```

Each generator

learns

the inverse transformation

of the other.

---

# <font color='orange'> 3. Two Generators and Two Discriminators </font>

A CycleGAN contains

four networks.

### Generator G

```
Horse

↓

Zebra
```

---

### Generator F

```
Zebra

↓

Horse
```

---

### Discriminator Dₓ

Distinguishes

real horses

from

generated horses.

---

### Discriminator Dᵧ

Distinguishes

real zebras

from

generated zebras.

Together,

the generators

and discriminators

learn

both domains simultaneously.

---

# <font color='orange'> 4. Why Two Generators? </font>

Suppose

we only learn

```
Horse

↓

Zebra
```

Many different zebra images

could correspond

to the same horse.

How can we ensure

the translated image

still represents

the original horse?

The solution is

to translate back.

```
Horse

↓

Zebra

↓

Horse
```

The recovered horse

should resemble

the original image.

---

# <font color='orange'> 5. Cycle Consistency </font>

Cycle consistency

requires

that

$$
\boxed{
F(G(x))
\approx
x
}
$$

and

$$
\boxed{
G(F(y))
\approx
y
}
$$

where

* \(x\) belongs to the horse domain,
* \(y\) belongs to the zebra domain.

This encourages

the translations

to preserve

the essential content

while changing

only

the domain-specific appearance.

---

# <font color='orange'> 6. Loss Function </font>

CycleGAN optimizes

three objectives.

### Adversarial Loss

Produces realistic images.

---

### Cycle Consistency Loss

Preserves content.

---

### Identity Loss (optional but often beneficial)

Encourages the generators to leave an image largely unchanged when it is already in the target domain, helping preserve colours and other attributes.

The total objective is typically written as

$$
\boxed{
L
=
L_{GAN}
+
\lambda
L_{cycle}
+
\gamma
L_{identity}
}
$$

where

\(\lambda\)

and

\(\gamma\)

control

the relative importance

of the different loss terms.

---

# <font color='orange'> 7. Information Flow </font>

Training proceeds as

```
Horse

↓

Generator G

↓

Fake Zebra

↓

Generator F

↓

Recovered Horse
```

The recovered horse

should match

the original horse.

The same process

occurs

starting from

a zebra image.

---

# <font color='orange'> 8. Why It Works </font>

Without

cycle consistency,

the generator

could simply produce

any realistic zebra.

Cycle consistency

forces

the translated image

to retain

the information needed

to reconstruct

the original horse.

Thus,

appearance changes,

while

underlying structure

is preserved.

---

# <font color='orange'> 11. Advantages </font>

CycleGAN provides several important benefits.

* No paired training data required.
* Learns bidirectional mappings.
* Preserves structural content.
* Flexible across many visual domains.
* Effective for domain adaptation.

---

# <font color='orange'> 12. Limitations </font>

Despite its success,

CycleGAN also has limitations.

* Training remains computationally expensive.
* Large geometric transformations are challenging.
* Cycle consistency does not guarantee semantic correctness.
* It inherits many optimization difficulties common to GANs.

# <font color='orange'> 13. Common Misconceptions </font>

### CycleGAN Requires Paired Images

False.

CycleGAN was specifically designed for **unpaired** image-to-image translation.

---

### Cycle Consistency Means Perfect Reconstruction

False.

Cycle consistency encourages approximate recovery of the original image but does not guarantee pixel-perfect reconstruction.

---

### CycleGAN Changes Only Colours

False.

Depending on the task, CycleGAN can modify textures, patterns, illumination, and other domain-specific characteristics while attempting to preserve overall scene structure.

---

### Two Generators Mean Double the Data

False.

The same unpaired datasets are used. The second generator provides the reverse mapping needed to enforce cycle consistency.

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| CycleGAN | GAN for unpaired image-to-image translation |
| Generator G | Maps domain X → domain Y |
| Generator F | Maps domain Y → domain X |
| Two Discriminators | Judge realism in each domain |
| Cycle Consistency | Translation to the other domain and back should recover the original image |
| Identity Loss | Helps preserve attributes already consistent with the target domain |
| Main Advantage | Learns translations without paired training data |
| Main Limitation | Computationally intensive and still inherits GAN training challenges |

> **Key Insight:** CycleGAN extends adversarial learning to unpaired image-to-image translation by simultaneously learning forward and reverse mappings between two domains. The cycle consistency constraint ensures that translated images preserve the essential content of the original while adopting the appearance of the target domain. This elegant idea removed the need for paired datasets and opened new possibilities for domain adaptation, style transfer, and scientific image translation.
