# <font color='blue'> Diffusion Models: Generating Data by Gradual Denoising </font>

Generative Adversarial Networks demonstrated that highly realistic images could be generated through adversarial learning. However, GANs often suffer from unstable optimization, mode collapse, and difficult hyperparameter tuning. Researchers therefore revisited an alternative idea inspired by statistical physics and denoising autoencoders: instead of learning to generate images directly, a model could learn to reverse a gradual process that destroys images by adding noise.

A **Diffusion Model** defines two complementary stochastic processes. The **forward diffusion process** progressively corrupts data by adding small amounts of random noise over many time steps until the data become nearly indistinguishable from pure Gaussian noise. The **reverse diffusion process** learns to remove this noise step by step, ultimately reconstructing realistic samples.

This simple but powerful framework has become one of the dominant approaches to generative modelling, producing high-quality images, audio, video, molecular structures, and scientific simulations.

# <font color='orange'> 2. The Central Idea </font>

A diffusion model

contains

two processes.

### Forward Diffusion

Adds noise.

```
Image

↓

Noise

↓

More Noise

↓

...

↓

Gaussian Noise
```

---

### Reverse Diffusion

Removes noise.

```
Gaussian Noise

↓

Less Noise

↓

Cleaner Image

↓

...

↓

Generated Image
```

Generation occurs

by running

the reverse process.

---

# <font color='orange'> 3. Forward Diffusion Process </font>

At each time step,

a small amount

of Gaussian noise

is added.

Mathematically,

$$
\boxed{
x_t
=
\sqrt{1-\beta_t}\,x_{t-1}
+
\sqrt{\beta_t}\,\epsilon
}
$$

where

* \(x_t\) is the noisy image,
* \(\beta_t\) is the noise schedule,
* \(\epsilon\sim\mathcal N(0,1)\).

After many steps,

the image

approaches

a standard Gaussian distribution.

---

# <font color='orange'> 4. Reverse Diffusion Process </font>

The neural network

receives

a noisy image

and predicts

how to remove

the added noise.

Conceptually,

```
Noisy Image

↓

Neural Network

↓

Predicted Noise

↓

Subtract Noise

↓

Cleaner Image
```

This procedure

is repeated

until

a realistic sample

emerges.

---

# <font color='orange'> 5. Learning the Noise </font>

A surprising insight

is that

the network

does **not**

usually predict

the clean image

directly.

Instead,

it predicts

the noise

that was added.

```
Noisy Image

↓

Network

↓

Predicted Noise
```

The predicted noise

is then removed

to obtain

a cleaner image.

This objective

has proven

more stable

than directly predicting

the clean image.

---

# <font color='orange'> 6. Training Objective </font>

Training minimizes

the difference

between

the true noise

and

the predicted noise.

A common loss is

Mean Squared Error.

$$
\boxed{
L
=
\left\|
\epsilon
-
\epsilon_\theta
(x_t,t)
\right\|^2
}
$$

where

* \(\epsilon\) is the true noise,
* \(\epsilon_\theta\) is the predicted noise.

The network therefore learns

how to denoise

images

at every noise level.

---

# <font color='orange'> 7. Noise Schedule </font>

The amount

of noise added

at each step

is controlled

by the

**noise schedule**

```
β₁

β₂

β₃

...

β_T
```

A good schedule

ensures

that information

is removed

gradually,

making

the reverse process

learnable.

---

Unlike VAEs,

# <font color='orange'> 14. Common Misconceptions </font>

### Diffusion Models Generate Images in One Step

False.

Generation proceeds through many successive denoising iterations.

---

### The Network Predicts the Final Image Directly

False.

Most diffusion models are trained to predict the added noise rather than the clean image itself.

---

### More Noise Always Improves the Model

False.

The noise schedule must be carefully designed. Excessive or poorly scheduled noise can make the reverse process difficult to learn.

---

### Diffusion Models Are Just Denoising Autoencoders

False.

Although both involve denoising, diffusion models define a probabilistic forward and reverse process over many timesteps. A denoising autoencoder typically learns a single-step corruption-and-reconstruction task.

# <font color='blue'> Latent Diffusion Models (LDMs): Efficient Generative Modelling in Latent Space </font>

Although diffusion models produce remarkably realistic images, they require repeated denoising operations directly on high-dimensional pixel representations. Since every denoising step processes the full image, training and inference are computationally expensive, especially for high-resolution images.

A **Latent Diffusion Model (LDM)** addresses this limitation by first compressing images into a lower-dimensional latent representation using an autoencoder. Diffusion is then performed entirely within this latent space rather than in pixel space. Once denoising is complete, the latent representation is decoded back into an image.

By combining representation learning with diffusion modelling, Latent Diffusion Models achieve image quality comparable to pixel-space diffusion while dramatically reducing computational cost. This architecture forms the basis of **Stable Diffusion** and many modern open-source image generation systems.

# <font color='blue'> Stable Diffusion: A Practical Latent Diffusion System </font>

Latent Diffusion Models demonstrated that high-quality image generation can be achieved efficiently by performing diffusion within a compressed latent space rather than directly on image pixels. Building upon this idea, **Stable Diffusion**, introduced by Stability AI, CompVis, and Runway in 2022, combines several deep learning components into a unified system capable of generating high-resolution images from natural language descriptions.

Rather than representing a fundamentally new learning algorithm, Stable Diffusion is a carefully engineered implementation of latent diffusion that integrates variational autoencoders, Transformer-based text encoders, U-Net denoising networks, cross-attention mechanisms, and guidance strategies into a practical image generation pipeline.

Today, Stable Diffusion has become one of the most influential open-source generative AI systems, enabling text-to-image generation, image editing, inpainting, outpainting, super-resolution, and numerous scientific applications.
