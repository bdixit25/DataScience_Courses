# <font color='blue'> Introduction to Graph Neural Networks (GNNs) </font>

Many deep learning architectures assume that data possess a regular structure. Convolutional Neural Networks operate on images represented as regular two-dimensional grids of pixels, while Recurrent Neural Networks and Transformers process ordered sequences such as text or time series. However, numerous real-world datasets do not naturally fit either representation. Social networks, transportation systems, molecular structures, communication networks, recommendation systems, and particle detector data are all more naturally represented as **graphs**.

A **Graph Neural Network (GNN)** is a class of neural network designed specifically to learn from graph-structured data. Rather than treating observations independently, GNNs explicitly model both the individual entities (nodes) and the relationships between them (edges). By repeatedly exchanging information among neighbouring nodes, GNNs learn representations that capture both local interactions and global graph structure.

Graph Neural Networks have become one of the most important architectures in modern machine learning, particularly in scientific computing, chemistry, recommendation systems, and high-energy physics.

# <font color='blue'> Graph Representation: Nodes, Edges, Adjacency Matrices, and Graph Features </font>

Before a Graph Neural Network can process a graph, the graph must first be represented mathematically. Unlike images, which naturally form regular grids of pixels, graphs are irregular structures consisting of entities connected by relationships. Consequently, graph learning relies on specialized mathematical representations that encode both the topology of the graph and the attributes of its nodes and edges.

This chapter introduces the fundamental mathematical objects used throughout graph machine learning, including adjacency matrices, degree matrices, feature matrices, graph Laplacians, and common graph representations.

These concepts provide the foundation for virtually every modern Graph Neural Network architecture.

---

A graph is written as

$$
\boxed{

}

* \(V\) is the set of nodes (vertices),
* \(E\) is the set of edges.

Example

```
A ----- B
|       |
|       |
C ----- D
```

```
V={A,B,C,D}
```

```
E={(A,B),(A,C),(B,D),(C,D)}
```

# <font color='blue'> Message Passing Neural Networks (MPNNs): The Core Mechanism of Graph Neural Networks </font>

Unlike images or sequences, graphs contain explicit relationships between entities. Consequently, learning on graphs requires information to flow not only through the neural network layers but also across the graph itself. The fundamental mechanism enabling this information exchange is known as **message passing**.

In a Message Passing Neural Network (MPNN), every node repeatedly exchanges information with its neighbouring nodes. During each iteration, a node receives messages from its neighbours, aggregates those messages, and updates its own representation. After multiple rounds of message passing, each node contains information not only about itself but also about increasingly distant parts of the graph.

This framework provides the conceptual foundation for nearly every modern Graph Neural Network architecture, including Graph Convolutional Networks (GCNs), GraphSAGE, Graph Attention Networks (GATs), and many Graph Transformers.

---

# <font color='orange'> 1. Motivation </font>

Suppose

we wish to classify

one detector hit.

```
Hit B
```

Looking only at

Hit B

may be insufficient.

Neighbouring hits contain

important information

about

* trajectory,
* timing,
* energy deposition,
* detector geometry.

The node should therefore

learn from

its neighbours.

---

# <font color='orange'> 2. The Central Idea </font>

Every node

performs

three operations.

```
Receive Messages

↓

Aggregate Messages

↓

Update Representation
```

These three steps

are repeated

across

the entire graph.

---

# <font color='orange'> 3. Initial Node Features </font>

Suppose

every node

starts with

its own feature vector

$$
\boxed{
h_i^{(0)}
}
$$

For a detector hit,

this may contain

* x,
* y,
* z,
* deposited energy,
* detector layer.

Initially,

each node

knows

only about itself.

---

# <font color='orange'> 4. Message Construction </font>

Each neighbouring node

constructs

a message.

Conceptually,

```
Neighbour

↓

Message Function

↓

Message
```

Mathematically,

a message

from node

\(j\)

to node

\(i\)

is written as

$$
\boxed{
m_{j\rightarrow i}
=
M
\left(
h_j,
e_{ij}
\right)
}
$$

where

* \(h_j\) is the neighbour's representation,
* \(e_{ij}\) represents optional edge features,
* \(M\) is the message function.

---

# <font color='orange'> 5. Message Aggregation </font>

Node

\(i\)

receives

messages

from

all neighbouring nodes.

```
Neighbour 1

↓

Neighbour 2

↓

Neighbour 3

↓

Aggregate
```

The aggregation function

must be

**permutation invariant**,

meaning the result does not depend on the order in which neighbours are processed.

Common aggregation functions include

* sum,
* mean,
* maximum.

Mathematically,

$$
\boxed{
m_i
=
\bigoplus_{j\in\mathcal N(i)}
m_{j\rightarrow i}
}
$$

where

\(\bigoplus\)

denotes

the aggregation operation.

---

# <font color='orange'> 6. Node Update </font>

After aggregation,

the node updates

its representation.

$$
\boxed{
h_i^{(k+1)}
=
U
\left(
h_i^{(k)},
m_i
\right)
}
$$

where

* \(U\) is the update function,
* \(k\) denotes the message-passing iteration.

The updated representation

contains information

from

both

the node itself

and

its neighbours.

---

# <font color='orange'> 7. One Round of Message Passing </font>

Consider

```
A ----- B ----- C
```

Initially

```
A knows A

B knows B

C knows C
```

After

one iteration

```
A knows A+B

B knows A+B+C

C knows B+C
```

Information has propagated

one hop

through the graph.

---

# <font color='orange'> 8. Multiple Rounds </font>

Suppose

we repeat

message passing.

After

two iterations

```
A knows A+B+C

B knows A+B+C

C knows A+B+C
```

Information has travelled

two hops.

In general,

after

\(k\)

message-passing steps,

each node

contains information

from nodes

within

\(k\) graph hops.

---

# <font color='orange'> 9. General Message Passing Equation </font>

Most GNNs

can be expressed

using the message passing framework.

For iteration

\(k\),

$$
\boxed{
m_i^{(k)}
=
\bigoplus_{j\in\mathcal N(i)}
M
\left(
h_i^{(k)},
h_j^{(k)},
e_{ij}
\right)
}
$$

followed by

$$
\boxed{
h_i^{(k+1)}
=
U
\left(
h_i^{(k)},
m_i^{(k)}
\right).
}
$$

Different GNN architectures

mainly differ

in

their choices

of

* message function,
* aggregation function,
* update function.

---

Conceptually,

one message-passing layer

resembles

```python
for node in graph:

    messages = aggregate(

        neighbors(node)

    )

    node = update(

        node,

        messages

    )
```

# <font color='orange'> 12. Advantages </font>

Message passing provides

* Naturally incorporates neighbourhood information.
* Works on arbitrary graph structures.

# <font color='orange'> 13. Limitations </font>

Repeated message passing

can introduce

new challenges.

* Oversmoothing: node representations become increasingly similar.
* Oversquashing: too much information is compressed into fixed-size vectors.
* Computational cost on very large graphs.
* Long-range dependencies may require many message-passing iterations.

These limitations motivated

attention-based GNNs

and

Graph Transformers.

# <font color='orange'> 14. Common Misconceptions </font>

### Message Passing Means Sending Raw Features

False.

Nodes typically exchange learned representations rather than raw input features.

---

### Every Neighbour Contributes Equally

Not always.

Basic GCNs often weight neighbours similarly (after normalization), but Graph Attention Networks learn different importance weights for different neighbours.

---

### One Message-Passing Layer Sees the Entire Graph

False.

One layer aggregates information only from immediate neighbours. Multiple layers are required to capture information from more distant nodes.

---

### Message Passing Is Unique to One GNN Architecture

False.

Message passing is a general framework. Different GNN architectures implement it in different ways.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Message | Information sent from one node to another |
| Aggregation | Combines messages from neighbouring nodes |
| Update | Computes the node's new representation |
| One Iteration | Information propagates one graph hop |
| Multiple Iterations | Information propagates across larger neighbourhoods |
| Permutation Invariance | Aggregation is independent of neighbour ordering |
| Main Advantage | Enables learning from graph structure |
| Main Limitation | Deep message passing can lead to oversmoothing and oversquashing |

> **Key Insight:** Message passing is the fundamental mechanism underlying Graph Neural Networks. Each node repeatedly exchanges information with its neighbours, aggregates the received messages, and updates its own representation. Through successive iterations, information propagates across the graph, allowing every node to incorporate both local and increasingly global structural information. Most modern GNN architectures differ primarily in how they implement the message, aggregation, and update functions within this general framework.
