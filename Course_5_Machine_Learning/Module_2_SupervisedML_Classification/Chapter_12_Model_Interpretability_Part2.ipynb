# <font color='blue'> Model-Agnostic Explanations </font>

Many modern machine learning algorithms, such as Random Forests, Gradient Boosting, XGBoost, CatBoost, and Deep Neural Networks, achieve outstanding predictive performance but possess highly complex internal structures that are difficult—or impossible—for humans to interpret directly.

Rather than attempting to understand the internal mathematics of these models, **model-agnostic explanation methods** treat the trained model as a **black box**. They seek to explain a model's behaviour solely by observing how its predictions change in response to changes in the input data.

This philosophy enables explanations to be generated for virtually any machine learning model, regardless of its internal architecture.

---

## <font color='orange'> 1. What Does "Model-Agnostic" Mean? </font>

A **model-agnostic** method is independent of the machine learning algorithm.

It does **not** require access to

* model parameters,
* coefficients,
* tree structures,
* gradients,
* internal weights.

Instead,

the model is viewed as a function

$$
\boxed{
f:\mathbf{x}\rightarrow\hat{y},
}
$$

where

* $\mathbf{x}$ represents the input features,
* $\hat{y}$ is the prediction.

Only the relationship between inputs and outputs is used to generate explanations.

---

## <font color='orange'> 2. The Black-Box Perspective </font>

Imagine a sealed box.

```
Features

↓

┌──────────────┐
│              │
│ Black-Box ML │
│    Model     │
│              │
└──────────────┘

↓

Prediction
```

The internal workings are hidden.

However,

we are free to

* modify the inputs,
* observe the outputs,
* infer how the model behaves.

This is precisely the strategy adopted by model-agnostic explanation methods.

---

## <font color='orange'> 3. Why Are Model-Agnostic Methods Needed? </font>

Suppose we train

```
CatBoost
```

or

```
XGBoost
```

with

```
500 Trees
```

Even though every tree can be inspected individually,

understanding the combined behaviour of hundreds of interacting trees is practically impossible.

Rather than analysing every tree,

model-agnostic methods investigate

how predictions respond

to changes in the input variables.

---

## <font color='orange'> 4. Local and Global Explanations </font>

Model-agnostic techniques generally fall into two categories.

### Global Explanations

Describe

how the model behaves

overall.

Examples include

* Feature Importance
* Partial Dependence Plots
* Global Surrogate Models

---

### Local Explanations

Explain

a single prediction.

Examples include

* LIME
* SHAP
* Local Surrogate Models

Both perspectives provide complementary insights into model behaviour.

---

## <font color='orange'> 5. Perturbation-Based Explanations </font>

Many model-agnostic methods operate by

**perturbing** the input data.

For example,

suppose we slightly increase

```
Age
```

while keeping all other variables fixed.

If the prediction changes substantially,

Age is likely to be an influential feature.

This simple principle underlies several explainability techniques.

---

## <font color='orange'> 6. Measuring Feature Importance </font>

One common question is

> **Which variables are most important?**

Model-agnostic methods answer this by measuring

how much predictive performance changes

when a feature is modified or removed.

If removing

```
Transaction Amount
```

causes prediction accuracy to decrease dramatically,

the feature is likely to be important.

This idea forms the basis of

**Permutation Feature Importance**.

---

## <font color='orange'> 7. Visualizing Feature Effects </font>

Beyond ranking variables,

we often wish to understand

how

a variable influences predictions.

For example,

```
Income

↓

Loan Approval Probability
```

Model-agnostic visualization techniques include

* Partial Dependence Plots (PDP),
* Individual Conditional Expectation (ICE),
* Accumulated Local Effects (ALE).

These methods reveal how predictions change as feature values vary.

---

## <font color='orange'> 8. Explaining Individual Predictions </font>

Global explanations answer

```
How does the model behave overall?
```

However,

many applications require understanding

```
Why was THIS prediction made?
```

For example,

```
Fraud Probability

=

97%
```

Why?

Local explanation methods identify

which features contributed

to this specific prediction.

This is the purpose of

LIME

and

SHAP.

---

## <font color='orange'> 9. Surrogate Models </font>

Suppose a Deep Neural Network is too complex to interpret directly.

Instead,

we train

a simple Decision Tree

to mimic its predictions.

Although the Decision Tree is only an approximation,

it provides a useful explanation of the original model.

This concept is known as a

**Surrogate Model.**

Surrogate models can be

* Global
* Local

depending on the region of the feature space being approximated.

---

## <font color='orange'> 10. Common Model-Agnostic Techniques </font>

Several explanation methods have become standard tools in Explainable AI.

| Method | Purpose |
| :--- | :--- |
| Permutation Feature Importance | Rank important features |
| Partial Dependence Plot (PDP) | Visualize average feature effects |
| Individual Conditional Expectation (ICE) | Visualize individual prediction behaviour |
| Accumulated Local Effects (ALE) | Alternative to PDP for correlated features |
| LIME | Local explanations using surrogate models |
| SHAP | Feature contributions based on game theory |

Each method provides a different perspective on model behaviour.

---

## <font color='orange'> 11. Advantages </font>

Model-agnostic methods offer several important benefits.

* Applicable to virtually any predictive model.
* No access to model internals required.
* Support both global and local explanations.
* Increase transparency.
* Improve trust.
* Useful for debugging complex models.

These techniques have become indispensable in modern Explainable AI.

---

## <font color='orange'> 12. Limitations </font>

Despite their flexibility,

model-agnostic methods also have limitations.

* Explanations are approximations.
* Computationally expensive.
* Different methods may produce different explanations.
* Some methods assume feature independence.
* Local explanations may not reflect global behaviour.

Consequently,

interpretations should always be considered alongside traditional model evaluation.

---

## <font color='orange'> 13. Practical Workflow </font>

A typical explainability workflow is

```
Train Black-Box Model

↓

Evaluate Predictive Performance

↓

Choose Explanation Scope

↓

Global

↓

Permutation Importance

PDP

ALE

```

or

```
Local

↓

LIME

SHAP

Local Surrogate
```

↓

Interpret Results

↓

Communicate Findings

The choice of explanation method depends on whether the objective is to understand the model as a whole or to explain individual predictions.

---

Model-agnostic explanation techniques are widely used in

* Healthcare

These methods help experts understand complex machine learning systems while maintaining their predictive performance.

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Model-Agnostic | Independent of the underlying algorithm |
| Black-Box Perspective | Explain predictions using only inputs and outputs |

> **Key Insight:** Model-agnostic explanation methods provide a powerful framework for understanding complex machine learning models without requiring access to their internal structure. By treating the model as a black box and analysing how predictions respond to carefully designed changes in the input data, these techniques can explain both overall model behaviour and individual predictions. Their flexibility allows the same explanation methods to be applied across a wide range of algorithms, from Decision Trees and Random Forests to Gradient Boosting, CatBoost, and Deep Neural Networks, making them a cornerstone of modern Explainable AI (XAI).

# <font color='blue'> Surrogate Models </font>

Many modern machine learning algorithms, such as Random Forests, XGBoost, CatBoost, and Deep Neural Networks, often achieve outstanding predictive performance but are difficult to interpret because of their complex internal structures. One practical solution is to replace the complex model—not permanently, but only for interpretation—with a simpler model that approximates its behaviour.

This simpler model is known as a **Surrogate Model**.

A surrogate model is trained **not on the original target values**, but on the **predictions of the black-box model**. If the surrogate accurately reproduces the behaviour of the original model, it can be used to understand how the original model makes decisions.

Surrogate models therefore provide an interpretable approximation of an otherwise opaque predictive system.

---

## <font color='orange'> 1. What is a Surrogate Model? </font>

Suppose we have trained a highly accurate

```
XGBoost Classifier
```

Although its predictions are excellent,

its internal reasoning is difficult to understand.

Instead of explaining the XGBoost model directly,

we train a simple model such as

```
Decision Tree

or

Linear Regression
```

using

* the original input variables,

and

* the predictions produced by the XGBoost model.

The simple model then serves as an explanation of the complex model.

---

## <font color='orange'> 2. General Workflow </font>

The overall process can be summarized as

```
Training Dataset

↓

Black-Box Model

↓

Predictions

↓

Train Interpretable Model

↓

Surrogate Model

↓

Interpret Predictions
```

Notice that

the surrogate never learns

the true labels.

Instead,

it learns to imitate

the original model.

---

## <font color='orange'> 3. Why Train on Predictions? </font>

Suppose the original dataset contains

$$
(\mathbf{x},y).
$$

The black-box model learns

$$
f(\mathbf{x}).
$$

The surrogate is trained using

$$
(\mathbf{x},f(\mathbf{x})).
$$

Thus,

its objective becomes

$$
\boxed{
g(\mathbf{x})
\approx
f(\mathbf{x}),
}
$$

where

* $f$ is the complex model,
* $g$ is the surrogate.

The surrogate approximates

the model,

not the data.

This distinction is fundamental.

---

## <font color='orange'> 4. Global Surrogate Models </font>

A **Global Surrogate Model** attempts to explain

the behaviour of the

entire machine learning model.

For example,

```
Random Forest

↓

Decision Tree Surrogate
```

The Decision Tree approximates

the global decision boundaries

learned by the Random Forest.

Global surrogates answer questions such as

* Which variables are most important?
* How does the model generally behave?
* Which decision rules dominate?

---

## <font color='orange'> 5. Local Surrogate Models </font>

Sometimes,

we are interested in

only one prediction.

For example,

```
Fraud Probability

=

97%
```

Rather than approximating

the entire model,

we approximate

only the small neighbourhood

around that observation.

This is called

a **Local Surrogate Model.**

LIME is the most well-known example.

---

## <font color='orange'> 6. Choosing the Surrogate Model </font>

Several interpretable algorithms can be used as surrogates.

### Decision Trees

Simple decision rules.

Easy visualization.

---

### Linear Regression

Suitable for approximately linear relationships.

Coefficients provide direct interpretation.

---

### Logistic Regression

Useful for classification.

Coefficients indicate feature influence.

---

### Rule-Based Models

Generate explicit

IF–THEN

decision rules.

The choice depends upon

the complexity of the original model

and the desired level of interpretability.

---

## <font color='orange'> 7. Fidelity </font>

A surrogate model is useful only if it closely approximates

the original model.

This property is called

**Fidelity.**

Fidelity measures

how well

the surrogate reproduces

the predictions of the black-box model.

High fidelity

↓

Good explanation.

Low fidelity

↓

Poor explanation.

Importantly,

high fidelity does **not** imply that the original model is correct—it only indicates that the surrogate accurately mimics the original model's predictions.

---

## <font color='orange'> 8. Measuring Fidelity </font>

Suppose

```
Random Forest

↓

Prediction

```

and

```
Decision Tree Surrogate

↓

Prediction
```

If both models agree

for

```
98%

of observations,
```

the surrogate possesses

high fidelity.

Common fidelity metrics include

* Accuracy
* F1 Score
* R² (for regression)
* Mean Squared Error

The evaluation is always performed

between

the surrogate predictions

and

the original model predictions.

---

## <font color='orange'> 9. Advantages of Surrogate Models </font>

Surrogate models provide several important benefits.

* Easy interpretation.
* Applicable to any machine learning algorithm.
* Global or local explanations.
* Visual decision rules.
* Low computational cost after training.
* Excellent communication tool.

These characteristics make surrogate models particularly useful when explaining predictive systems to non-technical audiences.

---

## <font color='orange'> 10. Limitations </font>

Surrogate models also possess several limitations.

* They are approximations.
* Important interactions may be omitted.
* Different surrogate models may produce different explanations.
* High interpretability often reduces fidelity.
* Local surrogates cannot explain global behaviour.

Consequently,

surrogate explanations should always be interpreted cautiously.

---

## <font color='orange'> 11. Interpretability vs Fidelity </font>

One of the central challenges is balancing

interpretability

against

fidelity.

```
Simple Model

↓

High Interpretability

↓

Lower Fidelity
```

```
More Complex Surrogate

↓

Higher Fidelity

↓

Lower Interpretability
```

Choosing the appropriate balance depends upon

the application.

## <font color='orange'> 14. Relationship to LIME </font>

LIME may be viewed as

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |

| Global Surrogate | Explains the entire model |
| Local Surrogate | Explains one prediction |
| Fidelity | Agreement between surrogate and original model |
| Common Surrogates | Decision Trees, Linear Regression, Logistic Regression |
| Main Limitation | Approximation rather than exact explanation |

> **Key Insight:** Surrogate models improve the interpretability of complex machine learning systems by training an inherently interpretable model to reproduce the predictions of a black-box model. Rather than learning the true target values, the surrogate learns to approximate the behaviour of the original model, enabling humans to inspect decision rules, coefficients, or feature relationships that would otherwise remain hidden. The usefulness of a surrogate depends on its **fidelity**, which measures how accurately it mimics the original model. Global surrogates explain overall model behaviour, while local surrogates focus on individual predictions, forming the conceptual basis for explanation techniques such as LIME.

# <font color='blue'> Permutation Feature Importance (PFI) </font>

Permutation Feature Importance (PFI) is a **model-agnostic** technique for measuring the importance of individual predictor variables in a trained machine learning model. Unlike model-specific feature importance measures, such as regression coefficients or Decision Tree impurity reductions, PFI can be applied to virtually any supervised learning algorithm.

The central idea is remarkably simple:

> **If randomly shuffling the values of a feature causes the model's predictive performance to deteriorate substantially, then that feature must contain important predictive information.**

Conversely, if shuffling a feature has little or no effect on the model's performance, the model relies only weakly—or not at all—on that feature.

Because PFI measures the effect of each feature on the final predictive performance, it provides a direct and intuitive assessment of feature importance.

---

## <font color='orange'> 1. Motivation </font>

Suppose we have trained

```
Random Forest
```

to predict fraudulent credit card transactions.

We now ask

> **Which variables are actually driving the predictions?**

The model itself may contain

```
500 Decision Trees
```

making direct interpretation difficult.

Rather than inspecting every tree,

PFI investigates

how the model behaves

when individual features are deliberately corrupted.

---

## <font color='orange'> 2. The Central Idea </font>

Suppose one predictor is

```
Transaction Amount
```

Normally,

the model uses this feature when making predictions.

Now imagine

randomly shuffling

its values.

```
Original

↓

10

50

120

35

80
```

↓

Random Permutation

```
120

10

80

35

50
```

The relationship between

Transaction Amount

and

Fraud

has now been destroyed,

while all other variables remain unchanged.

If model performance decreases substantially,

Transaction Amount must have been important.

---

## <font color='orange'> 3. Why Does Permutation Work? </font>

Machine learning models learn relationships between

features

and

the target variable.

Randomly permuting one feature

breaks this relationship.

The feature now contains

random noise

instead of meaningful information.

If predictions become much worse,

the model clearly depended upon that feature.

---

## <font color='orange'> 4. The General Algorithm </font>

Permutation Feature Importance proceeds as follows.

### Step 1

Train the machine learning model.

---

### Step 2

Evaluate its original performance.

For example,

```
Accuracy = 98.5%
```

---

### Step 3

Select one feature.

For example,

```
Transaction Amount
```

---

### Step 4

Randomly shuffle

only

that feature.

All remaining features remain unchanged.

---

### Step 5

Predict again using

the modified dataset.

---

### Step 6

Measure the reduction in performance.

Repeat

for every feature.

The largest reduction indicates

the most important variable.

---

## <font color='orange'> 5. Mathematical Definition </font>

Suppose

the original model achieves

performance

$$
M.
$$

After permuting feature

$j$,

the performance becomes

$$
M_j^{perm}.
$$

The importance is

$$
\boxed{
I_j
=
M
-
M_j^{perm}.
}
$$

Large values indicate

important features.

Small values indicate

unimportant features.

---

## <font color='orange'> 6. Example </font>

Suppose

the original model obtains

```
ROC-AUC = 0.992
```

Now,

permute

```
Transaction Amount.
```

The ROC-AUC becomes

```
0.921
```

Importance

$$
=
0.992-0.921
=
0.071.
$$

Now,

permute

```
Transaction Time.
```

ROC-AUC becomes

```
0.990.
```

Importance

$$
=
0.002.
$$

Clearly,

Transaction Amount

is considerably more influential.

---

## <font color='orange'> 7. Interpreting Importance Values </font>

Suppose

the following importance scores are obtained.

| Feature | Importance |
| :--- | ---: |
| Transaction Amount | 0.081 |
| V14 | 0.074 |
| V12 | 0.063 |
| V10 | 0.051 |
| Time | 0.004 |

Interpretation:

* Transaction Amount strongly influences predictions.
* V14 and V12 are also highly informative.
* Time contributes very little.

Notice that

PFI ranks

features,

not observations.

---

## <font color='orange'> 8. Choice of Performance Metric </font>

PFI can be computed using

any evaluation metric.

Examples include

Classification

* Accuracy
* Balanced Accuracy
* F1 Score
* ROC-AUC
* Precision-Recall AUC

Regression

* R²
* RMSE
* MAE
* MSE

The chosen metric should reflect

the objective of the modelling task.

For imbalanced datasets,

Precision-Recall AUC

or

F1 Score

is often preferable to Accuracy.

---

## <font color='orange'> 9. Global Interpretation </font>

Permutation Feature Importance is a

**global explanation technique.**

It answers

> **Which variables are generally important to the model?**

It does **not** explain

why

one particular prediction

was made.

For local explanations,

methods such as

LIME

or

SHAP

are more appropriate.

---

## <font color='orange'> 10. Advantages </font>

PFI offers several important advantages.

* Model-agnostic.
* Easy to understand.
* Easy to implement.
* Applicable to any predictive model.
* Directly linked to predictive performance.
* Useful for feature selection.

---

## <font color='orange'> 11. Limitations </font>

PFI also possesses several limitations.

### Correlated Features

Suppose

```
Age

and

Years of Employment
```

are strongly correlated.

Permuting

Age

may have little effect,

because

Years of Employment

still contains similar information.

PFI may therefore underestimate

the importance of correlated variables.

---

### Computational Cost

Every feature requires

new predictions.

Large datasets therefore increase

the computational expense.

---

### Global Only

PFI explains

overall model behaviour,

not

individual predictions.

---

> **Key Insight:** Permutation Feature Importance evaluates the contribution of each feature by deliberately destroying the relationship between that feature and the target variable through random permutation. If the model's predictive performance deteriorates substantially, the feature is considered important because the model relied on the information it contained. Since this procedure depends only on the model's predictions rather than its internal structure, Permutation Feature Importance is a model-agnostic technique that can be applied to virtually any supervised learning algorithm. Although computationally more expensive than model-specific importance measures and potentially affected by correlated predictors, it provides an intuitive, reliable, and widely applicable assessment of global feature importance.
