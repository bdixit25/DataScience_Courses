# <font color='blue'> Introduction to Support Vector Machines (SVM) </font>

Support Vector Machines (SVMs) are a powerful class of supervised machine learning algorithms used primarily for **classification**, although they can also be adapted for regression through **Support Vector Regression (SVR)**.

Unlike algorithms such as K-Nearest Neighbours (KNN), which classify observations based on nearby training examples, or Logistic Regression, which models the probability of class membership, SVMs approach the classification problem from an entirely different perspective.

The fundamental objective of an SVM is not merely to separate different classes, but to find the **optimal decision boundary** that maximizes the separation between them. This principle, known as the **Maximum Margin Principle**, gives SVMs excellent generalization performance and makes them particularly effective for high-dimensional datasets.

---

## <font color='orange'> 1. The Classification Problem </font>

Suppose we are given a dataset consisting of observations belonging to two classes,

$$
y \in \{-1,+1\},
$$

where

* \(+1\) represents the positive class,
* \(-1\) represents the negative class.

Each observation is represented by a feature vector

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n).
$$

The objective is to construct a classifier that correctly assigns the class label of a new observation.

Mathematically,

we seek a function

$$
f(\mathbf{x})
$$

such that

$$
f(\mathbf{x})>0
\quad\Rightarrow\quad
\text{Class }+1,
$$

and

$$
f(\mathbf{x})<0
\quad\Rightarrow\quad
\text{Class }-1.
$$

The surface separating these two classes is called the **decision boundary**.

---

## <font color='orange'> 2. Linear Classification </font>

Suppose the two classes can be separated by a straight line (in two dimensions) or a plane (in three dimensions).

A linear decision boundary can be written as

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0
}
$$

where

* \(\mathbf{w}\) is the weight vector,
* \(b\) is the bias (or intercept).

This equation represents

* a line in two dimensions,
* a plane in three dimensions,
* a hyperplane in higher dimensions.

SVM attempts to determine the values of

$$
\mathbf{w}
\quad\text{and}\quad
b
$$

that produce the best possible separation between the classes.

---

## <font color='orange'> 3. What Makes SVM Different? </font>

Many classifiers can separate two classes.

Consider the figure below.

```
● ● ● ● ●

----------------

○ ○ ○ ○ ○
```

A straight line clearly separates the two classes.

However,

there are infinitely many possible separating lines.

For example,

```
----------------

```

or

```
-----------------------

```

or

```
---------------------------
```

All classify the training data correctly.

The question therefore becomes

> **Which separating boundary should we choose?**

This is precisely the problem that Support Vector Machines solve.

---

## <font color='orange'> 4. The Maximum Margin Principle </font>

Rather than selecting an arbitrary separating hyperplane,

SVM chooses the hyperplane that leaves the **largest possible distance** between the two classes.

This distance is called the

**Margin**.

Maximizing the margin improves the robustness of the classifier and reduces the likelihood of overfitting.

Intuitively,

if future observations contain small measurement errors or random noise,

a classifier with a larger margin is more likely to classify them correctly.

---

## <font color='orange'> 5. Why Maximizing the Margin Helps </font>

Suppose two classifiers correctly separate the training data.

Classifier A has a very narrow separation,

while Classifier B has a much wider separation.

If a new observation lies close to the decision boundary,

Classifier A may classify it incorrectly after only a tiny perturbation.

Classifier B,

having a much larger margin,

is considerably more tolerant of small variations in the data.

Consequently,

the Maximum Margin Classifier usually exhibits better **generalization** on unseen data.

---

## <font color='orange'> 6. Support Vectors </font>

Interestingly,

the optimal separating hyperplane is **not determined by every training observation**.

Instead,

it depends only on a small subset of observations lying closest to the decision boundary.

These observations are called

**Support Vectors**.

They "support" the position of the optimal hyperplane.

If all other observations were removed,

the decision boundary would remain essentially unchanged.

---

## <font color='orange'> 7. Linear vs Nonlinear Problems </font>

Some datasets can be separated using a straight line.

Others cannot.

For example,

```
○ ○ ○ ○

  ● ●

○ ○ ○ ○
```

No straight line can separate these classes.

Fortunately,

Support Vector Machines can handle both

* linear problems,
* nonlinear problems,

through the use of **kernel functions**, which transform the data into higher-dimensional feature spaces where linear separation becomes possible.

The Kernel Trick is one of the defining features of SVMs and will be discussed in a later section.

---

## <font color='orange'> 8. Advantages of SVM </font>

Support Vector Machines possess several important advantages.

* Effective in high-dimensional spaces.
* Excellent performance when the number of features exceeds the number of observations.
* Robust against overfitting through margin maximization.
* Flexible through the use of different kernel functions.
* Uses only the support vectors to define the classifier.

---

## <font color='orange'> 9. Limitations of SVM </font>

Despite their strengths,

SVMs also have several limitations.

* Training can be computationally expensive for very large datasets.
* Performance depends on selecting suitable hyperparameters.
* Sensitive to feature scaling.
* Less interpretable than linear regression or decision trees.
* Choice of kernel function can significantly affect performance.

---

## <font color='orange'> 10. Applications of Support Vector Machines </font>

Support Vector Machines have been successfully applied to many classification problems, including

* Credit Card Fraud Detection
* Face Recognition
* Text Classification
* Handwritten Digit Recognition
* Medical Diagnosis
* Image Classification
* Bioinformatics
* Remote Sensing
* Spam Detection

Their ability to perform well in high-dimensional spaces has made SVMs particularly popular in fields where datasets contain many input variables.

---

| Training Speed | Moderate to Slow |
| Prediction Speed | Fast |

> **Key Insight:** Support Vector Machines differ fundamentally from most classification algorithms. Rather than simply finding a decision boundary that separates the classes, SVM identifies the **optimal separating hyperplane** by maximizing the margin between the nearest observations of opposite classes. This Maximum Margin Principle gives SVMs excellent generalization performance and makes them particularly effective for high-dimensional classification problems. The classifier itself is determined only by a small number of critical observations known as **support vectors**, which define the location of the optimal decision boundary.

# <font color='blue'> Linear Decision Boundaries and Hyperplanes </font>

The objective of a Support Vector Machine (SVM) is to separate observations belonging to different classes by constructing an optimal **decision boundary**. In two-dimensional problems, this decision boundary is simply a straight line. In three dimensions, it becomes a plane. More generally, in higher-dimensional feature spaces, the decision boundary is called a **hyperplane**.

Although the term *hyperplane* may sound intimidating, it is simply the higher-dimensional generalization of an ordinary line or plane.

---

## <font color='orange'> 1. What is a Hyperplane? </font>

Suppose an observation is represented by

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n).
$$

A hyperplane is defined by the linear equation

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0
}
$$

where

* $\mathbf{x}$ is the feature vector,
* $\mathbf{w}$ is the weight vector,
* $b$ is the bias (intercept).

This equation divides the feature space into two regions.

Points satisfying

$$
\mathbf{w}^T\mathbf{x}+b>0
$$

are assigned to one class,

while points satisfying

$$
\mathbf{w}^T\mathbf{x}+b<0
$$

are assigned to the other.

---

## <font color='orange'> 2. From a Line to a Hyperplane </font>

The idea of a hyperplane becomes clearer by considering different dimensions.

### One Dimension

A single feature corresponds to a number line.

The decision boundary is simply a point.

```
---------|---------
```

---

### Two Dimensions

With two features,

the decision boundary becomes a straight line.

```
      ○ ○ ○

-------------------

● ● ● ● ●
```

---

### Three Dimensions

With three features,

the decision boundary becomes an ordinary plane.

---

### Higher Dimensions

When the number of features exceeds three,

the decision boundary becomes a

**hyperplane**.

Although it cannot be visualized directly,

its mathematical description remains identical.

---

## <font color='orange'> 3. Deriving the Hyperplane Equation </font>

Consider the equation of a straight line,

$$
ax+by+c=0.
$$

Introducing vector notation,

define

$$
\mathbf{x}
=
\begin{bmatrix}
x_1\\
x_2
\end{bmatrix},
\qquad
\mathbf{w}
=
\begin{bmatrix}
w_1\\
w_2
\end{bmatrix}.
$$

The dot product becomes

$$
\mathbf{w}^T\mathbf{x}
=
w_1x_1+w_2x_2.
$$

Hence,

the equation of the line can be written as

$$
w_1x_1+w_2x_2+b=0.
$$

This generalizes naturally to

$$
n
$$

dimensions,

giving

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0.
}
$$

Thus,

the familiar equation of a straight line is simply a special case of the hyperplane equation.

---

## <font color='orange'> 4. The Weight Vector </font>

The vector

$$
\mathbf{w}
=
(w_1,w_2,\ldots,w_n)
$$

contains the coefficients associated with each feature.

Unlike Linear Regression,

these coefficients are **not** interpreted as the influence of individual features on the predicted value.

Instead,

they determine the **orientation** of the hyperplane.

An important geometric property is that

> **The weight vector is always perpendicular (normal) to the hyperplane.**

Consequently,

changing

$$
\mathbf{w}
$$

rotates the decision boundary.

---

## <font color='orange'> 5. The Bias Term </font>

The parameter

$$
b
$$

is called the

* bias,
* intercept,
* offset.

Its purpose is to translate the hyperplane without changing its orientation.

For example,

consider

$$
x+y=0.
$$

Now change the equation to

$$
x+y-3=0.
$$

The line shifts,

but its slope remains unchanged.

Thus,

$$
\mathbf{w}
$$

controls rotation,

while

$$
b
$$

controls translation.

---

## <font color='orange'> 6. Which Side of the Hyperplane? </font>

Every observation lies on one of three possibilities.

### On the Hyperplane

If

$$
\mathbf{w}^T\mathbf{x}+b=0,
$$

the observation lies exactly on the decision boundary.

---

### Positive Side

If

$$
\mathbf{w}^T\mathbf{x}+b>0,
$$

the observation belongs to the positive half-space.

---

### Negative Side

If

$$
\mathbf{w}^T\mathbf{x}+b<0,
$$

the observation belongs to the negative half-space.

The sign of

$$
\mathbf{w}^T\mathbf{x}+b
$$

therefore determines the predicted class.

---

## <font color='orange'> 7. Classification Rule </font>

For binary classification,

the prediction is

$$
\boxed{
\hat y
=
\operatorname{sign}
\left(
\mathbf{w}^T\mathbf{x}+b
\right)
}
$$

where

$$
\operatorname{sign}(z)
=
\begin{cases}
+1,& z>0,\\
-1,& z<0.
\end{cases}
$$

Thus,

classification reduces to evaluating a simple linear function.

---

## <font color='orange'> 8. Geometric Interpretation </font>

Suppose the feature space contains two classes.

```
○ ○ ○ ○ ○

----------------

● ● ● ● ●
```

The hyperplane separates the feature space into two half-spaces.

Each new observation is assigned according to the side of the hyperplane on which it lies.

Notice that the classifier does **not** estimate probabilities.

Instead,

it simply determines which side of the decision boundary contains the observation.

---

## <font color='orange'> 9. Distance from the Hyperplane </font>

The signed perpendicular distance of an observation

$$
\mathbf{x}
$$

from the hyperplane is

$$
\boxed{
d
=
\frac{
\mathbf{w}^T\mathbf{x}+b
}
{\|\mathbf{w}\|}
}
$$

where

$$
\|\mathbf{w}\|
=
\sqrt{
w_1^2+w_2^2+\cdots+w_n^2
}
$$

is the Euclidean norm of the weight vector.

This quantity plays a central role in Support Vector Machines because the algorithm seeks to **maximize the minimum distance** between the training observations and the decision boundary.

This minimum distance is known as the **margin**, which will be derived in the next section.

---

## <font color='orange'> 10. Comparison with Logistic Regression </font>

Although Logistic Regression and a linear SVM both use equations of the form

$$
\mathbf{w}^T\mathbf{x}+b,
$$

their objectives differ fundamentally.

| Logistic Regression | Support Vector Machine |
| :--- | :--- |
| Estimates probabilities | Finds maximum-margin boundary |
| Uses sigmoid function | Uses hyperplane geometry |
| Optimizes log-loss | Optimizes margin |
| Outputs probabilities | Outputs class labels (or decision scores) |

Thus,

the mathematical expressions appear similar,

but the optimization objectives are entirely different.

---

## <font color='orange'> 11. Why Hyperplanes Matter </font>

The hyperplane forms the foundation of every Support Vector Machine.

Once the hyperplane has been defined,

the remaining questions become

* Which hyperplane should be chosen?
* How can the margin be maximized?
* Which observations determine the solution?

These questions lead naturally to the concepts of

* Maximum Margin Classifier,
* Support Vectors,
* Soft Margin SVM,

which are discussed in the following sections.

---

## <font color='purple'> 12. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Hyperplane | Generalization of a line or plane to higher dimensions |
| Equation | $\mathbf{w}^T\mathbf{x}+b=0$ |
| Weight Vector | Determines the orientation of the hyperplane |
| Bias | Translates the hyperplane |
| Classification Rule | $\hat y=\operatorname{sign}(\mathbf{w}^T\mathbf{x}+b)$ |
| Distance to Hyperplane | $\dfrac{\mathbf{w}^T\mathbf{x}+b}{\|\mathbf{w}\|}$ |

> **Key Insight:** A Support Vector Machine classifies observations by constructing a **hyperplane** that divides the feature space into two regions. The weight vector determines the orientation of this hyperplane, while the bias controls its position. Every prediction is obtained simply by evaluating the sign of the linear function $\mathbf{w}^T\mathbf{x}+b$. Unlike many classifiers, however, SVM does not stop at finding any separating hyperplane—it seeks the one that maximizes the distance between the two classes. This leads directly to the concept of the **maximum margin classifier**, which forms the theoretical foundation of Support Vector Machines.

# <font color='blue'> The Maximum Margin Classifier </font>

In the previous section, we saw that many different hyperplanes can correctly separate two linearly separable classes. Since every one of these hyperplanes classifies the training data perfectly, an important question naturally arises:

> **Which hyperplane should be chosen?**

Support Vector Machines answer this question by selecting the hyperplane that **maximizes the separation between the two classes**. This separation is known as the **margin**, and the resulting classifier is called the **Maximum Margin Classifier**.

The Maximum Margin Principle is the fundamental idea behind Support Vector Machines and is responsible for their excellent ability to generalize to unseen data.

---

## <font color='orange'> 1. Multiple Separating Hyperplanes </font>

Consider a simple two-class dataset.

```
○ ○ ○ ○ ○


------------------ (A)


------------      (B)


-------------------------- (C)


● ● ● ● ●
```

Hyperplanes A, B and C all classify the training observations correctly.

Each one produces

$$
100\%
$$

training accuracy.

However,

their distances from the observations are different.

Some decision boundaries lie extremely close to one of the classes,

making them highly sensitive to small perturbations.

Others leave a much larger separation.

The natural question is therefore:

> Which hyperplane is most likely to classify future observations correctly?

---

## <font color='orange'> 2. Definition of the Margin </font>

The **margin** is the minimum perpendicular distance between the decision boundary and any training observation.

Suppose the separating hyperplane is

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=0.
}
$$

For any observation

$$
\mathbf{x},
$$

the perpendicular distance to the hyperplane is

$$
\boxed{
d
=
\frac{
|\mathbf{w}^T\mathbf{x}+b|
}
{\|\mathbf{w}\|}
}
$$

where

$$
\|\mathbf{w}\|
=
\sqrt{
w_1^2+w_2^2+\cdots+w_n^2
}
$$

is the Euclidean norm of the weight vector.

The **margin** is defined as the smallest of these distances.

---

## <font color='orange'> 3. Why Maximize the Margin? </font>

Suppose two classifiers perfectly separate the training data.

### Small Margin

```
○ ○ ○

--------

● ● ●
```

A small measurement error may cause a new observation to cross the decision boundary.

The classifier therefore becomes sensitive to noise.

---

### Large Margin

```
○ ○ ○



------------



● ● ●
```

The wider separation provides greater tolerance to small perturbations.

Consequently,

the classifier is more likely to generalize well to unseen data.

Maximizing the margin therefore reduces the risk of overfitting.

---

## <font color='orange'> 4. Margin Boundaries </font>

Instead of considering only one hyperplane,

SVM introduces two additional parallel hyperplanes,

defined by

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=1
}
$$

and

$$
\boxed{
\mathbf{w}^T\mathbf{x}+b=-1.
}
$$

The observations nearest the decision boundary lie on these two planes.

The region between them is called the

**Margin**.

Graphically,

```
○ ○ ○ ○
-------------------  w·x+b=1

===================  Decision Boundary

-------------------  w·x+b=-1
● ● ● ●
```

The decision boundary lies exactly halfway between the two margin boundaries.

---

## <font color='orange'> 5. Deriving the Margin Width </font>

The distance between two parallel hyperplanes

$$
\mathbf{w}^T\mathbf{x}+b=1
$$

and

$$
\mathbf{w}^T\mathbf{x}+b=-1
$$

is obtained from the distance formula.

The result is

$$
\boxed{
\text{Margin}
=
\frac{2}{\|\mathbf{w}\|}
}
$$

This is one of the most important equations in Support Vector Machines.

Notice that

* large

$$
\|\mathbf{w}\|
$$

produces a small margin,

while

* small

$$
\|\mathbf{w}\|
$$

produces a large margin.

Thus,

maximizing the margin is equivalent to minimizing

$$
\|\mathbf{w}\|.
$$

---

## <font color='orange'> 6. The Classification Constraints </font>

Suppose each training observation has class label

$$
y_i
\in
\{-1,+1\}.
$$

Correct classification requires

$$
\boxed{
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge1.
}
$$

This single equation represents both classes simultaneously.

### Positive Class

If

$$
y_i=+1,
$$

then

$$
\mathbf{w}^T\mathbf{x}_i+b
\ge1.
$$

---

### Negative Class

If

$$
y_i=-1,
$$

then

$$
\mathbf{w}^T\mathbf{x}_i+b
\le-1.
$$

Thus,

all correctly classified observations lie outside the margin.

---

## <font color='orange'> 7. The Optimization Problem </font>

Since

$$
\text{Margin}
=
\frac2{\|\mathbf{w}\|},
$$

maximizing the margin is equivalent to minimizing

$$
\|\mathbf{w}\|.
$$

For mathematical convenience,

the squared norm is minimized instead,

giving the optimization problem

$$
\boxed{
\min_{\mathbf{w},b}
\;
\frac12
\|\mathbf{w}\|^2
}
$$

subject to

$$
\boxed{
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge1,
\qquad
i=1,\ldots,m.
}
$$

This is known as the

**Primal Optimization Problem**

for a hard-margin Support Vector Machine.

---

## <font color='orange'> 8. Why the Factor of \(\frac12\)? </font>

The optimization objective is written as

$$
\frac12
\|\mathbf{w}\|^2
$$

instead of

$$
\|\mathbf{w}\|^2.
$$

The factor

$$
\frac12
$$

has no effect on the solution.

It is included solely because the derivative becomes simpler.

Indeed,

$$
\frac{\partial}{\partial\mathbf{w}}
\left(
\frac12
\|\mathbf{w}\|^2
\right)
=
\mathbf{w},
$$

rather than

$$
2\mathbf{w}.
$$

This simplifies the optimization procedure considerably.

---

## <font color='orange'> 9. Why This Improves Generalization </font>

Statistical learning theory shows that,

all else being equal,

classifiers with larger margins tend to have lower model complexity and better generalization.

Intuitively,

the classifier becomes less sensitive to

* measurement errors,
* noisy observations,
* small perturbations.

Consequently,

maximum-margin classifiers often perform better on unseen data than classifiers that merely minimize training error.

---

## <font color='orange'> 10. Hard Margin vs Soft Margin </font>

The Maximum Margin Classifier described above assumes that the two classes are

**perfectly linearly separable**.

In practice,

real datasets often contain

* overlapping classes,
* noisy observations,
* outliers.

The hard-margin formulation therefore becomes too restrictive.

To overcome this limitation,

Support Vector Machines introduce

**Soft Margins**,

which allow some observations to violate the margin constraints.

This leads naturally to the introduction of **slack variables** and the regularization parameter

$$
C,
$$

which will be discussed in the next section.

---

## <font color='purple'> 11. Conceptual Summary </font>

| Concept | Formula |
| :--- | :--- |
| Decision Boundary | $\mathbf{w}^T\mathbf{x}+b=0$ |
| Upper Margin | $\mathbf{w}^T\mathbf{x}+b=1$ |
| Lower Margin | $\mathbf{w}^T\mathbf{x}+b=-1$ |
| Margin Width | $\dfrac{2}{\|\mathbf{w}\|}$ |
| Classification Constraint | $y_i(\mathbf{w}^T\mathbf{x}_i+b)\ge1$ |
| Optimization Objective | $\displaystyle\min \frac12\|\mathbf{w}\|^2$ |

> **Key Insight:** The central idea of Support Vector Machines is not merely to separate two classes, but to identify the separating hyperplane that **maximizes the margin** between them. A larger margin produces a classifier that is more robust to noise and better able to generalize to unseen data. This requirement transforms the classification problem into a constrained optimization problem, where the objective is to minimize the squared norm of the weight vector while ensuring that every training observation lies on the correct side of the margin. This elegant optimization framework forms the mathematical foundation of Support Vector Machines.

# <font color='blue'> Support Vectors </font>

In the previous section, we derived the Maximum Margin Classifier and showed that the optimal decision boundary is obtained by maximizing the distance between the two classes.

An important question naturally follows:

> **Which training observations determine the position of the optimal hyperplane?**

Surprisingly,

the answer is **only a small subset of the training data**.

These critical observations are known as **Support Vectors**, and they give the Support Vector Machine its name.

---

## <font color='orange'> 1. What Are Support Vectors? </font>

Support vectors are the training observations that lie **closest to the decision boundary**.

More precisely,

they are the observations that lie exactly on the margin boundaries

$$
\mathbf{w}^T\mathbf{x}+b=1
$$

and

$$
\mathbf{w}^T\mathbf{x}+b=-1.
$$

These observations determine the location and orientation of the optimal separating hyperplane.

---

## <font color='orange'> 2. Visual Interpretation </font>

Consider the following simplified dataset.

```
          ○      ○

     ○          ○

-----------------------------  Margin

=============================  Decision Boundary

-----------------------------  Margin

      ●      ●

   ●         ●
```

The observations touching the two margin boundaries are

the **support vectors**.

The remaining observations,

although correctly classified,

do **not** influence the position of the hyperplane.

---

## <font color='orange'> 3. Why Are They Called "Support" Vectors? </font>

Imagine stretching a rope tightly between two walls.

The rope is held in place only by the objects that it touches.

Removing objects far from the rope has no effect.

Support vectors play exactly the same role.

They "support" the optimal hyperplane by preventing it from moving any further toward either class.

Hence the name

**Support Vector Machine**.

---

## <font color='orange'> 4. Mathematical Definition </font>

Recall the classification constraint

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge1.
$$

For most training observations,

the inequality is **strict**,

meaning

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
>
1.
$$

These observations lie outside the margin.

---

Support vectors satisfy the equality

$$
\boxed{
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
=
1.
}
$$

Thus,

they lie exactly on the margin boundary.

---

## <font color='orange'> 5. Three Types of Observations </font>

Every observation belongs to one of three categories.

### Outside the Margin

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
>
1.
$$

These observations are correctly classified and lie safely away from the decision boundary.

They are **not support vectors**.

---

### On the Margin

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
=
1.
$$

These observations define the margin.

They **are support vectors**.

---

### Inside the Margin

For a hard-margin SVM,

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
<
1
$$

is not permitted.

For a soft-margin SVM,

these observations are allowed and will also become support vectors, as discussed in the next chapter.

---

## <font color='orange'> 6. Why Most Training Observations Do Not Matter </font>

One of the most remarkable properties of SVMs is that

only the support vectors influence the solution.

Suppose a training observation lies far from the decision boundary.

Moving that observation slightly,

or even removing it,

usually does not change the optimal hyperplane.

Only when a support vector moves does the decision boundary change.

This makes SVM remarkably robust.

---

## <font color='orange'> 7. Geometric Interpretation </font>

Suppose two classes are perfectly separable.

The support vectors are simply the observations with the smallest perpendicular distance to the decision boundary.

Every other observation lies farther away.

The optimal hyperplane therefore "rests" against the support vectors.

If the support vectors move,

the hyperplane moves.

If distant observations move,

the hyperplane usually remains unchanged.

---

## <font color='orange'> 8. Relationship with the Margin </font>

Recall that the margin width is

$$
\boxed{
\text{Margin}
=
\frac{2}{\|\mathbf{w}\|}.
}
$$

The support vectors lie exactly at the boundaries of this margin.

They therefore determine

* the width of the margin,
* the orientation of the hyperplane,
* the final classifier.

Without the support vectors,

the concept of a maximum margin would not exist.

---

## <font color='orange'> 9. Support Vectors in the Optimization Problem </font>

When solving the optimization problem,

most observations contribute nothing to the final solution.

This becomes clear in the dual formulation of SVM,

where the weight vector can be written as

$$
\boxed{
\mathbf{w}
=
\sum_{i=1}^{m}
\alpha_i
y_i
\mathbf{x}_i
}
$$

where

$$
\alpha_i
$$

are the **Lagrange multipliers**.

For almost every observation,

$$
\alpha_i=0.
$$

Only the support vectors have

$$
\alpha_i>0.
$$

Consequently,

only support vectors contribute to the weight vector.

Although the full derivation of the dual problem is beyond the scope of this notebook, this result explains why the classifier depends only on the support vectors.

---

## <font color='orange'> 10. Practical Consequences </font>

Because only a small fraction of the training observations are support vectors,

SVMs often require much less information than one might expect.

For example,

a dataset containing

50,000

observations may produce only

300

support vectors.

The classifier is therefore determined by

less than

1%

of the training data.

---

## <font color='orange'> 11. Support Vectors in Scikit-Learn </font>

After fitting an SVM,

the support vectors can be inspected directly.

```python
model.support_vectors_
```

returns the feature values of every support vector.

The corresponding indices in the original dataset are

```python
model.support_
```

The total number of support vectors is

```python
len(model.support_)
```

These attributes are available for classifiers based on `SVC`.

---

## <font color='orange'> 12. Advantages of Depending Only on Support Vectors </font>

This property provides several important benefits.

* Robust to redundant training observations.
* Naturally ignores observations far from the decision boundary.
* Produces stable classifiers.
* Often improves generalization performance.
* Efficient prediction after training.

---

## <font color='orange'> 13. Limitations </font>

Although only support vectors determine the classifier,

identifying them requires solving the optimization problem over the **entire** training dataset.

Consequently,

training can become computationally expensive for very large datasets,

even though prediction is relatively fast.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Support Vector | Observation lying on the margin |
| Mathematical Condition | $y_i(\mathbf{w}^T\mathbf{x}_i+b)=1$ |
| Determines Hyperplane | Yes |
| Determines Margin | Yes |
| Distant Observations | Usually have no influence |
| Scikit-Learn | `support_vectors_` |

> **Key Insight:** The defining characteristic of a Support Vector Machine is that the optimal classifier is determined **only by the support vectors**—the observations lying closest to the decision boundary. These points define the margin and fix the position of the optimal hyperplane, while the vast majority of the training data have no direct influence on the final solution. This elegant property contributes to the robustness and strong generalization performance of SVMs, especially in high-dimensional classification problems.

# <font color='blue'> Soft-Margin Support Vector Machines and the Regularization Parameter \(C\) </font>

The Maximum Margin Classifier introduced in the previous sections assumes that every training observation can be separated perfectly by a single hyperplane. Such a classifier is known as a **Hard-Margin Support Vector Machine**.

Although mathematically elegant, this assumption is rarely satisfied in practice. Real-world datasets often contain

* measurement errors,
* noisy observations,
* overlapping classes,
* mislabeled samples.

Under these conditions, perfect separation may be impossible or may produce a classifier that overfits the training data.

To overcome these limitations, Support Vector Machines introduce the concept of a **Soft Margin**, which allows certain observations to violate the margin constraints while still seeking a classifier with good generalization performance.

---

## <font color='orange'> 1. Why Hard Margins Fail </font>

Consider the following dataset.

```
○ ○ ○ ○ ○

      ○

-------------------------

● ● ● ● ●
```

Suppose one observation from the positive class appears within the negative class due to measurement error or incorrect labeling.

A hard-margin SVM attempts to classify **every** observation correctly.

In this situation,

either

* no separating hyperplane exists,

or

* the classifier becomes extremely complicated in order to classify every observation correctly.

Both situations are undesirable.

---

## <font color='orange'> 2. The Idea Behind Soft Margins </font>

Instead of insisting that every observation lies outside the margin,

Soft-Margin SVM allows certain observations to

* lie inside the margin,
* lie on the wrong side of the decision boundary,
* even be misclassified.

The objective is no longer

> "Classify every training observation perfectly."

Instead,

the objective becomes

> "Find the hyperplane that achieves the best compromise between a large margin and a small number of classification errors."

---

## <font color='orange'> 3. Introducing Slack Variables </font>

To allow violations of the margin,

a new variable

$$
\xi_i
$$

(called the **slack variable**) is introduced for every training observation.

The classification constraint becomes

$$
\boxed{
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge
1-\xi_i
}
$$

where

$$
\xi_i
\ge0.
$$

The slack variable measures how much an observation violates the margin.

---

## <font color='orange'> 4. Interpretation of Slack Variables </font>

Different values of

$$
\xi_i
$$

have different meanings.

### No Violation

If

$$
\xi_i=0,
$$

then

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge1.
$$

The observation lies outside the margin and is correctly classified.

---

### Inside the Margin

If

$$
0<\xi_i<1,
$$

the observation lies inside the margin,

but remains correctly classified.

---

### On the Decision Boundary

If

$$
\xi_i=1,
$$

the observation lies exactly on the separating hyperplane.

---

### Misclassified Observation

If

$$
\xi_i>1,
$$

the observation lies on the wrong side of the decision boundary.

It is therefore misclassified.

---

## <font color='orange'> 5. Geometric Interpretation </font>

A soft-margin SVM permits observations to appear within the margin.

```
○ ○ ○ ○

---- Margin ----

    ○

===== Hyperplane =====

---- Margin ----

● ● ● ●
```

The observation inside the margin is assigned a positive slack variable.

Although the classifier no longer perfectly separates the training data,

it often performs much better on unseen observations.

---

## <font color='orange'> 6. The New Optimization Problem </font>

Recall that the hard-margin SVM minimizes

$$
\frac12
\|\mathbf{w}\|^2.
$$

Soft-margin SVM introduces an additional penalty for every margin violation.

The optimization problem becomes

$$
\boxed{
\min_{\mathbf{w},b,\boldsymbol{\xi}}
\;
\frac12
\|\mathbf{w}\|^2
+
C
\sum_{i=1}^{m}
\xi_i
}
$$

subject to

$$
y_i
(\mathbf{w}^T\mathbf{x}_i+b)
\ge
1-\xi_i,
$$

and

$$
\xi_i
\ge0.
$$

---

## <font color='orange'> 7. The Meaning of the Regularization Parameter \(C\) </font>

The parameter

$$
C
$$

controls the relative importance of

* maximizing the margin,
* minimizing classification errors.

It determines how severely violations of the margin are penalized.

---

### Large \(C\)

If

$$
C
$$

is very large,

classification errors become extremely expensive.

The optimization algorithm therefore attempts to classify almost every training observation correctly,

even if this results in a narrow margin.

Characteristics include

* lower training error,
* smaller margin,
* greater risk of overfitting.

---

### Small \(C\)

If

$$
C
$$

is small,

margin violations receive only a small penalty.

The optimization therefore prefers a wider margin,

even if several observations are misclassified.

Characteristics include

* larger margin,
* improved robustness,
* greater regularization,
* lower risk of overfitting.

---

## <font color='orange'> 8. Bias-Variance Trade-off </font>

The regularization parameter

$$
C
$$

controls the familiar Bias-Variance Trade-off.

| Small \(C\) | Large \(C\) |
| :---: | :---: |
| Large Margin | Narrow Margin |
| High Bias | Low Bias |
| Low Variance | High Variance |
| Better Generalization | Greater Risk of Overfitting |

Thus,

\(C\) plays a role similar to the regularization strength in Logistic Regression.

---

## <font color='orange'> 9. Hard Margin as a Special Case </font>

Suppose

$$
C
\rightarrow
\infty.
$$

Violating the margin becomes infinitely expensive.

The optimization therefore forces

$$
\xi_i=0
$$

for every observation,

recovering the hard-margin SVM.

Thus,

the hard-margin classifier is simply a limiting case of the soft-margin formulation.

---

## <font color='orange'> 10. Choosing \(C\) in Practice </font>

The optimal value of

$$
C
$$

depends on the dataset.

Typical candidate values include

$$
0.01,\;
0.1,\;
1,\;
10,\;
100.
$$

Rather than choosing

$$
C
$$

manually,

Cross-Validation is typically used.

For example,

Scikit-Learn provides

```python
GridSearchCV()
```

or

```python
RandomizedSearchCV()
```

to determine the value of

$$
C
$$

that yields the best validation performance.

---

## <font color='orange'> 11. Soft Margins and Support Vectors </font>

In a soft-margin SVM,

support vectors include

* observations lying on the margin,
* observations inside the margin,
* misclassified observations.

These observations have non-zero slack variables and directly influence the optimization.

Observations located far from the decision boundary continue to have little or no effect on the final classifier.

---

## <font color='orange'> 12. Practical Considerations </font>

Some useful guidelines include

* Begin with

```python
C=1
```

(the Scikit-Learn default).

* Use Cross-Validation to optimize

$$
C.
$$

* Large datasets with substantial noise often benefit from smaller values of

$$
C.
$$

* Nearly separable datasets may benefit from larger values of

$$
C.
$$

---

## <font color='orange'> 13. Relationship with Other Regularization Methods </font>

The parameter

$$
C
$$

serves the same purpose as regularization parameters used elsewhere in machine learning.

| Algorithm | Regularization Parameter |
| :--- | :--- |
| Ridge Regression | $\lambda$ |
| LASSO Regression | $\lambda$ |
| Elastic Net | $\lambda$ |
| Logistic Regression | $\lambda$ (or inverse \(C\) in Scikit-Learn) |
| Support Vector Machine | \(C\) |

Although the notation differs,

all regularization parameters balance model complexity against training error.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Hard Margin | Requires perfect separation |
| Soft Margin | Allows margin violations |
| Slack Variable | $\xi_i$ measures margin violation |
| Constraint | $y_i(\mathbf{w}^T\mathbf{x}_i+b)\ge1-\xi_i$ |
| Objective | $\frac12\|\mathbf{w}\|^2+C\sum\xi_i$ |
| Large \(C\) | Narrow margin, fewer errors, greater overfitting risk |
| Small \(C\) | Wide margin, more errors, better generalization |

> **Key Insight:** Real-world datasets are rarely perfectly separable. Soft-Margin Support Vector Machines overcome this limitation by introducing **slack variables**, which permit observations to violate the margin when necessary. The regularization parameter \(C\) controls the trade-off between maximizing the margin and minimizing classification errors. A large value of \(C\) prioritizes fitting the training data closely, whereas a small value encourages a wider margin and improved generalization. This balance between margin width and training error is one of the principal reasons why SVMs perform well on a wide variety of practical classification problems.

# <font color='blue'> The Kernel Trick </font>

The Support Vector Machine developed in the previous sections assumes that the two classes can be separated by a single hyperplane. Such classifiers are known as **Linear Support Vector Machines**.

Although many datasets are approximately linearly separable, numerous real-world classification problems exhibit highly nonlinear decision boundaries. Examples include

* concentric circles,
* spiral datasets,
* XOR classification,
* complex image recognition tasks.

A linear hyperplane cannot correctly separate these datasets.

The **Kernel Trick** provides an elegant solution by allowing SVMs to construct nonlinear decision boundaries **without explicitly transforming the data into higher-dimensional spaces**.

---

## <font color='orange'> 1. Why Linear SVMs Fail </font>

Consider the following dataset.

```
        ○ ○ ○

     ○       ○

        ● ●

     ○       ○

        ○ ○ ○
```

The positive class lies in the centre,

while the negative class surrounds it.

No straight line can separate these classes.

No matter how the line is rotated,

some observations will always be misclassified.

This is an example of a **nonlinearly separable dataset**.

---

## <font color='orange'> 2. A Simple Idea </font>

Suppose we could somehow transform the dataset into a different coordinate system.

Instead of describing each observation using

$$
(x_1,x_2),
$$

we introduce a new feature,

for example,

$$
z=x_1^2+x_2^2.
$$

The observations are now represented by

$$
(x_1,x_2,z).
$$

Although the original dataset was not linearly separable,

the transformed dataset may now be separated by a plane.

Thus,

a nonlinear problem in two dimensions can become a linear problem in three dimensions.

---

## <font color='orange'> 3. Feature Mapping </font>

More generally,

suppose each observation is transformed using a mapping

$$
\boxed{
\phi(\mathbf{x})
}
$$

which converts the original feature vector

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)
$$

into a higher-dimensional vector

$$
\phi(\mathbf{x}).
$$

The classifier then becomes

$$
\boxed{
\mathbf{w}^T\phi(\mathbf{x})+b=0.
}
$$

Although this equation appears identical to the linear SVM,

the transformation

$$
\phi
$$

allows the decision boundary in the original space to become highly nonlinear.

---

## <font color='orange'> 4. The Computational Problem </font>

The idea of feature mapping is powerful,

but it introduces a serious computational difficulty.

Suppose the original dataset contains

100

features.

A polynomial transformation of degree

4

may generate

thousands

or even

millions

of new features.

Training an SVM directly in this enormous feature space quickly becomes computationally impractical.

Fortunately,

Support Vector Machines avoid this problem entirely.

---

## <font color='orange'> 5. The Key Observation </font>

When solving the SVM optimization problem,

the transformed feature vectors

$$
\phi(\mathbf{x})
$$

never appear individually.

Instead,

they always appear inside an inner (dot) product,

$$
\phi(\mathbf{x}_i)^T
\phi(\mathbf{x}_j).
$$

This observation is crucial.

If we could compute this dot product **without explicitly computing**

$$
\phi(\mathbf{x}),
$$

then the expensive feature transformation could be avoided altogether.

---

## <font color='orange'> 6. The Kernel Function </font>

A **kernel function** is defined as

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\phi(\mathbf{x})^T
\phi(\mathbf{z})
}
$$

where

* $\mathbf{x}$ and $\mathbf{z}$ are two observations,
* $\phi$ represents the feature mapping.

The remarkable property of a kernel function is that it computes the inner product in the transformed feature space **without explicitly performing the transformation**.

This is known as the

**Kernel Trick**.

---

## <font color='orange'> 7. Why It Is Called a "Trick" </font>

Suppose the feature transformation generates

10,000

new variables.

Instead of calculating every transformed feature,

the kernel function directly computes

$$
K(\mathbf{x},\mathbf{z})
$$

using only the original variables.

Thus,

the algorithm behaves **as though** it were operating in a high-dimensional feature space,

while actually performing only a relatively simple calculation.

This remarkable shortcut is the Kernel Trick.

---

## <font color='orange'> 8. Geometric Interpretation </font>

The kernel function allows the classifier to construct a linear hyperplane in a higher-dimensional feature space.

When projected back into the original feature space,

this hyperplane becomes a nonlinear decision boundary.

For example,

a straight plane in three dimensions may appear as a circle in two dimensions.

Consequently,

the classifier can learn highly complex boundaries while still solving a linear optimization problem.

---

## <font color='orange'> 9. Common Kernel Functions </font>

Several kernel functions are widely used.

### Linear Kernel

Suitable for approximately linear datasets.

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\mathbf{x}^T\mathbf{z}
}
$$

---

### Polynomial Kernel

Introduces polynomial decision boundaries.

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
(\mathbf{x}^T\mathbf{z}+c)^d
}
$$

where

* $d$ is the polynomial degree,
* $c$ is a constant.

---

### Radial Basis Function (RBF)

The most commonly used nonlinear kernel.

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\exp
\left(
-\gamma
\|
\mathbf{x}-\mathbf{z}
\|^2
\right)
}
$$

where

$$
\gamma
$$

controls how rapidly similarity decreases with distance.

---

### Sigmoid Kernel

Inspired by neural networks.

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\tanh
(
\alpha
\mathbf{x}^T\mathbf{z}
+
c
)
}
$$

Although available,

it is used less frequently than the RBF kernel.

---

## <font color='orange'> 10. Advantages of the Kernel Trick </font>

The Kernel Trick provides several important benefits.

* Solves nonlinear classification problems.
* Avoids explicitly computing high-dimensional features.
* Makes complex decision boundaries computationally feasible.
* Often achieves excellent predictive performance.
* Flexible through the choice of kernel function.

---

## <font color='orange'> 11. Limitations </font>

Despite its power,

kernel methods also have limitations.

* Training time increases with dataset size.
* Selecting an appropriate kernel requires experimentation.
* Hyperparameter tuning becomes more important.
* Kernel SVMs are generally slower than linear SVMs on very large datasets.

---

## <font color='orange'> 12. Kernel Selection in Practice </font>

Some general guidelines are

| Dataset | Recommended Kernel |
| :--- | :--- |
| Approximately Linear | Linear |
| Unknown Structure | RBF |
| Polynomial Relationships | Polynomial |
| Text Classification | Linear |
| Image Classification | Usually RBF |

In practice,

the RBF kernel is often chosen as the default because of its flexibility.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Feature Mapping | $\phi(\mathbf{x})$ transforms data into a higher-dimensional space |
| Kernel Function | $K(\mathbf{x},\mathbf{z})=\phi(\mathbf{x})^T\phi(\mathbf{z})$ |
| Kernel Trick | Computes inner products without explicit feature mapping |
| Linear Kernel | Linear decision boundary |
| Polynomial Kernel | Polynomial decision boundary |
| RBF Kernel | Flexible nonlinear boundary |
| Sigmoid Kernel | Neural-network-inspired kernel |

> **Key Insight:** The Kernel Trick is one of the most elegant ideas in machine learning. Rather than explicitly transforming observations into a high-dimensional feature space, SVMs use a kernel function to compute inner products as if the transformation had already been performed. This allows Support Vector Machines to construct highly nonlinear decision boundaries while avoiding the enormous computational cost of generating high-dimensional features explicitly. As a result, kernel methods combine the mathematical simplicity of linear classifiers with the flexibility required to solve complex real-world classification problems.

# <font color='blue'> Common Kernel Functions in Support Vector Machines </font>

The Kernel Trick allows Support Vector Machines to solve nonlinear classification problems without explicitly transforming the input data into higher-dimensional feature spaces. This flexibility is achieved through the choice of a **kernel function**, which defines how similarity between two observations is measured.

Different kernel functions generate different decision boundaries. Consequently, selecting an appropriate kernel is one of the most important decisions when designing an SVM classifier.

Although numerous kernel functions have been proposed, four kernels are used most frequently:

* Linear Kernel
* Polynomial Kernel
* Radial Basis Function (RBF) Kernel
* Sigmoid Kernel

Each kernel has distinct mathematical properties and is suited to different types of classification problems.

---

## <font color='orange'> 1. Linear Kernel </font>

The simplest kernel is the **Linear Kernel**.

It is defined as

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\mathbf{x}^T\mathbf{z}
}
$$

where

* $\mathbf{x}$ and $\mathbf{z}$ are two feature vectors.

Since

$$
K(\mathbf{x},\mathbf{z})
=
\phi(\mathbf{x})^T\phi(\mathbf{z}),
$$

the linear kernel corresponds to

$$
\phi(\mathbf{x})
=
\mathbf{x}.
$$

In other words,

no transformation is performed.

The classifier therefore constructs a linear decision boundary directly in the original feature space.

### Advantages

* Fastest kernel.
* Few hyperparameters.
* Easy to interpret.
* Works well in high-dimensional datasets.

### Limitations

* Cannot model nonlinear decision boundaries.

### Typical Applications

* Text Classification
* Spam Detection
* Gene Expression Analysis
* Credit Risk Assessment

---

## <font color='orange'> 2. Polynomial Kernel </font>

The Polynomial Kernel introduces nonlinear interactions between features.

It is defined as

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
(\mathbf{x}^T\mathbf{z}+c)^d
}
$$

where

* $d$ is the polynomial degree,
* $c$ is a constant (often called **coef0**).

The polynomial degree determines the complexity of the decision boundary.

### Degree 1

$$
d=1
$$

reduces to the linear kernel.

### Degree 2

Produces quadratic decision boundaries.

### Higher Degrees

Increasing

$$
d
$$

produces progressively more flexible decision boundaries capable of modelling increasingly complex relationships.

However,

very large values of

$$
d
$$

may lead to overfitting.

### Advantages

* Captures feature interactions.
* Suitable for moderately nonlinear datasets.

### Limitations

* Computationally slower than the linear kernel.
* Requires tuning of both

$$
d
$$

and

$$
c.
$$

---

## <font color='orange'> 3. Radial Basis Function (RBF) Kernel </font>

The most widely used kernel is the

**Radial Basis Function (Gaussian) Kernel**.

It is defined as

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\exp
\left(
-\gamma
\|
\mathbf{x}-\mathbf{z}
\|^2
\right)
}
$$

where

$$
\gamma
$$

is a positive hyperparameter controlling the width of the Gaussian.

Unlike the previous kernels,

the RBF kernel depends on the Euclidean distance between observations.

Nearby observations have kernel values close to

$$
1,
$$

whereas distant observations have kernel values approaching

$$
0.
$$

---

### Effect of \(\gamma\)

The parameter

$$
\gamma
$$

controls how rapidly similarity decreases with distance.

### Small \(\gamma\)

The Gaussian function is wide.

Many observations influence one another,

producing a smooth decision boundary.

Characteristics:

* Higher Bias
* Lower Variance
* Lower risk of overfitting

---

### Large \(\gamma\)

The Gaussian function becomes narrow.

Only very nearby observations influence each prediction.

Characteristics:

* Lower Bias
* Higher Variance
* Greater risk of overfitting

Thus,

$$
\gamma
$$

plays a role analogous to model complexity.

---

### Advantages

* Extremely flexible.
* Handles highly nonlinear data.
* Performs well on many real-world datasets.

### Limitations

* Requires tuning of both

$$
C
$$

and

$$
\gamma.
$$

* Computationally slower than linear SVMs.

---

## <font color='orange'> 4. Sigmoid Kernel </font>

The Sigmoid Kernel originates from neural networks.

It is defined as

$$
\boxed{
K(\mathbf{x},\mathbf{z})
=
\tanh
(
\alpha
\mathbf{x}^T\mathbf{z}
+
c
)
}
$$

where

* $\alpha$ controls the slope,
* $c$ is an offset.

Its functional form resembles the activation function used in early neural networks.

Although historically important,

the Sigmoid Kernel is used much less frequently than the Linear or RBF kernels.

### Advantages

* Can model nonlinear decision boundaries.

### Limitations

* Sensitive to hyperparameter selection.
* Often performs worse than the RBF kernel.

---

## <font color='orange'> 5. Comparing Kernel Complexity </font>

The flexibility of the decision boundary increases approximately as follows:

| Kernel | Decision Boundary |
| :--- | :--- |
| Linear | Straight line / hyperplane |
| Polynomial (Low Degree) | Smooth curves |
| Polynomial (High Degree) | Complex curves |
| RBF | Highly flexible nonlinear boundaries |
| Sigmoid | Moderate nonlinear boundaries |

Increasing flexibility generally increases the risk of overfitting.

---

## <font color='orange'> 6. Hyperparameter Summary </font>

Different kernels introduce different hyperparameters.

| Kernel | Important Hyperparameters |
| :--- | :--- |
| Linear | \(C\) |
| Polynomial | \(C\), degree, coef0 |
| RBF | \(C\), \(\gamma\) |
| Sigmoid | \(C\), \(\alpha\), coef0 |

These parameters are usually optimized using Grid Search or Randomized Search with Cross-Validation.

---

## <font color='orange'> 7. Choosing a Kernel </font>

There is no universally optimal kernel.

Some practical guidelines include

### Linear Kernel

Choose when

* the dataset is approximately linearly separable,
* the number of features is very large,
* computational efficiency is important.

---

### Polynomial Kernel

Choose when

* interactions between variables are expected,
* the relationship is moderately nonlinear.

---

### RBF Kernel

Choose when

* the structure of the data is unknown,
* nonlinear boundaries are expected.

For many practical applications,

the RBF kernel provides an excellent default choice.

---

### Sigmoid Kernel

Useful primarily for specialised applications,

although it is rarely the best-performing kernel in practice.

| :--- | :--- | :--- |
| Linear | $\mathbf{x}^T\mathbf{z}$ | High-dimensional linear data |
| Polynomial | $(\mathbf{x}^T\mathbf{z}+c)^d$ | Moderate nonlinear relationships |
| RBF | $\exp(-\gamma\|\mathbf{x}-\mathbf{z}\|^2)$ | General-purpose nonlinear classification |
| Sigmoid | $\tanh(\alpha\mathbf{x}^T\mathbf{z}+c)$ | Neural-network-inspired models |

> **Key Insight:** The kernel function determines how similarity between observations is measured and therefore controls the shape of the decision boundary learned by a Support Vector Machine. Linear kernels are efficient for approximately linear problems, Polynomial kernels capture interactions between features, and the Radial Basis Function (RBF) kernel provides highly flexible nonlinear decision boundaries, making it the most widely used kernel in practice. Selecting an appropriate kernel and tuning its associated hyperparameters are essential steps in building an effective SVM classifier.

# <font color='blue'> Hyperparameter Tuning in Support Vector Machines </font>

The predictive performance of a Support Vector Machine depends not only on the choice of kernel, but also on several **hyperparameters** that control the complexity of the classifier.

Unlike model parameters (such as the weights of the hyperplane), **hyperparameters are specified before training begins** and determine how the learning algorithm constructs the decision boundary.

Selecting appropriate hyperparameter values is one of the most important steps in developing a high-performing SVM classifier.

---

## <font color='orange'> 1. Parameters vs Hyperparameters </font>

Before discussing specific SVM hyperparameters, it is useful to distinguish between **model parameters** and **hyperparameters**.

### Model Parameters

Model parameters are learned automatically during training.

Examples include

* weight vector

$$
\mathbf{w},
$$

* bias

$$
b.
$$

These values define the final decision boundary.

---

### Hyperparameters

Hyperparameters are chosen **before** training.

Examples include

* Regularization parameter

$$
C,
$$

* Kernel parameter

$$
\gamma,
$$

* Polynomial degree

$$
d.
$$

Hyperparameters determine the flexibility and complexity of the resulting classifier.

---

## <font color='orange'> 2. The Regularization Parameter \(C\) </font>

The regularization parameter

$$
C
$$

controls the balance between

* maximizing the margin,
* minimizing classification errors.

Recall the soft-margin optimization problem

$$
\boxed{
\min
\;
\frac12\|\mathbf{w}\|^2
+
C
\sum_{i=1}^{m}
\xi_i.
}
$$

---

### Large \(C\)

A large value of

$$
C
$$

assigns a large penalty to classification errors.

Characteristics include

* narrow margin,
* low training error,
* higher variance,
* increased risk of overfitting.

---

### Small \(C\)

A small value of

$$
C
$$

permits more margin violations.

Characteristics include

* wider margin,
* greater regularization,
* higher bias,
* improved robustness.

---

## <font color='orange'> 3. The Kernel Parameter \(\gamma\) </font>

The parameter

$$
\gamma
$$

appears only for nonlinear kernels such as the RBF kernel.

Recall

$$
K(\mathbf{x},\mathbf{z})
=
\exp
\left(
-\gamma
\|
\mathbf{x}-\mathbf{z}
\|^2
\right).
$$

It controls how rapidly similarity decreases with distance.

---

### Small \(\gamma\)

The Gaussian function becomes wide.

Each training observation influences a large region of the feature space.

Characteristics

* smooth decision boundary,
* higher bias,
* lower variance.

---

### Large \(\gamma\)

The Gaussian becomes narrow.

Each observation influences only its immediate neighbourhood.

Characteristics

* highly irregular boundary,
* lower bias,
* higher variance,
* increased overfitting.

---

## <font color='orange'> 4. Polynomial Degree </font>

For Polynomial Kernels,

the degree

$$
d
$$

determines the complexity of the decision boundary.

Low degrees produce smooth curves,

while high degrees generate increasingly complicated boundaries.

Typical values are

$$
d=2,
\;
3,
\;
4.
$$

Very large values often lead to severe overfitting.

---

## <font color='orange'> 5. The Parameter coef0 </font>

Polynomial and Sigmoid kernels include an additional parameter

```
coef0
```

through the expression

$$
(\mathbf{x}^T\mathbf{z}+c)^d.
$$

The constant

$$
c
$$

shifts the polynomial and influences the relative contribution of higher-order terms.

Although important,

its influence is usually smaller than that of

$$
C
$$

and

$$
\gamma.
$$

---

## <font color='orange'> 6. Underfitting and Overfitting </font>

Hyperparameters determine model complexity.

### Underfitting

Occurs when

* \(C\) is too small,
* \(\gamma\) is too small.

Characteristics

* smooth boundary,
* poor training accuracy,
* poor test accuracy.

---

### Overfitting

Occurs when

* \(C\) is too large,
* \(\gamma\) is too large.

Characteristics

* highly irregular boundary,
* excellent training accuracy,
* poor generalization.

The objective is to identify the intermediate region that minimizes the validation error.

---

## <font color='orange'> 7. Cross-Validation </font>

Rather than evaluating a model using a single train-test split,

Cross-Validation repeatedly partitions the training data into

* training subsets,
* validation subsets.

The model is trained multiple times,

each time using a different validation set.

The average validation score provides a much more reliable estimate of model performance.

The most common approach is

**k-Fold Cross-Validation**.

---

### Five-Fold Cross-Validation

The dataset is divided into

five

equal subsets.

Each subset serves as the validation set once,

while the remaining four subsets are used for training.

The average score becomes

$$
\boxed{
\text{CV Score}
=
\frac1k
\sum_{i=1}^{k}
S_i
}
$$

where

$$
S_i
$$

is the validation score from fold

$$
i.
$$

---

## <font color='orange'> 8. Grid Search </font>

Grid Search systematically evaluates every possible combination of hyperparameters.

For example,

```python
param_grid = {
    "C":[0.1,1,10,100],
    "gamma":[0.001,0.01,0.1],
    "kernel":["rbf"]
}
```

The algorithm trains an SVM using every possible combination.

The model producing the highest Cross-Validation score is selected.

Although computationally expensive,

Grid Search guarantees that every candidate is evaluated.

---

## <font color='orange'> 9. Randomized Search </font>

Instead of testing every possible combination,

Randomized Search samples only a subset of parameter combinations.

Example

```python
RandomizedSearchCV(
    estimator=model,
    param_distributions=param_grid,
    n_iter=20
)
```

Advantages include

* considerably faster,
* suitable for large parameter spaces,
* often finds solutions comparable to Grid Search.

---

## <font color='orange'> 10. Validation Curves </font>

Validation Curves illustrate how model performance changes as a single hyperparameter varies.

For example,

varying

$$
C
$$

while keeping every other parameter fixed.

Typical behaviour

```
Accuracy

|
|           Validation
|          /\
|         /  \
|        /    \
|_______/______\_________

          C
```

The peak corresponds to the optimal hyperparameter.

Validation curves are particularly useful for identifying

* underfitting,
* overfitting,
* optimal complexity.

---

## <font color='orange'> 11. Practical Workflow </font>

A typical SVM optimisation workflow is

1. Choose an appropriate kernel.

2. Standardize all features.

3. Perform Cross-Validation.

4. Use Grid Search or Randomized Search.

5. Evaluate the final model using an independent test set.

This procedure minimizes overfitting while maximizing predictive performance.

## <font color='purple'> 14. Conceptual Summary </font>

| \(\gamma\) | Controls locality of the RBF kernel |

> **Key Insight:** The predictive performance of a Support Vector Machine depends critically on selecting appropriate hyperparameters. The regularization parameter \(C\) controls the trade-off between maximizing the margin and minimizing classification errors, while kernel-specific parameters such as \(\gamma\) and the polynomial degree determine the complexity of the decision boundary. Systematic optimisation techniques such as Cross-Validation, Grid Search, and Randomized Search allow these hyperparameters to be selected objectively, producing classifiers that generalize well to unseen data while avoiding both underfitting and overfitting.

# <font color='blue'> The Dual Optimization Problem and Lagrange Multipliers </font>

In the previous sections, the Support Vector Machine was formulated as a constrained optimization problem. Although this **primal formulation** is sufficient for understanding the geometric intuition behind SVMs, it is **not** how SVMs are actually solved in practice.

Instead, the optimization problem is transformed into an equivalent **dual formulation**, which possesses several remarkable advantages:

* It depends only on inner products between observations.
* It naturally leads to the **Kernel Trick**.
* Only the support vectors contribute to the final classifier.
* The optimization problem becomes easier to solve numerically.

The dual formulation therefore forms the mathematical foundation of modern Support Vector Machines.

---

## <font color='orange'> 1. The Primal Optimization Problem </font>

Recall that the hard-margin Support Vector Machine seeks to maximize the margin while correctly classifying every observation.

The optimization problem is

$$
\boxed{
\min_{\mathbf{w},b}
\;
\frac12
\|\mathbf{w}\|^2
}
$$

subject to

$$
\boxed{
y_i(\mathbf{w}^T\mathbf{x}_i+b)\ge1,
\qquad
i=1,\ldots,m.
}
$$

This is called the **primal optimization problem**.

Notice that the variables being optimized are

* the weight vector

$$
\mathbf{w},
$$

* the bias

$$
b.
$$

---

## <font color='orange'> 2. Constrained Optimization </font>

Unlike ordinary optimization,

the solution cannot take arbitrary values.

Every training observation must satisfy the inequality

$$
y_i(\mathbf{w}^T\mathbf{x}_i+b)\ge1.
$$

Problems of this type are called

**constrained optimization problems**.

A standard mathematical technique for solving such problems is the method of

**Lagrange Multipliers**.

---

## <font color='orange'> 3. Lagrange Multipliers </font>

Suppose an optimization problem has the form

$$
\min f(x)
$$

subject to

$$
g(x)\ge0.
$$

Rather than solving this constrained problem directly,

we construct a new function called the

**Lagrangian**

$$
\boxed{
L
=
f(x)
-
\lambda g(x)
}
$$

where

$$
\lambda
\ge0
$$

is called a

**Lagrange Multiplier**.

The multiplier determines how strongly violations of the constraint are penalized.

---

## <font color='orange'> 4. Constructing the SVM Lagrangian </font>

Applying this procedure to the SVM optimization problem,

we obtain

$$
\boxed{
L(\mathbf{w},b,\boldsymbol{\alpha})
=
\frac12
\|\mathbf{w}\|^2
-
\sum_{i=1}^{m}
\alpha_i
\left[
y_i(\mathbf{w}^T\mathbf{x}_i+b)-1
\right]
}
$$

where

$$
\alpha_i
\ge0
$$

are the Lagrange multipliers.

Notice that

* one multiplier is introduced for every training observation,
* there are therefore

$$
m
$$

multipliers.

---

## <font color='orange'> 5. The Karush-Kuhn-Tucker (KKT) Conditions </font>

The optimal solution satisfies the

**Karush-Kuhn-Tucker (KKT) Conditions**.

Without deriving them in detail,

the most important conditions are

### Stationarity

$$
\frac{\partial L}{\partial\mathbf{w}}
=
0
$$

and

$$
\frac{\partial L}{\partial b}
=
0.
$$

---

### Complementary Slackness

$$
\boxed{
\alpha_i
\left[
y_i(\mathbf{w}^T\mathbf{x}_i+b)-1
\right]
=
0.
}
$$

This equation has profound consequences.

---

## <font color='orange'> 6. Deriving the Weight Vector </font>

Differentiating the Lagrangian with respect to

$$
\mathbf{w}
$$

gives

$$
\frac{\partial L}{\partial\mathbf{w}}
=
\mathbf{w}
-
\sum_i
\alpha_i
y_i
\mathbf{x}_i.
$$

Setting the derivative equal to zero,

$$
\boxed{
\mathbf{w}
=
\sum_{i=1}^{m}
\alpha_i
y_i
\mathbf{x}_i.
}
$$

This is one of the most important equations in Support Vector Machines.

Notice that

the weight vector is simply a weighted combination of the training observations.

---

## <font color='orange'> 7. Why Only Support Vectors Matter </font>

Recall the KKT condition

$$
\alpha_i
\left[
y_i(\mathbf{w}^T\mathbf{x}_i+b)-1
\right]
=
0.
$$

There are two possibilities.

### Case 1

If

$$
\alpha_i=0,
$$

then

the observation contributes nothing to

$$
\mathbf{w}.
$$

---

### Case 2

If

$$
\alpha_i>0,
$$

then

$$
y_i(\mathbf{w}^T\mathbf{x}_i+b)=1.
$$

The observation lies exactly on the margin.

Therefore,

only observations lying on the margin have non-zero

$$
\alpha_i.
$$

These are precisely the

**Support Vectors**.

---

## <font color='orange'> 8. The Dual Optimization Problem </font>

Substituting

$$
\mathbf{w}
=
\sum_i
\alpha_i
y_i
\mathbf{x}_i
$$

back into the Lagrangian eliminates

$$
\mathbf{w}
$$

entirely.

The optimization becomes

$$
\boxed{
\max_{\boldsymbol{\alpha}}
\;
\sum_i
\alpha_i
-
\frac12
\sum_i
\sum_j
\alpha_i
\alpha_j
y_i
y_j
(\mathbf{x}_i^T\mathbf{x}_j)
}
$$

subject to

$$
\alpha_i\ge0
$$

and

$$
\sum_i
\alpha_i
y_i
=
0.
$$

This is called the

**Dual Optimization Problem**.

---

## <font color='orange'> 9. The Most Important Observation </font>

Notice something remarkable.

The observations never appear individually.

Instead,

they always appear through the dot product

$$
\boxed{
\mathbf{x}_i^T\mathbf{x}_j.
}
$$

This observation makes the Kernel Trick possible.

Instead of computing

$$
\mathbf{x}_i^T\mathbf{x}_j,
$$

we simply replace it by

$$
\boxed{
K(\mathbf{x}_i,\mathbf{x}_j).
}
$$

The optimization algorithm itself remains unchanged.

---

## <font color='orange'> 10. Decision Function </font>

After solving for the

$$
\alpha_i,
$$

predictions are made using

$$
\boxed{
f(\mathbf{x})
=
\sum_i
\alpha_i
y_i
K(\mathbf{x}_i,\mathbf{x})
+
b.
}
$$

Only support vectors have

$$
\alpha_i>0.
$$

Consequently,

the prediction depends only on the support vectors.

---

## <font color='orange'> 11. Why the Dual Formulation is Preferred </font>

The dual formulation provides several advantages.

* Naturally incorporates kernel functions.
* Eliminates the need to compute high-dimensional feature vectors explicitly.
* Uses only support vectors.
* Often more efficient than solving the primal problem.

For these reasons,

virtually every practical SVM implementation solves the **dual optimization problem** rather than the primal formulation.

---

## <font color='purple'> 12. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Primal Problem | Optimizes \(\mathbf{w}\) and \(b\) directly |
| Lagrange Multipliers | Convert constrained optimization into an unconstrained form |
| KKT Conditions | Characterize the optimal solution |
| Dual Problem | Optimizes the multipliers \(\alpha_i\) |
| Weight Vector | \(\mathbf{w}=\sum_i\alpha_i y_i\mathbf{x}_i\) |
| Support Vectors | Observations with \(\alpha_i>0\) |
| Kernel Trick | Replace \(\mathbf{x}_i^T\mathbf{x}_j\) by \(K(\mathbf{x}_i,\mathbf{x}_j)\) |

> **Key Insight:** Although Support Vector Machines are introduced as maximum-margin classifiers, they are almost never solved using the primal optimization problem. Instead, the problem is reformulated using **Lagrange multipliers**, yielding the **dual optimization problem**. In this formulation, the data appear only through inner products between observations, allowing these products to be replaced by kernel functions. This elegant reformulation explains why only the support vectors determine the classifier and provides the mathematical foundation for the powerful Kernel Trick that enables SVMs to solve highly nonlinear classification problems.

# <font color='blue'> Support Vector Regression (SVR) </font>

Support Vector Machines were originally developed for **classification**, where the objective is to separate observations belonging to different classes. However, the same underlying principles can also be applied to **regression problems**, where the goal is to predict a continuous numerical value.

This extension is known as **Support Vector Regression (SVR)**.

Unlike ordinary regression techniques, which attempt to minimize the prediction error for every training observation, SVR introduces the idea of an **ε-insensitive region**, within which prediction errors are ignored. The model therefore focuses only on observations whose errors exceed a specified tolerance.

This results in a regression model that is often more robust to noise and outliers than traditional least-squares methods.

---

## <font color='orange'> 1. Regression vs Classification </font>

The objective of a classification algorithm is

$$
\mathbf{x}
\longrightarrow
\{-1,+1\},
$$

where each observation is assigned to one of several discrete classes.

In contrast,

regression seeks to estimate a continuous output

$$
\mathbf{x}
\longrightarrow
y,
$$

where

$$
y\in\mathbb{R}.
$$

Examples include

* predicting house prices,
* forecasting stock prices,
* estimating insurance costs,
* predicting temperatures.

---

## <font color='orange'> 2. The Regression Function </font>

SVR seeks a function

$$
\boxed{
f(\mathbf{x})
=
\mathbf{w}^T\mathbf{x}+b
}
$$

that approximates the target values.

Notice that this has exactly the same mathematical form as the decision function used for classification.

The difference lies entirely in the optimization objective.

---

## <font color='orange'> 3. The ε-Insensitive Tube </font>

Instead of forcing the prediction to pass through every observation,

SVR constructs a tube of width

$$
2\varepsilon
$$

around the regression function.

```
           ●

------------------  +ε

==================  Regression Line

------------------  -ε

      ●      ●
```

Predictions lying inside this tube are considered sufficiently accurate.

Consequently,

errors smaller than

$$
\varepsilon
$$

are ignored completely.

---

## <font color='orange'> 4. The ε-Insensitive Loss Function </font>

The defining feature of SVR is the

**ε-insensitive loss function**

defined as

$$
\boxed{
L_{\varepsilon}(y,f(\mathbf{x}))
=
\max
\left(
0,
|y-f(\mathbf{x})|-\varepsilon
\right).
}
$$

This loss behaves differently from MSE or MAE.

---

### Error Smaller than ε

If

$$
|y-f(\mathbf{x})|
\le
\varepsilon,
$$

then

$$
L_{\varepsilon}=0.
$$

No penalty is applied.

---

### Error Larger than ε

If

$$
|y-f(\mathbf{x})|
>
\varepsilon,
$$

only the amount exceeding

$$
\varepsilon
$$

is penalized.

Thus,

small prediction errors are ignored,

while larger deviations contribute linearly to the loss.

---

## <font color='orange'> 5. Introducing Slack Variables </font>

Just as in Soft-Margin SVM,

SVR introduces slack variables

$$
\xi_i
\quad\text{and}\quad
\xi_i^*.
$$

These variables measure deviations above and below the

ε-insensitive tube.

The constraints become

$$
y_i-f(\mathbf{x}_i)
\le
\varepsilon+\xi_i,
$$

and

$$
f(\mathbf{x}_i)-y_i
\le
\varepsilon+\xi_i^*,
$$

where

$$
\xi_i,\;
\xi_i^*
\ge0.
$$

---

## <font color='orange'> 6. The Optimization Problem </font>

The objective function becomes

$$
\boxed{
\min
\;
\frac12
\|\mathbf{w}\|^2
+
C
\sum_i
(\xi_i+\xi_i^*)
}
$$

subject to the constraints introduced above.

Notice the similarity to Soft-Margin SVM.

The optimization still seeks a simple model by minimizing

$$
\|\mathbf{w}\|^2,
$$

while simultaneously limiting prediction errors.

---

## <font color='orange'> 7. The Role of ε </font>

The parameter

$$
\varepsilon
$$

determines the width of the insensitive region.

---

### Small ε

A narrow tube.

The model attempts to fit the data closely.

Characteristics

* lower bias,
* higher variance,
* greater sensitivity to noise.

---

### Large ε

A wider tube.

More prediction errors are ignored.

Characteristics

* smoother regression function,
* greater robustness,
* increased regularization.

---

## <font color='orange'> 8. The Role of C </font>

The regularization parameter

$$
C
$$

plays exactly the same role as in classification.

Large

$$
C
$$

strongly penalizes observations lying outside the tube,

leading to

* smaller training errors,
* greater risk of overfitting.

Small

$$
C
$$

permits larger deviations,

leading to

* smoother regression,
* improved generalization.

---

## <font color='orange'> 9. Kernel Functions in SVR </font>

The regression function may also be nonlinear.

Replacing the dot product by a kernel function gives

$$
\boxed{
f(\mathbf{x})
=
\sum_i
(\alpha_i-\alpha_i^*)
K(\mathbf{x}_i,\mathbf{x})
+
b.
}
$$

Consequently,

SVR supports exactly the same kernel functions as SVM classification.

Examples include

* Linear Kernel,
* Polynomial Kernel,
* RBF Kernel,
* Sigmoid Kernel.

---

## <font color='orange'> 10. Advantages of SVR </font>

Support Vector Regression offers several important advantages.

* Robust to noisy observations.
* Controls model complexity naturally.
* Supports nonlinear regression through kernels.
* Often performs well in high-dimensional feature spaces.
* Good generalization performance.

---

## <font color='orange'> 11. Limitations of SVR </font>

Despite its strengths,

SVR also has limitations.

* Computationally expensive for large datasets.
* Sensitive to feature scaling.
* Requires tuning of

$$
C,
\;
\varepsilon,
\;
\gamma.
$$

* Less interpretable than linear regression.

---

## <font color='orange'> 12. SVR in Scikit-Learn </font>

Implementing SVR is almost identical to implementing SVM classification.

Example

```python
from sklearn.svm import SVR

model = SVR(
    kernel="rbf",
    C=10,
    epsilon=0.1,
    gamma="scale"
)

model.fit(X_train, y_train)

predictions = model.predict(X_test)
```

The hyperparameters have exactly the same interpretation discussed for SVM classification.

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |

> **Key Insight:** Support Vector Regression extends the maximum-margin philosophy of Support Vector Machines to continuous prediction problems. Rather than minimizing every prediction error, SVR introduces an **ε-insensitive loss function**, allowing small deviations between the predicted and actual values to be ignored. This creates a robust regression model that balances prediction accuracy with model simplicity. By combining the ε-insensitive loss, regularization through the parameter \(C\), and kernel functions for nonlinear modelling, SVR provides a flexible and powerful alternative to traditional regression techniques, particularly when robustness and generalization are more important than fitting every individual training observation exactly.
