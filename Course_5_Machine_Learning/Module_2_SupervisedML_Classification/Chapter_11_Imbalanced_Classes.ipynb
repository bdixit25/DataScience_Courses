# <font color='blue'> Modeling Imbalanced Classes </font>

Many real-world classification problems involve **highly imbalanced datasets**, where one class occurs much more frequently than the other. Examples include fraud detection, rare disease diagnosis, equipment failure prediction, network intrusion detection, and searches for rare physical processes in particle physics.

In these problems, a classifier that simply predicts the majority class can achieve an apparently excellent accuracy while completely failing to identify the minority class.

Consequently, imbalanced classification requires specialized evaluation metrics, sampling strategies, and learning algorithms that explicitly account for unequal class distributions.

---

## <font color='orange'> 1. What is Class Imbalance? </font>

A classification dataset is said to be **imbalanced** when one class contains substantially more observations than another.

For example,

| Class | Number of Samples |
| :--- | ---: |
| Legitimate Transaction | 284,315 |
| Fraudulent Transaction | 492 |

Although fraud represents the class of greatest interest,

it constitutes only

$$
\boxed{
\frac{492}{284807}
\approx
0.17\%
}
$$

of the dataset.

This severe imbalance creates significant challenges during model training.

---

## <font color='orange'> 2. Why is Accuracy Misleading? </font>

Suppose a classifier predicts

```
Every transaction is legitimate.
```

Its confusion matrix becomes

| | Predicted Legitimate | Predicted Fraud |
| :--- | ---: | ---: |
| Legitimate | 284315 | 0 |
| Fraud | 492 | 0 |

The accuracy is

$$
\boxed{
Accuracy
=
\frac{284315}{284807}
=
99.83\%.
}
$$

Despite this impressive accuracy,

the classifier detects

**zero fraud cases.**

The model is therefore useless.

---

## <font color='orange'> 3. Evaluation Metrics for Imbalanced Data </font>

For imbalanced datasets,

the following metrics are generally more informative than accuracy.

* Precision
* Recall (Sensitivity)
* F1 Score
* ROC-AUC
* Precision-Recall AUC
* Balanced Accuracy
* Matthews Correlation Coefficient (MCC)

Among these,

**Precision-Recall AUC** is often the preferred metric for extremely imbalanced problems.

---

## <font color='orange'> 4. Why Precision and Recall Matter </font>

Suppose

```
100 Fraud Cases
```

exist.

Model A detects

```
95 Fraud Cases
```

Recall

$$
=
95\%.
$$

However,

suppose it also incorrectly labels

```
1000 Legitimate Transactions
```

as fraud.

Precision becomes low.

Conversely,

a conservative model may achieve very high precision,

but detect only a handful of fraud cases.

Choosing the appropriate balance depends on the application.

## <font color='orange'> 7. Random Undersampling </font>

Random undersampling removes observations from the majority class.

```
Majority

100000

↓

5000
```

Advantages

* Fast
* Smaller datasets
* Reduced training time

Disadvantages

* Loss of information
* May remove useful observations

---

## <font color='orange'> 8. Random Oversampling </font>

Random oversampling duplicates minority observations.

```
Minority

500

↓

5000
```

Advantages

* No information loss
* Easy to implement

Disadvantages

* Increased overfitting
* Duplicate observations

---

## <font color='orange'> 9. Synthetic Oversampling (SMOTE) </font>

Instead of duplicating observations,

SMOTE (**Synthetic Minority Over-sampling Technique**) creates

new synthetic observations.

Suppose

```
Fraud A

Fraud B
```

exist.

SMOTE generates

```
Synthetic Fraud
```

between them.

This increases the minority class while reducing overfitting.

---

## <font color='orange'> 10. NearMiss </font>

NearMiss is an intelligent undersampling method.

Instead of removing majority observations randomly,

it removes observations that are far from the minority class,

retaining those near the decision boundary.

This often produces better classifiers than random undersampling.

---

## <font color='orange'> 11. Class Weighting </font>

Many classifiers allow different weights to be assigned to different classes.

Instead of changing the data,

the loss function is modified.

For example,

```
Fraud

↓

Weight = 10
```

```
Legitimate

↓

Weight = 1
```

Errors involving fraud become much more costly,

forcing the algorithm to pay greater attention to the minority class.

In Scikit-Learn,

this is often implemented using

```python
class_weight="balanced"
```

---

## <font color='orange'> 12. Stratified Sampling </font>

When splitting data into training and testing sets,

ordinary random sampling may accidentally produce

different class proportions.

Stratified sampling preserves

the original class distribution.

```python
train_test_split(
    X,
    y,
    stratify=y
)
```

This is almost always recommended for imbalanced classification.

---

## <font color='orange'> 13. Ensemble Methods for Imbalanced Data </font>

Several ensemble methods have been developed specifically for imbalanced datasets.

Examples include

* Balanced Random Forest
* EasyEnsemble
* RUSBoost
* Balanced Bagging

These methods combine

sampling

and

ensemble learning

to improve minority-class performance.

---

## <font color='orange'> 14. Threshold Moving </font>

Most classifiers predict probabilities.

The default threshold is

$$
0.5.
$$

However,

for fraud detection,

one may instead use

$$
0.2.
$$

Lowering the threshold

increases recall,

although precision usually decreases.

Choosing the threshold is therefore an application-dependent trade-off.

---

## <font color='orange'> 15. Practical Workflow </font>

A typical workflow for imbalanced classification is

```
Load Data

↓

Stratified Train/Test Split

↓

Scale Features (if required)

↓

Handle Class Imbalance
(
SMOTE /
Class Weights /
Undersampling
)

↓

Train Model

↓

Evaluate using

Precision

Recall

F1

ROC-AUC

PR-AUC

↓

Tune Threshold
```

> **Key Insight:** Imbalanced classification requires approaches that go beyond standard supervised learning. When the minority class is rare but important, conventional accuracy can provide a misleading assessment of model performance. Effective solutions combine appropriate evaluation metrics, such as Precision, Recall, F1 Score, and Precision–Recall AUC, with data-level techniques (e.g., SMOTE or undersampling), algorithm-level strategies (e.g., class weighting), or specialized ensemble methods. The choice of technique depends on the application, the degree of imbalance, and the relative cost of false positives and false negatives. Successful imbalanced classification therefore involves not only selecting a suitable learning algorithm but also carefully designing the entire modelling workflow.
