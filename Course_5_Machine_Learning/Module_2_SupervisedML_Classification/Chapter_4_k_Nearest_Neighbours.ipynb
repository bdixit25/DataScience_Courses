# <font color='blue'> Introduction to K-Nearest Neighbours (KNN) </font>

K-Nearest Neighbours (KNN) is one of the simplest supervised machine learning algorithms used for both **classification** and **regression**. Unlike Logistic Regression, which constructs an explicit mathematical model during training, KNN makes predictions by comparing new observations directly with previously observed training examples.

KNN is therefore known as an **instance-based learning algorithm** or a **lazy learning algorithm**, because almost all computation occurs during prediction rather than during training.

The central idea is remarkably intuitive:

> **Observations that are close together in feature space are likely to belong to the same class.**

This assumption is often called the **Principle of Local Similarity**.

---



## <font color='orange'> 1. The Basic Idea </font>

Suppose a new observation

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)
$$

is presented to the classifier.

Rather than fitting an equation,

KNN simply searches the training dataset and identifies the

$$
k
$$

nearest observations.

The predicted class is determined from the labels of these neighbouring observations.

For example,

suppose

$$
k=5.
$$

The five nearest neighbours have labels

| Neighbour | Class |
|:---:|:---:|
| 1 | Fraud |
| 2 | Genuine |
| 3 | Fraud |
| 4 | Fraud |
| 5 | Genuine |

The majority class is

Fraud.

Therefore,

the new observation is classified as

Fraud.

---



## <font color='orange'> 2. Why Is It Called "K-Nearest Neighbours"? </font>

The parameter

$$
k
$$

represents the number of neighbouring observations used when making a prediction.

For example,

| Value of \(k\) | Interpretation |
|:---:|:---|
| 1 | Only the closest neighbour is considered |
| 3 | Majority vote among the three nearest neighbours |
| 5 | Majority vote among five neighbours |
| 11 | Majority vote among eleven neighbours |

Thus,

the value of

$$
k
$$

controls how much local information is used during prediction.

---



## <font color='orange'> 3. Classification by Majority Voting </font>

Suppose

$$
k=7.
$$

The neighbouring observations are

| Class | Number of Neighbours |
|:---|---:|
| Fraud | 5 |
| Genuine | 2 |

Since

Fraud

receives the largest number of votes,

the classifier predicts

$$
\hat y=\text{Fraud}.
$$

This procedure is called

**Majority Voting**.

---



## <font color='orange'> 4. KNN for Regression </font>

Although KNN is commonly associated with classification,

it can also perform regression.

Instead of majority voting,

the prediction is simply the average of the neighbouring target values.

Suppose

$$
k=4,
$$

and the neighbouring house prices are

$$
£220000,
\quad
£235000,
\quad
£210000,
\quad
£225000.
$$

The prediction becomes

$$
\hat y
=
\frac{
220000+
235000+
210000+
225000
}
4
=
£222500.
$$

Thus,

classification uses

**majority voting**,

whereas regression uses

**averaging**.

---



## <font color='orange'> 5. Why KNN Is Called a Lazy Learning Algorithm </font>

Most machine learning algorithms perform substantial computation during training.

For example,

Logistic Regression estimates

$$
\beta
$$

by minimizing a loss function,

while Random Forest constructs hundreds of decision trees.

KNN does neither.

During training,

the algorithm simply stores the training dataset.

No mathematical model is fitted.

Training therefore consists of

```
Store Training Data
```

Prediction, however, requires

1. computing distances,
2. sorting the neighbours,
3. performing majority voting.

Consequently,

KNN has

* very fast training,
* relatively slow prediction.

---



## <font color='orange'> 6. Mathematical Formulation </font>

Suppose the training dataset contains

$$
m
$$

observations,

$$
(\mathbf{x}^{(1)},y^{(1)}),
\ldots,
(\mathbf{x}^{(m)},y^{(m)}).
$$

Given a new observation

$$
\mathbf{x},
$$

the algorithm

1. computes the distance

$$
d(\mathbf{x},\mathbf{x}^{(i)})
$$

to every training observation,

2. sorts the distances,

3. selects the

$$
k
$$

smallest distances,

4. predicts the majority class.

The choice of the distance function plays a crucial role in determining classifier performance.

---



## <font color='orange'> 7. Decision Boundary </font>

Unlike Logistic Regression,

KNN does not assume a linear decision boundary.

Instead,

the decision boundary is determined entirely by the distribution of the training observations.

Consequently,

KNN is capable of modelling highly nonlinear class boundaries.

This flexibility is one of its major strengths.

---



## <font color='orange'> 8. Parametric vs Non-Parametric Models </font>

Machine learning algorithms may be divided into

### Parametric Models

Examples

* Linear Regression
* Logistic Regression
* Naive Bayes

These assume a fixed mathematical form,

such as

$$
y
=
\beta^T\mathbf{x}.
$$

The number of model parameters is fixed,

regardless of dataset size.

---

### Non-Parametric Models

Examples

* KNN
* Decision Trees
* Random Forests

These do not assume a predefined functional form.

Their complexity naturally grows with the size of the dataset.

KNN belongs to this category because every training observation may influence future predictions.

---



## <font color='orange'> 9. Advantages of KNN </font>

KNN possesses several attractive properties.

* Extremely simple to understand.
* No assumptions regarding data distribution.
* Naturally handles nonlinear decision boundaries.
* No explicit training phase.
* Works well for small and medium-sized datasets.
* Can perform both classification and regression.

---



## <font color='orange'> 10. Limitations of KNN </font>

Despite its simplicity,

KNN also has important limitations.

* Slow prediction for large datasets.
* Sensitive to irrelevant features.
* Sensitive to feature scaling.
* Performance depends strongly on the choice of

$$
k.
$$

* Suffers from the **Curse of Dimensionality**, where distances become less meaningful in very high-dimensional spaces.

---



## <font color='orange'> 11. Typical Applications </font>

KNN is widely used in

* Credit Card Fraud Detection
* Medical Diagnosis
* Image Classification
* Handwritten Digit Recognition
* Recommendation Systems
* Pattern Recognition
* Document Classification
* Anomaly Detection

Although more sophisticated algorithms often outperform KNN on very large datasets, it remains an excellent baseline classifier because of its simplicity and interpretability.

---

## <font color='purple'> 12. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Learning Type | Supervised Learning |
| Model Type | Non-Parametric |
| Learning Strategy | Lazy Learning |
| Prediction Method | Majority Voting (Classification) |
| Regression Method | Average of Neighbours |
| Training Cost | Very Low |
| Prediction Cost | High |
| Decision Boundary | Nonlinear |

> **Key Insight:** K-Nearest Neighbours is fundamentally different from algorithms such as Logistic Regression. Instead of learning a mathematical model, KNN simply stores the training data and predicts the class of a new observation by examining its nearest neighbours in feature space. This makes KNN one of the simplest and most intuitive machine learning algorithms, while also allowing it to model highly complex decision boundaries without making assumptions about the underlying data distribution.

# <font color='blue'> Distance Metrics in K-Nearest Neighbours </font>

The defining characteristic of the K-Nearest Neighbours (KNN) algorithm is that predictions are based on the **distance** between observations. Before the algorithm can determine which observations are the "nearest", it must first define what is meant by **distance**.

In mathematics, a distance function (or **metric**) measures the degree of similarity between two observations in a multidimensional feature space. The choice of distance metric has a significant impact on the performance of a KNN classifier and should be selected according to the characteristics of the dataset.

Suppose two observations are represented by the feature vectors

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n)
$$

and

$$
\mathbf{y}
=
(y_1,y_2,\ldots,y_n).
$$

The objective of every distance metric is to compute

$$
d(\mathbf{x},\mathbf{y}),
$$

which quantifies how similar the two observations are.

Smaller distances indicate greater similarity.

---



## <font color='orange'> 1. Euclidean Distance </font>

The Euclidean distance is the most widely used distance metric and corresponds to the ordinary straight-line distance between two points.

### One Dimension

For a single variable,

$$
d
=
|x-y|.
$$

---

### Two Dimensions

Suppose two points are

$$
(x_1,x_2)
$$

and

$$
(y_1,y_2).
$$

Applying Pythagoras' theorem,

$$
d
=
\sqrt{
(x_1-y_1)^2
+
(x_2-y_2)^2
}.
$$

---

### Three Dimensions

Similarly,

$$
d
=
\sqrt{
(x_1-y_1)^2
+
(x_2-y_2)^2
+
(x_3-y_3)^2
}.
$$

---

### General Form

For

$$
n
$$

features,

$$
\boxed{
d_E
=
\sqrt{
\sum_{i=1}^{n}
(x_i-y_i)^2
}
}
$$

where

* \(x_i\) is the value of feature \(i\) for the first observation,
* \(y_i\) is the value of feature \(i\) for the second observation.

---

### Properties

* Most commonly used metric.
* Measures straight-line distance.
* Works well when features have similar scales.
* Sensitive to outliers.

---



## <font color='orange'> 2. Manhattan Distance </font>

The Manhattan distance is also called the

* Taxicab Distance
* City Block Distance

Instead of travelling diagonally,

movement occurs only horizontally and vertically,

similar to travelling through a city with rectangular streets.

The distance is therefore

$$
\boxed{
d_M
=
\sum_{i=1}^{n}
|x_i-y_i|
}
$$

---

### Example

Suppose

$$
(2,4)
$$

and

$$
(7,1).
$$

Then

$$
d_M
=
|2-7|
+
|4-1|
=
5+3
=
8.
$$

---

### Properties

* Less sensitive to outliers.
* Often preferred when features represent independent quantities.
* Produces diamond-shaped neighbourhoods.

---



## <font color='orange'> 3. Minkowski Distance </font>

Both Euclidean and Manhattan distances are special cases of a more general metric called the **Minkowski Distance**.

It is defined as

$$
\boxed{
d_p
=
\left(
\sum_{i=1}^{n}
|x_i-y_i|^p
\right)^{1/p}
}
$$

where

$$
p
>
0.
$$

---

### Special Cases

If

$$
p=1,
$$

then

$$
d_p=d_M,
$$

which is the Manhattan distance.

If

$$
p=2,
$$

then

$$
d_p=d_E,
$$

which is the Euclidean distance.

Therefore,

Minkowski distance provides a family of distance measures controlled by the parameter

$$
p.
$$

---



## <font color='orange'> 4. Chebyshev Distance </font>

Suppose

$$
p\rightarrow\infty.
$$

The Minkowski distance approaches the

**Chebyshev Distance**,

defined as

$$
\boxed{
d_C
=
\max_i
|x_i-y_i|
}
$$

Only the largest coordinate difference contributes to the distance.

---

### Example

Suppose

$$
(2,4,5)
$$

and

$$
(7,6,1).
$$

The coordinate differences are

$$
5,\;
2,\;
4.
$$

Therefore,

$$
d_C
=
5.
$$

---

### Applications

Chebyshev distance is useful when

* every feature has equal importance,
* the largest deviation dominates the comparison.

---



## <font color='orange'> 5. Mahalanobis Distance </font>

Euclidean distance assumes that every feature is independent.

However,

many datasets contain correlated variables.

Mahalanobis distance accounts for these correlations.

It is defined as

$$
\boxed{
d
=
\sqrt{
(\mathbf{x}-\mathbf{y})^T
\Sigma^{-1}
(\mathbf{x}-\mathbf{y})
}
}
$$

where

$$
\Sigma
$$

is the covariance matrix.

Unlike Euclidean distance,

Mahalanobis distance automatically adjusts for

* correlated variables,
* different feature variances.

---

### Applications

* Multivariate Statistics
* Anomaly Detection
* Finance
* Medical Diagnosis

---



## <font color='orange'> 6. Cosine Similarity </font>

Sometimes,

the direction of a vector is more important than its magnitude.

Cosine Similarity measures the angle between two vectors.

It is defined as

$$
\boxed{
\cos\theta
=
\frac{
\mathbf{x}\cdot\mathbf{y}
}
{
||\mathbf{x}||
\;
||\mathbf{y}||
}
}
$$

where

$$
||\mathbf{x}||
=
\sqrt{
\sum x_i^2
}.
$$

---

### Interpretation

| Cosine Similarity | Interpretation |
|:---:|:---|
| 1 | Identical direction |
| 0 | Orthogonal vectors |
| -1 | Opposite direction |

---

Cosine Distance is defined as

$$
\boxed{
d
=
1-\cos\theta.
}
$$

---

### Applications

* Text Classification
* Natural Language Processing
* Recommendation Systems
* Document Similarity

---



## <font color='orange'> 7. Hamming Distance </font>

Hamming distance is used for

categorical or binary variables.

It counts the number of positions at which two vectors differ.

Suppose

```
1011101
```

and

```
1001001
```

They differ in two positions,

therefore

$$
d_H=2.
$$

---

### Applications

* DNA Sequences
* Error Correcting Codes
* Binary Classification
* Computer Networks

---



## <font color='orange'> 8. Importance of Feature Scaling </font>

Distance metrics are strongly affected by feature scales.

Suppose a dataset contains

| Feature | Range |
|:---|---:|
| Age | 18–80 |
| Annual Income | £15,000–£250,000 |

Income dominates the Euclidean distance,

making Age almost irrelevant.

Therefore,

distance-based algorithms almost always require feature scaling.

The most common transformation is

$$
z
=
\frac{x-\mu}{\sigma},
$$

which standardizes every feature to

* Mean = 0
* Standard Deviation = 1.

---



## <font color='orange'> 9. Choosing the Appropriate Distance Metric </font>

There is no universally optimal distance metric.

The choice depends on the problem.

| Distance Metric | Best Used For |
|:---|:---|
| Euclidean | Continuous variables with similar scales |
| Manhattan | Robust to outliers |
| Minkowski | General-purpose metric |
| Chebyshev | Maximum deviation problems |
| Mahalanobis | Correlated variables |
| Cosine | Text mining and NLP |
| Hamming | Binary or categorical variables |

---



## <font color='orange'> 10. Computational Complexity </font>

Suppose the training dataset contains

$$
N
$$

observations,

each with

$$
d
$$

features.

Computing one Euclidean distance requires

$$
O(d)
$$

operations.

Comparing a new observation with the entire dataset therefore requires

$$
O(Nd).
$$

This explains why KNN becomes computationally expensive for large datasets.

---

## <font color='purple'> 11. Conceptual Summary </font>

| Metric | Formula | Typical Application |
| :--- | :--- | :--- |
| Euclidean | $\sqrt{\sum(x_i-y_i)^2}$ | General continuous data |
| Manhattan | $\sum|x_i-y_i|$ | Robust to outliers |
| Minkowski | $\left(\sum|x_i-y_i|^p\right)^{1/p}$ | General family of metrics |
| Chebyshev | $\max|x_i-y_i|$ | Maximum deviation |
| Mahalanobis | $\sqrt{(\mathbf{x}-\mathbf{y})^T\Sigma^{-1}(\mathbf{x}-\mathbf{y})}$ | Correlated variables |
| Cosine Distance | $1-\cos\theta$ | Text mining and recommendation systems |
| Hamming | Number of differing positions | Binary and categorical data |

> **Key Insight:** The performance of K-Nearest Neighbours depends fundamentally on how similarity between observations is measured. Different distance metrics emphasize different aspects of the data, such as absolute differences, geometric distance, correlation structure, or vector orientation. Since KNN bases every prediction on the notion of "nearest" neighbours, selecting an appropriate distance metric—and ensuring that features are properly scaled—is often as important as selecting the value of \(k\) itself.

# <font color='blue'> Choosing the Value of \(k\): The Bias-Variance Trade-off in K-Nearest Neighbours </font>

The performance of a K-Nearest Neighbours (KNN) classifier depends strongly on a single hyperparameter:

$$
k,
$$

the number of neighbouring observations used when making a prediction.

Choosing an appropriate value of

$$
k
$$

is one of the most important aspects of building an effective KNN classifier. If the value of

$$
k
$$

is too small, the classifier becomes overly sensitive to random fluctuations in the training data. Conversely, if

$$
k
$$

is too large, the classifier may become overly smooth and fail to capture genuine patterns.

The choice of

$$
k
$$

therefore represents a classic example of the **Bias-Variance Trade-off**.

---



## <font color='orange'> 1. What Does \(k\) Represent? </font>

Suppose a new observation is presented to the classifier.

Instead of examining every training example equally, KNN first identifies the

$$
k
$$

nearest observations.

The prediction is then determined by the majority class among these neighbours.

For example,

if

$$
k=5,
$$

the classifier considers only the five closest training observations.

If three belong to Class A and two belong to Class B,

the prediction becomes

$$
\hat y=\text{Class A}.
$$

Thus,

$$
k
$$

determines the size of the local neighbourhood used during prediction.

---



## <font color='orange'> 2. Small Values of \(k\) </font>

Suppose

$$
k=1.
$$

The prediction depends entirely on the nearest training observation.

Mathematically,

$$
\hat y
=
y_{\text{nearest}}.
$$

This produces an extremely flexible classifier.

### Advantages

* Captures highly complex decision boundaries.
* Detects local structure very well.

### Disadvantages

* Extremely sensitive to noise.
* Sensitive to outliers.
* Small changes in the training data may completely change the prediction.

Consequently,

small values of

$$
k
$$

produce **low bias** but **high variance**.

---



## <font color='orange'> 3. Large Values of \(k\) </font>

Now suppose

$$
k=101.
$$

Instead of considering one neighbour,

the classifier averages information over a much larger region.

The resulting decision boundary becomes much smoother.

### Advantages

* Less sensitive to noise.
* Stable predictions.
* Good generalization for smooth datasets.

### Disadvantages

* Ignores local structure.
* May smooth away important decision boundaries.

Large values of

$$
k
$$

therefore produce **high bias** but **low variance**.

---



## <font color='orange'> 4. Understanding Bias and Variance </font>

The effect of

$$
k
$$

can be understood through the Bias-Variance Trade-off.

### Low Bias

The model is flexible enough to represent the true relationship.

### High Bias

The model is overly simplistic and cannot capture the underlying structure.

---

### Low Variance

Predictions remain relatively unchanged when the training data changes.

### High Variance

Small changes in the training data produce large changes in the predictions.

---



## <font color='orange'> 5. Underfitting and Overfitting </font>

Choosing

$$
k
$$

incorrectly leads to either underfitting or overfitting.

### Very Small \(k\)

The classifier memorizes the training data.

Characteristics include

* Very low training error.
* High testing error.
* Poor generalization.

This is called

**Overfitting**.

---

### Very Large \(k\)

The classifier averages over too many observations.

Characteristics include

* High training error.
* High testing error.
* Decision boundary becomes excessively smooth.

This is called

**Underfitting**.

---



## <font color='orange'> 6. Graphical Interpretation </font>

As

$$
k
$$

increases,

the model complexity decreases.

The typical behaviour is

| Value of \(k\) | Model Behaviour |
|:---:|:---|
| Very Small | Overfitting |
| Moderate | Best Generalization |
| Very Large | Underfitting |

The testing error therefore follows a characteristic U-shaped curve,

similar to many other machine learning algorithms.

---



## <font color='orange'> 7. Odd vs Even Values of \(k\) </font>

For binary classification,

odd values of

$$
k
$$

are generally preferred.

For example,

if

$$
k=4,
$$

the neighbours may vote

| Class | Votes |
|:---|---:|
| Fraud | 2 |
| Genuine | 2 |

The prediction becomes ambiguous.

Using

$$
k=5
$$

eliminates ties,

making majority voting straightforward.

---



## <font color='orange'> 8. Choosing \(k\) Using Cross-Validation </font>

Rather than selecting

$$
k
$$

arbitrarily,

it is usually determined using **Cross-Validation**.

Suppose candidate values are

$$
k
=
1,
3,
5,
7,
9,
11,
15,
21.
$$

For each value,

the classifier is trained and evaluated using

k-fold Cross-Validation.

The value producing the highest validation score is selected.

This approach provides a more reliable estimate of the model's ability to generalize.

---



## <font color='orange'> 9. Validation Curves </font>

A useful way to visualize the effect of

$$
k
$$

is through a **Validation Curve**.

The horizontal axis represents

$$
k,
$$

while the vertical axis represents a performance metric such as

* Accuracy,
* Balanced Accuracy,
* F1 Score.

Typically,

* Training accuracy decreases as

$$
k
$$

increases,

while

* Validation accuracy initially improves,

reaches a maximum,

and then decreases.

The optimal value of

$$
k
$$

corresponds to the peak of the validation curve.

---



## <font color='orange'> 10. Rule-of-Thumb for Choosing \(k\) </font>

Although Cross-Validation is recommended,

a common heuristic is

$$
\boxed{
k
\approx
\sqrt{N}
}
$$

where

$$
N
$$

is the number of training observations.

For example,

if

$$
N=10\,000,
$$

then

$$
k
\approx
100.
$$

This rule provides only an initial estimate and should always be refined using validation.

---



## <font color='orange'> 11. Effect of Dataset Size </font>

The optimal value of

$$
k
$$

often depends on the size of the dataset.

### Small Dataset

A smaller value of

$$
k
$$

is often appropriate because fewer observations are available.

---

### Large Dataset

Larger values of

$$
k
$$

generally produce smoother decision boundaries and better generalization.

---



## <font color='orange'> 12. Practical Recommendations </font>

Some useful guidelines include

* Start with

$$
k=5
$$

or

$$
k=7.
$$

* Use odd values for binary classification.
* Standardize features before fitting KNN.
* Determine

$$
k
$$

using Cross-Validation.
* Avoid relying solely on training accuracy.

---



## <font color='orange'> 13. Relationship Between \(k\) and Model Complexity </font>

The relationship between

$$
k
$$

and model complexity is inverse.

| Small \(k\) | Large \(k\) |
|:---:|:---:|
| Complex decision boundary | Smooth decision boundary |
| Low Bias | High Bias |
| High Variance | Low Variance |
| Overfitting | Underfitting |

Thus,

increasing

$$
k
$$

acts as a form of regularization,

reducing model complexity.

---



## <font color='orange'> 14. Computational Considerations </font>

The value of

$$
k
$$

has relatively little impact on the computational cost of finding neighbours,

since the dominant cost lies in computing distances to all training observations.

However,

larger values of

$$
k
$$

slightly increase the cost of sorting and majority voting.

In practice,

this overhead is usually negligible compared with the distance calculations.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Choice of \(k\) | Effect |
|:---|:---|
| Small \(k\) | Flexible model, high variance, risk of overfitting |
| Large \(k\) | Smooth model, high bias, risk of underfitting |
| Moderate \(k\) | Best balance between bias and variance |
| Odd \(k\) | Avoids ties in binary classification |
| Cross-Validation | Preferred method for selecting \(k\) |

> **Key Insight:** The parameter \(k\) controls the complexity of the K-Nearest Neighbours classifier. Small values of \(k\) produce highly flexible models that closely follow the training data but are sensitive to noise and overfitting. Large values of \(k\) generate smoother decision boundaries that are more stable but may overlook important local structure, leading to underfitting. Selecting an appropriate value of \(k\) using cross-validation provides the best compromise between bias and variance, enabling the classifier to generalize effectively to unseen data.

# <font color='blue'> Weighted K-Nearest Neighbours (Distance-Weighted Voting) </font>

The standard K-Nearest Neighbours (KNN) algorithm assumes that **every neighbouring observation contributes equally** to the final prediction. Once the

$$
k
$$

nearest neighbours have been identified, each neighbour receives one vote, regardless of its distance from the query point.

Although this approach is simple, it ignores an important observation:

> **Nearby observations generally contain more useful information than distant observations.**

Weighted K-Nearest Neighbours addresses this limitation by assigning **larger weights to closer neighbours** and **smaller weights to more distant neighbours**.

---



## <font color='orange'> 1. Motivation </font>

Suppose the five nearest neighbours of a new observation have the following distances.

| Neighbour | Class | Distance |
|:---:|:---:|:---:|
| 1 | Fraud | 0.08 |
| 2 | Fraud | 0.12 |
| 3 | Genuine | 0.18 |
| 4 | Genuine | 2.15 |
| 5 | Genuine | 2.34 |

Using ordinary KNN,

the votes become

| Class | Votes |
|:---|---:|
| Fraud | 2 |
| Genuine | 3 |

The observation is classified as

**Genuine**.

However,

the two fraudulent neighbours are much closer than the three genuine neighbours.

Treating every neighbour equally may therefore produce a misleading prediction.

---



## <font color='orange'> 2. Basic Idea </font>

Instead of assigning each neighbour one vote,

each neighbour receives a weight

$$
w_i,
$$

which depends on its distance from the query point.

The closer the neighbour,

the larger its weight.

The predicted class becomes the class with the largest **weighted vote** rather than the largest number of neighbours.

---



## <font color='orange'> 3. Inverse Distance Weighting </font>

The most common weighting scheme is

**Inverse Distance Weighting**.

The weight assigned to neighbour

$$
i
$$

is

$$
\boxed{
w_i
=
\frac1{d_i}
}
$$

where

$$
d_i
$$

is the distance between the query point and neighbour

$$
i.
$$

Thus,

smaller distances produce larger weights.

---

### Example

Suppose the distances are

| Distance | Weight |
|:---:|:---:|
| 0.10 | 10.0 |
| 0.20 | 5.0 |
| 0.50 | 2.0 |
| 1.00 | 1.0 |

Notice that a neighbour located at

0.10

has ten times more influence than a neighbour located at

1.00.

---



## <font color='orange'> 4. Avoiding Division by Zero </font>

Suppose the query observation is identical to one of the training observations.

Then

$$
d_i=0.
$$

The weight becomes

$$
\frac10,
$$

which is undefined.

To avoid this problem,

the weight is usually modified as

$$
\boxed{
w_i
=
\frac1{d_i+\varepsilon}
}
$$

where

$$
\varepsilon
$$

is a very small positive number,

typically

$$
10^{-8}.
$$

---



## <font color='orange'> 5. Weighted Majority Voting </font>

Suppose there are

$$
C
$$

possible classes.

For each class,

the total vote is computed as

$$
\boxed{
V_c
=
\sum_{i\in c}
w_i
}
$$

where the summation includes only neighbours belonging to class

$$
c.
$$

The predicted class is then

$$
\boxed{
\hat y
=
\arg\max_c
V_c.
}
$$

Thus,

the class receiving the largest **weighted vote** becomes the prediction.

---



## <font color='orange'> 6. Numerical Example </font>

Suppose

$$
k=5,
$$

and the neighbours are

| Class | Distance |
|:---:|:---:|
| Fraud | 0.12 |
| Fraud | 0.20 |
| Genuine | 0.85 |
| Genuine | 1.40 |
| Genuine | 2.10 |

The corresponding inverse-distance weights are

| Class | Weight |
|:---:|:---:|
| Fraud | 8.33 |
| Fraud | 5.00 |
| Genuine | 1.18 |
| Genuine | 0.71 |
| Genuine | 0.48 |

Summing the weights,

Fraud receives

$$
8.33+5.00
=
13.33,
$$

while Genuine receives

$$
1.18+0.71+0.48
=
2.37.
$$

Although Genuine has more neighbours,

Fraud has the much larger weighted vote.

The classifier therefore predicts

**Fraud**.

---



## <font color='orange'> 7. Other Weighting Functions </font>

Inverse-distance weighting is not the only possibility.

Several alternative weighting functions exist.

### Inverse Square Distance

$$
\boxed{
w_i
=
\frac1{d_i^2}
}
$$

This places even greater emphasis on nearby observations.

---

### Gaussian Weighting

$$
\boxed{
w_i
=
\exp
\left(
-\frac{d_i^2}{2\sigma^2}
\right)
}
$$

where

$$
\sigma
$$

controls the width of the Gaussian.

Nearby observations dominate,

while distant neighbours contribute almost nothing.

---

### Uniform Weighting

The standard KNN algorithm corresponds to

$$
\boxed{
w_i=1.
}
$$

Every neighbour contributes equally.

---



## <font color='orange'> 8. Advantages of Weighted KNN </font>

Weighted voting offers several benefits.

* Nearby observations have greater influence.
* Reduces the effect of distant neighbours.
* Handles noisy datasets more effectively.
* Produces smoother decision boundaries.
* Often improves classification accuracy.

---



## <font color='orange'> 9. Limitations </font>

Weighted KNN also has some limitations.

* Sensitive to the choice of distance metric.
* Extremely small distances can dominate the prediction.
* Still suffers from the Curse of Dimensionality.
* Requires appropriate feature scaling.

---



## <font color='orange'> 10. Feature Scaling Remains Essential </font>

Distance weighting does **not** eliminate the need for feature scaling.

Suppose one feature ranges between

$$
0-1,
$$

while another ranges between

$$
0-10^6.
$$

The larger feature still dominates the distance calculation,

regardless of the weighting function.

Consequently,

StandardScaler or MinMaxScaler should still be applied before fitting a KNN model.

---



## <font color='orange'> 11. Weighted KNN in Scikit-Learn </font>

Scikit-Learn supports both standard and weighted KNN.

### Standard KNN

```python
KNeighborsClassifier(
    n_neighbors=5,
    weights="uniform"
)
```

Every neighbour contributes equally.

---

### Weighted KNN

```python
KNeighborsClassifier(
    n_neighbors=5,
    weights="distance"
)
```

Scikit-Learn automatically computes inverse-distance weights,

assigning larger influence to closer neighbours.

---



## <font color='orange'> 12. When Should Weighted KNN Be Used? </font>

Weighted KNN is generally preferred when

* neighbouring observations are unevenly distributed,
* classes overlap,
* noisy observations are present,
* local structure is important.

Uniform weighting may still perform well when

* observations are evenly spaced,
* datasets are small,
* noise levels are low.

---



## <font color='orange'> 13. Uniform vs Distance Weighting </font>

| Property | Uniform Weighting | Distance Weighting |
| :--- | :---: | :---: |
| Weight Assigned | Equal | Inversely proportional to distance |
| Nearby Points | Same importance | Greater importance |
| Distant Points | Same importance | Reduced importance |
| Noise Robustness | Moderate | Better |
| Typical Accuracy | Good | Often Better |

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Standard KNN | Equal voting by all neighbours |
| Weighted KNN | Votes weighted by distance |
| Common Weight | $w_i=\frac1{d_i}$ |
| Prediction | Largest weighted vote |
| Main Advantage | Nearby observations dominate |
| Scikit-Learn Option | `weights="distance"` |

> **Key Insight:** Standard K-Nearest Neighbours assumes that every neighbouring observation contributes equally to the prediction, regardless of its distance from the query point. Weighted KNN relaxes this assumption by assigning larger weights to closer neighbours, allowing observations that are most similar to the query point to have the greatest influence. In practice, distance-weighted voting often produces more accurate and robust classifiers, particularly for noisy datasets or when class boundaries overlap, and is therefore the recommended choice for many real-world applications.

# <font color='blue'> The Curse of Dimensionality and Feature Scaling in K-Nearest Neighbours </font>

K-Nearest Neighbours (KNN) is fundamentally a **distance-based learning algorithm**. Every prediction depends on identifying the observations that are closest to the query point. Consequently, the performance of KNN relies heavily on the assumption that **distance is a meaningful measure of similarity**.

Unfortunately, as the number of input features increases, this assumption gradually breaks down. In high-dimensional spaces, observations become increasingly sparse, distances between observations become more similar, and identifying truly "nearest" neighbours becomes much more difficult.

This phenomenon is known as the **Curse of Dimensionality**, and it is one of the principal limitations of KNN and many other distance-based machine learning algorithms.

---



## <font color='orange'> 1. What is Dimensionality? </font>

A **dimension** corresponds to one feature (or variable) in the dataset.

For example,

| Dataset | Number of Features |
| :--- | ---: |
| House Prices | 8 |
| Iris Dataset | 4 |
| Credit Card Fraud | 30 |
| MNIST Handwritten Digits | 784 |
| Colour Images (100 × 100 × 3) | 30,000 |

Thus,

the Credit Card Fraud dataset has

$$
30
$$

dimensions,

whereas the MNIST dataset has

$$
784
$$

dimensions.

---



## <font color='orange'> 2. Visualising Higher Dimensions </font>

The difficulty of high-dimensional spaces can be understood geometrically.

### One Dimension

Observations lie on a line.

```
-----------------------
```

---

### Two Dimensions

Observations lie on a plane.

```
•
     •
         •
   •
```

---

### Three Dimensions

Observations occupy a volume.

```
      •
   •
         •
```

---

As more dimensions are added,

the available volume increases extremely rapidly.

The observations therefore become increasingly spread apart.

---



## <font color='orange'> 3. Why Does Sparsity Increase? </font>

Suppose every feature ranges from

$$
0
\rightarrow1.
$$

### One Dimension

The total space has length

$$
1.
$$

---

### Two Dimensions

The total area becomes

$$
1^2=1.
$$

---

### Three Dimensions

The volume becomes

$$
1^3=1.
$$

---

Now suppose the range doubles,

$$
0
\rightarrow2.
$$

Then

| Dimension | Volume |
| :---: | ---: |
| 1 | 2 |
| 2 | 4 |
| 3 | 8 |
| 10 | 1024 |
| 30 | 1,073,741,824 |

Notice how rapidly the available space increases.

The training observations occupy only a tiny fraction of this enormous feature space.

The data therefore become **sparse**.

---



## <font color='orange'> 4. Distance Concentration </font>

One of the most surprising consequences of high dimensionality is that

**all distances begin to look similar.**

Suppose

the nearest neighbour is at distance

$$
5.2,
$$

while the furthest neighbour is at

$$
5.7.
$$

The difference is extremely small.

Consequently,

distinguishing between "near" and "far" observations becomes difficult.

This phenomenon is known as

**Distance Concentration**.

---



## <font color='orange'> 5. Why KNN Suffers </font>

Recall that KNN predicts the class of a new observation by selecting the

$$
k
$$

nearest neighbours.

If every observation lies at approximately the same distance,

there is no meaningful concept of

"nearest."

The classifier therefore becomes unstable,

and prediction accuracy deteriorates.

---



## <font color='orange'> 6. A Mathematical View </font>

The Euclidean distance between two observations is

$$
d
=
\sqrt{
\sum_{i=1}^{n}
(x_i-y_i)^2
}.
$$

As the number of dimensions

$$
n
$$

increases,

more squared terms contribute to the sum.

Even if each individual difference is small,

their cumulative contribution causes distances to become increasingly similar.

Thus,

distance loses its discriminating power.

---



## <font color='orange'> 7. Sample Size Requirements </font>

High-dimensional datasets require dramatically larger numbers of training observations.

Suppose each feature is divided into

10

intervals.

Then

| Dimensions | Grid Cells |
| :---: | ---: |
| 1 | 10 |
| 2 | 100 |
| 3 | 1000 |
| 5 | 100,000 |
| 10 | 10,000,000,000 |

To populate each region adequately,

the required dataset size grows **exponentially** with the number of dimensions.

This exponential growth is another manifestation of the Curse of Dimensionality.

---



## <font color='orange'> 8. Why Feature Scaling is Essential </font>

Distance calculations depend directly on the numerical values of the features.

Suppose a dataset contains

| Feature | Range |
| :--- | ---: |
| Age | 18–80 |
| Income | £20,000–£250,000 |

The Euclidean distance becomes

$$
d
=
\sqrt{
(\Delta\text{Age})^2
+
(\Delta\text{Income})^2
}.
$$

Since Income is several orders of magnitude larger,

it dominates the calculation,

making Age almost irrelevant.

The classifier therefore becomes biased toward the largest-scale feature.

---



## <font color='orange'> 9. Standardization </font>

To prevent one feature from dominating,

each variable is standardized.

The transformed feature is

$$
\boxed{
z
=
\frac{x-\mu}{\sigma}
}
$$

where

* \(\mu\) is the feature mean,
* \(\sigma\) is the standard deviation.

After transformation,

each feature has approximately

* Mean = 0
* Standard Deviation = 1.

---



## <font color='orange'> 10. Min-Max Normalization </font>

Another common transformation is

Min-Max Scaling,

defined by

$$
\boxed{
x'
=
\frac{x-x_{\min}}
{x_{\max}-x_{\min}}
}
$$

After scaling,

every feature lies within

$$
0
\le
x'
\le
1.
$$

---



## <font color='orange'> 11. StandardScaler vs MinMaxScaler </font>

| StandardScaler | MinMaxScaler |
| :--- | :--- |
| Mean = 0 | Minimum = 0 |
| Standard Deviation = 1 | Maximum = 1 |
| More robust for Gaussian-like data | Useful when feature bounds are important |
| Commonly used with KNN | Also widely used with KNN |

For most machine learning applications,

StandardScaler is generally preferred.

---



## <font color='orange'> 12. Reducing Dimensionality </font>

The Curse of Dimensionality can be mitigated by reducing the number of features.

Common approaches include

### Feature Selection

Retain only the most informative variables.

Examples include

* Mutual Information
* Recursive Feature Elimination
* Feature Importance from Random Forests

---

### Dimensionality Reduction

Construct new variables that summarize the original dataset.

The most common technique is

**Principal Component Analysis (PCA).**

Instead of

30

variables,

PCA may produce

5

principal components that retain

95%

of the information.

This greatly improves KNN performance.

---



## <font color='orange'> 13. Practical Recommendations </font>

When using KNN,

it is good practice to

* Standardize all features.
* Remove irrelevant variables.
* Consider PCA for high-dimensional datasets.
* Tune

$$
k
$$

using Cross-Validation.
* Experiment with different distance metrics.

---



## <font color='orange'> 14. Relationship to Other Algorithms </font>

The Curse of Dimensionality affects many machine learning algorithms,

particularly those based on distance.

Examples include

* K-Nearest Neighbours
* K-Means Clustering
* DBSCAN
* Hierarchical Clustering

Algorithms such as

* Decision Trees,
* Random Forests,
* Gradient Boosting,

are generally much less sensitive because they partition the feature space rather than relying directly on distances.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Dimensionality | Number of input features |
| Curse of Dimensionality | Performance degradation in high-dimensional spaces |
| Distance Concentration | Distances become increasingly similar |
| Sparsity | Observations occupy only a tiny fraction of feature space |
| Standardization | Mean = 0, Standard Deviation = 1 |
| Min-Max Scaling | Features scaled to [0,1] |
| PCA | Reduces dimensionality while retaining information |

> **Key Insight:** K-Nearest Neighbours assumes that nearby observations are genuinely similar. As the number of features increases, however, the feature space expands exponentially, causing observations to become sparse and pairwise distances to lose their discriminating power. This phenomenon, known as the **Curse of Dimensionality**, can significantly reduce the effectiveness of KNN. Proper feature scaling, careful feature selection, and dimensionality reduction techniques such as **Principal Component Analysis (PCA)** are therefore essential for maintaining good performance when applying KNN to high-dimensional datasets.
