# <font color='blue'> Confusion Matrix, Accuracy, Specificity, Precision, Recall and F1 Score </font>

After training a classification model, it is necessary to evaluate how well it performs on unseen data. Unlike regression, where the prediction error can be measured directly using quantities such as the Mean Squared Error (MSE), classification requires a different set of evaluation metrics because the output consists of **discrete class labels**.

The foundation of almost every classification metric is the **Confusion Matrix**, which summarizes the predictions made by the classifier.

---

## <font color='orange'> 1. The Confusion Matrix </font>

For a binary classification problem, every prediction falls into one of four possible categories.

| | **Predicted Positive** | **Predicted Negative** |
|:---|:---:|:---:|
| **Actual Positive** | True Positive (TP) | False Negative (FN) |
| **Actual Negative** | False Positive (FP) | True Negative (TN) |

Each quantity has a specific meaning.

### <font color='orange'> True Positive (TP) </font>

The model correctly predicts the positive class.

Example:

* A fraudulent transaction is correctly identified as fraud.

---

### <font color='orange'> True Negative (TN) </font>

The model correctly predicts the negative class.

Example:

* A genuine transaction is correctly classified as genuine.

---

### <font color='orange'> False Positive (FP) </font>

The model predicts a positive class when the true class is negative.

Example:

* A genuine customer is incorrectly flagged as fraud.

False positives are sometimes called **Type I Errors**.

---

### <font color='orange'> False Negative (FN) </font>

The model predicts a negative class when the true class is positive.

Example:

* A fraudulent transaction is missed.

False negatives are also known as **Type II Errors**.

---

## <font color='orange'> 2. Total Number of Observations </font>

The total number of observations is simply

$$
N = TP + TN + FP + FN.
$$

Every classification metric is derived from these four quantities.

---

## <font color='orange'> 3. Accuracy </font>

Accuracy measures the fraction of correctly classified observations.

It is defined as

$$
\boxed{
\text{Accuracy}
=
\frac{TP+TN}
{TP+TN+FP+FN}
}
$$

The value lies between

$$
0
\le
\text{Accuracy}
\le
1.
$$

An accuracy of

$$
0.95
$$

means

95% of all observations were classified correctly.

---

### Example

Suppose

| Quantity | Value |
|:---|---:|
| TP | 80 |
| TN | 900 |
| FP | 20 |
| FN | 10 |

Then

$$
\text{Accuracy}
=
\frac{80+900}{1010}
=
0.970.
$$

Therefore,

the classifier correctly predicts

97.0%

of all observations.

---

## <font color='orange'> 4. Why Accuracy Can Be Misleading </font>

Consider the Credit Card Fraud dataset.

| Class | Number |
|:---|---:|
| Genuine | 284315 |
| Fraud | 492 |

Suppose a classifier predicts

**every transaction as genuine.**

Then

$$
TP=0,
$$

$$
FP=0,
$$

$$
FN=492,
$$

$$
TN=284315.
$$

The resulting accuracy becomes

$$
\frac{284315}{284807}
=
99.83\%.
$$

Although the accuracy is almost

100%,

the classifier has failed completely because

it detects **zero fraudulent transactions.**

This demonstrates why **accuracy should never be used as the sole performance metric for imbalanced datasets.**

---

## <font color='orange'> 5. Precision </font>

Precision measures

> **Of all observations predicted to be positive, how many were actually positive?**

It is defined as

$$
\boxed{
\text{Precision}
=
\frac{TP}
{TP+FP}
}
$$

Precision is particularly important whenever

**false positives are expensive.**

Examples include

* Fraud detection
* Medical diagnosis
* Spam filtering

A high precision indicates that the model rarely raises false alarms.

---

### Example

Suppose

$$
TP=90,
$$

$$
FP=10.
$$

Then

$$
\text{Precision}
=
\frac{90}{100}
=
0.90.
$$

Thus,

90%

of all predicted fraud cases were genuine frauds.

---

## <font color='orange'> 6. Recall (Sensitivity or True Positive Rate) </font>

Recall answers a different question.

> **Of all actual positive observations, how many did the classifier identify?**

Mathematically,

$$
\boxed{
\text{Recall}
=
\frac{TP}
{TP+FN}
}
$$

Recall is also called

* Sensitivity
* True Positive Rate (TPR)

A classifier with high recall detects nearly all positive cases.

---

### Example

Suppose

$$
TP=90,
$$

$$
FN=30.
$$

Then

$$
\text{Recall}
=
\frac{90}{120}
=
0.75.
$$

The classifier successfully identifies

75%

of all fraud cases.

---

## <font color='orange'> 7. Precision vs Recall </font>

Precision and Recall often compete with one another.

Increasing Recall usually means lowering the classification threshold,

which identifies more positive cases,

but also increases the number of false positives.

Conversely,

increasing Precision usually requires raising the threshold,

which reduces false positives,

but misses more genuine positives.

Choosing between the two depends on the application.

---

## <font color='orange'> 8. Specificity (True Negative Rate) </font>

Specificity measures the classifier's ability to identify the negative class.

It is defined as

$$
\boxed{
\text{Specificity}
=
\frac{TN}
{TN+FP}
}
$$

Specificity answers

> **How well does the classifier identify negative observations?**

This metric is widely used in

* Medical diagnosis
* Quality control
* Manufacturing inspection

---

## <font color='orange'> 9. False Positive Rate </font>

The False Positive Rate is

$$
\boxed{
FPR
=
\frac{FP}
{FP+TN}
}
$$

Notice that

$$
FPR
=
1-\text{Specificity}.
$$

This quantity plays an important role in constructing ROC curves.

---

## <font color='orange'> 10. F1 Score </font>

Sometimes both Precision and Recall are equally important.

The **F1 Score** combines them into a single metric using the **harmonic mean**,

$$
\boxed{
F_1
=
\frac{2PR}
{P+R}
}
$$

or equivalently,

$$
\boxed{
F_1
=
\frac{2TP}
{2TP+FP+FN}
}
$$

The harmonic mean is used because it penalizes large differences between Precision and Recall.

A classifier achieves a high F1 Score only if **both Precision and Recall are high.**

---

## <font color='orange'> 11. Comparing the Metrics </font>

| Metric | Formula | Measures |
|:---|:---|:---|
| Accuracy | $\frac{TP+TN}{N}$ | Overall correctness |
| Precision | $\frac{TP}{TP+FP}$ | Reliability of positive predictions |
| Recall | $\frac{TP}{TP+FN}$ | Ability to detect positives |
| Specificity | $\frac{TN}{TN+FP}$ | Ability to detect negatives |
| F1 Score | $\frac{2PR}{P+R}$ | Balance between Precision and Recall |

---

## <font color='orange'> 12. Which Metric Should Be Used? </font>

Different applications require different metrics.

| Application | Most Important Metric |
|:---|:---|
| Credit Card Fraud | Recall / F1 Score |
| Cancer Diagnosis | Recall |
| Email Spam | Precision |
| Search Engines | Precision |
| Customer Churn | Recall |
| Balanced Datasets | Accuracy |

For fraud detection,

missing a fraudulent transaction is usually more costly than incorrectly flagging a genuine transaction.

Consequently,

Recall and F1 Score are often more informative than Accuracy.

---

## <font color='orange'> 13. Strengths and Limitations </font>

### Accuracy

**Advantages**

* Easy to interpret
* Suitable for balanced datasets

**Limitations**

* Misleading for imbalanced datasets

---

### Precision

**Advantages**

* Reduces false alarms

**Limitations**

* May miss many positive cases

---

### Recall

**Advantages**

* Detects nearly all positive observations

**Limitations**

* Often increases false positives

---

### F1 Score

**Advantages**

* Balances Precision and Recall
* Particularly useful for imbalanced datasets

**Limitations**

* Does not account for True Negatives

---

## <font color='purple'> 14. Conceptual Summary </font>

| Metric | High Value Indicates |
|:---|:---|
| Accuracy | Overall good performance |
| Precision | Few false positives |
| Recall | Few false negatives |
| Specificity | Few genuine cases incorrectly flagged |
| F1 Score | Good balance between Precision and Recall |

> **Key Insight:** Every classification metric measures a different aspect of model performance. While **Accuracy** evaluates overall correctness, it can be highly misleading for imbalanced datasets such as fraud detection. The **Confusion Matrix** provides the complete picture by separating predictions into True Positives, True Negatives, False Positives, and False Negatives. From these four quantities, metrics such as **Precision**, **Recall**, **Specificity**, and the **F1 Score** can be derived, each emphasizing different types of classification errors. Selecting the most appropriate metric depends on the practical cost of false positives and false negatives in the application under consideration.

# <font color='blue'> ROC and Precision-Recall Curves </font>

In the previous section, we introduced several metrics derived from the **Confusion Matrix**, including Accuracy, Precision, Recall, Specificity, and the F1 Score. Although these metrics are extremely useful, they all suffer from one important limitation:

> **They are computed for a single decision threshold.**

For Logistic Regression and many other classifiers, the model predicts a **probability** rather than a class label. The predicted class is obtained by selecting a threshold, typically

$$
P(y=1|\mathbf{x}) \ge 0.5.
$$

Changing this threshold changes the Confusion Matrix and therefore changes every performance metric.

Receiver Operating Characteristic (ROC) curves and Precision-Recall (PR) curves provide a much more comprehensive evaluation by measuring classifier performance over **all possible classification thresholds**.

---

## <font color='orange'> 1. Decision Threshold </font>

Suppose a Logistic Regression model predicts

| Observation | Fraud Probability |
|:---:|:---:|
| A | 0.98 |
| B | 0.91 |
| C | 0.74 |
| D | 0.41 |
| E | 0.08 |

Using the standard threshold

$$
0.5,
$$

the predictions become

| Probability | Prediction |
|:---:|:---:|
| 0.98 | Fraud |
| 0.91 | Fraud |
| 0.74 | Fraud |
| 0.41 | Genuine |
| 0.08 | Genuine |

Suppose the threshold is lowered to

$$
0.3.
$$

Now,

| Probability | Prediction |
|:---:|:---:|
| 0.98 | Fraud |
| 0.91 | Fraud |
| 0.74 | Fraud |
| 0.41 | Fraud |
| 0.08 | Genuine |

Notice that lowering the threshold increases the number of predicted fraud cases.

This generally

* increases Recall,
* decreases Precision,
* increases False Positives.

Changing the threshold therefore changes the classifier behaviour.

---

## <font color='orange'> 2. True Positive Rate (TPR) </font>

The **True Positive Rate** is simply another name for **Recall**.

It measures the fraction of positive observations correctly identified.

$$
\boxed{
TPR
=
\frac{TP}{TP+FN}
}
$$

The value ranges between

$$
0
\le
TPR
\le
1.
$$

A value close to one indicates that nearly every positive observation has been detected.

---

## <font color='orange'> 3. False Positive Rate (FPR) </font>

The False Positive Rate measures the fraction of negative observations that are incorrectly classified as positive.

$$
\boxed{
FPR
=
\frac{FP}{FP+TN}
}
$$

Notice the relationship

$$
FPR
=
1-\text{Specificity}.
$$

Unlike Recall,

we wish the False Positive Rate to remain as small as possible.

---

## <font color='orange'> 4. Constructing the ROC Curve </font>

To construct the ROC Curve,

the classification threshold is gradually decreased from

$$
1
\rightarrow
0.
$$

For every threshold,

compute

* True Positive Rate

$$
TPR
=
\frac{TP}{TP+FN}
$$

and

* False Positive Rate

$$
FPR
=
\frac{FP}{FP+TN}.
$$

Each threshold produces one point,

$$
(FPR,TPR).
$$

Joining all these points produces the **Receiver Operating Characteristic (ROC) Curve**.

---

## <font color='orange'> 5. Interpreting the ROC Curve </font>

A perfect classifier reaches

$$
(0,1),
$$

meaning

* zero false positives,
* all positive observations detected.

A completely random classifier follows the diagonal,

$$
TPR
=
FPR.
$$

The closer the ROC curve lies to the upper-left corner,

the better the classifier.

---

## <font color='orange'> 6. Area Under the ROC Curve (ROC-AUC) </font>

Rather than comparing complete ROC curves,

their performance is often summarized by the **Area Under the Curve (AUC).**

Mathematically,

$$
\boxed{
AUC
=
\int_0^1
TPR(FPR)\,
d(FPR)
}
$$

The value satisfies

$$
0
\le
AUC
\le
1.
$$

Typical interpretations are

| ROC-AUC | Interpretation |
|:---:|:---|
| 0.50 | Random guessing |
| 0.60–0.70 | Poor |
| 0.70–0.80 | Fair |
| 0.80–0.90 | Good |
| >0.90 | Excellent |
| 1.00 | Perfect classifier |

---

## <font color='orange'> 7. Limitations of ROC Curves </font>

ROC Curves are extremely useful for balanced datasets.

However,

they can produce overly optimistic evaluations for highly imbalanced datasets.

Consider fraud detection,

where

* Genuine Transactions = 284315
* Fraudulent Transactions = 492

Suppose

500 genuine transactions are falsely classified as fraud.

Then

$$
FPR
=
\frac{500}{284315}
=
0.0018.
$$

Although

500

customers have been incorrectly flagged,

the False Positive Rate remains extremely small.

The ROC curve therefore appears excellent,

even though the classifier may be impractical.

---

## <font color='orange'> 8. Precision-Recall Curves </font>

Precision-Recall Curves focus exclusively on the positive class.

Instead of plotting

$$
TPR
\quad\text{vs}\quad
FPR,
$$

they plot

$$
\boxed{
Precision
\quad\text{vs}\quad
Recall.
}
$$

Since fraud detection is primarily concerned with detecting fraudulent transactions,

PR Curves provide a more meaningful evaluation.

---

## <font color='orange'> 9. Constructing the Precision-Recall Curve </font>

For each classification threshold,

compute

$$
Precision
=
\frac{TP}{TP+FP}
$$

and

$$
Recall
=
\frac{TP}{TP+FN}.
$$

Plot

$$
Recall
$$

along the horizontal axis,

and

$$
Precision
$$

along the vertical axis.

As before,

varying the threshold generates the complete curve.

---

## <font color='orange'> 10. Average Precision (PR-AUC) </font>

The overall quality of a Precision-Recall Curve is summarized by its area,

known as

* Average Precision
* PR-AUC

$$
\boxed{
AP
=
\int_0^1
Precision(Recall)\,
d(Recall)
}
$$

Higher values indicate

* high Recall,
* high Precision,

simultaneously.

---

## <font color='orange'> 11. ROC vs Precision-Recall Curves </font>

| Property | ROC Curve | Precision-Recall Curve |
|:---|:---:|:---:|
| Horizontal Axis | False Positive Rate | Recall |
| Vertical Axis | True Positive Rate | Precision |
| Uses Negative Class | ✓ | ✗ |
| Uses Positive Class | ✓ | ✓ |
| Suitable for Balanced Data | Excellent | Good |
| Suitable for Imbalanced Data | Moderate | Excellent |

---

## <font color='orange'> 12. Which Curve Should You Use? </font>

The choice depends on the application.

### Balanced Dataset

Examples

* Image Recognition
* Species Classification
* Handwritten Digits

ROC Curves provide an excellent evaluation.

---

### Imbalanced Dataset

Examples

* Credit Card Fraud
* Medical Diagnosis
* Defect Detection
* Intrusion Detection

Precision-Recall Curves are generally preferred because they focus on correctly identifying the minority class.

---

## <font color='orange'> 13. Practical Interpretation </font>

Suppose two classifiers produce

| Model | ROC-AUC | PR-AUC |
|:---|:---:|:---:|
| Logistic Regression | 0.99 | 0.84 |
| Random Forest | 0.98 | 0.93 |

The ROC-AUC values suggest that Logistic Regression performs slightly better.

However,

the Precision-Recall Curve indicates that the Random Forest achieves much higher precision while maintaining good recall.

For fraud detection,

the Random Forest would generally be the preferred classifier.

---

## <font color='orange'> 14. Choosing the Decision Threshold </font>

Although

$$
0.5
$$

is commonly used,

there is no universal optimal threshold.

The threshold should be selected according to the application.

Examples include

| Application | Desired Property |
|:---|:---|
| Fraud Detection | High Recall |
| Cancer Screening | Very High Recall |
| Spam Detection | High Precision |
| Loan Approval | Balanced Precision and Recall |

The ROC and Precision-Recall Curves help identify an appropriate operating point for the classifier.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
|:---|:---|
| Decision Threshold | Converts probabilities into class labels |
| ROC Curve | TPR versus FPR |
| ROC-AUC | Overall classifier discrimination |
| Precision-Recall Curve | Precision versus Recall |
| PR-AUC | Overall performance on the positive class |
| Best Curve for Balanced Data | ROC |
| Best Curve for Imbalanced Data | Precision-Recall |

> **Key Insight:** Confusion matrix metrics evaluate a classifier at a **single decision threshold**, whereas ROC and Precision-Recall curves evaluate performance across **all possible thresholds**. While ROC curves measure the trade-off between the **True Positive Rate** and the **False Positive Rate**, Precision-Recall curves focus exclusively on the positive class and are therefore far more informative for highly imbalanced problems such as credit card fraud detection. Consequently, both ROC-AUC and PR-AUC should be reported, with particular emphasis on **PR-AUC** whenever the positive class is rare.

# <font color='blue'> Implementing ROC and Precision-Recall Curves in Scikit-Learn </font>

The previous section introduced the theoretical concepts behind **Receiver Operating Characteristic (ROC)** curves and **Precision-Recall (PR)** curves. In practice, these curves are generated using the **predicted probabilities** produced by a classifier rather than the predicted class labels.

Scikit-Learn provides built-in utilities that make generating these curves straightforward. Nevertheless, understanding the underlying calculations is essential for correctly interpreting the results.

---

## <font color='orange'> 1. Why Predicted Probabilities Are Needed </font>

Many classification algorithms do not simply predict

$$
0
\quad\text{or}\quad
1.
$$

Instead, they estimate the probability that an observation belongs to the positive class,

$$
P(y=1|\mathbf{x}).
$$

For example,

| Transaction | Predicted Probability |
|:---:|:---:|
| A | 0.97 |
| B | 0.82 |
| C | 0.54 |
| D | 0.29 |
| E | 0.08 |

The classifier converts these probabilities into class labels by applying a threshold,

typically

$$
0.5.
$$

However,

ROC and Precision-Recall curves evaluate the classifier over **every possible threshold**, not just 0.5.

---

## <font color='orange'> 2. Obtaining Prediction Probabilities </font>

After fitting the model,

predicted probabilities are obtained using

```python
y_prob = model.predict_proba(X_test)
```

For binary classification,

the output has the form

```python
[[0.97, 0.03],
 [0.22, 0.78],
 [0.85, 0.15],
 ...]
```

Each row contains

$$
\left[
P(y=0),
P(y=1)
\right].
$$

Since the probability of the positive class is required,

we extract the second column,

```python
y_prob = model.predict_proba(X_test)[:,1]
```

This vector is used to generate both ROC and Precision-Recall curves.

---

## <font color='orange'> 3. Computing the ROC Curve </font>

The ROC curve is generated by repeatedly varying the decision threshold.

Scikit-Learn performs this automatically using

```python
from sklearn.metrics import roc_curve
```

```python
fpr, tpr, thresholds = roc_curve(
    y_test,
    y_prob
)
```

The function returns three arrays.

### False Positive Rate

```python
fpr
```

contains

$$
\frac{FP}{FP+TN}.
$$

---

### True Positive Rate

```python
tpr
```

contains

$$
\frac{TP}{TP+FN}.
$$

---

### Thresholds

```python
thresholds
```

contains every probability threshold used to generate the ROC curve.

For each threshold,

a new confusion matrix is computed.

---

## <font color='orange'> 4. Plotting the ROC Curve </font>

The ROC curve is simply

```python
plt.plot(fpr,tpr)
```

The horizontal axis represents

False Positive Rate,

while the vertical axis represents

True Positive Rate.

A perfect classifier passes through

$$
(0,1).
$$

A random classifier follows the diagonal,

$$
TPR=FPR.
$$

---

## <font color='orange'> 5. Computing the ROC-AUC </font>

Rather than comparing complete curves,

their overall performance is summarized by the area underneath the curve.

Scikit-Learn computes this directly using

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(
    y_test,
    y_prob
)
```

Higher values indicate better class discrimination.

---

## <font color='orange'> 6. Computing the Precision-Recall Curve </font>

Precision-Recall curves are computed similarly.

```python
from sklearn.metrics import precision_recall_curve
```

```python
precision,
recall,
thresholds = precision_recall_curve(
    y_test,
    y_prob
)
```

The returned arrays contain

* Precision
* Recall
* Decision Threshold

for every operating point of the classifier.

---

## <font color='orange'> 7. Plotting the Precision-Recall Curve </font>

The curve is obtained by plotting

```python
plt.plot(recall,precision)
```

The horizontal axis represents

Recall,

while the vertical axis represents

Precision.

A classifier whose curve remains close to the upper-right corner performs well.

---

## <font color='orange'> 8. Computing Average Precision (PR-AUC) </font>

The area underneath the Precision-Recall curve is known as

Average Precision.

It is computed using

```python
from sklearn.metrics import average_precision_score

ap = average_precision_score(
    y_test,
    y_prob
)
```

This quantity is often reported as

* Average Precision
* AP
* PR-AUC

All refer to the same metric.

---

## <font color='orange'> 9. Built-In Visualization Tools </font>

Scikit-Learn includes dedicated plotting classes.

ROC Curve

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_predictions(
    y_test,
    y_prob
)
```

---

Precision-Recall Curve

```python
from sklearn.metrics import PrecisionRecallDisplay

PrecisionRecallDisplay.from_predictions(
    y_test,
    y_prob
)
```

These utilities automatically compute the curves and generate publication-quality figures.

---

## <font color='orange'> 10. Choosing the Optimal Threshold </font>

One advantage of ROC and PR curves is that they allow the decision threshold to be adjusted.

For example,

instead of

$$
0.5,
$$

one may choose

$$
0.3
$$

to improve Recall,

or

$$
0.8
$$

to improve Precision.

The choice depends on the application.

For fraud detection,

it is usually preferable to detect more fraudulent transactions,

even if this increases the number of false alarms.

---

## <font color='orange'> 11. Common Implementation Mistakes </font>

Several mistakes are frequently encountered.

### Using `predict()` Instead of `predict_proba()`

Incorrect

```python
roc_auc_score(y_test,y_pred)
```

Correct

```python
roc_auc_score(y_test,y_prob)
```

ROC curves require probabilities,

not predicted labels.

---

### Forgetting Feature Scaling

Algorithms such as

* Logistic Regression
* SVM
* KNN

should always be combined with

```python
StandardScaler()
```

using a Pipeline.

---

### Evaluating on the Training Set

ROC and Precision-Recall curves should always be generated using

```python
X_test
```

rather than

```python
X_train.
```

Otherwise,

the performance estimate becomes overly optimistic.

---

### Ignoring Class Imbalance

Accuracy may exceed

99%

for highly imbalanced datasets,

yet the classifier may detect almost no positive cases.

Always report

* Precision
* Recall
* F1 Score
* ROC-AUC
* PR-AUC

alongside Accuracy.

---

## <font color='orange'> 12. Recommended Evaluation Workflow </font>

A typical classification workflow proceeds as follows.

1. Split the dataset into training and testing sets.
2. Standardize the input features.
3. Train the classifier.
4. Predict class labels.
5. Predict class probabilities.
6. Compute the confusion matrix.
7. Calculate Accuracy, Precision, Recall, Specificity, and F1 Score.
8. Generate the ROC Curve.
9. Generate the Precision-Recall Curve.
10. Report both ROC-AUC and PR-AUC.

This workflow provides a comprehensive assessment of classifier performance.

---

## <font color='purple'> 13. Conceptual Summary </font>

| Step | Purpose |
|:---|:---|
| `predict()` | Predict class labels |
| `predict_proba()` | Estimate class probabilities |
| `roc_curve()` | Compute ROC coordinates |
| `roc_auc_score()` | Compute ROC-AUC |
| `precision_recall_curve()` | Compute Precision-Recall coordinates |
| `average_precision_score()` | Compute PR-AUC |
| `RocCurveDisplay` | Plot ROC Curve |
| `PrecisionRecallDisplay` | Plot Precision-Recall Curve |

> **Key Insight:** ROC and Precision-Recall curves provide a threshold-independent evaluation of a classifier by considering every possible decision threshold rather than a single operating point. In practice, these curves are generated using the predicted probabilities produced by the model. For balanced datasets, the ROC Curve and ROC-AUC provide an excellent measure of class discrimination. For highly imbalanced datasets, such as credit card fraud detection, the Precision-Recall Curve and Average Precision are generally more informative because they focus on the performance of the minority (positive) class. Together with the confusion matrix and derived metrics, these curves provide a complete and reliable assessment of classification performance.
