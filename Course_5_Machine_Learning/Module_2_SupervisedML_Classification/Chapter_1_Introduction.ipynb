# <font color='blue'> What is Classification? </font>

Classification is one of the two fundamental types of **supervised machine learning** problems. Unlike **regression**, where the objective is to predict a continuous numerical value, classification aims to predict the **category** or **class** to which an observation belongs.

In supervised learning, the model is provided with a labelled training dataset consisting of input variables (features) and their corresponding target labels (classes). The objective of the algorithm is to learn the relationship between the input features and the output labels so that it can accurately classify previously unseen observations.

Mathematically, if

$$
\mathbf{x} = (x_1,x_2,\ldots,x_n)
$$

represents a vector of input features, then a classifier attempts to learn a function

$$
f:\mathbf{x}\rightarrow y,
$$

where

$$
y\in\{1,2,\ldots,K\},
$$

and \(K\) represents the number of possible classes.

For binary classification,

$$
y\in\{0,1\},
$$

where the two classes typically represent the absence or presence of an event.

Examples include:

* Fraud vs Non-Fraud Transactions
* Spam vs Non-Spam Emails
* Malignant vs Benign Tumours
* Pass vs Fail
* Customer Churn vs Customer Retention

---

## <font color='orange'> 1. Classification vs Regression </font>

Although both classification and regression belong to supervised learning, they solve fundamentally different problems.

| Property | Regression | Classification |
| :--- | :--- | :--- |
| Target Variable | Continuous | Discrete |
| Example | House Price | Fraud Detection |
| Output | Real Number | Class Label |
| Typical Algorithms | Linear Regression, ElasticNet | Logistic Regression, SVM, Random Forest |
| Error Metric | MSE, MAE, RMSE | Accuracy, Precision, Recall, F1 |

For example,

Regression predicts

$$
\hat y = 253\,000
$$

where the output is a numerical value.

Classification predicts

$$
\hat y = \text{Fraud}
$$

or

$$
\hat y = 1.
$$

---

## <font color='orange'> 2. Binary Classification </font>

The simplest classification problem contains only two possible outcomes.

Examples include

* Fraud / Not Fraud
* Diseased / Healthy
* Positive / Negative
* Approved / Rejected

These classes are usually encoded numerically as

$$
y=
\begin{cases}
0,&\text{Negative Class}\\
1,&\text{Positive Class}
\end{cases}
$$

The learning algorithm estimates the probability that an observation belongs to the positive class,

$$
P(y=1|\mathbf{x}).
$$

A prediction is then made by comparing this probability with a decision threshold (commonly 0.5):

$$
\hat y=
\begin{cases}
1,&P(y=1|\mathbf{x})\ge0.5\\
0,&P(y=1|\mathbf{x})<0.5
\end{cases}
$$

---

## <font color='orange'> 3. Multi-Class Classification </font>

Some problems contain more than two classes.

Examples include

* Handwritten digit recognition (0–9)
* Species classification
* Language identification
* Image recognition

Here,

$$
y\in\{1,2,\ldots,K\}
$$

where

$$
K>2.
$$

The classifier predicts the class having the highest probability,

$$
\hat y
=
\arg\max_i P(y=i|\mathbf{x}).
$$

---

## <font color='orange'> 4. Features and Labels </font>

Every supervised learning dataset consists of two parts.

### <font color='orange'> Features </font>

The **features** are the measurable quantities describing each observation.

For the credit card fraud dataset,

| Feature | Description |
| :--- | :--- |
| Time | Time since first transaction |
| V1–V28 | Principal Components obtained using PCA |
| Amount | Transaction amount |

These variables are collectively written as

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n).
$$

---

### <font color='orange'> Labels </font>

The label represents the class assigned to each observation.

For fraud detection,

| Label | Meaning |
| :--- | :--- |
| 0 | Genuine Transaction |
| 1 | Fraudulent Transaction |

The learning algorithm attempts to discover the relationship

$$
\mathbf{x}
\longrightarrow
y.
$$

---

## <font color='orange'> 5. Decision Boundary </font>

Classification algorithms separate different classes by constructing a **decision boundary**.

In two dimensions, the decision boundary is simply a curve (or line) dividing the feature space into regions belonging to different classes.

For example,

Linear classifiers produce

$$
w_1x_1+w_2x_2+b=0,
$$

which is a straight line.

Nonlinear classifiers (Random Forests, Neural Networks, Kernel SVMs) produce much more complicated decision boundaries capable of modelling highly complex datasets.

The objective is to place the decision boundary so that future observations are classified correctly.

---

## <font color='orange'> 6. Probabilistic Classification </font>

Many modern classifiers do not simply predict a class.

Instead, they estimate the probability that an observation belongs to each class.

For binary classification,

$$
P(y=0|\mathbf{x})
+
P(y=1|\mathbf{x})
=
1.
$$

For example,

| Fraud Probability | Prediction |
| :---: | :--- |
| 0.08 | Genuine |
| 0.42 | Genuine |
| 0.91 | Fraud |

These probabilities allow the user to adjust the classification threshold depending on the application.

---

## <font color='orange'> 7. Common Classification Algorithms </font>

Many algorithms have been developed for classification.

| Algorithm | Characteristics |
| :--- | :--- |
| Logistic Regression | Simple linear probabilistic classifier |
| Support Vector Machine | Maximum-margin classifier |
| Decision Tree | Hierarchical rule-based classifier |
| Random Forest | Ensemble of decision trees |
| Gradient Boosting | Sequential ensemble learner |
| K-Nearest Neighbours | Instance-based learning |
| Naive Bayes | Probabilistic classifier based on Bayes' theorem |
| Neural Networks | Highly flexible nonlinear models |

Each algorithm constructs a different type of decision boundary and therefore performs differently depending on the dataset.

---

## <font color='orange'> 8. Challenges in Classification </font>

Real-world classification problems are rarely straightforward.

Some common challenges include:

### <font color='orange'> Class Imbalance </font>

Many datasets contain far more observations from one class than another.

For example,

| Class | Number of Samples |
| :--- | ---: |
| Genuine Transactions | 284,315 |
| Fraudulent Transactions | 492 |

A classifier trained on such a dataset may simply predict every transaction as genuine while still achieving very high accuracy.

---

### <font color='orange'> Overlapping Classes </font>

Sometimes the distributions of different classes overlap considerably, making perfect classification impossible.

---

### <font color='orange'> Noise </font>

Incorrect labels, measurement errors and missing values can all reduce classification performance.

---

### <font color='orange'> High-Dimensional Data </font>

Modern datasets often contain hundreds or thousands of features, increasing computational complexity and making feature selection increasingly important.

---

Common applications include:

## <font color='purple'> 10. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Goal | Predict a class label |
| Input | Feature vector \(\mathbf{x}\) |
| Output | Discrete category |
| Binary Classification | Two possible classes |
| Multi-Class Classification | More than two classes |
| Decision Boundary | Separates different classes |
| Probability Output | Confidence of class membership |
| Common Algorithms | Logistic Regression, SVM, Random Forest, Gradient Boosting, KNN |

> **Key Insight:** Classification is fundamentally a **decision-making problem**. Instead of predicting a numerical value, the model learns how to assign observations to one of several predefined categories by analysing patterns present in the training data. The quality of a classifier is determined not only by how often it predicts correctly, but also by its ability to generalize to previously unseen data.
