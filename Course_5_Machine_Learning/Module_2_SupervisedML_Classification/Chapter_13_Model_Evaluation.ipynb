# <font color='blue'> Partial Dependence Plots (PDP) </font>

Permutation Feature Importance (PFI) identifies which variables are important to a machine learning model, but it does not explain **how** those variables influence the model's predictions. Partial Dependence Plots (PDPs) address this limitation by visualizing the relationship between one (or two) input features and the model's predicted output.

A Partial Dependence Plot illustrates the **average effect** of changing one feature while averaging over all remaining features in the dataset. Consequently, PDPs provide valuable insight into how a machine learning model behaves globally without requiring knowledge of its internal structure.

Because PDPs rely only on the model's predictions, they are **model-agnostic** and can be applied to virtually any supervised learning algorithm.

---

## <font color='orange'> 1. Motivation </font>

Suppose we have trained an

```
XGBoost Classifier
```

to detect fraudulent credit card transactions.

Permutation Feature Importance identifies

```
Transaction Amount
```

as the most important variable.

However,

this immediately raises another question.

> **Does increasing the transaction amount increase or decrease the probability of fraud?**

Feature importance cannot answer this.

Partial Dependence Plots were developed specifically to answer such questions.

---

## <font color='orange'> 2. The Central Idea </font>

Rather than asking

> **Which feature is important?**

PDP asks

> **What happens to the model prediction when a particular feature changes?**

Suppose

```
Transaction Amount
```

takes values

```
£10

↓

£100

↓

£1000
```

For each value,

the model computes a prediction,

while averaging over

all other variables.

The resulting curve describes the average influence of

Transaction Amount

on the predicted probability of fraud.

---

## <font color='orange'> 3. The General Procedure </font>

Suppose we wish to study

```
Feature X₁
```

The procedure is

### Step 1

Choose a value

for

```
X₁.
```

---

### Step 2

Replace

every observation's value

of

```
X₁
```

with the chosen value.

---

### Step 3

Keep

all remaining variables

unchanged.

---

### Step 4

Predict

using the trained model.

---

### Step 5

Average

all predictions.

---

### Step 6

Repeat

for many values of

```
X₁.
```

The resulting curve is

the Partial Dependence Plot.

---

## <font color='orange'> 4. Mathematical Definition </font>

Suppose

the predictor variables are

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_p).
$$

We wish to investigate

feature

$$
x_s.
$$

The partial dependence function is

$$
\boxed{
PD(x_s)
=
E_{x_c}
\left[
f(x_s,x_c)
\right],
}
$$

where

* $f(\cdot)$ is the trained model,
* $x_c$ represents all remaining features,
* the expectation denotes averaging over the dataset.

Thus,

PDP estimates

the average prediction

as

one feature changes.

| £10 | 0.001 |
| £100 | 0.003 |
| £500 | 0.014 |
| £1000 | 0.052 |
| £2000 | 0.118 |

Interpretation

The model predicts

higher fraud probabilities

for increasingly large transactions.

Importantly,

this relationship reflects

the model's behaviour,

not necessarily

a causal relationship.

```
Prediction

│
│
│________
│
│
│        ________
└──────────── Feature
```

> **Key Insight:** Partial Dependence Plots extend feature importance analysis by showing **how** a feature influences a model's predictions rather than merely indicating **whether** it is important. By varying one feature while averaging predictions over the remaining features, PDPs reveal the model's average response to changes in that feature, making them particularly useful for identifying nonlinear relationships, thresholds, and monotonic trends. However, because they represent an average over the entire dataset, PDPs may conceal substantial differences between individual observations and can produce misleading results when features are strongly correlated. These limitations motivate the development of **Individual Conditional Expectation (ICE)** plots, which visualize feature effects at the level of individual observations.

# <font color='blue'> Individual Conditional Expectation (ICE) Plots </font>

Partial Dependence Plots (PDPs) provide valuable insight into the average relationship between a feature and a model's predictions. However, averaging predictions across the entire dataset can hide important differences between individual observations. In many machine learning applications, different observations may respond very differently to changes in the same feature.

Individual Conditional Expectation (ICE) plots were introduced to overcome this limitation. Rather than displaying a single average curve, an ICE plot displays **one curve for each individual observation**, illustrating how the model's prediction changes as a selected feature varies while all remaining features are held fixed.

Consequently, ICE plots provide a detailed view of individual model behaviour and reveal heterogeneous responses that are invisible in a standard Partial Dependence Plot.

---

## <font color='orange'> 1. Motivation </font>

Suppose a Partial Dependence Plot shows

```
Transaction Amount

↓

Fraud Probability
```

and indicates

that larger transactions generally have a higher predicted probability of fraud.

This is useful,

but it immediately raises another question.

> **Does every transaction behave in the same way?**

Perhaps

some customers

show almost no increase,

whereas

others exhibit a dramatic increase.

A single average curve cannot reveal these differences.

ICE plots were developed specifically to answer this question.

---

## <font color='orange'> 2. The Central Idea </font>

Instead of computing

the average prediction,

ICE computes

predictions

for

each observation individually.

Suppose we select

```
Transaction Amount.
```

For one customer,

the feature values become

```
£10

↓

£100

↓

£500

↓

£1000
```

while

every other feature

remains unchanged.

The resulting predictions form

one ICE curve.

This process is repeated

for every observation.

---

## <font color='orange'> 3. General Procedure </font>

Suppose we wish to investigate

```
Feature X₁.
```

The algorithm proceeds as follows.

### Step 1

Choose one observation.

---

### Step 2

Replace

the value of

```
X₁
```

with many different values.

---

### Step 3

Leave

all remaining variables

unchanged.

---

### Step 4

Predict

using the trained model.

---

### Step 5

Connect the predictions

to form

one ICE curve.

---

### Step 6

Repeat

for every observation.

The result is

many curves,

one for each observation.

---

## <font color='orange'> 4. Mathematical Definition </font>

Suppose

the trained model is

$$
f(\mathbf{x}),
$$

where

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_p).
$$

For observation

$$
i,
$$

the ICE function is

$$
\boxed{
ICE_i(x_s)
=
f(x_s,x_{c,i}),
}
$$

where

* $x_s$ is the feature of interest,
* $x_{c,i}$ denotes the remaining features for observation $i$.

Unlike PDP,

no averaging occurs.

Each observation therefore possesses

its own response curve.

---

## <font color='orange'> 5. PDP versus ICE </font>

Suppose we study

```
Age

↓

Disease Probability.
```

### PDP

```
One Curve

↓

Average Prediction
```

---

### ICE

```
Observation 1

↓

Curve 1
```

```
Observation 2

↓

Curve 2
```

```
Observation 3

↓

Curve 3
```

...

Each curve represents

one individual.

---

```
Prediction

│\      /
│ \    /
│  \  /
│   \/
│
└──────────── Feature
```

Interpretation

The PDP hides important variation.

---

### Crossing Curves

```
Prediction

│ \  /
│  \/
│  /\
│ /  \
└──────────── Feature
```

Interpretation

Strong interactions

between

the selected feature

and

other variables.

---

Suppose

Each curve is shifted

Only

is displayed.

This emphasizes

how the feature influences

each observation,

rather than the absolute prediction.

> **Key Insight:** Individual Conditional Expectation (ICE) plots extend Partial Dependence Plots by replacing a single average curve with one curve for every observation in the dataset. # <font color='blue'> LIME (Local Interpretable Model-Agnostic Explanations) </font>

Many modern machine learning algorithms, including Random Forests, XGBoost, CatBoost, and Deep Neural Networks, produce highly accurate predictions but often provide little insight into the reasoning behind an individual prediction. In many real-world applications, however, understanding **why a specific prediction was made** is just as important as the prediction itself.

LIME (**Local Interpretable Model-Agnostic Explanations**) is a technique introduced by **Ribeiro, Singh, and Guestrin (2016)** that explains individual predictions by approximating the complex model with a simple interpretable model in the neighbourhood of the observation being explained.

Rather than attempting to understand the entire machine learning model, LIME focuses only on a small region surrounding one observation. This local approximation allows complex models to be explained using familiar interpretable models such as linear regression or decision trees.

---

## <font color='orange'> 1. Motivation </font>

Suppose an XGBoost classifier predicts

```
Fraud Probability = 99.3%
```

for one credit card transaction.

The prediction is useful,

but a fraud analyst immediately asks

> **Why?**

Which variables caused

the model

to assign such a high probability?

Looking at

500 boosted trees

is practically impossible.

LIME was designed specifically to answer

this question.

---

## <font color='orange'> 2. The Central Idea </font>

Instead of explaining

the entire machine learning model,

LIME explains

only one prediction.

Suppose we wish to explain

Observation

```
x₀
```

LIME constructs

a simple interpretable model

that behaves similarly

to the original model

only near

```
x₀.
```

Far away from

```
x₀,

```

the approximation

is no longer expected to remain accurate.

---

## <font color='orange'> 3. Local Approximation </font>

Imagine a complicated nonlinear model.

```
Black-Box Model

↓

Highly Curved Decision Boundary
```

Now zoom into

one small neighbourhood.

```
Small Region

↓

Approximately Linear
```

Within this local region,

a simple linear model

can approximate

the behaviour of

the complex model.

This is precisely

the philosophy behind LIME.

---

## <font color='orange'> 4. The General Algorithm </font>

LIME proceeds through the following steps.

### Step 1

Choose

the observation

to explain.

---

### Step 2

Generate many

perturbed observations

near that point.

---

### Step 3

Predict

using the original machine learning model.

---

### Step 4

Assign

larger weights

to observations

closest to

the original observation.

---

### Step 5

Fit

a simple interpretable model

using

the weighted observations.

---

### Step 6

Use

the coefficients

or decision rules

of the surrogate model

as the explanation.

---

## <font color='orange'> 5. Perturbation Sampling </font>

Suppose

our original observation is

```
Age = 45

Income = £60,000

Credit Score = 720
```

LIME generates

nearby observations

such as

```
Age = 44

Income = £61,000

Credit Score = 718
```

```
Age = 46

Income = £58,500

Credit Score = 725
```

```
Age = 45

Income = £59,200

Credit Score = 721
```

These synthetic observations remain

close

to the original data point.

---

## <font color='orange'> 6. Local Weighting </font>

Not all perturbed observations

are equally informative.

Observations

very close

to

```
x₀
```

receive

larger weights.

More distant observations

receive

smaller weights.

A common weighting function is

$$
\boxed{
w(x)
=
\exp
\left(
-
\frac{D(x,x_0)^2}
{\sigma^2}
\right),
}
$$

where

* $D$ is a distance measure,
* $\sigma$ controls the neighbourhood size.

Nearby observations therefore dominate

the surrogate model.

---

## <font color='orange'> 7. The Surrogate Model </font>

LIME typically fits

a simple model

such as

* Linear Regression
* Logistic Regression
* Sparse Linear Model
* Small Decision Tree

The surrogate approximates

only

the local neighbourhood.

It is **not** intended

to replace

the original machine learning model.

---

## <font color='orange'> 8. Mathematical Formulation </font>

Suppose

the original model is

$$
f(\mathbf{x}).
$$

LIME seeks

an interpretable model

$$
g(\mathbf{x})
$$

that minimizes

$$
\boxed{
\mathcal L
(f,g,\pi_x)
+
\Omega(g),
}
$$

where

* $\mathcal L$ measures how well the surrogate approximates the black-box model locally,
* $\pi_x$ assigns larger weights to nearby observations,
* $\Omega(g)$ penalizes complex surrogate models.

Thus,

LIME simultaneously seeks

high local fidelity

and

high interpretability.

## <font color='orange'> 11. Advantages </font>

| Concept | Description |
| :--- | :--- |

> **Key Insight:** LIME explains an individual prediction by replacing the complex black-box model with a simple interpretable surrogate model in the immediate neighbourhood of the observation of interest. By generating perturbed samples, weighting nearby observations more heavily, and fitting a local linear or rule-based model, LIME identifies the features that most strongly influence a specific prediction. Rather than attempting to explain the entire machine learning model, LIME focuses exclusively on local behaviour, making it an effective and intuitive tool for understanding why a particular decision was made. However, because its explanations depend on the sampled neighbourhood and the surrogate approximation, they should be interpreted as faithful local approximations rather than exact descriptions of the original model.

# <font color='blue'> SHAP (SHapley Additive exPlanations) </font>

Modern machine learning models often achieve remarkable predictive accuracy but provide little insight into why a particular prediction was made. While methods such as LIME approximate the behaviour of a complex model locally using an interpretable surrogate, they do not guarantee that the explanation is unique, stable, or mathematically consistent.

SHAP (**SHapley Additive exPlanations**) addresses these limitations by applying concepts from **cooperative game theory** to machine learning. Introduced by **Scott Lundberg and Su-In Lee (2017)**, SHAP assigns every feature a contribution value that represents its influence on a prediction.

Unlike many earlier explanation techniques, SHAP provides both

* **Local explanations** (individual predictions),

and

* **Global explanations** (overall model behaviour),

within a single unified mathematical framework.

Today, SHAP is considered one of the most powerful and theoretically well-founded methods for interpreting machine learning models.

---

## <font color='orange'> 1. Motivation </font>

Suppose an

```
XGBoost Classifier
```

predicts

```
Fraud Probability

=

99.6%
```

The natural question becomes

> **Why?**

Perhaps

* Transaction Amount
* V14
* V12

all contributed.

But

how much

did each feature contribute?

SHAP was developed to answer

this question

fairly

and

consistently.

---

## <font color='orange'> 2. The Central Idea </font>

Imagine

every feature

is a player

participating in a cooperative game.

The "reward"

is

the model prediction.

Each player contributes

something

toward

the final outcome.

The question becomes

> **How should the reward be divided fairly among all players?**

Game theory answers this using

the

**Shapley Value**.

SHAP applies

exactly the same idea

to machine learning.

---

## <font color='orange'> 3. Cooperative Game Theory </font>

Suppose

three players

cooperate

to win

£300.

```
Player A

Player B

Player C
```

Clearly,

each player

did not necessarily contribute equally.

Shapley Values calculate

the

average marginal contribution

of every player

over

all possible orders

in which the coalition could be formed.

Machine learning simply replaces

players

with

features.

---

## <font color='orange'> 4. Features as Players </font>

Suppose we have

```
Age

Income

Credit Score
```

The prediction becomes

the "reward."

SHAP computes

how much

each feature contributes

toward moving the prediction

away from

the average prediction.

Thus,

every prediction is decomposed into

individual feature contributions.

---

## <font color='orange'> 5. The Baseline Prediction </font>

Every SHAP explanation begins with

a baseline.

The baseline is

the expected model prediction

over

the training data.

For example,

suppose

```
Average Fraud Probability

=

0.003
```

This is

the starting point.

Individual features then

increase

or

decrease

this value.

---

## <font color='orange'> 6. Additive Explanation </font>

Suppose

the baseline prediction is

$$
E[f(x)].
$$

SHAP expresses

every prediction

as

$$
\boxed{
f(x)
=
E[f(x)]
+
\sum_{i=1}^{p}
\phi_i,
}
$$

where

* $E[f(x)]$ is the baseline prediction,
* $\phi_i$ is the SHAP value of feature $i$.

Thus,

each prediction equals

the baseline

plus

the contributions

from every feature.

---

## <font color='orange'> 7. What is a SHAP Value? </font>

A SHAP value represents

how much

one feature

changes

the prediction.

Positive SHAP value

↓

Increases

the prediction.

Negative SHAP value

↓

Decreases

the prediction.

Zero

↓

No contribution.

For example,

| Feature | SHAP Value |
| :--- | ---: |
| Transaction Amount | +0.28 |
| V14 | +0.19 |
| V12 | +0.15 |
| Time | -0.02 |

The fraud probability increases because

Transaction Amount,

V14,

and

V12

push

the prediction upward.

---

## <font color='orange'> 8. Computing Shapley Values </font>

Suppose we have

three features.

The contribution of

Feature A

depends upon

whether

it joins

the coalition

before

or

after

Features B and C.

Therefore,

SHAP computes

the feature contribution

for

every possible ordering

of the features.

The final SHAP value

is

the average marginal contribution

across

all permutations.

This guarantees

a fair allocation

of importance.

---

## <font color='orange'> 9. Desirable Properties </font>

SHAP satisfies several important mathematical properties.

### Local Accuracy

The feature contributions

exactly sum

to

the prediction.

---

### Consistency

If a feature becomes

more influential,

its SHAP value

cannot decrease.

---

### Missingness

A feature

that contributes nothing

receives

a SHAP value of zero.

These properties distinguish SHAP

from many earlier explanation techniques.

---

## <font color='orange'> 10. Local Explanations </font>

SHAP explains

one prediction.

Example

```
Fraud Probability

=

98%
```

Explanation

```
Baseline

↓

0.3%

↓

Amount

+22%

↓

V14

+41%

↓

V12

+35%

↓

Prediction

98%
```

This provides

a complete explanation

of one observation.

---

## <font color='orange'> 11. Global Explanations </font>

SHAP also explains

the entire model.

Global feature importance

is obtained

by averaging

the absolute SHAP values

over

all observations.

Large average SHAP values

indicate

globally important variables.

Thus,

one framework provides

both

local

and

global interpretation.

## <font color='orange'> 13. TreeSHAP </font>

Computing

exact Shapley Values

requires evaluating

all possible feature subsets,

which becomes computationally infeasible

for large numbers of features.

Tree-based models,

however,

permit

an efficient algorithm

known as

**TreeSHAP**.

TreeSHAP computes

exact SHAP values

for

Decision Trees,

Random Forests,

XGBoost,

LightGBM,

and

CatBoost

in polynomial time,

making SHAP practical for modern ensemble methods.

## <font color='orange'> 15. Limitations </font>

Despite its strengths,

SHAP also has limitations.

### Computational Cost

Exact SHAP values

become expensive

for general machine learning models.

---

### Correlated Features

Highly correlated variables

may share

their contributions,

making interpretation

more difficult.

---

### Interpretation

SHAP explains

how the model behaves,

not

causal relationships.

A large SHAP value

does **not** imply

that changing the feature

will necessarily change

the real-world outcome.

## <font color='purple'> 19. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| SHAP | SHapley Additive exPlanations |
| Foundation | Cooperative Game Theory |
| Main Idea | Fairly distribute the prediction among the features |
| Local Explanation | Explains one prediction |
| Global Explanation | Average absolute SHAP values |
| Mathematical Property | Prediction = Baseline + Sum of SHAP Values |
| Specialized Algorithm | TreeSHAP for tree-based models |
| Main Strength | Consistent, theoretically grounded feature attribution |

> **Key Insight:** SHAP interprets a machine learning prediction by viewing each feature as a player in a cooperative game and the model's prediction as the reward generated by their combined contribution. Using Shapley values from game theory, SHAP fairly allocates the difference between the baseline prediction and the final prediction among all input features. This additive decomposition provides intuitive local explanations for individual predictions while also enabling global interpretation through the aggregation of feature contributions across the dataset. Because SHAP satisfies desirable mathematical properties such as local accuracy, consistency, and missingness, and because efficient algorithms such as TreeSHAP exist for tree-based models, it has become the preferred explanation framework for modern machine learning systems.
