# <font color='blue'> XGBoost (Extreme Gradient Boosting) </font>

XGBoost, short for **Extreme Gradient Boosting**, is one of the most successful and widely used machine learning algorithms for structured (tabular) data. Developed by **Tianqi Chen** in 2014, XGBoost extends the Gradient Boosting framework by introducing regularization, second-order optimization, efficient tree construction, and parallel computation.

Although Gradient Boosting already provides excellent predictive performance, it can be computationally expensive and prone to overfitting. XGBoost addresses these limitations through a series of algorithmic and engineering improvements, making it significantly faster, more scalable, and often more accurate.

Today, XGBoost is widely used in finance, healthcare, engineering, scientific research, and has been the foundation of numerous winning solutions in machine learning competitions.

---



## <font color='orange'> 1. Motivation </font>

Recall that Gradient Boosting constructs an additive model

$$
F_M(x)=F_{M-1}(x)+\gamma h_M(x),
$$

where each new Decision Tree learns the residual errors of the current ensemble.

Although highly effective,

Gradient Boosting has several limitations.

* Training is computationally expensive.
* Trees may overfit.
* Large datasets require considerable memory.
* Hyperparameter tuning becomes increasingly important.

XGBoost was specifically designed to overcome these limitations while preserving the fundamental principles of Gradient Boosting.

---



## <font color='orange'> 2. The Central Idea </font>

The philosophy of XGBoost is simple.

> **Build Gradient Boosting models more accurately, more efficiently, and with better generalization.**

It achieves this by introducing

* Regularization,
* Second-order optimization,
* Efficient split finding,
* Parallel tree construction,
* Missing value handling,
* Tree pruning.

The underlying boosting framework remains unchanged.

---



## <font color='orange'> 3. Additive Model </font>

Like Gradient Boosting,

XGBoost builds an additive model.

After

$$
M
$$

iterations,

the prediction becomes

$$
\boxed{
\hat y
=
\sum_{m=1}^{M}
f_m(x),
}
$$

where

each

$$
f_m(x)
$$

is a Decision Tree.

Each successive tree corrects the remaining prediction errors.

---



## <font color='orange'> 4. Regularized Objective Function </font>

One of the most important innovations introduced by XGBoost is the addition of a regularization term.

Instead of minimizing only the prediction loss,

XGBoost minimizes

$$
\boxed{
\mathcal L
=
\sum_i
L(y_i,\hat y_i)
+
\sum_k
\Omega(f_k),
}
$$

where

* $L$ is the prediction loss,
* $\Omega$ penalizes overly complex trees.

This regularization reduces overfitting and improves generalization.

---



## <font color='orange'> 5. Tree Complexity Penalty </font>

The regularization term is

$$
\boxed{
\Omega(f)
=
\gamma T
+
\frac{\lambda}{2}
\sum_j w_j^2,
}
$$

where

* $T$ is the number of leaf nodes,
* $w_j$ is the prediction value assigned to leaf $j$,
* $\gamma$ penalizes additional leaves,
* $\lambda$ controls L2 regularization.

The algorithm therefore prefers simpler trees unless more complex trees produce a sufficiently large improvement in prediction accuracy.

---



## <font color='orange'> 6. Second-Order Optimization </font>

Ordinary Gradient Boosting uses only the

**first derivative**

of the loss function.

XGBoost additionally uses the

**second derivative (Hessian).**

Using a second-order Taylor expansion,

the objective function becomes

$$
\boxed{
\mathcal L
\approx
\sum_i
\left[
g_i f(x_i)
+
\frac12
h_i
f(x_i)^2
\right]
+
\Omega(f),
}
$$

where

* $g_i$ is the gradient,
* $h_i$ is the Hessian.

Using both derivatives allows more accurate optimization and faster convergence.

---



## <font color='orange'> 7. Tree Pruning </font>

Unlike ordinary Gradient Boosting,

XGBoost first grows a large tree,

then removes branches that provide insufficient improvement.

This process is known as

**post-pruning**.

A split is retained only if

its reduction in the objective function exceeds

$$
\gamma.
$$

This strategy generally produces smaller,

better-generalizing trees.

---



## <font color='orange'> 8. Handling Missing Values </font>

Many datasets contain missing observations.

Rather than requiring imputation beforehand,

XGBoost automatically learns

the optimal direction for missing values at every split.

Consequently,

missing values can often be handled directly during training.

---



## <font color='orange'> 9. Parallel Processing </font>

Although trees are still constructed sequentially,

many operations within each tree can be parallelized.

Examples include

* split evaluation,
* histogram construction,
* feature scanning.

Consequently,

XGBoost trains substantially faster than traditional Gradient Boosting on modern multicore processors.

---



## <font color='orange'> 10. Shrinkage (Learning Rate) </font>

Like Gradient Boosting,

XGBoost updates the model using

$$
\boxed{
F_m(x)
=
F_{m-1}(x)
+
\eta
f_m(x),
}
$$

where

$$
\eta
$$

is the learning rate.

Smaller learning rates generally

* improve generalization,
* reduce overfitting,
* require more trees.

---



## <font color='orange'> 11. Column Sampling </font>

Another important innovation is

random feature sampling.

Instead of evaluating every predictor,

XGBoost may randomly select only a subset.

This is conceptually similar to Random Forests,

reducing correlation between trees and improving robustness.

Common parameters include

```python
colsample_bytree

colsample_bylevel

colsample_bynode
```

# <font color='blue'> LightGBM (Light Gradient Boosting Machine) </font>

LightGBM (Light Gradient Boosting Machine) is an advanced Gradient Boosting framework developed by **Microsoft Research** in 2017. It was specifically designed to address the computational limitations of traditional Gradient Boosting and XGBoost when working with very large datasets containing millions of observations and thousands of features.

Although LightGBM follows the same additive boosting framework as Gradient Boosting and XGBoost, it introduces several important algorithmic innovations that dramatically reduce computational cost while maintaining excellent predictive performance.

Its principal innovations include

* Histogram-based learning,
* Leaf-wise tree growth,
* Gradient-based One-Side Sampling (GOSS),
* Exclusive Feature Bundling (EFB).

These improvements make LightGBM one of the fastest and most memory-efficient Gradient Boosting implementations available today.

---



## <font color='orange'> 1. Motivation </font>

Although XGBoost significantly improves upon traditional Gradient Boosting, training can still become computationally expensive when

* the dataset contains millions of observations,
* there are thousands of predictor variables,
* many boosting iterations are required.

For example,

```
10 million observations

×

500 features
```

Evaluating every possible split for every feature quickly becomes computationally demanding.

LightGBM was designed to solve this problem without sacrificing predictive accuracy.

---



## <font color='orange'> 2. The Central Idea </font>

LightGBM accelerates Gradient Boosting by improving

* how data are represented,
* how trees are constructed,
* how observations are sampled,
* how features are stored.

Rather than changing the underlying boosting philosophy,

LightGBM optimizes nearly every computational step.

---



## <font color='orange'> 3. Histogram-Based Learning </font>

Traditional Decision Trees evaluate every possible split.

Suppose a feature contains

```
10,000

different values.
```

Evaluating every threshold becomes expensive.

LightGBM instead groups feature values into

**histogram bins.**

For example,

```
Original Values

↓

10,000 values

↓

255 histogram bins
```

The algorithm searches only among the histogram boundaries.

Consequently,

split evaluation becomes dramatically faster.

---



## <font color='orange'> 4. Why Histograms are Faster </font>

Suppose a feature contains

```
1

2

3

...

10000
```

Instead of evaluating

9,999

possible split locations,

LightGBM may first create

```
255 bins
```

Only

254

candidate split locations are then examined.

The computational complexity is reduced substantially while maintaining nearly identical predictive performance.

---



## <font color='orange'> 5. Leaf-Wise Tree Growth </font>

Perhaps the most distinctive feature of LightGBM is its

**leaf-wise tree growth strategy.**

Traditional algorithms such as XGBoost typically grow trees

```
Level-by-Level.
```

```
Root

↓

Two Children

↓

Four Children

↓

Eight Children
```

Every level is expanded uniformly.

---

LightGBM instead grows

```
Leaf-by-Leaf.
```

Only the leaf producing

the largest reduction in the loss function

is expanded.

```
Root

↓

Best Leaf

↓

Best Leaf

↓

Best Leaf
```

The tree therefore becomes asymmetric.

This often produces

lower prediction error

using fewer nodes.

---



## <font color='orange'> 6. Level-Wise vs Leaf-Wise Growth </font>

| Property | Level-Wise | Leaf-Wise |
| :--- | :---: | :---: |
| Tree Shape | Symmetric | Asymmetric |
| Speed | Moderate | Faster |
| Accuracy | Good | Often Higher |
| Risk of Overfitting | Lower | Higher |

Leaf-wise growth generally achieves better predictive performance,

although careful regularization is required to avoid overfitting.

---



## <font color='orange'> 7. Gradient-Based One-Side Sampling (GOSS) </font>

During boosting,

not every observation contributes equally.

Many observations already possess

very small residual errors.

These observations provide relatively little new information.

LightGBM therefore keeps

* observations with **large gradients**,

while randomly sampling only a fraction of observations with

* **small gradients**.

This strategy is called

**Gradient-Based One-Side Sampling (GOSS).**

The algorithm focuses computation on the observations that remain difficult to predict.

---



## <font color='orange'> 8. Exclusive Feature Bundling (EFB) </font>

Many high-dimensional datasets contain sparse features.

Often,

two features are

never simultaneously non-zero.

For example,

```
Feature A

1 0 0 0
```

```
Feature B

0 1 0 0
```

Since these variables never overlap,

LightGBM combines them into

a single bundled feature.

This technique,

called

**Exclusive Feature Bundling (EFB),**

reduces memory usage while preserving information.

---



## <font color='orange'> 9. Mathematical Framework </font>

Like Gradient Boosting,

LightGBM minimizes the regularized objective

$$
\boxed{
\mathcal L
=
\sum_i
L(y_i,\hat y_i)
+
\Omega(f),
}
$$

where

$$
\Omega(f)
$$

penalizes model complexity.

The optimization procedure therefore remains identical to XGBoost,

only the computational implementation differs.

# <font color='blue'> CatBoost (Categorical Boosting) </font>

CatBoost (Categorical Boosting) is a Gradient Boosting framework developed by **Yandex** and released in 2018. Unlike most machine learning algorithms, which require categorical variables to be manually encoded before training, CatBoost was specifically designed to handle categorical data directly.

Traditional encoding techniques, such as Label Encoding and One-Hot Encoding, often increase dimensionality, introduce artificial ordering, or create information leakage. CatBoost addresses these problems through two major innovations:

* **Ordered Target Encoding**
* **Ordered Boosting**

These techniques allow CatBoost to achieve excellent predictive performance while reducing overfitting and improving generalization, particularly on datasets containing many categorical variables.

---



## <font color='orange'> 1. Motivation </font>

Many real-world datasets contain categorical variables.

Examples include

* Gender
* Country
* Occupation
* Product Category
* Customer Segment

Unlike numerical variables,

these values possess

**no natural numerical ordering.**

For example,

```
Country

↓

UK

France

Germany
```

Assigning

```
UK = 1

France = 2

Germany = 3
```

incorrectly implies

$$
Germany > France > UK,
$$

which has no statistical meaning.

---



## <font color='orange'> 2. Why Are Categorical Variables Difficult? </font>

Most machine learning algorithms require numerical inputs.

Therefore,

categorical variables must usually be transformed.

Common approaches include

### Label Encoding

```
Red → 0

Blue → 1

Green → 2
```

Problem:

Artificial ordering.

---

### One-Hot Encoding

```
Red

↓

[1,0,0]
```

```
Blue

↓

[0,1,0]
```

Problem:

Produces many additional features.

High-cardinality variables may generate thousands of sparse columns.

---



## <font color='orange'> 3. Target Encoding </font>

A more informative strategy is

**Target Encoding.**

Suppose

```
City
```

contains

```
London

Manchester

Leeds
```

Rather than assigning arbitrary numbers,

each category is replaced by

its average target value.

For example,

| City | Fraud Rate |
| :--- | ---: |
| London | 0.031 |
| Manchester | 0.012 |
| Leeds | 0.024 |

This often provides much better predictive performance.

---



## <font color='orange'> 4. The Information Leakage Problem </font>

Ordinary Target Encoding introduces a serious problem.

Suppose

```
London
```

appears in the current observation.

Its encoded value is computed using

the target value of

that same observation.

The model therefore indirectly observes

the answer

during training.

This phenomenon is called

**Target Leakage**

or

**Prediction Shift.**

It often leads to unrealistically optimistic performance estimates.

---



## <font color='orange'> 5. Ordered Target Encoding </font>

CatBoost solves this problem using

**Ordered Target Encoding.**

Rather than computing category statistics using

all observations,

CatBoost processes the dataset in a random order.

For each observation,

the encoded value is computed using

**only previous observations.**

```
Observation 1

↓

No previous information
```

```
Observation 2

↓

Uses Observation 1
```

```
Observation 3

↓

Uses Observations 1–2
```

The current observation never contributes to its own encoded value.

Information leakage is therefore eliminated.

---



## <font color='orange'> 6. Ordered Boosting </font>

Traditional Gradient Boosting computes residuals using

models trained on

the complete dataset.

This introduces another subtle form of prediction bias.

CatBoost instead performs

**Ordered Boosting.**

Each prediction is generated using

only models trained on

earlier observations.

Consequently,

every prediction resembles

true unseen data.

This substantially reduces overfitting.

---



## <font color='orange'> 7. Symmetric Trees </font>

Unlike XGBoost and LightGBM,

CatBoost constructs

**Oblivious (Symmetric) Decision Trees.**

Every level uses

the same splitting feature.

For example,

```
Root

↓

Income
```

```
Level 2

↓

Age

Age
```

Every node at the same depth performs

the same split.

The resulting trees are

* balanced,
* computationally efficient,
* less prone to overfitting.

---



## <font color='orange'> 8. Mathematical Framework </font>

Like Gradient Boosting,

CatBoost minimizes

$$
\boxed{
\mathcal L
=
\sum_i
L(y_i,\hat y_i)
+
\Omega(f),
}
$$

where

$$
\Omega(f)
$$

regularizes tree complexity.

The boosting framework therefore remains identical to Gradient Boosting,

while the treatment of categorical variables differs substantially.
