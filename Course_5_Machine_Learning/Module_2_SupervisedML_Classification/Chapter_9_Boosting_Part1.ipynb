# <font color='blue'> AdaBoost (Adaptive Boosting) </font>

AdaBoost, short for **Adaptive Boosting**, is one of the earliest and most influential ensemble learning algorithms. Introduced by **Yoav Freund and Robert Schapire (1997)**, AdaBoost constructs a strong classifier by sequentially combining many weak learners.

Unlike Bagging, where every model contributes equally, AdaBoost adapts the learning process by assigning greater emphasis to observations that are difficult to classify. At each iteration, incorrectly classified observations receive larger weights, encouraging the next learner to focus its attention on these challenging cases.

The final prediction is obtained through a **weighted vote**, where more accurate learners receive greater influence.

---



## <font color='orange'> 1. Motivation </font>

Suppose we train a Decision Stump to classify fraudulent transactions.

```
Training Data

↓

Decision Stump

↓

92 Correct

8 Incorrect
```

The eight incorrectly classified observations represent the difficult cases.

Rather than ignoring these mistakes,

AdaBoost increases their importance.

The next learner is therefore encouraged to classify these observations correctly.

Over successive iterations,

the ensemble gradually improves its performance.

---



## <font color='orange'> 2. Why "Adaptive"? </font>

The name **Adaptive** refers to the fact that the algorithm continuously adapts to previous mistakes.

Initially,

every observation is assigned equal importance.

```
Observation

↓

Weight

↓

1/N
```

After each learner,

the weights are updated.

Incorrectly classified observations receive larger weights,

while correctly classified observations receive smaller weights.

Consequently,

future learners concentrate on the most difficult observations.

---



## <font color='orange'> 3. Weak Learners </font>

AdaBoost usually employs

**Decision Stumps**

as its base learners.

A Decision Stump contains only one split.

```
Root

↓

Leaf      Leaf
```

Although a stump is a very weak classifier,

combining hundreds of them produces an extremely powerful model.

---



## <font color='orange'> 4. Initial Sample Weights </font>

Suppose the training dataset contains

$$
N
$$

observations.

Initially,

every observation receives equal weight,

$$
\boxed{
w_i=\frac1N,
}
$$

where

$$
i=1,\ldots,N.
$$

Thus,

all observations contribute equally to the first learner.

---



## <font color='orange'> 5. Training the First Weak Learner </font>

The first Decision Stump is trained using the weighted dataset.

Initially,

all weights are equal,

so this is equivalent to ordinary Decision Tree training.

After fitting,

the stump produces predictions

```
Correct

Correct

Incorrect

Correct

Incorrect
```

The misclassified observations are identified for the next stage.

---



## <font color='orange'> 6. Weighted Classification Error </font>

Unlike ordinary classification,

AdaBoost evaluates the **weighted error**

rather than the ordinary error rate.

The weighted error is

$$
\boxed{
\epsilon
=
\sum_{i=1}^{N}
w_i
I(y_i\neq\hat y_i),
}
$$

where

* $w_i$ is the weight of observation $i$,
* $I(\cdot)$ is the indicator function.

Only incorrectly classified observations contribute to the error.

---



## <font color='orange'> 7. Learner Weight (\(\alpha\)) </font>

After computing the weighted error,

AdaBoost assigns the learner a weight

$$
\alpha.
$$

The learner weight is

$$
\boxed{
\alpha
=
\frac12
\ln
\left(
\frac{1-\epsilon}{\epsilon}
\right).
}
$$

This equation has several important consequences.

### Small Error

If

$$
\epsilon
\approx
0,
$$

then

$$
\alpha
$$

becomes large.

The learner receives a large vote.

---

### Large Error

If

$$
\epsilon
=
0.5,
$$

then

$$
\alpha
=
0.
$$

The learner contributes nothing,

since it performs no better than random guessing.

---



## <font color='orange'> 8. Updating Sample Weights </font>

The observation weights are updated using

$$
\boxed{
w_i
\leftarrow
w_i
\exp
(-\alpha y_i h_i),
}
$$

where

* $y_i$ is the true label,
* $h_i$ is the prediction of the learner.

In practice,

this means

### Correct Classification

Weight decreases.

---

### Incorrect Classification

Weight increases.

Thus,

future learners focus increasingly on difficult observations.

---



## <font color='orange'> 9. Weight Normalization </font>

After updating,

the weights no longer sum to one.

Therefore,

AdaBoost normalizes them,

$$
\boxed{
w_i
\leftarrow
\frac{w_i}
{\sum_j w_j}.
}
$$

The weights once again represent a probability distribution.

---



## <font color='orange'> 10. Repeating the Process </font>

The algorithm repeats

```
Train Weak Learner

↓

Compute Error

↓

Compute α

↓

Update Sample Weights

↓

Normalize

↓

Repeat
```

Each new learner concentrates increasingly on observations that previous learners found difficult.

---



## <font color='orange'> 11. Final Prediction </font>

After

$$
M
$$

iterations,

the final prediction is obtained using a weighted vote.

$$
\boxed{
F(x)
=
\operatorname{sign}
\left(
\sum_{m=1}^{M}
\alpha_m
h_m(x)
\right).
}
$$

Unlike Bagging,

every learner contributes with a different weight.

More accurate learners exert greater influence.

---



## <font color='orange'> 12. Intuition </font>

Imagine a classroom.

The teacher asks a question.

Most students answer correctly.

The teacher then spends additional time helping the students who answered incorrectly.

After learning,

the class performs better.

AdaBoost follows the same philosophy.

Each learner devotes increasing attention to the observations that remain difficult.

---

## <font color='purple'> 18. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| AdaBoost | Adaptive Boosting |
| Base Learner | Usually Decision Stump |
| Training Strategy | Sequential |
| Sample Weights | Updated after every iteration |
| Learner Weight | Depends on weighted error |

| Primary Objective | Reduce bias by correcting previous mistakes |

> **Key Insight:** AdaBoost constructs a powerful classifier by sequentially combining many weak learners, typically Decision Stumps. Unlike Bagging, which trains models independently, AdaBoost continuously adapts to previous prediction errors by increasing the importance of misclassified observations and decreasing the importance of correctly classified ones. Each learner is assigned a weight based on its classification accuracy, allowing stronger learners to contribute more heavily to the final prediction. Through this adaptive weighting mechanism, AdaBoost transforms a collection of simple models into a highly accurate ensemble while illustrating the fundamental principle underlying all modern boosting algorithms.

# <font color='blue'> Gradient Boosting </font>

Gradient Boosting is one of the most powerful ensemble learning algorithms for both classification and regression. Proposed by **Jerome Friedman (1999, 2001)**, Gradient Boosting extends the fundamental idea of Boosting by viewing model construction as a numerical optimization problem.

Unlike AdaBoost, which modifies the weights assigned to training observations, Gradient Boosting constructs an additive model by **iteratively minimizing a differentiable loss function**. At each iteration, a new weak learner is trained to predict the errors made by the current ensemble.

Consequently, each successive learner improves the overall model by correcting the remaining prediction errors, gradually converging toward the optimal predictor.

---



## <font color='orange'> 1. Motivation </font>

Suppose we wish to predict house prices.

Our first model produces

| House | True Price | Prediction |
| :--- | ---: | ---: |
| A | £300,000 | £280,000 |
| B | £450,000 | £470,000 |
| C | £350,000 | £330,000 |

Although the model captures the general trend,

prediction errors remain.

Instead of constructing an entirely new model,

Gradient Boosting asks

> **Can we build another model that predicts these remaining errors?**

If the errors can be predicted,

they can also be corrected.

---



## <font color='orange'> 2. The Central Idea </font>

Gradient Boosting builds models sequentially.

```
Model 1

↓

Prediction

↓

Residual Errors

↓

Model 2 learns residuals

↓

Updated Prediction

↓

New Residuals

↓

Model 3 learns remaining residuals

↓

...
```

Each learner is responsible only for the information that previous learners failed to capture.

---



## <font color='orange'> 3. Residual Errors </font>

Suppose the target value is

$$
y_i,
$$

and the current prediction is

$$
\hat{y}_i.
$$

The residual is

$$
\boxed{
r_i
=
y_i
-
\hat y_i.
}
$$

The residual measures how much prediction error remains.

Positive residual

↓

Prediction too small.

Negative residual

↓

Prediction too large.

Zero residual

↓

Perfect prediction.

---



## <font color='orange'> 4. Building an Additive Model </font>

Suppose

$$
F_0(x)
$$

is the initial model.

Gradient Boosting updates the model iteratively.

After one iteration,

$$
\boxed{
F_1(x)
=
F_0(x)
+
\gamma_1 h_1(x),
}
$$

where

* $h_1(x)$ is a weak learner,
* $\gamma_1$ is its contribution.

After

$$
M
$$

iterations,

the complete model becomes

$$
\boxed{
F_M(x)
=
F_0(x)
+
\sum_{m=1}^{M}
\gamma_m h_m(x).
}
$$

Thus,

Gradient Boosting gradually improves the prediction through a sequence of small corrections.

---



## <font color='orange'> 5. Why is it Called "Gradient" Boosting? </font>

The objective of machine learning is to minimize a loss function,

$$
L(y,\hat y).
$$

Rather than directly minimizing

$$
L,
$$

Gradient Boosting moves in the direction of

the **negative gradient**,

which represents the direction of greatest decrease.

Mathematically,

the pseudo-residuals become

$$
\boxed{
r_i^{(m)}
=
-
\left[
\frac{\partial
L(y_i,F(x_i))}
{\partial F(x_i)}
\right].
}
$$

Each new Decision Tree is trained to predict these pseudo-residuals.

This is why the algorithm is called

**Gradient Boosting**.

---



## <font color='orange'> 6. The Optimization Perspective </font>

Recall gradient descent from optimization.

```
Loss Function

↓

Compute Gradient

↓

Move Downhill

↓

Repeat
```

Gradient Boosting follows exactly the same philosophy.

Instead of updating

model parameters,

it updates

the prediction function itself.

Each weak learner represents one optimization step toward minimizing the loss function.

---



## <font color='orange'> 7. Weak Learners </font>

Like AdaBoost,

Gradient Boosting typically uses

small Decision Trees.

These trees usually have

```
Depth = 3
```

or

```
Depth = 4.
```

Unlike AdaBoost,

Decision Stumps are not always optimal.

Slightly deeper trees often model residuals more effectively.

---



## <font color='orange'> 8. The Gradient Boosting Algorithm </font>

The algorithm proceeds as follows.

### Step 1

Initialize the model

$$
F_0(x).
$$

For regression,

this is usually

the mean target value.

---

### Step 2

Compute residuals

$$
r_i^{(m)}.
$$

---

### Step 3

Fit a Decision Tree to the residuals.

---

### Step 4

Determine the optimal contribution

$$
\gamma_m.
$$

---

### Step 5

Update the model

$$
F_m(x)
=
F_{m-1}(x)
+
\gamma_m h_m(x).
$$

---

### Step 6

Repeat until

* maximum iterations,

or

* convergence.

---



## <font color='orange'> 9. Learning Rate (Shrinkage) </font>

A key improvement over AdaBoost is the

**Learning Rate**.

Instead of adding the entire correction,

Gradient Boosting adds only a fraction.

$$
\boxed{
F_m(x)
=
F_{m-1}(x)
+
\eta
\gamma_m h_m(x),
}
$$

where

$$
0<\eta<1
$$

is the learning rate.

Smaller learning rates

* improve generalization,
* reduce overfitting,
* usually require more trees.

---



## <font color='orange'> 10. Loss Functions </font>

One major advantage of Gradient Boosting is that it can optimize many different loss functions.

Regression

* Mean Squared Error
* Mean Absolute Error
* Huber Loss

Classification

* Logistic Loss
* Exponential Loss
* Deviance

This flexibility explains why Gradient Boosting applies to a wide variety of problems.

## <font color='orange'> 16. AdaBoost vs Gradient Boosting </font>

| Property | AdaBoost | Gradient Boosting |
| :--- | :---: | :---: |
| Learns | Sample weights | Residuals / gradients |
| Optimization | Exponential weighting | Gradient descent |
| Weak Learners | Decision Stumps | Small Decision Trees |
| Learning Rate | Optional | Fundamental |
| Loss Function | Exponential Loss | Many differentiable losses |
| Flexibility | Moderate | Very High |

Gradient Boosting may therefore be viewed as a generalized optimization framework,

where AdaBoost is a special case.

## <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Gradient Boosting | Sequential optimization algorithm |
| Weak Learners | Small Decision Trees |
| Learning Strategy | Predict residual errors |
| Objective | Minimize a differentiable loss function |
| Optimization | Gradient Descent in function space |
| Learning Rate | Controls contribution of each learner |
| Modern Extensions | XGBoost, LightGBM, CatBoost |

> **Key Insight:** Gradient Boosting views ensemble learning as an optimization problem. Rather than reweighting observations, each new weak learner is trained to predict the residual errors—or more generally, the negative gradient of a differentiable loss function—produced by the current ensemble. These learners are added sequentially, gradually refining the prediction function through a series of small corrective updates. By combining additive modelling with gradient-based optimization, Gradient Boosting provides a flexible framework capable of optimizing a wide variety of loss functions, forming the theoretical foundation of modern algorithms such as **XGBoost**, **LightGBM**, and **CatBoost**.
