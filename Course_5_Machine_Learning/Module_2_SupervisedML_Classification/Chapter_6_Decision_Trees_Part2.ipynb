# <font color='blue'> Overfitting and Underfitting in Decision Trees </font>

A Decision Tree should not simply memorize the training data—it should learn the underlying relationships that allow it to make accurate predictions for previously unseen observations. Achieving this objective requires an appropriate balance between **model complexity** and **generalization**.

If a Decision Tree is too simple, it may fail to capture important patterns within the data, resulting in **underfitting**. Conversely, if the tree becomes excessively complex, it may begin modelling random fluctuations and statistical noise, leading to **overfitting**.

Understanding these two phenomena is essential for constructing Decision Trees that perform well on both training and testing datasets.

---



## <font color='orange'> 1. Generalization </font>

The primary objective of supervised learning is **generalization**.

Rather than simply producing accurate predictions for the training data,

the model should also perform well on

* validation data,
* testing data,
* future unseen observations.

A Decision Tree that generalizes well has learned the underlying structure of the data instead of memorizing individual training examples.

---



## <font color='orange'> 2. Underfitting </font>

A Decision Tree **underfits** when it is too simple to capture the true relationship between the input variables and the target variable.

Typical causes include

* very small tree depth,
* excessive pruning,
* overly restrictive stopping criteria.

For example,

```python
max_depth = 1
```

produces only a single decision rule.

Such a model cannot represent complex nonlinear relationships.

---

### Characteristics of Underfitting

* High training error.
* High testing error.
* High bias.
* Low variance.
* Poor predictive performance.

The model fails to learn sufficient information from the training data.

---



## <font color='orange'> 3. Example of Underfitting </font>

Suppose the true relationship is highly nonlinear,

but the Decision Tree consists of only one split.

```
Root

↓

Leaf      Leaf
```

Regardless of the amount of available data,

the model remains incapable of representing the underlying complexity.

The predictions therefore remain inaccurate for both the training and testing datasets.

---



## <font color='orange'> 4. Overfitting </font>

A Decision Tree **overfits** when it becomes excessively complex.

Instead of learning the true underlying relationship,

the tree begins modelling

* measurement noise,
* random fluctuations,
* unusual observations,
* outliers.

Training accuracy becomes extremely high,

but testing accuracy decreases.

---

### Characteristics of Overfitting

* Very low training error.
* High testing error.
* Low bias.
* High variance.
* Poor generalization.

Although the model appears highly accurate during training,

its predictive performance on unseen data deteriorates.

---



## <font color='orange'> 5. Example of Overfitting </font>

Suppose a tree is allowed to grow without restriction.

```
Root

↓

Many Splits

↓

Many Splits

↓

Hundreds of Leaves
```

Eventually,

each leaf may contain only one or two observations.

The tree has effectively memorized the training dataset.

Consequently,

small changes in the data may produce entirely different trees.

---



## <font color='orange'> 6. Comparing Underfitting and Overfitting </font>

| Property | Underfitting | Overfitting |
| :--- | :--- | :--- |
| Model Complexity | Too Low | Too High |
| Training Error | High | Very Low |
| Testing Error | High | High |
| Bias | High | Low |
| Variance | Low | High |
| Generalization | Poor | Poor |

The objective is to avoid both extremes.

---



## <font color='orange'> 7. Error Curves </font>

As tree complexity increases,

the training error decreases continuously,

whereas the testing error initially decreases before eventually increasing.

```
Prediction Error

^

|

|              Testing Error

|                 /\

|                /  \

|               /    \

|              /      \

|-------------/--------\------------>

         Training Error

          Tree Complexity
```

The minimum of the testing error corresponds to the optimal model complexity.

---



## <font color='orange'> 8. Causes of Underfitting </font>

Common causes include

* very small `max_depth`,
* large `min_samples_split`,
* large `min_samples_leaf`,
* excessive pruning,
* insufficient predictor variables.

In these situations,

the tree lacks sufficient flexibility to represent the underlying relationship.

---



## <font color='orange'> 9. Causes of Overfitting </font>

Common causes include

* unrestricted tree growth,
* extremely deep trees,
* very small leaf nodes,
* noisy datasets,
* outliers,
* insufficient training data.

The tree begins modelling random variation rather than meaningful patterns.

---



## <font color='orange'> 10. Detecting Underfitting and Overfitting </font>

The easiest method is to compare

* training performance,

with

* validation or testing performance.

### Underfitting

```
Training Accuracy

↓

Low

Testing Accuracy

↓

Low
```

---

### Overfitting

```
Training Accuracy

↓

Very High

Testing Accuracy

↓

Significantly Lower
```

A large performance gap is often an indication of overfitting.

---



## <font color='orange'> 11. Preventing Underfitting </font>

Possible strategies include

* increasing `max_depth`,
* reducing pruning,
* lowering `min_samples_leaf`,
* lowering `min_samples_split`,
* adding informative features.

These changes increase the flexibility of the model.

---



## <font color='orange'> 12. Preventing Overfitting </font>

Several strategies help reduce overfitting.

### Restrict Tree Growth

Limit

```python
max_depth
```

---

### Increase Leaf Size

Increase

```python
min_samples_leaf
```

---

### Increase Split Requirements

Increase

```python
min_samples_split
```

---

### Apply Pruning

Remove branches that contribute little predictive value.

---

### Use Cross-Validation

Select hyperparameters that minimize validation error rather than training error.

---

### Ensemble Methods

Algorithms such as

* Random Forests,
* Gradient Boosting,

greatly reduce overfitting while maintaining high predictive performance.

---



## <font color='orange'> 13. Example in Scikit-Learn </font>

An overly flexible tree

```python
DecisionTreeClassifier(
    max_depth=None
)
```

may overfit.

A more constrained tree

```python
DecisionTreeClassifier(
    max_depth=6,
    min_samples_leaf=10,
    min_samples_split=20
)
```

is often better able to generalize.

The optimal hyperparameters should always be determined using cross-validation.

---



## <font color='orange'> 14. Practical Workflow </font>

A typical approach is

1. Train an unrestricted Decision Tree.
2. Evaluate the training and validation performance.
3. If overfitting is observed, gradually increase the regularization by adjusting

   * `max_depth`,
   * `min_samples_leaf`,
   * `min_samples_split`,
   * pruning parameters.

4. Select the model with the best validation performance.

This iterative procedure balances model complexity against predictive accuracy.

---



## <font color='orange'> 15. Relationship to Ensemble Methods </font>

One reason why ensemble methods outperform a single Decision Tree is that they help control variance.

For example,

* **Random Forests** reduce variance by averaging many Decision Trees.
* **Gradient Boosting** reduces bias by sequentially correcting previous prediction errors.

Although the individual trees may still overfit,

the ensemble often generalizes much better than any single tree.

---

## <font color='purple'> 16. Conceptual Summary </font>

| Aspect | Underfitting | Good Fit | Overfitting |
| :--- | :---: | :---: | :---: |
| Tree Depth | Too Small | Moderate | Too Large |
| Training Error | High | Low | Very Low |
| Validation Error | High | Lowest | High |
| Bias | High | Balanced | Low |
| Variance | Low | Balanced | High |
| Generalization | Poor | Excellent | Poor |

> **Key Insight:** The predictive performance of a Decision Tree depends critically on achieving the appropriate level of model complexity. Trees that are too shallow suffer from **underfitting**, failing to capture important relationships within the data, whereas excessively deep trees exhibit **overfitting** by memorizing the training observations rather than learning generalizable patterns. By carefully controlling tree complexity through stopping criteria, pruning, and cross-validation, a Decision Tree can achieve an effective balance between bias and variance, resulting in strong predictive performance on previously unseen data.

# <font color='blue'> Cost-Complexity Pruning (Post-Pruning) </font>

Although Decision Trees are capable of modelling highly complex relationships, unrestricted tree growth often leads to **overfitting**. Deep trees may perfectly classify the training data while performing poorly on previously unseen observations.

One effective strategy for reducing overfitting is **post-pruning**, where the tree is first allowed to grow to its maximum size and is then simplified by systematically removing branches that contribute little predictive value.

The most widely used post-pruning technique is **Cost-Complexity Pruning**, originally introduced in the **Classification and Regression Trees (CART)** algorithm and implemented in **Scikit-Learn** through the parameter

```python
ccp_alpha
```

Cost-Complexity Pruning balances **prediction accuracy** against **model complexity**, producing a simpler tree with improved generalization.

---



## <font color='orange'> 1. Why is Pruning Necessary? </font>

Suppose a Decision Tree is allowed to grow without restriction.

```
Root

↓

Many Splits

↓

Many More Splits

↓

Hundreds of Leaves
```

Eventually,

the tree may contain hundreds or even thousands of terminal nodes.

Although the training error approaches zero,

many branches represent

* statistical noise,
* measurement error,
* random fluctuations,
* outliers.

These branches rarely improve prediction on unseen data.

Instead,

they increase model variance and reduce generalization.

---



## <font color='orange'> 2. What is Pruning? </font>

Pruning removes unnecessary branches from a Decision Tree.

Rather than constructing a smaller tree initially,

the algorithm

1. grows a large tree,
2. identifies branches that contribute little predictive value,
3. removes those branches.

The resulting tree is

* smaller,
* easier to interpret,
* less likely to overfit.

---



## <font color='orange'> 3. Pre-Pruning vs Post-Pruning </font>

There are two general approaches.

### Pre-Pruning

The growth of the tree is restricted during construction.

Typical parameters include

* `max_depth`
* `min_samples_split`
* `min_samples_leaf`
* `max_leaf_nodes`

---

### Post-Pruning

The tree is first grown completely.

After construction,

branches are removed according to an optimization criterion.

Cost-Complexity Pruning belongs to this category.

---



## <font color='orange'> 4. Motivation </font>

Suppose two trees achieve similar predictive performance.

```
Large Tree

Training Accuracy = 99.9%

Validation Accuracy = 94%
```

```
Smaller Tree

Training Accuracy = 98.8%

Validation Accuracy = 95%
```

Although the first tree fits the training data more closely,

the second tree generalizes better.

Cost-Complexity Pruning therefore prefers the simpler tree.

---



## <font color='orange'> 5. The Cost-Complexity Objective Function </font>

The CART algorithm introduces a regularized objective function,

$$
\boxed{
R_\alpha(T)
=
R(T)
+
\alpha |T|,
}
$$

where

* $R(T)$ is the training error of tree $T$,
* $|T|$ is the number of terminal (leaf) nodes,
* $\alpha$ is the complexity parameter.

This objective balances two competing goals:

* minimizing prediction error,
* minimizing model complexity.

---



## <font color='orange'> 6. Interpretation of the Objective Function </font>

The objective function consists of two terms.

### Prediction Error

$$
R(T)
$$

measures how well the tree fits the training data.

Smaller values indicate better predictive performance.

---

### Complexity Penalty

$$
\alpha |T|
$$

penalizes large trees.

Increasing the number of leaf nodes increases the penalty.

The parameter

$$
\alpha
$$

controls the strength of this regularization.

---



## <font color='orange'> 7. The Complexity Parameter (\(\alpha\)) </font>

The parameter

$$
\alpha
$$

determines how aggressively branches are removed.

### Small \(\alpha\)

```
Weak Penalty

↓

Large Tree

↓

Lower Bias

Higher Variance
```

The tree remains relatively complex.

---

### Large \(\alpha\)

```
Strong Penalty

↓

Small Tree

↓

Higher Bias

Lower Variance
```

Many branches are removed,

producing a simpler model.

---



## <font color='orange'> 8. How Pruning Works </font>

The pruning algorithm proceeds iteratively.

```
Grow Complete Tree

↓

Evaluate Every Internal Node

↓

Compute Increase in Cost

↓

Remove Weakest Branch

↓

Repeat
```

At each iteration,

the branch whose removal produces the smallest increase in prediction error is pruned.

The process continues until further pruning would significantly reduce predictive performance.

---



## <font color='orange'> 9. Example </font>

Suppose an internal node contains

```
Parent

↓

Left Subtree

Right Subtree
```

If replacing both subtrees with a single leaf only slightly increases the training error,

the subtree is removed.

```
Parent

↓

Leaf
```

The resulting tree is simpler while maintaining nearly identical predictive accuracy.

---



## <font color='orange'> 10. Choosing the Optimal \(\alpha\) </font>

The optimal value of

$$
\alpha
$$

is usually selected using **Cross-Validation**.

The procedure is

1. Train multiple trees using different values of

```python
ccp_alpha
```

2. Evaluate each model on the validation set.

3. Select the value producing the lowest validation error.

This approach balances prediction accuracy and model complexity.

---



## <font color='orange'> 11. Cost-Complexity Pruning in Scikit-Learn </font>

Scikit-Learn implements Cost-Complexity Pruning through

```python
ccp_alpha
```

For example,

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="gini",
    ccp_alpha=0.005,
    random_state=42
)
```

Increasing

```python
ccp_alpha
```

produces progressively smaller Decision Trees.

---

Scikit-Learn provides a convenient method for computing the pruning path.

```python
path = model.cost_complexity_pruning_path(
    X_train,
    y_train
)

ccp_alphas = path.ccp_alphas
impurities = path.impurities
```

The returned values describe the sequence of increasingly simplified trees.

---



## <font color='orange'> 13. Effect of Pruning </font>

Pruning generally produces

* fewer nodes,
* fewer leaves,
* lower variance,
* improved interpretability,
* better generalization.

Although the training accuracy decreases slightly,

the testing accuracy often improves.

---



## <font color='orange'> 14. Advantages of Cost-Complexity Pruning </font>

Cost-Complexity Pruning provides several benefits.

* Reduces overfitting.
* Improves generalization.
* Produces simpler models.
* Improves interpretability.
* Reduces prediction time.
* Provides principled regularization.

These advantages explain why it remains the standard pruning technique in modern Decision Tree implementations.

---



## <font color='orange'> 15. Limitations </font>

Although highly effective,

Cost-Complexity Pruning has several limitations.

* Requires selecting an appropriate value of

  $$
  \alpha.
  $$

* May remove useful branches if

  $$
  \alpha
  $$

  is too large.

* Usually requires Cross-Validation to determine the optimal pruning strength.

Nevertheless,

it remains one of the most successful regularization techniques for Decision Trees.

---



## <font color='orange'> 16. Relationship to Regularization </font>

Cost-Complexity Pruning is conceptually similar to regularization methods used elsewhere in machine learning.

| Algorithm | Regularization Parameter |
| :--- | :--- |
| Ridge Regression | $\lambda \sum \beta_j^2$ |
| LASSO | $\lambda \sum |\beta_j|$ |
| Elastic Net | Combined $L_1$ and $L_2$ penalty |
| Decision Tree | $\alpha |T|$ |

In every case,

the objective is to prevent overly complex models from overfitting the training data.

---

## <font color='purple'> 17. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Pruning | Removing unnecessary branches |
| Objective | Reduce complexity while preserving predictive accuracy |
| Method | Grow first, simplify afterwards |
| Cost Function | $R_\alpha(T)=R(T)+\alpha|T|$ |
| Complexity Parameter | $\alpha$ (`ccp_alpha`) |
| Effect of Increasing $\alpha$ | Smaller trees, lower variance, higher bias |
| Primary Benefit | Improved generalization |

> **Key Insight:** Cost-Complexity Pruning is the standard post-pruning technique used in CART Decision Trees. Rather than preventing the tree from growing, it first constructs a large tree and then systematically removes branches that contribute little predictive value. By minimizing the regularized objective function \(R_\alpha(T)=R(T)+\alpha|T|\), the algorithm balances training accuracy against model complexity. The complexity parameter \(\alpha\) acts as a regularization term, encouraging simpler trees that are less prone to overfitting and more likely to generalize well to previously unseen data.

# <font color='blue'> Decision Tree Regression </font>

Decision Trees are not limited to classification problems. They can also be used to predict **continuous numerical variables**, such as house prices, insurance premiums, stock prices, temperatures, or energy consumption.

When applied to continuous target variables, the resulting model is known as a **Decision Tree Regressor**.

Although the overall tree-building algorithm remains the same as that used for classification, the splitting criterion changes fundamentally. Instead of minimizing measures such as **Entropy** or **Gini Impurity**, regression trees seek to minimize the variation of the target values within each node.

The objective is to partition the feature space into regions where the target variable is as homogeneous as possible.

---



## <font color='orange'> 1. Classification vs Regression Trees </font>

The primary difference lies in the type of target variable.

### Classification Tree

Predicts

$$
y \in \{1,2,\ldots,K\}
$$

where the output belongs to one of several discrete classes.

Examples include

* Fraud Detection
* Disease Classification
* Email Spam Detection

---

### Regression Tree

Predicts

$$
y\in\mathbb{R},
$$

where the output is a continuous numerical value.

Examples include

* House Prices
* Insurance Charges
* Sales Forecasting
* Temperature Prediction

---



## <font color='orange'> 2. The Basic Idea </font>

Suppose we wish to predict house prices.

Initially,

the dataset contains

```
All Houses

↓

One Large Region
```

The Decision Tree searches for the feature and threshold that best separates houses with different prices.

For example,

```
Floor Area < 1800 sq ft ?
```

The resulting regions contain houses with increasingly similar prices.

The splitting process then continues recursively.

---



## <font color='orange'> 3. Prediction at a Leaf Node </font>

Unlike classification trees,

a regression tree does **not** predict a class label.

Instead,

each terminal node predicts the **average target value** of the observations contained within that leaf.

Suppose a leaf contains

```
£320,000

£340,000

£310,000

£330,000
```

The prediction becomes

$$
\boxed{
\hat y
=
\frac{320000+340000+310000+330000}{4}
=
325000.
}
$$

Every new observation reaching this leaf receives the prediction

```
£325,000.
```

---



## <font color='orange'> 4. Why Use the Mean? </font>

Suppose a leaf contains

$$
N
$$

observations,

with target values

$$
y_1,y_2,\ldots,y_N.
$$

The predicted value is

$$
\boxed{
\hat y
=
\frac{1}{N}
\sum_{i=1}^{N}
y_i.
}
$$

The arithmetic mean minimizes the **Mean Squared Error (MSE)** within the leaf.

Consequently,

it is the optimal prediction under the squared-error loss function.

---



## <font color='orange'> 5. Measuring Node Impurity </font>

Classification trees measure impurity using

* Entropy,
* Gini Impurity,
* Misclassification Error.

Regression trees instead measure the spread of numerical target values.

Common impurity measures include

* Variance,
* Mean Squared Error (MSE),
* Mean Absolute Error (MAE) (less common).

The objective remains identical:

**produce child nodes with smaller impurity than the parent node.**

---



## <font color='orange'> 6. Variance of a Node </font>

Suppose a node contains

$$
N
$$

observations.

Its variance is

$$
\boxed{
Var
=
\frac{1}{N}
\sum_{i=1}^{N}
(y_i-\bar y)^2,
}
$$

where

$$
\bar y
$$

is the mean target value within the node.

Large variance indicates that the observations differ substantially.

Small variance indicates that they are highly similar.

---



## <font color='orange'> 7. Mean Squared Error (MSE) </font>

Most regression tree implementations,

including Scikit-Learn,

minimize the Mean Squared Error.

For a node,

the impurity becomes

$$
\boxed{
MSE
=
\frac{1}{N}
\sum_{i=1}^{N}
(y_i-\bar y)^2.
}
$$

Notice that this expression is identical to the node variance.

Thus,

variance reduction and MSE reduction are mathematically equivalent.

---



## <font color='orange'> 8. Choosing the Best Split </font>

Suppose a candidate split divides the parent node into two child nodes,

$$
R_1
$$

and

$$
R_2.
$$

The weighted impurity becomes

$$
\boxed{
I_{\text{split}}
=
\frac{N_1}{N}
MSE(R_1)
+
\frac{N_2}{N}
MSE(R_2).
}
$$

The algorithm selects the split producing the smallest weighted MSE.

---



## <font color='orange'> 9. Example </font>

Suppose a node contains house prices

```
£200k

£210k

£220k

£900k
```

The variance is large because one observation differs substantially.

A candidate split separates

```
£200k

£210k

£220k
```

from

```
£900k.
```

Both child nodes now exhibit much smaller variance.

The split is therefore highly desirable.

---



## <font color='orange'> 10. Piecewise Constant Predictions </font>

Regression trees produce **piecewise constant functions**.

For example,

```
Income < £40k

↓

Prediction = £18,000
```

```
Income ≥ £40k

↓

Prediction = £42,000
```

The prediction remains constant throughout each region of the feature space.

Unlike Linear Regression,

Decision Trees do not produce smooth continuous functions.

Instead,

they create a collection of rectangular regions,

each associated with a constant prediction.

---



## <font color='orange'> 11. Recursive Construction </font>

The recursive algorithm remains unchanged.

```
Compute Node Variance

↓

Evaluate Every Split

↓

Choose Minimum MSE

↓

Create Two Child Nodes

↓

Repeat
```

The only difference from classification trees is the impurity measure.

---



## <font color='orange'> 12. Advantages of Regression Trees </font>

Regression trees possess several attractive properties.

* Capture nonlinear relationships.
* Handle interactions automatically.
* No feature scaling required.
* Robust to monotonic transformations.
* Easy to interpret.
* Naturally handle mixed feature types.

---



## <font color='orange'> 13. Limitations </font>

Regression trees also have several weaknesses.

* Piecewise constant predictions.
* Poor extrapolation outside the training range.
* High variance.
* Sensitive to noisy data.
* Prone to overfitting.

These limitations motivate ensemble methods such as

* Random Forest Regression,
* Gradient Boosting Regression,
* XGBoost Regression.

---



## <font color='orange'> 14. Regression Trees in Scikit-Learn </font>

A regression tree may be implemented as follows.

```python
from sklearn.tree import DecisionTreeRegressor

model = DecisionTreeRegressor(
    criterion="squared_error",
    max_depth=5,
    random_state=42
)

model.fit(X_train,y_train)
```

The default criterion

```python
"squared_error"
```

minimizes the Mean Squared Error within each node.

Alternative criteria include

```python
"friedman_mse"

"absolute_error"

"poisson"
```

depending on the regression problem.

---



## <font color='orange'> 15. Classification vs Regression Trees </font>

| Property | Classification Tree | Regression Tree |
| :--- | :--- | :--- |
| Target Variable | Discrete classes | Continuous values |
| Prediction | Majority class | Mean target value |
| Impurity Measure | Entropy, Gini, Misclassification Error | Variance, MSE |
| Objective | Maximize node purity | Minimize prediction variance |
| Output | Class label | Numerical value |

---

## <font color='purple'> 16. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Target | Continuous numerical variable |
| Prediction | Mean value within each leaf |
| Impurity Measure | Variance or Mean Squared Error |
| Objective | Minimize weighted node variance |
| Decision Boundary | Piecewise constant regions |
| Advantages | Nonlinear modelling, interpretability |
| Limitations | Piecewise predictions, overfitting |

> **Key Insight:** Decision Tree Regression extends the Decision Tree framework from classification to continuous prediction problems. Rather than minimizing class impurity, regression trees minimize the variability of the target values within each node, typically using the **Mean Squared Error (MSE)** or **variance**. Each terminal node predicts the average target value of the observations it contains, resulting in a piecewise constant approximation to the underlying regression function. Although simple and highly interpretable, regression trees often exhibit high variance, motivating the use of ensemble methods such as **Random Forest Regression** and **Gradient Boosting Regression**, which combine many regression trees to achieve substantially improved predictive performance.
