# <font color='blue'> Introduction to Logistic Regression </font>

Logistic Regression is one of the most fundamental supervised learning algorithms used for **binary classification**. Despite its name, logistic regression is **not a regression algorithm** in the traditional sense. Instead, it is a **probabilistic classification algorithm** that estimates the probability that an observation belongs to a particular class.

For binary classification problems, Logistic Regression predicts the probability

$$
P(y=1|\mathbf{x}),
$$

where

* \(y=1\) denotes the positive class,
* \(y=0\) denotes the negative class,
* \(\mathbf{x}\) represents the vector of input features.

Instead of directly predicting the class label, Logistic Regression first predicts the probability of an observation belonging to the positive class. The class label is then obtained by applying a decision threshold, usually

$$
P(y=1|\mathbf{x}) \ge 0.5.
$$

---

## <font color='orange'> 1. Why Can't We Use Linear Regression? </font>

At first glance, one might consider using Linear Regression for classification.

A linear regression model predicts

$$
\hat{y}
=
\beta_0
+
\beta_1x_1
+
\beta_2x_2
+\cdots+
\beta_nx_n.
$$

The prediction can take **any real value**

$$
-\infty
<
\hat y
<
+\infty.
$$

For example,

| Prediction | Interpretation |
| :---: | :--- |
| -2.4 | Impossible Probability |
| 0.35 | Possible Probability |
| 1.87 | Impossible Probability |

Since probabilities must satisfy

$$
0
\le
P
\le
1,
$$

linear regression clearly cannot represent probabilities.

Furthermore, the relationship between the input variables and the probability of belonging to a class is generally **nonlinear**, while linear regression assumes a linear relationship.

---

## <font color='orange'> 2. The Logistic (Sigmoid) Function </font>

To solve this problem, Logistic Regression transforms the linear prediction into a probability using the **logistic function**, also called the **sigmoid function**.

The sigmoid function is defined as

$$
\sigma(z)
=
\frac{1}{1+e^{-z}},
$$

where

$$
z
=
\beta_0
+
\beta_1x_1
+\cdots+
\beta_nx_n.
$$

Therefore,

$$
P(y=1|\mathbf{x})
=
\frac{1}
{1+e^{-(\beta_0+\beta_1x_1+\cdots+\beta_nx_n)}}.
$$

The sigmoid function has several important properties:

* Output always lies between 0 and 1.
* Smooth and differentiable.
* Monotonically increasing.
* Easily optimized using gradient-based methods.

---

## <font color='orange'> 3. Properties of the Sigmoid Function </font>

The sigmoid curve has an S-shaped appearance.

Some important values are

| \(z\) | \(\sigma(z)\) |
| :---: | :---: |
| \(-\infty\) | 0 |
| -2 | 0.119 |
| -1 | 0.269 |
| 0 | 0.500 |
| 1 | 0.731 |
| 2 | 0.881 |
| \(+\infty\) | 1 |

Notice that

$$
\sigma(0)=0.5.
$$

This naturally defines the decision boundary.

---

## <font color='orange'> 4. From Linear Scores to Probabilities </font>

Instead of predicting the class directly, Logistic Regression computes

$$
z
=
\beta^T\mathbf{x}
+\beta_0.
$$

This quantity is called the **linear score**.

The probability is then obtained using

$$
P(y=1|\mathbf{x})
=
\sigma(z).
$$

Thus,

$$
z
\longrightarrow
\sigma(z)
\longrightarrow
P(y=1).
$$

This transformation compresses all real numbers into the interval

$$
[0,1].
$$

---

## <font color='orange'> 5. Odds and Odds Ratio </font>

Before introducing the logistic equation, statisticians define the **odds** of an event.

If

$$
P=p,
$$

then

$$
\text{Odds}
=
\frac{p}{1-p}.
$$

For example,

If

$$
P=0.80,
$$

then

$$
\text{Odds}
=
\frac{0.8}{0.2}
=
4.
$$

This means the event is four times more likely to occur than not occur.

---

## <font color='orange'> 6. Log-Odds (The Logit Function) </font>

The odds are always positive.

To obtain a quantity that can vary over the entire real line, we take the logarithm,

$$
\log
\left(
\frac{p}{1-p}
\right).
$$

This quantity is called the **logit**.

Logistic Regression assumes that the log-odds vary linearly with the input variables,

$$
\log
\left(
\frac{p}{1-p}
\right)
=
\beta_0
+
\beta_1x_1
+\cdots+
\beta_nx_n.
$$

This is the fundamental equation of Logistic Regression.

---

## <font color='orange'> 7. Deriving the Logistic Equation </font>

Starting from

$$
\log
\left(
\frac{p}{1-p}
\right)
=
z,
$$

where

$$
z
=
\beta_0+\beta_1x_1+\cdots+\beta_nx_n,
$$

take the exponential of both sides,

$$
\frac{p}{1-p}
=
e^z.
$$

Multiplying by

$$
1-p,
$$

gives

$$
p
=
e^z(1-p).
$$

Expanding,

$$
p
=
e^z
-
pe^z.
$$

Collecting the terms containing \(p\),

$$
p(1+e^z)
=
e^z.
$$

Therefore,

$$
p
=
\frac{e^z}{1+e^z}.
$$

Multiplying numerator and denominator by

$$
e^{-z},
$$

gives the familiar sigmoid equation,

$$
p
=
\frac{1}
{1+e^{-z}}.
$$

Hence,

$$
P(y=1|\mathbf{x})
=
\frac{1}
{1+e^{-(\beta_0+\beta_1x_1+\cdots+\beta_nx_n)}}.
$$

---

## <font color='orange'> 8. Decision Boundary </font>

The classifier predicts

$$
\hat y
=
1
$$

whenever

$$
P(y=1)\ge0.5.
$$

Since

$$
\sigma(0)=0.5,
$$

the decision boundary occurs when

$$
\beta_0
+
\beta_1x_1
+\cdots+
\beta_nx_n
=
0.
$$

Thus, Logistic Regression is still a **linear classifier**, even though it predicts nonlinear probabilities.

---

## <font color='orange'> 9. Advantages of Logistic Regression </font>

Some important advantages include

* Fast to train.
* Easy to interpret.
* Produces probabilities.
* Works well for linearly separable data.
* Computationally inexpensive.
* Robust baseline classifier.

---

## <font color='orange'> 10. Limitations </font>

Despite its popularity, Logistic Regression has several limitations.

* Assumes a linear decision boundary.
* Sensitive to highly correlated features.
* Cannot model complex nonlinear relationships without feature engineering.
* Performance decreases when classes overlap significantly.
* May struggle with severe class imbalance unless class weighting or resampling techniques are used.

---

## <font color='orange'> 11. Typical Applications </font>

Logistic Regression is widely used in

* Credit Card Fraud Detection
* Medical Diagnosis
* Credit Risk Analysis
* Customer Churn Prediction
* Email Spam Detection
* Marketing Response Prediction
* Loan Approval
* Quality Control

---

## <font color='purple'> 12. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Model Type | Supervised Classification |
| Output | Probability |
| Activation Function | Sigmoid |
| Decision Boundary | Linear |
| Optimization | Maximum Likelihood Estimation |

> **Key Insight:** Logistic Regression does **not** directly model the class labels. Instead, it models the **probability** that an observation belongs to a particular class by assuming that the **log-odds** are a linear function of the input variables. The sigmoid function then transforms these log-odds into valid probabilities between 0 and 1, making Logistic Regression one of the simplest yet most powerful probabilistic classifiers.

# <font color='blue'> Classification with Logistic Regression </font>

In the previous section, Logistic Regression was introduced as a probabilistic classifier that predicts the probability

$$
P(y=1|\mathbf{x})
=
\frac{1}{1+e^{-z}},
$$

where

$$
z=\beta_0+\beta_1x_1+\cdots+\beta_nx_n.
$$

The next question is **how the parameters**

$$
\beta=(\beta_0,\beta_1,\ldots,\beta_n)
$$

**are determined from the training data.**

Unlike Linear Regression, which minimizes the Mean Squared Error (MSE), Logistic Regression estimates its parameters by **maximizing the likelihood** of observing the training data.

---

## <font color='orange'> 1. The Probabilistic Interpretation </font>

Suppose a training dataset contains

$$
m
$$

observations,

$$
(\mathbf{x}^{(1)},y^{(1)}),
(\mathbf{x}^{(2)},y^{(2)}),
\ldots,
(\mathbf{x}^{(m)},y^{(m)}).
$$

Each label satisfies

$$
y^{(i)}
\in
\{0,1\}.
$$

For every observation, Logistic Regression predicts

$$
p^{(i)}
=
P(y^{(i)}=1|\mathbf{x}^{(i)}).
$$

Since there are only two possible classes,

$$
P(y=0)=1-p.
$$

Therefore,

$$
P(y)=
\begin{cases}
p,&y=1\\
1-p,&y=0.
\end{cases}
$$

This can be written compactly as

$$
P(y)
=
p^y(1-p)^{1-y}.
$$

This expression is known as the **Bernoulli probability distribution**, which forms the probabilistic foundation of Logistic Regression.

---

## <font color='orange'> 2. The Likelihood Function </font>

The objective of the learning algorithm is to choose the parameters

$$
\beta
$$

that make the observed training data **most probable**.

Assuming all observations are statistically independent,

the probability of observing the complete dataset is

$$
L(\beta)
=
\prod_{i=1}^{m}
P(y^{(i)}|\mathbf{x}^{(i)}).
$$

Substituting the Bernoulli distribution,

$$
L(\beta)
=
\prod_{i=1}^{m}
\left(p^{(i)}\right)^{y^{(i)}}
\left(1-p^{(i)}\right)^{1-y^{(i)}}.
$$

This is called the **Likelihood Function**.

The parameters of Logistic Regression are chosen to maximize this quantity.

---

## <font color='orange'> 3. Why Use the Log-Likelihood? </font>

The likelihood is a product of many probabilities,

which quickly becomes extremely small,

leading to numerical underflow.

Instead, we maximize the logarithm of the likelihood.

Since

$$
\log(ab)
=
\log a+\log b,
$$

the product becomes a sum,

making optimization much easier.

Taking the logarithm gives

$$
\ell(\beta)
=
\log L(\beta)
=
\sum_{i=1}^{m}
\left[
y^{(i)}
\log p^{(i)}
+
(1-y^{(i)})
\log(1-p^{(i)})
\right].
$$

This is called the **Log-Likelihood Function**.

---

## <font color='orange'> 4. From Log-Likelihood to the Cost Function </font>

Machine learning algorithms are generally formulated as **minimization problems** rather than maximization problems.

Therefore, we define the cost function as the negative average log-likelihood,

$$
J(\beta)
=
-\frac1m
\sum_{i=1}^{m}
\left[
y^{(i)}
\log p^{(i)}
+
(1-y^{(i)})
\log(1-p^{(i)})
\right].
$$

This function is known as

* Binary Cross-Entropy Loss
* Log Loss
* Logistic Loss

All three names refer to the same objective function.

---

## <font color='orange'> 5. Why Mean Squared Error is Not Used </font>

One might ask why Logistic Regression does not minimize

$$
MSE
=
\frac1m
\sum
(y-\hat y)^2.
$$

Although mathematically possible,

combining the sigmoid function with the squared error produces a **non-convex optimization problem**.

A non-convex function contains multiple local minima,

making optimization much more difficult.

In contrast,

Binary Cross-Entropy produces a convex cost function,

ensuring that gradient-based optimization converges to a unique global minimum.

---

## <font color='orange'> 6. Understanding Cross-Entropy Loss </font>

Consider a single observation.

The loss is

$$
J
=
-
\left[
y\log p
+
(1-y)\log(1-p)
\right].
$$

There are only two possible cases.

### <font color='orange'> Case 1: True Class = 1 </font>

If

$$
y=1,
$$

then

$$
J
=
-\log p.
$$

Examples:

| Predicted Probability | Loss |
| :---: | :---: |
| 0.99 | 0.010 |
| 0.90 | 0.105 |
| 0.50 | 0.693 |
| 0.10 | 2.303 |
| 0.01 | 4.605 |

Notice that

**confident incorrect predictions are heavily penalized.**

---

### <font color='orange'> Case 2: True Class = 0 </font>

If

$$
y=0,
$$

then

$$
J
=
-\log(1-p).
$$

Again,

large penalties occur whenever the classifier confidently predicts the wrong class.

---

## <font color='orange'> 7. Gradient Descent </font>

The parameters

$$
\beta
$$

are obtained using Gradient Descent.

Each iteration updates the coefficients according to

$$
\beta_j
\leftarrow
\beta_j
-
\alpha
\frac{\partial J}{\partial\beta_j},
$$

where

* \(\alpha\) is the learning rate,
* \(J\) is the cost function.

For Logistic Regression,

the gradient simplifies to

$$
\frac{\partial J}{\partial\beta_j}
=
\frac1m
\sum_{i=1}^{m}
(p^{(i)}-y^{(i)})
x_j^{(i)}.
$$

This elegant expression is one of the reasons Logistic Regression is computationally efficient.

---

## <font color='orange'> 8. Decision Threshold </font>

Logistic Regression predicts probabilities.

To convert probabilities into class labels,

a threshold must be selected.

The standard choice is

$$
0.5.
$$

Thus,

$$
\hat y=
\begin{cases}
1,&p\ge0.5\\
0,&p<0.5
\end{cases}
$$

However,

the threshold is **not fixed**.

For highly imbalanced datasets,

such as credit card fraud detection,

a lower threshold (for example 0.2)

may improve recall by identifying more fraudulent transactions,

although this typically increases the number of false positives.

---

## <font color='orange'> 9. Regularization </font>

Large coefficient values may cause overfitting.

Regularization introduces a penalty term to discourage excessively large coefficients.

### <font color='orange'> L2 Regularization (Ridge) </font>

$$
J(\beta)
=
J_0(\beta)
+
\frac{\lambda}{2m}
\sum_{j=1}^{n}
\beta_j^2.
$$

This shrinks coefficients toward zero,

improving generalization.

---

### <font color='orange'> L1 Regularization (LASSO) </font>

$$
J(\beta)
=
J_0(\beta)
+
\frac{\lambda}{m}
\sum_{j=1}^{n}
|\beta_j|.
$$

Unlike Ridge Regression,

LASSO can force some coefficients to become exactly zero,

performing automatic feature selection.

---

### <font color='orange'> Elastic Net </font>

Elastic Net combines both penalties,

$$
J
=
J_0
+
\lambda_1
\sum |\beta_j|
+
\lambda_2
\sum \beta_j^2.
$$

It is particularly useful when the predictors are highly correlated.

---

## <font color='orange'> 10. Advantages of Maximum Likelihood Estimation </font>

Maximum Likelihood Estimation provides several desirable statistical properties.

Under suitable assumptions,

the estimated parameters are

* Consistent
* Efficient
* Asymptotically unbiased
* Approximately normally distributed for large datasets

These properties make Logistic Regression a statistically well-founded classification method.

---

## <font color='orange'> 11. Summary of the Learning Procedure </font>

The complete Logistic Regression training algorithm proceeds as follows:

1. Initialize the model parameters randomly.
2. Compute the linear score

   $$
   z=\beta^T\mathbf{x}.
   $$

3. Apply the sigmoid function,

   $$
   p=\sigma(z).
   $$

4. Compute the Binary Cross-Entropy Loss.
5. Evaluate the gradient.
6. Update the parameters using Gradient Descent.
7. Repeat until convergence.

---

## <font color='purple'> 12. Conceptual Summary </font>

| Concept | Purpose |
| :--- | :--- |
| Bernoulli Distribution | Models binary outcomes |
| Likelihood | Probability of observing the data |
| Log-Likelihood | Easier optimization |
| Cross-Entropy Loss | Objective function |
| Gradient Descent | Parameter optimization |
| Decision Threshold | Converts probabilities into class labels |
| Regularization | Reduces overfitting |

> **Key Insight:** Logistic Regression learns by finding the model parameters that maximize the probability of observing the training data. This is achieved through **Maximum Likelihood Estimation**, which leads naturally to the **Binary Cross-Entropy (Log Loss)** objective function. Gradient-based optimization is then used to minimize this loss, producing a classifier that estimates probabilities while maintaining a convex optimization problem with a unique global optimum.

# <font color='blue'> Logistic Regression for Multi-Class Problems </font>

The Logistic Regression model introduced previously is naturally designed for **binary classification**, where the target variable has only two possible outcomes,

$$
y\in\{0,1\}.
$$

However, many real-world machine learning problems involve **multiple classes**. Examples include handwritten digit recognition (0–9), image classification, language identification, and species classification.

To solve these problems, Logistic Regression must be extended so that it can estimate the probability of an observation belonging to **more than two classes**.

Suppose there are

$$
K
$$

possible classes,

$$
y\in\{1,2,\ldots,K\},
$$

where

$$
K>2.
$$

The classifier now estimates

$$
P(y=i|\mathbf{x}),
\qquad
i=1,\ldots,K.
$$

The predicted class is simply the one having the highest probability,

$$
\hat y
=
\arg\max_i
P(y=i|\mathbf{x}).
$$

---

## <font color='orange'> 1. Why Binary Logistic Regression Cannot Be Used Directly </font>

Binary Logistic Regression predicts

$$
P(y=1|\mathbf{x})
=
\frac{1}
{1+e^{-z}}.
$$

Since

$$
P(y=0)=1-P(y=1),
$$

only **two probabilities** can be represented.

Suppose we wish to classify handwritten digits,

$$
y
\in
\{0,1,2,\ldots,9\}.
$$

There are now ten possible outcomes.

A single sigmoid function cannot simultaneously represent ten probabilities because

$$
P(y=0)+P(y=1)\neq1
$$

is no longer sufficient.

A different strategy is therefore required.

---

## <font color='orange'> 2. One-vs-Rest (OvR) Classification </font>

The simplest extension is called **One-vs-Rest (OvR)**, also known as **One-vs-All**.

Suppose there are

$$
K
$$

classes.

Instead of training one classifier,

we train

$$
K
$$

independent binary classifiers.

Each classifier learns

$$
\text{Class }i
\quad
\text{vs}
\quad
\text{All Remaining Classes}.
$$

For example,

consider a four-class problem,

| Classifier | Positive Class | Negative Class |
| :--- | :--- | :--- |
| Model 1 | Cat | Dog, Bird, Horse |
| Model 2 | Dog | Cat, Bird, Horse |
| Model 3 | Bird | Cat, Dog, Horse |
| Model 4 | Horse | Cat, Dog, Bird |

Each model independently predicts a probability.

Finally,

the classifier having the largest probability is selected,

$$
\hat y
=
\arg\max_i
P_i.
$$

---

## <font color='orange'> 3. Advantages of One-vs-Rest </font>

One-vs-Rest has several attractive properties.

* Easy to understand.
* Simple to implement.
* Requires only binary Logistic Regression.
* Efficient when the number of classes is moderate.

Consequently,

OvR remains the default implementation in many machine learning libraries.

---

## <font color='orange'> 4. Limitations of One-vs-Rest </font>

Because each classifier is trained independently,

the predicted probabilities

$$
P_1,P_2,\ldots,P_K
$$

are **not guaranteed to sum to one**.

For example,

$$
P=
(0.72,
0.61,
0.48,
0.31).
$$

Notice that

$$
0.72+0.61+0.48+0.31
=
2.12.
$$

These are therefore **confidence scores**, not true probabilities.

---

## <font color='orange'> 5. One-vs-One (OvO) Classification </font>

Another strategy is **One-vs-One**.

Instead of comparing each class against all others,

a separate classifier is trained for every pair of classes.

The total number of classifiers becomes

$$
\frac{K(K-1)}{2}.
$$

For example,

if

$$
K=5,
$$

the number of binary classifiers required is

$$
\frac{5(5-1)}2
=
10.
$$

Each classifier votes for one class.

The class receiving the largest number of votes becomes the prediction.

---

## <font color='orange'> 6. Softmax Regression </font>

The most elegant extension of Logistic Regression is **Softmax Regression**, also known as

* Multinomial Logistic Regression
* Multi-Class Logistic Regression

Instead of fitting several binary classifiers,

Softmax Regression learns all classes simultaneously.

Suppose there are

$$
K
$$

classes.

Each class has its own linear score,

$$
z_i
=
\beta_i^T\mathbf{x}.
$$

The probability of class

$$
i
$$

is

$$
P(y=i|\mathbf{x})
=
\frac{e^{z_i}}
{\sum_{j=1}^{K}e^{z_j}}.
$$

This equation is known as the **Softmax Function**.

---

## <font color='orange'> 7. Properties of the Softmax Function </font>

The Softmax function has several desirable mathematical properties.

### All probabilities are positive

$$
P_i>0.
$$

---

### All probabilities sum to one

$$
\sum_{i=1}^{K}
P_i
=
1.
$$

---

### Largest score receives largest probability

If

$$
z_i>z_j,
$$

then

$$
P_i>P_j.
$$

Thus,

Softmax naturally converts arbitrary scores into valid probability distributions.

---

## <font color='orange'> 8. Example of Softmax </font>

Suppose the model computes

$$
z=
(2.4,1.1,0.5).
$$

The exponential scores are

$$
e^z=
(11.02,
3.00,
1.65).
$$

The denominator becomes

$$
11.02+3.00+1.65
=
15.67.
$$

Therefore,

$$
P=
(0.703,
0.191,
0.106).
$$

Notice that

$$
0.703+0.191+0.106
=
1.
$$

The classifier therefore predicts

Class 1.

---

## <font color='orange'> 9. Loss Function for Softmax Regression </font>

Binary Logistic Regression minimizes Binary Cross-Entropy.

For multiple classes,

the appropriate objective becomes the **Categorical Cross-Entropy Loss**,

$$
J
=
-
\sum_{i=1}^{K}
y_i
\log(P_i).
$$

For an entire dataset,

$$
J
=
-\frac1m
\sum_{j=1}^{m}
\sum_{i=1}^{K}
y_i^{(j)}
\log(P_i^{(j)}).
$$

This is the standard loss function used in nearly all modern neural network classifiers.

---

## <font color='orange'> 10. Training Procedure </font>

Softmax Regression is trained similarly to binary Logistic Regression.

For each iteration,

1. Compute the linear scores,

   $$
   z_i=\beta_i^T\mathbf{x}.
   $$

2. Apply the Softmax transformation.

3. Compute the categorical cross-entropy loss.

4. Compute the gradient.

5. Update the parameters using Gradient Descent.

6. Repeat until convergence.

---

## <font color='orange'> 11. OvR vs Softmax </font>

| Property | One-vs-Rest | Softmax |
| :--- | :---: | :---: |
| Number of Models | \(K\) | 1 |
| Probabilities Sum to 1 | ✗ | ✓ |
| Computational Cost | Lower | Slightly Higher |
| Mathematical Elegance | Moderate | Excellent |
| Used in Neural Networks | ✗ | ✓ |

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Binary Logistic Regression | Two classes |
| Multi-Class Classification | More than two classes |
| One-vs-Rest | Trains one classifier per class |
| One-vs-One | Trains one classifier for every pair of classes |
| Softmax Regression | Learns all classes simultaneously |
| Softmax Function | Converts scores into probabilities |
| Loss Function | Categorical Cross-Entropy |

> **Key Insight:** Binary Logistic Regression predicts the probability of one class versus another. For problems involving multiple classes, the algorithm must be extended. While **One-vs-Rest** constructs several independent binary classifiers, **Softmax Regression** directly models all classes simultaneously, producing a valid probability distribution whose probabilities always sum to one. Because of its mathematical elegance and probabilistic interpretation, Softmax Regression has become the standard classification layer in modern machine learning and deep learning models.

# <font color='blue'> Implementing Logistic Regression Models </font>

Having understood the mathematical foundations of Logistic Regression, the next step is to implement the model on real-world datasets. Fortunately, libraries such as **Scikit-Learn** provide efficient implementations that require only a few lines of code. However, obtaining a good classifier involves considerably more than simply calling the `fit()` function.

A complete Logistic Regression workflow generally consists of

1. Data preprocessing
2. Splitting the data into training and testing sets
3. Feature scaling
4. Model construction
5. Model training
6. Hyperparameter tuning
7. Model evaluation
8. Interpretation of the fitted model

Each of these stages contributes to the overall performance of the classifier.

---

## <font color='orange'> 1. Preparing the Dataset </font>

The first step is to separate the dataset into the **feature matrix**

$$
X
$$

and the **target vector**

$$
y.
$$

Mathematically,

$$
X=
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1n}\\
x_{21} & x_{22} & \cdots & x_{2n}\\
\vdots & \vdots & \ddots & \vdots\\
x_{m1} & x_{m2} & \cdots & x_{mn}
\end{bmatrix},
$$

while

$$
y=
\begin{bmatrix}
y_1\\
y_2\\
\vdots\\
y_m
\end{bmatrix}.
$$

In Python,

```python
X = df.drop("Class", axis=1)
y = df["Class"]
```

## <font color='orange'> 2. Train-Test Split </font>

To evaluate the performance of the classifier on unseen data, the dataset is divided into two subsets.

* Training Set
* Testing Set

Typically,

* 80% Training
* 20% Testing

This is performed using

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)
```

The option

```python
stratify=y
```

is especially important for **imbalanced datasets**, as it preserves the proportion of each class in both the training and testing sets.

## <font color='orange'> 3. Why Feature Scaling is Necessary </font>

Logistic Regression is an optimization-based algorithm.

Gradient Descent updates each parameter according to

$$
\beta_j
\leftarrow
\beta_j
-
\alpha
\frac{\partial J}{\partial\beta_j}.
$$

If one feature ranges from

$$
0-1
$$

while another ranges from

$$
0-10^6,
$$

the optimization process becomes inefficient because one variable dominates the gradients.

Feature scaling ensures that every feature contributes approximately equally during optimization.

The most common approach is **Standardization**,

$$
x'
=
\frac{x-\mu}{\sigma},
$$

where

* \(\mu\) is the mean,
* \(\sigma\) is the standard deviation.

After scaling,

each feature has approximately

* Mean = 0
* Standard Deviation = 1

---

## <font color='orange'> 4. StandardScaler </font>

Scikit-Learn performs standardization using

```python
from sklearn.preprocessing import StandardScaler
```

The transformed feature is

$$
z
=
\frac{x-\bar x}{s}.
$$

Using a **Pipeline** ensures that scaling is performed correctly on both the training and testing data.

```python
Pipeline([
    ("scaler", StandardScaler()),
    ("logreg", LogisticRegression())
])
```

## <font color='orange'> 5. Choosing the Solver </font>

The Logistic Regression optimization problem can be solved using several numerical algorithms.

| Solver | Characteristics |
| :--- | :--- |
| lbfgs | Fast and stable (default) |
| liblinear | Good for small datasets and L1 regularization |
| saga | Supports Elastic Net and very large datasets |
| newton-cg | Suitable for multinomial classification |
| sag | Fast for large datasets |

The default choice,

```python
solver="lbfgs"
```

works well for most binary and multiclass classification problems.

## <font color='orange'> 6. Regularization </font>

Logistic Regression includes regularization by default.

The optimization problem becomes

$$
J
=
J_0
+
\lambda R(\beta),
$$

where

$$
R(\beta)
$$

is the regularization penalty.

Three common choices are

### Ridge (L2)

$$
R
=
\sum \beta_j^2
$$

### LASSO (L1)

$$
R
=
\sum |\beta_j|
$$

### Elastic Net

$$
R
=
\lambda_1
\sum |\beta_j|
+
\lambda_2
\sum \beta_j^2.
$$

Regularization reduces overfitting by discouraging excessively large coefficients.

---

## <font color='orange'> 7. The Regularization Parameter \(C\) </font>

Scikit-Learn uses

```python
C
```

instead of

$$
\lambda.
$$

The relationship is

$$
C
=
\frac1\lambda.
$$

Therefore,

| Small \(C\) | Large Regularization |
| Large \(C\) | Weak Regularization |

Typical values include

```python
C=[0.01,0.1,1,10,100]
```

## <font color='orange'> 10. Handling Class Imbalance </font>

Many real-world datasets contain far more observations from one class than another.

For example,

| Class | Number of Samples |
| :--- | ---: |
| Genuine | 284315 |
| Fraud | 492 |

Without correction,

the classifier becomes biased toward the majority class.

A common solution is

```python
class_weight="balanced"
```

which automatically assigns

$$
w_i
=
\frac{N}
{kN_i},
$$

where

* \(N\) is the total number of observations,
* \(k\) is the number of classes,
* \(N_i\) is the number of observations belonging to class \(i\).

The minority class therefore receives a larger weight during optimization.

## <font color='orange'> 13. Interpreting the Coefficients </font>

Unlike many machine learning algorithms,

Logistic Regression remains highly interpretable.

The fitted model is

$$
\log
\left(
\frac{p}{1-p}
\right)
=
\beta_0
+
\sum
\beta_i x_i.
$$

Each coefficient represents the change in the **log-odds** associated with a one-unit increase in the corresponding feature.

Interpretation:

| Coefficient | Meaning |
| :---: | :--- |
| Positive | Increases probability of positive class |
| Negative | Decreases probability of positive class |
| Large Magnitude | Strong influence |
| Near Zero | Weak influence |
