# <font color='blue'> Random Forests </font>

Random Forests are one of the most powerful and widely used ensemble learning algorithms for both classification and regression. Introduced by **Leo Breiman (2001)**, a Random Forest extends the Bagging algorithm by introducing an additional source of randomness during tree construction.

While Bagging trains multiple Decision Trees using different bootstrap samples, the trees often remain highly correlated because they repeatedly select the same strong predictor variables when determining their splits.

Random Forests overcome this limitation by forcing every Decision Tree to consider only a **random subset of predictor variables** at each split.

This simple modification significantly reduces the correlation between trees, allowing the ensemble to achieve a greater reduction in prediction variance than Bagging alone.

---

## <font color='orange'> 1. Motivation </font>

Recall that Bagging constructs multiple Decision Trees using different bootstrap samples.

```
Training Data

↓

Bootstrap Sample 1

↓

Tree 1
```

```
Training Data

↓

Bootstrap Sample 2

↓

Tree 2
```

```
Training Data

↓

Bootstrap Sample 3

↓

Tree 3
```

Although each tree receives different observations,

the trees frequently identify the same important predictor variables.

Consequently,

their structures often become very similar.

This correlation limits the benefit obtained by averaging.

---

## <font color='orange'> 2. The Correlation Problem </font>

Suppose one feature,

```
Income
```

is overwhelmingly more informative than every other feature.

Almost every Decision Tree begins with

```
Income < £50,000 ?
```

```
Tree 1

↓

Income
```

```
Tree 2

↓

Income
```

```
Tree 3

↓

Income
```

Although later splits differ,

the trees remain highly correlated.

Highly correlated trees tend to make similar prediction errors.

Averaging similar predictions provides only limited improvement.

---

## <font color='orange'> 3. The Key Idea Behind Random Forests </font>

Random Forests introduce an additional source of randomness.

Rather than evaluating

**every feature**

at every split,

each Decision Tree considers only a **random subset of features**.

Suppose the dataset contains

```
20 features.
```

At a particular node,

the algorithm randomly selects only

```
5 features.
```

The best split is then chosen **only among those five features**.

Different nodes receive different random subsets.

This forces different trees to grow in different ways,

substantially increasing model diversity.

---

## <font color='orange'> 4. Bagging vs Random Forest </font>

Bagging performs

```
Bootstrap Sampling

↓

Decision Trees

↓

Averaging
```

Random Forest performs

```
Bootstrap Sampling

↓

Random Feature Selection

↓

Decision Trees

↓

Averaging
```

The additional feature randomness is the defining characteristic of Random Forests.

---

## <font color='orange'> 5. Random Subspace Method </font>

The technique of selecting random subsets of predictor variables is known as the

**Random Subspace Method**.

Suppose a dataset contains

$$
p
$$

features.

Instead of evaluating all

$$
p
$$

variables,

the algorithm randomly chooses

$$
m
$$

features,

where

$$
m<p.
$$

Only these

$$
m
$$

features are examined when searching for the optimal split.

The feature subset changes independently at every node.

---

## <font color='orange'> 6. Mathematical Description </font>

Suppose

* the dataset contains

$$
p
$$

features,

* each node randomly selects

$$
m
$$

features.

The split optimization becomes

$$
\boxed{
(j^*,s^*)
=
\arg\min_{j\in\mathcal{M},\,s}
I(j,s),
}
$$

where

$$
\mathcal{M}
$$

represents the randomly selected subset of

$$
m
$$

features.

Notice that the search is no longer performed over all variables.

---

## <font color='orange'> 7. Why Random Features Help </font>

Suppose two features are highly informative.

Without feature randomization,

every tree repeatedly chooses

```
Income

↓

Credit Score
```

The trees become similar.

With Random Forests,

one tree might begin with

```
Income
```

another with

```
Age
```

another with

```
Balance
```

Each tree explores a different part of the feature space.

The prediction errors become less correlated,

making averaging considerably more effective.

---

## <font color='orange'> 8. Complete Random Forest Algorithm </font>

The Random Forest algorithm proceeds as follows.

### Step 1

Generate

$$
B
$$

bootstrap samples.

---

### Step 2

Train one Decision Tree on each bootstrap sample.

---

### Step 3

At every node,

randomly select

$$
m
$$

predictor variables.

---

### Step 4

Choose the best split only among those variables.

---

### Step 5

Grow each tree completely (or until stopping criteria are met).

---

### Step 6

Aggregate the predictions.

Classification

↓

Majority Vote

Regression

↓

Average Prediction

---

## <font color='orange'> 9. Out-of-Bag (OOB) Error </font>

Like Bagging,

Random Forest automatically produces

Out-of-Bag samples.

Each observation omitted from a bootstrap sample is predicted by the trees that did not observe it.

Combining these predictions produces the

**Out-of-Bag Error**,

which provides an unbiased estimate of test performance without requiring a separate validation dataset.

This measure is called the

# <font color='blue'> Feature Importance in Tree-Based Models </font>

One of the greatest strengths of Decision Trees and their ensemble variants is that they naturally provide estimates of **feature importance**. Unlike many statistical and machine learning models, which often behave as "black boxes", tree-based algorithms can quantify the contribution of each predictor variable to the overall predictive performance of the model.

Feature importance helps answer questions such as

* Which variables are most influential?
* Which predictors contribute little to the model?
* Which variables could potentially be removed?
* Which physical quantities should receive further investigation?

These insights make tree-based methods particularly attractive in scientific research, finance, medicine, and engineering.

---

## <font color='orange'> 1. What is Feature Importance? </font>

Feature importance is a numerical measure that quantifies how useful a predictor variable is during model construction.

A variable that repeatedly produces large improvements in prediction accuracy is assigned a high importance score.

Conversely,

variables that are rarely selected or produce only small improvements receive lower scores.

For example,

```
Feature

↓

Importance

Income

0.37

Credit Score

0.26

Age

0.18

Balance

0.12

Gender

0.07
```

Although every variable contributes,

some variables clearly influence the predictions much more strongly than others.

---

## <font color='orange'> 2. Why Do Trees Naturally Measure Importance? </font>

Every split in a Decision Tree reduces the impurity of the current node.

Suppose the root node has impurity

$$
I_{parent}.
$$

After splitting,

the weighted impurity becomes

$$
I_{children}.
$$

The improvement produced by the split is

$$
\boxed{
\Delta I
=
I_{parent}
-
I_{children}.
}
$$

Every time a feature is selected,

it receives credit for this impurity reduction.

The more often a feature creates large reductions,

the more important it becomes.

---

## <font color='orange'> 3. Mean Decrease in Impurity (MDI) </font>

The most common importance measure is

**Mean Decrease in Impurity (MDI)**,

sometimes called

**Gini Importance**.

The importance of feature

$$
j
$$

is computed by summing the impurity reductions over every node that uses that feature.

Mathematically,

$$
\boxed{
FI_j
=
\sum_{t \in T_j}
\frac{N_t}{N}
\Delta I_t,
}
$$

where

* $T_j$ denotes all nodes split using feature $j$,
* $N_t$ is the number of observations reaching node $t$,
* $N$ is the total number of observations,
* $\Delta I_t$ is the impurity reduction produced by that split.

The weighting by

$$
\frac{N_t}{N}
$$

ensures that splits affecting many observations contribute more than splits occurring deep within the tree.

---

## <font color='orange'> 4. Feature Importance in Random Forests </font>

A Random Forest contains

many Decision Trees.

Each tree produces its own feature importance estimates.

The final importance becomes the average over all trees.

$$
\boxed{
FI_j
=
\frac1B
\sum_{b=1}^{B}
FI_j^{(b)},
}
$$

where

$$
B
$$

is the number of trees.

Averaging produces much more stable importance estimates than a single Decision Tree.

---

## <font color='orange'> 5. Example </font>

Suppose a Random Forest contains

100 trees.

Feature

```
Income
```

is selected

* at the root of many trees,
* near the top of several others,
* produces large reductions in Gini Impurity.

Its importance becomes

```
0.42
```

Another feature,

```
Gender,
```

appears only occasionally,

producing small impurity reductions.

Its importance becomes

```
0.03.
```

The ranking therefore becomes

| Feature | Importance |
| :--- | ---: |
| Income | 0.42 |
| Credit Score | 0.24 |
| Age | 0.18 |
| Balance | 0.13 |
| Gender | 0.03 |

---

## <font color='orange'> 6. Normalized Importance </font>

Scikit-Learn normalizes feature importances so that

$$
\boxed{
\sum_j FI_j=1.
}
$$

Consequently,

each importance score may be interpreted as the fraction of the total impurity reduction attributed to that feature.

---

## <font color='orange'> 7. Visualizing Feature Importance </font>

Feature importance is typically displayed as a horizontal bar chart.

```
Income         ███████████████

Credit Score   █████████

Age            ██████

Balance        ████

Gender         █
```

Such plots provide an immediate understanding of which variables dominate the model.

---

## <font color='orange'> 8. Permutation Importance </font>

An alternative approach is

**Permutation Importance**.

Instead of examining the tree structure,

Permutation Importance measures how much model performance deteriorates when the values of one feature are randomly shuffled.

The procedure is

1. Train the model.
2. Compute the baseline prediction accuracy.
3. Randomly shuffle one feature.
4. Recompute the prediction accuracy.
5. Measure the performance decrease.

Large performance degradation indicates an important feature.

Little change indicates an unimportant feature.

---

## <font color='orange'> 9. Why Does Permutation Importance Work? </font>

Suppose

```
Income
```

is highly predictive.

Randomly shuffling its values destroys the relationship between

```
Income
```

and the target variable.

Prediction accuracy decreases substantially.

Now suppose

```
Hair Colour
```

contains little useful information.

Randomly shuffling it changes almost nothing.

Prediction accuracy remains nearly unchanged.

The larger the performance degradation,

the greater the feature importance.

---

## <font color='orange'> 10. Mean Decrease in Impurity vs Permutation Importance </font>

| Property | MDI | Permutation |
| :--- | :---: | :---: |
| Uses Tree Structure | ✔ | ✘ |
| Uses Validation Data | ✘ | ✔ |
| Computational Cost | Low | Higher |
| Sensitive to Correlated Features | Yes | Less |
| Model Agnostic | ✘ | ✔ |

Permutation Importance can be applied to

* Random Forests,
* Gradient Boosting,
* Neural Networks,
* Support Vector Machines,
* Linear Models.

MDI is specific to tree-based methods.

---

## <font color='orange'> 11. Limitations of Feature Importance </font>

Feature importance should be interpreted carefully.

### Correlated Predictors

If two predictors contain nearly identical information,

the model may distribute importance between them.

Each variable therefore appears less important than expected.

---

### High-Cardinality Features

Variables with many unique values have more potential split points.

As a result,

MDI may assign them disproportionately high importance.

---

### Importance ≠ Causality

A large importance score indicates predictive usefulness,

not a causal relationship.

A feature may be highly predictive while having no direct causal influence on the target variable.

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Feature Importance | Contribution of a predictor to the model |

> **Key Insight:** Feature importance provides a quantitative measure of how much each predictor contributes to the predictive performance of a tree-based model. Decision Trees and Random Forests naturally estimate importance through the **Mean Decrease in Impurity (MDI)**, which accumulates the reduction in node impurity produced by each feature across the entire forest. An alternative approach, **Permutation Importance**, measures the degradation in model performance when a feature is randomly shuffled, providing a model-agnostic estimate of predictive relevance. Although feature importance is invaluable for model interpretation and variable selection, it should be interpreted cautiously, particularly in the presence of correlated predictors, and should not be regarded as evidence of a causal relationship.

# <font color='blue'> Boosting: Motivation and Fundamental Principles </font>

Boosting is one of the most influential developments in modern machine learning. Unlike Bagging, which improves predictive performance by averaging many independently trained models, Boosting constructs an ensemble **sequentially**, where each new model is trained to correct the mistakes made by the previous models.

The central philosophy of Boosting can be summarized as

> **Rather than treating every training observation equally, progressively focus learning on the examples that are hardest to predict.**

This strategy allows Boosting to produce highly accurate models while using relatively simple base learners. Today, many state-of-the-art algorithms—including **AdaBoost**, **Gradient Boosting**, **XGBoost**, **LightGBM**, and **CatBoost**—are based on this principle.

---

## <font color='orange'> 1. Motivation </font>

Suppose we train a Decision Tree to classify credit card transactions.

Although the model correctly classifies most observations,

it still makes mistakes.

```
Transaction 1

↓

Correct
```

```
Transaction 2

↓

Correct
```

```
Transaction 3

↓

Incorrect
```

```
Transaction 4

↓

Incorrect
```

A conventional machine learning algorithm simply accepts these prediction errors.

Boosting asks a different question:

> **Can we train another model that concentrates specifically on the observations that were previously misclassified?**

The answer is yes.

This simple idea forms the foundation of Boosting.

---

## <font color='orange'> 2. The Core Idea </font>

Instead of training many independent models,

Boosting constructs models sequentially.

```
Training Data

↓

Model 1

↓

Prediction Errors

↓

Model 2

↓

Remaining Errors

↓

Model 3

↓

Remaining Errors

↓

...
```

Each new learner attempts to improve the performance of the current ensemble.

Rather than replacing previous models,

their predictions are combined to produce the final prediction.

---

## <font color='orange'> 3. Sequential Learning </font>

The defining characteristic of Boosting is

**sequential learning**.

Unlike Bagging,

where every model can be trained simultaneously,

Boosting requires

```
Model 1

↓

Model 2

↓

Model 3

↓

Model 4
```

Each learner depends upon the performance of its predecessor.

Consequently,

Boosting cannot generally be parallelized as easily as Bagging.

---

## <font color='orange'> 4. Weak Learners </font>

Boosting is based on the surprising observation that

many **weak learners**

can be combined into a highly accurate predictor.

A weak learner performs only slightly better than random guessing.

For Decision Trees,

the most common weak learner is a

**Decision Stump**,

which is simply a tree containing a single split.

```
Root

↓

Leaf      Leaf
```

Although a Decision Stump is extremely simple,

Boosting combines many such learners into a powerful predictive model.

---

## <font color='orange'> 5. Why Weak Learners? </font>

One might expect stronger models to produce better ensembles.

However,

very complex models often

* overfit,
* make highly correlated errors,
* leave little opportunity for later learners to improve.

Simple learners,

on the other hand,

leave meaningful residual errors that subsequent models can correct.

Consequently,

Boosting generally performs best using relatively simple base learners.

---

## <font color='orange'> 6. Error Correction </font>

Suppose the first learner produces

```
100 Predictions

↓

90 Correct

10 Incorrect
```

The second learner focuses primarily on those

```
10 Incorrect
```

observations.

After combining both learners,

perhaps only

```
4 Incorrect
```

remain.

The third learner now concentrates on those remaining difficult observations.

The prediction error therefore decreases progressively as more learners are added.

---

## <font color='orange'> 7. Additive Learning </font>

Boosting constructs the final model incrementally.

Suppose

$$
f_1(\mathbf{x})
$$

is the first learner.

The second learner

$$
f_2(\mathbf{x})
$$

is added,

producing

$$
F_2(\mathbf{x})
=
f_1(\mathbf{x})
+
f_2(\mathbf{x}).
$$

Continuing,

the complete ensemble becomes

$$
\boxed{
F(\mathbf{x})
=
\sum_{m=1}^{M}
\alpha_m
f_m(\mathbf{x}),
}
$$

where

* $f_m(\mathbf{x})$ is the $m^{\text{th}}$ learner,
* $\alpha_m$ is its contribution (weight).

The ensemble therefore grows gradually,

one learner at a time.

---

## <font color='orange'> 8. Bias Reduction </font>

Recall the Bias–Variance decomposition,

$$
\text{Error}
=
\text{Bias}^2
+
\text{Variance}
+
\text{Noise}.
$$

Unlike Bagging,

which primarily reduces

```
Variance,
```

Boosting mainly reduces

```
Bias.
```

Each learner improves the approximation to the true underlying relationship,

gradually reducing systematic prediction error.

---

## <font color='orange'> 9. Bagging vs Boosting </font>

The philosophies of Bagging and Boosting are fundamentally different.

| Property | Bagging | Boosting |
| :--- | :---: | :---: |
| Training | Parallel | Sequential |
| Data | Bootstrap Samples | Same Dataset (reweighted or residuals) |
| Goal | Reduce Variance | Reduce Bias |
| Base Learners | Independent | Dependent |
| Combination | Averaging / Voting | Weighted Sum |
| Parallelizable | ✔ | ✘ |

Both methods improve predictive performance,

but they do so in very different ways.

---

## <font color='orange'> 10. General Boosting Algorithm </font>

Although different Boosting algorithms implement different update rules,

their overall structure is remarkably similar.

```
Train Weak Learner

↓

Evaluate Errors

↓

Increase Focus on Difficult Observations

↓

Train Next Learner

↓

Combine Predictions

↓

Repeat
```

Each iteration improves the current ensemble.

---

## <font color='orange'> 11. Advantages of Boosting </font>

Boosting offers several important advantages.

* High predictive accuracy.
* Reduces systematic errors.
* Excellent generalization.
* Handles complex nonlinear relationships.
* Often outperforms Bagging.
* Works well with relatively simple base learners.

Many of today's highest-performing machine learning algorithms are based on Boosting.

---

## <font color='orange'> 12. Limitations </font>

Boosting also has several limitations.

* Sequential training increases computation time.
* More sensitive to noisy data and outliers.
* Requires careful hyperparameter tuning.
* May overfit if too many learners are added.
* More difficult to parallelize.

These challenges motivate the use of techniques such as learning-rate regularization and early stopping.

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Boosting | Sequential ensemble learning |

> **Key Insight:** Boosting improves predictive performance by constructing an ensemble of weak learners sequentially, with each new learner focusing on correcting the mistakes made by the existing ensemble. Unlike Bagging, which reduces variance through parallel averaging, Boosting progressively reduces bias by refining the model at every iteration. This additive learning strategy enables a collection of simple models—often shallow Decision Trees—to achieve remarkable predictive accuracy. Modern algorithms such as AdaBoost, Gradient Boosting, XGBoost, LightGBM, and CatBoost are all built upon this fundamental principle, making Boosting one of the most influential paradigms in contemporary machine learning.
