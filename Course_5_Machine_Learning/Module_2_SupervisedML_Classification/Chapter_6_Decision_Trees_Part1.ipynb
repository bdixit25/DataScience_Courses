# <font color='blue'> Introduction to Decision Trees </font>

Decision Trees are among the most intuitive and widely used supervised machine learning algorithms. They can be applied to both **classification** and **regression** problems and are particularly valued for their simplicity, interpretability, and ability to model nonlinear relationships without requiring complex mathematical transformations.

Unlike algorithms such as Logistic Regression or Support Vector Machines, which learn mathematical equations describing the relationship between variables, Decision Trees learn a sequence of **if-then decision rules** that recursively partition the feature space into increasingly homogeneous subsets.

The resulting model closely resembles the decision-making process used by humans and can therefore be visualized and interpreted with relative ease.

---

## <font color='orange'> 1. Supervised Learning </font>

Decision Trees belong to the family of **supervised learning algorithms**.

During training, the algorithm receives

* a set of input features

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_n),
$$

and

* the corresponding target values

$$
y.
$$

The objective is to learn a mapping

$$
f(\mathbf{x})
\rightarrow
y
$$

that can accurately predict the target value for previously unseen observations.

---

## <font color='orange'> 2. Classification and Regression Trees </font>

Decision Trees can solve two fundamentally different problems.

### Classification Trees

The target variable is categorical.

Examples include

* Spam or Not Spam
* Fraud or Non-Fraud
* Benign or Malignant Tumour

The objective is to predict the class label.

---

### Regression Trees

The target variable is continuous.

Examples include

* House Prices
* Insurance Charges
* Temperature Prediction

The objective is to predict a numerical value.

Although the overall tree-building algorithm remains similar, the splitting criteria differ between classification and regression.

---

## <font color='orange'> 3. The Fundamental Idea </font>

Rather than fitting a mathematical equation, a Decision Tree repeatedly asks a sequence of questions about the data.

For example,

```
Is Income > £60,000?

        Yes                No
         |                  |
Owns House?           Age > 35?
     |                    |
```

Each question partitions the dataset into smaller subsets.

The algorithm continues asking questions until the observations within each subset become sufficiently similar.

This process is known as

**Recursive Binary Splitting**.

---

## <font color='orange'> 4. Why Are They Called Trees? </font>

The structure resembles an upside-down tree.

```
            Root

        /           \

    Internal      Internal
      Node          Node

    /     \       /      \

 Leaf   Leaf   Leaf    Leaf
```

The tree consists of

* one Root Node,
* several Internal Nodes,
* Terminal Nodes (Leaves).

Each branch represents a decision rule,

while each leaf contains the final prediction.

---

## <font color='orange'> 5. Decision Rules </font>

Each internal node contains a rule of the form

$$
x_j<c
$$

or

$$
x_j\ge c,
$$

where

* $x_j$ is one of the input variables,
* $c$ is a threshold.

For example,

```
Age < 45 ?
```

or

```
Income > £80,000 ?
```

The observation follows one branch depending upon whether the condition is satisfied.

---

## <font color='orange'> 6. Recursive Partitioning </font>

Each split divides the feature space into smaller regions.

Suppose the original dataset contains

1000

observations.

The first split might divide the data into

```
600 observations

400 observations
```

Each subset is then split again,

producing progressively smaller regions.

The splitting process continues recursively until a stopping criterion is satisfied.

---

## <font color='orange'> 7. Learning Without Equations </font>

One of the defining characteristics of Decision Trees is that they do **not** assume any mathematical relationship between the features and the target variable.

Unlike

* Linear Regression,
* Logistic Regression,
* Support Vector Machines,

Decision Trees require

* no linearity,
* no normality,
* no feature scaling.

Instead,

they simply learn decision rules directly from the training data.

---

## <font color='orange'> 8. Advantages of Decision Trees </font>

Decision Trees possess several attractive properties.

* Easy to understand.
* Highly interpretable.
* Naturally handles nonlinear relationships.
* Works with numerical and categorical variables.
* Requires little data preprocessing.
* Produces explicit decision rules.

These advantages explain why Decision Trees are widely used in finance, medicine, engineering, and many other disciplines.

---

## <font color='orange'> 9. Limitations of Decision Trees </font>

Despite their simplicity,

Decision Trees also exhibit several weaknesses.

* Prone to overfitting.
* Small changes in the training data may produce different trees.
* High variance.
* Often less accurate than ensemble methods.

Fortunately,

these limitations can be overcome through methods such as

* Random Forests,
* Gradient Boosting,
* XGBoost,

which combine many Decision Trees into a single predictive model.

---

## <font color='orange'> 10. Applications </font>

Decision Trees have been successfully applied in numerous domains.

Examples include

* Medical diagnosis
* Credit approval
* Fraud detection
* Customer segmentation
* Insurance risk assessment
* Loan approval
* Equipment fault diagnosis
* Predictive maintenance

Their transparent decision-making process makes them particularly attractive in applications requiring explainable artificial intelligence (XAI).

---

## <font color='purple'> 11. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Learning Type | Supervised Learning |
| Tasks | Classification and Regression |
| Model | Tree of decision rules |
| Splitting Strategy | Recursive Binary Splitting |
| Decision Rule | Feature threshold comparison |
| Output | Class label or numerical value |
| Major Strength | Interpretability |
| Major Weakness | High variance and overfitting |

> **Key Insight:** Decision Trees differ fundamentally from many other machine learning algorithms because they do not learn a mathematical equation relating the input variables to the target. Instead, they recursively partition the feature space using a sequence of simple decision rules, creating a tree-like structure that is both intuitive and highly interpretable. Their ability to model nonlinear relationships without requiring feature scaling or distributional assumptions makes them one of the most versatile algorithms in supervised learning and the foundation of many modern ensemble techniques such as Random Forests and Gradient Boosting.

# <font color='blue'> Structure of a Decision Tree </font>

A Decision Tree derives its name from its graphical structure, which closely resembles an upside-down tree. Every Decision Tree, regardless of whether it is used for classification or regression, consists of a collection of connected **nodes** joined by **branches**. Each node represents a decision, while each branch represents one possible outcome of that decision.

Understanding these structural components is essential before studying how Decision Trees are constructed and optimized.

---

## <font color='orange'> 1. Anatomy of a Decision Tree </font>

A Decision Tree consists of five basic elements:

* Root Node
* Internal Nodes
* Branches
* Leaf Nodes (Terminal Nodes)
* Decision Paths

A simplified Decision Tree is illustrated below.

```
                      Root Node
                  Is Age < 40?

                  /          \
               Yes            No
                |              |
          Internal Node    Internal Node
        Income > £50k?     Owns House?

          /      \           /      \
       Yes       No       Yes      No
        |         |         |        |
     Leaf      Leaf      Leaf     Leaf
```

Each observation begins at the root and travels through the tree until it reaches a terminal node.

---

## <font color='orange'> 2. Root Node </font>

The **Root Node** is the highest node in the tree.

It represents the entire training dataset before any partitioning has taken place.

The first split made by the algorithm always occurs at the root.

For example,

```
Is Age < 40?
```

Every observation is evaluated using this rule.

Depending on the outcome,

each observation moves to one of the child nodes.

A Decision Tree always contains **exactly one** root node.

---

## <font color='orange'> 3. Internal Nodes </font>

An **Internal Node** (sometimes called a Decision Node) performs another test after the initial split.

Each internal node contains a decision rule such as

$$
x_j < c
$$

or

$$
x_j \ge c,
$$

where

* $x_j$ is one of the predictor variables,
* $c$ is a threshold chosen by the learning algorithm.

For example,

```
Income > £50,000?
```

Every observation arriving at that node is evaluated again.

If the stopping criteria have not yet been satisfied, additional internal nodes are created.

---

## <font color='orange'> 4. Branches </font>

Branches connect one node to another.

Each branch represents one possible outcome of a decision.

For a binary tree,

each node has two branches.

For example,

```
Age < 40 ?

      Yes
       |

      Node

       |

      No
```

Every observation follows exactly one branch at each node.

As the observation moves through the tree,

the feature space is partitioned into progressively smaller regions.

---

## <font color='orange'> 5. Leaf Nodes (Terminal Nodes) </font>

A **Leaf Node** is the final node reached by an observation.

Leaf nodes contain no further decision rules.

Instead,

they produce the final prediction.

For a **classification tree**,

the leaf predicts the class label,

for example,

```
Fraud
```

or

```
Non-Fraud
```

For a **regression tree**,

the leaf predicts a numerical value,

for example,

```
Predicted House Price
=
£325,000
```

Every path through the tree ends at a leaf node.

---

## <font color='orange'> 6. Parent and Child Nodes </font>

Nodes are connected through parent-child relationships.

Suppose

```
           Parent

          /      \

      Child    Child
```

The upper node is called the **parent**,

while the lower nodes are its **children**.

Every node,

except the root,

has exactly one parent.

Internal nodes may have multiple children,

whereas leaf nodes have none.

---

## <font color='orange'> 7. Sibling Nodes </font>

Nodes sharing the same parent are called

**Sibling Nodes**.

For example,

```
          Income > 50?

          /         \

      Node A      Node B
```

Node A and Node B are siblings because they originate from the same split.

---

## <font color='orange'> 8. Decision Paths </font>

A **Decision Path** is the sequence of decisions followed by an observation from the root to a leaf.

Consider the following tree.

```
Age < 40?

        Yes

Income > 50?

        No

Owns House?

        Yes

Approved
```

The decision path becomes

```
Age < 40

↓

Income ≤ 50

↓

Owns House

↓

Loan Approved
```

Each observation follows exactly one decision path.

---

## <font color='orange'> 9. Binary Trees </font>

Most machine learning implementations construct

**Binary Decision Trees**.

Every split divides the data into exactly two subsets.

Mathematically,

each split partitions the feature space into

$$
R_1
$$

and

$$
R_2.
$$

Binary trees are computationally efficient and simplify the optimization process.

Although trees with multiple branches are possible,

binary trees are overwhelmingly preferred in modern machine learning libraries.

---

## <font color='orange'> 10. Tree Depth </font>

The **depth** of a tree is the maximum number of splits required to reach a leaf node.

For example,

```
Root

↓

Node

↓

Node

↓

Leaf
```

has depth

$$
3.
$$

Shallow trees

* are simpler,
* have higher bias,
* are less likely to overfit.

Deep trees

* model increasingly complex relationships,
* have lower bias,
* have higher variance,
* are more susceptible to overfitting.

Tree depth therefore plays an important role in controlling model complexity.

---

## <font color='orange'> 11. Number of Leaves </font>

The number of leaf nodes determines the number of regions into which the feature space is partitioned.

Suppose a tree contains

16

leaf nodes.

The feature space has therefore been divided into

16

distinct regions,

each with its own prediction.

Increasing the number of leaves generally increases model complexity.

---

## <font color='orange'> 12. Example of a Complete Prediction </font>

Suppose the tree predicts whether a loan should be approved.

```
Age < 35?

       Yes

Income > £60k?

       Yes

Credit Score > 700?

       No

Reject Loan
```

An applicant with

* Age = 30,
* Income = £75,000,
* Credit Score = 680,

would follow the path

```
Root

↓

Yes

↓

Yes

↓

No

↓

Reject
```

Notice that the prediction requires only a sequence of simple logical comparisons.

No complicated mathematical equations are evaluated.

---

## <font color='orange'> 13. Why This Structure is Useful </font>

The hierarchical structure of Decision Trees provides several important advantages.

* Easy to visualize.
* Highly interpretable.
* Mirrors human decision making.
* Naturally handles nonlinear relationships.
* Produces explicit decision rules.

These characteristics explain why Decision Trees remain popular in fields requiring explainable machine learning.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Component | Description |
| :--- | :--- |
| Root Node | Top-most node containing the full dataset |
| Internal Node | Intermediate node containing a decision rule |
| Branch | Connection representing one outcome of a decision |
| Leaf Node | Terminal node containing the prediction |
| Parent Node | Node immediately above another node |
| Child Node | Node created after a split |
| Sibling Nodes | Nodes sharing the same parent |
| Decision Path | Sequence of decisions from root to leaf |
| Tree Depth | Maximum number of splits from root to leaf |

> **Key Insight:** A Decision Tree is a hierarchical collection of decision rules that recursively partitions the feature space into smaller and increasingly homogeneous regions. Every prediction begins at the **root node**, follows a unique sequence of **branches** through a series of **internal decision nodes**, and terminates at a **leaf node**, where the final prediction is made. This intuitive structure closely resembles human reasoning, making Decision Trees among the most interpretable machine learning models and forming the foundation of many modern ensemble algorithms.

# <font color='blue'> Recursive Binary Splitting </font>

The construction of a Decision Tree begins with the entire training dataset contained within the **root node**. The objective is to repeatedly divide (or **split**) this dataset into smaller and increasingly homogeneous subsets until each terminal node contains observations that are as similar as possible.

This procedure is known as **Recursive Binary Splitting**, and it forms the fundamental learning algorithm behind Decision Trees.

The term

* **Recursive** means that the same splitting procedure is applied repeatedly to every newly created node.
* **Binary** means that each split divides the data into exactly two subsets.

By repeatedly partitioning the feature space, the algorithm constructs a hierarchical sequence of decision rules that ultimately produces the final prediction.

---

## <font color='orange'> 1. The Basic Idea </font>

Suppose a dataset contains several predictor variables,

$$
\mathbf{x}
=
(x_1,x_2,\ldots,x_p),
$$

together with a target variable

$$
y.
$$

Initially,

every observation belongs to a single region,

$$
R.
$$

The first task is to determine

* which feature should be used,
* where that feature should be split.

For example,

```
Age < 40 ?
```

or

```
Income > £60,000 ?
```

The chosen split divides the original region into two new regions,

$$
R_1
\quad\text{and}\quad
R_2.
$$

The algorithm then repeats exactly the same process within each new region.

---

## <font color='orange'> 2. Binary Partitioning of the Feature Space </font>

Every split partitions the feature space into two disjoint regions.

Suppose a split is performed using feature

$$
x_j
$$

at threshold

$$
s.
$$

The resulting regions are

$$
\boxed{
R_1(j,s)
=
\{
\mathbf{x}
:
x_j<s
\}
}
$$

and

$$
\boxed{
R_2(j,s)
=
\{
\mathbf{x}
:
x_j\ge s
\}.
}
$$

Every observation belongs to exactly one of these regions.

No observation can belong to both simultaneously.

---

## <font color='orange'> 3. Example </font>

Suppose the feature

```
Age
```

is selected,

and the threshold is

```
40 years.
```

The split becomes

```
                Age < 40 ?

               /          \

             Yes          No

         Region R₁     Region R₂
```

Every observation with

```
Age < 40
```

moves into

$$
R_1,
$$

while every remaining observation moves into

$$
R_2.
$$

The process then repeats independently within each region.

---

## <font color='orange'> 4. Why Binary Splits? </font>

Modern Decision Tree algorithms almost always use binary splits because they

* simplify the optimization,
* produce efficient algorithms,
* are easy to visualize,
* naturally extend to ensemble methods.

Although multi-way splits are possible,

binary partitioning generally produces more flexible trees.

---

## <font color='orange'> 5. Searching for the Best Split </font>

For each predictor variable,

the algorithm considers many possible threshold values.

Suppose

```
Age
```

contains

```
18, 25, 30, 42, 51, 60
```

Possible candidate thresholds include

```
21.5

27.5

36

46.5

55.5
```

Every possible threshold is evaluated.

The split producing the greatest improvement is selected.

Thus,

the algorithm performs an exhaustive search over candidate split locations.

---

## <font color='orange'> 6. Measuring Split Quality </font>

Not every split improves the tree.

A good split should produce child nodes that are **more homogeneous** than the parent node.

For a classification problem,

this means observations belonging predominantly to a single class.

For a regression problem,

this means observations having similar numerical values.

Mathematically,

the algorithm minimizes an impurity function,

denoted

$$
I(R).
$$

The best split is the one that produces the largest reduction in impurity.

---

## <font color='orange'> 7. Recursive Construction </font>

Once the best split has been selected,

the algorithm applies the same procedure independently to both child nodes.

For example,

```
                    Root

               Income > 60?

                /        \

             Yes          No

        Age > 45?     Age > 30?

         /    \         /     \

      Leaf   Leaf    Leaf    Leaf
```

Notice that every child node becomes the root of a smaller subtree.

The algorithm therefore calls itself repeatedly.

This repeated self-application is precisely what is meant by

**recursion**.

---

## <font color='orange'> 8. Mathematical Representation </font>

Suppose a node contains observations belonging to region

$$
R.
$$

The algorithm searches over every feature

$$
j
$$

and every possible threshold

$$
s.
$$

The objective is

$$
\boxed{
(j^*,s^*)
=
\arg\min_{j,s}
\left[
I(R_1(j,s))
+
I(R_2(j,s))
\right].
}
$$

In words,

the algorithm chooses

* the feature,
* the threshold,

that minimize the combined impurity of the two child nodes.

This optimization lies at the heart of Decision Tree learning.

---

## <font color='orange'> 9. Recursive Splitting Example </font>

Suppose the original dataset contains

1000

observations.

The algorithm may proceed as follows.

```
1000 observations

↓

Split on Income

↓

600           400

↓

Split again

↓

320 280     150 250

↓

Continue...
```

Eventually,

every branch terminates at a leaf node.

---

## <font color='orange'> 10. When Does the Splitting Stop? </font>

Recursive splitting cannot continue indefinitely.

Typical stopping conditions include

* maximum tree depth reached,
* minimum number of observations in a node,
* node impurity becomes zero,
* further splits produce negligible improvement.

These stopping criteria prevent excessively deep trees that overfit the training data.

The stopping rules will be discussed in detail later.

---

## <font color='orange'> 11. Recursive Binary Splitting Algorithm </font>

The complete algorithm may be summarized as follows.

```
Start with entire dataset

↓

Evaluate every feature

↓

Evaluate every threshold

↓

Choose best split

↓

Create two child nodes

↓

Repeat for each child

↓

Continue until stopping criterion

↓

Return completed tree
```

This recursive procedure automatically discovers increasingly informative decision rules without requiring any prior knowledge of the relationships within the data.

---

## <font color='orange'> 12. Computational Complexity </font>

Suppose

* there are

$$
p
$$

features,

* and

$$
n
$$

training observations.

At each node,

the algorithm must examine numerous candidate thresholds for every feature.

Consequently,

constructing deep trees can become computationally expensive.

Fortunately,

modern implementations employ several optimization techniques that significantly accelerate this search.

---

## <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Recursive | Apply the same splitting procedure repeatedly |
| Binary | Every split produces exactly two child nodes |
| Region | Subset of the feature space |
| Split | Partition of one region into two smaller regions |
| Threshold | Numerical value defining the split |
| Objective | Minimize impurity after splitting |
| Search Strategy | Evaluate every candidate feature and threshold |
| Result | Hierarchical sequence of decision rules |

> **Key Insight:** Recursive Binary Splitting is the fundamental learning algorithm underlying Decision Trees. Beginning with the entire dataset, the algorithm systematically searches over every feature and every possible threshold to identify the split that produces the greatest reduction in impurity. Each split partitions the feature space into two increasingly homogeneous regions, and the same procedure is then applied recursively to each child node. Through this repeated partitioning process, the Decision Tree automatically discovers a hierarchy of decision rules that captures complex relationships within the data while remaining highly interpretable.

# <font color='blue'> Measuring Node Impurity </font>

The objective of a Decision Tree is to partition the feature space into regions that contain observations with similar target values. To determine whether a proposed split is useful, the algorithm requires a quantitative measure of how "pure" or "homogeneous" each node is.

This quantity is known as the **node impurity**.

A node is considered **pure** if all observations within that node belong to the same class (for classification) or have very similar numerical values (for regression). Conversely, a node is considered **impure** if it contains a mixture of different classes or highly variable target values.

The Decision Tree algorithm therefore seeks splits that **reduce node impurity as much as possible**.

---

## <font color='orange'> 1. What is Node Impurity? </font>

Consider a binary classification problem with two classes:

* Fraud
* Non-Fraud

Suppose a node contains the following observations.

```
Fraud

Fraud

Fraud

Fraud

Fraud
```

Every observation belongs to the same class.

The node is therefore perfectly pure.

---

Now consider another node.

```
Fraud

Fraud

Non-Fraud

Fraud

Non-Fraud
```

The observations belong to multiple classes.

This node is considerably more impure.

A good Decision Tree attempts to transform impure nodes into increasingly pure child nodes.

---

## <font color='orange'> 2. Measuring Class Proportions </font>

Suppose a node contains

$$
N
$$

observations belonging to

$$
K
$$

classes.

The proportion of observations belonging to class

$$
k
$$

is

$$
\boxed{
p_k
=
\frac{N_k}{N},
}
$$

where

* $N_k$ is the number of observations belonging to class $k$,
* $N$ is the total number of observations in the node.

These class probabilities form the basis of every impurity measure used in Decision Trees.

---

## <font color='orange'> 3. Characteristics of a Good Impurity Measure </font>

An impurity measure should satisfy several intuitive properties.

### Pure Node

If every observation belongs to the same class,

$$
p_1=1,
$$

then

the impurity should be

$$
0.
$$

---

### Mixed Node

If observations are evenly distributed between classes,

the impurity should be as large as possible.

For binary classification,

this occurs when

$$
p_1=p_2=0.5.
$$

---

### Symmetry

The impurity should depend only on the proportions of the classes,

not on the class labels themselves.

Thus,

```
70% Fraud

30% Non-Fraud
```

should produce exactly the same impurity as

```
30% Fraud

70% Non-Fraud.
```

---

## <font color='orange'> 4. Example of Node Purity </font>

Suppose a node contains

20

observations.

### Node A

```
20 Fraud

0 Non-Fraud
```

Class probabilities become

$$
p_{Fraud}=1,
$$

$$
p_{Non-Fraud}=0.
$$

This node is perfectly pure.

---

### Node B

```
10 Fraud

10 Non-Fraud
```

The probabilities become

$$
p_{Fraud}=0.5,
$$

$$
p_{Non-Fraud}=0.5.
$$

This represents the maximum possible impurity for a binary classification problem.

---

### Node C

```
18 Fraud

2 Non-Fraud
```

Although not perfectly pure,

most observations belong to one class.

Its impurity therefore lies somewhere between Node A and Node B.

---

## <font color='orange'> 5. Why Minimize Impurity? </font>

Suppose a parent node contains

```
50 Fraud

50 Non-Fraud
```

This node is highly impure.

A candidate split produces

```
Left Child

48 Fraud

5 Non-Fraud
```

and

```
Right Child

2 Fraud

45 Non-Fraud
```

Each child node is much more homogeneous than the original parent node.

This is considered an excellent split because it greatly reduces impurity.

Conversely,

if a split produces two child nodes with similar class mixtures,

the split provides little useful information.

---

## <font color='orange'> 6. General Objective of Tree Construction </font>

Suppose

$$
I(R)
$$

denotes the impurity of region

$$
R.
$$

After a candidate split,

the parent node is divided into two child regions,

$$
R_1
$$

and

$$
R_2.
$$

The Decision Tree algorithm seeks the split that minimizes the weighted impurity of the child nodes,

$$
\boxed{
I_{split}
=
\frac{N_1}{N}I(R_1)
+
\frac{N_2}{N}I(R_2),
}
$$

where

* $N_1$ is the number of observations in the left child,
* $N_2$ is the number of observations in the right child,
* $N=N_1+N_2$.

The weights ensure that larger child nodes contribute proportionally more to the overall impurity.

---

## <font color='orange'> 7. Impurity Reduction </font>

Rather than minimizing the child impurity directly,

many Decision Tree algorithms maximize the **reduction in impurity**,

defined as

$$
\boxed{
\Delta I
=
I_{parent}
-
I_{split}.
}
$$

The larger

$$
\Delta I,
$$

the better the split.

A perfect split would produce completely pure child nodes,

maximizing the impurity reduction.

---

## <font color='orange'> 8. Classification vs Regression Trees </font>

The concept of impurity differs slightly depending on the prediction task.

### Classification Trees

Impurity measures the **mixture of class labels**.

Common impurity measures include

* Entropy,
* Gini Impurity,
* Misclassification Error.

---

### Regression Trees

Instead of class labels,

the objective is to reduce the variability of numerical target values.

Typical measures include

* Variance,
* Mean Squared Error (MSE),
* Mean Absolute Error (MAE).

These measures will be discussed later when Decision Tree Regression is introduced.

---

## <font color='orange'> 9. Common Impurity Measures </font>

Several mathematical functions satisfy the desired properties of an impurity measure.

The most commonly used are

| Impurity Measure | Common Algorithm |
| :--- | :--- |
| Entropy | ID3, C4.5 |
| Information Gain | ID3, C4.5 |
| Gini Impurity | CART |
| Misclassification Error | Tree Pruning and Evaluation |

Although these measures differ mathematically,

their common objective is to identify the split producing the greatest increase in node purity.

---

## <font color='orange'> 10. Visual Interpretation </font>

Node impurity may be viewed as a measure of uncertainty.

```
Pure Node

████████████████

↓

Low Uncertainty

↓

Low Impurity
```

```
Mixed Node

██████░░████░░░

↓

High Uncertainty

↓

High Impurity
```

A Decision Tree attempts to move progressively from highly uncertain nodes toward perfectly homogeneous terminal nodes.

---

## <font color='purple'> 11. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Node Purity | Degree to which observations belong to one class |
| Node Impurity | Degree of class mixture within a node |
| Pure Node | Contains observations from a single class |
| Impure Node | Contains observations from multiple classes |
| Split Quality | Reduction in impurity after splitting |
| Objective | Maximize impurity reduction |
| Classification Trees | Use class-based impurity measures |
| Regression Trees | Use variance-based impurity measures |

> **Key Insight:** The construction of a Decision Tree is fundamentally an optimization problem in which every candidate split is evaluated according to how much it reduces **node impurity**. A good split partitions the data into child nodes that are more homogeneous than the parent node, thereby increasing the certainty of future predictions. Although several impurity measures exist—including **Entropy**, **Gini Impurity**, and **Misclassification Error**—they all share the same objective: to identify the decision rule that produces the greatest improvement in node purity while recursively partitioning the feature space.

# <font color='blue'> Entropy </font>

Entropy is one of the most important concepts in Decision Tree learning. It provides a quantitative measure of the **uncertainty**, **disorder**, or **impurity** within a collection of observations.

Originally introduced by **Claude Shannon** in his pioneering work on Information Theory, entropy measures how unpredictable the outcome of a random event is. In the context of Decision Trees, entropy measures how mixed the class labels are within a node.

A node containing observations from a single class has **zero entropy**, whereas a node containing an equal mixture of classes has **maximum entropy**.

Decision Tree algorithms therefore seek splits that reduce entropy as much as possible.

---

## <font color='orange'> 1. Intuition Behind Entropy </font>

Imagine tossing a coin.

### Case 1: Two-Headed Coin

Suppose the coin always lands on heads.

```
Heads

Heads

Heads

Heads
```

There is no uncertainty.

Every outcome is known in advance.

Consequently,

the entropy is

$$
0.
$$

---

### Case 2: Fair Coin

Now consider an ordinary coin.

```
Heads

Tails
```

Before tossing the coin,

both outcomes are equally likely.

There is maximum uncertainty.

Consequently,

the entropy is at its maximum.

---

The same idea applies to Decision Trees.

If every observation belongs to one class,

there is no uncertainty.

If observations are evenly divided among classes,

there is maximum uncertainty.

---

## <font color='orange'> 2. Measuring Class Probabilities </font>

Suppose a node contains

$$
K
$$

classes.

The probability that an observation belongs to class

$$
k
$$

is

$$
\boxed{
p_k
=
\frac{N_k}{N},
}
$$

where

* $N_k$ is the number of observations in class $k$,
* $N$ is the total number of observations.

Since these are probabilities,

they satisfy

$$
\sum_{k=1}^{K}p_k=1.
$$

---

## <font color='orange'> 3. Information Content </font>

Before defining entropy,

we first introduce the concept of **information content**.

Suppose an event occurs with probability

$$
p.
$$

The information obtained when observing that event is defined as

$$
\boxed{
I(p)
=
-\log_2(p).
}
$$

This definition satisfies several desirable properties.

### Certain Event

If

$$
p=1,
$$

then

$$
I(1)
=
-\log_2(1)
=
0.
$$

No new information is gained because the outcome was already certain.

---

### Rare Event

Suppose

$$
p=0.01.
$$

Then

$$
I(0.01)
=
-\log_2(0.01)
\approx
6.64.
$$

Rare events contain much more information because they are unexpected.

Thus,

**unlikely events carry more information than common events.**

---

## <font color='orange'> 4. Derivation of Entropy </font>

Since each class occurs with probability

$$
p_k,
$$

its information content is

$$
-\log_2(p_k).
$$

The average information contained within the node is therefore

$$
\boxed{
H
=
\sum_{k=1}^{K}
p_k
I(p_k).
}
$$

Substituting

$$
I(p_k)
=
-\log_2(p_k)
$$

gives Shannon's entropy equation,

$$
\boxed{
H
=
-
\sum_{k=1}^{K}
p_k
\log_2(p_k).
}
$$

This is the entropy used in Decision Trees.

---

## <font color='orange'> 5. Binary Classification </font>

For two classes,

the entropy simplifies to

$$
\boxed{
H
=
-
p
\log_2(p)
-
(1-p)
\log_2(1-p).
}
$$

where

$$
p
$$

is the probability of one class.

This equation is widely used in binary classification problems.

---

## <font color='orange'> 6. Example 1 — Pure Node </font>

Suppose a node contains

```
20 Fraud

0 Non-Fraud
```

Then

$$
p=1.
$$

Therefore,

$$
H
=
-
1\log_2(1)
-
0\log_2(0).
$$

Using the convention

$$
0\log(0)=0,
$$

we obtain

$$
\boxed{
H=0.
}
$$

A perfectly pure node has zero entropy.

---

## <font color='orange'> 7. Example 2 — Maximum Uncertainty </font>

Suppose

```
10 Fraud

10 Non-Fraud
```

Then

$$
p=0.5.
$$

Substituting,

$$
H
=
-
0.5
\log_2(0.5)
-
0.5
\log_2(0.5).
$$

Since

$$
\log_2(0.5)=-1,
$$

we obtain

$$
H
=
1.
$$

This is the maximum possible entropy for binary classification.

---

## <font color='orange'> 8. Example 3 — Intermediate Purity </font>

Suppose

```
18 Fraud

2 Non-Fraud
```

Then

$$
p=0.9.
$$

The entropy becomes

$$
H
=
-
0.9
\log_2(0.9)
-
0.1
\log_2(0.1)
\approx
0.469.
$$

This value lies between

$$
0
$$

and

$$
1,
$$

indicating moderate impurity.

---

## <font color='orange'> 9. Properties of Entropy </font>

Entropy possesses several important mathematical properties.

### Non-Negative

Entropy is always

$$
H\ge0.
$$

---

### Maximum for Equal Probabilities

Entropy reaches its largest value when all classes occur with equal probability.

For binary classification,

this occurs when

$$
p=0.5.
$$

---

### Minimum for Pure Nodes

Entropy becomes

$$
0
$$

only when every observation belongs to one class.

---

### Symmetry

Entropy depends only on the probabilities,

not on the labels themselves.

Thus,

```
70% Fraud

30% Non-Fraud
```

produces exactly the same entropy as

```
30% Fraud

70% Non-Fraud.
```

---

## <font color='orange'> 10. Entropy Curve </font>

For binary classification,

the entropy curve has the characteristic shape

```
Entropy

1.0       ●

         / \

        /   \

       /     \

0.0 ●-----------●

    0   0.5    1

      Class Probability
```

The maximum occurs when the classes are equally probable.

---

## <font color='orange'> 11. Why Decision Trees Minimize Entropy </font>

The objective of a Decision Tree is to create increasingly pure nodes.

Consequently,

the algorithm searches for the split producing the largest reduction in entropy.

A good split transforms

```
Mixed Node

↓

Pure Child Nodes
```

thereby reducing uncertainty and improving predictive accuracy.

The amount by which entropy decreases is called the **Information Gain**, which is discussed in the next section.

---

## <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Entropy | Measure of uncertainty or impurity |
| Formula | $H=-\sum p_k\log_2(p_k)$ |
| Pure Node | Entropy = 0 |
| Maximum Entropy | Equal class probabilities |
| Units | Bits of information |
| Used By | ID3 and C4.5 Decision Trees |

> **Key Insight:** Entropy provides a mathematical measure of uncertainty within a Decision Tree node. Derived from Shannon's Information Theory, it quantifies how mixed the class labels are by computing the average information required to identify the class of an observation. Pure nodes have zero entropy because their outcomes are completely predictable, while nodes containing equally probable classes exhibit maximum entropy. Decision Tree algorithms therefore seek splits that minimize entropy, thereby producing increasingly homogeneous nodes and reducing uncertainty throughout the tree.

# <font color='blue'> Information Gain </font>

Entropy measures the amount of uncertainty or impurity within a node. However, when constructing a Decision Tree, the objective is not simply to measure impurity—it is to determine **how much the impurity decreases after a candidate split**.

This reduction in entropy is known as **Information Gain**.

Information Gain provides a quantitative measure of how much "information" is obtained by partitioning the data according to a particular feature. The larger the Information Gain, the more useful the split.

The **ID3** and **C4.5** Decision Tree algorithms therefore select the feature that produces the **maximum Information Gain** at each node.

---

## <font color='orange'> 1. Motivation </font>

Suppose a parent node contains a mixture of two classes.

```
Parent Node

Fraud
Fraud
Fraud
Non-Fraud
Non-Fraud
Fraud
```

This node has relatively high entropy because the class labels are mixed.

Now suppose we split the node according to

```
Amount > £500 ?
```

producing

```
Left Child

Fraud
Fraud
Fraud
Fraud
```

and

```
Right Child

Non-Fraud
Non-Fraud
```

Both child nodes are perfectly pure.

The uncertainty has therefore been greatly reduced.

This reduction is precisely the **Information Gain**.

---

## <font color='orange'> 2. Parent Entropy </font>

Suppose the parent node contains

$$
N
$$

observations.

Its entropy is

$$
\boxed{
H(\text{Parent})
=
-
\sum_{k=1}^{K}
p_k
\log_2(p_k),
}
$$

where

$$
p_k
$$

is the proportion of observations belonging to class

$$
k.
$$

This represents the uncertainty before splitting.

---

## <font color='orange'> 3. Child Entropy </font>

After splitting,

the parent node is divided into two child nodes,

$$
R_1
$$

and

$$
R_2.
$$

Each child has its own entropy,

$$
H(R_1)
$$

and

$$
H(R_2).
$$

Since the child nodes may contain different numbers of observations,

their entropies must be weighted according to their sizes.

The weighted entropy after splitting is

$$
\boxed{
H_{\text{split}}
=
\frac{N_1}{N}
H(R_1)
+
\frac{N_2}{N}
H(R_2),
}
$$

where

* $N_1$ is the number of observations in the left child,
* $N_2$ is the number of observations in the right child,
* $N=N_1+N_2$.

This quantity represents the remaining uncertainty after the split.

---

## <font color='orange'> 4. Definition of Information Gain </font>

Information Gain is defined as the difference between the parent entropy and the weighted entropy of the child nodes.

Thus,

$$
\boxed{
IG
=
H(\text{Parent})
-
H_{\text{split}}.
}
$$

Substituting the weighted entropy,

$$
\boxed{
IG
=
H(\text{Parent})
-
\left[
\frac{N_1}{N}
H(R_1)
+
\frac{N_2}{N}
H(R_2)
\right].
}
$$

This is the standard Information Gain equation used in Decision Tree learning.

---

## <font color='orange'> 5. Interpretation </font>

Information Gain measures how much uncertainty has been removed by performing a split.

Three situations are possible.

### Large Information Gain

The child nodes are much purer than the parent node.

The split is highly informative.

---

### Small Information Gain

The child nodes remain almost as impure as the parent.

The split provides little useful information.

---

### Zero Information Gain

The split does not reduce entropy at all.

Such a split should never be selected.

---

## <font color='orange'> 6. Example Calculation </font>

Suppose a parent node contains

```
9 Fraud

5 Non-Fraud
```

The parent entropy is

$$
H(\text{Parent})
=
0.940.
$$

After splitting,

the child nodes become

```
Left

6 Fraud

1 Non-Fraud
```

and

```
Right

3 Fraud

4 Non-Fraud.
```

The corresponding entropies are

$$
H(R_1)
=
0.592,
$$

$$
H(R_2)
=
0.985.
$$

Since each child contains

7

observations,

the weighted entropy becomes

$$
H_{\text{split}}
=
\frac{7}{14}(0.592)
+
\frac{7}{14}(0.985)
=
0.789.
$$

Therefore,

$$
IG
=
0.940
-
0.789
=
0.151.
$$

The split has reduced the uncertainty by

0.151 bits.

---

## <font color='orange'> 7. Why the Weighted Average? </font>

Consider two candidate splits.

### Split A

```
95 observations

5 observations
```

### Split B

```
50 observations

50 observations
```

The larger child node should contribute more to the overall impurity because it contains more observations.

Therefore,

the child entropies are weighted by

$$
\frac{N_i}{N}.
$$

Without weighting,

small nodes could disproportionately influence the splitting criterion.

---

## <font color='orange'> 8. Choosing the Best Feature </font>

Suppose the algorithm evaluates three candidate features.

| Feature | Information Gain |
| :--- | ---: |
| Age | 0.12 |
| Income | 0.31 |
| Credit Score | 0.21 |

Since

$$
0.31
$$

is the largest value,

the algorithm selects

```
Income
```

for the next split.

This process is repeated recursively throughout the tree.

---

## <font color='orange'> 9. Recursive Tree Construction </font>

At every internal node,

the algorithm repeats the following procedure.

1. Compute the entropy of the current node.
2. Evaluate every candidate feature.
3. Compute the Information Gain for every possible split.
4. Select the split with the largest Information Gain.
5. Repeat for the resulting child nodes.

The tree therefore grows greedily,

making the locally optimal decision at each stage.

---

## <font color='orange'> 10. Advantages of Information Gain </font>

Information Gain possesses several desirable properties.

* Easy to compute.
* Strong theoretical foundation in Information Theory.
* Produces highly interpretable trees.
* Encourages increasingly pure child nodes.
* Widely used in the ID3 and C4.5 algorithms.

---

## <font color='orange'> 11. Limitations of Information Gain </font>

Information Gain exhibits one important weakness.

It tends to favour features with **many unique values**.

For example,

suppose one feature is

```
Customer ID
```

where every observation has a different value.

Splitting on this feature creates many tiny, perfectly pure nodes,

resulting in an artificially large Information Gain.

However,

such a split has poor predictive value because it merely memorizes the training data.

To address this bias,

the C4.5 algorithm introduces the **Gain Ratio**, which normalizes the Information Gain by the intrinsic information of the split.

---

## <font color='orange'> 12. Information Gain in Scikit-Learn </font>

Decision Trees using Information Gain are implemented by selecting

```python
criterion="entropy"
```

or

```python
criterion="log_loss"
```

For example,

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)
```

During training,

Scikit-Learn computes the Information Gain for every candidate split and selects the one that maximizes the reduction in entropy.

## <font color='orange'> 13. Information Gain vs Entropy </font>

| Quantity | Purpose |
| :--- | :--- |
| Entropy | Measures uncertainty within a node |
| Information Gain | Measures the reduction in uncertainty after splitting |

Entropy describes the impurity of a single node,

whereas Information Gain evaluates the usefulness of a proposed split.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Parent Entropy | Uncertainty before splitting |
| Child Entropy | Uncertainty after splitting |
| Information Gain | Reduction in entropy |
| Formula | $IG = H(\text{Parent}) - H_{\text{split}}$ |
| Objective | Maximize Information Gain |
| Used By | ID3 and C4.5 Decision Trees |
| Limitation | Biased towards features with many distinct values |

> **Key Insight:** Information Gain quantifies the improvement achieved by splitting a node. It is defined as the reduction in entropy between the parent node and its child nodes, thereby measuring how much uncertainty is removed by a candidate split. Decision Tree algorithms such as **ID3** and **C4.5** evaluate every possible feature and threshold combination, selecting the split that maximizes Information Gain. Although this criterion effectively produces increasingly pure nodes, it can exhibit a bias towards attributes with many unique values, motivating the introduction of alternative measures such as the **Gain Ratio** and **Gini Impurity**.

# <font color='blue'> Gini Impurity </font>

While **Entropy** measures impurity using concepts from Information Theory, the **Gini Impurity** measures impurity using probabilities. It quantifies the likelihood that a randomly selected observation from a node would be **incorrectly classified** if it were assigned a class label according to the class distribution within that node.

The Gini Impurity is the default splitting criterion used by the **Classification and Regression Tree (CART)** algorithm and by **Scikit-Learn's `DecisionTreeClassifier`**.

Like entropy, Gini Impurity assigns a value of **zero** to perfectly pure nodes and larger values to increasingly mixed nodes.

The Decision Tree algorithm therefore seeks splits that minimize the Gini Impurity of the child nodes.

---

## <font color='orange'> 1. Intuition Behind Gini Impurity </font>

Consider a node containing observations from two classes.

```
Fraud

Fraud

Fraud

Fraud

Fraud
```

Every observation belongs to the same class.

If we randomly assign class labels according to the class distribution,

every observation will be classified correctly.

The impurity is therefore

$$
0.
$$

---

Now consider

```
Fraud

Fraud

Non-Fraud

Fraud

Non-Fraud
```

Since both classes are present,

random assignment occasionally produces incorrect classifications.

The node is therefore impure.

---

The more evenly distributed the classes become,

the larger the probability of making an incorrect classification,

and therefore the larger the Gini Impurity.

---

## <font color='orange'> 2. Class Probabilities </font>

Suppose a node contains

$$
K
$$

classes.

The probability that an observation belongs to class

$$
k
$$

is

$$
\boxed{
p_k
=
\frac{N_k}{N},
}
$$

where

* $N_k$ is the number of observations belonging to class $k$,
* $N$ is the total number of observations.

These probabilities satisfy

$$
\sum_{k=1}^{K}p_k=1.
$$

---

## <font color='orange'> 3. Derivation of the Gini Impurity </font>

Suppose we randomly assign a class label to an observation according to the class probabilities within a node.

The probability of assigning class

$$
k
$$

is

$$
p_k.
$$

The probability that the assigned label is **incorrect** is

$$
1-p_k.
$$

Therefore,

the probability of incorrectly classifying an observation as class

$$
k
$$

is

$$
p_k(1-p_k).
$$

Summing over all classes gives

$$
\boxed{
G
=
\sum_{k=1}^{K}
p_k(1-p_k).
}
$$

Expanding,

$$
G
=
\sum_{k=1}^{K}
p_k
-
\sum_{k=1}^{K}
p_k^2.
$$

Since

$$
\sum_{k=1}^{K}p_k=1,
$$

the equation simplifies to

$$
\boxed{
G
=
1-
\sum_{k=1}^{K}
p_k^2.
}
$$

This is the standard Gini Impurity equation.

---

## <font color='orange'> 4. Binary Classification </font>

For two classes,

$$
p
$$

and

$$
1-p,
$$

the Gini Impurity becomes

$$
\boxed{
G
=
1-
p^2-
(1-p)^2.
}
$$

Expanding,

$$
\boxed{
G
=
2p(1-p).
}
$$

This is the expression most commonly encountered in binary classification.

---

## <font color='orange'> 5. Example 1 — Pure Node </font>

Suppose a node contains

```
20 Fraud

0 Non-Fraud
```

Then

$$
p=1.
$$

Therefore,

$$
G
=
1-
1^2-
0^2
=
0.
$$

A perfectly pure node has zero Gini Impurity.

---

## <font color='orange'> 6. Example 2 — Maximum Impurity </font>

Suppose

```
10 Fraud

10 Non-Fraud
```

Then

$$
p=0.5.
$$

The Gini Impurity becomes

$$
G
=
1-
0.5^2-
0.5^2.
$$

Thus,

$$
G
=
0.5.
$$

Unlike entropy,

whose maximum is

$$
1,
$$

the maximum Gini Impurity for binary classification is

$$
0.5.
$$

---

## <font color='orange'> 7. Example 3 — Mostly Pure Node </font>

Suppose

```
18 Fraud

2 Non-Fraud
```

Then

$$
p=0.9.
$$

Therefore,

$$
G
=
1-
0.9^2-
0.1^2
=
0.18.
$$

The impurity is relatively small,

reflecting that the node is nearly pure.

---

## <font color='orange'> 8. Properties of Gini Impurity </font>

Gini Impurity possesses several desirable mathematical properties.

### Pure Node

If one class has probability

$$
1,
$$

then

$$
G=0.
$$

---

### Maximum Impurity

The maximum impurity occurs when all classes are equally likely.

For binary classification,

$$
p=0.5,
$$

giving

$$
G=0.5.
$$

---

### Symmetry

Interchanging class labels does not change the Gini Impurity.

For example,

```
70% Fraud

30% Non-Fraud
```

produces exactly the same impurity as

```
30% Fraud

70% Non-Fraud.
```

---

## <font color='orange'> 9. Choosing the Best Split </font>

Suppose a parent node has impurity

$$
G_{parent}.
$$

After splitting,

the weighted impurity becomes

$$
\boxed{
G_{split}
=
\frac{N_1}{N}G(R_1)
+
\frac{N_2}{N}G(R_2).
}
$$

The reduction in impurity is

$$
\boxed{
\Delta G
=
G_{parent}
-
G_{split}.
}
$$

The CART algorithm selects the split that maximizes

$$
\Delta G.
$$

---

## <font color='orange'> 10. Example of Selecting a Split </font>

Suppose three candidate features produce

| Feature | Weighted Gini |
| :--- | ---: |
| Age | 0.31 |
| Income | 0.18 |
| Credit Score | 0.26 |

The smallest weighted impurity is

$$
0.18.
$$

Therefore,

the Decision Tree selects

```
Income
```

for the next split.

---

## <font color='orange'> 11. Gini Impurity vs Entropy </font>

Both measures quantify node impurity,

but they differ mathematically.

| Property | Entropy | Gini Impurity |
| :--- | :---: | :---: |
| Formula | $-\sum p_k\log_2(p_k)$ | $1-\sum p_k^2$ |
| Origin | Information Theory | Probability Theory |
| Maximum (Binary) | 1 | 0.5 |
| Computation | Logarithms | Squaring only |
| Speed | Slightly slower | Slightly faster |

Since Gini Impurity avoids logarithmic calculations,

it is generally computationally more efficient.

---

## <font color='orange'> 12. Which is Better? </font>

In practice,

both impurity measures often produce very similar Decision Trees.

Entropy may produce slightly more balanced trees,

whereas Gini Impurity often isolates the dominant class more rapidly.

For most practical applications,

the difference in predictive performance is small.

Consequently,

Scikit-Learn adopts Gini Impurity as the default criterion because of its computational efficiency.

---

| Concept | Description |
| :--- | :--- |
| Gini Impurity | Probability of incorrectly classifying a randomly chosen observation |

> **Key Insight:** Gini Impurity measures the probability of making an incorrect classification when observations are labelled according to the class distribution within a node. A perfectly pure node has zero Gini Impurity, while increasingly mixed class distributions produce larger impurity values. During tree construction, the CART algorithm evaluates every candidate split by computing the weighted Gini Impurity of the resulting child nodes and selects the split that produces the greatest reduction in impurity. Because it requires only simple arithmetic operations rather than logarithms, Gini Impurity is computationally efficient and has become the default splitting criterion in many modern machine learning libraries, including Scikit-Learn.

# <font color='blue'> Misclassification Error </font>

The **Misclassification Error** is one of the simplest measures of node impurity used in Decision Trees. Unlike **Entropy** and **Gini Impurity**, which quantify uncertainty using logarithms or probabilities, the Misclassification Error simply measures the proportion of observations that would be incorrectly classified if every observation within a node were assigned to the **majority class**.

Although this measure is intuitive and easy to interpret, it is relatively insensitive to small changes in class probabilities. Consequently, it is rarely used as the splitting criterion during tree construction, but it plays an important role in **tree pruning** and **model evaluation**.

---

## <font color='orange'> 1. Intuition Behind Misclassification Error </font>

Suppose a node contains the following observations.

```
Fraud

Fraud

Fraud

Fraud

Fraud
```

Every observation belongs to the same class.

If we classify every observation as

```
Fraud
```

then every prediction is correct.

The classification error is therefore

$$
0.
$$

---

Now consider another node.

```
Fraud

Fraud

Fraud

Non-Fraud

Non-Fraud
```

The majority class is

```
Fraud.
```

If we classify every observation as

```
Fraud,
```

then two observations are classified incorrectly.

The misclassification error is therefore

$$
\frac{2}{5}=0.4.
$$

---

## <font color='orange'> 2. Majority Class Prediction </font>

Suppose a node contains

$$
K
$$

classes.

The probability of class

$$
k
$$

is

$$
p_k.
$$

The Decision Tree predicts the class with the largest probability,

namely

$$
\boxed{
\max_k(p_k).
}
$$

This is called the **majority class**.

---

## <font color='orange'> 3. Derivation of Misclassification Error </font>

If the majority class has probability

$$
\max_k(p_k),
$$

then the remaining observations are classified incorrectly.

The Misclassification Error therefore becomes

$$
\boxed{
E
=
1-
\max_k(p_k).
}
$$

This simple equation measures the proportion of observations that would receive an incorrect prediction.

---

## <font color='orange'> 4. Binary Classification </font>

For binary classification,

let

$$
p
$$

represent the probability of one class.

The majority class has probability

$$
\max(p,1-p).
$$

Therefore,

the Misclassification Error becomes

$$
\boxed{
E
=
1-
\max(p,1-p).
}
$$

---

## <font color='orange'> 5. Example 1 — Pure Node </font>

Suppose a node contains

```
20 Fraud

0 Non-Fraud
```

The majority class probability is

$$
1.
$$

Therefore,

$$
E
=
1-1
=
0.
$$

A perfectly pure node has zero Misclassification Error.

---

## <font color='orange'> 6. Example 2 — Equal Class Distribution </font>

Suppose a node contains

```
10 Fraud

10 Non-Fraud
```

The largest class probability is

$$
0.5.
$$

Thus,

$$
E
=
1-0.5
=
0.5.
$$

This is the maximum possible Misclassification Error for a binary classification problem.

---

## <font color='orange'> 7. Example 3 — Mostly Pure Node </font>

Suppose

```
18 Fraud

2 Non-Fraud
```

The majority class probability is

$$
0.9.
$$

Therefore,

$$
E
=
1-0.9
=
0.1.
$$

The node is nearly pure,

and the impurity is correspondingly small.

---

## <font color='orange'> 8. Properties of Misclassification Error </font>

Misclassification Error satisfies several useful properties.

### Pure Node

If every observation belongs to one class,

$$
E=0.
$$

---

### Maximum Impurity

For binary classification,

the largest possible impurity occurs when

$$
p=0.5,
$$

giving

$$
E=0.5.
$$

---

### Symmetry

Interchanging the class labels leaves the impurity unchanged.

Thus,

```
70% Fraud

30% Non-Fraud
```

and

```
30% Fraud

70% Non-Fraud
```

produce exactly the same Misclassification Error.

---

## <font color='orange'> 9. Misclassification Error After Splitting </font>

Suppose a parent node is divided into two child nodes,

$$
R_1
$$

and

$$
R_2.
$$

The weighted Misclassification Error becomes

$$
\boxed{
E_{\text{split}}
=
\frac{N_1}{N}E(R_1)
+
\frac{N_2}{N}E(R_2),
}
$$

where

$$
N_1+N_2=N.
$$

The preferred split is the one producing the smallest weighted error.

---

## <font color='orange'> 10. Comparison with Entropy and Gini </font>

Although all three impurity measures identify pure nodes correctly,

they respond differently to changes in class probabilities.

Consider a binary classification problem.

| Majority Class Probability | Misclassification Error | Gini Impurity | Entropy |
| :---: | :---: | :---: | :---: |
| 1.0 | 0.00 | 0.00 | 0.00 |
| 0.9 | 0.10 | 0.18 | 0.47 |
| 0.8 | 0.20 | 0.32 | 0.72 |
| 0.7 | 0.30 | 0.42 | 0.88 |
| 0.6 | 0.40 | 0.48 | 0.97 |
| 0.5 | 0.50 | 0.50 | 1.00 |

Notice that

* Entropy changes most rapidly,
* Gini lies between the two,
* Misclassification Error changes least rapidly.

---

## <font color='orange'> 11. Why Isn't Misclassification Error Used for Splitting? </font>

Suppose two candidate splits produce

```
Split A

90% Fraud

10% Non-Fraud
```

and

```
Split B

80% Fraud

20% Non-Fraud.
```

Their Misclassification Errors are

```
0.10

0.20
```

respectively.

Now suppose another split improves the purity slightly,

changing

```
90%

↓

92%.
```

The Misclassification Error changes only from

```
0.10

↓

0.08.
```

This relatively small change makes Misclassification Error **less sensitive** to improvements in node purity.

By contrast,

Entropy and Gini Impurity respond much more strongly,

allowing the algorithm to distinguish between candidate splits more effectively.

For this reason,

modern Decision Tree algorithms almost always use Entropy or Gini Impurity when growing the tree.

---

## <font color='orange'> 12. Role in Tree Pruning </font>

Although rarely used during tree construction,

Misclassification Error is useful during **post-pruning**.

When simplifying an overgrown Decision Tree,

the algorithm evaluates whether removing a subtree significantly increases the classification error.

If the increase is negligible,

the subtree can be safely removed,

resulting in a simpler model with improved generalization.

---

## <font color='orange'> 13. Practical Usage </font>

Modern machine learning libraries generally provide

* **Gini Impurity**
* **Entropy**

as splitting criteria.

Misclassification Error is primarily encountered in

* theoretical discussions,
* pruning algorithms,
* model evaluation.

It is uncommon to use it directly as the impurity criterion during tree growth.

---

## <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Misclassification Error | Fraction of observations incorrectly classified using the majority class |
| Formula | $E = 1 - \max_k(p_k)$ |
| Pure Node | Error = 0 |
| Maximum (Binary) | Error = 0.5 |
| Sensitivity | Lower than Entropy and Gini |
| Primary Use | Tree pruning and evaluation |
| Rarely Used For | Tree construction |

> **Key Insight:** Misclassification Error measures the proportion of observations that would be incorrectly classified if every observation within a node were assigned to the majority class. Although it is simple to interpret, it is relatively insensitive to small improvements in node purity compared with Entropy and Gini Impurity. Consequently, modern Decision Tree algorithms rarely use Misclassification Error when selecting splits. Instead, it is primarily employed during **tree pruning** and **model evaluation**, where its direct relationship to classification accuracy provides an intuitive measure of predictive performance.

# <font color='blue'> Choosing the Best Split </font>

The construction of a Decision Tree consists of repeatedly partitioning the training data into increasingly homogeneous subsets. At every internal node, the algorithm must determine **which feature** should be used to split the data and **where** that feature should be divided.

This process is known as **split selection**.

The objective is to identify the feature and threshold that produce the **largest reduction in impurity**. Every possible split is evaluated, and the split yielding the greatest improvement is selected.

This procedure is repeated recursively until the stopping criteria are satisfied.

---

## <font color='orange'> 1. The Objective of Split Selection </font>

Suppose a node contains

$$
N
$$

training observations.

The algorithm must answer two questions:

1. Which feature should be used?

2. What threshold should divide that feature?

For example,

```
Age < 40 ?
```

or

```
Income > £50,000 ?
```

or

```
Credit Score < 720 ?
```

Among all possible candidate splits,

the algorithm chooses the one that produces the largest increase in node purity.

---

## <font color='orange'> 2. Candidate Features </font>

Suppose the training dataset contains

$$
p
$$

input variables,

$$
x_1,x_2,\ldots,x_p.
$$

Every feature is considered as a potential splitting variable.

For example,

```
Age

Income

Credit Score

Balance

Loan Amount
```

Each feature is evaluated independently.

The feature producing the best split is selected.

---

## <font color='orange'> 3. Candidate Thresholds </font>

For every numerical feature,

multiple threshold values are examined.

Suppose the feature

```
Age
```

contains

```
22

31

36

42

50

65
```

Possible split locations occur halfway between adjacent observations.

Therefore,

the candidate thresholds become

```
26.5

33.5

39

46

57.5
```

Each threshold partitions the observations into

```
Left Child

Age < threshold
```

and

```
Right Child

Age ≥ threshold.
```

---

## <font color='orange'> 4. Evaluating Every Candidate Split </font>

Suppose

```
Age
```

has five candidate thresholds,

```
Income
```

has six,

and

```
Credit Score
```

has four.

The algorithm evaluates every possible combination.

```
Age

↓

26.5

33.5

39

46

57.5

↓

Compute impurity reduction
```

The same procedure is repeated for every feature.

The split producing the greatest reduction in impurity is selected.

---

## <font color='orange'> 5. Mathematical Formulation </font>

Suppose a node occupies region

$$
R.
$$

For feature

$$
x_j
$$

and threshold

$$
s,
$$

the split partitions the observations into

$$
R_1(j,s)
=
\{
x_j<s
\}
$$

and

$$
R_2(j,s)
=
\{
x_j\ge s
\}.
$$

The weighted impurity after splitting becomes

$$
\boxed{
I_{\text{split}}
=
\frac{N_1}{N}
I(R_1)
+
\frac{N_2}{N}
I(R_2),
}
$$

where

* $I(\cdot)$ denotes the impurity measure,
* $N_1$ and $N_2$ are the numbers of observations in each child node,
* $N=N_1+N_2$.

---

## <font color='orange'> 6. Optimization Objective </font>

The Decision Tree searches over

* every feature,

and

* every threshold.

The optimal split is therefore

$$
\boxed{
(j^*,s^*)
=
\arg\min_{j,s}
I_{\text{split}}(j,s).
}
$$

Equivalently,

the algorithm may maximize the impurity reduction,

$$
\boxed{
\Delta I
=
I(R)
-
I_{\text{split}}.
}
$$

Both formulations produce exactly the same optimal split.

---

## <font color='orange'> 7. Example </font>

Suppose the current node contains

```
100 observations
```

Three candidate features are evaluated.

| Feature | Weighted Gini |
| :--- | ---: |
| Age | 0.31 |
| Income | 0.18 |
| Credit Score | 0.27 |

Since

```
Income
```

produces the smallest impurity,

the algorithm chooses

```
Income
```

for the next split.

---

## <font color='orange'> 8. Why Use Weighted Impurity? </font>

Suppose one split produces

```
95 observations

5 observations
```

while another produces

```
50 observations

50 observations.
```

The larger child node contains far more data,

so its impurity should contribute proportionally more to the overall split quality.

Therefore,

the weighted impurity is computed as

$$
\frac{N_i}{N}
I(R_i).
$$

Without weighting,

very small child nodes could dominate the optimization.

---

## <font color='orange'> 9. Greedy Optimization </font>

Decision Trees use a **greedy algorithm**.

At each node,

the algorithm selects the split that is **locally optimal**.

It does **not** attempt to examine every possible future tree.

For example,

```
Current Node

↓

Choose Best Split

↓

Repeat
```

The algorithm never revisits previous decisions.

Consequently,

the resulting tree is not guaranteed to be the globally optimal tree,

but it can be constructed efficiently.

---

## <font color='orange'> 10. Computational Complexity </font>

Suppose

* there are

$$
p
$$

features,

* and

$$
n
$$

observations.

For each feature,

multiple thresholds must be evaluated.

A naive implementation would require examining many candidate splits,

making the computational cost substantial for large datasets.

Modern Decision Tree implementations therefore employ efficient sorting and search algorithms to reduce the computational burden.

---

## <font color='orange'> 11. Numerical vs Categorical Features </font>

### Numerical Features

A threshold is selected,

for example,

```
Income < £60,000
```

---

### Categorical Features

The algorithm partitions categories into groups.

For example,

```
Region

↓

North

South

↓

East

West
```

Some Decision Tree implementations handle categorical variables directly,

while others require them to be encoded numerically before training.

---

## <font color='orange'> 12. Recursive Construction </font>

Once the best split has been chosen,

the parent node is replaced by two child nodes.

Each child then becomes the root of a smaller Decision Tree.

```
Root

↓

Best Split

↓

Child

Child

↓

Repeat
```

The identical optimization procedure is applied recursively until the stopping criteria are reached.

This repeated application gives rise to the complete Decision Tree.

---

## <font color='orange'> 13. Summary of the Split Selection Algorithm </font>

The complete algorithm proceeds as follows.

```
Start with the current node

↓

Evaluate every feature

↓

Evaluate every candidate threshold

↓

Compute weighted impurity

↓

Select the split with minimum impurity

↓

Create two child nodes

↓

Repeat recursively
```

This process continues until no further beneficial splits can be made.

---

## <font color='orange'> 14. Split Selection in Scikit-Learn </font>

Scikit-Learn automatically performs this optimization during training.

For example,

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="gini",
    random_state=42
)

model.fit(X_train, y_train)
```

Internally,

the algorithm evaluates thousands of candidate splits and selects the one that minimizes the weighted impurity at each node.

The entire optimization is performed automatically.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Candidate Feature | Predictor variable considered for splitting |
| Candidate Threshold | Possible split location within a feature |
| Objective | Minimize weighted impurity |
| Optimization | Search over every feature and threshold |
| Strategy | Greedy optimization |
| Child Nodes | Created after the optimal split |
| Tree Construction | Recursive repetition of the same process |

> **Key Insight:** Choosing the best split is the core optimization step in Decision Tree learning. At every node, the algorithm evaluates every feature and every possible threshold, computes the weighted impurity of the resulting child nodes, and selects the split that maximizes the reduction in impurity. Because this optimization is performed recursively throughout the tree using a greedy strategy, Decision Trees can efficiently discover complex nonlinear decision boundaries while remaining highly interpretable.

# <font color='blue'> Building a Decision Tree: The Recursive Algorithm </font>

Having introduced **Recursive Binary Splitting**, **Node Impurity**, and **Split Selection**, we can now describe the complete procedure used to construct a Decision Tree.

The Decision Tree learning algorithm follows a **top-down recursive approach**. Beginning with the entire training dataset at the root node, the algorithm repeatedly selects the best split, partitions the data into two child nodes, and then applies exactly the same procedure to each child.

This recursive process continues until a stopping criterion is satisfied, producing the final Decision Tree.

---

## <font color='orange'> 1. The Overall Strategy </font>

The algorithm begins with the complete training dataset,

$$
D
=
\{(\mathbf{x}_1,y_1),(\mathbf{x}_2,y_2),\ldots,(\mathbf{x}_N,y_N)\}.
$$

Initially,

all observations belong to a single node,

called the **root node**.

The objective is to repeatedly divide this dataset into increasingly homogeneous subsets.

The resulting structure resembles an inverted tree.

```
Entire Dataset

↓

Root Node

↓

Split

↓

Child Nodes

↓

Further Splits

↓

Leaf Nodes
```

---

## <font color='orange'> 2. Recursive Nature of the Algorithm </font>

A Decision Tree is built recursively.

This means that exactly the same procedure is applied repeatedly to every newly created node.

Suppose the root node is divided into

```
Root

↓

Left Child

Right Child
```

Each child is then treated as an entirely new Decision Tree.

The algorithm searches for the best split within each child independently.

This recursive process continues until no further splitting is required.

---

## <font color='orange'> 3. Step 1 — Begin with the Root Node </font>

Initially,

every observation belongs to the root node.

```
1000 observations

↓

Root
```

The impurity of the root node is computed using

* Entropy,

or

* Gini Impurity,

depending on the chosen criterion.

---

## <font color='orange'> 4. Step 2 — Search for the Best Split </font>

The algorithm evaluates

* every feature,

and

* every possible threshold.

For each candidate split,

the weighted impurity is calculated.

The split producing the largest reduction in impurity is selected.

Mathematically,

$$
(j^*,s^*)
=
\arg\min_{j,s}
I_{\text{split}}(j,s).
$$

This becomes the first decision rule in the tree.

---

## <font color='orange'> 5. Step 3 — Partition the Dataset </font>

After selecting the optimal split,

the observations are divided into two subsets.

Suppose the split is

```
Age < 40 ?
```

The data become

```
Age < 40

↓

Left Child
```

and

```
Age ≥ 40

↓

Right Child
```

Each child node now contains a smaller subset of the original observations.

---

## <font color='orange'> 6. Step 4 — Repeat the Process </font>

Each child node now becomes the root of a smaller subtree.

The algorithm again

1. computes the node impurity,
2. searches every feature,
3. evaluates every threshold,
4. selects the best split,
5. creates new child nodes.

This recursive procedure continues until a stopping condition is reached.

---

## <font color='orange'> 7. Worked Example </font>

Suppose a dataset contains

```
100 observations.
```

### Initial Tree

```
100

↓

Root
```

---

### First Split

```
100

↓

Income > £50k

↓

60            40
```

---

### Second Split

```
60

↓

Age < 35

↓

35        25
```

and

```
40

↓

Credit Score > 700

↓

28        12
```

---

### Third Split

Each of the four child nodes is examined independently.

The recursion continues until every terminal node satisfies the stopping criteria.

---

## <font color='orange'> 8. Recursive Algorithm (Pseudocode) </font>

The Decision Tree learning algorithm may be summarized as follows.

```text
BuildTree(Node)

    Compute node impurity

    If stopping criterion satisfied

        Return Leaf

    Else

        Search every feature

        Search every threshold

        Find split with minimum impurity

        Divide data into two child nodes

        BuildTree(Left Child)

        BuildTree(Right Child)
```

Notice that

```
BuildTree()
```

calls itself repeatedly.

This self-referencing procedure is the defining characteristic of recursion.

---

## <font color='orange'> 9. Visual Representation of Recursion </font>

```
Root

↓

Split

↓

Left            Right

↓

Split           Split

↓

Leaves         Leaves
```

Every internal node follows exactly the same sequence of operations.

Only the subset of observations changes.

---

## <font color='orange'> 10. Greedy Nature of Tree Construction </font>

Decision Trees use a **greedy algorithm**.

At every node,

the algorithm selects the split that appears best **at that moment**.

It does **not** search all possible tree structures.

Consequently,

the resulting tree is not guaranteed to be globally optimal.

However,

this greedy strategy dramatically reduces computational complexity while producing excellent practical performance.

---

## <font color='orange'> 11. Why Doesn't the Algorithm Find the Global Optimum? </font>

Suppose two possible first splits exist.

```
Split A

↓

Excellent now

↓

Average later
```

```
Split B

↓

Average now

↓

Excellent later
```

The greedy algorithm always chooses

```
Split A
```

because it immediately produces the largest impurity reduction.

It does not examine whether

```
Split B
```

might eventually produce a better overall tree.

Searching every possible tree would require an astronomically large number of combinations and is computationally infeasible.

---

## <font color='orange'> 12. Computational Complexity </font>

Suppose

* the dataset contains

$$
n
$$

observations,

* and

$$
p
$$

features.

At every node,

the algorithm evaluates many candidate thresholds for every feature.

Modern Decision Tree implementations therefore employ efficient algorithms based on

* sorted feature values,
* recursive partitioning,
* optimized search strategies.

These improvements allow large Decision Trees to be trained efficiently.

---

## <font color='orange'> 13. Advantages of Recursive Tree Construction </font>

The recursive approach offers several important advantages.

* Simple algorithmic structure.
* Naturally handles nonlinear decision boundaries.
* Efficient implementation.
* Easily extended to ensemble methods.
* Produces interpretable decision rules.

This recursive strategy forms the basis of

* Random Forests,
* Gradient Boosting,
* XGBoost,
* LightGBM,
* CatBoost.

---

Tree construction is performed automatically during training.

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)

model.fit(X_train, y_train)
```

```python
fit()
```

The user does not need to implement the recursive algorithm manually.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Starting Point | Entire training dataset |
| Root Node | Contains all observations |
| Split Selection | Choose feature and threshold minimizing impurity |
| Child Nodes | Created after each split |
| Recursion | Apply the same algorithm to each child node |
| Greedy Strategy | Select the locally optimal split |
| Termination | Stop when stopping criteria are satisfied |
| Final Output | Fully constructed Decision Tree |

> **Key Insight:** Decision Tree learning is fundamentally a recursive divide-and-conquer algorithm. Beginning with the entire training dataset, the algorithm repeatedly searches for the split that produces the greatest reduction in impurity, partitions the data into increasingly homogeneous subsets, and applies the same procedure recursively to every child node. This process continues until predefined stopping criteria are reached, producing a hierarchical structure of decision rules that can model highly complex nonlinear relationships while remaining intuitive and interpretable.

# <font color='blue'> Stopping Criteria in Decision Trees </font>

If a Decision Tree is allowed to grow without restriction, it will continue splitting the training data until every leaf node contains only a single observation or all observations within each leaf belong to the same class. Although this minimizes the training error, it usually results in **overfitting**, where the model memorizes the training data instead of learning the underlying patterns.

To prevent excessive growth, Decision Tree algorithms employ a collection of **stopping criteria** that determine when further splitting is no longer beneficial.

These stopping criteria control the **complexity** of the tree and therefore play a crucial role in balancing **bias** and **variance**.

---

## <font color='orange'> 1. Why Can't Trees Grow Forever? </font>

Suppose we begin with

```
1000 observations
```

After repeated recursive splitting, the tree might eventually reach

```
Leaf 1 : 1 observation

Leaf 2 : 1 observation

Leaf 3 : 1 observation

...

Leaf 1000 : 1 observation
```

Every training observation is classified perfectly.

The training accuracy becomes

```
100%
```

However,

the model has simply memorized the training dataset.

When presented with new observations,

its predictive performance often deteriorates significantly.

This phenomenon is known as **overfitting**.

---

## <font color='orange'> 2. Purpose of Stopping Criteria </font>

Stopping criteria determine whether a node should

* continue splitting,

or

* become a leaf node.

Their primary objectives are

* reduce overfitting,
* improve generalization,
* control tree complexity,
* reduce computational cost,
* improve interpretability.

---

## <font color='orange'> 3. Pure Nodes </font>

The simplest stopping criterion occurs when a node is already perfectly pure.

For example,

```
Fraud

Fraud

Fraud

Fraud
```

Since every observation belongs to the same class,

there is no uncertainty.

The impurity is

$$
I=0.
$$

No further split can improve this node,

so recursion stops automatically.

---

## <font color='orange'> 4. Maximum Tree Depth (`max_depth`) </font>

The **depth** of a tree is the maximum number of splits from the root node to any leaf node.

Example:

```
Root

↓

Node

↓

Node

↓

Leaf
```

has depth

$$
3.
$$

By specifying

```python
max_depth = 5
```

the algorithm is forbidden from creating nodes deeper than five levels.

### Advantages

* Controls model complexity.
* Prevents extremely deep trees.
* Reduces overfitting.
* Faster prediction.

### Disadvantages

If the depth is too small,

the model may underfit.

---

## <font color='orange'> 5. Minimum Samples Required to Split (`min_samples_split`) </font>

A split should only occur if sufficient data are available.

Suppose

```python
min_samples_split = 20
```

Then

```
25 observations

↓

Split Allowed
```

but

```
15 observations

↓

No Split
```

The node immediately becomes a leaf.

This prevents unreliable decisions based upon very small datasets.

---

## <font color='orange'> 6. Minimum Samples per Leaf (`min_samples_leaf`) </font>

Sometimes a split produces extremely small child nodes.

For example,

```
Parent

100 observations

↓

99          1
```

Although technically valid,

the leaf containing

```
1 observation
```

is unlikely to generalize well.

Using

```python
min_samples_leaf = 10
```

ensures that every terminal node contains at least ten observations.

This generally produces smoother decision boundaries.

---

## <font color='orange'> 7. Maximum Number of Leaf Nodes (`max_leaf_nodes`) </font>

Rather than limiting depth,

we may directly limit the number of terminal nodes.

For example,

```python
max_leaf_nodes = 20
```

Once twenty leaves have been created,

tree construction stops automatically,

regardless of the remaining impurity.

This provides another mechanism for controlling model complexity.

---

## <font color='orange'> 8. Minimum Impurity Decrease (`min_impurity_decrease`) </font>

Not every split produces a meaningful improvement.

Suppose a candidate split reduces the Gini Impurity from

```
0.412

↓

0.410
```

The improvement is negligible.

Using

```python
min_impurity_decrease = 0.01
```

requires every split to reduce impurity by at least

```
0.01.
```

Otherwise,

the node becomes a leaf.

This prevents unnecessary tree growth.

---

## <font color='orange'> 9. Early Stopping </font>

Stopping criteria collectively implement a form of **early stopping**.

Rather than allowing the tree to grow completely,

construction terminates once further improvements become insignificant.

This reduces

* overfitting,
* computational cost,
* memory usage.

---

## <font color='orange'> 10. Trade-Off Between Bias and Variance </font>

Stopping criteria directly influence the **Bias–Variance Trade-off**.

### Weak Restrictions

```
Large Tree

↓

Low Bias

↓

High Variance
```

The model fits the training data extremely well,

but may overfit.

---

### Strong Restrictions

```
Small Tree

↓

High Bias

↓

Low Variance
```

The model becomes simpler,

but may fail to capture important relationships.

The objective is to choose stopping criteria that balance these competing effects.

---

## <font color='orange'> 11. Example in Scikit-Learn </font>

A Decision Tree with several stopping criteria may be created as follows.

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=6,
    min_samples_split=20,
    min_samples_leaf=10,
    max_leaf_nodes=30,
    min_impurity_decrease=0.001,
    random_state=42
)
```

During training,

the algorithm automatically checks every stopping condition before creating a new split.

---

## <font color='orange'> 12. Choosing Appropriate Values </font>

There is no universally optimal set of stopping parameters.

Instead,

they are usually determined using

* Cross-Validation,
* Grid Search,
* Random Search,
* Bayesian Optimization.

The optimal values depend upon

* dataset size,
* noise level,
* number of features,
* desired model complexity.

---

## <font color='orange'> 13. Summary of Common Stopping Parameters </font>

| Hyperparameter | Purpose |
| :--- | :--- |
| `max_depth` | Limits the maximum depth of the tree |
| `min_samples_split` | Minimum observations required before splitting |
| `min_samples_leaf` | Minimum observations allowed in every leaf |
| `max_leaf_nodes` | Maximum number of terminal nodes |
| `min_impurity_decrease` | Minimum impurity reduction required for a split |

Together,

these parameters determine the overall complexity of the Decision Tree.

---

## <font color='orange'> 14. Pre-Pruning vs Post-Pruning </font>

Stopping criteria are examples of **pre-pruning**, where restrictions are imposed **before** or **during** tree construction.

An alternative strategy is **post-pruning**, where the tree is first allowed to grow fully and is then simplified by removing branches that contribute little to predictive performance.

Both approaches seek to improve generalization while reducing unnecessary complexity.

Post-pruning will be discussed in detail in a later section.

---

## <font color='purple'> 15. Conceptual Summary </font>

| Stopping Criterion | Effect |
| :--- | :--- |
| Pure Node | Stop splitting when impurity becomes zero |
| `max_depth` | Limits tree height |
| `min_samples_split` | Prevents splitting very small nodes |
| `min_samples_leaf` | Prevents tiny terminal nodes |
| `max_leaf_nodes` | Limits total number of leaves |
| `min_impurity_decrease` | Ignores insignificant improvements |
| Overall Goal | Improve generalization and reduce overfitting |

> **Key Insight:** A Decision Tree that grows without restriction will almost always overfit the training data by memorizing individual observations. Stopping criteria provide a principled mechanism for controlling the complexity of the tree by limiting its growth according to depth, node size, impurity reduction, or the number of terminal nodes. These hyperparameters balance the trade-off between bias and variance, enabling the model to capture meaningful patterns while maintaining strong predictive performance on previously unseen data.

# <font color='blue'> Tree Depth and Model Complexity </font>

One of the most important factors determining the performance of a Decision Tree is its **depth**. Tree depth controls the complexity of the model, influencing its ability to learn patterns from the training data while also affecting its ability to generalize to previously unseen observations.

A shallow tree may be too simple to capture important relationships within the data, leading to **underfitting**. Conversely, an excessively deep tree may memorize the training dataset, resulting in **overfitting**.

Understanding how tree depth influences model complexity is therefore essential for selecting an appropriate Decision Tree architecture.

---

## <font color='orange'> 1. What is Tree Depth? </font>

The **depth** of a Decision Tree is defined as the maximum number of decision rules (or splits) encountered when travelling from the **root node** to the deepest **leaf node**.

For example,

```
Root

↓

Node

↓

Node

↓

Leaf
```

contains

three

splits,

so the tree depth is

$$
\boxed{
\text{Depth}=3.
}
$$

A larger depth allows the tree to represent increasingly complex decision boundaries.

---

## <font color='orange'> 2. Model Complexity </font>

Model complexity describes the ability of a machine learning model to capture increasingly complicated relationships within the data.

For Decision Trees,

complexity increases primarily through

* larger tree depth,
* more leaf nodes,
* more decision rules.

As complexity increases,

the tree partitions the feature space into progressively smaller regions.

---

## <font color='orange'> 3. Shallow Trees </font>

Consider a tree with

```python
max_depth = 1
```

```
Root

↓

Leaf      Leaf
```

Only one decision rule is available.

Such a model is extremely simple.

Although it is unlikely to overfit,

it often fails to capture nonlinear relationships within the data.

Consequently,

both the training error and the testing error remain relatively large.

This phenomenon is known as **underfitting**.

---

## <font color='orange'> 4. Deep Trees </font>

Now consider

```python
max_depth = 20
```

The tree contains many levels of branching.

```
Root

↓

Split

↓

Split

↓

Split

↓

...

↓

Leaf
```

The Decision Tree can now create extremely detailed partitions of the feature space.

Training accuracy generally improves.

However,

the tree also begins modelling random fluctuations and statistical noise.

This behaviour is characteristic of **overfitting**.

---

## <font color='orange'> 5. Relationship Between Depth and Error </font>

As tree depth increases,

the **training error** almost always decreases.

This occurs because every additional split enables the model to fit the training data more closely.

Mathematically,

if

$$
d
$$

denotes the tree depth,

then

$$
J_{\text{train}}(d)
$$

is generally a decreasing function of

$$
d.
$$

However,

the testing or cross-validation error behaves differently.

Initially,

increasing depth improves generalization,

but beyond a certain point,

the model begins fitting noise rather than signal.

Consequently,

the testing error eventually increases.

---

## <font color='orange'> 6. The Bias–Variance Trade-off </font>

Tree depth directly controls the balance between **bias** and **variance**.

### Small Depth

* High bias
* Low variance
* Underfitting

The model is too simple.

It cannot adequately represent the underlying relationship.

---

### Large Depth

* Low bias
* High variance
* Overfitting

The model becomes excessively sensitive to the training data.

Small changes in the dataset may produce substantially different trees.

---

### Intermediate Depth

An intermediate tree depth usually provides the best compromise,

producing both

* low bias,

and

* low variance.

This corresponds to the optimal level of model complexity.

---

## <font color='orange'> 7. Conceptual Error Curve </font>

The relationship between model complexity and prediction error can be summarized conceptually.

```
Prediction Error

^

|

|          CV Error

|            /\

|           /  \

|          /    \

|         /      \

|        /        \

|-------/----------\---------------->

      Training Error

      Model Complexity
```

As complexity increases,

* the training error continually decreases,
* the validation error initially decreases,
* eventually the validation error increases due to overfitting.

The minimum of the validation curve corresponds to the optimal tree depth.

---

## <font color='orange'> 8. Effect on the Decision Boundary </font>

### Shallow Tree

```
Simple

Large Regions

↓

High Bias
```

The feature space is partitioned into only a few large regions.

---

### Deep Tree

```
Many Small Regions

↓

Complex Boundary

↓

High Variance
```

The feature space becomes highly fragmented,

allowing the model to fit extremely complicated patterns.

---

## <font color='orange'> 9. Influence on Interpretability </font>

Decision Trees are popular because they are highly interpretable.

However,

interpretability decreases as tree depth increases.

A shallow tree

```
Depth = 3
```

may contain only a handful of decision rules,

making it easy to understand.

A deep tree

```
Depth = 30
```

may contain hundreds of nodes,

making interpretation considerably more difficult.

Thus,

greater predictive flexibility often comes at the expense of interpretability.

---

## <font color='orange'> 10. Selecting the Optimal Tree Depth </font>

The optimal depth is usually determined using **Cross-Validation**.

The general procedure is

1. Train several trees using different values of

```python
max_depth
```

2. Evaluate each model using cross-validation.

3. Select the depth producing the smallest validation error.

This procedure helps ensure good generalization performance.

---

## <font color='orange'> 11. Example in Scikit-Learn </font>

Tree depth can be controlled directly.

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)
```

Increasing

```python
max_depth
```

produces a more flexible model,

whereas reducing it creates a simpler tree.

---

## <font color='orange'> 12. Practical Guidelines </font>

The following recommendations are commonly used in practice.

* Begin with a relatively unrestricted tree.
* Evaluate performance using cross-validation.
* Reduce the depth if overfitting occurs.
* Increase the depth if the model underfits.
* Tune additional parameters such as

  * `min_samples_split`,
  * `min_samples_leaf`,
  * `max_leaf_nodes`.

These parameters work together to control model complexity.

---

## <font color='orange'> 13. Relationship to Ensemble Methods </font>

Tree depth remains important even when using ensemble methods.

### Random Forests

Individual trees are often allowed to grow relatively deep,

while averaging reduces variance.

---

### Gradient Boosting

Individual trees are intentionally kept shallow,

typically

```python
max_depth = 3
```

or

```python
max_depth = 4,
```

because many weak learners are combined sequentially.

Thus,

tree depth continues to play an important role even in advanced ensemble algorithms.

## <font color='purple'> 14. Conceptual Summary </font>

| Tree Depth | Model Complexity | Bias | Variance | Typical Behaviour |
| :--- | :--- | :---: | :---: | :--- |
| Very Small | Low | High | Low | Underfitting |
| Moderate | Medium | Balanced | Balanced | Good Generalization |
| Very Large | High | Low | High | Overfitting |

> **Key Insight:** Tree depth is one of the primary determinants of Decision Tree complexity. Increasing the depth allows the model to learn progressively more intricate decision boundaries, reducing bias but increasing variance. Conversely, restricting the depth produces simpler, more interpretable models that may underfit the data. The optimal tree depth is therefore the one that minimizes the validation error by balancing the competing effects of bias and variance, enabling the Decision Tree to generalize effectively to previously unseen observations.
