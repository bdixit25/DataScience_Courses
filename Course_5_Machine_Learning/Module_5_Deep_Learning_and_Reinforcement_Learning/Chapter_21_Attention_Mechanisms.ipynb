# <font color='blue'> Attention Mechanisms: Learning Where to Focus </font>

Recurrent neural networks process sequences one element at a time while maintaining a hidden state that summarizes previously observed information. Although architectures such as Long Short-Term Memory (LSTM) networks and Gated Recurrent Units (GRUs) significantly improve the learning of long-range dependencies, they still compress the information contained in an entire input sequence into a limited hidden representation. As input sequences become longer, this fixed-size representation increasingly limits the amount of information that can be retained.

The **Attention Mechanism** addresses this limitation by allowing the model to dynamically access different parts of the input sequence whenever an output is produced. Rather than relying exclusively on a single hidden state, the model learns to assign different levels of importance—or **attention weights**—to different input elements.

Originally developed for neural machine translation, attention has become one of the most influential ideas in modern deep learning and forms the conceptual foundation of the Transformer architecture.

# <font color='orange'> 2. The Fixed-Length Bottleneck </font>

An encoder-decoder RNN performs

```
Input Sequence

↓

Encoder

↓

Single Vector

↓

Decoder

↓

Output Sequence
```

Regardless of whether the input contains

10 words

or

1000 words,

the encoder produces

one fixed-length representation.

This information bottleneck

limits performance

on long sequences.

---



# <font color='orange'> 3. The Central Idea </font>

Instead of relying only on

one vector,

the decoder

can examine

all encoder hidden states.

Conceptually,

```
Encoder Outputs

↓

Attention

↓

Relevant Information

↓

Decoder
```

The decoder therefore decides

where to focus

at each output step.

---



# <font color='orange'> 4. Encoder Hidden States </font>

Suppose the encoder processes

```
x₁

↓

x₂

↓

x₃

↓

...

↓

xₙ
```

Instead of keeping

only the final hidden state,

attention preserves

every hidden state

```
h₁

h₂

h₃

...

hₙ
```

Each hidden state

contains information

about one part

of the input sequence.

---



# <font color='orange'> 5. Query, Key, and Value </font>

Modern attention mechanisms

are based on

three vectors.

### Query (Q)

"What information am I looking for?"

---

### Key (K)

"What information does this input contain?"

---

### Value (V)

"What information should be returned?"

The decoder compares

the query

with every key

to determine

which values

should receive attention.

---



# <font color='orange'> 6. Computing Attention Scores </font>

The similarity

between

a query

and

each key

is computed.

Conceptually,

```
Query

↓

Compare

↓

Key₁

Key₂

Key₃

...

↓

Attention Scores
```

Higher scores

indicate

greater relevance.

Different scoring functions exist, including additive (Bahdanau) and multiplicative (Luong or dot-product) attention.

---



# <font color='orange'> 7. Softmax Normalization </font>

The attention scores

are converted

into probabilities

using

Softmax.

$$
\boxed{
\alpha_i
=
\frac{
e^{s_i}
}{
\sum_j e^{s_j}
}
}
$$

where

* \(s_i\) is the attention score,
* \(\alpha_i\) is the attention weight.

The weights satisfy

$$
\sum_i
\alpha_i
=
1.
$$

Thus,

they form

a probability distribution

over the input sequence.

---



# <font color='orange'> 8. Context Vector </font>

The final output

of the attention mechanism

is called

the **context vector**.

It is computed

as a weighted sum

of the value vectors.

$$
\boxed{
c
=
\sum_i
\alpha_i
V_i
}
$$

Input elements

receiving higher attention weights

contribute more

to the context vector.

| Type | Main Idea |

# <font color='orange'> 11. Advantages </font>

Attention provides several important benefits.

# <font color='orange'> 12. Limitations </font>

Despite its strengths,

* Standard attention requires comparing many pairs of sequence elements.
* The original attention mechanism still relied on recurrent encoder-decoder architectures.

These limitations motivated

# <font color='orange'> 14. Common Misconceptions </font>

### Attention Replaces the Entire Neural Network

False.

The original attention mechanism is an additional component used alongside encoder-decoder RNNs.

---

### The Model Looks at Only One Input Word

False.

Attention assigns a weight to every input element. Some receive higher weights than others, but all can contribute to the final context vector.

---

### Attention Weights Are Fixed

False.

Attention weights are computed dynamically for every output step and depend on both the input sequence and the current query.

---

### Attention Solves Every Sequence-Modelling Problem

False.

Attention greatly improves long-range dependency modelling, but efficient processing of very long sequences remains an active area of research.

| :--- | :--- |
| Attention | Mechanism for dynamically focusing on relevant parts of the input |
| Query | Represents what information is currently needed |
| Key | Represents what information each input element contains |
| Value | Information returned by each input element |

> **Key Insight:** The attention mechanism transforms sequence modelling by allowing the model to dynamically access the entire input sequence whenever an output is generated. Instead of compressing all information into a single hidden representation, attention computes relevance scores between a query and all input elements, producing a context vector that emphasizes the most informative parts of the sequence. This simple idea dramatically improved long-range dependency modelling and laid the conceptual foundation for the Transformer architecture.

# <font color='blue'> The Transformer Architecture: Attention Is All You Need </font>

Although attention mechanisms significantly improved recurrent neural networks by allowing the decoder to focus dynamically on different parts of the input sequence, the underlying architecture still relied on sequential processing. Since each hidden state depends on the previous one, RNNs, LSTMs, and GRUs cannot process all elements of a sequence simultaneously. This sequential dependency limits computational efficiency and makes training on very long sequences difficult.

In 2017, Ashish Vaswani and collaborators introduced the **Transformer** architecture in the landmark paper *Attention Is All You Need*. Instead of recurrence, the Transformer relies entirely on **self-attention**, allowing every element of a sequence to interact directly with every other element in parallel.

The Transformer dramatically improved both computational efficiency and modelling capability, becoming the foundation of nearly all modern large language models and many state-of-the-art systems in computer vision, speech recognition, reinforcement learning, and scientific machine learning.

# <font color='orange'> 1. Motivation </font>

Recall

an RNN.

```
Word₁

↓

Word₂

↓

Word₃

↓

Word₄
```

Each word

must wait

for the previous one.

Training therefore

cannot be fully parallelized.

This limits

both

training speed

and

---



# <font color='orange'> 2. The Central Idea </font>

Instead of processing

words sequentially,

the Transformer processes

the entire sequence

simultaneously.

Conceptually,

```
Input Sequence

↓

Self-Attention

↓

Output Sequence
```

Every token

can immediately interact

with every other token.

---



# <font color='orange'> 3. Self-Attention </font>

Unlike ordinary attention,

where a decoder attends to encoder outputs,

**self-attention**

allows every input token

to attend

to every other input token.

Conceptually,

```
Word₁

↔

Word₂

↔

Word₃

↔

Word₄
```

Every token exchanges information

with every other token.

---



# <font color='orange'> 4. Query, Key and Value </font>

Each token

is transformed into

three vectors.

```
Token

↓

Query (Q)

↓

Key (K)

↓

Value (V)
```

These vectors are learned

through linear transformations.

---



# <font color='orange'> 5. Scaled Dot-Product Attention </font>

The similarity

between tokens

is computed using

the dot product.

The attention weights are

$$
\boxed{
\mathrm{Attention}(Q,K,V)
=
\mathrm{Softmax}
\left(
\frac{QK^T}
{\sqrt{d_k}}
\right)
V
}
$$

where

* \(Q\) = Query matrix,
* \(K\) = Key matrix,
* \(V\) = Value matrix,
* \(d_k\) = dimension of the key vectors.

The scaling factor

\(\sqrt{d_k}\)

prevents excessively large dot products,

which helps stabilize the softmax computation and training.

---



# <font color='orange'> 6. Why Scaling Is Needed </font>

Suppose

the feature dimension

is large.

Dot products

naturally become larger.

Large values entering

Softmax

produce extremely peaked probability distributions,

leading to very small gradients.

Dividing by

$$
\sqrt{d_k}
$$

keeps the values

within a suitable numerical range,

making optimization more stable.

---



# <font color='orange'> 7. Multi-Head Attention </font>

Rather than using

one attention mechanism,

the Transformer uses

multiple attention heads.

Conceptually,

```
Input

↓

Head 1

↓

Head 2

↓

Head 3

↓

...

↓

Concatenate

↓

Linear Layer
```

Each head

can learn

different relationships.

For example,

one head may capture

syntax,

another

long-range dependencies,

and another

semantic relationships.

---



# <font color='orange'> 8. Positional Encoding </font>

Since Transformers process

all tokens simultaneously,

they have

no inherent notion

of word order.

Positional encodings

provide

information about position.

Conceptually,

```
Word Embedding

+

Position Encoding

↓

Transformer Input
```

The original Transformer uses deterministic sinusoidal positional encodings, although many modern models learn positional embeddings instead.

---



# <font color='orange'> 9. Encoder Block </font>

A Transformer encoder block consists of

```
Input

↓

Multi-Head Self-Attention

↓

Add & Layer Normalization

↓

Feed-Forward Network

↓

Add & Layer Normalization

↓

Output
```

Residual (skip) connections

help stabilize training,

while layer normalization

improves optimization.

---



# <font color='orange'> 10. Decoder Block </font>

The decoder contains

three major components.

```
Masked Self-Attention

↓

Encoder-Decoder Attention

↓

Feed-Forward Network
```

Masked attention prevents

future tokens

from being observed

during sequence generation.

This ensures

autoregressive prediction.

---

Unlike RNNs,

Transformers process

```
Word₁

Word₂

Word₃

Word₄

↓

Simultaneously
```

# <font color='orange'> 16. Common Misconceptions </font>

### Transformers Have No Memory

False.

Transformers do not maintain a recurrent hidden state, but they model dependencies through self-attention over the input (and, in autoregressive generation, over previously generated tokens).

---

### Self-Attention Is the Same as Ordinary Attention

False.

Ordinary attention typically relates a decoder state to encoder outputs.

Self-attention allows elements within the same sequence to attend to one another.

---

### Positional Encoding Is Optional

Not in the standard Transformer.

Without positional information, a vanilla Transformer cannot distinguish different orderings of the same tokens.

Some modern architectures use learned or relative positional encodings rather than the original sinusoidal form.

---

### Transformers Always Outperform Every Other Model

False.

Transformers achieve state-of-the-art performance on many tasks, but CNNs, GNNs, or recurrent models may be preferable depending on the data modality, dataset size, latency constraints, or available computational resources.
