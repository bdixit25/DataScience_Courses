# <font color='blue'> Backpropagation: Learning Through the Chain Rule </font>

Training a neural network requires determining how each weight and bias contributes to the overall prediction error. While gradient descent provides the rule for updating the network parameters, it does not explain how the required gradients are computed. For modern neural networks containing millions of parameters, computing each derivative independently would be computationally infeasible.

The **backpropagation algorithm** solves this problem by efficiently applying the **chain rule of calculus** to compute the gradient of the loss function with respect to every trainable parameter in the network. Rather than treating each parameter independently, backpropagation exploits the layered structure of neural networks, propagating error information backward from the output layer toward the input layer.

Backpropagation is therefore the mathematical engine that enables deep neural networks to learn. Combined with forward propagation and gradient-based optimization, it forms the foundation of virtually all modern deep learning systems.

---



# <font color='orange'> 1. Where We Are in the Learning Process </font>

Training a neural network consists of four main steps.

```
Input

↓

Forward Propagation

↓

Prediction

↓

Loss Function

↓

Backpropagation

↓

Gradients

↓

Gradient Descent

↓

Update Weights
```

Notice that

backpropagation

does **not**

update the parameters.

It only computes

the gradients.

---



# <font color='orange'> 2. Why Computing Gradients is Difficult </font>

Suppose a network contains

```
10 million parameters.
```

Computing

every derivative independently

would require enormous computational effort.

Instead,

we exploit

the layered structure

of the network.

Each layer depends only on

the previous layer,

allowing gradients to be computed efficiently using the chain rule.

---



# <font color='orange'> 3. The Chain Rule </font>

Backpropagation is based entirely on

the chain rule from calculus.

Suppose

$$
y=f(u),
$$

and

$$
u=g(x).
$$

Then

$$
\boxed{
\frac{dy}{dx}
=
\frac{dy}{du}
\cdot
\frac{du}{dx}
}
$$

Rather than differentiating the entire function directly,

we differentiate each intermediate step

and multiply the results together.

This simple rule makes deep learning possible.

---



# <font color='orange'> 4. Computational Graphs </font>

A neural network computation can be represented as

a computational graph.

Example

```
Input

↓

Weighted Sum

↓

Activation

↓

Loss
```

Each node represents

a mathematical operation.

Forward propagation evaluates

the graph

from top to bottom.

Backpropagation traverses

the graph

in reverse,

applying the chain rule at each step.

---



# <font color='orange'> 5. Forward Pass vs Backward Pass </font>

The two passes perform different tasks.

### Forward Pass

```
Input

↓

Prediction

↓

Loss
```

Purpose

* compute activations,
* compute predictions,
* compute the loss.

---

### Backward Pass

```
Loss

↓

Gradients

↓

Hidden Layers

↓

Input Layer
```

Purpose

* compute gradients,
* determine how each parameter influenced the loss.

---



# <font color='orange'> 6. A Simple Example </font>

Suppose

$$
z=wx+b,
$$

and

$$
\hat y=f(z),
$$

where

\(f\)

is an activation function.

The loss depends on

\(\hat y\).

To compute

$$
\frac{\partial L}{\partial w},
$$

we apply the chain rule.

$$
\boxed{
\frac{\partial L}{\partial w}
=
\frac{\partial L}{\partial \hat y}
\cdot
\frac{\partial \hat y}{\partial z}
\cdot
\frac{\partial z}{\partial w}
}
$$

Each derivative is simple,

and their product gives the desired gradient.

This is the central idea behind backpropagation.

---



# <font color='orange'> 7. Error Propagation Through Layers </font>

Suppose the network contains

three layers.

Conceptually,

```
Input

↓

Hidden Layer 1

↓

Hidden Layer 2

↓

Output

↓

Loss
```

During backpropagation,

the error flows

in the opposite direction.

```
Loss

↓

Output

↓

Hidden Layer 2

↓

Hidden Layer 1

↓

Input
```

Each layer receives

gradient information

from the layer above it.

---



# <font color='orange'> 8. Computing Gradients Layer by Layer </font>

For every layer,

backpropagation computes

```
Loss

↓

Gradient of Activation

↓

Gradient of Weighted Sum

↓

Gradient of Weights

↓

Gradient of Bias
```

These gradients are then passed

to the previous layer.

This recursive procedure allows gradients for all parameters to be computed efficiently.

---



# <font color='orange'> 9. Updating the Parameters </font>

Once the gradients are available,

gradient descent performs the update.

For the weights,

$$
\boxed{
W
\leftarrow
W
-
\eta
\frac{\partial L}{\partial W}
}
$$

For the biases,

$$
\boxed{
b
\leftarrow
b
-
\eta
\frac{\partial L}{\partial b}
}
$$

Thus,

backpropagation

computes the gradients,

while

gradient descent

uses them.

---



# <font color='orange'> 10. Why Backpropagation is Efficient </font>

A naive approach would repeatedly compute

the same derivatives.

Backpropagation avoids this by

reusing

intermediate gradient calculations.

Each derivative is computed only once,

making the algorithm computationally efficient even for very deep networks.

Without this efficiency,

modern deep learning would be impractical.

---



# <font color='orange'> 11. Automatic Differentiation </font>

Modern deep learning libraries

such as TensorFlow and PyTorch

implement

**automatic differentiation**.

Instead of manually deriving gradients,

the framework

constructs the computational graph,

applies the chain rule automatically,

and computes all required gradients.

Consequently,

practitioners rarely implement backpropagation by hand.

Understanding the algorithm, however,

remains essential for diagnosing training problems and interpreting optimization behaviour.

---



# <font color='orange'> 12. Practical Implementation in TensorFlow </font>

TensorFlow computes gradients using

`GradientTape`.

```python
import tensorflow as tf

with tf.GradientTape() as tape:

    predictions = model(X)

    loss = loss_function(y, predictions)

gradients = tape.gradient(

    loss,

    model.trainable_variables

)

optimizer.apply_gradients(

    zip(

        gradients,

        model.trainable_variables

    )

)
```

The framework performs the entire backpropagation algorithm automatically.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Backpropagation Updates the Weights

False.

Backpropagation computes

the gradients.

The optimizer updates the weights.

---

### The Error Travels Back Through the Data

False.

The training data remain unchanged.

Only gradient information propagates backward through the computational graph.

---

### Backpropagation Requires Numerical Differentiation

False.

Backpropagation computes

exact analytical gradients

using the chain rule,

making it both accurate and computationally efficient.

---



# <font color='orange'> 14. Limitations </font>

Although extremely successful,

backpropagation can encounter difficulties.

* Vanishing gradients.
* Exploding gradients.
* Large computational requirements for very deep networks.
* Dependence on differentiable activation functions.

These challenges motivate many of the optimization techniques studied later in this module.

> **Key Insight:** Backpropagation is the algorithm that enables deep neural networks to learn. By applying the chain rule of calculus to the computational graph, it efficiently computes the gradient of the loss function with respect to every weight and bias in the network. These gradients are then used by optimization algorithms such as gradient descent to update the network parameters. Although modern deep learning frameworks perform backpropagation automatically, understanding its underlying principles is essential for interpreting neural network training and diagnosing optimization challenges.
