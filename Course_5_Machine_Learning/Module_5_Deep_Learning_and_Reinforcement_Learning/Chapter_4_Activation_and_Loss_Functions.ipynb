# <font color='blue'> Activation Functions in Neural Networks </font>

Each neuron in a neural network first computes a weighted sum of its inputs and then applies a mathematical transformation known as an **activation function**. Although the weighted sum is a linear operation, the activation function introduces nonlinearity into the network, allowing neural networks to learn complex relationships that cannot be represented by linear models alone.

The importance of activation functions cannot be overstated. Without nonlinear activations, adding additional hidden layers would provide no additional expressive power, as the entire network would reduce to a single linear transformation. The remarkable success of deep learning therefore depends critically on the use of suitable nonlinear activation functions.

This chapter explains why activation functions are necessary, introduces the most widely used activation functions in modern deep learning, compares their properties, and discusses appropriate choices for different machine learning tasks.

---



# <font color='orange'> 1. Why Do We Need Activation Functions? </font>

Consider a neuron that computes

$$
z=\mathbf{w}^T\mathbf{x}+b.
$$

If the neuron outputs

$$
a=z,
$$

then the neuron performs

only a linear transformation.

Now suppose we stack many such layers.

Conceptually,

```
Input

↓

Linear Layer

↓

Linear Layer

↓

Linear Layer

↓

Output
```

Intuitively,

this appears more powerful.

Surprisingly,

it is not.

---



# <font color='orange'> 2. Why Multiple Linear Layers Do Not Help </font>

Suppose the first layer computes

$$
\mathbf{h}
=
W_1\mathbf{x}+b_1.
$$

The second layer computes

$$
\mathbf{y}
=
W_2\mathbf{h}+b_2.
$$

Substituting,

$$
\begin{aligned}
\mathbf{y}
&=
W_2(W_1\mathbf{x}+b_1)+b_2 \\
&=
(W_2W_1)\mathbf{x}
+
(W_2b_1+b_2).
\end{aligned}
$$

Notice that

$$
W_2W_1
$$

is simply another matrix,

and

$$
W_2b_1+b_2
$$

is simply another bias vector.

Therefore,

the two-layer network becomes

$$
\boxed{
\mathbf{y}
=
W\mathbf{x}+b,
}
$$

which is **still a linear model**.

The same argument extends to any number of purely linear layers.

Therefore,

without nonlinear activation functions,

deep neural networks offer **no additional expressive power** over ordinary linear models.

---



# <font color='orange'> 3. The Purpose of an Activation Function </font>

An activation function introduces

**nonlinearity**

into the network.

Conceptually,

```
Weighted Sum

↓

Activation Function

↓

Nonlinear Output
```

This allows neural networks to

* learn nonlinear decision boundaries,
* approximate highly complex functions,
* construct hierarchical feature representations.

Without activation functions,

modern deep learning would not exist.

---



# <font color='orange'> 4. The Step Function </font>

The earliest activation function was

the **step function**,

used by the perceptron.

$$
\boxed{
f(z)=
\begin{cases}
1,&z\ge0,\\
0,&z<0.
\end{cases}
}
$$

---

### Advantages

* Simple.
* Suitable for binary decisions.

---

### Limitations

* Not differentiable.
* Cannot be used with gradient-based optimization.
* Rarely used in modern deep learning.

---



# <font color='orange'> 5. The Sigmoid Function </font>

The sigmoid activation is

$$
\boxed{
\sigma(z)
=
\frac{1}{1+e^{-z}}
}
$$

Its output lies between

```
0

and

1.
```

---

### Properties

* Smooth.
* Differentiable.
* Can be interpreted as a probability.

---

### Advantages

* Well suited for binary classification outputs.

---

### Limitations

* Saturates for large positive or negative inputs.
* Suffers from the **vanishing gradient problem**.
* Hidden layers trained with sigmoid often learn slowly.

---



# <font color='orange'> 6. Hyperbolic Tangent (tanh) </font>

The hyperbolic tangent function is

$$
\boxed{
\tanh(z)
=
\frac{e^z-e^{-z}}
{e^z+e^{-z}}
}
$$

Its outputs lie between

```
-1

and

1.
```

---

### Advantages

* Zero-centered outputs.
* Often converges faster than sigmoid in hidden layers.

---

### Limitations

* Also suffers from vanishing gradients for large \(|z|\).

---



# <font color='orange'> 7. Rectified Linear Unit (ReLU) </font>

The Rectified Linear Unit is

$$
\boxed{
\mathrm{ReLU}(z)
=
\max(0,z)
}
$$

Conceptually,

```
Negative Input

↓

0
```

```
Positive Input

↓

Identity
```

---

### Advantages

* Computationally efficient.
* Reduces the vanishing gradient problem.
* Sparse activations.
* Fast training.

ReLU has become the default activation for hidden layers in many neural networks.

---

### Limitations

Some neurons may become permanently inactive if they consistently receive negative inputs.

This phenomenon is known as the

**dying ReLU problem**.

---



# <font color='orange'> 8. Leaky ReLU </font>

Leaky ReLU modifies ReLU by allowing a small negative slope.

$$
\boxed{
f(z)=
\begin{cases}
z,&z>0,\\
\alpha z,&z\le0,
\end{cases}
}
$$

where

$$
\alpha
\approx
0.01.
$$

This reduces the likelihood of neurons becoming permanently inactive.

---



# <font color='orange'> 9. Exponential Linear Unit (ELU) </font>

ELU further modifies the negative region.

$$
\boxed{
f(z)=
\begin{cases}
z,&z>0,\\
\alpha(e^z-1),&z\le0.
\end{cases}
}
$$

Advantages include

* smoother behaviour near zero,
* negative outputs closer to zero mean,
* improved optimization in some problems.

---



# <font color='orange'> 10. Softmax </font>

Softmax is used

primarily for

**multiclass classification**.

For class \(i\),

$$
\boxed{
P_i
=
\frac{e^{z_i}}
{\sum_j e^{z_j}}
}
$$

The outputs satisfy

* every probability lies between 0 and 1,
* the probabilities sum to one.

Softmax therefore converts arbitrary scores into a valid probability distribution over multiple classes.

---



# <font color='orange'> 11. Comparing Activation Functions </font>

| Activation | Output Range | Typical Use |
| :--- | :---: | :--- |
| Step | 0, 1 | Historical perceptrons |
| Sigmoid | (0, 1) | Binary output layer |
| Tanh | (-1, 1) | Occasionally hidden layers |
| ReLU | [0, ∞) | Default hidden layers |
| Leaky ReLU | (-∞, ∞) | Hidden layers when avoiding dying ReLU |
| ELU | (-α, ∞) | Hidden layers |
| Softmax | Probabilities | Multiclass output layer |

---



# <font color='orange'> 12. Choosing an Activation Function </font>

General recommendations are

| Task | Recommended Activation |
| :--- | :--- |
| Hidden Layers | ReLU (default) |
| Hidden Layers (ReLU issues) | Leaky ReLU or ELU |
| Binary Classification Output | Sigmoid |
| Multiclass Classification Output | Softmax |
| Regression Output | Linear (no activation) |

These recommendations reflect common modern practice, although specialized architectures may use alternative activation functions.

---

# <font color='orange'> 14. Common Misconceptions </font>

### More Hidden Layers Alone Make Networks Powerful

False.

Without nonlinear activation functions,

multiple linear layers collapse into a single linear transformation.

---

### Sigmoid Should Always Be Used

False.

Although historically important,

ReLU and its variants are generally preferred for hidden layers because they train more efficiently.

---

### Activation Functions Change During Training

False.

The **weights** and **biases** are learned.

The activation function itself is chosen by the practitioner and remains fixed throughout training.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Activation Function | Introduces nonlinearity into a neural network |
| Step Function | Historical activation used by the perceptron |
| Sigmoid | Smooth function producing outputs in (0, 1) |
| Tanh | Zero-centered activation with outputs in (-1, 1) |
| ReLU | Default activation for many hidden layers |
| Leaky ReLU | Mitigates the dying ReLU problem |
| ELU | Smooth alternative to ReLU |
| Softmax | Produces probability distributions for multiclass classification |
| Main Purpose | Enable neural networks to learn nonlinear functions |

> **Key Insight:** Activation functions are the mathematical mechanism that gives deep neural networks their expressive power. By introducing nonlinearity after each weighted sum, they allow networks to learn complex decision boundaries and hierarchical feature representations that cannot be captured by linear models alone. Among modern activation functions, ReLU has become the standard choice for hidden layers due to its computational efficiency and favourable optimization properties, while sigmoid and softmax remain the preferred choices for binary and multiclass output layers, respectively.

# <font color='blue'> Loss Functions in Deep Learning </font>

A neural network learns by making predictions and comparing them with the corresponding true target values. To improve its predictions, the network requires a quantitative measure of how far its outputs differ from the desired values. This measure is provided by the **loss function**, which assigns a numerical value to the prediction error. During training, the optimization algorithm seeks to adjust the network's parameters so that this loss becomes as small as possible.

The choice of loss function depends on the nature of the learning problem. Regression tasks typically measure the difference between continuous numerical values, whereas classification tasks compare predicted probabilities with discrete class labels. Selecting an appropriate loss function is therefore essential because it directly influences the optimization process and the behaviour of the trained model.

This chapter introduces the concept of loss functions, distinguishes between loss and cost functions, presents the most widely used loss functions in modern deep learning, and discusses their practical implementation using TensorFlow and Keras.

---



# <font color='orange'> 1. Why Do We Need a Loss Function? </font>

Suppose a neural network predicts

```
Predicted House Price

↓

£248,000
```

while the true value is

```
£250,000.
```

The prediction is close,

but

**how close?**

The network requires

a numerical measure

of prediction quality.

This numerical measure is called

the **loss**.

---



# <font color='orange'> 2. The Learning Process </font>

Training a neural network follows a simple cycle.

```
Input

↓

Prediction

↓

Loss Function

↓

Compute Error

↓

Update Weights

↓

Repeat
```

Every improvement in the network ultimately aims to reduce the loss.

---



# <font color='orange'> 3. Loss vs Cost Function </font>

Although the terms are often used interchangeably,

there is a useful distinction.

### Loss Function

Measures the error for

a **single training example**.

$$
\boxed{
L(y,\hat{y})
}
$$

---

### Cost (Objective) Function

Measures the average loss

over an entire dataset

or mini-batch.

$$
\boxed{
J
=
\frac{1}{n}
\sum_{i=1}^{n}
L(y_i,\hat{y}_i)
}
$$

Modern optimization algorithms minimize

the cost function.

---



# <font color='orange'> 4. Mean Squared Error (MSE) </font>

For regression,

the most common loss function is

the **Mean Squared Error**.

For a single observation,

$$
\boxed{
L
=
(y-\hat y)^2
}
$$

For an entire dataset,

$$
\boxed{
J
=
\frac1n
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
}
$$

---

### Interpretation

Large prediction errors

are penalized much more heavily

than small errors.

Example

```
Error = 10

↓

Loss = 100
```

```
Error = 2

↓

Loss = 4
```

This makes MSE sensitive to outliers.

---

### Typical Applications

* House price prediction
* Temperature forecasting
* Scientific regression problems

---



# <font color='orange'> 5. Mean Absolute Error (MAE) </font>

Another common regression loss is

the **Mean Absolute Error**.

For one example,

$$
\boxed{
L
=
|y-\hat y|
}
$$

For the dataset,

$$
\boxed{
J
=
\frac1n
\sum_{i=1}^{n}
|y_i-\hat y_i|
}
$$

---

### Advantages

* Easy to interpret.
* Less sensitive to outliers than MSE.

---

### Limitations

The absolute value is not differentiable at zero, making optimization mathematically less convenient than MSE. In practice, modern optimization libraries handle this appropriately.

---



# <font color='orange'> 6. Why MSE Is Often Preferred </font>

Although MAE is more robust to outliers,

MSE has two important advantages.

* It is smooth and differentiable everywhere.
* Large errors receive stronger penalties.

For these reasons,

MSE is widely used for training regression networks.

---



# <font color='orange'> 7. Binary Cross-Entropy </font>

Regression losses are not appropriate

for classification.

Instead,

binary classification typically uses

**Binary Cross-Entropy (BCE)**.

$$
\boxed{
L
=
-
\left[
y\log(\hat y)
+
(1-y)\log(1-\hat y)
\right]
}
$$

where

* \(y\in\{0,1\}\),
* \(\hat y\) is the predicted probability.

---

### Interpretation

The loss becomes

very small

when the predicted probability matches the true class,

and

very large

when the model is confidently wrong.

Example

```
True Label = 1

Prediction = 0.99

↓

Very Small Loss
```

```
True Label = 1

Prediction = 0.01

↓

Very Large Loss
```

---



# <font color='orange'> 8. Categorical Cross-Entropy </font>

For multiclass classification,

the standard loss is

**Categorical Cross-Entropy**.

Suppose there are

\(K\)

classes.

The loss is

$$
\boxed{
L
=
-
\sum_{i=1}^{K}
y_i
\log(\hat y_i)
}
$$

where

* \(y_i\) is the one-hot encoded target,
* \(\hat y_i\) is the predicted probability.

This loss is used together with

the **Softmax** activation function.

---



# <font color='orange'> 9. Sparse Categorical Cross-Entropy </font>

One-hot encoding is not always convenient.

Instead,

class labels may simply be integers.

Example

```
Dog → 0

Cat → 1

Horse → 2
```

Sparse Categorical Cross-Entropy

accepts

integer labels directly,

avoiding the need for one-hot encoding.

The underlying mathematical objective remains equivalent to categorical cross-entropy.

---



# <font color='orange'> 10. Choosing the Right Loss Function </font>

The choice depends entirely on the learning task.

| Problem | Typical Output Activation | Recommended Loss |
| :--- | :--- | :--- |
| Regression | Linear | Mean Squared Error (MSE) or Mean Absolute Error (MAE) |
| Binary Classification | Sigmoid | Binary Cross-Entropy |
| Multiclass Classification (One-Hot Labels) | Softmax | Categorical Cross-Entropy |
| Multiclass Classification (Integer Labels) | Softmax | Sparse Categorical Cross-Entropy |

Using an inappropriate combination of activation function and loss can significantly impair training.

---



# <font color='orange'> 11. Loss During Training </font>

As training progresses,

the objective is

```
Epoch

↓

Lower Loss

↓

Better Predictions
```

A typical training curve therefore shows

```
Loss

│

│●

│ ●

│  ●

│   ●

│    ●

└────────────────

Epoch
```

A steadily decreasing loss generally indicates that the network is learning useful patterns, although validation loss should also be monitored to detect overfitting.

---

# <font color='orange'> 13. Common Misconceptions </font>

### Lower Training Loss Always Means a Better Model

False.

A network may continue reducing the training loss while simultaneously overfitting the training data.

Validation performance must also be monitored.

---

### Accuracy and Loss Measure the Same Thing

False.

Accuracy measures the proportion of correct predictions.

Loss measures

how confident

and

how incorrect

the predictions are.

Two models can achieve the same accuracy while having very different losses.

---

### The Loss Function Changes During Training

False.

The form of the loss function remains fixed.

Only the network parameters (weights and biases) are updated to reduce its value.

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Loss Function | Measures prediction error for a single example |
| Cost Function | Average loss over a dataset or mini-batch |
| MSE | Standard regression loss that strongly penalizes large errors |
| MAE | Regression loss based on absolute errors |
| Binary Cross-Entropy | Standard loss for binary classification |
| Categorical Cross-Entropy | Standard loss for multiclass classification with one-hot labels |
| Sparse Categorical Cross-Entropy | Multiclass loss for integer labels |
| Main Goal | Quantify prediction error and provide the objective for optimization |

> **Key Insight:** The loss function provides the numerical objective that guides the learning process in deep neural networks. During training, the network repeatedly computes predictions, evaluates the resulting loss, and adjusts its parameters to reduce that loss. Different machine learning tasks require different loss functions: regression typically uses Mean Squared Error or Mean Absolute Error, whereas classification relies on variants of Cross-Entropy Loss. Choosing an appropriate loss function is essential because it directly determines how the network learns from data.
