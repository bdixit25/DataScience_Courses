# <font color='blue'> Introduction to Sequence Models </font>

Many machine learning problems involve data that possess an inherent sequential structure. Unlike images, where neighbouring pixels are arranged in two-dimensional space, sequential data consist of ordered observations whose interpretation depends on both their values and their positions within the sequence. Examples include natural language, speech, financial time series, sensor measurements, biological sequences, and scientific event streams.

Traditional feedforward neural networks assume that every input is independent of all others. While this assumption is appropriate for many classification and regression tasks, it becomes inadequate when previous observations influence future predictions. Sequence models address this limitation by introducing mechanisms that allow information to persist across multiple time steps.

This chapter introduces sequential data, explains why conventional neural networks struggle with sequence modelling, and motivates the development of Recurrent Neural Networks (RNNs), Long Short-Term Memory (LSTM) networks, Gated Recurrent Units (GRUs), and Transformers.

# <font color='blue'> Recurrent Neural Networks (RNNs): Learning from Sequential Data </font>

Many real-world problems involve sequences in which earlier observations influence later predictions. Traditional feedforward neural networks assume that each input is independent and therefore cannot naturally model temporal or sequential dependencies. This limitation motivated the development of **Recurrent Neural Networks (RNNs)**, the first class of neural networks specifically designed to process sequential data.

Unlike feedforward networks, an RNN maintains a **hidden state**, which acts as a memory of previously observed inputs. As each element of a sequence is processed, the hidden state is updated and passed to the next time step, allowing information to propagate through the sequence.

Although later architectures such as Long Short-Term Memory (LSTM) networks and Transformers have largely replaced vanilla RNNs in many applications, understanding RNNs is essential because they introduce the fundamental concepts of recurrence, hidden states, and sequence modelling.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to predict

the next word.

Sentence

```
The Higgs boson was discovered in
```

The next word

depends on

all previous words,

not only

the last one.

A conventional neural network

cannot remember

earlier inputs.

Instead,

we require

a model with memory.

---

# <font color='orange'> 2. The Central Idea </font>

An RNN processes

one element

of a sequence

at a time.

At each time step,

it receives

* the current input,
* the previous hidden state.

Conceptually,

```
Current Input

+

Previous Memory

↓

RNN

↓

Output

+

Updated Memory
```

The updated memory

is passed

to the next time step.

---

# <font color='orange'> 3. Hidden State </font>

The hidden state

stores information

about everything

the network has observed so far.

Denote the hidden state by

$$
h_t.
$$

At time step

\(t\),

the hidden state summarizes

the sequence

from

```
1

↓

2

↓

...

↓

t
```

It therefore acts

as the network's memory.

---

# <font color='orange'> 4. Recurrent Connection </font>

Unlike feedforward networks,

the output of the hidden layer

is fed back

into the network.

Conceptually,

```
        hₜ₋₁
          │
          ▼
xₜ ───► RNN ───► hₜ
            │
            ▼
           yₜ
```

The feedback connection

creates

the recurrence

that gives the architecture

its name.

---

# <font color='orange'> 5. Mathematical Formulation </font>

The hidden state is updated according to

$$
\boxed{
h_t
=
f
\left(
W_{xh}x_t
+
W_{hh}h_{t-1}
+
b_h
\right)
}
$$

where

* \(x_t\) is the current input,
* \(h_{t-1}\) is the previous hidden state,
* \(W_{xh}\) connects the input to the hidden state,
* \(W_{hh}\) connects the previous hidden state to the current one,
* \(b_h\) is the bias,
* \(f(\cdot)\) is typically **tanh** or **ReLU**.

The output is

$$
\boxed{
y_t
=
g
\left(
W_{hy}h_t
+
b_y
\right)
}
$$

where \(g(\cdot)\) depends on the task (e.g., softmax for classification).

---

# <font color='orange'> 6. Unrolling Through Time </font>

Although an RNN contains

a recurrent loop,

during training

it is helpful to visualize it

as a sequence of identical networks.

Conceptually,

```
x₁ → [RNN] → y₁
       │
       ▼
x₂ → [RNN] → y₂
       │
       ▼
x₃ → [RNN] → y₃
       │
       ▼
...
```

Each block

shares

the same weights.

This representation is called

**unrolling through time**.

---

# <font color='orange'> 7. Shared Parameters </font>

Unlike a deep feedforward network,

an RNN uses

the same parameters

at every time step.

Conceptually,

```
Time 1

↓

W
```

```
Time 2

↓

Same W
```

```
Time 3

↓

Same W
```

Parameter sharing

reduces

the total number of learnable parameters

and allows the model to process sequences of arbitrary length.

# <font color='orange'> 9. Advantages </font>

RNNs introduced several important ideas.

* Learn sequential dependencies.
* Process variable-length sequences.
* Share parameters across time.
* Maintain a hidden memory state.
* Suitable for many sequence-learning problems.

These innovations laid the foundation

for modern sequence modelling.

---

# <font color='orange'> 10. Limitations </font>

Despite their historical importance,

vanilla RNNs suffer from several limitations.

* Difficulty learning long-range dependencies.
* Vanishing gradients.
* Exploding gradients.
* Sequential computation limits parallelism.
* Slow training on long sequences.

These limitations motivated

LSTM

and

GRU

architectures.

# <font color='orange'> 11. TensorFlow Implementation </font>

TensorFlow provides a simple RNN layer.

```python
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.SimpleRNN(

        128,

        input_shape=(None, 64)

    ),

    tf.keras.layers.Dense(

        10,

        activation="softmax"

    )

])
```

Here,

* `None` represents a variable sequence length,
* `64` is the number of features at each time step,
* `128` is the hidden-state dimension.

# <font color='orange'> 13. Common Misconceptions </font>

### RNNs Store the Entire Sequence Perfectly

False.

The hidden state is a compressed representation of past information, not an exact copy of every previous input.

---

### Every Time Step Uses Different Weights

False.

An RNN shares the same weight matrices across all time steps.

---

### RNNs Can Learn Arbitrarily Long Dependencies

False.

Vanilla RNNs struggle with long-range dependencies due to vanishing and exploding gradients.

---

### RNNs Have Been Completely Replaced

Not entirely.

While Transformers dominate many NLP tasks, RNNs, LSTMs, and GRUs are still used in certain embedded systems, streaming applications, and resource-constrained environments.

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Recurrent Neural Network | Neural network designed for sequential data |
| Hidden State (\(h_t\)) | Memory summarizing previous inputs |
| Recurrence | Hidden state passed from one time step to the next |
| Shared Parameters | Same weights reused across all time steps |
| Unrolling | Viewing the RNN as a sequence of repeated cells |
| Main Advantage | Learns temporal dependencies |
| Main Limitation | Difficulty learning long-range dependencies |

> **Key Insight:** Recurrent Neural Networks extend traditional neural networks by introducing a hidden state that carries information from one time step to the next. This recurrent structure enables the model to process variable-length sequences and capture temporal dependencies using a shared set of parameters. Although vanilla RNNs struggle with long-term dependencies due to optimization difficulties, they introduced the fundamental concepts of memory and recurrence that underpin later architectures such as LSTMs, GRUs, and modern sequence models.
