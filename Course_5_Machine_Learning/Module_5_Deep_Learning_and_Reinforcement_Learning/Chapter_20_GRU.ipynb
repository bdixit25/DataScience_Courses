# <font color='blue'> Gated Recurrent Units (GRUs) </font>

Long Short-Term Memory (LSTM) networks successfully address many of the optimization challenges faced by vanilla Recurrent Neural Networks by introducing a dedicated memory cell together with multiple gating mechanisms. Although highly effective, LSTMs contain a relatively complex architecture with three gates and two internal states, resulting in increased computational cost and a larger number of trainable parameters.

The **Gated Recurrent Unit (GRU)**, introduced by Kyunghyun Cho and collaborators in 2014, simplifies the LSTM architecture while preserving its ability to model long-range dependencies. Instead of maintaining separate hidden and cell states, a GRU uses a single hidden state together with two gates: the **update gate** and the **reset gate**.

Despite its simpler design, the GRU often achieves performance comparable to LSTMs while requiring fewer parameters and faster training, making it an attractive choice for many sequence modelling tasks.

---



# <font color='orange'> 1. Motivation </font>

Recall the LSTM.

It contains

* hidden state,
* cell state,
* forget gate,
* input gate,
* output gate.

Although powerful,

this architecture is relatively complex.

Researchers therefore asked

```
Can we simplify

the architecture

without sacrificing

performance?
```

The result was

the GRU.

---



# <font color='orange'> 2. The Central Idea </font>

Unlike an LSTM,

a GRU maintains

only one state.

```
Hidden State

hₜ
```

There is

no separate cell state.

Furthermore,

three gates become

only two.

```
Update Gate

Reset Gate
```

This reduces

both

the number of parameters

and

the computational cost.

---



# <font color='orange'> 3. GRU Architecture </font>

A GRU receives

```
Current Input

xₜ
```

and

```
Previous Hidden State

hₜ₋₁
```

It produces

```
Updated Hidden State

hₜ
```

The hidden state

acts as both

short-term

and

long-term memory.

---



# <font color='orange'> 4. Update Gate </font>

The **update gate**

controls

how much previous information

should be retained.

It is defined by

$$
\boxed{
z_t
=
\sigma
\left(
W_zx_t
+
U_zh_{t-1}
+
b_z
\right)
}
$$

Interpretation

```
z ≈ 1

↓

Keep previous memory
```

```
z ≈ 0

↓

Replace with new information
```

The update gate therefore plays a role similar to the combined action of the LSTM's forget and input gates.

---



# <font color='orange'> 5. Reset Gate </font>

The **reset gate**

determines

how much past information

should be ignored

when computing the new candidate hidden state.

It is defined by

$$
\boxed{
r_t
=
\sigma
\left(
W_rx_t
+
U_rh_{t-1}
+
b_r
\right)
}
$$

Interpretation

```
r ≈ 0

↓

Ignore previous memory
```

```
r ≈ 1

↓

Use previous memory
```

This gate allows the model to forget irrelevant historical information when appropriate.

---



# <font color='orange'> 6. Candidate Hidden State </font>

Using the reset gate,

the GRU computes

a candidate hidden state.

$$
\boxed{
\tilde{h}_t
=
\tanh
\left(
W_hx_t
+
U_h
(r_t
\odot
h_{t-1})
+
b_h
\right)
}
$$

Only the retained portion

of the previous hidden state

contributes

to this candidate representation.

---



# <font color='orange'> 7. Hidden State Update </font>

The final hidden state

combines

the previous hidden state

and

the candidate state.

$$
\boxed{
h_t
=
(1-z_t)
\odot
\tilde{h}_t
+
z_t
\odot
h_{t-1}
}
$$

The update gate

determines

how much information

comes from

old memory

versus

new information.

---



# <font color='orange'> 8. Information Flow </font>

The GRU performs

the following sequence.

```
Previous Hidden State

↓

Reset Gate

↓

Candidate Hidden State

↓

Update Gate

↓

Updated Hidden State
```

Unlike the LSTM,

there is

no separate memory cell.

---



# <font color='orange'> 9. GRU vs LSTM </font>

| Property | LSTM | GRU |
| :--- | :---: | :---: |
| Hidden State | ✓ | ✓ |
| Cell State | ✓ | ✗ |
| Forget Gate | ✓ | ✗ |
| Input Gate | ✓ | ✗ |
| Output Gate | ✓ | ✗ |
| Update Gate | ✗ | ✓ |
| Reset Gate | ✗ | ✓ |
| Number of Parameters | Larger | Smaller |
| Training Speed | Slower | Faster |

# <font color='orange'> 10. Advantages </font>

GRUs provide several benefits.

* Simpler architecture.
* Fewer trainable parameters.
* Faster training.
* Good performance on many sequence tasks.
* Effective modelling of long-term dependencies.

Because of these advantages,

GRUs are widely used

when computational efficiency is important.

---



# <font color='orange'> 11. Limitations </font>

Despite their simplicity,

GRUs also have limitations.

* Sequential computation limits parallelism.
* Performance varies depending on the dataset and task.
* May not always outperform LSTMs.
* Modern Transformer architectures outperform both GRUs and LSTMs on many large-scale language tasks.

---



# <font color='orange'> 12. TensorFlow Implementation </font>

As with LSTMs,

the sequence length

may vary.

# <font color='orange'> 14. Common Misconceptions </font>

### GRUs Always Outperform LSTMs

False.

Performance depends on the dataset, sequence length, and application. In many cases, the two architectures achieve similar accuracy.

---

### GRUs Cannot Learn Long-Term Dependencies

False.

GRUs were specifically designed to improve the learning of long-range dependencies compared with vanilla RNNs, although their mechanism differs from that of LSTMs.

---

### GRUs Contain a Cell State

False.

A GRU maintains only a hidden state. The long-term memory mechanism is incorporated directly into the hidden-state updates.

---

### Fewer Parameters Always Means Better Performance

False.

A smaller model may train faster and require less memory, but a larger model can sometimes capture more complex patterns.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| GRU | Simplified gated recurrent neural network |
| Hidden State | Combined short-term and long-term memory |
| Update Gate | Controls how much previous information is retained |
| Reset Gate | Controls how much previous information is used when computing new information |
| Candidate Hidden State | Proposed new representation before the final update |
| Main Advantage | Simpler and faster than an LSTM |
| Main Limitation | Still processes sequences sequentially and is less parallelizable than Transformers |

> **Key Insight:** Gated Recurrent Units simplify the LSTM architecture by merging the hidden state and cell state into a single representation and replacing three gates with two. The update and reset gates enable the network to regulate the flow of information, allowing it to model long-term dependencies while reducing computational complexity. GRUs often achieve performance comparable to LSTMs with fewer parameters, making them a popular choice for efficient sequence modelling.
