# <font color='blue'> Introduction to Semantic Segmentation </font>

Image classification assigns a single label to an entire image, while object detection identifies and localizes objects using bounding boxes. Although bounding boxes provide approximate object locations, they do not precisely describe object boundaries. Many applications require much finer spatial understanding, where every pixel in the image must be assigned a semantic meaning.

**Semantic Segmentation** addresses this challenge by performing **pixel-wise classification**. Instead of predicting one label for the entire image or one label per bounding box, a semantic segmentation model predicts the class of every individual pixel. The output is therefore a dense map describing the semantic content of the entire image.

Semantic segmentation is fundamental to autonomous driving, medical image analysis, robotics, remote sensing, agriculture, industrial inspection, and numerous scientific imaging applications.

This chapter introduces the motivation for semantic segmentation, explains pixel-wise classification, discusses segmentation masks and evaluation metrics, and prepares the reader for architectures such as Fully Convolutional Networks (FCNs), U-Net, and DeepLab.

# <font color='orange'> 7. Evaluation Metrics </font>

Pixel-wise accuracy

is useful,

but often insufficient.

Suppose

is background.

A model predicting

only background

would still achieve

95% accuracy.

Instead,

**Intersection over Union (IoU)**

for each class.

For one class,

$$
\boxed{
\mathrm{IoU}
=
\frac{\text{True Positive}}
{\text{True Positive}
+
\text{False Positive}
+
\text{False Negative}}
}
$$

The average across all classes

is called

**Mean IoU (mIoU)**.

# <font color='blue'> Fully Convolutional Networks (FCNs) </font>

Traditional convolutional neural networks such as AlexNet, VGGNet, and ResNet were designed primarily for image classification. These architectures progressively reduce the spatial dimensions of feature maps before using fully connected layers to predict a single class label. While highly effective for classification, this design discards much of the spatial information required for dense prediction tasks such as semantic segmentation.

In 2015, Jonathan Long, Evan Shelhamer, and Trevor Darrell introduced **Fully Convolutional Networks (FCNs)**, the first deep learning architecture capable of performing end-to-end semantic segmentation. The key innovation was the replacement of fully connected layers with convolutional layers, allowing the network to produce spatially dense predictions for images of arbitrary size.

FCNs established the encoder-decoder paradigm that forms the foundation of many modern semantic segmentation architectures, including U-Net, SegNet, and DeepLab.

---

# <font color='orange'> 1. Motivation </font>

Consider a classification CNN.

```
Image

↓

Convolution

↓

Pooling

↓

Fully Connected

↓

Class Label
```

The fully connected layers

collapse

the spatial information

into a single vector.

This is ideal for

classification,

but unsuitable for

pixel-wise prediction.

---

# <font color='orange'> 2. The Central Idea </font>

Instead of

fully connected layers,

FCNs use

only

convolutional layers.

Conceptually,

```
Image

↓

Convolution

↓

Convolution

↓

Convolution

↓

Prediction Map
```

Because no fully connected layers are used,

the network preserves

the spatial arrangement

of the image.

---

# <font color='orange'> 3. Why Remove Fully Connected Layers? </font>

Suppose

a feature map is

```
7 × 7 × 512
```

A fully connected layer converts this into

```
4096
```

numbers.

The original

7×7

spatial structure

is lost.

Instead,

a convolutional layer produces

another feature map,

preserving

spatial relationships.

---

# <font color='orange'> 4. Images of Arbitrary Size </font>

Because FCNs contain

only convolutional layers,

they are not restricted

to a fixed input size.

Examples include

```
224 × 224
```

```
512 × 512
```

```
1024 × 768
```

The same network

can process

all of these images.

This flexibility is one of the major advantages of FCNs.

---

# <font color='orange'> 5. The Encoder </font>

The first part of the network

is called

the **encoder**.

Conceptually,

```
Image

↓

Conv

↓

Pool

↓

Conv

↓

Pool

↓

Feature Maps
```

The encoder extracts

high-level semantic features

while reducing

the spatial resolution.

---

# <font color='orange'> 6. The Decoder </font>

The feature maps produced by the encoder

are much smaller

than the original image.

To recover

pixel-level predictions,

the network must increase

the spatial resolution.

This is the role of

the **decoder**.

Conceptually,

```
Feature Maps

↓

Upsampling

↓

Prediction Map
```

---

# <font color='orange'> 7. Upsampling </font>

The decoder enlarges

small feature maps.

Suppose

the encoder produces

```
32 × 32
```

features.

Upsampling transforms them into

```
64 × 64

↓

128 × 128

↓

256 × 256
```

Eventually,

the output

matches

the original image resolution.

---

# <font color='orange'> 8. Transposed Convolution </font>

One common upsampling technique

is the

**transposed convolution**

(sometimes called a **deconvolution**, although this term is technically inaccurate).

Conceptually,

```
Small Feature Map

↓

Transposed Convolution

↓

Larger Feature Map
```

Unlike ordinary convolution,

transposed convolution

increases

the spatial dimensions.

It contains

learnable parameters,

allowing the network to learn

how to reconstruct

fine spatial details.

---

# <font color='orange'> 9. Skip Connections </font>

Repeated pooling

causes

fine image details

to disappear.

FCNs recover

this information

using

**skip connections**.

Conceptually,

```
Early Feature Map

──────────────►

↓

Encoder

↓

Decoder

↓

Combine Features
```

The decoder combines

high-resolution features

from early layers

with

high-level semantic information

from deeper layers.

This improves boundary accuracy.

---

# <font color='orange'> 10. Complete FCN Pipeline </font>

The overall architecture becomes

```
Input Image

↓

Encoder

↓

Compressed Feature Maps

↓

Decoder

↓

Upsampling

↓

Pixel-wise Predictions

↓

Segmentation Mask
```

Every pixel

receives

a class label.

---

# <font color='orange'> 11. Advantages </font>

FCNs introduced several important ideas.

* End-to-end semantic segmentation.
* Images of arbitrary size.
* Dense pixel-wise prediction.
* Learnable upsampling.
* Foundation for modern segmentation networks.

These ideas remain central

to semantic segmentation today.

---

# <font color='orange'> 12. Limitations </font>

Despite their importance,

early FCNs have limitations.

* Coarse segmentation boundaries.
* Loss of spatial detail after repeated pooling.
* Limited multi-scale feature fusion.

These challenges motivated

U-Net,

SegNet,

and

# <font color='orange'> 14. Common Misconceptions </font>

### FCNs Remove Convolutions

False.

FCNs remove the **fully connected layers**, not the convolutional layers.

---

### Transposed Convolution Reverses Convolution

False.

Despite the historical name "deconvolution," transposed convolution is **not** the mathematical inverse of convolution. It is a learnable operation that increases spatial resolution.

---

### FCNs Produce Bounding Boxes

False.

FCNs produce **segmentation masks**, assigning a class label to every pixel rather than predicting object locations with bounding boxes.

---

### FCNs Work Only for Fixed Image Sizes

False.

Because they contain only convolutional operations, FCNs naturally process images of varying spatial dimensions.

| :--- | :--- |
| Fully Convolutional Network | CNN containing only convolutional layers |
| Main Innovation | Replace fully connected layers with convolutions |
| Encoder | Extracts semantic features while reducing resolution |
| Decoder | Restores spatial resolution |
| Transposed Convolution | Learnable upsampling operation |
| Skip Connections | Recover fine spatial details from early layers |

> **Key Insight:** Fully Convolutional Networks transformed convolutional neural networks from image classifiers into dense prediction models by replacing fully connected layers with convolutional layers. This allows images of arbitrary size to be processed while preserving spatial information throughout the network. Combined with learnable upsampling and skip connections, FCNs established the encoder-decoder framework that underpins many modern semantic segmentation architectures.

# <font color='blue'> U-Net: High-Resolution Semantic Segmentation </font>

Fully Convolutional Networks demonstrated that convolutional neural networks could perform end-to-end semantic segmentation. However, repeated pooling operations significantly reduce spatial resolution, causing fine object boundaries and small structures to be lost. Although upsampling restores the image size, much of the detailed localization information has already disappeared.

In 2015, researchers at the University of Freiburg introduced **U-Net**, an encoder-decoder architecture designed specifically for biomedical image segmentation. The defining feature of U-Net is the use of **long skip connections**, which directly transfer high-resolution feature maps from the encoder to the decoder. These connections allow the decoder to recover fine spatial details while simultaneously exploiting the high-level semantic information learned by deeper layers.

Although originally developed for biomedical imaging, U-Net has become one of the most influential segmentation architectures in computer vision and scientific imaging.

---

# <font color='orange'> 1. Motivation </font>

Recall the FCN architecture.

```
Image

↓

Encoder

↓

Compressed Feature Maps

↓

Decoder

↓

Segmentation Mask
```

During encoding,

repeated pooling removes

fine spatial information.

This leads to

```
Blurry Boundaries

↓

Poor Localization
```

U-Net addresses this problem

through feature reuse.

---

# <font color='orange'> 2. The U-Shaped Architecture </font>

The network has a symmetric

encoder-decoder

structure.

Conceptually,

```
Input

↓

Encoder

↓

Bottleneck

↓

Decoder

↓

Output
```

The overall shape resembles

the letter

```
U
```

which gives the architecture its name.

---

# <font color='orange'> 3. Encoder (Contracting Path) </font>

The encoder extracts increasingly abstract features.

Each stage performs

```
Convolution

↓

ReLU

↓

Convolution

↓

ReLU

↓

Max Pooling
```

Pooling reduces

the spatial dimensions,

while increasing

the number of feature channels.

Example

```
572 × 572

↓

284 × 284

↓

142 × 142

↓

71 × 71
```

---

# <font color='orange'> 4. Bottleneck </font>

At the deepest point,

the network has

the smallest spatial resolution

and the largest number of feature channels.

Conceptually,

```
Small Feature Maps

↓

Rich Semantic Information
```

The bottleneck captures

high-level contextual information

about the image.

---

# <font color='orange'> 5. Decoder (Expanding Path) </font>

The decoder gradually restores

the original resolution.

Each stage performs

```
Upsampling

↓

Convolution

↓

ReLU

↓

Convolution

↓

ReLU
```

Eventually,

the feature maps

return to

the original image size.

---

# <font color='orange'> 6. Long Skip Connections </font>

This is the defining innovation

of U-Net.

Instead of relying only

on the compressed bottleneck representation,

the decoder also receives

high-resolution encoder features.

Conceptually,

```
Encoder Layer

──────────────►

↓

Bottleneck

↓

Decoder

↓

Concatenate
```

These long skip connections

preserve fine spatial information.

---

# <font color='orange'> 7. Concatenation </font>

Unlike ResNet,

which performs

addition,

U-Net performs

feature-map concatenation.

Conceptually,

```
Encoder Features

+

Decoder Features

↓

Concatenate

↓

Convolution
```

The decoder therefore has access to

both

high-level semantics

and

fine image details.

---

# <font color='orange'> 8. Why Skip Connections Help </font>

Suppose

the encoder detects

an object boundary

before pooling.

Without a skip connection,

this boundary information

may be lost.

With a skip connection,

the decoder receives

the original high-resolution features,

allowing it to reconstruct

sharp object boundaries.

---

# <font color='orange'> 9. Complete U-Net Pipeline </font>

The overall workflow becomes

```
Input Image

↓

Encoder

↓

Bottleneck

↓

Decoder

↓

Pixel-wise Predictions

↓

Segmentation Mask
```

Long skip connections

link corresponding encoder

and decoder stages

at matching resolutions.

---

# <font color='orange'> 10. Advantages </font>

U-Net provides several important benefits.

* Excellent localization accuracy.
* Sharp object boundaries.
* Effective with relatively small datasets.
* End-to-end training.
* Widely applicable beyond biomedical imaging.

Its balanced combination of semantic understanding and spatial precision explains its enduring popularity.

---

# <font color='orange'> 11. Limitations </font>

Despite its success,

U-Net also has limitations.

* High memory usage due to feature-map concatenation.
* Computational cost increases with image size.
* Original architecture processes images at a single scale.
* Later architectures improve multi-scale context.

These limitations motivated architectures such as DeepLab and UNet++.

The concatenation step

# <font color='blue'> DeepLab: Multi-Scale Semantic Segmentation with Atrous Convolutions </font>

Semantic segmentation requires both precise localization and rich contextual understanding. While encoder-decoder architectures such as U-Net recover fine spatial details through skip connections, repeated pooling operations still reduce feature-map resolution and may limit the network's ability to capture large-scale context efficiently.

The **DeepLab** family of architectures addresses these challenges by introducing **atrous (dilated) convolutions**, which enlarge the receptive field without reducing spatial resolution or increasing the number of learnable parameters. Later versions further introduced **Atrous Spatial Pyramid Pooling (ASPP)**, allowing the network to capture contextual information at multiple spatial scales simultaneously.

DeepLab has become one of the most influential semantic segmentation architectures and remains a benchmark for dense prediction tasks.

---

# <font color='orange'> 1. Motivation </font>

Suppose we wish to classify

one pixel.

Looking only at

its immediate neighbours

may be insufficient.

```
Pixel

↓

Small Context

↓

Ambiguous Prediction
```

We instead require

both

local details

and

global context.

---

# <font color='orange'> 2. Receptive Field </font>

Recall

the receptive field.

It is

the region of the input image

that influences

one output feature.

Conceptually,

```
Input Image

↓

Convolution Layers

↓

Output Pixel
```

The deeper the network,

the larger

the receptive field.

However,

pooling also reduces

the spatial resolution.

---

# <font color='orange'> 3. Atrous (Dilated) Convolution </font>

Instead of increasing

the kernel size,

DeepLab inserts

spaces

between kernel elements.

For example,

a standard

```
3 × 3
```

kernel becomes

```
x   x   x

  x   x

x   x   x
```

where the blank positions represent skipped input locations.

The kernel still contains

only

nine learnable weights,

but now covers

a much larger region

of the image.

---

# <font color='orange'> 4. Dilation Rate </font>

The spacing between kernel elements

is controlled by

the **dilation rate**.

### Dilation = 1

Standard convolution.

```
xxx
xxx
xxx
```

---

### Dilation = 2

```
x . x . x

. . . . .

x . x . x

. . . . .

x . x . x
```

---

### Dilation = 3

Even larger spacing,

producing

an even larger receptive field.

Increasing the dilation rate enlarges the receptive field without adding parameters.

---

# <font color='orange'> 5. Why Atrous Convolutions Help </font>

Traditional methods

increase the receptive field

by

```
Pooling

↓

Lower Resolution
```

Atrous convolutions instead achieve

```
Larger Receptive Field

↓

Same Resolution
```

This is particularly valuable

for semantic segmentation,

where precise localization is essential.

---

# <font color='orange'> 6. Atrous Spatial Pyramid Pooling (ASPP) </font>

Different objects

appear

at different scales.

DeepLab therefore applies

multiple atrous convolutions

in parallel.

Conceptually,

```
Feature Map

├── Dilation 6

├── Dilation 12

├── Dilation 18

└── Image-Level Pooling

↓

Concatenate

↓

Output Features
```

This module is called

**Atrous Spatial Pyramid Pooling (ASPP).**

---

# <font color='orange'> 7. Multi-Scale Context </font>

Each ASPP branch

captures

different spatial scales.

| Branch | Captures |
| :--- | :--- |
| Small dilation | Fine local details |
| Medium dilation | Intermediate structures |
| Large dilation | Global context |
| Image pooling | Scene-level information |

Together,

they provide

a rich multi-scale representation.

---

# <font color='orange'> 8. DeepLab Pipeline </font>

A simplified DeepLab architecture is

```
Input Image

↓

CNN Backbone

↓

Atrous Convolutions

↓

ASPP

↓

Decoder

↓

Segmentation Mask
```

Modern versions

often use

ResNet

or

Xception

as the backbone.

---

# <font color='orange'> 9. DeepLab Versions </font>

The DeepLab family evolved over several years.

| Version | Major Innovation |
| :--- | :--- |

# <font color='orange'> 10. Advantages </font>

DeepLab provides several important benefits.

* Large receptive field.
* Preserves spatial resolution.
* Multi-scale context.
* Excellent segmentation accuracy.
* Flexible backbone architectures.

It performs particularly well

---

* Very large dilation rates can produce "gridding" artifacts, where sampled input locations become sparse.

# <font color='orange'> 14. Common Misconceptions </font>

### Atrous Convolution Adds More Parameters

False.

Atrous convolution uses the same kernel weights as a standard convolution; it simply spaces them farther apart.

---

### Larger Dilation Always Improves Performance

False.

Excessively large dilation rates may skip important local information and can introduce gridding artifacts. Multiple dilation rates are therefore often combined in ASPP.

---

### DeepLab Eliminates the Need for an Encoder

False.

DeepLab still relies on a CNN backbone (encoder) to extract meaningful feature representations before applying atrous convolutions and ASPP.

---

### DeepLab Is Only for Natural Images

False.

DeepLab has been successfully applied to medical imaging, remote sensing, microscopy, and other scientific imaging tasks.

> **Key Insight:** DeepLab advances semantic segmentation by replacing repeated spatial downsampling with atrous convolutions, allowing the receptive field to grow while preserving feature-map resolution. Combined with Atrous Spatial Pyramid Pooling, which captures information across multiple spatial scales, DeepLab achieves highly accurate pixel-wise segmentation in complex scenes. These innovations have made the DeepLab family one of the most influential architectures in modern semantic segmentation.

# <font color='blue'> Mask R-CNN: Instance Segmentation </font>

Object detection predicts the location and class of objects using bounding boxes, while semantic segmentation assigns a semantic class label to every pixel in an image. However, semantic segmentation does not distinguish between different instances of the same object class. For example, if an image contains three people, every person pixel receives the same label.

**Instance segmentation** combines the strengths of object detection and semantic segmentation by simultaneously identifying individual objects, localizing them, and predicting a separate segmentation mask for each instance.

Introduced in 2017 by Kaiming He and collaborators, **Mask R-CNN** extends Faster R-CNN by adding a third prediction branch that generates pixel-wise segmentation masks. The architecture also introduces **RoI Align**, which replaces RoI Pooling and significantly improves localization accuracy.

Mask R-CNN has become one of the most influential instance segmentation architectures in modern computer vision.

# <font color='orange'> 4. Why RoI Pooling Was Not Enough </font>

Recall

Fast R-CNN

used

RoI Pooling.

RoI Pooling

rounds

feature coordinates

to integer locations.

This small approximation

has little effect

for bounding boxes,

but significantly degrades

pixel-level segmentation.

---

# <font color='orange'> 5. RoI Align </font>

Mask R-CNN replaces

RoI Pooling

with

**RoI Align**.

Instead of rounding,

RoI Align

uses

bilinear interpolation

to sample feature values

at precise locations.

Conceptually,

```
Feature Map

↓

RoI Align

↓

Accurate Feature Extraction
```

This preserves

fine spatial detail.

# <font color='orange'> 13. Common Misconceptions </font>

### Mask R-CNN Replaces Faster R-CNN

False.

Mask R-CNN extends Faster R-CNN by adding an additional segmentation branch.

---

### Semantic Segmentation and Instance Segmentation Are the Same

False.

Semantic segmentation labels every pixel by class.

Instance segmentation additionally separates different objects belonging to the same class.

---

### RoI Align Is Identical to RoI Pooling

False.

RoI Align avoids coordinate quantization by using interpolation, improving pixel-level localization.

---

### The Mask Branch Predicts One Mask for the Entire Image

False.

Mask R-CNN predicts one mask for each detected object (region of interest).

| Mask R-CNN | Extension of Faster R-CNN for instance segmentation |
| RoI Align | Precise feature extraction using interpolation |
| Three Prediction Heads | Classification, bounding box regression, mask prediction |
| Mask Loss | Pixel-wise loss for each detected object |
| Main Advantage | Simultaneous detection and pixel-level segmentation |
| Main Limitation | Higher computational cost than one-stage approaches |

> **Key Insight:** Mask R-CNN extends Faster R-CNN by adding a dedicated mask prediction branch, enabling simultaneous object detection and instance segmentation. By replacing RoI Pooling with the more precise RoI Align operation, the architecture preserves fine spatial information necessary for accurate pixel-level masks. This elegant multi-task framework allows each detected object to receive its own class label, bounding box, and segmentation mask, making Mask R-CNN one of the foundational architectures for instance-aware computer vision.
