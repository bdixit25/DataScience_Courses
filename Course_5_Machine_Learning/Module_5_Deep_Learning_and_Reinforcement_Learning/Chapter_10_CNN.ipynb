# <font color='blue'> Motivation for Convolutional Neural Networks (CNNs) </font>

Artificial Neural Networks (ANNs), particularly Multilayer Perceptrons (MLPs), have demonstrated remarkable success on structured data such as tabular datasets. However, when applied directly to image data, fully connected networks become computationally inefficient and fail to exploit the spatial structure inherent in images. As image resolution increases, the number of trainable parameters grows rapidly, leading to excessive memory requirements, slow training, and severe overfitting.

Convolutional Neural Networks (CNNs) were developed to overcome these limitations. Rather than treating every pixel as an independent input, CNNs exploit the local spatial relationships between neighbouring pixels through the use of convolutional filters and weight sharing. These architectural innovations dramatically reduce the number of trainable parameters while enabling the network to automatically learn hierarchical visual features.

This chapter motivates the development of CNNs by examining the shortcomings of fully connected networks for image analysis and introducing the key principles that distinguish convolutional architectures.

---



# <font color='orange'> 1. Images Are Different from Tabular Data </font>

Consider a typical tabular dataset.

Changing one feature,

does not directly affect the meaning of another feature.

Images are fundamentally different.

An image is a structured arrangement of pixels,

where neighbouring pixels are strongly correlated.

Ignoring this spatial structure discards valuable information.

# <font color='orange'> 8. Three Key Ideas Behind CNNs </font>

CNNs learn increasingly complex features.

Conceptually,

```
Pixels

↓

Edges

↓

Corners

↓

Textures

↓

Object Parts

↓

Objects
```

Higher layers build upon the features learned by earlier layers.

# <font color='blue'> Images as Tensors </font>

Computers do not interpret images as visual objects. Instead, an image is represented numerically as a collection of pixel intensity values arranged in a structured grid. Understanding tensor representations is fundamental to Convolutional Neural Networks (CNNs), because convolution, pooling, and feature extraction all operate directly on image tensors. Before studying these operations, it is therefore essential to understand how images are stored, represented, and manipulated mathematically.

This chapter introduces pixel values, grayscale and colour image representations, tensor dimensions, image batches, and the tensor conventions used in modern deep learning frameworks.

# <font color='blue'> The Convolution Operation </font>

The defining operation of a Convolutional Neural Network (CNN) is the **convolution**. Unlike a fully connected layer, which connects every input feature to every neuron, convolution operates locally by applying a small learnable filter to neighbouring regions of an image. This allows the network to detect meaningful visual patterns while preserving the spatial structure of the input.

Convolution dramatically reduces the number of trainable parameters through **weight sharing**, enabling CNNs to process high-resolution images efficiently. Moreover, by repeatedly applying multiple filters throughout successive layers, CNNs automatically learn hierarchical visual representations, progressing from simple edges and textures to complex object parts and complete objects.

This chapter introduces the mathematical formulation of convolution, explains kernels and feature maps, illustrates the sliding-window mechanism through numerical examples, and discusses why convolution is particularly effective for image analysis.

---



# <font color='orange'> 1. What is a Convolution? </font>

A convolution applies

a small matrix,

called a

**kernel** (or **filter**),

to different regions of an image.

Conceptually,

```
Image

↓

Small Filter

↓

Slide Across Image

↓

Feature Map
```

The same filter is reused

everywhere in the image.

---



# <font color='orange'> 2. The Kernel (Filter) </font>

A kernel is a small matrix of learnable weights.

Typical sizes include

```
3 × 3

5 × 5

7 × 7
```

Example

$$
\boxed{
K=
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}
}
$$

Initially,

the kernel values are random.

During training,

backpropagation learns

the most useful values automatically.

---



# <font color='orange'> 3. The Sliding Window </font>

The kernel is placed over a small region of the image.

```
□□□□□□□□□□□□□□□□

□□■□

□□□■

■□□□

□□□□□□□□□□□□□□□□
```

It computes

one output value,

then slides

to the next position.

Conceptually,

```
Patch

↓

Multiply

↓

Sum

↓

One Output Pixel
```

This process repeats

until the entire image has been scanned.

---



# <font color='orange'> 4. The Dot Product </font>

At each position,

the kernel and image patch are multiplied element-wise,

then summed.

Suppose

the image patch is

$$
\begin{bmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
7 & 8 & 9
\end{bmatrix}
$$

and the kernel is

$$
\begin{bmatrix}
1 & 0 & -1\\
1 & 0 & -1\\
1 & 0 & -1
\end{bmatrix}.
$$

The output becomes

$$
\begin{aligned}
&(1)(1)+(2)(0)+(3)(-1)\\
+&(4)(1)+(5)(0)+(6)(-1)\\
+&(7)(1)+(8)(0)+(9)(-1).
\end{aligned}
$$

Evaluating,

$$
=
1-3+4-6+7-9
=
-6.
$$

This single number forms one element of the output feature map.

---



# <font color='orange'> 5. Mathematical Definition </font>

For an input image \(I\) and kernel \(K\),

the convolution output at location \((i,j)\) is

$$
\boxed{
Y(i,j)
=
\sum_m
\sum_n
I(i+m,j+n)
K(m,n)
}
$$

where

* \(I\) is the input image,
* \(K\) is the kernel,
* \(Y\) is the resulting feature map.

In practice, deep learning libraries implement the closely related cross-correlation operation, which uses the same expression without flipping the kernel.

---



# <font color='orange'> 6. Feature Maps </font>

The output produced by a convolution is called a

**feature map**

(or **activation map**).

Conceptually,

```
Input Image

↓

Convolution

↓

Feature Map
```

Each feature map highlights

where a particular visual pattern appears in the image.

---



# <font color='orange'> 7. Why Do Kernels Detect Features? </font>

Different kernels respond strongly to different patterns.

Examples include

* vertical edges,
* horizontal edges,
* diagonal edges,
* textures,
* corners.

During training,

the network automatically learns

which kernels are most useful.

Unlike classical computer vision,

the filters are **not handcrafted**.

---



# <font color='orange'> 8. Multiple Filters </font>

A CNN rarely uses

only one kernel.

Instead,

many filters are learned simultaneously.

Example

```
Input Image

↓

Filter 1

↓

Feature Map 1

```

```
Input Image

↓

Filter 2

↓

Feature Map 2

```

```
Input Image

↓

Filter 3

↓

Feature Map 3
```

If

32 filters

are used,

the convolutional layer produces

32 feature maps.

Each filter specializes in detecting different visual features.

---



# <font color='orange'> 9. Parameter Efficiency </font>

Suppose

a convolution uses

a

```
3 × 3
```

kernel.

The number of learnable weights is

$$
3 \times 3 = 9.
$$

Including one bias,

only

```
10 parameters
```

are learned,

regardless of whether the input image is

```
32 × 32

or

224 × 224.
```

This contrasts sharply with fully connected layers,

whose parameter count grows directly with the input size.

---



# <font color='orange'> 10. Why Weight Sharing Matters </font>

The same kernel is reused

across the entire image.

Conceptually,

```
One Filter

↓

Top Left

↓

Centre

↓

Bottom Right
```

A vertical edge remains

a vertical edge

regardless of its position.

Weight sharing enables the network to detect the same pattern anywhere in the image while using only a small number of parameters.

---

* `kernel_size=(3,3)` specifies the size of each kernel.

---



# <font color='orange'> 12. Common Misconceptions </font>

### The Kernel Is Fixed

False.

Unlike classical image processing,

CNN kernels are learned automatically through backpropagation.

---

### A Larger Kernel Is Always Better

False.

Small kernels (such as \(3 \times 3\)) are often preferred because stacking multiple small convolutions provides a larger effective receptive field with fewer parameters and more nonlinearities.

---

### Every Filter Learns the Same Feature

False.

Different filters learn different visual patterns.

Some specialize in edges,

others in textures,

corners,

or more complex structures.

---

### Convolution Changes the Image

Not exactly.

Convolution produces a **new representation** (a feature map) that emphasizes specific patterns in the input image.

---

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Convolution | Local operation that applies a learnable kernel to an image |
| Kernel (Filter) | Small matrix of trainable weights |
| Sliding Window | Moves the kernel across the image |
| Dot Product | Computes one output value at each position |
| Feature Map | Output highlighting where a feature is present |

| Main Advantage | Efficient feature extraction with relatively few parameters |

> **Key Insight:** Convolution is the fundamental operation underlying Convolutional Neural Networks. By applying small, learnable filters across local image regions, CNNs efficiently detect visual patterns while preserving spatial structure. Weight sharing dramatically reduces the number of trainable parameters, and multiple filters enable the network to learn increasingly sophisticated feature representations. These properties make convolution exceptionally well suited for image analysis and form the mathematical foundation of modern computer vision.

# <font color='blue'> Filters, Feature Maps, and Channels </font>

The convolution operation applies a learnable filter (kernel) to an image and produces a new representation known as a **feature map**. In practical convolutional neural networks, however, images usually contain multiple colour channels, and each convolutional layer typically learns dozens or even hundreds of filters simultaneously. Consequently, understanding how filters interact with image channels is essential for understanding the architecture of modern CNNs.

This chapter explains the relationship between filters, feature maps, and channels, derives the dimensions of convolutional kernels for multi-channel inputs, discusses how the number of learnable parameters is computed, and illustrates why deeper convolutional layers naturally produce increasingly rich feature representations.

---



# <font color='orange'> 1. From One Filter to One Feature Map </font>

Suppose we have

one grayscale image

and

one filter.

Conceptually,

```
Input Image

↓

One Filter

↓

One Feature Map
```

A single filter searches for

one particular visual pattern.

For example,

it may learn to detect

* vertical edges,
* horizontal edges,
* corners,
* textures.

Each filter therefore produces

exactly

one feature map.

---



# <font color='orange'> 2. Multiple Filters </font>

Suppose we instead learn

four filters.

Conceptually,

```
Input Image

↓

Filter 1

↓

Feature Map 1
```

```
Input Image

↓

Filter 2

↓

Feature Map 2
```

```
Input Image

↓

Filter 3

↓

Feature Map 3
```

```
Input Image

↓

Filter 4

↓

Feature Map 4
```

Each filter specializes in detecting

a different visual feature.

The outputs are stacked together,

forming the output tensor.

---



# <font color='orange'> 3. Understanding Channels </font>

Recall that

an RGB image contains

three channels.

Conceptually,

```
Red

Green

Blue
```

Each channel is itself

a two-dimensional image.

Thus,

an RGB image has shape

$$
\boxed{
H \times W \times 3
}
$$

where

* \(H\) is the height,
* \(W\) is the width,
* 3 represents the colour channels.

---



# <font color='orange'> 4. How Does a Filter Operate on an RGB Image? </font>

A common misconception is that

a

```
3 × 3
```

filter is applied independently to each colour channel.

In reality,

each filter spans

**all input channels**.

For an RGB image,

a filter has dimensions

$$
\boxed{
3 \times 3 \times 3.
}
$$

Conceptually,

```
Red Channel

↓

Green Channel

↓

Blue Channel

↓

One 3×3×3 Filter

↓

One Feature Map
```

The filter combines information

from all three colour channels simultaneously.

---



# <font color='orange'> 5. Computing One Output Pixel </font>

Suppose the filter size is

$$
3 \times 3 \times 3.
$$

For one image location,

the filter covers

* 9 values from the Red channel,
* 9 values from the Green channel,
* 9 values from the Blue channel.

The convolution computes

27 multiplications,

adds the results,

adds a bias,

and produces

one output value.

Thus,

the colour information is fused into a single response.

---



# <font color='orange'> 6. Why Does One Filter Produce Only One Output Channel? </font>

Although the filter processes

all input channels,

it produces

only one numerical response

at each spatial location.

Consequently,

```
One Filter

↓

One Feature Map

↓

One Output Channel
```

Therefore,

the number of output channels

equals

the number of filters.

---



# <font color='orange'> 7. Example </font>

Suppose the input tensor has shape

$$
224 \times 224 \times 3
$$

and the convolutional layer contains

```
32 filters.
```

Each filter has size

$$
3 \times 3 \times 3.
$$

The output becomes

approximately

$$
224 \times 224 \times 32
$$

(assuming padding preserves the spatial dimensions).

Notice

* input channels = 3,
* output channels = 32.

The output channels correspond

to the number of learned filters.

---



# <font color='orange'> 8. Parameter Calculation </font>

Suppose

the convolution uses

```
64 filters
```

of size

```
3 × 3
```

on an RGB image.

Each filter contains

$$
3 \times 3 \times 3
=
27
$$

weights,

plus

one bias.

Therefore,

each filter learns

$$
27+1=28
$$

parameters.

The entire layer therefore learns

$$
64 \times 28
=
1792
$$

parameters.

Notice how this number

is completely independent

of the image resolution.

Whether the input image is

```
32×32

or

1024×1024,
```

the parameter count remains

1792.

This remarkable property explains why CNNs scale efficiently to large images.

---



# <font color='orange'> 9. Deeper Convolutional Layers </font>

After the first convolution,

the output tensor may have shape

```
224 × 224 × 32.
```

The next convolutional layer now receives

32 input channels.

Consequently,

its filters become

$$
\boxed{
3 \times 3 \times 32.
}
$$

If this layer learns

# <font color='blue'> Padding in Convolutional Neural Networks </font>

The convolution operation applies a kernel to local regions of an input image to produce a feature map. Although convolution efficiently extracts meaningful visual features, it also reduces the spatial dimensions of the input whenever the kernel extends beyond the image boundaries. After several convolutional layers, this continual reduction may remove important spatial information and severely limit the depth of the network.

**Padding** addresses this problem by adding additional pixels around the border of the input image before performing the convolution. By appropriately choosing the amount of padding, convolutional neural networks can preserve spatial dimensions, improve feature extraction near image boundaries, and enable the construction of much deeper architectures.

This chapter explains why convolution reduces image size, introduces the concept of padding, derives the general output size formula, and discusses the most common padding strategies used in modern CNNs.

---



# <font color='orange'> 1. Why Does Convolution Reduce Image Size? </font>

Consider a

```
5 × 5
```

image

and a

```
3 × 3
```

kernel.

The kernel cannot be centred on pixels near the image boundary because part of the kernel would lie outside the image.

Consequently,

the kernel can only occupy

```
3 × 3
```

valid positions.

Therefore,

```
5 × 5

↓

3 × 3 Convolution

↓

3 × 3 Output
```

The output becomes smaller than the input.

---



# <font color='orange'> 2. A Simple Numerical Example </font>

Suppose

the input image has size

```
7 × 7
```

and we apply

a

```
3 × 3
```

kernel.

Without padding,

the output size is

$$
7-3+1=5.
$$

Thus,

```
Input

7 × 7

↓

Output

5 × 5
```

Every convolution reduces each spatial dimension by

```
Kernel Size − 1.
```

---



# <font color='orange'> 3. The Problem with Repeated Convolutions </font>

Suppose we repeatedly apply

```
3 × 3
```

convolutions.

```
224 × 224

↓

222 × 222

↓

220 × 220

↓

218 × 218

↓

...
```

Eventually,

the feature maps become

very small.

Important information,

particularly near the image boundaries,

may be lost.

---



# <font color='orange'> 4. The Central Idea of Padding </font>

Padding adds

extra pixels

around the image border

before convolution.

Conceptually,

Without padding

```
□□□□□

□□□□□

□□□□□

□□□□□

□□□□□
```

With padding

```
□□□□□□□

□□□□□□

□□□□□□

□□□□□□

□□□□□□

□□□□□□

□□□□□□□
```

The additional border allows the kernel to process pixels near the edges.

---



# <font color='orange'> 5. Zero Padding </font>

The most common padding strategy is

**zero padding**.

Extra pixels with value

```
0
```

are added around the image.

Example

Original image

$$
\begin{bmatrix}
1 & 2 & 3\\
4 & 5 & 6\\
7 & 8 & 9
\end{bmatrix}
$$

After one-pixel padding

$$
\boxed{
\begin{bmatrix}
0&0&0&0&0\\
0&1&2&3&0\\
0&4&5&6&0\\
0&7&8&9&0\\
0&0&0&0&0
\end{bmatrix}
}
$$

The original image is surrounded by zeros.

---



# <font color='orange'> 6. Valid Padding </font>

When no padding is applied,

the convolution is called

**valid convolution**.

Conceptually,

```
Padding = 0

↓

Output Smaller
```

TensorFlow

```python
padding="valid"
```

This is the default behaviour in many convolution implementations.

---



# <font color='orange'> 7. Same Padding </font>

Modern CNNs often use

**same padding**.

The amount of padding is chosen so that

the output has

the same spatial dimensions

as the input

(for stride = 1).

Conceptually,

```
Input

224 × 224

↓

Convolution

↓

Output

224 × 224
```

TensorFlow

```python
padding="same"
```

This has become the standard choice in many deep learning architectures.

---



# <font color='orange'> 8. General Output Size Formula </font>

Suppose

* input size

$$
N,
$$

* kernel size

$$
K,
$$

* padding

$$
P,
$$

* stride

$$
S.
$$

The output size is

$$
\boxed{
\text{Output}
=
\left\lfloor
\frac{N-K+2P}{S}
\right\rfloor
+1
}
$$

where

* \(N\) = input size,
* \(K\) = kernel size,
* \(P\) = padding,
* \(S\) = stride.

This formula is one of the most important equations in CNNs.

---



# <font color='orange'> 9. Example Calculation </font>

Suppose

```
Input = 32 × 32

Kernel = 3 × 3

Padding = 1

Stride = 1
```

The output size becomes

$$
\frac{32-3+2(1)}{1}+1
=
32.
$$

Thus,

```
32 × 32

↓

Convolution

↓

32 × 32
```

The spatial dimensions are preserved.

---



# <font color='orange'> 10. Why Padding Helps </font>

Padding offers several important advantages.

* Preserves spatial dimensions.
* Allows deeper CNN architectures.
* Improves feature extraction near image boundaries.
* Prevents rapid shrinking of feature maps.

These properties explain why same padding is commonly used in modern CNNs.

---



# <font color='orange'> 11. Limitations </font>

Padding is not entirely without drawbacks.

* Artificial border pixels are introduced.
* Excessive padding may slightly increase computational cost.
* Zero padding may introduce boundary artefacts, although these are generally minor in practice.

Alternative padding methods,

such as reflection or replication padding,

are sometimes used in image generation and restoration tasks.

---

```python
layer = tf.keras.layers.Conv2D(

    filters=32,

    kernel_size=(3,3),

    padding="same"

)
```

Only the `padding` argument changes.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Padding Adds Information to the Image

False.

Padding does not create new visual information.

It simply extends the image boundaries so that edge pixels participate more fully in the convolution.

---

### Same Padding Means No Padding

False.

"Same" refers to preserving the output size, not to the absence of padding.

In most cases,

zero padding is added automatically.

---

### Padding Is Always Necessary

False.

Some architectures intentionally reduce spatial dimensions by using valid convolutions.

The choice depends on the design objectives of the network.

---

### Padding Only Works with 3×3 Kernels

False.

Padding can be applied with kernels of any size.

The required amount depends on the kernel dimensions and the desired output size.

---

| Concept | Description |
| :--- | :--- |
| Padding | Adding extra pixels around an image before convolution |
| Zero Padding | Border pixels are assigned the value zero |
| Valid Padding | No padding; output becomes smaller |
| Same Padding | Padding chosen to preserve spatial dimensions (for stride = 1) |

| Main Advantage | Preserves spatial information and enables deeper CNNs |
| Main Limitation | Introduces artificial border pixels |

> **Key Insight:** Convolution naturally reduces the spatial dimensions of an image because the kernel cannot be centred on boundary pixels without extending beyond the image. Padding overcomes this limitation by adding additional pixels around the image before convolution, allowing edge information to be incorporated into the feature extraction process. In modern CNNs, same padding is widely used because it preserves spatial dimensions, facilitates the construction of deep architectures, and improves feature learning near image boundaries.

# <font color='blue'> Stride in Convolutional Neural Networks </font>

The convolution operation extracts local features by sliding a kernel across an input image. In previous chapters, we assumed that the kernel moved one pixel at a time. Although this is the most common choice, the movement of the kernel can be adjusted to control both the computational cost and the spatial resolution of the resulting feature maps.

The number of pixels by which the kernel moves between successive applications is called the **stride**. Larger strides reduce the spatial dimensions of the output feature map, providing an efficient mechanism for downsampling while simultaneously extracting features.

This chapter introduces the concept of stride, derives the general output-size formula including stride, explains the relationship between stride and downsampling, and discusses its practical implementation in modern deep learning frameworks.

---



# <font color='orange'> 1. What is Stride? </font>

The **stride** is the number of pixels

the convolution kernel moves

after each operation.

Conceptually,

```
Stride = 1

□→□→□→□
```

The kernel moves

one pixel

at a time.

---

If

```
Stride = 2
```

the kernel moves

```
□→→□→→□
```

skipping every other pixel.

Thus,

larger strides produce fewer output values.

---



# <font color='orange'> 2. Stride = 1 </font>

Suppose

the input image is

```
7 × 7
```

and the kernel is

```
3 × 3.
```

With

```
Stride = 1
```

the kernel visits

every valid position.

Conceptually,

```
■■■....

.■■■...

..■■■..

...■■■.

....
```

This produces

the largest possible feature map.

Stride 1 preserves the maximum amount of spatial information.

---



# <font color='orange'> 3. Stride = 2 </font>

Now suppose

```
Stride = 2.
```

The kernel jumps

two pixels

after every convolution.

Conceptually,

```
■■■..

..■■■

....
```

Many intermediate positions are skipped.

The resulting feature map is therefore smaller.

This simultaneously performs

feature extraction

and

downsampling.

---



# <font color='orange'> 4. Why Use Larger Strides? </font>

Larger strides offer several advantages.

```
Larger Stride

↓

Fewer Convolutions

↓

Smaller Feature Maps

↓

Lower Computation
```

However,

this comes at the cost of

reduced spatial resolution.

Choosing an appropriate stride therefore involves balancing efficiency against information preservation.

---



# <font color='orange'> 5. Output Size Formula </font>

Suppose

* input size

$$
N,
$$

* kernel size

$$
K,
$$

* padding

$$
P,
$$

* stride

$$
S.
$$

The output size is

$$
\boxed{
\text{Output}
=
\left\lfloor
\frac{N-K+2P}{S}
\right\rfloor
+1
}
$$

This equation combines

kernel size,

padding,

and

stride,

making it one of the fundamental formulas in CNNs.

---



# <font color='orange'> 6. Numerical Example </font>

Suppose

```
Input = 32

Kernel = 3

Padding = 1

Stride = 2
```

The output becomes

$$
\begin{aligned}
\text{Output}
&=
\left\lfloor
\frac{32-3+2}{2}
\right\rfloor
+1\\
&=
\left\lfloor
15.5
\right\rfloor
+1\\
&=
15+1\\
&=
16.
\end{aligned}
$$

Thus,

```
32 × 32

↓

16 × 16
```

The spatial dimensions have been reduced by approximately one half.

---



# <font color='orange'> 7. Stride Performs Downsampling </font>

Increasing the stride naturally reduces

the spatial resolution

of the feature maps.

Conceptually,

```
224 × 224

↓

Stride 2

↓

112 × 112

↓

Stride 2

↓

56 × 56
```

Many modern CNNs intentionally use strided convolutions to reduce computational cost while retaining important features.

---



# <font color='orange'> 8. Stride vs Pooling </font>

Both stride

and

pooling

reduce the spatial dimensions of feature maps.

However,

they operate differently.

| Method | Learns Parameters? | Performs Downsampling? |
| :--- | :---: | :---: |
| Strided Convolution | ✓ | ✓ |
| Pooling | ✗ | ✓ |

A strided convolution simultaneously learns features and reduces resolution,

whereas pooling performs only the downsampling operation.

---



# <font color='orange'> 9. Choosing the Stride </font>

Common choices include

| Stride | Typical Use |
| :---: | :--- |
| 1 | Feature extraction |
| 2 | Downsampling |
| 3 or larger | Rare in modern CNNs |

Most convolutional layers use

```
Stride = 1
```

while selected layers use

```
Stride = 2
```

to reduce the spatial dimensions.

---



# <font color='orange'> 10. Practical Implementation in TensorFlow </font>

### Stride = 1

```python
import tensorflow as tf

layer = tf.keras.layers.Conv2D(

    filters=32,

    kernel_size=(3,3),

    strides=(1,1),

    padding="same"

)
```

---

### Stride = 2

```python
layer = tf.keras.layers.Conv2D(

    filters=32,

    kernel_size=(3,3),

    strides=(2,2),

    padding="same"

)
```

Changing only the `strides` argument alters the output resolution.

---



# <font color='orange'> 11. Advantages </font>

Using stride provides several practical benefits.

* Reduces computational cost.
* Produces smaller feature maps.
* Enables deeper CNN architectures.
* Performs feature extraction and downsampling simultaneously.

These properties make stride an important design parameter in modern CNNs.

---



# <font color='orange'> 12. Limitations </font>

Larger strides also have disadvantages.

* Reduced spatial resolution.
* Possible loss of fine image details.
* Excessively large strides may skip important visual features.

Consequently,

very large strides are seldom used.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Larger Strides Always Improve Performance

False.

Although larger strides reduce computation,

they also discard spatial information.

A balance must be maintained.

---

### Stride Changes the Number of Parameters

False.

The stride affects

how often

the kernel is applied,

not

the number of learnable kernel weights.

---

### Stride and Padding Are Independent

False.

Both parameters jointly determine the output dimensions through the convolution output-size formula.

---

### Strided Convolution Is Identical to Pooling

False.

Both reduce spatial dimensions,

but only strided convolution learns feature detectors.

Pooling contains no trainable parameters.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Stride | Number of pixels the kernel moves after each convolution |
| Stride = 1 | Maximum spatial resolution |
| Stride = 2 | Simultaneous feature extraction and downsampling |
| Output Size | \(\left\lfloor\frac{N-K+2P}{S}\right\rfloor+1\) |
| Main Advantage | Reduces computation while preserving learned features |
| Main Limitation | Larger strides discard more spatial information |

> **Key Insight:** The stride determines how far a convolutional kernel moves between successive applications. Increasing the stride reduces the spatial dimensions of the output feature maps, lowering computational cost while performing downsampling. Because stride controls only the movement of the kernel—not its parameters—it provides an efficient mechanism for balancing computational efficiency against spatial resolution. Together with kernel size and padding, stride determines the output dimensions of every convolutional layer in a CNN.

# <font color='blue'> Pooling Layers in Convolutional Neural Networks </font>

Convolutional layers extract increasingly sophisticated visual features from an input image. However, repeatedly applying convolutions while preserving the full spatial resolution results in large feature maps that require substantial computational resources. Furthermore, many neighbouring activations contain redundant information, making it unnecessary to retain every value.

**Pooling** is a downsampling operation that reduces the spatial dimensions of feature maps while preserving their most informative characteristics. Unlike convolution, pooling introduces **no trainable parameters**. Instead, it summarizes local neighbourhoods using simple mathematical operations such as taking the maximum or average value.

Pooling improves computational efficiency, reduces memory requirements, and introduces a degree of robustness to small translations and distortions in the input image. Consequently, pooling has become a standard component of many convolutional neural network architectures.

This chapter introduces the motivation for pooling, explains Max Pooling and Average Pooling, discusses Global Average Pooling, and demonstrates their implementation in TensorFlow and Keras.

---



# <font color='orange'> 1. Why Do We Need Pooling? </font>

Suppose a convolutional layer produces

```
224 × 224 × 64
```

feature maps.

These contain

over three million values.

Many neighbouring activations are highly similar.

Conceptually,

```
Large Feature Map

↓

Many Similar Values

↓

Redundant Information
```

Pooling compresses these feature maps

while preserving the most important information.

---



# <font color='orange'> 2. The Central Idea </font>

Pooling divides the feature map

into small regions,

called

**pooling windows**,

and replaces each region

with a single representative value.

Conceptually,

```
2 × 2 Window

8   4

2   6

↓

One Value
```

The window then moves

across the feature map,

similar to convolution.

---



# <font color='orange'> 3. Max Pooling </font>

The most common pooling method is

**Max Pooling**.

For each pooling window,

the maximum value is selected.

Example

```
8   4

2   6
```

becomes

```
8
```

Another example

```
1   5

7   3
```

becomes

```
7
```

The strongest activation is preserved,

while the remaining values are discarded.

---



# <font color='orange'> 4. Average Pooling </font>

Instead of selecting

the maximum value,

Average Pooling computes

the arithmetic mean.

Example

```
8   4

2   6
```

becomes

$$
\frac{8+4+2+6}{4}
=
5.
$$

Average Pooling preserves

the overall intensity

rather than the strongest response.

---



# <font color='orange'> 5. Pooling Window and Stride </font>

The most common configuration is

```
Window = 2 × 2

Stride = 2
```

Conceptually,

```
□□□□

□□□□

□□□□

□□□□

↓

□□

□□
```

Every

```
2 × 2
```

region becomes

one output value.

Thus,

both spatial dimensions

are reduced by approximately one half.

---



# <font color='orange'> 6. Max Pooling vs Average Pooling </font>

| Property | Max Pooling | Average Pooling |
| :--- | :---: | :---: |
| Output | Largest value | Mean value |
| Preserves strongest feature | ✓ | ✗ |
| Smooths activations | ✗ | ✓ |
| Most common in CNNs | ✓ | Less common |

Max Pooling is generally preferred for image classification because it preserves the most prominent features.

---



# <font color='orange'> 7. Why Does Pooling Help? </font>

Pooling provides several practical advantages.

```
Feature Map

↓

Smaller Feature Map

↓

Less Computation

↓

Better Efficiency
```

Pooling also introduces

a degree of robustness

to small translations.

Suppose

an edge shifts

by one pixel.

Max Pooling often produces

the same output,

making the network less sensitive

to small changes in object position.

---



# <font color='orange'> 8. Global Average Pooling </font>

Modern CNNs often replace

fully connected layers

with

**Global Average Pooling (GAP).**

Instead of pooling small windows,

GAP averages

each entire feature map.

Suppose

the final convolution produces

```
7 × 7 × 512.
```

Global Average Pooling computes

one number

for each channel,

producing

```
1 × 1 × 512.
```

This dramatically reduces

the number of parameters

and helps reduce overfitting.

Architectures such as **ResNet**, **MobileNet**, and **EfficientNet** commonly use Global Average Pooling.

---



# <font color='orange'> 9. Pooling Has No Learnable Parameters </font>

Unlike convolution,

pooling does not learn filters.

Conceptually,

```
Convolution

↓

Learn Weights
```

```
Pooling

↓

Simple Mathematical Operation
```

Pooling therefore contributes

no trainable parameters

to the network.

---



# <font color='orange'> 10. Practical Implementation in TensorFlow </font>

### Max Pooling

```python
import tensorflow as tf

layer = tf.keras.layers.MaxPooling2D(

    pool_size=(2,2),

    strides=(2,2)

)
```

---

### Average Pooling

```python
layer = tf.keras.layers.AveragePooling2D(

    pool_size=(2,2),

    strides=(2,2)

)
```

---

### Global Average Pooling

```python
layer = tf.keras.layers.GlobalAveragePooling2D()
```

TensorFlow provides dedicated layers for each pooling operation.

---



# <font color='orange'> 11. Advantages </font>

Pooling offers several important benefits.

* Reduces computational cost.
* Reduces memory usage.
* Introduces translation robustness.
* Helps reduce overfitting.
* Simplifies feature maps before classification.

These properties explain why pooling has historically been a central component of CNNs.

---



# <font color='orange'> 12. Limitations </font>

Pooling also has limitations.

* Some spatial information is inevitably lost.
* Fine image details may disappear.
* Excessive pooling may reduce feature maps too aggressively.

Because of these limitations,

some modern architectures replace pooling with

**strided convolutions**,

which perform learned downsampling.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Pooling Learns Features

False.

Pooling simply summarizes existing feature maps.

Feature learning occurs in the convolutional layers.

---

### Max Pooling Always Gives Better Results

Not necessarily.

Although Max Pooling is more common,

Average Pooling and Global Average Pooling perform better in certain architectures and applications.

---

### Pooling Changes the Number of Channels

False.

Pooling reduces only the

height

and

width

of the feature maps.

The number of channels remains unchanged.

For example,

```
64 × 64 × 32

↓

32 × 32 × 32
```

---

### Every CNN Uses Pooling

False.

Some modern architectures (e.g., certain variants of ResNet, EfficientNet, ConvNeXt, and Vision Transformers) reduce spatial dimensions using **strided convolutions** instead of—or in addition to—traditional pooling layers.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Pooling | Downsampling operation applied to feature maps |
| Max Pooling | Selects the maximum value in each pooling window |
| Average Pooling | Computes the mean value in each pooling window |
| Global Average Pooling | Averages each entire feature map into a single value |
| Trainable Parameters | None |
| Main Advantage | Reduces computation while preserving important information |
| Main Limitation | Some spatial information is discarded |

> **Key Insight:** Pooling is a parameter-free operation that reduces the spatial dimensions of feature maps while preserving their most informative characteristics. Max Pooling retains the strongest activations, Average Pooling summarizes local regions, and Global Average Pooling compresses entire feature maps into single values. By lowering computational cost, reducing memory requirements, and introducing robustness to small translations, pooling has become a fundamental component of many convolutional neural network architectures, although some modern networks increasingly employ learned downsampling through strided convolutions.

# <font color='blue'> Building a Complete Convolutional Neural Network </font>

A Convolutional Neural Network (CNN) is constructed by combining several fundamental building blocks into a single end-to-end architecture capable of learning hierarchical visual representations. Earlier chapters introduced these components individually, including convolution, activation functions, pooling, padding, and fully connected layers. This chapter demonstrates how these components are assembled into a complete neural network for image classification.

Modern CNNs progressively transform raw pixel values into increasingly abstract feature representations. Early layers detect simple visual patterns such as edges, intermediate layers identify textures and object parts, and deeper layers recognise complete objects. Finally, a classifier converts these learned features into class probabilities.

This chapter presents the standard CNN architecture, explains the role of each layer, follows the flow of data through the network, and demonstrates how to implement a complete CNN using TensorFlow and Keras.

# <font color='orange'> 1. The Overall CNN Pipeline </font>

```
Input Image

↓

Convolution

↓

Activation (ReLU)

↓

Pooling

↓

Repeat Convolutional Blocks

↓

Flatten or Global Average Pooling

↓

Fully Connected Layer

↓

Softmax

↓

Predicted Class
```

Each stage has a specific purpose.

# <font color='orange'> 2. Input Layer </font>

The network receives an image tensor.

Example

MNIST

```
28 × 28 × 1
```

CIFAR-10

```
32 × 32 × 3
```

ImageNet

# <font color='orange'> 3. First Convolutional Block </font>

# <font color='orange'> 4. Stacking Convolutional Blocks </font>

A single convolutional layer is insufficient for recognising complex objects.

multiple convolutional blocks are stacked.

Conceptually,

```
Input

↓

Conv

↓

Pool

↓

Conv

↓

Pool

↓

Conv

↓

Pool
```

Each successive layer receives increasingly abstract feature maps.

---



# <font color='orange'> 5. Hierarchical Feature Learning </font>

One of the defining strengths of CNNs is their ability to learn features hierarchically.

```
Pixels

↓

Edges

↓

Corners

↓

Textures

↓

Object Parts

↓

Complete Objects
```

For example,

a face recognition CNN may learn

```
Edges

↓

Eyes

↓

Nose

↓

Face
```

without being explicitly programmed to detect these structures.

---



# <font color='orange'> 6. Flattening </font>

Eventually,

the feature maps must be converted into a format suitable for a classifier.

Traditionally,

this is achieved using

**Flatten**.

Example

```
7 × 7 × 64

↓

Flatten

↓

3136
```

The three-dimensional tensor becomes a one-dimensional vector.

This vector serves as the input to one or more fully connected layers.

---



# <font color='orange'> 7. Global Average Pooling </font>

Many modern CNNs replace

Flatten

with

**Global Average Pooling (GAP).**

Conceptually,

```
7 × 7 × 512

↓

Global Average Pooling

↓

512
```

Advantages of GAP include

* far fewer parameters,
* reduced overfitting,
* improved generalization.

Architectures such as ResNet, EfficientNet, and MobileNet commonly employ GAP.

---



# <font color='orange'> 8. Fully Connected Classifier </font>

After feature extraction,

the network performs classification using one or more dense layers.

Conceptually,

```
Extracted Features

↓

Dense Layer

↓

Dense Layer

↓

Output Layer
```

These layers combine the learned visual features to make the final prediction.

---



# <font color='orange'> 9. Softmax Output </font>

For multi-class classification,

the final layer usually employs

Softmax.

Example

```
Cat

0.05
```

```
Dog

0.90
```

```
Horse

0.03
```

```
Bird

0.02
```

The class with the highest probability is selected as the prediction.

---



# <font color='orange'> 10. Example Architecture </font>

A simple CNN for MNIST classification might be

```
28 × 28 × 1

↓

Conv(32)

↓

ReLU

↓

MaxPool

↓

Conv(64)

↓

ReLU

↓

MaxPool

↓

Flatten

↓

Dense(128)

↓

Softmax(10)
```

Although relatively small,

this network already performs remarkably well on handwritten digit recognition.

---



# <font color='orange'> 11. Forward Propagation Through a CNN </font>

The complete forward pass is

```
Image

↓

Convolution

↓

Feature Maps

↓

ReLU

↓

Pooling

↓

Repeat

↓

Flatten / GAP

↓

Dense Layer

↓

Softmax

↓

Prediction
```

Every layer transforms the data into a progressively more informative representation.

---



# <font color='orange'> 12. Practical Implementation in TensorFlow </font>

```python
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.Conv2D(
        32,
        (3,3),
        activation="relu",
        padding="same",
        input_shape=(28,28,1)
    ),

    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Conv2D(
        64,
        (3,3),
        activation="relu",
        padding="same"
    ),

    tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(
        128,
        activation="relu"
    ),

    tf.keras.layers.Dense(
        10,
        activation="softmax"
    )

])
```

This architecture combines all of the CNN components introduced in previous chapters.

# <font color='orange'> 13. Why Do CNNs Work So Well? </font>

CNNs are effective because they combine several complementary ideas.

* Local feature extraction through convolution.
* Parameter efficiency through weight sharing.
* Progressive abstraction through multiple convolutional layers.
* Downsampling through pooling or strided convolutions.
* Nonlinear feature learning through activation functions.
* Final decision making through dense classification layers.

Together,

these components enable CNNs to learn rich visual representations from raw pixels.

---



# <font color='orange'> 14. Common Misconceptions </font>

### Every CNN Must Use Flatten

False.

Modern architectures often replace Flatten with Global Average Pooling, which significantly reduces the number of trainable parameters.

---

### CNNs Learn Entire Objects in the First Layer

False.

Early layers learn simple features such as edges and corners.

Object recognition emerges only after multiple convolutional layers.

---

### More Convolutional Layers Always Improve Performance

False.

Deeper networks generally require appropriate architectural innovations such as residual connections, normalization, and careful optimization to remain effective.

---

### The Fully Connected Layers Perform Feature Extraction

False.

Most feature extraction occurs within the convolutional layers.

The fully connected layers primarily combine the extracted features to perform classification.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Convolutional Block | Convolution + Activation + Pooling |
| Feature Hierarchy | Progression from edges to complete objects |
| Flatten | Converts feature maps into a vector |
| Global Average Pooling | Replaces Flatten with channel-wise averaging |
| Dense Layers | Perform classification using extracted features |
| Softmax | Produces class probabilities |
| Main Goal | Transform raw pixels into accurate class predictions |

> **Key Insight:** A Convolutional Neural Network is constructed by stacking convolutional blocks that progressively transform raw pixel values into increasingly abstract feature representations. Early layers detect simple visual patterns, while deeper layers combine these into meaningful object-level representations. The extracted features are then passed to a classifier, typically consisting of fully connected layers or Global Average Pooling followed by a Softmax output. This hierarchical architecture enables CNNs to achieve remarkable performance across a wide range of computer vision tasks.
