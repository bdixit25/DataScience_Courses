# <font color='blue'> Long Short-Term Memory (LSTM) Networks </font>

Vanilla Recurrent Neural Networks introduced the concept of maintaining a hidden state to process sequential data. However, during Backpropagation Through Time (BPTT), gradients propagated across many time steps often become extremely small or excessively large, making it difficult for the network to learn long-range dependencies. As a result, conventional RNNs tend to forget information from the distant past.

To address this limitation, Sepp Hochreiter and Jürgen Schmidhuber introduced the **Long Short-Term Memory (LSTM)** architecture in 1997. Rather than relying solely on a hidden state, LSTMs introduce a dedicated **cell state**, which serves as a long-term memory, together with several **gates** that regulate the flow of information into, within, and out of this memory.

These gating mechanisms enable LSTMs to preserve important information over long sequences while selectively forgetting irrelevant information, making them one of the most successful sequence modelling architectures prior to the widespread adoption of Transformers.

---



# <font color='orange'> 1. Motivation </font>

Suppose we wish to process

```
The Nobel Prize in Physics was awarded to ...

↓

Many words later ...

↓

... Higgs.
```

The prediction depends on information

encountered much earlier.

A vanilla RNN often loses

this information.

LSTMs were designed to

remember important information

for long periods of time.

---



# <font color='orange'> 2. The Central Idea </font>

Instead of maintaining only

a hidden state,

an LSTM maintains

two states.

```
Hidden State

hₜ
```

and

```
Cell State

Cₜ
```

The hidden state

contains the current output,

while the cell state

stores long-term memory.

---



# <font color='orange'> 3. Overall LSTM Cell </font>

An LSTM cell receives

```
Current Input

xₜ
```

```
Previous Hidden State

hₜ₋₁
```

```
Previous Cell State

Cₜ₋₁
```

and produces

```
New Hidden State

hₜ
```

```
New Cell State

Cₜ
```

The information flow is controlled by

three gates.

---



# <font color='orange'> 4. Forget Gate </font>

The first gate decides

what information

should be discarded.

It computes

$$
\boxed{
f_t
=
\sigma
\left(
W_f[h_{t-1},x_t]
+
b_f
\right)
}
$$

where

* \(\sigma\) is the sigmoid function,
* \(f_t\) contains values between 0 and 1.

Interpretation

```
0

↓

Forget Completely
```

```
1

↓

Keep Completely
```

---



# <font color='orange'> 5. Input Gate </font>

Next,

the LSTM determines

what new information

should be stored.

The input gate computes

$$
\boxed{
i_t
=
\sigma
\left(
W_i[h_{t-1},x_t]
+
b_i
\right)
}
$$

A candidate memory

is also computed

$$
\boxed{
\tilde{C}_t
=
\tanh
\left(
W_C[h_{t-1},x_t]
+
b_C
\right)
}
$$

The gate decides

how much of this candidate

should be added

to the memory.

---



# <font color='orange'> 6. Updating the Cell State </font>

The new memory combines

the previous memory

with

the candidate memory.

$$
\boxed{
C_t
=
f_t
\odot
C_{t-1}
+
i_t
\odot
\tilde{C}_t
}
$$

where

\(\odot\)

denotes element-wise multiplication.

Conceptually,

```
Old Memory

↓

Forget Some

↓

Add New Information

↓

Updated Memory
```

---



# <font color='orange'> 7. Output Gate </font>

Finally,

the LSTM determines

what information

should become

the hidden state.

The output gate computes

$$
\boxed{
o_t
=
\sigma
\left(
W_o[h_{t-1},x_t]
+
b_o
\right)
}
$$

The hidden state becomes

$$
\boxed{
h_t
=
o_t
\odot
\tanh(C_t)
}
$$

Only selected parts

of the memory

become visible

to the next layer

or time step.

---



# <font color='orange'> 8. Information Flow </font>

An LSTM performs

the following sequence.

```
Previous Memory

↓

Forget Gate

↓

Input Gate

↓

Updated Memory

↓

Output Gate

↓

Hidden State
```

Each gate

controls

one aspect

of memory management.

---



# <font color='orange'> 9. Why LSTMs Reduce Vanishing Gradients </font>

The cell state

provides

a nearly linear path

through time.

Conceptually,

```
C₁

↓

C₂

↓

C₃

↓

C₄
```

Rather than repeatedly transforming

the memory,

the network can preserve it

over many time steps,

greatly improving gradient flow.

This makes learning long-term dependencies much easier than in a vanilla RNN, although gradients can still degrade in extremely long sequences.

---



# <font color='orange'> 10. Advantages </font>

LSTMs introduced several important improvements.

* Learn long-term dependencies.
* Reduce vanishing gradients.
* Process variable-length sequences.
* Learn when to remember and when to forget.
* Widely applicable across sequence-learning tasks.

For many years,

they represented

the state of the art

in sequence modelling.

---



# <font color='orange'> 11. Limitations </font>

Despite their success,

LSTMs also have limitations.

* More parameters than vanilla RNNs.
* Slower training.
* Sequential computation limits parallelism.
* Training on extremely long sequences remains challenging.
* Transformers now outperform LSTMs on many large-scale NLP tasks.

---



# <font color='orange'> 12. TensorFlow Implementation </font>

TensorFlow provides an LSTM layer.

```python
import tensorflow as tf

model = tf.keras.Sequential([

    tf.keras.layers.LSTM(

        128,

        input_shape=(None, 64)

    ),

    tf.keras.layers.Dense(

        10,

        activation="softmax"

    )

])
```

Here,

* `None` represents a variable sequence length,
* `64` is the number of features per time step,
* `128` is the number of LSTM units.

# <font color='orange'> 14. Common Misconceptions </font>

### LSTMs Never Forget Information

False.

The forget gate explicitly learns which information should be discarded. Forgetting irrelevant information is an essential part of effective sequence modelling.

---

### The Cell State and Hidden State Are Identical

False.

The **cell state** stores long-term memory, whereas the **hidden state** represents the current output passed to the next time step and subsequent layers.

---

### LSTMs Completely Solve the Vanishing Gradient Problem

False.

LSTMs substantially mitigate the problem through their gated memory mechanism, but they do not eliminate all optimization challenges, especially for extremely long sequences.

---

### Every Gate Makes a Binary Decision

False.

The gates produce continuous values between 0 and 1, allowing information to be partially remembered or partially forgotten.

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| LSTM | Recurrent neural network with gated memory |
| Cell State (\(C_t\)) | Long-term memory pathway |
| Hidden State (\(h_t\)) | Current output of the network |
| Forget Gate | Decides what information to discard |
| Input Gate | Controls what new information enters memory |
| Candidate Cell State | Proposed new memory content |
| Output Gate | Determines what information becomes the hidden state |
| Main Advantage | Learns long-range dependencies much more effectively than vanilla RNNs |
| Main Limitation | Sequential computation limits scalability compared with Transformers |

> **Key Insight:** Long Short-Term Memory networks overcome the principal weakness of vanilla RNNs by introducing a dedicated memory cell together with learnable gating mechanisms that regulate information flow. The forget, input, and output gates enable the network to selectively preserve useful information while discarding irrelevant details, substantially improving the ability to learn long-range dependencies. This innovation made LSTMs the dominant architecture for sequence modelling for nearly two decades and laid important conceptual foundations for modern attention-based models.
