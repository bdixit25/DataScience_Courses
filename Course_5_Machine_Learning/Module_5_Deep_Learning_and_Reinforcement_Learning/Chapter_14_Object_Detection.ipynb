# <font color='blue'> Introduction to Object Detection </font>

Image classification determines the primary object present in an image by assigning a single class label. While this is sufficient for many applications, real-world scenes often contain multiple objects at different locations and scales. Understanding such scenes requires not only identifying the objects present but also determining their precise locations within the image.

**Object Detection** addresses this problem by simultaneously performing **classification** and **localization**. For each detected object, the model predicts both its class (e.g., car, person, bicycle) and a bounding box specifying its position within the image.

Object detection forms the foundation of numerous modern applications, including autonomous driving, robotics, surveillance, medical imaging, satellite image analysis, industrial inspection, and many scientific imaging tasks.

This chapter introduces the fundamentals of object detection, explains bounding boxes and evaluation metrics, and prepares the reader for modern object detection architectures such as R-CNN, Faster R-CNN, and YOLO.

# <font color='orange'> 3. Bounding Boxes </font>

Most object detectors represent locations

using

rectangular

**bounding boxes**.

Conceptually,

```
+-------------+

|             |

|     CAT     |

|             |

+-------------+
```

The rectangle encloses

the detected object.

---



# <font color='orange'> 4. Bounding Box Representation </font>

A bounding box is commonly represented by

four numbers.

One common parameterization is

$$
\boxed{
(x_{\min},\; y_{\min},\; x_{\max},\; y_{\max})
}
$$

where

* \(x_{\min}\) = left boundary,
* \(y_{\min}\) = upper boundary,
* \(x_{\max}\) = right boundary,
* \(y_{\max}\) = lower boundary.

Another common representation is

$$
\boxed{
(x_c,\; y_c,\; w,\; h)
}
$$

where

* \(x_c,y_c\) denote the box centre,
* \(w\) is the width,
* \(h\) is the height.

Different object detection architectures use different parameterizations internally.

# <font color='orange'> 7. Intersection over Union (IoU) </font>

How do we determine

whether a predicted box

matches

the true object?

Object detection uses

**Intersection over Union (IoU).**

Conceptually,

```
Ground Truth Box

+

Predicted Box

↓

Overlap
```

The IoU is defined as

$$
\boxed{
\mathrm{IoU}
=
\frac{\text{Area of Intersection}}
{\text{Area of Union}}
}
$$

Its value lies between

0

and

1.

---



# <font color='orange'> 8. Interpreting IoU </font>

| IoU | Interpretation |
| :---: | :--- |
| 0 | No overlap |
| 0.5 | Moderate overlap |
| 0.75 | Very good overlap |
| 1.0 | Perfect prediction |

Many benchmarks

consider

```
IoU ≥ 0.5
```

to represent a successful detection, although modern evaluations often report performance across multiple IoU thresholds.

---



# <font color='orange'> 9. Precision and Recall </font>

Object detection must balance

false positives

and

false negatives.

Precision measures

how many predicted objects

are correct.

$$
\boxed{
\mathrm{Precision}
=
\frac{TP}{TP+FP}
}
$$

Recall measures

how many true objects

are detected.

$$
\boxed{
\mathrm{Recall}
=
\frac{TP}{TP+FN}
}
$$

Both are important.

---



# <font color='orange'> 10. Mean Average Precision (mAP) </font>

The standard evaluation metric for object detection is

**Mean Average Precision (mAP).**

For each object class,

we compute

Average Precision (AP),

which summarizes the precision–recall curve.

The final metric is

the mean over all classes.

$$
\boxed{
\mathrm{mAP}
=
\frac{1}{N}
\sum_{i=1}^{N}
AP_i
}
$$

where

\(N\)

is the number of object classes.

Modern benchmarks often report values such as **mAP@0.5** or **mAP@[0.5:0.95]**, indicating the IoU thresholds used during evaluation.

# <font color='orange'> 13. Common Misconceptions </font>

### Object Detection and Image Classification Are the Same

False.

Image classification predicts

**what** is present.

Object detection predicts

**what**

and

**where**.

---

### Bounding Boxes Are Always Perfect

False.

Bounding boxes approximate object locations.

Their accuracy is evaluated using IoU.

---

### Higher Confidence Always Means Better Localization

False.

Confidence estimates how likely the predicted class is correct.

Localization quality is measured separately using metrics such as IoU.

---

### Object Detection Finds Only One Object

False.

Modern detectors routinely identify many objects within a single image.

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Object Detection | Simultaneous classification and localization |
| Localization | Determining an object's position |
| Bounding Box | Rectangle enclosing a detected object |
| Confidence Score | Estimated probability that a detection is correct |
| IoU | Measures overlap between predicted and true bounding boxes |
| Precision | Fraction of predicted detections that are correct |
| Recall | Fraction of true objects successfully detected |
| mAP | Standard evaluation metric for object detection |

> **Key Insight:** Object detection extends image classification by predicting both the identity and location of every object in an image. Each detection consists of a class label, a confidence score, and a bounding box. Performance is evaluated using localization metrics such as Intersection over Union (IoU) together with precision, recall, and mean Average Precision (mAP). These concepts form the foundation of modern object detection architectures such as R-CNN, Faster R-CNN, and YOLO.
