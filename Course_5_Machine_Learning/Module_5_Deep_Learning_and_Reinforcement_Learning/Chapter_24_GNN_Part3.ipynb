# <font color='blue'> Graph Transformers: Extending Transformer Attention to Graphs </font>

Graph Attention Networks introduced learnable attention coefficients between neighbouring nodes, significantly improving the flexibility of graph representation learning. However, standard GATs restrict attention to nodes connected by graph edges. While this local attention is effective for many tasks, some applications require modelling long-range interactions between distant nodes.

**Graph Transformers** extend the Transformer architecture to graph-structured data. Instead of relying solely on local message passing, Graph Transformers use self-attention mechanisms that allow nodes to exchange information across much larger portions of the graph while incorporating structural information such as graph connectivity, shortest-path distances, edge features, or positional encodings.

By combining the expressive power of Transformers with graph-specific representations, Graph Transformers have become one of the leading architectures for complex graph learning tasks.

# <font color='blue'> Introduction to Geometric Deep Learning: Learning from Symmetry and Geometry </font>

Classical deep learning architectures were designed for specific types of data. Convolutional Neural Networks exploit the grid structure of images, Recurrent Neural Networks and Transformers process sequences, and Graph Neural Networks operate on graphs. These architectures achieve remarkable performance because they are tailored to the underlying structure of the data.

However, many scientific datasets possess a richer geometric structure. Molecular systems, physical simulations, particle detectors, 3D point clouds, and robotic environments are governed by geometric relationships and physical symmetries. Effective learning in these domains requires models that respect these symmetries rather than ignoring them.

**Geometric Deep Learning (GDL)** is a unifying framework that studies deep learning on non-Euclidean domains while explicitly incorporating geometric structure and symmetry into neural network design.

Rather than viewing CNNs, GNNs, and Transformers as unrelated architectures, Geometric Deep Learning shows that they are all special cases of a common mathematical framework.

# <font color='blue'> Invariance and Equivariance: The Mathematical Language of Symmetry </font>

Many successful neural network architectures derive their effectiveness from exploiting symmetries present in the data. Images are naturally translated across a plane, molecules can be rotated in three-dimensional space, graphs may be represented using different node orderings, and physical systems are often governed by continuous symmetries. A neural network that explicitly respects these symmetries typically requires less training data, generalizes more effectively, and produces physically meaningful predictions.

Two central mathematical concepts describe how neural networks should respond to transformations: **invariance** and **equivariance**. Although closely related, they describe fundamentally different behaviours. Understanding this distinction is essential for modern computer vision, graph learning, robotics, molecular modelling, and scientific machine learning.

---



# <font color='orange'> 1. Motivation </font>

Suppose

we classify

an image.

```
Cat
```

Now rotate

the image

by

90°.

```
Rotated Cat
```

Should the prediction

change?

No.

It is still

a cat.

The model

**invariant**

to this transformation.

---

Now consider

```
Move Object Right

↓

Detected Box

also moves right
```

The prediction

should change

in exactly

the same way

as the input.

This behaviour

is called

**equivariance**.

}

* particle tracking,

# <font color='blue'> Group Theory for Machine Learning: The Mathematics of Symmetry </font>

Symmetry plays a central role in mathematics, physics, and modern machine learning. Physical laws remain unchanged under transformations such as translations, rotations, and Lorentz transformations. Likewise, many machine learning problems involve data whose essential properties should remain unchanged—or transform predictably—under similar transformations.

The mathematical language used to describe symmetry is **group theory**. Geometric Deep Learning builds upon this framework by designing neural network architectures that are invariant or equivariant with respect to specific transformation groups.

This chapter introduces the fundamental concepts of group theory needed to understand modern equivariant neural networks.

$$

### Associativity

The grouping

$$

> **Key Insight:** Group theory provides the mathematical language of symmetry. By identifying the transformation group underlying a learning problem and designing neural networks that are invariant or equivariant with respect to that group, Geometric Deep Learning produces models that are more data-efficient, robust, and consistent with the underlying structure of the physical world.

# <font color='blue'> SE(3)-Equivariant Neural Networks: Learning in Three-Dimensional Space </font>

Many scientific datasets consist of objects embedded in three-dimensional space. Molecules, proteins, robotic systems, astronomical observations, and particle detector events are naturally described by spatial coordinates. Conventional Graph Neural Networks can model relationships between entities but generally do not exploit the continuous geometric symmetries of three-dimensional space.

**SE(3)-Equivariant Neural Networks** explicitly incorporate the symmetries of the **Special Euclidean Group**, consisting of rotations and translations in three dimensions. Rather than learning these symmetries from data, the architecture guarantees that its internal representations transform consistently under rotations and translations.

By embedding geometric symmetry directly into the model, SE(3)-equivariant networks achieve improved data efficiency, stronger generalization, and physically meaningful predictions for many scientific applications.

# <font color='orange'> 2. The Special Euclidean Group </font>

The group

SE(3)

contains

all rigid-body motions

in three dimensions.

Each transformation consists of

* a rotation,
* a translation.

Mathematically,

a transformation is written as

$$
\boxed{
x'
=
Rx+t
}
$$

where

* \(R\) is a rotation matrix,
* \(t\) is a translation vector.

Distances

and

angles

remain unchanged.

---



# <font color='orange'> 3. Why Ordinary GNNs Are Insufficient </font>

Suppose

each node

contains

3D coordinates.

```
(x,y,z)
```

A standard GNN

treats these

as ordinary numbers.

It does not know

that rotating

every coordinate

should produce

a corresponding rotation

of geometric quantities.

Consequently,

the network must learn

rotation symmetry

from data,

which is inefficient.

---



# <font color='orange'> 4. Equivariance </font>

Suppose

we rotate

the input graph.

An SE(3)-equivariant network satisfies

$$
\boxed{
f(Rx+t)
=
Rf(x)+t
}
$$

This means

the output

transforms

exactly

as expected.

The network

respects

the underlying geometry.

# <font color='orange'> 14. Common Misconceptions </font>

### Rotation Invariance and Rotation Equivariance Are the Same

False.

Rotation invariance means the output does not change after rotation.

Rotation equivariance means the output rotates consistently with the input.

---

### Ordinary GNNs Automatically Respect 3D Geometry

False.

Standard GNNs can process coordinates but do not inherently preserve rotational or translational symmetry.

---

### Every Geometric Network Uses Tensor Algebra

False.

Architectures such as EGNN achieve equivariance without explicitly using spherical harmonics or tensor fields.

---

### SE(3)-Equivariant Models Are Useful Only in Chemistry

False.

They are increasingly important in robotics, astronomy, climate science, medical imaging, and particle physics.
