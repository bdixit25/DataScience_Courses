# <font color='blue'> Graph Convolutional Networks (GCNs): Extending Convolutions to Graphs </font>

Convolutional Neural Networks perform feature extraction by applying the same local filter across every spatial location in an image. This operation relies on the regular grid structure of images, where each pixel has a fixed neighbourhood. Graphs, however, are irregular structures in which different nodes may have different numbers of neighbours. Consequently, the classical convolution operation cannot be applied directly.

A **Graph Convolutional Network (GCN)** generalizes convolution to graph-structured data by allowing each node to aggregate information from its neighbouring nodes. Rather than sliding a convolutional kernel across a regular grid, a GCN updates each node representation using information from its graph neighbourhood.

Introduced by Kipf and Welling (2017), Graph Convolutional Networks became one of the most influential architectures in graph machine learning and remain the foundation for many later Graph Neural Networks.

---



# <font color='orange'> 1. Motivation </font>

Suppose

we wish

to classify

one detector hit.

```
Hit B
```

The hit itself

contains

its own features,

but

nearby hits

provide additional information.

Instead of learning

from

only

one node,

we would like

each node

to learn

from

its neighbours.

This is

the graph analogue

of convolution.

---



# <font color='orange'> 2. The Central Idea </font>

A Graph Convolutional Layer

updates

every node.

Conceptually,

```
Neighbour Features

↓

Aggregate

↓

Linear Transformation

↓

Activation

↓

Updated Node Features
```

Every node

performs

this computation

simultaneously.

---



# <font color='orange'> 3. Self-Loops </font>

Suppose

we aggregate

only

neighbouring nodes.

Then

the node's own features

would disappear.

To avoid this,

we add

a **self-loop**

to every node.

```
A

↓

A
```

Mathematically,

the adjacency matrix

becomes

$$
\boxed{
\tilde A
=
A
+
I
}
$$

where

\(I\)

is the identity matrix.

Now

every node

is connected

to itself.

---



# <font color='orange'> 4. Degree Matrix </font>

Using

the modified adjacency matrix,

we compute

the corresponding degree matrix

$$
\boxed{
\tilde D
}
$$

whose diagonal entries

contain

the degrees

after

adding self-loops.

---



# <font color='orange'> 5. Why Normalization Is Needed </font>

Suppose

one node

has

3 neighbours,

while another

has

300 neighbours.

Simply summing

their neighbour features

would produce

very different scales.

To prevent

high-degree nodes

from dominating,

we normalize

the adjacency matrix.

The normalized adjacency matrix is

$$
\boxed{
\hat A
=
\tilde D^{-1/2}
\tilde A
\tilde D^{-1/2}
}
$$

This symmetric normalization

balances

the contributions

from nodes

with different degrees.

---



# <font color='orange'> 6. The GCN Layer Equation </font>

The update rule

for one GCN layer

is

$$
\boxed{
H^{(l+1)}
=
\sigma
\left(
\hat A
H^{(l)}
W^{(l)}
\right)
}
$$

where

* \(H^{(l)}\) is the node feature matrix at layer \(l\),
* \(W^{(l)}\) is the learnable weight matrix,
* \(\hat A\) is the normalized adjacency matrix,
* \(\sigma\) is an activation function such as ReLU.

This equation

is the defining equation

of a Graph Convolutional Network.

---



# <font color='orange'> 7. Understanding the Equation </font>

The computation occurs

in three stages.

### Step 1

Neighbour aggregation

$$
\hat A
H^{(l)}
$$

Every node

collects

information

from itself

and

its neighbours.

---

### Step 2

Linear transformation

$$
W^{(l)}
$$

Learns

which features

are important.

---

### Step 3

Activation

$$
\sigma
$$

Introduces

nonlinearity,

allowing

complex representations

to be learned.

---



# <font color='orange'> 8. Multiple GCN Layers </font>

One layer

collects

information

from

one-hop neighbours.

```
A --- B --- C
```

After

one layer,

B knows

A

and

C.

After

two layers,

A learns

about C

through B.

Thus,

capture

larger graph neighbourhoods.

---



# <font color='orange'> 9. Worked Example </font>

Consider

```
A ----- B ----- C
```

Suppose

each node

has

one feature.

```
A = 2

B = 5

C = 8
```

After graph aggregation,

each node

contains

a weighted combination

of

its own value

and

its neighbours' values.

The learnable weight matrix

then transforms

these aggregated features,

producing

new node representations.

Although real GCNs operate on vectors and normalized adjacency matrices, the underlying idea is simply to mix information from connected nodes before applying a learned transformation.

### GCNs Apply Ordinary CNN Filters

False.

GCNs generalize convolution through neighbourhood aggregation rather than sliding kernels.

---

### Every Neighbour Contributes Equally

Not exactly.

The normalized adjacency matrix weights contributions according to graph connectivity, but the weights are not learned individually. Learned neighbour importance is introduced later by Graph Attention Networks.

### More GCN Layers Always Improve Performance

False.

Too many layers can cause oversmoothing, where node representations become nearly indistinguishable.

False.

GCNs naturally support graphs with varying numbers of nodes and edges.

# <font color='blue'> GraphSAGE: Inductive Representation Learning on Large Graphs </font>

Graph Convolutional Networks demonstrated that graph convolution provides an effective framework for learning node representations. However, standard GCNs assume that the complete graph structure is available during training and inference, making them less suitable for extremely large or evolving graphs. Furthermore, GCNs are primarily **transductive**, meaning they learn representations for nodes that are already present in the training graph.

**GraphSAGE (Graph Sample and Aggregate)** addresses these limitations by introducing neighbourhood sampling and learnable aggregation functions. Rather than aggregating information from every neighbour, GraphSAGE samples a fixed-size subset of neighbours, greatly improving scalability. More importantly, GraphSAGE learns an aggregation function that can be applied to entirely new nodes or graphs, enabling **inductive learning**.

These innovations have made GraphSAGE one of the most widely used GNN architectures in industrial recommendation systems, social networks, and scientific applications.

# <font color='blue'> Graph Attention Networks (GATs): Learning Which Neighbours Matter </font>

Graph Convolutional Networks aggregate information from neighbouring nodes using a normalized adjacency matrix, while GraphSAGE introduces neighbour sampling and flexible aggregation functions. However, these methods generally do not learn the relative importance of individual neighbours. In many real-world graphs, some neighbours provide highly informative signals, whereas others contribute little useful information.

A **Graph Attention Network (GAT)** addresses this limitation by introducing the attention mechanism into graph learning. Rather than assigning fixed weights based solely on graph connectivity, a GAT learns **attention coefficients** that determine how strongly each neighbouring node should influence the updated node representation.

Introduced by Veličković et al. (2018), Graph Attention Networks combine the flexibility of attention mechanisms with message passing, allowing the network to learn which graph relationships are most important for a given task.

---



# <font color='orange'> 1. Motivation </font>

Suppose

a node

has

five neighbours.

```
Neighbour 1

Neighbour 2

Neighbour 3

Neighbour 4

Neighbour 5
```

Should they all contribute

equally?

Probably not.

Some neighbours

contain much more

useful information.

A GAT

learns

these importance weights

automatically.

---



# <font color='orange'> 2. The Central Idea </font>

Instead of

simple aggregation,

each edge

receives

an attention weight.

Conceptually,

```
Neighbour Features

↓

Attention Scores

↓

Weighted Aggregation

↓

Updated Node
```

The network decides

which neighbours

deserve

more attention.

---



# <font color='orange'> 3. Computing Attention Scores </font>

Suppose

node

\(i\)

has neighbour

\(j\).

The network first computes

an unnormalized attention score

$$
\boxed{
e_{ij}
=
a
\left(
Wh_i,
Wh_j
\right)
}
$$

where

* \(W\) is a learnable linear transformation,
* \(a(\cdot)\) is a small neural network (often a single feed-forward layer),
* \(h_i\) and \(h_j\) are the node features.

This score measures

how relevant

node

\(j\)

is

to node

\(i\).

---



# <font color='orange'> 4. Softmax Normalization </font>

The attention scores

must become

probabilities.

This is achieved

using

Softmax.

$$
\boxed{
\alpha_{ij}
=
\frac{
\exp(e_{ij})
}
{
\sum_{k\in\mathcal N(i)}
\exp(e_{ik})
}
}
$$

The coefficients satisfy

$$
\sum_j
\alpha_{ij}
=
1.
$$

Thus,

each neighbour

receives

a learned importance weight.

---



# <font color='orange'> 5. Node Update </font>

The updated node representation

is computed

using

a weighted sum

of neighbour features.

$$
\boxed{
h_i'
=
\sigma
\left(
\sum_{j\in\mathcal N(i)}
\alpha_{ij}
Wh_j
\right)
}
$$

Neighbours

with larger attention weights

contribute more strongly.

---



# <font color='orange'> 6. Multi-Head Attention </font>

Just as in

Transformers,

GATs often use

multiple attention heads.

Conceptually,

```
Attention Head 1

↓

Attention Head 2

↓

Attention Head 3

↓

Combine Outputs
```

Each head

may learn

different aspects

of the graph.

For hidden layers, the outputs are often concatenated, while for the final layer they are commonly averaged.

---



# <font color='orange'> 7. Information Flow </font>

Each node performs

```
Neighbour Features

↓

Linear Transformation

↓

Attention Scores

↓

Softmax

↓

Weighted Aggregation

↓

Activation

↓

Updated Node
```

This replaces

# <font color='blue'> Graph Transformers: Extending Transformer Attention to Graphs </font>

Graph Attention Networks introduced learnable attention coefficients between neighbouring nodes, significantly improving the flexibility of graph representation learning. However, standard GATs restrict attention to nodes connected by graph edges. While this local attention is effective for many tasks, some applications require modelling long-range interactions between distant nodes.

**Graph Transformers** extend the Transformer architecture to graph-structured data. Instead of relying solely on local message passing, Graph Transformers use self-attention mechanisms that allow nodes to exchange information across much larger portions of the graph while incorporating structural information such as graph connectivity, shortest-path distances, edge features, or positional encodings.

By combining the expressive power of Transformers with graph-specific representations, Graph Transformers have become one of the leading architectures for complex graph learning tasks.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Graph Transformers Ignore Graph Structure

False.

Graph structure is incorporated through positional encodings, edge information, masking strategies, or structural biases.

---

### Every Node Always Attends to Every Other Node

Not necessarily.

Some architectures use full attention, while others restrict or sparsify attention to improve efficiency.

---

### Graph Transformers Replace All GNNs

False.

Graph Convolutional Networks, GraphSAGE, and Graph Attention Networks remain competitive for many tasks, especially when computational efficiency is important.

---

### Graph Transformers Do Not Use Message Passing

Not entirely.

Although attention replaces traditional neighbourhood aggregation, many Graph Transformer architectures can still be interpreted within the broader message-passing framework.

---

# <font color='purple'> 14. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Graph Transformer | Transformer architecture adapted for graphs |
| Self-Attention | Learns interactions between graph nodes |
| Structural Encoding | Injects graph topology into attention |
| Global Attention | Captures long-range dependencies |
| Multi-Head Attention | Learns multiple interaction patterns |
| Main Advantage | Models complex graph relationships beyond local neighbourhoods |
| Main Limitation | Higher computational cost, especially on large graphs |

> **Key Insight:** Graph Transformers extend the Transformer architecture to graph-structured data by combining self-attention with graph-specific structural information. Unlike traditional GNNs, which primarily aggregate information from neighbouring nodes, Graph Transformers can model long-range interactions directly, making them highly expressive for complex relational learning tasks in domains such as molecular modelling, recommendation systems, and particle physics.
