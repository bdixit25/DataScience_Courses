# <font color='blue'> Normalizing Flows: Exact Likelihood-Based Generative Models </font>

Most generative models transform simple latent variables into realistic data. However, many models either do not provide an explicit probability density (GANs) or optimize only an approximation to the true likelihood (VAEs).

**Normalizing Flows** solve this problem by constructing a sequence of **invertible transformations** that gradually reshape a simple probability distribution into the target data distribution. Because every transformation is invertible, probability densities can be computed exactly using the **change-of-variables theorem**.

This combination of exact likelihood evaluation and efficient sampling makes Normalizing Flows particularly attractive for scientific machine learning, density estimation, anomaly detection, and particle physics.

# <font color='orange'> 2. Base Distribution </font>

Normalizing Flows

start

with

a simple probability distribution.

Usually,

$$
\boxed{
z
\sim
\mathcal N(0,I)
}
$$

Sampling

from

this distribution

is easy.

The challenge

is transforming it

into

the complex data distribution.

---



# <font color='orange'> 3. Invertible Transformations </font>

Suppose

we apply

a function

$$
\boxed{
x=f(z)
}
$$

The function

must satisfy

two conditions.

* It is invertible.
* Its Jacobian determinant is efficiently computable.

This allows

both

sampling

and

likelihood evaluation.

---



# <font color='orange'> 4. The Change-of-Variables Formula </font>

The central equation

of Normalizing Flows

is

$$
\boxed{
p_X(x)
=
p_Z(z)
\left|
\det
\left(
\frac{\partial z}{\partial x}
\right)
\right|
}
$$

where

* \(z=f^{-1}(x)\),
* the determinant accounts for how volumes change under the transformation.

This equation

makes

exact likelihood computation

possible.

---



# <font color='orange'> 5. Multiple Flow Layers </font>

One transformation

is rarely sufficient.

Instead,

we compose

many invertible functions.

$$
\boxed{
z
\rightarrow
f_1
\rightarrow
f_2
\rightarrow
\cdots
\rightarrow
f_K
\rightarrow
x
}
$$

Each layer

gradually reshapes

the distribution.

---

```
Observed Data

↓

Inverse Flow

↓

Gaussian
```

Using

the change-of-variables theorem,

its exact likelihood

can be computed.

# <font color='orange'> 11. Advantages </font>

Normalizing Flows provide

* exact likelihoods,
* exact sampling,
* invertible mappings,
* stable optimization,
* probabilistic interpretation.

---



# <font color='orange'> 12. Limitations </font>

Challenges include

* architectural constraints,
* computationally expensive Jacobians,
* invertibility requirements,
* reduced architectural flexibility compared with GANs.

---



# <font color='orange'> 13. Common Misconceptions </font>

### Normalizing Flows Approximate the Likelihood

False.

They compute the likelihood exactly (within floating-point precision), provided the transformations are invertible and the Jacobian determinant is computed exactly.

---

### Every Neural Network Can Be Used as a Flow

False.

Flow layers must be invertible and have tractable Jacobian determinants.

---

### Flows Are Only Used for Images

False.

They are widely applied to density estimation, scientific inference, molecular modelling, and particle physics.

---

### Flows Always Produce Better Images Than GANs

False.

Historically, GANs often generated sharper images, while flows excelled in probabilistic modelling and exact likelihood estimation.

| Exact Likelihood | One of the defining strengths of flows |

> **Key Insight:** Normalizing Flows transform a simple probability distribution into a complex data distribution using a sequence of invertible neural network transformations. By exploiting the change-of-variables theorem, they provide both exact likelihood evaluation and efficient sampling, making them uniquely valuable for probabilistic modelling and scientific machine learning.
