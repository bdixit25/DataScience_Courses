# <font color='blue'> Point Cloud Neural Networks: Learning Directly from Sets of 3D Points </font>

Many real-world datasets are naturally represented as collections of points embedded in two- or three-dimensional space. Unlike images, these points do not lie on a regular grid, and unlike graphs, explicit connections between points may not be known in advance. Examples include LiDAR scans, molecular structures, astronomical catalogues, medical imaging, and particle detector hits.

A **point cloud** is simply an unordered set of points, where each point may possess geometric coordinates and additional attributes such as colour, intensity, energy, or time.

Point Cloud Neural Networks are deep learning architectures designed to operate directly on these unordered sets while respecting their geometric properties. Rather than converting point clouds into images or manually constructing graphs, these models learn directly from the raw spatial data.

# <font color='blue'> PointNet: Deep Learning Directly on Point Clouds </font>

Traditional deep learning architectures operate on structured data such as images or sequences. Point clouds, however, are unordered sets of points with no regular spatial arrangement. A successful neural network for point clouds must therefore satisfy a crucial requirement: its output must be independent of the ordering of the input points.

**PointNet**, introduced by Qi et al. (2017), was the first deep neural network capable of learning directly from raw point clouds without converting them into voxel grids, meshes, or graphs. Its key innovation was the combination of shared feature extraction with a symmetric aggregation function, guaranteeing permutation invariance.

PointNet remains one of the foundational architectures in geometric deep learning and inspired many later point-cloud models.

# <font color='blue'> PointNet++: Hierarchical Learning of Local Geometry </font>

PointNet demonstrated that deep neural networks could learn directly from unordered point clouds by combining shared feature extraction with a permutation-invariant pooling operation. However, because each point is processed independently before global aggregation, PointNet has only a limited ability to capture local geometric structure.

**PointNet++** extends PointNet by introducing a hierarchical architecture that learns features from progressively larger neighbourhoods of points. Rather than treating the entire point cloud as a single set, PointNet++ recursively partitions the cloud into overlapping local regions and applies PointNet to each region.

This hierarchical strategy is conceptually similar to Convolutional Neural Networks, where small local filters are combined to build increasingly abstract representations.

# <font color='blue'> Dynamic Graph CNN (DGCNN): Learning Dynamic Relationships Between Points </font>

PointNet++ introduced hierarchical neighbourhood learning by grouping nearby points according to their spatial coordinates. However, these neighbourhoods remain largely determined by the original geometry of the point cloud. As learning progresses, the feature representation of each point evolves, suggesting that the notion of neighbourhood should evolve as well.

**Dynamic Graph CNN (DGCNN)** addresses this limitation by dynamically constructing a graph in the learned feature space at every layer. Rather than relying on a fixed geometric neighbourhood, DGCNN repeatedly recomputes nearest neighbours based on the current feature embeddings. This enables the network to discover increasingly meaningful relationships as learning proceeds.

The core building block of DGCNN is **EdgeConv**, a graph-based operation that combines local feature differences with message passing to learn expressive geometric representations.

# <font color='orange'> 4. EdgeConv </font>

The key innovation

is

**EdgeConv**.

Instead of learning

only

from node features,

EdgeConv learns

from

relationships

between neighbouring points.

For a point

\(i\)

and one of its neighbours

\(j\),

an edge feature is computed as

$$
\boxed{
e_{ij}
=
h_\Theta
\left(
x_i,\,
x_j-x_i
\right)
}
$$

where

* \(x_i\) is the feature vector of point \(i\),
* \(x_j-x_i\) represents the relative feature difference,
* \(h_\Theta\) is a learnable neural network.

The relative difference captures how neighbouring points differ, making the representation sensitive to local geometry.

---



# <font color='orange'> 5. Aggregation </font>

Each point

receives

EdgeConv features

from

its neighbours.

These are combined

using

a symmetric aggregation function,

typically

max pooling.

$$
\boxed{
x_i'
=
\max_{j\in\mathcal N(i)}
e_{ij}
}
$$

This preserves

permutation invariance

while capturing

local neighbourhood information.

---



# <font color='orange'> 6. Dynamic Graph Update </font>

After

every point

has

a new feature vector.

The graph

is then rebuilt.

```
Updated Features

↓

New k-NN Graph

↓

New EdgeConv Layer
```

Thus,

the notion

of

"neighbour"

evolves

throughout

training.

---

```
Current Features

↓

k-NN Search

↓

Edge Construction

↓

Shared MLP

↓

Max Aggregation

↓

Updated Features
```

This resembles

except

changes

between layers.

# <font color='blue'> Point Transformers: Attention-Based Learning on Point Clouds </font>

PointNet introduced permutation-invariant learning on unordered point clouds, while PointNet++ incorporated hierarchical local geometry. Dynamic Graph CNN further improved representation learning through evolving neighbourhood graphs. Despite these advances, neighbourhood aggregation remained largely based on fixed or manually defined operations.

**Point Transformers** introduce the attention mechanism into point cloud learning. Rather than treating neighbouring points equally, the network learns attention weights that determine how much each neighbouring point contributes to the updated representation.

By combining geometric information with self-attention, Point Transformers achieve state-of-the-art performance on many point cloud tasks and represent one of the most influential modern architectures in geometric deep learning.

# <font color='blue'> Learning Rate Scheduling: Controlling How Neural Networks Learn </font>

Training a neural network involves iteratively adjusting millions or even billions of parameters in order to minimize a loss function. One of the most important hyperparameters governing this optimization process is the **learning rate**. The learning rate determines the size of each parameter update and therefore controls both the speed and stability of learning.

Choosing a good learning rate is often more important than choosing the exact optimization algorithm. A learning rate that is too large may cause training to diverge, while a learning rate that is too small can lead to painfully slow convergence or poor local minima.

Rather than keeping the learning rate fixed throughout training, modern deep learning almost always employs **learning rate schedules**, which adapt the learning rate as optimization progresses.

---



# <font color='orange'> 1. Motivation </font>

Suppose

we are

descending

a mountain.

```
Large Steps

↓

Fast

↓

Risk Overshooting
```

Alternatively,

```
Tiny Steps

↓

Very Safe

↓

Very Slow
```

The ideal strategy

is

* large steps initially,
* smaller steps near the optimum.

Learning-rate schedules implement exactly this idea.

---



# <font color='orange'> 2. The Learning Rate </font>

Recall

gradient descent.

The parameter update is

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta
\nabla_\theta L
}
$$

where

* \(\theta\) denotes the model parameters,
* \(L\) is the loss,
* \(\eta\) is the learning rate.

The learning rate

controls

how far

we move

along

the gradient.

---



# <font color='orange'> 3. Large vs Small Learning Rates </font>

### Large Learning Rate

Advantages

* faster learning,
* rapid exploration.

Disadvantages

* oscillations,
* unstable training,
* possible divergence.

---

### Small Learning Rate

Advantages

* stable optimization,
* precise convergence.

Disadvantages

* slow learning,
* longer training times,
* greater risk of getting stuck.

---



# <font color='orange'> 4. Why Scheduling Helps </font>

Instead of using

one learning rate,

we gradually reduce it.

Conceptually,

```
Epoch

↓

Learning Rate

↓

Large → Small
```

This combines

fast exploration

with

careful refinement.

---



# <font color='orange'> 5. Step Decay </font>

One of the simplest schedules

reduces

the learning rate

after fixed intervals.

Example

```
Epochs 1–30

↓

0.01
```

```
Epochs 31–60

↓

0.001
```

```
Epochs 61–90

↓

0.0001
```

This approach

is easy to implement

but introduces abrupt changes.

---



# <font color='orange'> 6. Exponential Decay </font>

Instead of sudden drops,

the learning rate

decays smoothly.

$$
\boxed{
\eta_t
=
\eta_0
e^{-kt}
}
$$

where

\(k\)

controls

the decay rate.

---



# <font color='orange'> 7. Cosine Annealing </font>

Modern deep learning

often uses

cosine annealing.

The learning rate

follows

a cosine curve

from

its maximum

to

its minimum.

$$
\boxed{
\eta_t
=
\eta_{\min}
+
\frac12
(\eta_{\max}-\eta_{\min})
\left(
1+\cos
\frac{\pi t}{T}
\right)
}
$$

The decay

is smooth

and

typically improves convergence.

---



# <font color='orange'> 8. Learning Rate Warmup </font>

Large models,

especially Transformers,

often begin

with

a very small

learning rate.

```
Small

↓

Increase

↓

Peak

↓

Decay
```

Warmup

helps stabilize

early optimization

before switching

to the main schedule.

---



# <font color='orange'> 9. One-Cycle Policy </font>

The One-Cycle schedule

first

increases

the learning rate,

then

decreases it

below

its initial value.

```
Increase

↓

Peak

↓

Rapid Decay
```

This strategy,

introduced by Leslie Smith,

often accelerates convergence

and improves generalization.

---



# <font color='orange'> 10. TensorFlow Illustration </font>

TensorFlow

provides several built-in schedules.

Example

```python
schedule = tf.keras.optimizers.schedules.CosineDecay(

    initial_learning_rate=1e-3,

    decay_steps=10000
)
```

The optimizer

uses

the schedule

instead of

a fixed learning rate.

---



# <font color='orange'> 11. Applications </font>

Learning-rate scheduling

is used in

* CNNs,
* Transformers,
* Graph Neural Networks,
* Large Language Models,
* Reinforcement Learning,
* Scientific Machine Learning.

It is essentially universal in modern deep learning.

---



# <font color='orange'> 12. Advantages </font>

Learning-rate schedules provide

* faster convergence,
* improved stability,
* better final accuracy,
* reduced sensitivity to initialization,
* more efficient optimization.

---



# <font color='orange'> 13. Limitations </font>

Schedules also introduce

additional hyperparameters,

such as

* decay rate,
* minimum learning rate,
* warmup duration,
* cycle length.

Selecting these values

often requires experimentation.

---



# <font color='orange'> 14. Common Misconceptions </font>

### A Smaller Learning Rate Is Always Better

False.

Very small learning rates can make training prohibitively slow and may prevent effective exploration of the loss landscape.

---

### Every Model Should Use the Same Schedule

False.

Different architectures and datasets often benefit from different scheduling strategies.

---

### Warmup Is Only Useful for Transformers

False.

Although especially important for Transformers and other very deep networks, warmup can also stabilize the early stages of training for other architectures.

---

### Learning Rate Scheduling Replaces Good Optimizers

False.

The optimizer and the learning-rate schedule complement each other. Most state-of-the-art models rely on both.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Learning Rate | Step size in parameter updates |
| Fixed Learning Rate | Constant throughout training |
| Step Decay | Reduces the learning rate at predefined epochs |
| Exponential Decay | Smooth exponential reduction |
| Cosine Annealing | Smooth cosine-based decay |
| Warmup | Gradual increase at the start of training |
| One-Cycle Policy | Increase then decrease learning rate |
| Main Advantage | Faster, more stable optimization |
| Main Limitation | Additional hyperparameters to tune |

> **Key Insight:** The learning rate controls how quickly a neural network updates its parameters during optimization. Modern deep learning rarely uses a fixed learning rate; instead, carefully designed learning-rate schedules such as cosine annealing, warmup, and the One-Cycle policy enable faster convergence, greater stability, and improved generalization across a wide range of architectures.

# <font color='blue'> Modern Optimization Algorithms: SGD, Momentum, Adam, AdamW, Lion, and Beyond </font>

Training a neural network involves minimizing a loss function by iteratively updating millions or even billions of model parameters. The algorithm responsible for computing these parameter updates is called the **optimizer**.

While the learning rate determines the size of each update, the optimizer determines **how the gradients are interpreted** and **how previous optimization history influences future updates**. Modern optimizers accelerate convergence, improve stability, and help neural networks escape poor local minima or flat regions of the loss landscape.

This chapter introduces the evolution of optimization algorithms, from classical Gradient Descent to state-of-the-art methods such as AdamW and Lion.

---



# <font color='orange'> 1. Motivation </font>

Suppose

we wish

to descend

a mountain.

A naïve strategy

takes

the steepest downhill step

at every iteration.

```
Current Position

↓

Gradient

↓

Step
```

This works,

but

may be

slow,

unstable,

or trapped

in poor regions.

Modern optimizers

use

additional information

to improve

the descent.

---



# <font color='orange'> 2. Gradient Descent </font>

The simplest optimizer

updates

the parameters

using

the gradient.

$$
\boxed{
\theta_{t+1}
=
\theta_t
-
\eta
\nabla_\theta L
}
$$

where

* \(\theta\) denotes the parameters,
* \(L\) is the loss,
* \(\eta\) is the learning rate.

Advantages

* mathematically simple.

Disadvantages

* requires the full dataset each step,
* computationally expensive,
* slow convergence.

---



# <font color='orange'> 3. Stochastic Gradient Descent (SGD) </font>

Instead of

using

the entire dataset,

SGD updates

the parameters

using

small mini-batches.

```
Mini-batch

↓

Gradient

↓

Update
```

Advantages

* much faster,
* scalable,
* introduces useful gradient noise.

Disadvantages

* noisy updates,
* may oscillate.

---



# <font color='orange'> 4. Momentum </font>

Imagine

rolling

a heavy ball

downhill.

Instead of

responding only

to

the current slope,

the ball

also carries

its previous velocity.

Momentum introduces

a velocity term

$$
\boxed{
v_t
=
\beta v_{t-1}
+
\nabla_\theta L
}
$$

The parameters become

$$
\boxed{
\theta
=
\theta
-
\eta v_t
}
$$

Momentum

smooths

the optimization trajectory

and accelerates

movement

along consistent directions.

---



# <font color='orange'> 5. Nesterov Momentum </font>

Nesterov Accelerated Gradient

improves

ordinary momentum

by

looking ahead

before computing

the gradient.

Instead of asking

> "Where am I?"

it asks

> "Where will momentum take me next?"

This often provides

faster convergence

and better stability.

---



# <font color='orange'> 6. RMSProp </font>

Different parameters

may require

different learning rates.

RMSProp

tracks

the moving average

of

squared gradients.

Large gradients

receive

smaller updates.

Small gradients

receive

larger updates.

Thus,

every parameter

has

its own

adaptive learning rate.

---



# <font color='orange'> 7. Adam </font>

Adam

combines

two ideas.

* Momentum
* RMSProp

It maintains

both

the first moment

(mean)

and

the second moment

(variance)

of the gradients.

Parameter updates

are therefore

adaptive

and

stable.

Adam became

the default optimizer

for many deep learning applications.

---



# <font color='orange'> 8. AdamW </font>

Standard Adam

implements

weight decay

in a way

that is coupled

to

the adaptive gradient updates.

AdamW

decouples

weight decay

from

the gradient update,

providing

better regularization.

Today,

AdamW

is the optimizer

used by

most modern Transformers,

Vision Transformers,

and

Large Language Models.

---



# <font color='orange'> 9. Lion Optimizer </font>

Lion

(Layer-wise Optimization with Sign Momentum)

is a more recent optimizer.

Instead of

using

gradient magnitudes,

it updates

parameters

using

the **sign**

of the momentum.

Advantages include

* lower memory usage,
* competitive performance,
* faster training

for some large models.

Lion remains

an active research topic

and is not yet as universally adopted as AdamW.

---



# <font color='orange'> 10. TensorFlow Illustration </font>

TensorFlow provides

built-in implementations.

Examples

```python
optimizer = tf.keras.optimizers.SGD(
    learning_rate=0.01
)
```

```python
optimizer = tf.keras.optimizers.Adam(
    learning_rate=1e-3
)
```

```python
optimizer = tf.keras.optimizers.AdamW(
    learning_rate=1e-3,
    weight_decay=1e-4
)
```

These optimizers

can be combined

with

learning-rate schedules.

---



# <font color='orange'> 11. Applications </font>

Optimizer choice depends

on

the task.

Common choices include

* SGD + Momentum for image classification,
* Adam for many research prototypes,
* AdamW for Transformers,
* Lion for selected large-scale models.

Graph Neural Networks

typically use

Adam

or

AdamW.

---



# <font color='orange'> 12. Advantages </font>

Modern optimizers provide

* faster convergence,
* adaptive learning,
* improved stability,
* reduced manual tuning,
* better large-scale performance.

---



# <font color='orange'> 13. Limitations </font>

Adaptive optimizers

introduce

additional hyperparameters,

such as

* momentum coefficients,
* weight decay,
* epsilon values.

Furthermore,

no single optimizer

is optimal

for every problem.

---



# <font color='orange'> 14. Common Misconceptions </font>

### Adam Always Outperforms SGD

False.

SGD with Momentum often achieves better final generalization for image classification tasks, particularly when trained for sufficiently long.

---

### Adaptive Learning Rates Remove the Need for Scheduling

False.

Learning-rate schedules remain important even when using Adam or AdamW.

---

### More Complex Optimizers Are Always Better

False.

The simplest optimizer that performs well for a given problem is often the best choice.

---

### AdamW Is Just Adam with Weight Decay

Not exactly.

The key innovation is that the weight decay is **decoupled** from the adaptive gradient update rather than being implemented as an L2 penalty inside the gradient.

---

# <font color='purple'> 15. Conceptual Summary </font>

| Optimizer | Main Idea | Typical Use |
| :--- | :--- | :--- |
| Gradient Descent | Full-batch optimization | Teaching, small datasets |
| SGD | Mini-batch updates | Classical deep learning |
| SGD + Momentum | Velocity-based updates | CNNs, ResNets |
| Nesterov | Look-ahead momentum | Faster SGD variants |
| RMSProp | Adaptive learning rates | RNNs, earlier deep learning |
| Adam | Momentum + adaptive scaling | General-purpose optimizer |
| AdamW | Adam + decoupled weight decay | Transformers, ViTs, LLMs |
| Lion | Sign-based momentum updates | Emerging large-scale models |

> **Key Insight:** Modern optimizers improve upon basic gradient descent by incorporating information from previous gradients and adapting learning rates for individual parameters. While SGD with Momentum remains highly competitive for many vision tasks, AdamW has become the standard optimizer for most modern Transformer-based architectures due to its stability and effective regularization.

# <font color='blue'> Mixed-Precision Training: Faster Deep Learning with Less Memory </font>

Modern deep learning models often contain hundreds of millions or even billions of parameters. Training these models efficiently requires substantial computational resources and memory. One of the most important advances in practical deep learning has been the use of **mixed-precision training**, which combines different numerical precisions to accelerate computation while maintaining numerical stability.

Rather than performing every computation using 32-bit floating-point arithmetic (FP32), mixed-precision training strategically uses lower-precision formats such as **FP16** or **BF16** for most operations while retaining FP32 where higher numerical accuracy is required.

This approach significantly reduces memory consumption and increases computational throughput on modern hardware without sacrificing model quality.

---



# <font color='orange'> 1. Motivation </font>

Suppose

a neural network

contains

one billion

parameters.

If every parameter

uses

32 bits,

memory requirements

become enormous.

What if

many computations

could instead use

16 bits?

Memory usage

would nearly halve,

and modern GPUs

could perform

many more operations

per second.

---



# <font color='orange'> 2. Floating-Point Precision </font>

Common numerical formats include

| Format | Bits |
| :--- | ---: |
| FP64 | 64 |
| FP32 | 32 |
| FP16 | 16 |
| BF16 | 16 |

Lower precision

requires

less memory

and

faster computation.

However,

it also

reduces

numerical accuracy.

---



# <font color='orange'> 3. Why FP32 Became Standard </font>

Historically,

deep learning

used

FP32

because

it provides

sufficient numerical precision

for

gradient-based optimization.

FP64

is rarely needed

because

it doubles

memory usage

while offering

little improvement

for most neural networks.

---



# <font color='orange'> 4. FP16 </font>

FP16

stores

numbers

using

16 bits.

Advantages include

* reduced memory,
* higher throughput,
* Tensor Core acceleration,
* faster training.

Potential disadvantages include

* smaller numerical range,
* gradient underflow,
* overflow for large values.

These issues

require

special handling.

---



# <font color='orange'> 5. BF16 </font>

Brain Floating Point (BF16)

also uses

16 bits,

but

allocates

more bits

to the exponent

than FP16.

Advantages include

* larger dynamic range,
* fewer overflow problems,
* improved numerical stability.

BF16

has become

the preferred format

on many modern accelerators,

including Google TPUs and newer NVIDIA GPUs.

---



# <font color='orange'> 6. Mixed Precision </font>

Instead of choosing

one precision,

mixed precision

combines

multiple formats.

Typically,

```
Forward Pass

↓

FP16 / BF16
```

```
Backward Pass

↓

FP16 / BF16
```

```
Master Weights

↓

FP32
```

Critical quantities,

such as

optimizer states

and

master parameters,

remain

in FP32.

---



# <font color='orange'> 7. Loss Scaling </font>

FP16

may produce

very small gradients

that underflow

to zero.

To avoid this,

the loss

is multiplied

by

a scaling factor

before backpropagation.

```
Loss

↓

Multiply

↓

Backpropagation

↓

Divide Gradients
```

This technique

is called

**loss scaling**.

Modern frameworks

perform

dynamic loss scaling

automatically.

---



# <font color='orange'> 8. Tensor Cores </font>

Modern NVIDIA GPUs

contain

specialized hardware

called

**Tensor Cores**.

Tensor Cores

perform

matrix multiplication

extremely efficiently

using

FP16,

BF16,

or

TensorFloat-32 (TF32).

Mixed precision

allows

the hardware

to operate

at maximum efficiency.

---



# <font color='orange'> 9. TensorFlow Illustration </font>

No further changes

is used for

# <font color='orange'> 12. Limitations </font>

Fortunately,

# <font color='orange'> 13. Common Misconceptions </font>

False.
