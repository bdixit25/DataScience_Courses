# <font color='blue'> Batch, Stochastic, and Mini-Batch Gradient Descent </font>

Gradient descent updates the parameters of a neural network by moving them in the direction that reduces the loss function. An important practical question, however, is **how much training data should be used to compute each gradient update?** Should the gradients be calculated using the entire training dataset, a single training example, or a subset of the data?

The answer leads to three principal training strategies: **Batch Gradient Descent**, **Stochastic Gradient Descent (SGD)**, and **Mini-Batch Gradient Descent**. These methods differ primarily in the amount of data used to estimate the gradient during each update, leading to different trade-offs in computational cost, convergence speed, memory requirements, and optimization behaviour.

This chapter introduces these three approaches, compares their characteristics, explains the concepts of epochs, batches, and iterations, and discusses why mini-batch gradient descent has become the standard optimization strategy in modern deep learning.

---



# <font color='orange'> 1. Why Does Batch Size Matter? </font>

Suppose a training dataset contains

```
1,000,000 examples.
```

Should we

* process all one million examples before updating the weights?

or

* update the weights after every individual example?

Both approaches are possible,

but they have different computational and statistical properties.

---



# <font color='orange'> 2. Batch Gradient Descent </font>

In **Batch Gradient Descent**,

the gradient is computed using

the **entire training dataset**.

Conceptually,

```
Entire Dataset

↓

Forward Propagation

↓

Loss

↓

Backpropagation

↓

Gradient

↓

Update Weights
```

Only one parameter update occurs after processing all training examples.

---

### Advantages

* Stable gradient estimates.
* Smooth convergence.
* Deterministic updates.

---

### Limitations

* Computationally expensive.
* Requires large memory.
* Slow for large datasets.

Batch Gradient Descent is therefore rarely used for modern large-scale deep learning.

---



# <font color='orange'> 3. Stochastic Gradient Descent (SGD) </font>

In **Stochastic Gradient Descent**,

the parameters are updated

after processing

**one training example at a time**.

Conceptually,

```
One Example

↓

Forward Pass

↓

Loss

↓

Backpropagation

↓

Update Weights
```

The process repeats for every training example.

---

### Advantages

* Very fast updates.
* Low memory requirements.
* Can escape shallow local minima and saddle points due to noisy gradients.

---

### Limitations

* Highly noisy optimization path.
* Loss fluctuates significantly.
* May require many updates before converging.

---



# <font color='orange'> 4. Mini-Batch Gradient Descent </font>

Mini-Batch Gradient Descent combines the advantages of Batch Gradient Descent and SGD.

Instead of using

```
Entire Dataset
```

or

```
One Example,
```

it processes

a small subset,

called a **mini-batch**.

Typical batch sizes are

```
16

32

64

128

256
```

Conceptually,

```
Mini-Batch

↓

Forward Pass

↓

Loss

↓

Backpropagation

↓

Update Weights
```

This process repeats until all mini-batches have been processed.

---



# <font color='orange'> 5. Comparing the Three Methods </font>

| Method | Data Used Per Update | Update Frequency |
| :--- | :--- | :--- |
| Batch Gradient Descent | Entire dataset | Once per epoch |
| Stochastic Gradient Descent | One example | Every example |
| Mini-Batch Gradient Descent | Small subset | Every mini-batch |

Mini-batch training has become the standard because it balances computational efficiency with reliable optimization.

---



# <font color='orange'> 6. Epochs, Batches, and Iterations </font>

These three terms are closely related.

### Epoch

One **epoch** means that

every training example

has been processed once.

---

### Batch

A **batch**

is the group of examples

used for one gradient computation.

---

### Iteration

An **iteration**

is one parameter update.

Suppose

```
Training Examples = 10,000

Batch Size = 100
```

Then

```
Iterations per Epoch

=

10,000 / 100

=

100
```

Thus,

one epoch consists of

100 iterations.

# <font color='orange'> 8. Why Mini-Batches Work So Well </font>

Mini-batches provide several practical advantages.

* Efficient use of GPU parallelism.
* Lower memory usage than full-batch training.
* Less noisy than pure SGD.
* Faster convergence than batch gradient descent in many practical settings.

These properties explain why nearly all modern deep learning frameworks use mini-batch training by default.

---



# <font color='orange'> 9. Choosing the Batch Size </font>

There is no universally optimal batch size.

Common choices include

| Batch Size | Characteristics |
| :---: | :--- |
| 16–32 | More frequent updates, noisier gradients |
| 64–128 | Common compromise between speed and stability |
| 256+ | Efficient on powerful hardware but may require more memory |

The optimal value depends on

* dataset size,
* available GPU memory,
* network architecture,
* optimization algorithm.

# <font color='orange'> 12. Common Misconceptions </font>

### One Epoch Means One Update

False.

The number of updates depends on the batch size.

A single epoch may contain many iterations.

---

### Larger Batch Sizes Always Train Faster

False.

Larger batches improve hardware utilization but require more memory and may not always improve convergence.

---

### SGD Always Uses the Entire Dataset

False.

Despite its name,

Stochastic Gradient Descent updates the parameters after processing **one** training example.

Mini-batch gradient descent is the approach most commonly referred to as "SGD" in many deep learning libraries.

# <font color='purple'> 13. Conceptual Summary </font>

| Concept | Description |
| :--- | :--- |
| Batch Gradient Descent | Uses the entire dataset for each update |
| Stochastic Gradient Descent | Uses one training example per update |
| Mini-Batch Gradient Descent | Uses a small subset of the data for each update |
| Epoch | One complete pass through the training dataset |
| Batch | Group of examples processed together |
| Iteration | One parameter update |
| Standard Practice | Mini-batch gradient descent |
| Main Goal | Efficiently estimate gradients while balancing computational cost and convergence behaviour |

> **Key Insight:** The distinction between batch, stochastic, and mini-batch gradient descent lies in how much data are used to estimate the gradient before updating the network parameters. Batch Gradient Descent provides stable but computationally expensive updates, while Stochastic Gradient Descent offers rapid but noisy updates. Mini-Batch Gradient Descent combines the strengths of both approaches by using small subsets of the training data, making it the standard training strategy for modern deep learning due to its efficiency, scalability, and favourable optimization behaviour.
